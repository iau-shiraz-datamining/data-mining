<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: سنجش رفتار سفرهای شهری در «اوبر» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    سامانه درخواستی خودرو «اوبر» برای تحلیل رفتار سفرهای شهری از دو ابزار مکمل داده‌کاوی استفاده می‌کند. هنگامی که مسئله بررسی هم‌زمانی استفاده از سرویس‌های کیفی مختلف (مانند انتخاب سرویس UberX همراه با کرایه صندلی کودک) باشد، از <b>تحلیل ارتباط (Association Analysis)</b> روی داده‌های گسسته استفاده می‌شود. اما زمانی که هدف بررسی رابطه متغیرهای پیوسته و عددی (مانند رابطه بین «مسافت سفر» و «مبلغ کرایه» یا «مدت زمان انتظار» و «امتیاز مسافر») باشد، سیستم به سراغ <b>تحلیل همبستگی (Correlation Analysis)</b> می‌رود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    فرمول ضریب همبستگی پیرسون (Pearson Correlation Coefficient):<br>
    جهت و شدت رابطه خطی بین دو متغیر پیوسته عددی X و Y بر اساس کوواریانس آن‌ها تقسیم بر حاصل‌ضرب انحراف معیارهایشان محاسبه می‌شود:
  </p>

  <div style="text-align: center; margin: 15px 0;">
    $$r_{XY} = \frac{\text{Cov}(X, Y)}{\sigma_X \sigma_Y} = \frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum_{i=1}^{n} (x_i - \bar{x})^2} \sqrt{\sum_{i=1}^{n} (y_i - \bar{y})^2}}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    تفسیر مقادیر ضریب همبستگی پیرسون ($r$):
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    • همبستگی مثبت کامل ($r = +1.0$): افزایش مستقیم یک متغیر با افزایش متغیر دیگر (مثل مسافت سفر و مبلغ کرایه ثابت).
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    • عدم همبستگی خطی ($r = 0.0$): عدم وجود هیچ‌گونه رابطه خطی سیستماتیک بین دو متغیر عددی.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    • همبستگی منفی کامل ($r = -1.0$): افزایش یک متغیر با کاهش متغیر دیگر (مثل زمان انتظار تا سوار شدن و رضایت مسافر).
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: اوبر با گذار از داده‌های گسسته سبد خدمات به متغیرهای پیوسته عددی، می‌تواند نرخ کرایه‌های پویا (Surge Pricing) و مدل‌های پیش‌بینی زمان رسیدن (ETA) را با دقت بالای آماری تنظیم کند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: انتقال از قواعد کیفی هم‌وقوعی به روابط دقیق و کمی خطی بین متغیرهای پیوسته سیستم
    </div>
  </p>

</div>

```python
display_academic_text(
    "سنجش رفتار سفرهای شهری در «اوبر»: از قواعد گسسته (Association) تا همبستگی پیوسته (Correlation) 🚕", 
    "main_title"
)

display_academic_text(
    "سامانه هوشمند «اوبر» (Uber) برای تحلیل رفتار مسافران از دو ابزار متفاوتِ ریاضی استفاده می‌کند. "
    "برای کشف الگو در <b>داده‌های کِیفی و گُسسته (Categorical Data)</b> (مثلاً انتخاب سرویس <i>UberX</i> به صورت همزمان با <i>صندلی کودک</i>) از الگوریتم‌های <b>تحلیل ارتباط (Association Rules)</b> مانند Apriori استفاده می‌شود. "
    "اما زمانی که هدف، بررسی ارتباط خطی میان <b>متغیرهایِ پیوسته و عددی (Continuous Data)</b> (مانند <i>مسافت سفر در برابر مبلغ کرایه</i>) باشد، سیستم به سراغ <b>تحلیل همبستگی (Correlation Analysis)</b> می‌رود.", 
    "paragraph"
)

display_academic_text("فرمول ضریب همبستگی پیرسون (Pearson Correlation Coefficient):", "sub_title")
display_academic_text(
    "جهت و شدتِ رابطهٔ خَطیِ بین دو متغیرِ پیوستهٔ عددی ($X$ و $Y$)، بر اساس کوواریانس (Covariance) آن‌ها تقسیم بر حاصل‌ضربِ انحراف‌معیارهایشان (Standard Deviations) محاسبه می‌شود:", 
    "paragraph"
)

display_academic_text(
    r"$$r_{XY} = \frac{\text{Cov}(X,Y)}{\sigma_X \sigma_Y} = \frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum_{i=1}^{n} (x_i - \bar{x})^2} \sqrt{\sum_{i=1}^{n} (y_i - \bar{y})^2}}$$", 
    "paragraph"
)

display_academic_text("تفسیر مقادیر ضریب همبستگی ($r$):", "sub_title")
display_academic_text(
    "• <b>همبستگی مثبت و مستقیم ($+1.0$):</b> افزایش یک متغیر دقیقاً باعث افزایش متغیر دیگر می‌شود (مثل <i>طول مسافت</i> و <i>هزینه سفر</i>).<br>"
    "• <b>عدم همبستگی و استقلال کامل ($0.0$):</b> هیچ رابطهٔ خطی میان دو متغیر وجود ندارد.<br>"
    "• <b>همبستگی منفی و معکوس ($-1.0$):</b> افزایش یک متغیر منجر به کاهش متغیر دیگر می‌گردد (مثل <i>زمان انتظار مسافر</i> و <i>امتیاز ثبت‌شده برای راننده</i>).", 
    "paragraph"
)


# ==========================================================
# 3. SECTION 2: PHASE 1 - ASSOCIATION ANALYSIS (CATEGORICAL)
# ==========================================================
display_academic_text("فاز اول: کاوشِ الگوهایِ تصمیم‌گیری (داده‌های گسسته/کیفی)", "sub_title")

# Educational Step 1: Discrete Categorical Data
# Users combining specific service features in one trip.
services_data = [
    ['UberX', 'Child Seat', 'Airport Trip'],
    ['UberXL', 'Extra Luggage'],
    ['UberX', 'Airport Trip'],
    ['UberBlack', 'Pet Friendly', 'Airport Trip'],
    ['UberX', 'Child Seat', 'Airport Trip'],
    ['UberXL', 'Extra Luggage', 'Airport Trip'],
    ['UberX', 'Child Seat'],
    ['UberBlack', 'Airport Trip']
]

te = TransactionEncoder()
df_categorical = pd.DataFrame(te.fit(services_data).transform(services_data), columns=te.columns_)

# Extract frequent itemsets and rules
frequent_itemsets = apriori(df_categorical, min_support=0.25, use_colnames=True)
rules = association_rules(frequent_itemsets, metric='confidence', min_threshold=0.60)

# Format rules for UI
rules['antecedents_str'] = rules['antecedents'].apply(lambda x: ', '.join(list(x)))
rules['consequents_str'] = rules['consequents'].apply(lambda x: ', '.join(list(x)))
rules['rule'] = rules['antecedents_str'] + ' -> ' + rules['consequents_str']

df_rules_display = rules[['rule', 'support', 'confidence', 'lift']].sort_values(by='lift', ascending=False)

display_side_by_side_datasets(
    [df_rules_display.head(5).reset_index(drop=True)],
    ["قواعدِ انجمنیِ استخراج‌شده از داده‌های گسسته (تصمیمات کِیفی مسافر)"]
)


# ==========================================================
# 4. SECTION 3: PHASE 2 - CORRELATION ANALYSIS (CONTINUOUS)
# ==========================================================
display_academic_text("فاز دوم: محاسبهٔ همبستگیِ خطی (داده‌های پیوسته/کَمّی)", "sub_title")

# Educational Step 2: Continuous Numerical Data Simulation
# We simulate a real-world scenario where distance directly affects 
# fare, and wait time negatively affects driver ratings.
n_rides = 100

trip_distance = np.random.uniform(2, 25, n_rides) # Distance in km

# Fare = Distance * Base Rate + Noise (Strong Positive Correlation)
fare_amount = trip_distance * 2.5 + np.random.normal(3, 2, n_rides) 

wait_time = np.random.uniform(1, 15, n_rides) # Wait time in minutes

# Rating = Base Score - (Wait Time * Penalty) + Noise (Strong Negative Correlation)
driver_rating = 5.0 - (wait_time * 0.15) + np.random.normal(0, 0.2, n_rides) 
driver_rating = np.clip(driver_rating, 1.0, 5.0) # Ratings logically bounded [1, 5]

df_numerical = pd.DataFrame({
    'Distance_km': trip_distance,
    'Fare_USD': fare_amount,
    'Wait_Time_min': wait_time,
    'Rating': driver_rating
})

# Educational Step 3: Pearson Correlation Matrix
# Calculates the linear relationship coefficient (r) between ALL continuous variables.
corr_matrix = df_numerical.corr(method='pearson')

display_side_by_side_datasets(
    [df_numerical.head(6).reset_index(drop=True), corr_matrix.reset_index().rename(columns={'index': 'Variables'})],
    [
        "نمونه رکوردهای داده‌های پیوسته (سفرهای شهری)", 
        "ماتریس همبستگی پیرسون (Pearson Correlation Matrix)"
    ]
)


# ==========================================================
# 5. SECTION 4: ADVANCED ACADEMIC VISUALIZATIONS
# ==========================================================
display_academic_text("مصورسازی تحلیلی: خطِ روندِ خطی (Linear Trend) و نقشه حرارتی (Heatmap)", "sub_title")

fig, axes = plt.subplots(1, 2, figsize=(18, 6.5), dpi=135)

# ----------------------------------------------------------
# Subplot 1: Scatter Plot of Continuous Variables (Distance vs Fare)
# ----------------------------------------------------------
scatter = axes[0].scatter(
    df_numerical['Distance_km'],
    df_numerical['Fare_USD'],
    c=df_numerical['Fare_USD'], # Color mapping based on Fare
    cmap='Blues',               # Academic Blue Theme
    edgecolor='#2c3e50',
    alpha=0.9,
    linewidth=1.0,
    s=120                       # Bubble size
)

# Educational Step 4: Calculate and draw the Linear Trendline (Line of Best Fit)
m, b = np.polyfit(df_numerical['Distance_km'], df_numerical['Fare_USD'], 1)
pearson_r = corr_matrix.loc["Distance_km", "Fare_USD"]

axes[0].plot(
    df_numerical['Distance_km'], 
    m * df_numerical['Distance_km'] + b, 
    color='#e74c3c', # Red Alert trendline for contrast
    linestyle='-', 
    linewidth=3.0, 
    label=f'Line of Best Fit (Pearson $r = {pearson_r:.2f}$)'
)

axes[0].annotate(
    "Perfect Positive Correlation:\nNotice how tightly the points cluster\naround the red linear trendline.\nAs distance increases, fare reliably increases.",
    xy=(15, m * 15 + b), xytext=(3, 60),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title('Figure 1: Positive Linear Correlation (Trip Distance vs Fare Amount)', fontsize=14, fontweight='bold', pad=15)
axes[0].set_xlabel('Trip Distance Driven (Kilometers)', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Total Fare Paid ($USD)', fontsize=13, fontweight='bold')
axes[0].grid(True, linestyle='--', alpha=0.5)
axes[0].legend(loc='lower right', frameon=True, fontsize=11, edgecolor='#bdc3c7')

# ----------------------------------------------------------
# Subplot 2: Correlation Matrix Heatmap
# ----------------------------------------------------------
# Using RdBu_r (Red-Blue reversed) for a standard academic correlation heatmap
# Deep Blue = Positive Correlation (+1.0)
# Deep Red = Negative Correlation (-1.0)
sns.heatmap(
    corr_matrix, 
    ax=axes[1], 
    annot=True,        # Write the data value in each cell
    fmt=".2f",         # 2 decimal places
    cmap="RdBu",       # Academic standard for Correlations
    vmin=-1.0, 
    vmax=1.0, 
    center=0, 
    square=True, 
    linewidths=2, 
    linecolor='white',
    annot_kws={"size": 13, "fontweight": "bold"},
    cbar_kws={'label': 'Pearson Correlation Coefficient ($r$)'}
)

axes[1].set_title('Figure 2: Comprehensive Pearson Correlation Matrix Heatmap', fontsize=14, fontweight='bold', pad=15)
# Rotate x-labels for better readability
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=25, ha='right', fontweight='bold', fontsize=11)
axes[1].set_yticklabels(axes[1].get_yticklabels(), rotation=0, fontweight='bold', fontsize=11)

for ax in axes:
    # Heatmap doesn't have standard spines to remove, so we only apply to Scatter
    if ax == axes[0]:
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)

plt.subplots_adjust(wspace=0.25)
plt.show()

# ==========================================================
# 6. SECTION 5: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک برای طراحی سیستم‌های قیمت‌گذاری (Strategic Implication):</b><br>"
    r"با بررسی دقیق گراف ۱ و ماتریس همبستگی پیرسون (گراف ۲)، مشاهده می‌کنیم که متغیر «مسافت سفر» دارای <b>همبستگی خطی مثبتِ بسیار قوی ($r = 0.99$)</b> با «مبلغ کرایه» است. (آبی تیره در ماتریس).<br>"
    r"در نقطهٔ مقابل، متغیر «زمانِ انتظار مسافر» دارای <b>همبستگیِ منفیِ معنادار ($r = -0.96$)</b> با «امتیاز نهایی راننده (Rating)» است؛ "
    r"یعنی هرچه زمانِ انتظار بیشتر طول بکشد (رشد متغیر $X$)، امتیاز راننده با شیبی تند کاهش می‌یابد (اُفت متغیر $Y$) که با رنگ قرمز تیره در ماتریس مشخص شده است.<br>"
    r"سیستم‌های اَبَرهوشمندِ اوبر (Uber) با ترکیبِ <i>تحلیلِ ارتباطِ گسسته</i> (مثلاً شناسایی نیاز به ماشین بزرگ در فرودگاه با استفاده از Apriori) و <i>تحلیلِ همبستگیِ پیوسته</i> (فرمول پیرسون)، "
    r"الگوریتم‌هایِ بی‌نقصِ <b>قیمت‌گذاریِ پویا (Dynamic Pricing / Surge Pricing)</b> را در زمان واقعی اجرا می‌کنند."
)
display_academic_text(insight_conclusion, "insight")
```
