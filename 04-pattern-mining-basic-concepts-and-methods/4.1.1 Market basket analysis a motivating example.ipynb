<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: تحلیل سبد خرید و پیشنهاد هوشمند کالا در «دیجی‌کالا» 🛒
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    تیم داده‌کاوی «دیجی‌کالا» روزانه میلیون‌ها سبد خرید نهایی‌شده را بررسی می‌کند تا ببیند کدام محصولات بیشتر همراه با یکدیگر خریداری می‌شوند. اگر سیستم متوجه شود که خریداران گوشی موبایل معمولاً محافظ صفحه نمایش (گلس) و قاب محافظ را هم در همان سفارش قرار می‌دهند، می‌تواند از این الگو برای پیشنهاد خودکار محصولات مکمل (Cross-selling) و ایجاد بسته‌های تخفیفی (Bundling) استفاده کند. این تحلیل به <b>تحلیل سبد خرید (Market Basket Analysis)</b> معروف است.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و معیارهای اصلی در استخراج قواعد ارتباطی (Association Rules):
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. پشتیبانی (Support):<br>
    نسبت سبدهایی که شامل هر دو کالای X و Y هستند به کل سبدهای خرید ثبت‌شده در دیجی‌کالا. این معیار فراوانی هم‌زمان دو کالا را نشان می‌دهد:
  </p>

  <div style="text-align: center; margin: 15px 0;">
    $$\text{Support}(X \rightarrow Y) = \frac{\text{Count}(X \cup Y)}{N}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. اطمینان (Confidence):<br>
    احتمال اینکه مشتری کالای Y را بخرد، به شرطی که قبلاً کالای X را به سبد خرید خود اضافه کرده باشد. این معیار درصد صحت و قابلیت اعتماد قاعده را می‌سنجد:
  </p>

  <div style="text-align: center; margin: 15px 0;">
    $$\text{Confidence}(X \rightarrow Y) = \frac{\text{Support}(X \cup Y)}{\text{Support}(X)}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. لیفت یا قدرت ارتباط (Lift):<br>
    نسبت اطمینان واقعی قاعده به احتمال خرید تصادفی و مستقل کالای Y. اگر لیفت بزرگ‌تر از ۱ باشد (Lift > 1)، نشان می‌دهد که خرید کالای X به‌طور معناداری شانس خرید کالای Y را افزایش می‌دهد:
  </p>

  <div style="text-align: center; margin: 15px 0;">
    $$\text{Lift}(X \rightarrow Y) = \frac{\text{Confidence}(X \rightarrow Y)}{\text{Support}(Y)}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: دیجی‌کالا با شناسایی قواعدی که لیفت بالا و اطمینان مناسبی دارند، بخش «خریداران این محصول، کالاهای زیر را هم خریده‌اند» را در پایین صفحه محصولات نمایش می‌دهد و فروش محصولات جانبی را به شدت افزایش می‌دهد.<br><br>

    </div>
  </p>

```python
display_academic_text(
    "تحلیل سبد خرید و پیشنهاد هوشمند کالا در «دیجی‌کالا» (Market Basket Analysis) 🛒", 
    "main_title"
)

display_academic_text(
    "تیم داده‌کاوی «دیجی‌کالا» روزانه میلیون‌ها سبد خرید نهایی‌شده را بررسی می‌کند تا دریابد کدام محصولات بیشتر همراه با یکدیگر خریداری می‌شوند. "
    "این فرآیند به <b>تحلیل سبد خرید (Market Basket Analysis)</b> معروف است.", 
    "paragraph"
)

display_academic_text("معیارهای اصلی در استخراج قواعد ارتباطی (Association Rules):", "sub_title")

display_academic_text(
    "<b>۱. پشتیبانی (Support):</b> فراوانی هم‌زمانِ دو کالا در کُل سبدهای خرید سیستم.<br>"
    r"$$\text{Support}(X \rightarrow Y) = \frac{\text{Count}(X \cup Y)}{N}$", 
    "paragraph"
)

display_academic_text(
    "<b>۲. اطمینان (Confidence):</b> درصد صحت قاعده؛ یعنی اگر ک کالای X خریده شود، با چه احتمالی ک کالای Y نیز در سبد حضور دارد.<br>"
    r"$$\text{Confidence}(X \rightarrow Y) = \frac{\text{Support}(X \cup Y)}{\text{Support}(X)}$$", 
    "paragraph"
)

display_academic_text(
    "<b>۳. لیفت (Lift):</b> میزانِ قدرت و کششِ قاعده. اگر $\text{Lift} > 1$ باشد، یعنی خرید X شانسِ خرید Y را به شدت افزایش می‌دهد (کالای مُمکّن و مکمل).", 
    "paragraph"
)
display_academic_text(
    r"$$\text{Lift}(X \rightarrow Y) = \frac{\text{Confidence}(X \rightarrow Y)}{\text{Support}(Y)}$$", 
    "paragraph"
)


# ==========================================================
# 3. SECTION 2: DATA PREPARATION & ONE-HOT ENCODING
# ==========================================================
display_academic_text("مرحله نخست: آماده‌سازی سبدهای خرید و رمزگذاری ماتریسی (One-Hot Encoding)", "sub_title")

transactions = [
    ['Mobile', 'Screen_Protector', 'Case'],
    ['Mobile', 'Screen_Protector', 'Charger'],
    ['Mobile', 'Case', 'Headphones'],
    ['Screen_Protector', 'Case'],
    ['Mobile', 'Screen_Protector', 'Case', 'Powerbank'],
    ['Mobile', 'Charger'],
    ['Screen_Protector', 'Case', 'Powerbank'],
    ['Mobile', 'Screen_Protector', 'Case'],
    ['Mobile', 'Headphones'],
    ['Screen_Protector', 'Case'],
]

df_raw_baskets = pd.DataFrame({
    'Order_ID': range(1, len(transactions) + 1),
    'Purchased_Items': [', '.join(basket) for basket in transactions]
})

# Convert list of lists into boolean matrix (One-Hot Encoding) for Apriori
te = TransactionEncoder()
te_array = te.fit(transactions).transform(transactions)
df_baskets_encoded = pd.DataFrame(te_array, columns=te.columns_)

display_side_by_side_datasets(
    [df_raw_baskets.head(6), df_baskets_encoded.astype(int).head(6)],
    [
        "نمونه سبدهای خرید خام دیجی‌کالا", 
        "ماتریس One-Hot Encoding (آماده برای اجرای Apriori)"
    ]
)


# ==========================================================
# 4. SECTION 3: EXECUTING APRIORI & EXTRACTING RULES
# ==========================================================
display_academic_text("مرحله دوم: اجرای الگوریتم Apriori و استخراج قواعد طلایی کراس‌سِلینگ", "sub_title")

# Find Frequent Itemsets (Min Support = 20%)
frequent_itemsets = apriori(
    df_baskets_encoded, 
    min_support=0.20, 
    use_colnames=True
)
frequent_itemsets['itemset_len'] = frequent_itemsets['itemsets'].apply(len)
frequent_itemsets_sorted = frequent_itemsets.sort_values(by='support', ascending=False)

df_freq_display = frequent_itemsets_sorted.copy()
df_freq_display['itemsets'] = df_freq_display['itemsets'].apply(lambda x: ', '.join(list(x)))

# Generate Association Rules (Min Confidence = 50%)
rules = association_rules(
    frequent_itemsets, 
    metric='confidence', 
    min_threshold=0.50
)

rules['antecedents_str'] = rules['antecedents'].apply(lambda x: ', '.join(list(x)))
rules['consequents_str'] = rules['consequents'].apply(lambda x: ', '.join(list(x)))
rules['rule_label'] = rules['antecedents_str'] + ' -> ' + rules['consequents_str']

top_rules_df = rules[['rule_label', 'support', 'confidence', 'lift']].sort_values(by='lift', ascending=False)

display_side_by_side_datasets(
    [df_freq_display.head(6).reset_index(drop=True), top_rules_df.head(6).reset_index(drop=True)],
    [
        "مجموعه‌های پرتکرار (حداقل پشتیبانی ۲۰٪)", 
        "قواعد ارتباطی استخراج‌شده (مرتب‌شده بر اساس ضریب Lift)"
    ]
)


# ==========================================================
# 5. SECTION 4: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
display_academic_text("مرحله سوم: مصورسازی تحلیلیِ قواعد ارتباطی (Support vs Confidence)", "sub_title")

fig, axes = plt.subplots(1, 2, figsize=(18, 6.5), dpi=135)

# ----------------------------------------------------------
# Subplot 1: Scatter plot (Support vs Confidence colored by Lift)
# ----------------------------------------------------------
scatter = axes[0].scatter(
    rules['support'],
    rules['confidence'],
    c=rules['lift'],
    s=rules['lift'] * 300,  # Bubble size represents Lift power
    cmap='Blues',           # Academic Blue Palette
    edgecolor='#2c3e50',
    alpha=0.9,
    linewidth=1.5
)

axes[0].annotate(
    "Rule Strength:\nBubbles positioned higher (Confidence)\nand further right (Support) with darker\nshades (Lift) represent high-value cross-sell rules.",
    xy=(0.4, 0.8), xytext=(0.2, 0.5),
    fontsize=11, fontweight='bold', color='#2980b9',
    arrowprops=dict(arrowstyle="->", color="#2980b9", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#ecf0f1", ec="#bdc3c7", lw=1.5)
)

axes[0].set_title('Figure 1: Association Rules Map (Support vs Confidence)', fontsize=14, fontweight='bold', pad=15)
axes[0].set_xlabel('Support Frequency', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Confidence Ratio', fontsize=13, fontweight='bold')
axes[0].grid(True, linestyle='--', alpha=0.5)

cbar = fig.colorbar(scatter, ax=axes[0], shrink=0.8, pad=0.05)
cbar.set_label('Lift Ratio Power', fontweight='bold', fontsize=11)

# ----------------------------------------------------------
# Subplot 2: Top Association Rules by Lift (Bar Chart)
# ----------------------------------------------------------
top_rules_bar = rules.sort_values(by='lift', ascending=True).tail(6)

bars = axes[1].barh(
    top_rules_bar['rule_label'],
    top_rules_bar['lift'],
    color='#1abc9c', # Turquoise
    edgecolor='#2c3e50',
    linewidth=1.5,
    height=0.55
)

axes[1].axvline(
    1.0,
    color='#e74c3c',
    linestyle='--',
    linewidth=2.5,
    label='Baseline Correlation (Lift = 1.0)'
)

axes[1].annotate(
    "Positive Correlation:\nAny rule with a Lift greater than 1.0\nindicates items that attract each other\nstrongly in customer baskets.",
    xy=(1.3, 1.0), xytext=(1.5, 2.0),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[1].set_title('Figure 2: Top Cross-Selling Rules Ranked by Lift', fontsize=14, fontweight='bold', pad=15)
axes[1].set_xlabel('Lift Ratio (Higher = Stronger Complement)', fontsize=13, fontweight='bold')
axes[1].set_ylabel('Association Rule', fontsize=13, fontweight='bold')
axes[1].legend(loc='lower right', frameon=True, fontsize=11, edgecolor='#bdc3c7')
axes[1].grid(axis='x', linestyle='--', alpha=0.5)
axes[1].set_yticklabels(top_rules_bar['rule_label'], fontweight='bold', fontsize=11)

for i, val in enumerate(top_rules_bar['lift']):
    axes[1].text(
        val + 0.03, 
        i, 
        f'{val:.2f}x', 
        va='center', fontweight='bold', fontsize=12, color='#2c3e50'
    )

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# 6. SECTION 5: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک در مهندسی داده و بازاریابی دیجیتال (Strategic Implication):</b><br>"
    r"کشفِ قواعدی که هم‌زمان دارایِ <b>لیفت بالا (Lift > 1)</b> و <b>اطمینان مناسب (Confidence)</b> هستند، یک معدن طلا برای دیجی‌کالا محسوب می‌شود.<br>"
    r"همان‌طور که در خروجی‌ها مشاهده می‌کنید، قاعدهٔ <code>(Screen_Protector -> Case)</code> دارای بالاترین ضریب لیفت است؛ یعنی خرید محافظ صفحه، احتمالِ خرید قاب گوشی را به‌شدت بالا می‌برد. "
    r"تیم مارکتینگ می‌تواند با نمایش هوشمندانهٔ این کالاها در بخش <i>«خریداران این محصول، کالاهای زیر را هم خریده‌اند»</i>، تکنیکِ ارزشمندِ <b>Cross-Selling</b> را با بالاترین نرخ موفقیت اجرا کرده "
    r"و سبد خریدِ مشتری را به طور ارگانیک و سیستماتیک بزرگ‌تر کند."
)
display_academic_text(insight_conclusion, "insight")
```
