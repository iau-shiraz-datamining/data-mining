<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: بهینه‌سازی و مقیاس‌پذیری الگوریتم Apriori در پلتفرم «آمازون» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    پلتفرم خرده‌فروشی «آمازون» روزانه صدها میلیون تراکنش خرید را پردازش می‌کند. اجرای الگوریتم Apriori ساده روی این حجم عظیم از داده‌ها، به دلیل پیمایش‌های مکرر کل پایگاه‌داده (Database Scans) و انفجار تعداد کاندیداها، پردازش را با بن‌بست مواجه می‌کند. برای حل این مشکل، مهندسان داده آمازون از استراتژی‌های بهینه‌سازی محاسباتی برای افزایش سرعت تحلیل سبد خرید استفاده می‌کنند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    معادله شتاب اجرای موازی (Speedup Ratio):<br>
    اگر کار محاسباتی بین P پردازنده یا گره توزیع شود، زمان اجرای الگوریتم طبق رابطه زیر کاهش می‌یابد:
  </p>

  <div style="text-align: center; margin: 15px 0;">
    $$\text{Speedup} = \frac{T_{\text{sequential}}}{T_{\text{parallel}}} = \frac{T_1}{\frac{T_1}{P} + T_{\text{overhead}}}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    تکنیك‌های ۴ گانه بهبود کارایی الگوریتم Apriori در مقیاس آمازون:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. هرس مبتنی بر هش (Hash-based Pruning):<br>
    با نگاشت کاندیداهای ۲-آیتمی به یک جدول هش (Hash Table)، کاندیداهایی که سطل (Bucket) مربوط به آن‌ها دارای مقدار کمتر از حد آستانه است، پیش از شمارش دقیق حذف می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. نمونه‌گیری و کاهش تراکنش‌ها (Sampling & Transaction Reduction):<br>
    به جای بررسی کل میلیاردها تراکنش، الگوریتم روی یک نمونه تصادفی نمایندگی‌کننده (Random Sample) اجرا می‌شود و تراکنش‌های فاقد آیتم‌های پرتکرار از پیمایش‌های بعدی حذف می‌گردند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. موازی‌سازی و پردازش توزیع‌شده (Parallelization):<br>
    توزیع بار شمارش کاندیداها بین گره‌های محاسباتی سرورهای ابری (AWS) با رویکرد MapReduce تا زمان اجرا به‌طور خطی کاهش یابد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. شمارش پویای آیتم‌ها (Dynamic Itemset Counting - DIC):<br>
    اضافه‌کردن کاندیداهای جدید در هر نقطه از پیمایش پایگاه‌داده بدون نیاز به انتظار برای پایان کامل یک دور پیمایش.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: آمازون با به‌کارگیری تکنیک‌های موازی‌سازی و نمونه‌گیری، زمان محاسبه مجموعه‌های پرتکرار را از چندین ساعت به چند ثانیه کاهش داده و پیشنهادات آنی لحظه خرید را ممکن می‌سازد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: کاهش چشمگیر زمان پردازش و مصرف حافظه رام با بهینه‌سازی ساختار داده و موازی‌سازی
    </div>
  </p>

</div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import time
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori
from IPython.display import display, HTML

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})
plt.rcParams['figure.dpi'] = 135
np.random.seed(42)

def display_academic_text(text, type="normal"):
    """Renders text in a strict academic format."""
    styles = {
        "main_title": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 24px; "
            "font-weight: bold; text-align: center; border-bottom: 2px solid #34495e; "
            "padding-bottom: 15px; margin: 30px 0;"
        ),
        "sub_title": (
            "font-family: Tahoma, sans-serif; color: #34495e; font-size: 19px; "
            "font-weight: bold; border-right: 5px solid #2980b9; padding-right: 15px; "
            "margin: 40px 0px 20px 0px; background-color: #ecf0f1; padding: 12px; "
            "border-radius: 4px;"
        ),
        "paragraph": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 16px; "
            "margin-bottom: 20px; line-height: 1.8; text-align: justify;"
        ),
        "insight": (
            "font-family: Tahoma, sans-serif; color: #16a085; font-size: 16px; "
            "font-weight: bold; background-color: #e8f8f5; padding: 20px; "
            "border-left: 5px solid #1abc9c; margin: 30px 0; line-height: 1.8; "
            "text-align: justify;"
        )
    }
    display(HTML(f"<div dir='rtl' style='text-align: right; {styles[type]}'>{text}</div>"))

def get_protected_css():
    return """
    <style>
    .academic-table-wrapper { 
        direction: ltr !important; overflow-x: auto !important; width: 100% !important; 
        background-color: #fff; border: 1px solid #bdc3c7; border-radius: 4px; 
        box-shadow: 0 3px 6px rgba(0,0,0,0.08); 
    }
    table.strict-tbl { 
        border-collapse: collapse !important; width: 100% !important; margin: 0 !important; 
        font-family: inherit !important; font-size: 14px !important; 
        table-layout: auto !important; border: none !important; 
    }
    table.strict-tbl thead th { 
        background-color: #2c3e50 !important; color: #ffffff !important; 
        padding: 12px 18px !important; text-align: center !important; 
        border: 1px solid #34495e !important; white-space: nowrap !important; font-weight: bold !important; 
    }
    table.strict-tbl tbody td { 
        padding: 10px 18px !important; text-align: center !important; 
        border: 1px solid #bdc3c7 !important; white-space: nowrap !important; color: #2c3e50 !important; 
    }
    table.strict-tbl tbody tr:nth-child(even) td { background-color: #f8f9fa !important; }
    table.strict-tbl tbody tr:hover td { background-color: #e8ecef !important; }
    </style>
    """

def display_side_by_side_datasets(dfs, titles):
    col_html_list = []
    for i, (df, title) in enumerate(zip(dfs, titles)):
        table_html = df.to_html(index=False, classes=['strict-tbl'], border=0, float_format=lambda x: f"{x:.2f}")
        col_html = f"""
        <div class="flex-column" style="flex: 1 1 0; min-width: 0; display: flex; flex-direction: column;">
            <div style="background-color: #2c3e50; color: white; padding: 12px; text-align: center; font-family: Tahoma, sans-serif; font-weight: bold; font-size: 15px; border-radius: 4px 4px 0 0;">
                {title}
            </div>
            <div class="academic-table-wrapper" style="margin-top: 0; border-top: none;">
                {table_html}
            </div>
        </div>
        """
        col_html_list.append(col_html)
    
    html = f"<div style='display: flex; flex-direction: row; justify-content: space-between; gap: 20px; width: 100%; max-width: 100%; margin: 20px 0 35px 0; box-sizing: border-box;'>{''.join(col_html_list)}</div>"
    display(HTML(get_protected_css() + html))
```

```python
display_academic_text(
    "بهینه‌سازی و مقیاس‌پذیری (Scalability) الگوریتم Apriori در پلتفرم «آمازون» 🚀", 
    "main_title"
)

display_academic_text(
    "پلتفرم خرده‌فروشی «آمازون» روزانه صدها میلیون تراکنش را پردازش می‌کند. اجرای الگوریتم Aprioriِ کلاسیک روی این حجم عظیم از داده‌ها، "
    "به دلیل پیمایش‌های مکرر در جداول و <b>انفجار حافظه رم در فاز تولید کاندیداها (Candidate Generation Explosion)</b> پردازش را با بن‌بست و Crash مواجه می‌کند. "
    "از این رو، استراتژی‌های بهینه‌سازی محاسباتی الزامی است.", 
    "paragraph"
)

display_academic_text("معادلهٔ شتاب اجرای موازی (Speedup Ratio) بر اساس قانون Amdahl:", "sub_title")
display_academic_text(
    r"$$\text{Speedup} = \frac{T_{\text{sequential}}}{T_{\text{parallel}}} = \frac{T_1}{\frac{T_1}{P} + T_{\text{overhead}}}$$", 
    "paragraph"
)

display_academic_text("تکنیک‌های بهبود کارایی در محیط‌های واقعی Big Data:", "sub_title")
display_academic_text(
    "۱. <b>تخصیص بهینه حافظه (Low-Memory Iterator Mode):</b> استفاده از تکرارگرها (Iterators) به جای لود کردن کُل ماتریس در RAM.<br>"
    "۲. <b>هرس مبتنی بر هش (Hash-based Pruning):</b> حذف سریع کاندیداها پیش از شمارشِ کاملِ آن‌ها.<br>"
    "۳. <b>نمونه‌گیری (Sampling):</b> اجرای الگوریتم روی درصدِ کُوچکی از داده‌ها به منظور رسیدن به تقریبِ ریاضی سریع.<br>"
    "۴. <b>شمارش پویای الگوها (Dynamic Itemset Counting - DIC):</b> اضافه کردن کاندیداها در حین پیمایشِ پایگاه‌داده به جای تولید بلوکی.", 
    "paragraph"
)


# ==========================================================
# 3. SECTION 2: GENERATING SYNTHETIC BIG DATA
# ==========================================================
display_academic_text("مرحله نخست: تولید دیتاستِ مقیاس‌یافته (Scaled-up Dataset)", "sub_title")

# Educational Step 1: Create a Scaled-up Benchmark Dataset
# We simulate 3,000 random Amazon baskets with varying lengths.
items = [f'Product_{i}' for i in range(1, 26)]
num_transactions = 3000

dataset = []
for _ in range(num_transactions):
    basket_size = np.random.randint(3, 9)
    # Sampling without replacement to avoid duplicate items in one basket
    basket = list(np.random.choice(items, size=basket_size, replace=False))
    dataset.append(basket)

df_raw_baskets = pd.DataFrame({
    'Order_ID': range(1, num_transactions + 1),
    'Basket_Items': [', '.join(basket) for basket in dataset]
})

# Educational Step 2: One-Hot Encoding
te = TransactionEncoder()
te_data = te.fit(dataset).transform(dataset)
df_transactions = pd.DataFrame(te_data, columns=te.columns_)

display_side_by_side_datasets(
    [df_raw_baskets.head(5).reset_index(drop=True)],
    [f"نمونه تراکنش‌های شبیه‌سازی‌شده آمازون (حاوی {num_transactions:,} سبد)"]
)


# ==========================================================
# 4. SECTION 3: BENCHMARKING APRIORI OPTIMIZATIONS
# ==========================================================
display_academic_text("مرحله دوم: بنچمارک استراتژی‌های بهینه‌سازی", "sub_title")

# ----------------------------------------------------
# Educational Step 3: Standard vs Low-Memory Benchmark
# 'low_memory=True' in mlxtend uses an iterator-based approach instead of 
# creating a massive NxM matrix in RAM. It trades a slight CPU overhead 
# to prevent RAM explosion.
# ----------------------------------------------------
support_thresholds = [0.10, 0.08, 0.05, 0.03]
execution_times_standard = []
execution_times_low_memory = []
frequent_counts = []

for supp in support_thresholds:
    # 1. Standard Apriori (High Memory Consumption - Matrix in RAM)
    t0 = time.perf_counter()
    res_std = apriori(
        df_transactions, 
        min_support=supp, 
        use_colnames=True, 
        low_memory=False
    )
    t1 = time.perf_counter()
    execution_times_standard.append(t1 - t0)
    
    # 2. Optimized Low-Memory Apriori (Iterator-based Generation)
    t2 = time.perf_counter()
    res_opt = apriori(
        df_transactions, 
        min_support=supp, 
        use_colnames=True, 
        low_memory=True
    )
    t3 = time.perf_counter()
    execution_times_low_memory.append(t3 - t2)
    
    frequent_counts.append(len(res_std))

# ----------------------------------------------------
# Educational Step 4: Transaction Reduction (Sampling) Benchmark
# Sampling reduces N (Total transactions). It heavily speeds up computation
# but introduces a margin of error (we might miss border-line itemsets).
# ----------------------------------------------------
sample_sizes = [0.25, 0.50, 0.75, 1.00]
sampling_times = []
sampling_itemsets_found = []
target_supp = 0.05

for frac in sample_sizes:
    sample_df = df_transactions.sample(frac=frac, random_state=42)
    
    t_start = time.perf_counter()
    res_sample = apriori(
        sample_df, 
        min_support=target_supp, 
        use_colnames=True, 
        low_memory=True
    )
    t_end = time.perf_counter()
    
    sampling_times.append(t_end - t_start)
    sampling_itemsets_found.append(len(res_sample))


# ==========================================================
# 5. SECTION 4: ADVANCED ACADEMIC VISUALIZATIONS (ACADEMIC BLUE)
# ==========================================================
display_academic_text("مصورسازی تحلیلی: تأثیر موازنهٔ منابع (Trade-offs) بر سرعت پردازش", "sub_title")

fig, axes = plt.subplots(1, 2, figsize=(18, 6.5), dpi=135)

# ----------------------------------------------------------
# Subplot 1: Execution Time vs Minimum Support (Standard vs Low-Memory)
# ----------------------------------------------------------
axes[0].plot(
    [f'{s*100:.0f}%' for s in support_thresholds],
    execution_times_standard,
    marker='o',
    color='#e74c3c', # Red
    linewidth=3.0,
    markersize=8,
    label='Standard Mode (High RAM)'
)
axes[0].plot(
    [f'{s*100:.0f}%' for s in support_thresholds],
    execution_times_low_memory,
    marker='s',
    color='#1abc9c', # Turquoise
    linewidth=3.0,
    linestyle='--',
    markersize=8,
    label='Low-Memory Mode (Iterators)'
)

axes[0].set_title('Figure 1: Exponential Growth of Execution Time vs Minimum Support', fontsize=14, fontweight='bold', pad=15)
axes[0].set_xlabel('Minimum Support Threshold (%)', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Execution Time (Seconds)', fontsize=13, fontweight='bold')
axes[0].grid(True, linestyle='--', alpha=0.5)
axes[0].legend(loc='upper left', frameon=True, fontsize=11, edgecolor='#bdc3c7')
axes[0].tick_params(axis='both', labelsize=12)

# Annotate points clearly
for i, txt in enumerate(execution_times_standard):
    axes[0].annotate(
        f'{txt:.3f}s', 
        (i, execution_times_standard[i]), 
        textcoords="offset points", 
        xytext=(0, 10), 
        ha='center', 
        fontweight='bold', 
        color='#c0392b',
        fontsize=11
    )

# ----------------------------------------------------------
# Subplot 2: Impact of Sampling Strategy (Dual Axis Combo Chart)
# ----------------------------------------------------------
color_bar = '#2980b9'  # Classic Blue (Time)
color_line = '#f39c12' # Orange (Itemsets Count)

ax2_left = axes[1]
ax2_right = ax2_left.twinx()  # Create a secondary Y-axis

bars = ax2_left.bar(
    [f'{int(f*100)}%' for f in sample_sizes],
    sampling_times,
    color=color_bar,
    edgecolor='#2c3e50',
    linewidth=2.0,
    width=0.45,
    label='Execution Time (Seconds)'
)

lines = ax2_right.plot(
    [f'{int(f*100)}%' for f in sample_sizes],
    sampling_itemsets_found,
    color=color_line,
    marker='D',
    linewidth=3.5,
    markersize=8,
    label='Frequent Itemsets Count'
)


ax2_left.set_title('Figure 2: Effect of Data Reduction (Sampling Trade-off)', fontsize=14, fontweight='bold', pad=15)
ax2_left.set_xlabel('Dataset Sample Ratio (%)', fontsize=13, fontweight='bold')
ax2_left.set_ylabel('Execution Time (Seconds)', color=color_bar, fontsize=13, fontweight='bold')
ax2_right.set_ylabel('Frequent Itemsets Discovered (Count)', color='#d35400', fontsize=13, fontweight='bold')
ax2_left.grid(axis='y', linestyle='--', alpha=0.5)
ax2_left.tick_params(axis='x', labelsize=12)
ax2_left.tick_params(axis='y', labelcolor=color_bar, labelsize=11)
ax2_right.tick_params(axis='y', labelcolor='#d35400', labelsize=11)

# Value annotations for the Bar Chart
for bar in bars:
    yval = bar.get_height()
    ax2_left.text(
        bar.get_x() + bar.get_width() / 2, 
        yval + (max(sampling_times) * 0.02), 
        f'{yval:.3f}s', 
        ha='center', 
        va='bottom', 
        fontweight='bold',
        fontsize=12,
        color=color_bar
    )

# Clean up axes
axes[0].spines['top'].set_visible(False)
axes[0].spines['right'].set_visible(False)
ax2_left.spines['top'].set_visible(False)
ax2_right.spines['top'].set_visible(False)
# Note: We keep the right spine for ax2_right because it serves as the secondary Y-axis.

# Merge Legends for Subplot 2
lines_1, labels_1 = ax2_left.get_legend_handles_labels()
lines_2, labels_2 = ax2_right.get_legend_handles_labels()
ax2_left.legend(lines_1 + lines_2, labels_1 + labels_2, loc='upper left', frameon=True, fontsize=11, edgecolor='#bdc3c7')

plt.subplots_adjust(wspace=0.35)
plt.show()

# ==========================================================
# 6. SECTION 5: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک در مقیاس‌پذیری کلان‌داده (Strategic Implication):</b><br>"
    r"طراحی الگوریتم در محیط‌های Big Data همیشه نیازمندِ مصالحه‌گری (Trade-offs) است.<br>"
    r"در گراف ۲ (Figure 2) که یک نمودار دوگانه (Dual-Axis) است، به زیبایی اثبات می‌شود که استراتژیِ <b>«نمونه‌گیری آماری (Sampling)»</b> یک موازنهٔ فوق‌العاده ایجاد می‌کند. "
    r"با کاهش سهم داده‌ها به ۵۰٪ (کاهش ارتفاع میله‌های آبی)، زمان اجرای الگوریتم تقریباً نصف می‌شود؛ درحالی‌که خط نارنجی نشان می‌دهد تعداد الگوهای کشف‌شده اُفتِ چشمگیری نداشته و دقت ریاضی همچنان پایدار مانده است.<br>"
    r"در پلتفرم‌های عظیمی نظیر آمازون، <b>یک تقریبِ ۹۵ درصدی که در ۱۰ دقیقه اجرا شود، همیشه به یک پردازشِ ۱۰۰ درصدی که ۳ روز سرورها را قفل کند، ارجحیت دارد!</b>"
)
display_academic_text(insight_conclusion, "insight")
```
