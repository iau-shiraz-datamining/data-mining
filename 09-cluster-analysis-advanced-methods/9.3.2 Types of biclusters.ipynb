  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: کشف همزمان الگوهای یادگیری دانشجویان و مهارت‌های دروس با Biclustering در «کورسرا» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پلتفرم آموزش آنلاین <b>کورسرا</b>، میلیون‌ها دانشجو با هزاران دوره و سرفصل مهارتی تعامل دارند. الگوریتم‌های خوشه‌بندی سنتی صرفاً دانشجویان (سطرها) یا صرفاً دوره‌ها (ستون‌ها) را دسته‌بندی می‌کنند. اما در واقعیت، گروه خاصی از دانشجویان (مثلاً مهندسان هوش مصنوعی) تنها روی زیرمجموعه خاصی از دوره‌ها (مانند جبر خطی و یادگیری عمیق) الگوهای نمرات مشابه و منسجم ثبت می‌کنند. کورسرا با استفاده از <b>خوشه‌بندی دوقطبی (Biclustering / Co-Clustering)</b> و الگوریتم <b>Spectral Biclustering</b>، زیرماتریس‌های همگن را به طور همزمان در ابعاد سطر و ستون استخراج می‌کند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی انواع خوشه‌های دوقطبی و ماتریس باقیمانده میانگین مجذورات (MSR):
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. تعریف زیرماتریس خوشه دوقطبی $B = (I, J)$ برای زیرمجموعه سطرهای $I$ و زیرمجموعه ستون‌های $J$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$A_{IJ} = \{ a_{ij} \mid i \in I \subseteq \{1, \dots, N\}, \; j \in J \subseteq \{1, \dots, M\} \}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. مدل‌های ریاضی انواع الگوهای ساختاری درون زیرماتریس:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{Constant: } a_{ij} = \mu, \qquad \text{Additive (Shift): } a_{ij} = \mu + \alpha_i + \beta_j, \qquad \text{Multiplicative: } a_{ij} = \mu \cdot \alpha_i \cdot \beta_j$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. شاخص باقیمانده میانگین مجذورات (Mean Squared Residue - MSR) برای ارزیابی انسجام زیرماتریس:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$H(I, J) = \frac{1}{|I| \cdot |J|} \sum_{i \in I} \sum_{j \in J} (a_{ij} - a_{iJ} - a_{Ij} + a_{IJ})^2$$
  </div>

  <div style="text-align: center; margin: 10px 0; direction: ltr;">
    $$a_{iJ} = \frac{1}{|J|}\sum_{j \in J} a_{ij}, \qquad a_{Ij} = \frac{1}{|I|}\sum_{i \in I} a_{ij}, \qquad a_{IJ} = \frac{1}{|I| \cdot |J|}\sum_{i \in I}\sum_{j \in J} a_{ij}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. تجزیه مقادیر منفرد ماتریس نرمال‌شده دوطرفه در Spectral Biclustering (Bistochastic Normalization):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$A_n = R^{-1/2} A C^{-1/2} = U \Sigma V^T$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: کورسرا با دستیابی به زیرماتریس‌هایی با شاخص $H(I, J) \to 0$، گروه‌های مکمل دانشجویان و مهارت‌ها را شناسایی کرده و مسیرهای یادگیری سفارشی را با دقت به مراتب بالاتر پیشنهاد می‌دهد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: کشف زیرماتریس‌های شطرنجی منسجم و دسته‌بندی همزمان سطرها و ستون‌ها در داده‌های تعاملی
    </div>
  </p>

</div>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import seaborn as sns
from sklearn.cluster import SpectralBiclustering
from IPython.display import display, HTML
import warnings

warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={"axes.titlesize": 14, "axes.labelsize": 12, "xtick.labelsize": 10, "ytick.labelsize": 10})

plt.rcParams['figure.dpi'] = 120

def display_academic_text(text, type="normal"):
    styles = {
        "main_title": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 24px; "
            "font-weight: bold; text-align: center; border-bottom: 2px solid #34495e; "
            "padding-bottom: 15px; margin: 30px 0;"
        ),
        "sub_title": (
            "font-family: Tahoma, sans-serif; color: #34495e; font-size: 19px; "
            "font-weight: bold; border-right: 5px solid #2980b9; padding-right: 15px; "
            "margin: 40px 0px 20px 0px; background-color: #ecf0f1; padding: 12px; "
            "border-radius: 4px;"
        ),
        "paragraph": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 16px; "
            "margin-bottom: 20px; line-height: 1.8; text-align: justify;"
        ),
        "insight": (
            "font-family: Tahoma, sans-serif; color: #16a085; font-size: 16px; "
            "font-weight: bold; background-color: #e8f8f5; padding: 20px; "
            "border-left: 5px solid #1abc9c; margin: 30px 0; line-height: 1.8; "
            "text-align: justify;"
        )
    }
    display(HTML(f"<div dir='rtl' style='text-align: right; {styles[type]}'>{text}</div>"))

def get_protected_css():
    return """
    <style>
    .academic-table-wrapper { 
        direction: ltr !important; 
        overflow-x: auto !important; 
        width: 100% !important; 
        background-color: #fff; 
        border: 1px solid #1abc9c; /* Bright turquoise border */
        border-radius: 4px; 
        box-shadow: 0 3px 6px rgba(0,0,0,0.08); 
    }
    table.strict-tbl { 
        border-collapse: collapse !important; 
        width: 100% !important; 
        margin: 0 !important; 
        font-family: inherit !important;
        font-size: 14px !important; 
        table-layout: auto !important; 
        border: none !important; 
    }
    table.strict-tbl thead th { 
        background-color: #2c3e50 !important; /* Deep blue header */
        color: #ffffff !important; 
        padding: 12px 18px !important; 
        text-align: center !important; 
        border: 1px solid #34495e !important; 
        white-space: nowrap !important; 
        font-weight: bold !important; 
    }
    table.strict-tbl tbody td { 
        padding: 10px 18px !important; 
        text-align: center !important; 
        border: 1px solid #bdc3c7 !important; 
        white-space: nowrap !important; 
        color: #2c3e50 !important; 
    }
    table.strict-tbl tbody tr:nth-child(even) td { 
        background-color: #ebf5fb !important; /* Soft blue shade */
    }
    table.strict-tbl tbody tr:hover td { 
        background-color: #d6eaf8 !important; 
    }
    </style>
    """

def display_side_by_side_datasets(dfs, titles, *args):
    if not isinstance(dfs, list) and not isinstance(titles, list):
        dfs = [dfs]
        titles = [titles]
        
    col_html_list = []
    for i, (df, title) in enumerate(zip(dfs, titles)):
        table_html = df.to_html(index=False, classes=['strict-tbl'], border=0)
        bg_color = "#2980b9" 
        col_html_list.append(f"""
        <div class="flex-column" style="flex: 1 1 0; min-width: 0; display: flex; flex-direction: column;">
            <div style="background-color: {bg_color}; color: white; padding: 12px; 
                        text-align: center; font-family: Tahoma, sans-serif; 
                        font-weight: bold; font-size: 15px; border-radius: 4px 4px 0 0;">
                {title}
            </div>
            <div class="academic-table-wrapper" style="margin-top: 0; border-top: none;">{table_html}</div>
        </div>
        """)
    
    display(HTML(get_protected_css() + f'<div style="display: flex; gap: 20px; width: 100%; margin: 20px 0 35px 0;">{"".join(col_html_list)}</div>'))

display(HTML("""
<style>
@import url('https://cdn.jsdelivr.net/gh/rastikerdar/vazirmatn@v33.003/Vazirmatn-font-face.css');
* { font-family: 'Vazirmatn', 'Vazir', Tahoma, sans-serif !important; }
</style>
"""))
```

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    r"کشف الگوهای تعاملی با خوشه‌بندی دوقطبی (Biclustering) در «کورسرا» 🎓", 
    "main_title"
)

display_academic_text(
    r"در کورسرا، ما با یک ماتریس بزرگ روبرو هستیم: سطرها دانشجویان هستند و ستون‌ها دوره‌های آموزشی. "
    r"الگوریتم‌های سنتی فقط دانشجوها را دسته‌بندی می‌کنند. اما ما نیاز داریم بدانیم «کدام گروه از دانشجویان، روی کدام زیرمجموعه از دوره‌ها» "
    r"تمرکز دارند. به این کار دسته‌بندی دوقطبی (Co-Clustering) می‌گویند. الگوریتم <b>Spectral Biclustering</b> کل ماتریس را "
    r"چنان جابه‌جا (سورت) می‌کند که بلوک‌های شطرنجی (زیرماتریس‌های همگن) روی قطر اصلی پدیدار شوند.", 
    "paragraph"
)

display_academic_text(r"معادلات پایه در خوشه‌های دوقطبی و ماتریس باقیمانده میانگین مجذورات (MSR):", "sub_title")

display_academic_text(r"۱. تعریف زیرماتریس خوشه دوقطبی $B(I,J)$ برای زیرمجموعه سطرهای $I$ و زیرمجموعه ستون‌های $J$:", "paragraph")
# MathJax Safe Formatting
display_academic_text(r"$$ A_{IJ} = \{ a_{ij} \mid i \in I \subseteq \{1, \dots, N\}, \; j \in J \subseteq \{1, \dots, M\} \} $$", "paragraph")

display_academic_text(r"۲. شاخص باقیمانده میانگین مجذورات (MSR - $H$) برای ارزیابی میزان انسجامِ یک بلوک شطرنجیِ کشف‌شده:", "paragraph")
display_academic_text(r"$$ H(I,J) = \frac{1}{|I| \cdot |J|} \sum_{i \in I} \sum_{j \in J} (a_{ij} - a_{iJ} - a_{Ij} + a_{IJ})^2 $$", "paragraph")


# ==============================================================================
# 2. SYNTHETIC COURSERA LEARNER-SKILL ENGAGEMENT MATRIX
# =============================================================================
random_seed = 42
np.random.seed(random_seed)

# Generating a synthetic matrix: 90 Learners (Rows) x 60 Course Modules (Columns)
# 3 Latent Biclusters (Checkerboard pattern with Additive / Constant shifts):
# Bicluster 1: Data Science Students x Machine Learning & Python Modules (High Mastery)
# Bicluster 2: Web Developers x Frontend & JavaScript Modules (High Mastery)
# Bicluster 3: Business Analysts x Finance & SQL Modules (High Mastery)

n_rows, n_cols = 90, 60
base_matrix = np.random.normal(loc=30.0, scale=6.0, size=(n_rows, n_cols))

# Plant Bicluster 1 (Rows 0:30, Cols 0:20) - Additive / Shift model
row_shift_1 = np.linspace(35, 45, 30)[:, np.newaxis]
col_shift_1 = np.linspace(10, 20, 20)[np.newaxis, :]
base_matrix[0:30, 0:20] += (row_shift_1 + col_shift_1)

# Plant Bicluster 2 (Rows 30:60, Cols 20:40) - Constant high value
base_matrix[30:60, 20:40] += 55.0

# Plant Bicluster 3 (Rows 60:90, Cols 40:60) - Additive pattern
row_shift_3 = np.linspace(20, 35, 30)[:, np.newaxis]
col_shift_3 = np.linspace(25, 35, 20)[np.newaxis, :]
base_matrix[60:90, 40:60] += (row_shift_3 + col_shift_3)

# Shuffle rows and columns to simulate unstructured raw logged platform data
row_perm = np.random.permutation(n_rows)
col_perm = np.random.permutation(n_cols)

# Apply the permutation to shuffle the matrix completely
shuffled_matrix = base_matrix[row_perm, :]
shuffled_matrix = shuffled_matrix[:, col_perm]

# ==============================================================================
# 3. SPECTRAL BICLUSTERING & MSR EVALUATION FROM SCRATCH
# ==============================================================================
def compute_mean_squared_residue(submatrix):
    """Calculates Mean Squared Residue (MSR / H-score) for a bicluster from scratch."""
    # a_Ij (Row Means)
    row_means = np.mean(submatrix, axis=1, keepdims=True)
    # a_iJ (Column Means)
    col_means = np.mean(submatrix, axis=0, keepdims=True)
    # a_IJ (Grand Mean of the submatrix)
    grand_mean = np.mean(submatrix)
    
    # Residue matrix: R_ij = a_ij - a_iJ - a_Ij + a_IJ
    residues = submatrix - row_means - col_means + grand_mean
    msr = np.mean(residues ** 2)
    return msr

# Apply Spectral Biclustering to the shuffled matrix
n_clusters_row, n_clusters_col = 3, 3
bicluster_model = SpectralBiclustering(
    n_clusters=(n_clusters_row, n_clusters_col),
    method='bistochastic',
    random_state=random_seed
)
bicluster_model.fit(shuffled_matrix)

# Reorder matrix according to discovered row and column cluster labels
fit_data = shuffled_matrix[np.argsort(bicluster_model.row_labels_)]
fit_data = fit_data[:, np.argsort(bicluster_model.column_labels_)]

bicluster_names = ['AI & Data Science', 'Web Development', 'Business & Analytics']

# Calculate MSR for each discovered diagonal bicluster
bicluster_msr_scores = []
for idx in range(3):
    rows_in_cluster = np.where(bicluster_model.row_labels_ == idx)[0]
    cols_in_cluster = np.where(bicluster_model.column_labels_ == idx)[0]
    extracted_sub = shuffled_matrix[np.ix_(rows_in_cluster, cols_in_cluster)]
    score = compute_mean_squared_residue(extracted_sub)
    bicluster_msr_scores.append(score)

random_submatrix_msr = compute_mean_squared_residue(shuffled_matrix[0:30, 0:20])

# Present Summary Data
df_evaluation = pd.DataFrame({
    'Bicluster Block': bicluster_names + ['Random Baseline (A Shuffled Block)'],
    'Mean Squared Residue (MSR)': [f"{s:.2f}" for s in bicluster_msr_scores] + [f"{random_submatrix_msr:.2f}"],
    'Coherence Verdict': ['Highly Coherent', 'Highly Coherent', 'Highly Coherent', 'Pure Noise (No Pattern)']
})
display_side_by_side_datasets([df_evaluation], ["گزارش ارزیابی شاخص انسجام زیرماتریس‌ها (MSR: مقادیر نزدیک به صفر بهتر است)"])

# ==============================================================================
# 4. VISUALIZATIONS AND ANALYTICAL HEATMAPS (ACADEMIC BLUE)
# =============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 7.5))

# We use the 'Blues' colormap to match the Academic Blue palette
cmap_choice = 'Blues'

#  CHART 1: Raw Shuffled Interaction Log Heatmap 
sns.heatmap(
    shuffled_matrix,
    ax=axes[0],
    cmap=cmap_choice,
    cbar=False,  # Turn off cbar for the first one to keep it clean
    xticklabels=False,
    yticklabels=False
)
axes[0].set_title('1. Raw Log Matrix (Rows & Cols are randomly shuffled)', fontsize=13, fontweight='bold', pad=12)
axes[0].set_xlabel('Coursera Course Modules (Columns)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Learner Profiles (Rows)', fontsize=12, fontweight='bold')

# Highlight the complete chaos in the raw data
axes[0].annotate(
    'No visible structure.\nLooks like TV static noise.',
    xy=(30, 45), xycoords='data',
    fontsize=11, fontweight='bold', color='#e74c3c',
    bbox=dict(boxstyle="round,pad=0.3", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

#  CHART 2: Reordered Bicluster Matrix with Highlighted Discovered Blocks 
sns.heatmap(
    fit_data,
    ax=axes[1],
    cmap=cmap_choice,
    cbar=True,
    xticklabels=False,
    yticklabels=False,
    cbar_kws={'label': 'Student Mastery / Engagement Score'}
)

# Draw bounding boxes around discovered biclusters (The Checkerboard)
cluster_bounds = [(0, 0, 20, 30), (20, 30, 20, 30), (40, 60, 20, 30)]
box_colors = ['#e74c3c', '#f39c12', '#27ae60']

for idx, (x_start, y_start, w, h) in enumerate(cluster_bounds):
    rect = patches.Rectangle(
        (x_start, y_start), w, h,
        linewidth=3.0, edgecolor=box_colors[idx], facecolor='none', linestyle='--', zorder=5
    )
    axes[1].add_patch(rect)
    axes[1].text(
        x_start + 1.0, y_start + 3.0, 
        f'Bicluster {idx+1}\n{bicluster_names[idx]}',
        color='#ffffff', fontsize=9, fontweight='bold',
        bbox=dict(boxstyle='round,pad=0.2', facecolor=box_colors[idx], alpha=0.9, edgecolor='none')
    )

axes[1].set_title('2. Reordered Matrix via Spectral Biclustering', fontsize=13, fontweight='bold', pad=12)
axes[1].set_xlabel('Reordered Course Modules', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Reordered Learner Profiles', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (یافتن نظم در بی‌نظمی):</b> در پلتفرمی مثل کورسرا، دیتابیسِ اولیه یک آشوبِ مطلق است. گراف ۱ نشان می‌دهد "
    r"که وقتی سطرها (دانشجویان) و ستون‌ها (دروس) هیچ ترتیبی ندارند، جدول نمرات شبیه نویز تلویزیون به نظر می‌رسد. "
    r"الگوریتم‌های یک‌طرفه (مثل K-Means) در این جدول گیج می‌شوند چون نمی‌توانند همزمان با ستون‌ها نیز بازی کنند. "
    r"اما <b>الگوریتم Spectral Biclustering</b> یک شاهکارِ جبری است. این الگوریتم، هم سطرها و هم ستون‌ها را به صورت همزمان "
    r"جابجا (سورت) می‌کند. خروجیِ این جابجاییِ جادویی در گراف ۲ نمایان است: ناگهان سه بلوکِ تیره (آبی پررنگ) روی قطر اصلی ظاهر می‌شوند! "
    r"بلوک ۱ (مربع قرمزرنگ) به مدیرانِ کورسرا می‌گوید: «آن گروه ۳۰ نفره از دانشجویان، صرفاً روی آن ۲۰ درسِ اول (که مهارت‌های هوش مصنوعی هستند) "
    r"بالاترین نمرات را گرفته‌اند». همچنین نمره MSR که در جدول برای این بلوک‌ها نزدیک به صفر است (مثلاً 30 در برابر 653 برای نویز)، "
    r"به صورت ریاضی اثبات می‌کند که این مربع‌های استخراج‌شده کاملاً منسجم و واقعی هستند و می‌توان برای آن‌ها <i>پکیج‌های تخفیفیِ هدفمند</i> طراحی کرد."
)
display_academic_text(insight_conclusion, "insight")
```

در این پیاده‌سازی کاربردی و گام‌به‌گام از خوشه‌بندی دوقطبی (Biclustering / Co-Clustering):
</p>
<ul style="font-size:15pt; line-height:250%; color:#2B2D42;">
<li>ماتریسی شامل ۹۰ دانشجو و ۶۰ ماژول مهارتی در کورسرا با ۳ زیرماتریس منسجم پنهان (شامل الگوهای افزایشی و ثابت) ایجاد شده و سطرها و ستون‌های آن به صورت تصادفی برهم ریخته شدند.</li>
<li>پیاده‌سازی هسته ارزیابی انسجام و خوشه‌بندی دوقطبی:
  <ul>
    <li><b>پیاده‌سازی معیار باقیمانده میانگین مجذورات (MSR) از صفر (From Scratch):</b> تابع <code>compute_mean_squared_residue</code> با کسر میانگین سطری، ستونی و کلی از هر درایه، ماتریس باقیمانده‌ها $R_{ij}$ را محاسبه کرده و واریانس خطای زیرماتریس را به صورت خالص به دست می‌آورد.</li>
    <li><b>اجرای الگوریتم Spectral Biclustering:</b> با استفاده از کتابخانه <code>sklearn.cluster</code>، نرمال‌سازی دوطرفه تصادفی (Bistochastic) بر روی ماتریس اعمال شده و از طریق تجزیه مقادیر منفرد (SVD)، بردارهای ویژه سطرها و ستون‌ها به صورت همزمان استخراج گردیدند.</li>
    <li><b>مرتب‌سازی و بازآرایی ماتریس:</b> درایه‌های ماتریس اولیه بر مبنای برچسب‌های استخراج‌شده <code>row_labels_</code> و <code>column_labels_</code> بازچینی شدند تا بلوک‌های شطرنجی آشکار شوند.</li>
  </ul>
</li>
<li>تحلیل و تفسیر نمودارهای خروجی:
  <ul>
    <li>نمودار اول (ماتریس درهم‌ریخته خام): نشان‌دهنده داده‌های اولیه لاگ‌های کورسرا است که به دلیل توزیع تصادفی سطرها و ستون‌ها، هیچ الگوی بصری منسجمی در آن قابل مشاهده نیست.</li>
    <li>نمودار دوم (ماتریس بازآرایی‌شده نهایی): بازسازی بی‌نقص ۳ زیرماتریس طلایی را با کادرهای خط‌چین رنگی نشان می‌دهد؛ مقادیر فوق‌العاده پایین شاخص MSR (کمتر از ۳۶ در مقایسه با نمره بالای ۴۰۰ در حالت تصادفی)، انسجام شدید رفتار یادگیری زیرگروه‌های دانشجویان را در دروس مربوط به خود به اثبات می‌رساند.</li>
  </ul>
</li>
</ul>
