  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: خوشه‌بندی طیفی و تفکیک الگوهای غیرخطی تعامل در «تیک‌تاک» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پلتفرم ویدیویی <b>تیک‌تاک</b>، الگوهای وایرال شدن محتوا و زنجیره بازنشر ویدیوها ساختارهای پیچیده، غیرمحدب و درهم‌تنیده‌ای (مانند حلقه‌های متحدالمرکز بازنشر یا منیفولدهای غیرخطی موج‌های ترند) ایجاد می‌کنند. الگوریتم‌های سنتی مبتنی بر مرکز نظیر K-Means به دلیل فرض کروی بودن و تحدب خوشه‌ها، در تفکیک این ساختارها کاملاً شکست می‌خورند. تیک‌تاک با بهره‌گیری از <b>Spectral Clustering</b>، داده‌ها را به عنوان گره‌های یک گراف شباهت مدل کرده و با استفاده از بردارهای ویژه <b>Graph Laplacian</b>، داده‌های پیچیده را به فضایی منتقل می‌کند که در آن خوشه‌ها به صورت خطی تفکیک‌پذیر می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی خوشه‌بندی طیفی و لاپلاسیان گراف:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. ماتریس شباهت و وابستگی گراف (Affinity Matrix $W$) با تابع هسته گوسی RBF:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$W_{ij} = \exp \left( -\frac{\| x_i - x_j \|_2^2}{2 \sigma^2} \right) \quad (\forall i \neq j), \qquad W_{ii} = 0$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. ماتریس درجه گراف (Degree Matrix $D$) و ماتریس لاپلاسیان متقارن نرمال‌شده ($L_{\text{sym}}$):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$D_{ii} = \sum_{j=1}^{N} W_{ij}, \qquad L_{\text{sym}} = D^{-1/2} (D - W) D^{-1/2} = I - D^{-1/2} W D^{-1/2}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. فرمول کمینه‌سازی برش نرمال‌شده گراف (Normalized Cut - NCut Objective):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{NCut}(A_1, \dots, A_K) = \frac{1}{2} \sum_{k=1}^{K} \frac{\text{cut}(A_k, \bar{A}_k)}{\text{vol}(A_k)} = \text{Tr}(Y^T L_{\text{sym}} Y) \quad \text{subject to } Y^T Y = I$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. نگاشت به فضای طیفی و نرمال‌سازی سطری بردارهای ویژه ($U \in \mathbb{R}^{N \times K}$):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$Y_{ij} = \frac{u_{ij}}{\sqrt{\sum_{k=1}^{K} u_{ik}^2}} \quad \implies \quad \| Y_{i, :} \|_2 = 1$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: تیک‌تاک با نگاشت داده‌های غیرخطی تعامل کاربران بر روی بردارهای ویژه لاپلاسیان، منیفولدهای درهم‌تنیده را به نقاط مجزا روی کره واحد تبدیل کرده و الگوریتم‌های پیشنهاد محتوا را بر مبنای جوامع واقعی ترندها بهینه‌سازی می‌کند.<br><br>

    </div>
  </p>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    r"تفکیک الگوهای غیرخطی تعامل در «تیک‌تاک» (Spectral Clustering & Graph Laplacian) 🎵", 
    "main_title"
)

display_academic_text(
    r"در شبکه‌هایی مثل تیک‌تاک، یک ترند به صورت کُروی پخش نمی‌شود؛ بلکه مثل امواج آب که در اثر پرتاب سنگ ایجاد می‌شوند، "
    r"به صورت حلقه‌های متحدالمرکز (پخش‌کنندگان اولیه $\rightarrow$ ری‌اکشن‌ها $\rightarrow$ تماشاگران عادی) گسترش می‌یابد. "
    r"الگوریتم K-Means در این داده‌های غیرمحدب (Non-Convex) صددرصد شکست می‌خورد. <b>خوشه‌بندی طیفی (Spectral Clustering)</b> "
    r"نقاط را به شکل یک گراف (شبکه) می‌بیند و با محاسبه ماتریس لاپلاسیان گراف، این حلقه‌های درهم‌تنیده را در فضای «طیفی» "
    r"از هم باز می‌کند تا به صورت خطی قابل تفکیک شوند.", 
    "paragraph"
)

display_academic_text(r"معادلات پایه در خوشه‌بندی طیفی و لاپلاسیان گراف (Ng, Jordan, Weiss):", "sub_title")

display_academic_text(r"۱. ماتریس شباهت گراف (Affinity Matrix $W$) با استفاده از هسته گوسی RBF:", "paragraph")
# MathJax Safe Formatting
display_academic_text(r"$$ W_{ij} = \exp\left(-\frac{\|x_i - x_j\|_2^2}{2\sigma^2}\right) \quad (\forall i \neq j), \qquad W_{ii} = 0 $$", "paragraph")

display_academic_text(r"۲. ماتریس لاپلاسیان متقارن نرمال‌شده ($L_{\text{sym}}$):", "paragraph")
display_academic_text(r"$$ D_{ii} = \sum_{j=1}^{N} W_{ij}, \qquad L_{\text{sym}} = D^{-1/2} (D - W) D^{-1/2} $$", "paragraph")

display_academic_text(r"۳. نگاشت به فضای طیفی با استفاده از بردارهای ویژه ($U$) و نرمال‌سازی سطری (کُره واحد):", "paragraph")
display_academic_text(r"$$ Y_{ij} = \frac{u_{ij}}{\sqrt{\sum_{k=1}^{K} u_{ik}^2}} \quad \implies \quad \| Y_{i, :} \|_2 = 1 $$", "paragraph")

# ==============================================================================
# 2. TIKTOK ENGAGEMENT MANIFOLD DATASET (NON-CONVEX CONCENTRIC RINGS)
# ==============================================================================
random_seed = 42
np.random.seed(random_seed)

# Simulating 400 TikTok creator interaction profiles in 2 concentric rings:
# Inner Ring: Core Content Creators (High engagement density, viral origin)
# Outer Ring: Secondary Consumer Wave (Echo chamber spread across broader network)

X_tiktok, y_true_manifold = make_circles(
    n_samples=400,
    factor=0.35, # Distance between rings
    noise=0.06,  # Noise in the data
    random_state=random_seed
)
n_samples = len(X_tiktok)

# ==============================================================================
# 3. SPECTRAL CLUSTERING ENGINE IMPLEMENTED FROM SCRATCH
# ==============================================================================
class SpectralClusteringFromScratch:
    """Normalized Spectral Clustering (Ng, Jordan, Weiss algorithm 2002) implemented from scratch."""
    
    def __init__(self, n_clusters=2, gamma=12.0):
        self.n_clusters = n_clusters
        self.gamma = gamma # Width of the Gaussian Kernel
        self.affinity_matrix_ = None
        self.laplacian_sym_ = None
        self.eigenvalues_ = None
        self.eigenvectors_ = None
        self.spectral_embedding_ = None
        self.labels_ = None

    def fit(self, X):
        n = X.shape[0]

        # Step 1: Compute pairwise RBF Affinity Matrix W
        # W_ij = exp(-gamma * ||x_i - x_j||^2)
        dists_sq = np.sum((X[:, np.newaxis, :] - X[np.newaxis, :, :]) ** 2, axis=2)
        W = np.exp(-self.gamma * dists_sq)
        np.fill_diagonal(W, 0.0)  # Zero diagonal (no self-loops in the graph)
        self.affinity_matrix_ = W

        # Step 2: Compute Degree Matrix D and Normalized Symmetric Laplacian L_sym
        degrees = np.sum(W, axis=1)
        d_inv_sqrt = 1.0 / np.sqrt(np.fmax(degrees, 1e-12))
        D_inv_sqrt = np.diag(d_inv_sqrt)

        # L_sym = I - D^(-1/2) W D^(-1/2)
        L_sym = np.eye(n) - np.dot(np.dot(D_inv_sqrt, W), D_inv_sqrt)
        self.laplacian_sym_ = L_sym

        # Step 3: Eigendecomposition of L_sym
        eigenvalues, eigenvectors = np.linalg.eigh(L_sym)
        self.eigenvalues_ = eigenvalues

        # Step 4: Select eigenvectors corresponding to K smallest eigenvalues
        k_indices = np.argsort(eigenvalues)[:self.n_clusters]
        U = eigenvectors[:, k_indices]
        self.eigenvectors_ = U

        # Step 5: Normalize rows to unit length: Y_ij = U_ij / sqrt(sum_k U_ik^2)
        # This projects all points onto the unit sphere!
        row_norms = np.linalg.norm(U, axis=1, keepdims=True)
        Y = U / np.fmax(row_norms, 1e-12)
        self.spectral_embedding_ = Y

        # Step 6: Apply K-Means on the normalized spectral embedding Y
        kmeans = KMeans(n_clusters=self.n_clusters, init='k-means++', n_init=10, random_state=random_seed)
        self.labels_ = kmeans.fit_predict(Y)

        return self

# Fit Custom Spectral Clustering
spectral_custom = SpectralClusteringFromScratch(n_clusters=2, gamma=15.0)
spectral_custom.fit(X_tiktok)
spectral_labels = spectral_custom.labels_

# Baseline: Standard K-Means directly on raw geometric coordinates
kmeans_baseline = KMeans(n_clusters=2, init='k-means++', n_init=10, random_state=random_seed)
kmeans_labels = kmeans_baseline.fit_predict(X_tiktok)


# Calculate and Present Accuracies in an Academic Table
accuracy_spectral = np.max([np.mean(spectral_labels == y_true_manifold), np.mean(spectral_labels == (1 - y_true_manifold))]) * 100
accuracy_kmeans = np.max([np.mean(kmeans_labels == y_true_manifold), np.mean(kmeans_labels == (1 - y_true_manifold))]) * 100

df_benchmark = pd.DataFrame([
    {"Clustering Algorithm": "Standard K-Means (Baseline)", "Topological Accuracy": f"{accuracy_kmeans:.1f}%", "Verdict": "Catastrophic Failure (Splits the rings)"},
    {"Clustering Algorithm": "Spectral Clustering (From Scratch)", "Topological Accuracy": f"{accuracy_spectral:.1f}%", "Verdict": "Perfect Geometric Recovery"}
])
display_side_by_side_datasets([df_benchmark], ["معیارسنجی عملکرد: K-Means در برابر لاپلاسیان گراف در داده‌های غیرمحدب"])

# ==============================================================================
# 4. VISUALIZATIONS AND ANALYTICAL CHARTS (ACADEMIC BLUE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

# Academic Blue Palette
palette = ['#2c3e50', '#2980b9']

#  CHART 1: Geometric Input Space (Standard K-Means Failure vs Spectral Success) 
# Subplot: Spectral Clustering Result on Non-Convex Rings
for k in range(2):
    mask = spectral_labels == k
    axes[0].scatter(
        X_tiktok[mask, 0], X_tiktok[mask, 1],
        s=45, color=palette[k], alpha=0.85, edgecolor='white', linewidth=0.5,
        label=f'Spectral Cluster {k+1} ({"Core Origin" if k==1 else "Viral Echo Wave"})'
    )

# Add decision boundary artifact line of failed K-Means for comparison
km_center1, km_center2 = kmeans_baseline.cluster_centers_
midpoint = (km_center1 + km_center2) / 2.0
axes[0].axline(midpoint, slope=-1.0, color='#e74c3c', linestyle=':', linewidth=2.5, label='Blind K-Means Linear Cut')

axes[0].set_title('1. TikTok Interaction Manifold: Successful Spectral Recovery', fontsize=13, fontweight='bold', pad=12)
axes[0].set_xlabel('Creator Influence Metric ($X_1$)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Viral Broadcast Latency ($X_2$)', fontsize=12, fontweight='bold')
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='lower left')
axes[0].grid(True, linestyle='--', alpha=0.6)

#  CHART 2: Spectral Eigenspace Transformation & Fiedler Vector Embedding 
# Plot the 2nd Eigenvector (Fiedler Vector v2) which achieves 1D linear separation
# The 1st eigenvector is constant, the 2nd contains the actual clustering information!
fiedler_vector = spectral_custom.spectral_embedding_[:, 1]
sample_indices = np.arange(n_samples)

axes[1].scatter(
    sample_indices, fiedler_vector,
    c=spectral_labels,
    cmap=plt.matplotlib.colors.ListedColormap(palette),
    s=55, alpha=0.85, edgecolor='black', linewidth=0.4,
    label='Transformed Eigenspace Samples'
)

# Threshold separation line in eigenspace (The K-Means cut in the new space)
axes[1].axhline(0.0, color='#e74c3c', linestyle='-.', linewidth=2.5, label='Optimal Spectral Hyperplane (Threshold = 0.0)')

axes[1].annotate(
    'Outer Ring Mapped to (+1.0)\nInner Ring Mapped to (-1.0)',
    xy=(n_samples // 2, 0.0),
    xytext=(n_samples // 2 - 80, 0.45),
    fontsize=10, fontweight='bold', color='#2c3e50',
    arrowprops=dict(arrowstyle="->", color="#c0392b", lw=2),
    bbox=dict(boxstyle="round,pad=0.3", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[1].set_title('2. Spectral Eigenspace (Fiedler Vector $v_2$): Perfect Linear Separation', fontsize=13, fontweight='bold', pad=12)
axes[1].set_xlabel('TikTok User Sample Index (Arbitrary Order)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Spectral Coordinate $Y$ (Eigenvector $v_2$)', fontsize=12, fontweight='bold')
axes[1].set_ylim(-1.3, 1.3)
axes[1].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='lower right')
axes[1].grid(True, linestyle='--', alpha=0.6)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک :</b> در گراف ۱، رفتار کاربران تیک‌تاک شبیه دو حلقهٔ تودرتو است (داده‌های غیرمحدب). "
    r"الگوریتم K-Means که یک الگوریتم خطی است، برای تفکیک آن‌ها یک خطِ صافِ مورب (خط‌چین قرمز) وسطِ نمودار می‌کشد و هر دو حلقه را به طرز فاجعه‌باری از وسط نصف می‌کند (دقت ۵۰٪). "
    r"اما <b>Spectral Clustering</b> داده‌ها را به شکل یک گرافِ پیوسته در نظر می‌گیرد. او با محاسبه <b>لاپلاسیانِ گراف</b> و استخراج بردارهای ویژه (Eigenvectors)، "
    r"این حلقه‌های درهم‌پیچیده را در یک بُعدِ جدید «باز می‌کند». "
    r"به گراف ۲ نگاه کنید! این گراف همان بُعدِ جدید (بردار فیدلر $v_2$) است. تمامِ نقاطِ متعلق به حلقهٔ بیرونی (سرمه‌ای) به عدد $+1.0$ منتقل شده‌اند "
    r"و تمامِ نقاطِ متعلق به حلقهٔ داخلی (آبی کلاسیک) به عدد $-1.0$ پرتاب شده‌اند! "
    r"در این فضای جدیدِ طیفی، یک خط‌کشِ ساده روی نقطهٔ صفر (خط‌چین قرمز) می‌تواند این دو جامعهٔ کاربری را با <b>دقت ۱۰۰٪</b> از هم جدا کند."
)
display_academic_text(insight_conclusion, "insight")
```
