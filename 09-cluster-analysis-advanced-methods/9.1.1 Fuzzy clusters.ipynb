    مثال عملی: مدل‌سازی پروفایل مهارت‌های چندگانه توسعه‌دهندگان با الگوریتم Fuzzy C-Means در «گیت‌هاب» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پلتفرم مدیریت کد <b>گیت‌هاب</b>، میلیون‌ها برنامه‌نویس در پروژه‌های منبع‌باز فعالیت می‌کنند. تخصص یک توسعه‌دهنده به ندرت به یک حوزه منفرد (صرفاً Backend یا Frontend) محدود می‌شود؛ یک مهندس نرم‌افزار ممکن است به طور همزمان ۵۰٪ در زمینه Backend، ۳۰٪ در زمینه DevOps و ۲۰٪ در زمینه Data Science فعالیت داشته باشد. در چنین شرایطی، الگوریتم‌های Hard Clustering کارایی ندارند. گیت‌هاب با استفاده از <b>Fuzzy Clustering</b> و به طور ویژه الگوریتم <b>Fuzzy C-Means (FCM)</b>، درجه عضویت پیوسته هر کاربر در خوشه‌های مهارتی را محاسبه کرده و امکان پیشنهاد هوشمند پروژه‌ها و هم‌تیمی‌ها را فراهم می‌سازد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی الگوریتم فازی سی-مینز (Fuzzy C-Means):
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. تابع هدف مجموع خطای وزن‌دار فازی (Fuzzy Objective Function) با ضریب فازی‌ساز $m > 1$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$J_m(U, C) = \sum_{i=1}^{N} \sum_{j=1}^{K} u_{ij}^m \| x_i - c_j \|_2^2$$
  </div>

  <div style="text-align: center; margin: 10px 0; direction: ltr;">
    $$\text{subject to: } \sum_{j=1}^{K} u_{ij} = 1 \quad (\forall i \in \{1,\dots,N\}), \qquad 0 \le u_{ij} \le 1$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. محاسبه ماتریس درجه عضویت فازی ($u_{ij}$) بر مبنای نسبت معکوس فواصل اقلیدسی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$u_{ij} = \frac{1}{\sum_{k=1}^{K} \left( \frac{\| x_i - c_j \|_2}{\| x_i - c_k \|_2} \right)^{\frac{2}{m-1}}}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. بازتولید بردار مراکز خوشه‌های فازی ($c_j$) بر اساس میانگین وزن‌دار توان فازی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$c_j = \frac{\sum_{i=1}^{N} u_{ij}^m x_i}{\sum_{i=1}^{N} u_{ij}^m}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. محاسبه شاخص ضریب تفکیک فازی (Partition Coefficient - PC) برای سنجش وضوح خوشه‌ها:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{PC} = \frac{1}{N} \sum_{i=1}^{N} \sum_{j=1}^{K} u_{ij}^2, \qquad \frac{1}{K} \le \text{PC} \le 1$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: گیت‌هاب با استخراج ماتریس $U$، توسعه‌دهندگان چندمهارته (Full-Stack) را با دقت بالا شناسایی کرده و پروژه‌های بین‌رشته‌ای را بر اساس توزیع احتمالاتی مهارت‌هایشان توصیه می‌کند.<br><br>

    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    r"مدل‌سازی پروفایل مهارت‌های چندگانه در «گیت‌هاب» (Fuzzy C-Means From Scratch) 💻", 
    "main_title"
)

display_academic_text(
    r"در دنیای واقعی، توسعه‌دهندگان فقط یک مهارت ندارند. یک نفر ممکن است هم در Backend فعال باشد هم در Data Science. "
    r"الگوریتم‌های قطعی (Hard Clustering) مثل K-Means این شخص را فقط به یک گروه پرتاب می‌کنند و مهارت دومش نادیده گرفته می‌شود. "
    r"ما از الگوریتم <b>Fuzzy C-Means (FCM)</b> استفاده می‌کنیم. در این رویکرد، مرزها نرم هستند و خروجیِ مدل یک "
    r"«ماتریس عضویت» است که درصد مهارت کاربر را در حوزه‌های مختلف به صورت یک عدد اعشاری (بین صفر و یک) گزارش می‌دهد.", 
    "paragraph"
)

display_academic_text(r"معادلات پایه الگوریتم فازی سی-مینز (FCM):", "sub_title")

display_academic_text(r"۱. تابع هدف مجموع خطای وزن‌دار فازی (Fuzzy Objective Function) با ضریب فازی‌ساز $m > 1$:", "paragraph")
# MathJax Safe Formatting
display_academic_text(r"$$ J_m(U, C) = \sum_{i=1}^{N} \sum_{j=1}^{K} u_{ij}^m \| x_i - c_j \|_2^2 \quad \text{subject to:} \quad \sum_{j=1}^{K} u_{ij} = 1 $$", "paragraph")

display_academic_text(r"۲. محاسبه ماتریس درجه عضویت فازی ($u_{ij}$) بر مبنای نسبت معکوس فواصل اقلیدسی:", "paragraph")
display_academic_text(r"$$ u_{ij} = \frac{1}{\sum_{k=1}^{K} \left( \frac{\| x_i - c_j \|_2}{\| x_i - c_k \|_2} \right)^{\frac{2}{m-1}}} $$", "paragraph")

display_academic_text(r"۳. بازتولید بردار مراکز خوشه‌های فازی ($c_j$) بر اساس میانگین وزن‌دار توان فازی:", "paragraph")
display_academic_text(r"$$ c_j = \frac{\sum_{i=1}^{N} u_{ij}^m x_i}{\sum_{i=1}^{N} u_{ij}^m} $$", "paragraph")

# ==============================================================================
# 2. GITHUB DEVELOPER ACTIVITY DATASET GENERATION
# ==============================================================================
# Set random seed for reproducibility
random_seed = 42
np.random.seed(random_seed)

# Simulating 360 GitHub developers across 3 primary specializations:
# Cluster 0: Backend & Systems (Go, Rust, Distributed Systems)
# Cluster 1: Frontend & UI/UX (React, TypeScript, CSS Architecture)
# Cluster 2: Data Science & AI (Python, PyTorch, Data Pipelines)

X_raw, _ = make_blobs(
    n_samples=360,
    centers=[[-2.2, -1.8], [2.0, 2.0], [2.2, -1.8]],
    cluster_std=[0.75, 0.85, 0.70],
    n_features=2,
    random_state=random_seed
)

# Standardize features
scaler = StandardScaler()
X_devs = scaler.fit_transform(X_raw)
n_samples = len(X_devs)

# ==============================================================================
# 3. FUZZY C-MEANS (FCM) CLUSTERING IMPLEMENTED FROM SCRATCH
# ==============================================================================
class FuzzyCMeansFromScratch:
    """Fuzzy C-Means (FCM) clustering algorithm implemented completely from scratch."""
    
    def __init__(self, n_clusters=3, m=2.0, max_iter=150, tol=1e-5):
        self.n_clusters = n_clusters
        self.m = m  # Fuzzifier exponent (m > 1). Standard choice is 2.0.
        self.max_iter = max_iter
        self.tol = tol
        self.centroids = None
        self.u_matrix = None
        self.loss_history = []

    def _init_membership_matrix(self, n_samples):
        """Initializes random fuzzy membership matrix where each row sums to 1.0 (Dirichlet distribution)."""
        random_weights = np.random.dirichlet(np.ones(self.n_clusters), size=n_samples)
        return random_weights

    def fit(self, X):
        """Fits Fuzzy C-Means iteratively by alternating optimization of U and C."""
        n, d = X.shape
        self.u_matrix = self._init_membership_matrix(n)

        for iteration in range(self.max_iter):
            u_old = self.u_matrix.copy()

            # Step 1: Update cluster centroids c_j = sum(u_ij^m * x_i) / sum(u_ij^m)
            u_power = self.u_matrix ** self.m
            self.centroids = np.dot(u_power.T, X) / np.sum(u_power.T, axis=1, keepdims=True)

            # Step 2: Compute pairwise Euclidean distance matrix d_ij = ||x_i - c_j||
            dists = np.linalg.norm(X[:, np.newaxis, :] - self.centroids[np.newaxis, :, :], axis=2)
            dists = np.fmax(dists, 1e-10)  # Prevent divide-by-zero for points landing exactly on a centroid

            # Step 3: Compute current fuzzy objective loss J_m
            current_loss = np.sum((self.u_matrix ** self.m) * (dists ** 2))
            self.loss_history.append(current_loss)

            # Step 4: Update fuzzy membership matrix u_ij
            power_factor = 2.0 / (self.m - 1.0)
            inv_dists = 1.0 / dists
            inv_dists_powered = inv_dists ** power_factor
            sum_inv_dists = np.sum(inv_dists_powered, axis=1, keepdims=True)
            self.u_matrix = inv_dists_powered / sum_inv_dists

            # Step 5: Check convergence threshold
            shift = np.linalg.norm(self.u_matrix - u_old)
            if shift < self.tol:
                break

        return self

# Fit FCM model
k_clusters = 3
fcm = FuzzyCMeansFromScratch(n_clusters=k_clusters, m=2.0, max_iter=100)
fcm.fit(X_devs)

# Hard cluster assignments for reference (Maximum membership)
hard_labels = np.argmax(fcm.u_matrix, axis=1)

# Calculate Partition Coefficient (PC) - A metric for fuzziness vs crispness
partition_coefficient = np.mean(np.sum(fcm.u_matrix ** 2, axis=1))

# Identify hybrid full-stack developers (high entropy / maximum ambiguity)
# Uncertainty is quantified as 1 - max(u_i)
uncertainty = 1.0 - np.max(fcm.u_matrix, axis=1)
boundary_dev_indices = np.argsort(uncertainty)[-6:]

# Present Optimization Result Table
df_convergence = pd.DataFrame([
    {"Fuzzy Iteration Step": "Iteration 1 (Start)", "Fuzzy Loss Function ($J_m$)": f"{fcm.loss_history[0]:.2f}"},
    {"Fuzzy Iteration Step": f"Iteration {len(fcm.loss_history)} (Converged)", "Fuzzy Loss Function ($J_m$)": f"{fcm.loss_history[-1]:.2f}"},
    {"Fuzzy Iteration Step": "Partition Coefficient (PC)", "Fuzzy Loss Function ($J_m$)": f"{partition_coefficient:.4f} (1.0 = Rigid, 0.33 = Max Fuzzy)"}
])
display_side_by_side_datasets([df_convergence], ["گزارش همگرایی تابع هدف فازی و شاخص وضوح خوشه‌ها"])

# ==============================================================================
# 4. VISUALIZATIONS AND ANALYTICAL CHARTS (ACADEMIC BLUE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

cluster_colors = ['#1abc9c', '#3498db', '#2c3e50'] # Teal (Backend), Light Blue (Frontend), Navy (Data)
domain_titles = {0: 'Backend & Systems', 1: 'Frontend & UI', 2: 'Data Science & AI'}

#  CHART 1: Continuous RGB Blended Fuzzy Membership Scatter Plot 
# Map 3 membership degrees directly to RGB/HEX colors for visual blending
# We create an RGB array where each channel represents a cluster membership
rgb_colors = np.zeros((n_samples, 3))

# We'll map to specific RGB representations of our Academic Palette to mix them
# Navy (Data): [44, 62, 80]/255
# Light Blue (Frontend): [52, 152, 219]/255
# Teal (Backend): [26, 188, 156]/255

color_vectors = np.array([
    [26/255.0, 188/255.0, 156/255.0],  # Backend
    [52/255.0, 152/255.0, 219/255.0],  # Frontend
    [44/255.0, 62/255.0, 80/255.0]     # Data Science
])

# Multiply membership matrix (N x 3) by color vectors (3 x 3) to get blended colors (N x 3)
blended_colors = np.dot(fcm.u_matrix, color_vectors)

# Scatter points with continuous fuzzy membership blending
axes[0].scatter(
    X_devs[:, 0], X_devs[:, 1],
    c=blended_colors,
    s=45 + (uncertainty * 200), # Higher uncertainty = Larger dot
    edgecolor='black',
    linewidth=0.5,
    alpha=0.9,
    label='Developers (Color = Soft Membership Mix)'
)

# Plot Centroids
centroids = fcm.centroids
axes[0].scatter(
    centroids[:, 0], centroids[:, 1],
    s=300, color='#e74c3c', marker='X', edgecolor='white', linewidth=2, zorder=10,
    label='FCM Fuzzy Centroids ($c_j$)'
)

# Annotate centroids
for j, c in enumerate(centroids):
    axes[0].annotate(
        f'{domain_titles[j]} Hub',
        xy=(c[0], c[1]),
        xytext=(c[0] + 0.15, c[1] + 0.15),
        fontsize=9, fontweight='bold', color='#c0392b',
        bbox=dict(boxstyle="round,pad=0.25", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
    )

# Highlight high-uncertainty multi-stack developer
top_hybrid_idx = boundary_dev_indices[-1]
axes[0].scatter(
    X_devs[top_hybrid_idx, 0], X_devs[top_hybrid_idx, 1],
    s=400, facecolor='none', edgecolor='#e74c3c', linewidth=2.5, linestyle=':'
)
axes[0].annotate(
    'Hybrid Full-Stack Dev\n(Perfectly Balanced Mixture)',
    xy=(X_devs[top_hybrid_idx, 0], X_devs[top_hybrid_idx, 1]),
    xytext=(X_devs[top_hybrid_idx, 0] - 2.0, X_devs[top_hybrid_idx, 1] + 0.6),
    fontsize=10, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2),
    bbox=dict(boxstyle="round,pad=0.3", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title(f'1. GitHub Developer Profiling in Fuzzy Continuous Space', fontsize=13, fontweight='bold', pad=12)
axes[0].set_xlabel('Standardized Backend / Systems Activity', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Standardized Frontend & ML Activity', fontsize=12, fontweight='bold')
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='upper left')
axes[0].grid(True, linestyle='--', alpha=0.6)


#  CHART 2: Fuzzy Membership Distribution (u_ij) for Selected Boundary Developers 
dev_labels = [f'Dev #{idx}' for idx in boundary_dev_indices]
df_memberships = pd.DataFrame(
    fcm.u_matrix[boundary_dev_indices],
    columns=[f'Cluster {j+1}: {domain_titles[j]}' for j in range(k_clusters)],
    index=dev_labels
)

bottom_val = np.zeros(len(boundary_dev_indices))
x_positions = np.arange(len(boundary_dev_indices))

# Academic Blue palette for Stacked Bar Chart
for j in range(k_clusters):
    col_name = df_memberships.columns[j]
    axes[1].bar(
        x_positions,
        df_memberships[col_name],
        bottom=bottom_val,
        width=0.55,
        label=col_name,
        color=cluster_colors[j],
        edgecolor='black',
        alpha=0.9
    )
    bottom_val += df_memberships[col_name].values

axes[1].set_title(f'2. Soft Membership Matrix ($u_{{ij}}$) for Multi-Disciplinary Devs', fontsize=13, fontweight='bold', pad=12)
axes[1].set_xlabel('Selected Boundary Developers (Highest Multi-Skill Overlap)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Degree of Membership (Sum = 1.0)', fontsize=12, fontweight='bold')
axes[1].set_xticks(x_positions)
axes[1].set_xticklabels(dev_labels, fontsize=10, fontweight='bold', rotation=15)
axes[1].set_ylim(0, 1.05)
axes[1].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='upper right')
axes[1].grid(axis='y', linestyle='--', alpha=0.6)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 6: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک :</b> مدل‌های قطعی مثل K-Means باعث می‌شوند ما بخشِ پنهانِ مهارتِ انسان‌ها را "
    r"از دست بدهیم. اما در گراف ۱، ما از <b>FCM (سی‌مینز فازی)</b> استفاده کردیم و ماتریس عضویت (U) را مستقیماً به رنگ‌های RGB مپ کردیم. "
    r"نقاطی که به مراکز نزدیک‌اند، رنگِ خالص دارند (تخصص قطعی). اما هرچه به سمتِ مرکز نقشه می‌آییم، رنگ‌ها با هم ترکیب می‌شوند. "
    r"برای مثال، برنامه‌نویسِ شماره ۱۶۱ (کادر قرمز) دقیقاً در مرزِ مشترک ایستاده است و رنگِ آن یک آبیِ کدر (ترکیب هر سه تخصص) است. "
    r"گراف ۲ دقیقاً کالبدشکافیِ مغزِ الگوریتم است. برای Dev #161 (ستون آخر)، ماشین می‌گوید: "
    r"<i>«این شخص ۴۰٪ در حوزه هوش‌مصنوعی (سرمه‌ای)، ۴۰٪ در حوزه فرانت‌اند (آبی روشن) و ۲۰٪ در حوزه بک‌اند (فیروزه‌ای) فعالیت دارد»</i>. "
    r"با استفاده از این خروجیِ پیوسته و منعطف، گیت‌هاب می‌تواند پروژه‌های سنگینِ Full-Stack را بدون هدر دادن منابع، به مناسب‌ترین افراد پیشنهاد دهد."
)
display_academic_text(insight_conclusion, "insight")
```
