<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: محاسبه فاصله و مسیریابی سفرهای «اوبر» (Uber) 📏
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    الگوریتم‌های مسیریابی «اوبر» (Uber) برای تخمین فاصله بین مسافر و راننده، از مختصات جغرافیایی (طول و عرض جغرافیایی) استفاده می‌کنند. بسته به نوع وسیله نقلیه (تاکسی، پهپاد یا پیک پیاده)، سیستم از حالت‌های مختلف <b>فاصله مینکوفسکی (Minkowski Distance)</b> بهره می‌برد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام اول: فرمول کلی فاصله مینکوفسکی</b><br>
    فاصله مینکوفسکی یک فرمول جامع برای محاسبه عدم تشابه و فاصله در فضاهای چندبعدی عددی است:
  </p>
  
  <div style="text-align: center; margin: 15px 0;">
    $$d(x, y) = \left( \sum_{i=1}^{n} |x_i - y_i|^p \right)^{\frac{1}{p}}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در این فرمول، پارامتر $p$ نوع فاصله را تعیین می‌کند:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۱. حالت $p = 1$ — فاصله منهتن (Manhattan Distance / City Block):</b><br>
    حرکت در خیابان‌های شطرنجی شهر (مثل نیویورک) که نمی‌توان از روی ساختمان‌ها پرید! خودرو مجبور است فقط در جهت افقی و عمودی حرکت کند:<br>
    <div style="text-align: center; margin: 8px 0;">
      $$d_{\text{Manhattan}}(x, y) = |x_1 - x_2| + |y_1 - y_2|$$
    </div>
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۲. حالت $p = 2$ — فاصله اقلیدسی (Euclidean Distance):</b><br>
    مستقیم‌ترین مسیر بین دو نقطه ("As the crow flies" یا مانند پرواز کلاغ). این فاصله برای پهپادهای تحویل کالا (Uber Eats Drones) یا خطوط هوایی استفاده می‌شود:<br>
    <div style="text-align: center; margin: 8px 0;">
      $$d_{\text{Euclidean}}(x, y) = \sqrt{(x_1 - x_2)^2 + (y_1 - y_2)^2}$$
    </div>
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام دوم: مثال عددی ساده</b><br>
    فرض کنید موقعیت راننده اوبر $(1, 2)$ و موقعیت مسافر $(4, 6)$ باشد:<br>
    &nbsp;&nbsp;&nbsp;• <b>فاصله منهتن ($p=1$):</b> $|1 - 4| + |2 - 6| = 3 + 4 = 7$ کیلومتر.<br>
    &nbsp;&nbsp;&nbsp;• <b>فاصله اقلیدسی ($p=2$):</b> $\sqrt{(1 - 4)^2 + (2 - 6)^2} = \sqrt{9 + 16} = 5$ کیلومتر.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> فاصله منهتن همیشه بزرگ‌تر یا مساوی فاصله اقلیدسی است. اوبر برای خودروها از $p=1$ (مسیر خیابانی) و برای تخمین پهپادی از $p=2$ استفاده می‌کند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: یکپارچه‌سازی معیارهای فاصله در یک فرمول ریاضی با تنظیم پارامتر p
    </div>
  </p>

</div>
