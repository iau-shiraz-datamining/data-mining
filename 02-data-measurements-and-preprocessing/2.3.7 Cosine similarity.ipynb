  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: سیستم پیشنهاد مقاله و تحلیل متون در «مدیوم» (Medium) 📐
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    پلتفرم «مدیوم» (Medium) می‌خواهد بر اساس مقاله‌ای که کاربر در حال خواندن آن است، مقالات مشابه را پیشنهاد دهد. هر مقاله به یک <b>بردار از تکرار کلمات (Word Count / TF-IDF Vector)</b> تبدیل می‌شود. اما یک چالش بزرگ وجود دارد: یک مقاله ممکن است بسیار کوتاه (۲۰۰ کلمه) و مقاله دیگر بسیار طولانی (۲۰۰۰ کلمه) باشد، در حالی که هر دو درباره یک موضوع (مثلاً هوش مصنوعی) نوشته شده‌اند!
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام اول: چرا فاصله‌های اقلیدسی یا فاصله مینکوفسکی در اینجا شکست می‌خورند؟</b><br>
    اگر از فاصله اقلیدسی استفاده کنیم، طولانی بودن مقاله دوم باعث می‌شود تعداد کلمات آن بسیار زیاد شود و فاصله اقلیدسی عدد بزرگی نشان دهد! یعنی سیستم به‌اشتباه فکر می‌کند این دو مقاله هیچ ربطی به هم ندارند، فقط چون طول آن‌ها متفاوت است.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام دوم: راه‌حل — سنجش زاویه به‌جای اندازه‌گیری طول (شباهت کسینوسی)</b><br>
    شباهت کسینوسی <b>طول بردارها (اندازه مقاله) را کاملاً نادیده می‌گیرد</b> و فقط <b>زاویه بین بردارها (جهت موضوعی)</b> را می‌سنجد. فرمول ریاضی آن ضرب داخلی دو بردار تقسیم بر حاصل‌ضرب اندازه آن‌هاست:
  </p>

  <div style="text-align: center; margin: 15px 0;">
    $$\text{Cosine Similarity}(A, B) = \frac{A \cdot B}{\|A\| \|B\|} = \frac{\sum_{i=1}^{n} A_i B_i}{\sqrt{\sum_{i=1}^{n} A_i^2} \sqrt{\sum_{i=1}^{n} B_i^2}}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام سوم: تفسیر بازه عددی</b><br>
    کسینوس زاویه بین دو بردار همیشه عددی در بازه <b>۱- تا ۱+</b> است (در تحلیل متون چون فراوانی کلمات منفی نمی‌شود، بازه بین <b>۰ تا ۱</b> خواهد بود):<br>
    &nbsp;&nbsp;&nbsp;• <b>زاویه ۰ درجه ($\cos(0^\circ) = 1$):</b> دو بردار کاملاً هم‌جهت هستند (شباهت ۱۰۰٪ کامل موضوعی).<br>
    &nbsp;&nbsp;&nbsp;• <b>زاویه ۹۰ درجه ($\cos(90^\circ) = 0$):</b> دو بردار عمود بر هم هستند و هیچ کلمه یا موضوع مشترکی ندارند.<br>
    &nbsp;&nbsp;&nbsp;• <b>زاویه ۱۸۰ درجه ($\cos(180^\circ) = -1$):</b> دو بردار کاملاً در جهت عکس یکدیگر قرار دارند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> مقاله کوتاه ۲۰۰ کلمه‌ای و مقاله طولانی ۲۰۰۰ کلمه‌ای درباره هوش مصنوعی زاویه بسیار کوچکی با هم می‌سازند و شباهت کسینوسی آن‌ها نزدیک به <b>۱.۰۰</b> می‌شود؛ بنابراین سیستم مدیوم آن‌ها را به‌درستی پیشنهادی عالی برای یکدیگر تشخیص می‌دهد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: سنجش شباهت موضوعی متون مستقل از طول و بزرگی بردارها
    </div>
  </p>

</div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})

plt.rcParams['figure.dpi'] = 135

np.random.seed(42)
```

```python
display_academic_text(
    "معیارهای شباهت در تحلیل متون: خنثی‌سازی طول سند با شباهت کسینوسی (Cosine Similarity) 📝", 
    "main_title"
)

display_academic_text(
    "در کاوش داده‌های متنی (Text Mining)، اسناد خام پیش از ورود به الگوریتم، به <i>بردارهایی از فراوانی کلمات (Bag of Words)</i> تبدیل می‌شوند. "
    "استفاده از فواصل مکان‌محور (نظیر اقلیدسی) در این فضای برداریِ خلوت (Sparse)، منجر به طبقه‌بندی اشتباه می‌گردد؛ زیرا یک مقالهٔ طولانی "
    "فاصله‌ای نجومی با یک مقالهٔ کوتاه اما <i>هم‌موضوع</i> پیدا خواهد کرد. معیار <b>شباهت کسینوسی</b> با تمرکز انحصاری بر زاویه میان بردارها و "
    "نادیده گرفتن <b>طول یا بزرگی (Magnitude)</b> آن‌ها، این نقص محاسباتی را در سیستم‌های بازیابی اطلاعات جبران می‌نماید.", 
    "paragraph"
)

display_academic_text(
    "معادله شباهت کسینوسی (نسبت ضرب داخلی به حاصل‌ضرب اندازه بردارها):", 
    "sub_title"
)

# Utilizing Raw Strings (r"") for exact LaTeX rendering in Jupyter
display_academic_text(
    r"$$Cosine(A, B) = \cos(\theta) = \frac{A \cdot B}{\|A\| \|B\|} = \frac{\sum_{i=1}^{n} A_i B_i}{\sqrt{\sum_{i=1}^{n} A_i^2} \sqrt{\sum_{i=1}^{n} B_i^2}}$$", 
    "paragraph"
)

# ==========================================================
# SECTION 2: DATA VECTORIZATION AND SIMILARITY COMPUTATION
# ==========================================================
display_academic_text(
    "مرحله نخست: استخراج بردارهای فراوانی و محاسبه ماتریس شباهت", 
    "sub_title"
)

display_academic_text(
    "در این فاز، متون خام پردازش شده و به ماتریس ویژگی‌های عددی تبدیل می‌گردند. سپس الگوریتم شباهت کسینوسی بر روی این ماتریس اعمال می‌شود "
    "تا <i>زاویهٔ موضوعیِ مقالات</i> نسبت به یکدیگر استخراج شده و در قالب یک ماتریس متقارن رندر گردد.", 
    "paragraph"
)

# Step 1: Initialize the raw text corpus representing Medium articles
articles = [
    "Python machine learning artificial intelligence AI code",                                 
    "Deep machine learning algorithms with Python code and artificial intelligence AI models",  
    "Italian pizza recipe baking cheese and tomato dough"                                       
]
doc_names = ["Doc_1 (Short AI)", "Doc_2 (Long AI)", "Doc_3 (Pizza Recipe)"]

# Step 2: Convert the text corpus into a Bag of Words (BoW) numerical matrix
# CountVectorizer tokenizes the strings and counts the frequency of each distinct word
vectorizer = CountVectorizer()
word_count_matrix_sparse = vectorizer.fit_transform(articles)

# Convert the sparse matrix into a dense pandas DataFrame for visualization
df_vectors = pd.DataFrame(
    word_count_matrix_sparse.toarray(),
    index=doc_names,
    columns=vectorizer.get_feature_names_out()
)
df_vectors.index.name = 'Document'

# Step 3: Compute the Pairwise Cosine Similarity Matrix
# Calculates dot products normalized by vector magnitudes (L2 norms)
cos_sim_matrix_array = cosine_similarity(word_count_matrix_sparse)

# Package the resulting numpy array into a labeled DataFrame
df_cosine_sim = pd.DataFrame(
    cos_sim_matrix_array, 
    index=doc_names, 
    columns=doc_names
)
df_cosine_sim.index.name = 'Document'

# Reset indices for clean HTML display without blank columns
df_vec_disp = df_vectors.reset_index()
df_cos_disp = df_cosine_sim.reset_index().rename(columns={'Document': 'Target / Ref'})

# Render the feature extraction matrix and the resulting similarity matrix side-by-side
display_side_by_side_datasets(
    [df_vec_disp, df_cos_disp],
    [
        "جدول ۱: ماتریس بردارهای فراوانی کلمات (Bag of Words)", 
        "جدول ۲: ماتریس شباهت کسینوسی (۱.۰ = کاملاً هم‌جهت)"
    ]
)

# ==========================================================
# SECTION 3: ANALYTICAL VISUALIZATION (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی - اثبات استقلالِ زاویه از بزرگی بردار", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکی زیر مکانیزم عملکرد شباهت کسینوسی را تشریح می‌نماید. در نمودار سمت چپ (نقشه حرارتی)، مشخص است که مقالهٔ اول و دوم "
    "علی‌رغم اختلاف طول، <b>شباهت موضوعی بسیار بالایی (۰.۷۹)</b> دارند. نمودار سمت راست، اندازه ($L_2$-Norm) بردارهای هر مقاله را نشان می‌دهد. "
    "با وجود اینکه بردارِ مقالهٔ دوم بسیار بزرگتر از مقالهٔ اول است، <i>ضریب کسینوسی با موفقیت این اختلافِ مقیاس را فیلتر کرده</i> و "
    "تنها <b>هم‌راستاییِ معنایی</b> را استخراج نموده است.", 
    "paragraph"
)

# Initialize a wide dual-axis canvas for comparative visual analytics
fig, axes = plt.subplots(1, 2, figsize=(19, 6.5), dpi=135)

# --- SUBPLOT 1: Heatmap of Pairwise Cosine Similarity ---
# Using Academic Blues palette. Dark Blue = 1.0 (Identical Direction)
sns.heatmap(
    df_cosine_sim, 
    annot=True, 
    cmap='Blues', 
    fmt='.3f', 
    vmin=0, vmax=1, 
    ax=axes[0], 
    linewidths=2.0, 
    linecolor='white',
    cbar_kws={'label': 'Cosine Similarity Score'},
    annot_kws={"size": 15, "weight": "bold"}
)

axes[0].annotate(
    "Semantic Alignment:\nDoc_1 and Doc_2 are highly similar (0.791)\nbecause they point in the same 'direction'\n(sharing tech words), regardless of their lengths.",
    xy=(1.5, 0.5), xytext=(3.0, 1.2),
    fontsize=11, fontweight='bold', color='#16a085', ha='center',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5),
    clip_on=False
)

axes[0].annotate(
    "Orthogonality:\nDoc_3 shares zero words with Doc_1.\nThey are mathematically orthogonal (90 degrees),\nyielding a cosine similarity of exactly 0.000.",
    xy=(2.5, 0.5), xytext=(3.0, 2.8),
    fontsize=11, fontweight='bold', color='#c0392b', ha='center',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5),
    clip_on=False
)

axes[0].set_title('Figure 1: Topology of Cosine Similarity Matrix', fontdict={'fontsize': 15, 'fontweight': 'bold'}, pad=15)
axes[0].set_xlabel('Target Document (Vector B)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_ylabel('Reference Document (Vector A)', fontdict={'fontsize': 13, 'fontweight': 'bold'})

axes[0].set_xticklabels(axes[0].get_xticklabels(), fontweight='bold', fontsize=11, rotation=15)
axes[0].set_yticklabels(axes[0].get_yticklabels(), fontweight='bold', fontsize=11, rotation=0, va='center')

# --- SUBPLOT 2: Bar Plot of Document Vector Magnitudes (L2 Norm) ---
# Calculate the L2 Norm (Euclidean length) of each document vector
l2_norms = np.linalg.norm(df_vectors.values, axis=1)
df_norms = pd.DataFrame({
    'Document': doc_names, 
    'Vector_Magnitude': l2_norms
})

sns.barplot(
    data=df_norms, 
    x='Document', 
    y='Vector_Magnitude', 
    ax=axes[1], 
    palette=['#2980b9', '#2c3e50', '#1abc9c'], # Academic Blue Variants
    edgecolor='#2c3e50', 
    linewidth=1.5,
    width=0.5
)

axes[1].set_title('Figure 2: Vector Magnitudes ($L_2$ Norm) of Articles', fontdict={'fontsize': 15, 'fontweight': 'bold'}, pad=15)
axes[1].set_xlabel('Document Topology', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_ylabel('Absolute Vector Magnitude ||A||', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_xticklabels(axes[1].get_xticklabels(), fontweight='bold', fontsize=11, rotation=15)
axes[1].grid(axis='y', linestyle='--', alpha=0.5)

# Add value labels on top of the bars for exact magnitude reference
for p in axes[1].patches:
    axes[1].annotate(
        format(p.get_height(), '.2f'), 
        (p.get_x() + p.get_width() / 2., p.get_height()), 
        ha='center', va='center', xytext=(0, 12), textcoords='offset points',
        fontweight='bold', fontsize=12, color='#2c3e50'
    )

# Clear top and right spines for a clean academic aesthetic
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی عملیات (Strategic Implication):</b><br>"
    r"محاسباتِ گرافیکی بالا اثبات می‌نماید که در معماریِ <b>سیستم‌های بازیابی اطلاعات (Information Retrieval)</b>، به کارگیری فواصلِ وابسته به مقیاس (نظیر اقلیدسی) "
    r"کارایی الگوریتم را مخدوش می‌سازد؛ زیرا فاصلهٔ مقالهٔ کوتاه ۱ با مقالهٔ بلند ۲ (به دلیل تفاوت چشمگیر در بزرگیِ بردارها - گراف ۲)، در فضای دکارتی بسیار زیاد محاسبه می‌شود!<br><br>"
    r"در مقابل، <b>شباهت کسینوسی</b> با تقسیمِ حاصل‌ضربِ داخلی بر اندازهٔ بردارها، اثرِ <i>طول مقاله (Document Length)</i> را به صورتِ کامل خنثی نموده "
    r"و سیستم را قادر می‌سازد تا هم‌خطی و جهت‌گیری معنایی (زاویهٔ بردارها) را صرف‌نظر از حجم کلمات، با دقتِ قطعی استخراج نماید."
)
display_academic_text(insight_conclusion, "insight")
```
