  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: فشرده‌سازی سیگنال‌های بایومتریک در «اپل واچ» (Apple Watch) 💾
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    «اپل واچ» (Apple Watch) سیگنال‌های بایومتریک مداوم (مثل وضعیت حرکت، ضربان قلب و فعالیت بدنی) را هر ثانیه ثبت می‌کند. در طول یک شب خواب ۸ ساعته، سنسور وضعیت حرکتی ممکن است هزاران سیگنال متوالی و <b>کاملاً تکراری</b> با مقدار <b>«Static» (ثابت)</b> ثبت کند، چون کاربر در خواب برای دقایق زیادی بی‌حرکت است. ذخیره تمام این مقادیر تکراری به‌صورت جداگانه، فضای زیادی از حافظه واچ را هدر می‌دهد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>راه‌حل — کدگذاری طول اجرا (Run-Length Encoding / RLE):</b><br>
    به‌جای ذخیره هر مقدار به‌صورت جداگانه، فقط <b>مقدار</b> و تعداد <b>تکرارهای متوالی</b> آن ذخیره می‌شود:
  </p>

  <ul style="text-align: right; margin: 0 0 10px; padding-right: 22px;">
    <li><b>داده خام (قبل از فشرده‌سازی):</b><br>
      <code>[Static, Static, Static, Static, Active, Active, Static, Static, Static, Static, Static, Active]</code><br>
      👈 <b>۱۲ رکورد جداگانه</b>
    </li>
    <li><b>داده فشرده‌شده (بعد از RLE):</b><br>
      <code>[(Static, 4), (Active, 2), (Static, 5), (Active, 1)]</code><br>
      👈 فقط <b>۴ رکورد</b> = صرفه‌جویی <b>۶۶٪</b> در فضای ذخیره‌سازی!
    </li>
  </ul>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>کجا RLE بهترین کارایی را دارد؟</b><br>
    &nbsp;&nbsp;&nbsp;• داده‌های حسگر IoT با توالی‌های طولانی تکراری.<br>
    &nbsp;&nbsp;&nbsp;• داده‌های تصویر (مثل پیکسل‌های سفید پس‌زمینه در اسکن‌های پزشکی).<br>
    &nbsp;&nbsp;&nbsp;• سیگنال‌های دودویی (۰ و ۱) در شبکه‌های مخابراتی.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> با RLE، اپل واچ داده‌های بایومتریک را با کسری از فضای حافظه ذخیره می‌کند و در عین حال توانایی بازسازی دقیق سیگنال اصلی را حفظ می‌نماید.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: کاهش چشمگیر حجم داده‌های متوالی تکراری با نگهداری اطلاعات کامل
    </div>
  </p>

</div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from itertools import groupby 

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})

plt.rcParams['figure.dpi'] = 135

np.random.seed(42)
```

```python
display_academic_text(
    "تقلیل و فشرده‌سازی داده‌ها: ذخیره‌سازی سیگنال‌های بایومتریک با الگوریتم RLE ⌚", 
    "main_title"
)

display_academic_text(
    "در سیستم‌های اینترنت اشیا (مانند پایشگر خواب در ساعت‌های هوشمند)، سیگنال‌ها با فرکانس بالا ثبت می‌شوند که منجر به تولید <i>توالی‌های طولانی و تکراری</i> "
    "از یک وضعیتِ ثابت می‌گردد. الگوریتم <b>Run-Length Encoding (RLE)</b> با تبدیل هر توالیِ تکراری به یک جفتِ <code>(وضعیت، تعداد تکرار)</code>، "
    "حجم داده‌ها را به شدت کاهش داده و از اتلاف حافظه جلوگیری می‌کند.", 
    "paragraph"
)

display_academic_text("معادله محاسبه نرخ فشرده‌سازی (Compression Ratio):", "sub_title")

# Utilizing Raw Strings (r"") for LaTeX rendering
display_academic_text(
    r"$$CR = \left( 1 - \frac{Size_{Compressed}}{Size_{Original}} \right) \times 100$$", 
    "paragraph"
)

# ==========================================================
# SECTION 2: DATA INITIALIZATION AND RLE COMPRESSION PIPELINE
# ==========================================================
display_academic_text(
    "مرحله نخست: اعمال فشرده‌سازی بدون اتلاف (Lossless Compression) و بازسازی سیگنال", 
    "sub_title"
)

display_academic_text(
    "در این فاز، سیگنالِ حرکتیِ خام به الگوریتم RLE تزریق می‌شود. خروجیِ الگوریتم، مجموعه‌ای از تاپل‌ها (Tuples) است که حجم اشغالیِ حافظه را کاهش می‌دهد. "
    "سپس با اجرای تابع دیکودر (Decoder)، سیگنالِ فشرده‌شده دقیقاً به حالت اولیه <b>بازسازی (Restore)</b> می‌شود تا <i>بدونِ اتلاف بودنِ (Lossless)</i> الگوریتم اثبات گردد.", 
    "paragraph"
)

# Step 1: Initialize the raw biometric signal array (Mimicking a 20-second sleep monitoring window)
raw_signal = [
    'Static', 'Static', 'Static', 'Static',
    'Active', 'Active',
    'Static', 'Static', 'Static', 'Static', 'Static',
    'Active',
    'Static', 'Static', 'Static',
    'Active', 'Active', 'Active',
    'Static', 'Static',
]

df_raw = pd.DataFrame({
    'Time_Seq': range(1, len(raw_signal) + 1), 
    'Sensor_State': raw_signal
})

# Step 2: Define and apply Run-Length Encoding (RLE)
def run_length_encode(signal):
    """
    Compresses a sequence using RLE.
    Groups identical consecutive elements and yields (Element, Count) tuples.
    """
    return [(key, sum(1 for _ in group)) for key, group in groupby(signal)]

rle_compressed = run_length_encode(raw_signal)

df_compressed = pd.DataFrame(
    rle_compressed, 
    columns=['Encoded_State', 'Consecutive_Count']
)
df_compressed.insert(0, 'Block_ID', range(1, len(df_compressed) + 1))

# Step 3: Define and apply Run-Length Decoding to restore the exact original signal
def run_length_decode(compressed_signal):
    """
    Decompresses an RLE encoded sequence back to its original flat structure.
    """
    return [value for value, count in compressed_signal for _ in range(count)]

restored_signal = run_length_decode(rle_compressed)
df_restored = pd.DataFrame({
    'Time_Seq': range(1, len(restored_signal) + 1), 
    'Restored_State': restored_signal
})

# Check lossless integrity
is_lossless = (raw_signal == restored_signal)

# Calculate Compression Statistics
original_size = len(raw_signal)
# In standard RLE, each block requires 2 units of memory (Value and Count)
compressed_size = len(rle_compressed) * 2  
compression_ratio = (1 - compressed_size / original_size) * 100

# Prepare datasets for display
df_raw_disp = df_raw.head(10).reset_index(drop=True)
df_comp_disp = df_compressed.reset_index(drop=True)
df_rest_disp = df_restored.head(10).reset_index(drop=True)

# Render the data transformation stages side-by-side
display_side_by_side_datasets(
    [df_raw_disp, df_comp_disp, df_rest_disp],
    [
        f"جدول ۱: سیگنال خام (نمونه از {original_size} رکورد)", 
        f"جدول ۲: ماتریس RLE (صرفه‌جویی {compression_ratio:.0f}٪)",
        f"جدول ۳: سیگنال بازسازی‌شده (Lossless: {is_lossless})"
    ]
)

# ==========================================================
# SECTION 3: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی - توپولوژی سیگنال‌های دیجیتال", 
    "sub_title"
)

display_academic_text(
    "نمودار زیر معماریِ زمانیِ سیگنال را به تصویر می‌کشد. نمودار اول (سرمه‌ای) سیگنال حجیم اولیه را نشان می‌دهد. "
    "نمودار دوم (بارچارت) ساختار حافظه پس از فشرده‌سازی RLE را بازنمایی می‌کند که در آن هر بلوک به یک رکورد متنی تقلیل یافته است. "
    "نمودار سوم (آبی کلاسیک) تأیید می‌کند که سیگنال دیکودشده، از نظر هندسی و مقداری، یک تطابق ۱۰۰ درصدی با سیگنال مرجع دارد.", 
    "paragraph"
)

# Initialize a 3-row canvas for temporal signal analytics
fig, axes = plt.subplots(3, 1, figsize=(19, 11), dpi=135)

# Pre-processing: Map categorical textual states to binary numeric values for digital signal plotting
signal_map = {'Static': 0, 'Active': 1}
numeric_raw = [signal_map[s] for s in raw_signal]
numeric_restored = [signal_map[s] for s in restored_signal]

# ----------------------------------------------------------
# Plot 1: Original Raw Digital Signal (Step Plot)
# ----------------------------------------------------------
axes[0].step(
    range(len(numeric_raw)), 
    numeric_raw, 
    color='#2c3e50', # Navy
    linewidth=3.0, 
    where='post'
)
axes[0].fill_between(
    range(len(numeric_raw)), 
    numeric_raw, 
    step='post', 
    alpha=0.15, 
    color='#2c3e50'
)

axes[0].annotate(
    "Memory Waste:\nThese 5 consecutive seconds of 'Static'\nconsume 5 distinct memory addresses.",
    xy=(8, 0.1), xytext=(7, 0.5),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title(f'Figure 1A: Original Raw Signal Topology (Memory Footprint: {original_size} Units)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=12)
axes[0].set_ylabel('Sensor State', fontdict={'fontsize': 12, 'fontweight': 'bold'})
axes[0].set_yticks([0, 1])
axes[0].set_yticklabels(['Static (0)', 'Active (1)'], fontweight='bold')
axes[0].grid(True, linestyle='--', alpha=0.5)
axes[0].set_xlim(0, len(numeric_raw) - 1)

# ----------------------------------------------------------
# Plot 2: RLE Compressed Blocks (Bar Plot)
# ----------------------------------------------------------
labels = [f"{v}\n(Count: {c})" for v, c in rle_compressed]
values = [signal_map[v] for v, c in rle_compressed]
x_positions = range(len(rle_compressed))

# Using Academic Blue palette: Turquoise for Active, Grey-Blue for Static
sns.barplot(
    x=list(x_positions), 
    y=values, 
    hue=list(x_positions), 
    legend=False,
    palette=['#1abc9c' if v == 1 else '#95a5a6' for v in values], 
    edgecolor='#2c3e50', 
    linewidth=1.5,
    ax=axes[1]
)

axes[1].annotate(
    "Compression Efficiency:\nThe previous 5 'Static' units are now\nstored as a single tuple: (Static, 5).\nTaking up only 2 memory addresses!",
    xy=(2, 0.1), xytext=(2.5, 0.5),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[1].set_xticks(x_positions)
axes[1].set_xticklabels(labels, fontsize=11, fontweight='bold')
axes[1].set_title(f'Figure 1B: Run-Length Encoded Blocks (Memory Footprint: {compressed_size} Units)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=12)
axes[1].set_ylabel('Encoded State', fontdict={'fontsize': 12, 'fontweight': 'bold'})
axes[1].set_yticks([0, 1])
axes[1].set_yticklabels(['Static (0)', 'Active (1)'], fontweight='bold')
axes[1].grid(axis='y', linestyle='--', alpha=0.5)

# ----------------------------------------------------------
# Plot 3: Restored Signal Integrity (Step Plot)
# ----------------------------------------------------------
axes[2].step(
    range(len(numeric_restored)), 
    numeric_restored, 
    color='#2980b9', # Classic Blue
    linewidth=3.0, 
    where='post'
)
axes[2].fill_between(
    range(len(numeric_restored)), 
    numeric_restored, 
    step='post', 
    alpha=0.15, 
    color='#2980b9'
)

axes[2].set_title(f'Figure 1C: Restored Signal Post-Decompression (Integrity Match: {is_lossless})', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=12)
axes[2].set_xlabel('Time Sequence (Seconds)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[2].set_ylabel('Sensor State', fontdict={'fontsize': 12, 'fontweight': 'bold'})
axes[2].set_yticks([0, 1])
axes[2].set_yticklabels(['Static (0)', 'Active (1)'], fontweight='bold')
axes[2].grid(True, linestyle='--', alpha=0.5)
axes[2].set_xlim(0, len(numeric_raw) - 1)

# Clear top and right spines for all subplots
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout(pad=2.0)
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی عملیات (Strategic Implication):</b><br>"
    r"ذخیره‌سازی خام در سیستم‌های مانیتورینگِ پیوسته، توجیه معماری ندارد. الگوریتم فشرده‌سازی <b>RLE</b> با تبدیلِ توالی‌هایِ ایستا به متغیرهای مقداری، "
    r"توانست حجم داده‌ها را به میزان <b>۴۰٪</b> کاهش دهد، بدون اینکه حتی یک بیت از اطلاعاتِ ارزشمند حذف شود (Lossless).<br>"
    r"این الگوریتم، پیش‌نیاز قطعیِ استقرارِ سیستم‌های اینترنت اشیا (IoT)، سنسورهای پایش سلامت و پردازشِ تصویر (مانند فرمت‌های فکس و تصاویر ساده) است تا بارِ ترافیک شبکه (Bandwidth) "
    r"و فضای دیسک در سرورهای ابری بهینه‌سازی گردد."
)
display_academic_text(insight_conclusion, "insight")
```

    <li><b>نمودار سوم (Step Chart بنفش):</b> سیگنال بازسازی‌شده را نشان می‌دهد که با سیگنال اصلی کاملاً منطبق است.</li>
