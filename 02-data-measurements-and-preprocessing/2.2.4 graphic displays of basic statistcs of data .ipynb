  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    پلتفرم «دیوار» روزانه هزاران آگهی اجاره و فروش مسکن را منتشر می‌کند. برای تحلیل این حجم زیاد از داده‌ها، تیم داده‌کاوی نیاز دارد تا ویژگی‌هایی مثل قیمت، متراژ و موقعیت مکانی را به‌صورت خلاصه و قابل فهم بررسی کند. برای این کار از <b>نمایش‌های گرافیکی آمار پایه</b> استفاده می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۱. هیستوگرام (Histogram) — بررسی توزیع قیمت‌ها</b><br>
    با رسم هیستوگرام برای قیمت اجاره ماهانه آگهی‌ها (به تومان) و تقسیم آن‌ها به سطل‌های مشخص، می‌توان شکل توزیع قیمت را دید. معمولاً توزیع قیمت اجاره در شهرها <b>چولگی به راست (Right-Skewed)</b> دارد؛ یعنی بیشتر آگهی‌ها در رنج اقتصادی قرار می‌گیرند و تعداد کمی از آن‌ها (مثل خانه‌های لوکس) قیمت بسیار بالاتری دارند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۲. نمودار جعبه‌ای (Box Plot) — مقایسه قیمت برحسب نوع ملک</b><br>
    با استفاده از نمودار جعبه‌ای می‌توان قیمت اجاره را برحسب نوع ملک (آپارتمان، خانه ویلایی و استودیو) با هم مقایسه کرد. این نمودار علاوه بر نمایش <b>میانه (Median)</b> و <b>۵۰٪ مرکزی داده‌ها (IQR)</b>، به‌راحتی <b>نقاط پرت (Outliers)</b> را هم نشان می‌دهد؛ مثلاً آگهی‌های اجاره بسیار گران‌تر از میانگین بازار.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۳. نمودار پراکندگی (Scatter Plot) — بررسی رابطه متراژ و قیمت</b><br>
    فرضیه‌ای رایج در بازار مسکن این است: «هرچه متراژ ملک بیشتر باشد، قیمت اجاره آن نیز بالاتر می‌رود». برای بررسی این رابطه، محور X را <b>متراژ (متر مربع)</b> و محور Y را <b>قیمت اجاره (تومان)</b> قرار می‌دهیم. هر آگهی به‌صورت یک نقطه روی نمودار نمایش داده می‌شود. اگر نقاط به‌صورت کلی روند صعودی داشته باشند، نشان‌دهنده یک <b>رابطه مثبت</b> بین متراژ و قیمت است.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> با این نمایش‌های گرافیکی، می‌توان بدون بررسی تک‌تک آگهی‌ها، روند کلی بازار مسکن را در یک نگاه درک کرد و تصمیم‌گیری منطقی‌تری داشت.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: تبدیل داده‌های خام آگهی‌ها به درک بصری از توزیع، داده‌های پرت و روابط بین متغیرها
    </div>
  </p>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display, HTML

import warnings
warnings.filterwarnings('ignore')
sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})

plt.rcParams['figure.dpi'] = 135

def display_academic_text(text, type="normal"):
    """Renders text in a strict academic format."""
    styles = {
        "main_title": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 24px; "
            "font-weight: bold; text-align: center; border-bottom: 2px solid #34495e; "
            "padding-bottom: 15px; margin: 30px 0;"
        ),
        "sub_title": (
            "font-family: Tahoma, sans-serif; color: #34495e; font-size: 19px; "
            "font-weight: bold; border-right: 5px solid #2980b9; padding-right: 15px; "
            "margin: 40px 0px 20px 0px; background-color: #ecf0f1; padding: 12px; "
            "border-radius: 4px;"
        ),
        "paragraph": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 16px; "
            "margin-bottom: 20px; line-height: 1.8; text-align: justify;"
        ),
        "insight": (
            "font-family: Tahoma, sans-serif; color: #16a085; font-size: 16px; "
            "font-weight: bold; background-color: #e8f8f5; padding: 20px; "
            "border-left: 5px solid #1abc9c; margin: 30px 0; line-height: 1.8; "
            "text-align: justify;"
        )
    }
    display(HTML(f"<div dir='rtl' style='text-align: right; {styles[type]}'>{text}</div>"))

def get_protected_css():
    """Returns essential CSS for robust and clean table formatting."""
    return """
    <style>
    .academic-table-wrapper { 
        direction: ltr !important; 
        overflow-x: auto !important; 
        width: 100% !important; 
        background-color: #fff; 
        border: 1px solid #bdc3c7; 
        border-radius: 4px; 
        box-shadow: 0 3px 6px rgba(0,0,0,0.08); 
    }
    table.strict-tbl { 
        border-collapse: collapse !important; 
        width: 100% !important; 
        margin: 0 !important; 
        font-family: inherit !important;
        font-size: 14px !important; 
        table-layout: auto !important; 
        border: none !important; 
    }
    table.strict-tbl thead th { 
        background-color: #2c3e50 !important; 
        color: #ffffff !important; 
        padding: 12px 18px !important; 
        text-align: center !important; 
        border: 1px solid #34495e !important; 
        white-space: nowrap !important; 
        font-weight: bold !important; 
    }
    table.strict-tbl tbody td { 
        padding: 10px 18px !important; 
        text-align: center !important; 
        border: 1px solid #bdc3c7 !important; 
        white-space: nowrap !important; 
        color: #2c3e50 !important; 
    }
    table.strict-tbl tbody tr:nth-child(even) td { 
        background-color: #f8f9fa !important; 
    }
    table.strict-tbl tbody tr:hover td { 
        background-color: #e8ecef !important; 
    }
    </style>
    """

def display_side_by_side_datasets(dfs, titles, *args):
    """Renders single or multiple pandas DataFrames side-by-side gracefully."""
    if not isinstance(dfs, list) and not isinstance(titles, list):
        if len(args) >= 2:
            dfs = [dfs, args[0]]
            titles = [titles, args[1]]
        else:
            dfs = [dfs]
            titles = [titles]
            
    if len(dfs) == 1:
        df = dfs[0]
        title = titles[0]
        table_html = df.to_html(index=False, classes=['strict-tbl'], border=0, float_format=lambda x: f"{x:.2f}" if isinstance(x, (float, int)) else x)
        html = f"""
        <div style="width: 100%; margin: 20px 0 35px 0;">
            <div dir="rtl" style="background-color: #2c3e50; color: white; padding: 12px; 
                                 text-align: center; font-family: Tahoma, sans-serif; 
                                 font-weight: bold; font-size: 15px; border-radius: 4px 4px 0 0;">
                {title}
            </div>
            <div class="academic-table-wrapper" style="margin-top: 0; border-top: none;">
                {table_html}
            </div>
        </div>
        """
        display(HTML(get_protected_css() + html))
        return
    
    col_html_list = []
    for i, (df, title) in enumerate(zip(dfs, titles)):
        table_html = df.to_html(index=False, classes=['strict-tbl'], border=0, float_format=lambda x: f"{x:.2f}" if isinstance(x, (float, int)) else x)
        bg_color = "#2c3e50" 
        col_html = f"""
        <div class="flex-column" style="flex: 1 1 0; min-width: 0; display: flex; flex-direction: column;">
            <div style="background-color: {bg_color}; color: white; padding: 12px; 
                        text-align: center; font-family: Tahoma, sans-serif; 
                        font-weight: bold; font-size: 15px; border-radius: 4px 4px 0 0;">
                {title}
            </div>
            <div class="academic-table-wrapper" style="margin-top: 0; border-top: none;">
                {table_html}
            </div>
        </div>
        """
        col_html_list.append(col_html)
    
    columns_joined = "".join(col_html_list)
    html = f"""
    <div style="display: flex; flex-direction: row; justify-content: space-between; 
                gap: 20px; width: 100%; max-width: 100%; margin: 20px 0 35px 0;
                box-sizing: border-box;">
        {columns_joined}
    </div>
    """
    display(HTML(get_protected_css() + html))

np.random.seed(42)
```

```python
display_academic_text(
    "آمار توصیفی گرافیکی: بازنمایی بصری توزیع و پراکندگی در بازار مسکن 🏢", 
    "main_title"
)

display_academic_text(
    "پیش از استقرار الگوریتم‌های پیچیدهٔ یادگیری ماشین، ارزیابی ساختاریِ پایگاه داده از طریق <b>آمار پایهٔ گرافیکی (Graphical EDA)</b> امری الزامی است. "
    "در این سناریو، پویاییِ بازار اجاره مسکن (با نگاهی به املاک و ویلاهای کلان‌شهری نظیر شیراز) مدل‌سازی شده است. هدف، درکِ سریعِ توزیع قیمت‌ها، "
    "شناسایی داده‌های پرت (نظیر املاک فوق‌لوکس) و کشف همبستگی خطی میان متراژ و بهای اجاره می‌باشد. نمایش‌های گرافیکی، میلیون‌ها رکورد عددی را به "
    "الگوهای قابلِ تفسیرِ بصری ترجمه می‌نمایند.", 
    "paragraph"
)

# ==========================================================
# SECTION 2: DATA INITIALIZATION AND SCALING
# ==========================================================
display_academic_text(
    "مرحله نخست: شبیه‌سازی فضای بازار و استخراج آمار توصیفی (Descriptive Statistics)", 
    "sub_title"
)

display_academic_text(
    "در این فاز، داده‌های بازار مسکن با فرضِ <b>چولگی به راست (Right-Skewness)</b> در قیمت‌ها شبیه‌سازی گردیده و چند رکوردِ ناهنجار (Outlier) "
    "به صورت دستی به سیستم تزریق می‌شود تا عملکرد الگوریتم‌های مصورسازی به چالش کشیده شود. همچنین متغیر قیمت به مقیاس <i>«میلیون تومان»</i> تبدیل می‌گردد "
    "تا خوانایی داده‌ها در جداول و محورهای گرافیکی بهینه‌سازی شود.", 
    "paragraph"
)

# Step 1: Generate a synthetic dataset representing real estate rental listings
n_samples = 150
data = {
    'Listing_ID': range(1001, 1001 + n_samples),
    'Area_M2': np.random.uniform(50, 250, n_samples),
    'Property_Type': np.random.choice(['Apartment', 'Villa', 'Studio'], size=n_samples, p=[0.6, 0.25, 0.15]),
    'Distance_Center_KM': np.random.uniform(1.0, 20.0, n_samples)
}

# Step 2: Generate monthly rent with a mathematically right-skewed distribution.
# An exponential distribution models real-world economic realities where most assets fall in the lower/middle tier.
base_price = np.random.exponential(scale=6_000_000, size=n_samples) + (data['Area_M2'] * 60_000)

# Convert the raw numerical values to 'Million Toman' to prevent extreme axis labels
data['Price_Million_Toman'] = np.round(base_price / 1_000_000, 1)

# Step 3: Explicitly inject extreme luxury outliers to test Box Plot robustness
data['Price_Million_Toman'][5] = 85.0
data['Price_Million_Toman'][12] = 110.0
data['Property_Type'][12] = 'Villa'

df_raw = pd.DataFrame(data)

# Step 4: Create a statistical summary dataframe isolating Area and Price.
# The describe() function computes central tendency, dispersion, and shape of the distribution.
df_summary = df_raw[['Area_M2', 'Price_Million_Toman']].describe().round(2).reset_index()
df_summary.rename(columns={'index': 'Statistical Metric'}, inplace=True)

# Render a sample of raw observations alongside the descriptive statistics
display_side_by_side_datasets(
    [df_raw.head(6), df_summary],
    [
        "جدول ۱: رکوردهای ثبت‌شده آگهی‌های املاک (نمونه)", 
        "جدول ۲: ماتریس آمار توصیفی (Descriptive Stats)"
    ]
)

# ==========================================================
# SECTION 3: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: استنتاج گرافیکی و تحلیل بصری فضای ویژگی‌ها (Visual Analytics)", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکیِ سه‌گانهٔ زیر، ابزارهای کلیدی آمار توصیفی را به کار می‌گیرد. <b>هیستوگرام</b> تراکم قیمتی را در قالب سطل‌های داده (Bins) نمایش می‌دهد. "
    "<b>نمودار جعبه‌ای (Boxplot)</b>، توزیع چارک‌ها را تفکیک نموده و نقاط خارج از سبیل‌ها (Whiskers)، داده‌های پرت را ایزوله می‌کنند. "
    "در نهایت، <b>نمودار پراکندگی (Scatterplot)</b> وابستگیِ خطیِ احتمالی میان متراژ و قیمت را به تصویر می‌کشد.", 
    "paragraph"
)

# Initialize a 1x3 subplot grid with an extended width for comprehensive visual analysis
fig, axes = plt.subplots(1, 3, figsize=(20, 6.5), dpi=135)

academic_palette = ['#2980b9', '#1abc9c', '#2c3e50'] # Classic Blue, Turquoise, Navy

# ----------------------------------------------------------
# PLOT 1: Histogram (Data Distribution & Skewness)
# ----------------------------------------------------------
sns.histplot(
    data=df_raw, 
    x='Price_Million_Toman', 
    bins=20, 
    kde=True, 
    ax=axes[0], 
    color='#2c3e50', 
    edgecolor='white', 
    linewidth=1.2,
    line_kws={'linewidth': 2.5, 'color': '#e74c3c'}
)

axes[0].annotate(
    "Right-Skewed Distribution:\nThe long tail to the right indicates\nthat while most rents are affordable,\na few expensive properties pull\nthe mean upwards.",
    xy=(45, 10), xytext=(35, 20),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title('Figure 1: Monthly Rent Distribution', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[0].set_xlabel('Monthly Rent (Million Toman)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_ylabel('Frequency (Number of Listings)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].grid(True, linestyle='--', alpha=0.5)

# ----------------------------------------------------------
# PLOT 2: Box Plot (Outlier Detection and IQR)
# ----------------------------------------------------------
sns.boxplot(
    data=df_raw, 
    x='Property_Type', 
    y='Price_Million_Toman', 
    ax=axes[1], 
    palette=academic_palette, 
    linewidth=2.0, 
    fliersize=8,
    flierprops=dict(marker='X', markerfacecolor='#e74c3c', markeredgecolor='#c0392b')
)

axes[1].annotate(
    "Luxury Outlier Detected!\nThis data point lies far beyond the\nUpper Whisker (Q3 + 1.5*IQR).",
    xy=(1, 108), xytext=(0.5, 80),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[1].set_title('Figure 2: Rent Dispersion & Outliers', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[1].set_xlabel('Property Classification', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_ylabel('Monthly Rent (Million Toman)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].grid(axis='y', linestyle='--', alpha=0.5)

# ----------------------------------------------------------
# PLOT 3: Scatter Plot (Bivariate Relationship Mapping)
# ----------------------------------------------------------
sns.scatterplot(
    data=df_raw, 
    x='Area_M2', 
    y='Price_Million_Toman', 
    hue='Property_Type', 
    ax=axes[2], 
    palette=academic_palette, 
    s=90, 
    edgecolor='white', 
    linewidth=1.0, 
    alpha=0.85
)

axes[2].set_title('Figure 3: Area vs. Rent Correlation', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[2].set_xlabel('Property Area (Square Meters)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[2].set_ylabel('Monthly Rent (Million Toman)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[2].legend(title='Property Type', loc='upper left', frameon=True, facecolor='#ffffff', edgecolor='#bdc3c7')
axes[2].grid(True, linestyle='--', alpha=0.5)

# Remove the top and right spines from all axes for a cleaner academic aesthetic
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی تحلیل بصری (Strategic Implication):</b><br><br>"
    r"تحلیل گرافیکی مؤید آن است که بخش عمده‌ای از عرضهٔ بازار در هستهٔ اقتصادی متمرکز گردیده و توزیع قیمت دارای یک <b>چولگی به راست (Right-Skewness)</b> است (گراف ۱). "
    r"همچنین، نمودار جعبه‌ای (گراف ۲) با موفقیت توانست ویلاهای فوق‌لوکسِ ناهنجار را به عنوان <i>Outlier</i> از بدنهٔ اصلی داده‌ها تفکیک نماید. <br><br>"
    r"این شفافیتِ بصری، تیم داده‌کاوی را قادر می‌سازد تا پیش از استقرار مدل‌های رگرسیون قیمت‌گذاری، <b>استراتژی برخورد با داده‌های پرت</b> (نظیر حذف آن‌ها یا استفاده از Robust Scaler) "
    r"و نرمال‌سازی متغیرها (نظیر Log-Transformation برای رفع چولگی) را به صورت کاملاً ساختاریافته و علمی تدوین نماید."
)
display_academic_text(insight_conclusion, "insight")
```

در این مثال:
<ul>
<li>ما یک مجموعه داده شبیه‌سازی‌شده از آگهی‌های اجاره مسکن در پلتفرم <b>دیوار</b> ساختیم. این مجموعه شامل ویژگی‌های <b>'Area_M2'</b> (متراژ به متر مربع)، <b>'Price_Toman'</b> (قیمت اجاره ماهانه به تومان)، <b>'Property_Type'</b> (نوع ملک) و چند آگهی لوکس به‌عنوان داده‌ی پرت است.</li>
<li><b>هیستوگرام (Histogram):</b> با متد <b>plt.hist()</b> قیمت‌ها را بر حسب میلیون تومان در ۱۵ سطل (Bin) نمایش کردیم. این نمودار چولگی توزیع قیمت به سمت راست را به‌خوبی نشان می‌دهد (بیشتر آگهی‌ها اقتصادی و تعداد کمی گران‌قیمت هستند).</li>
<li><b>نمودار جعبه‌ای (Box Plot):</b> با متد <b>plt.boxplot()</b> توزیع قیمت اجاره را به تفکیک نوع ملک رسم کردیم. خط میانی داخل جعبه، <b>میانه قیمت</b> هر گروه را نشان می‌دهد و نقاط خارج از باله‌ها، <b>داده‌های پرت (Outliers)</b> را مشخص می‌کنند.</li>
<li><b>نمودار پراکندگی (Scatter Plot):</b> با متد <b>plt.scatter()</b> رابطه بین دو متغیر پیوسته «متراژ» و «قیمت اجاره» را بررسی کردیم. محور افقی متراژ (متر مربع) و محور عمودی قیمت اجاره (میلیون تومان) قرار گرفته تا روند صعودی رابطه مثبت بین این دو متغیر به‌خوبی دیده شود.</li>
</ul>
