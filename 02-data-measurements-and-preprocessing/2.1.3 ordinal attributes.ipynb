<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: رتبه‌بندی سطح اشتراک و رضایت کاربران در «فودباکس» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در استارتاپ «فودباکس»، با متغیرهایی روبه‌رو هستیم که علاوه بر داشتن برچسب، دارای یک <b>ترتیب و رتبه‌بندی طبیعی و معنادار</b> هستند. برای مثال، سطوح اشتراک کاربران شامل «برنزی»، «نقره‌ای»، «طلایی» و «پلاتینیم» یا حجم وعده‌های غذایی شامل «بزرگ»، «متوسط» و «کوچک» است. به این متغیرها <b>ویژگی‌های ترتیبی (Ordinal Attributes)</b> می‌گویند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام اول: تفاوت کلیدی با ویژگی‌های اسمی</b><br>
    در ویژگی‌های اسمی (مثل نوع رژیم کیتو یا گیاه‌خواری)، هیچ ترتیبی وجود نداشت. اما در اینجا مشخص است که اشتراک «طلایی» رتبه‌ای بالاتر از «نقره‌ای» دارد. با این حال، فاصله ریاضی دقیق بین برنزی تا نقره‌ای لزوماً با فاصله نقره‌ای تا طلایی برابر نیست؛ تنها چیزی که برای ما حتمی است، <b>جهت و ترتیب رتبه‌هاست</b>.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام دوم: خطای بزرگ One-Hot Encoding در متغیرهای ترتیبی</b><br>
    اگر برای این متغیرها از One-Hot Encoding استفاده کنیم، ترتیب ارزشمند بین رتبه‌ها را کاملاً نابود می‌کنیم و مدل نمی‌فهمد که کاربر طلایی وفادارتر از کاربر برنزی است! از طرف دیگر، اگر اعداد را به‌صورت تصادفی اختصاص دهیم (مثلاً برنزی=۴ و طلایی=۱)، ترتیب منطقی را معکوس و خراب کرده‌ایم.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام سوم: راه‌حل — کدگذاری ترتیبی دستی (Ordinal Mapping)</b><br>
    بهترین روش، نگاشت دستی اعداد صحیح بر اساس ترتیب واقعی آنهاست؛ به‌طوری که عدد بزرگ‌تر نشان‌دهنده رتبه بالاتر باشد: <br>
    &nbsp;&nbsp;&nbsp;• برنزی 👈 ۱ | نقره‌ای 👈 ۲ | طلایی 👈 ۳ | پلاتینیم 👈 ۴
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> الگوریتم‌های داده‌کاوی (مثل درخت‌های تصمیم) با فهمیدن این ترتیب، می‌توانند پیش‌بینی کنند که با ارتقای کاربر از رتبه ۲ به ۳، احتمال تمدید اشتراک او چقدر افزایش می‌یابد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: حفظ دقیق ترتیب منطقی داده‌ها در قالب اعداد صحیحِ رتبه‌دار
    </div>
  </p>

</div>

```python
display_academic_text(
    "پیش‌پردازش داده‌ها: حفظ یکپارچگی سلسله‌مراتب در ویژگی‌های ترتیبی (Ordinal Encoding) 📊", 
    "main_title"
)

display_academic_text(
    "متغیرهای ترتیبی (Ordinal Variables) علاوه بر تمایزِ دسته‌ای، دارای یک توالی و <b>رتبه‌بندی معنادار (Meaningful Ranking)</b> می‌باشند. "
    "در این سناریو، یک پلتفرم خدمات غذایی نیازمند تبدیل سطوح وفاداری (Loyalty Tiers) و حجم سفارشات به فرمت عددی است. "
    "استفاده از روش‌های رمزگذاری متقارن (نظیر One-Hot Encoding) منجر به <i>تخریبِ این سلسله‌مراتب</i> و از بین رفتنِ فواصلِ منطقی می‌گردد؛ "
    "لذا استراتژی <b>نگاشت ترتیبی (Ordinal Mapping)</b> جهت حفظ ارزش‌های پیش‌بینانه در دستور کار قرار می‌گیرد.", 
    "paragraph"
)

# ==========================================================
# SECTION 2: DATA INITIALIZATION & ORDINAL ENGINEERING
# ==========================================================
display_academic_text(
    "مرحله نخست: بارگذاری داده‌های خام و اعمال نگاشت رتبه‌ای (Rank Mapping)", 
    "sub_title"
)

display_academic_text(
    "در این فاز، مقادیر متنی بر اساس جایگاهِ منطقیِ خود ارزش‌گذاری می‌شوند. به هر سطح، یک عددِ صحیح متناسب با رتبهٔ آن "
    "اختصاص می‌یابد تا الگوریتم‌های مدل‌سازی بتوانند جهت‌گیریِ ارتقا (Upgrade) یا تنزلِ وضعیت (Downgrade) کاربر را به‌دقت تحلیل نمایند.", 
    "paragraph"
)

# Step 1: Initialize the raw database containing ordinal textual attributes
raw_data = {
    'User_ID': [201, 202, 203, 204, 205],
    'Loyalty_Tier': ['Bronze', 'Gold', 'Silver', 'Platinum', 'Bronze'],
    'Meal_Portion': ['Small', 'Large', 'Medium', 'Large', 'Small']
}
df_raw = pd.DataFrame(raw_data)

# Step 2: Create a copy of the dataframe to store the ordinally encoded features
df_processed = df_raw.copy()

# Step 3: Define explicit mapping dictionaries that capture the natural rank of each category.
# Higher integer values strictly correspond to higher conceptual ranks (Monotonicity).
tier_mapping_logic = {
    'Bronze': 1, 
    'Silver': 2, 
    'Gold': 3, 
    'Platinum': 4
}

portion_mapping_logic = {
    'Small': 1, 
    'Medium': 2, 
    'Large': 3
}

# Step 4: Apply the ordinal encoding map to transform textual data into ranked numerical vectors
df_processed['Loyalty_Tier_Rank'] = df_processed['Loyalty_Tier'].map(tier_mapping_logic)
df_processed['Meal_Portion_Rank'] = df_processed['Meal_Portion'].map(portion_mapping_logic)

# Isolate the transformed columns for comparative display
df_encoded_view = df_processed[['User_ID', 'Loyalty_Tier_Rank', 'Meal_Portion_Rank']]

# Render the raw and engineered datasets side-by-side to visually demonstrate the structural transformation
display_side_by_side_datasets(
    [df_raw, df_encoded_view],
    [
        "جدول ۱: پایگاه داده خام (متغیرهای ترتیبیِ متنی)", 
        "جدول ۲: پایگاه داده مهندسی‌شده (رتبه‌های عددیِ محفوظ)"
    ]
)

# ==========================================================
# SECTION 3: ANALYTICAL VISUALIZATION (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلیِ سلسله‌مراتبِ ارزش‌گذاری‌شده", 
    "sub_title"
)

display_academic_text(
    "جهت اثباتِ حفظِ ساختار ترتیبی (Monotonicity)، نمودارهای نردبانی و تصاعدیِ زیر، ارزش عددیِ اختصاص‌یافته به رده‌های اشتراک و حجم وعده‌ها را نمایش می‌دهند. "
    "این بازنمایی گرافیکی تأیید می‌کند که <b>فاصله و تقدم منطقی</b> بین سطوحِ کاربران پس از پیش‌پردازش، کاملاً دست‌نخورده باقی مانده است.", 
    "paragraph"
)

# Initialize a wide dual-plot canvas for optimal academic presentation
fig, axes = plt.subplots(1, 2, figsize=(19, 6.5), dpi=135)

# --- CHART 1: Loyalty Tier Progression (Step Plot) ---
tiers = list(tier_mapping_logic.keys())
tier_ranks = list(tier_mapping_logic.values())

# Step plot to emphasize the discrete, ranked leaps between tiers
axes[0].step(
    tiers, tier_ranks, 
    where='mid', color='#2c3e50', linewidth=3.5, 
    marker='o', markersize=12, markerfacecolor='#1abc9c', markeredgecolor='#2c3e50', markeredgewidth=2
)

axes[0].annotate(
    "Rank Preservation:\nThe algorithm now mathematically understands\nthat 'Platinum' (4) is conceptually greater\nthan 'Bronze' (1) by exactly 3 levels.",
    xy=(2, 3), xytext=(0.5, 3.5),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[0].set_title('Figure 1: Hierarchical Integrity of Loyalty Tiers', fontdict={'fontsize': 15, 'fontweight': 'bold'}, pad=15)
axes[0].set_ylabel('Assigned Numerical Rank (Magnitude)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_xlabel('Original Textual Categories', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_yticks(tier_ranks)
axes[0].grid(True, linestyle='--', alpha=0.5)

# --- CHART 2: Meal Portion Progression (Line/Scatter Plot) ---
portions = list(portion_mapping_logic.keys())
portion_ranks = list(portion_mapping_logic.values())

axes[1].plot(
    portions, portion_ranks, 
    color='#2980b9', linewidth=3.5, linestyle='--',
    marker='D', markersize=12, markerfacecolor='#e74c3c', markeredgecolor='#2c3e50', markeredgewidth=2
)

axes[1].annotate(
    "Monotonic Relationship:\nThe transition from Small to Large is strictly increasing,\navoiding the pitfall of orthogonal (One-Hot) isolation.",
    xy=(1, 2), xytext=(1.2, 1.2),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[1].set_title('Figure 2: Monotonic Transformation of Meal Portions', fontdict={'fontsize': 15, 'fontweight': 'bold'}, pad=15)
axes[1].set_ylabel('Assigned Capacity Rank', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_xlabel('Original Textual Capacity', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_yticks(portion_ranks)
axes[1].grid(True, linestyle='--', alpha=0.5)

# Clean up top and right spines for both charts
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی عملیات (Strategic Implication):</b><br><br>"
    r"تصور کنید اگر ویژگیِ <code>Loyalty_Tier</code> را به روش متقارن (One-Hot) کدگذاری می‌کردیم، فاصلهٔ بین یک مشتری «برنز» تا «پلاتینیوم» "
    r"دقیقاً برابر با فاصلهٔ «برنز» تا «نقره‌ای» می‌شد! این یک فاجعهٔ تحلیلی است. <br><br>"
    r"اتخاذ رویکردِ <b>نگاشت ترتیبی دستی (Ordinal Mapping)</b>، از تخریب اطلاعاتِ ارزشمندِ سلسله‌مراتبی جلوگیری می‌نماید. "
    r"این تکنیک (که در گراف‌های ۱ و ۲ با شیب‌های صعودی و نردبانی اثبات شد)، الگوریتم‌های هوش مصنوعی (مثل Random Forest یا XGBoost) را مجهز می‌سازد "
    r"تا وابستگی‌های متوالی نظیر <i>«ارتقای سطح اشتراک»</i> یا <i>«افزایش حجم مصرفی»</i> را درک کرده و پیش‌بینی‌های مدل را دقیقاً بر مبنای منطق اولویت‌بندی، بهینه‌سازی کنند."
)
display_academic_text(insight_conclusion, "insight")
```
