  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: ترسیم نقشه دو بعدی سبک‌های موسیقی در «اسپاتیفای» با الگوریتم t-SNE 🗜️
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    سیستم هوش مصنوعی «اسپاتیفای» (Spotify) برای هر آهنگ ده‌ها ویژگی صوتی پیچیده (مانند <code>Danceability</code>، <code>Energy</code>، <code>Acousticness</code>، <code>Speechiness</code>، <code>Valence</code> و <code>Tempo</code>) استخراج می‌کند. ساختار و مرز بین سبک‌های مختلف موسیقی (کلاسیک، هپ‌هاپ، الکترونیک و راک) پیچیده و **غیرخطی (Nonlinear Manifold)** است.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>چرا روش‌های خطی مانند PCA در اینجا کافی نیستند؟</b><br>
    روش PCA تنها خطوط و صفحات راست را برای کاهش ابعاد فرض می‌کند. وقتی داده‌ها دارای منیفولدهای خمیده یا خوشه‌های تداخل‌یافته غیرخطی باشند، PCA آن‌ها را روی هم تصویر کرده و مرز بین سبک‌های موسیقی را کاملاً در هم می‌ریزد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>راه‌حل — کاهش ابعاد غیرخطی با t-SNE:</b><br>
    الگوریتم **t-SNE** (تعبیه همسایه تصادفی با توزیع t) یک روش غیرخطی فوق‌العاده قدرتمند است که بر **حفظ شباهت‌های محلی (Local Similarities)** تمرکز دارد. این الگوریتم تضمین می‌کند آهنگ‌هایی که در فضای چندبعدی همسایه نزدیک هم بوده‌اند، در نقشه ۲ بعدی جدید نیز حتماً کنار هم باقی بمانند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>نحوه کارکرد t-SNE به زبان ساده:</b><br>
    &nbsp;&nbsp;&nbsp;۱. فواصل بین تمام جفت‌آهنگ‌ها در فضای چندبعدی به **توزیع‌های احتمال تشابه** تبدیل می‌شود.<br>
    &nbsp;&nbsp;&nbsp;۲. در فضای ۲ بعدی جدید، نقاط طوری جابه‌جا می‌شوند که **واگرایی KL (KL Divergence)** بین توزیع‌های احتمال اولیه و جدید به حداقل برسد.<br>
    &nbsp;&nbsp;&nbsp;۳. استفاده از **توزیع t-Student** در فضای ۲ بعدی مانع از ازدحام نقاط (Crowding Problem) می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> t-SNE داده‌های ۶ بعدی صوت را به یک نقشه ۲ بعدی تبدیل می‌کند که در آن خوشه‌های سبک‌های مختلف موسیقی کاملاً مجزا و تفکیک‌شده دیده می‌شوند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: بصری‌سازی ۲ بعدی و تفکیک شفاف خوشه‌های غیرخطی بدون تخریب همسایگی‌های محلی
    </div>
  </p>

</div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.manifold import TSNE 

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})

plt.rcParams['figure.dpi'] = 135

np.random.seed(42)
```

```python
display_academic_text(
    "کاهش ابعاد غیرخطی: ترسیم منیفولدهای پیچیده صوتی با الگوریتم t-SNE 🎵", 
    "main_title"
)

display_academic_text(
    "بر مبنای اصول پیش‌پردازش کلان‌داده‌ها، ساختار ارتباط میان ویژگی‌های صوتیِ سبک‌های موسیقی (Spotify Features) ماهیتی کاملاً <b>غیرخطی (Non-Linear)</b> دارد. "
    "روش‌های خطی مانند PCA در تفکیک این فضاها ناکارآمدند. الگوریتم <b>t-SNE (t-Distributed Stochastic Neighbor Embedding)</b> با تبدیل فواصلِ چندبُعدی "
    "به توزیع‌های احتمالاتی و کمینه‌سازی واگرایی کولبک-لایبلر (KL Divergence)، <i>همسایگی‌های محلی (Local Structures)</i> را حفظ نموده و مرزهای درهم‌پیچیده را در قالبِ خوشه‌های کاملاً ایزوله در فضای ۲ بُعدی نمایش می‌دهد.", 
    "paragraph"
)

display_academic_text(
    r"تابع هزینه الگوریتم t-SNE (کمینه‌سازی واگرایی KL میان توزیع احتمالاتِ فضایِ اصلی $p_{j|i}$ و فضایِ تقلیل‌یافته دوبعدی $q_{j|i}$):", 
    "sub_title"
)

display_academic_text(
    r"$$Cost = D_{KL}(P \parallel Q) = \sum_i \sum_j p_{j|i} \log \left( \frac{p_{j|i}}{q_{j|i}} \right)$$", 
    "paragraph"
)


# ==========================================================
# SECTION 2: DATA GENERATION & t-SNE DIMENSIONALITY REDUCTION
# ==========================================================
display_academic_text(
    "مرحله نخست: شبیه‌سازی ویژگی‌های صوتی و اجرای تعبیه غیرخطی t-SNE", 
    "sub_title"
)

display_academic_text(
    "در این فاز، داده‌های صوتی برای ۴ سبک مختلف موسیقی (کلاسیک، الکترونیک/EDM، هیپ‌هاپ و راک) در یک <b>فضای ۶ بُعدی (6D Space)</b> شبیه‌سازی می‌شوند. "
    "سپس الگوریتم t-SNE با تنظیم پارامتر <code>perplexity=30</code> (تنظیم شعاع همسایگی محلی)، این فضای ۶ بعدی را به مختصاتِ ۲ بُعدی فشرده‌سازی کرده و خوشه‌های متمایز را استخراج می‌نماید.", 
    "paragraph"
)

# Step 1: Generate synthetic Spotify music features dataset (6 Dimensions, 4 Genres)
n_samples_per_genre = 80

# 1. Classical: Low Energy, High Acousticness, Low Speechiness
classical = np.random.normal(loc=[0.2, 0.9, 0.1, 0.3, 0.2, 80], scale=0.1, size=(n_samples_per_genre, 6))

# 2. EDM/Electronic: High Energy, High Danceability, Low Acousticness
edm = np.random.normal(loc=[0.9, 0.1, 0.8, 0.9, 0.7, 128], scale=0.1, size=(n_samples_per_genre, 6))

# 3. Hip-Hop: High Speechiness, High Danceability, Medium Energy
hiphop = np.random.normal(loc=[0.6, 0.3, 0.8, 0.8, 0.8, 95], scale=0.1, size=(n_samples_per_genre, 6))

# 4. Rock: High Energy, Low Speechiness, Low Acousticness
rock = np.random.normal(loc=[0.85, 0.15, 0.5, 0.5, 0.6, 120], scale=0.1, size=(n_samples_per_genre, 6))

# Combine datasets and assign genre labels
X_music = np.vstack([classical, edm, hiphop, rock])
genres = (
    ['Classical'] * n_samples_per_genre + 
    ['EDM'] * n_samples_per_genre + 
    ['Hip-Hop'] * n_samples_per_genre + 
    ['Rock'] * n_samples_per_genre
)

feature_cols = ['Energy', 'Acousticness', 'Danceability', 'Valence', 'Speechiness', 'Tempo']
df_music = pd.DataFrame(X_music, columns=feature_cols)
df_music['Genre'] = genres

# Step 2: Apply Non-linear Dimensionality Reduction using t-SNE
# random_state ensures full reproducibility of the stochastic gradient descent optimization
tsne = TSNE(
    n_components=2, 
    perplexity=30, 
    random_state=42, 
    max_iter=1000
)
X_tsne = tsne.fit_transform(df_music[feature_cols])

# Create DataFrame for t-SNE 2D projections
df_tsne = pd.DataFrame(X_tsne, columns=['t-SNE Dimension 1', 't-SNE Dimension 2'])
df_tsne['Genre'] = genres

# Create a genre distribution summary report table
df_genre_counts = pd.DataFrame({
    'Music Genre': ['Classical', 'EDM', 'Hip-Hop', 'Rock'],
    'Sample Count': [n_samples_per_genre] * 4,
    'Cluster Separation': ['High', 'High', 'High', 'High']
})

# Display preparation to prevent empty index columns in HTML
df_music_disp = df_music.head(5).reset_index(drop=True)
df_tsne_disp = df_tsne.head(5).reset_index(drop=True)
df_counts_disp = df_genre_counts.reset_index(drop=True)

# Render the initial 6D data sample, reduced 2D t-SNE coordinates, and distribution summary
display_side_by_side_datasets(
    [df_music_disp, df_tsne_disp, df_counts_disp],
    [
        "جدول ۱: داده‌های خام صوتی ۶ بُعدی (فضای Feature)", 
        "جدول ۲: مختصاتِ جدید در فضای ۲ بُعدی (t-SNE)",
        "جدول ۳: توزیع خوشه‌های سبک‌های موسیقی"
    ]
)

# ==========================================================
# SECTION 3: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی - نقشه‌برداری دو بُعدیِ سبک‌های موسیقی", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکیِ زیر <i>قدرت شگفت‌انگیزِ الگوریتم t-SNE</i> را در تفکیک خوشه‌های غیرخطی به نمایش می‌گذارد. هر نقطه نشان‌دهندهٔ یک قطعه موسیقی است که "
    "بر اساس ویژگی‌های صوتیِ خود در فضای دو بُعدی جایگذاری شده است. تفکیک کامل و عدم همپوشانی میان رنگ‌های متعلق به سبک‌های مختلف "
    "(کلاسیک در سمت راست، راک و EDM و هیپ‌هاپ در نواحی مستقل)، <b>اثبات عینیِ موفقیتِ تعبیهٔ غیرخطی (Non-linear Embedding) در حفظ همسایگی‌های محلی</b> است.", 
    "paragraph"
)

# Initialize a plot canvas optimized for scatter cluster visualization
plt.figure(figsize=(13, 8), dpi=135)

# Define an explicit academic color palette mapping to the music genres
# Integrating the requested Academic Blue palette alongside distinguishable cluster colors
genre_colors = {
    'Classical': '#1abc9c', # Turquoise
    'EDM': '#e67e22',       # Orange (Warm contrast for electronic)
    'Hip-Hop': '#2c3e50',   # Navy Blue
    'Rock': '#2980b9',      # Classic Blue
}

# Utilizing seaborn scatterplot for modern aesthetic compliance
sns.scatterplot(
    data=df_tsne,
    x='t-SNE Dimension 1',
    y='t-SNE Dimension 2',
    hue='Genre',
    palette=genre_colors,
    alpha=0.9,
    s=120, # Larger dots for visibility
    edgecolor='white',
    linewidth=1.2,
    zorder=5
)

plt.annotate(
    "High Energy Proximity:\nRock and EDM are clustered closer together\nbecause they share 'High Tempo'\nand 'High Energy' features in 6D space.",
    xy=(-35, -5), xytext=(-40, 15),
    fontsize=11, fontweight='bold', color='#2980b9',
    arrowprops=dict(arrowstyle="->", color="#2980b9", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#ecf0f1", ec="#2980b9", lw=1.5)
)

# Enhance plot labeling, titles, and structural configuration
plt.title('Figure 1: Non-Linear Manifold Projection (Spotify Music Genres Mapping via t-SNE)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
plt.xlabel('t-SNE Projected Dimension 1 (Abstract Axis)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
plt.ylabel('t-SNE Projected Dimension 2 (Abstract Axis)', fontdict={'fontsize': 13, 'fontweight': 'bold'})

plt.legend(
    title='Music Genre Clusters', 
    title_fontproperties={'weight':'bold'},
    loc='lower right', 
    frameon=True, 
    fontsize=12,
    edgecolor='#bdc3c7'
)

plt.grid(True, linestyle='--', alpha=0.5, zorder=0)

# Make ticks bold
plt.xticks(fontweight='bold', fontsize=11)
plt.yticks(fontweight='bold', fontsize=11)

# Clear top and right spines for a clean academic aesthetic
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی عملیات (Strategic Implication):</b><br>"
    r"در مواجهه با پایگاه‌های داده‌ای که دارای ساختارهای <b>منیفولدِ درهم‌پیچیده و غیرخطی</b> هستند (نظیر ویژگی‌های صوتیِ چندبُعدیِ اسپاتیفای)، "
    r"استفاده از روش‌های خطی سنتی مانند PCA ناکارآمد است و صرفاً به تولید یک ابرِ دادهٔ درهم‌ریخته می‌انجامد.<br>"
    r"الگوریتم <b>t-SNE</b> با تمرکز بر حفظ <i>همسایگی‌های محلی (Local Neighborhoods)</i> و به حداقل رساندنِ واگرایی آماریِ KL، موفق شد "
    r"بدون اینکه برچسبِ سبک‌ها را بداند (Unsupervised)، مرزهای معنایی میانِ دسته‌ها را (همان‌طور که در گراف مشخص است) با <b>تفکیک‌پذیریِ مطلق (Absolute Separation)</b> ترسیم نماید. "
    r"این امر پیش‌نیازِ بی‌بدیل در طراحی سیستم‌های توصیه‌گرِ پیشرفته (Recommender Systems) است."
)
display_academic_text(insight_conclusion, "insight")
```
