  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: ادغام پایگاه‌های داده ناهمگن در «اسپاتیفای» (Spotify) 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    پلتفرم «اسپاتیفای» (Spotify) داده‌های خود را در چند سیستم مجزا نگهداری می‌کند:
  </p>

  <ul style="text-align: right; margin: 0 0 10px; padding-right: 22px;">
    <li><b>منبع اول (پایگاه‌داده پخش موسیقی):</b> شامل لیست کاربران و رفتار شنیداری آن‌ها (تعداد ساعات پخش و سبک محبوب).</li>
    <li><b>منبع دوم (پایگاه‌داده مالی و حسابداری):</b> شامل تراکنش‌های پرداختی، نوع اشتراک (پریمیوم / رایگان) و آدرس صورت‌حساب.</li>
  </ul>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>چالش یکپارچه‌سازی داده‌ها (Data Integration):</b><br>
    تیم تحلیل الگوریتم برای پاسخ به سوالاتی مثل «آیا کاربران پریمیوم ساعات پخش بیشتری نسبت به کاربران رایگان دارند؟» نمی‌تواند از جدول‌های جداگانه استفاده کند. داده‌ها باید با استفاده از یک **کلید مشترک (Primary/Foreign Key)** مانند <code>User_ID</code> با هم ادغام (Merge / Join) شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>حل تعارضات و ناسازگاری‌ها در فرایند ادغام:</b><br>
    در زمان یکپارچه‌سازی داده‌ها، ۳ چالش اصلی رخ می‌دهد که باید حل شوند:
  </p>
  <ul style="text-align: right; margin: 0 0 10px; padding-right: 22px;">
    <li><b>تعارض نام ویژگی‌ها (Schema Matching):</b> ستون کشور در جدول اول <code>Country</code> و در جدول دوم <code>Region_Code</code> نام دارد که باید یکسان‌سازی شوند.</li>
    <li><b>عدم تطابق کلیدها (Data Redundancy):</b> ممکن است کاربری در جدول پخش وجود داشته باشد اما هنوز هیچ تراکنش مالی ثبت نکرده باشد؛ که نوع ادغام (Inner, Left, Outer Join) این موضوع را مدیریت می‌کند.</li>
    <li><b>تعارض مقادیر (Value Conflict):</b> مقادیر یکسان با فرمت‌های متفاوت (مثلاً ثبت تاریخ میلادی در یک منبع و شمسی در منبع دیگر).</li>
  </ul>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> یکپارچه‌سازی داده‌ها، دیدگاه ۳۶۰ درجه‌ای و جامع از تمام ابعاد رفتار کاربر در اختیار تیم داده‌کاوی اسپاتیفای قرار می‌دهد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: ترکیب منابع داده مجزا و حل تعارضات ساختاری بر اساس کلید مشترک شناسه کاربر
    </div>
  </p>

```python
display_academic_text(
    "پیش‌پردازش داده‌ها: یکپارچه‌سازی پایگاه‌های اطلاعاتی (Data Integration) و حل تعارضات شماتیک 🔗", 
    "main_title"
)

display_academic_text(
    "بر مبنای اصول معماری داده، اطلاعات مستقر در <b>سیلوهای مجزا (Data Silos)</b> فاقد ارزشِ تحلیلیِ ترکیبی هستند. "
    "در این سناریو، به منظور کشف وابستگی میان رفتار شنیداری (Streaming Activity) و وضعیت مالی (Financial Status) کاربران، "
    "دو پایگاه دادهٔ <i>ناهمگن</i> از طریق یک <b>کلید مشترک (Primary Key)</b> ادغام می‌شوند. چالش‌های تعارضِ نامِ ویژگی‌ها (Schema Conflict) "
    "و رکوردهای معلق (Dangling Tuples) در حین این عملیات مدیریت می‌گردند.", 
    "paragraph"
)

display_academic_text(
    "منطق جبر رابطه‌ای (Relational Algebra) برای عملیات Left Outer Join:", 
    "sub_title"
)

# Utilizing Raw Strings (r"") for LaTeX rendering of Relational Algebra
display_academic_text(
    r"$$T_{Integrated} = T_{Streaming} \ \text{LEFT OUTER JOIN} \ T_{Financial} \ \text{ON} \ (User\_ID)$$", 
    "paragraph"
)


# ==========================================================
# SECTION 2: DATA SOURCES INITIALIZATION AND INTEGRATION
# ==========================================================
display_academic_text(
    "مرحله نخست: بارگذاری منابع مجزا و اجرای عملیات یکپارچه‌سازی", 
    "sub_title"
)

display_academic_text(
    "عملیات <b>Left Outer Join</b> تضمین می‌کند که تمامی کاربرانِ پلتفرم پخش زنده حفظ شوند، حتی اگر سابقهٔ خریدی در سیستمِ مالی نداشته باشند (تولید مقادیر گمشده). "
    "همچنین کاربران موجود در سیستم مالی که هرگز موسیقی گوش نداده‌اند (مانند کاربر ۱۰۶) به صورت خودکار فیلتر می‌شوند.", 
    "paragraph"
)

# Step 1: Initialize Source Dataset 1 (Streaming Activity Logs)
data_streaming = {
    'User_ID': [101, 102, 103, 104, 105],
    'Monthly_Hours_Streamed': [45.5, 12.0, 88.2, 5.0, 62.1],
    'Favorite_Genre': ['Pop', 'Rock', 'Hip-Hop', 'Jazz', 'Pop'],
}
df_streaming = pd.DataFrame(data_streaming)

# Step 2: Initialize Source Dataset 2 (Financial Subscription Database)
# Notice Schema Conflict: 'Region_Code' instead of 'Country'
# Notice Dangling Tuples: User 104 is missing, User 106 is exclusively in this table
data_financial = {
    'User_ID': [101, 102, 103, 105, 106],  
    'Subscription_Plan': ['Premium', 'Free', 'Premium', 'Premium', 'Free'],
    'Monthly_Fee_USD': [9.99, 0.00, 9.99, 9.99, 0.00],
    'Region_Code': ['US', 'UK', 'US', 'CA', 'DE'],  
}
df_financial = pd.DataFrame(data_financial)

# Step 3: Perform Data Integration (Left Outer Join on 'User_ID')
df_integrated = pd.merge(
    df_streaming, 
    df_financial, 
    on='User_ID', 
    how='left'
)

# Step 4: Resolve Schema Conflicts (Attribute Name Standardization)
df_integrated = df_integrated.rename(columns={'Region_Code': 'Country'})

# Step 5: Handle Missing Values derived from the Join operation (User 104 lacks financial data)
df_integrated['Subscription_Plan'] = df_integrated['Subscription_Plan'].fillna('Unknown / Unregistered')
df_integrated['Monthly_Fee_USD'] = df_integrated['Monthly_Fee_USD'].fillna(0.00)

# Display preparation to prevent empty index columns in HTML
df_stream_disp = df_streaming.reset_index(drop=True)
df_fin_disp = df_financial.reset_index(drop=True)
df_int_disp = df_integrated.reset_index(drop=True)

# Display the independent sources
display_side_by_side_datasets(
    [df_stream_disp, df_fin_disp],
    [
        "منبع اول: پایگاه داده رفتار پخش (Streaming Logs)", 
        "منبع دوم: پایگاه داده تراکنش‌ها (Financial System)"
    ]
)

# Display the finalized Integrated Dataset
display_side_by_side_datasets(
    [df_int_disp],
    [
        "مخزن داده یکپارچه‌شده (Unified Data Warehouse) پس از اجرای JOIN و پاک‌سازی"
    ]
)

# ==========================================================
# SECTION 3: ANALYTICAL VISUALIZATION (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: استخراج دانش تحلیلی از پایگاه داده یکپارچه‌شده", 
    "sub_title"
)

display_academic_text(
    "هدف نهایی از یکپارچه‌سازی داده‌ها، پاسخ به <b>سوالاتِ ترکیبیِ کسب‌وکار</b> است. بوم گرافیکی زیر با بهره‌گیری از جدولِ ادغام‌شده، "
    "تأثیر نوع اشتراکِ مالی (از جدول دوم) را بر میزان ساعاتِ پخش موسیقی (از جدول اول) تحلیل می‌نماید. این نمودار ثابت می‌کند که ایجاد یک دیدگاه ۳۶۰ درجه، "
    "چگونه الگوهای پنهان را نمایان می‌سازد.", 
    "paragraph"
)

# Initialize a plot canvas optimized for business analytics
plt.figure(figsize=(14, 6), dpi=135)

# Plotting the cross-domain relationship: Streaming Hours vs. Subscription Plan
sns.barplot(
    data=df_integrated, 
    x='Subscription_Plan', 
    y='Monthly_Hours_Streamed', 
    hue='Subscription_Plan',
    legend=False,
    palette=['#2980b9', '#1abc9c', '#7f8c8d'], # Academic Blues & Grays
    edgecolor='#2c3e50', 
    linewidth=2.0,
    errorbar=None # Simplified for direct mean representation
)

# Intelligent Annotations to explain the Cross-Domain Insight
plt.annotate(
    "The Magic of Data Integration:\nBy combining the Financial Table with the Streaming Table,\nwe immediately discovered that Premium users listen\nalmost 5x more than Free/Unregistered users!",
    xy=(0.0, 65.2), xytext=(0.5, 75),
    fontsize=11, fontweight='bold', color='#2c3e50',
    arrowprops=dict(arrowstyle="->", color="#34495e", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.5", fc="#ecf0f1", ec="#bdc3c7", lw=1.5)
)

# Enhance plot labeling, titles, and structural configuration
plt.title('Figure 1: Cross-Domain Analytics (Financial Status vs. Streaming Engagement)', fontdict={'fontsize': 15, 'fontweight': 'bold'}, pad=15)
plt.xlabel('Financial Subscription Tier (From Table 2)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
plt.ylabel('Average Monthly Streamed Hours (From Table 1)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
plt.grid(axis='y', linestyle='--', alpha=0.5, zorder=0)

# Make tick labels bold
plt.xticks(fontweight='bold', fontsize=12)

# Add exact value annotations on top of the bars for precise reporting
ax = plt.gca()
for p in ax.patches:
    if p.get_height() > 0:
        ax.annotate(
            f"{p.get_height():.1f} Hrs", 
            (p.get_x() + p.get_width() / 2., p.get_height()), 
            ha='center', va='center', xytext=(0, 12), textcoords='offset points',
            fontweight='bold', fontsize=12, color='#2c3e50'
        )

# Clear top and right spines for a clean academic aesthetic
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی عملیات (Strategic Implication):</b><br>"
    r"یکپارچه‌سازی منابعِ ناهمگن، تنها چسباندن جداول به یکدیگر نیست؛ بلکه نیازمند استراتژی‌های قطعی جهت <b>حل تعارضات ساختاری (Schema Matching)</b> (مثل تبدیل <code>Region_Code</code> به <code>Country</code>) "
    r"و مدیریتِ <b>خلأهای اطلاعاتیِ</b> ناشی از رکورد‌های معلق (Dangling Tuples) است.<br>"
    r"ادغام موفقیت‌آمیز این داده‌ها، به تیم تحلیلگر اجازه داد تا همبستگیِ پنهان میانِ <i>«پرداخت حق اشتراک در سیستم مالی»</i> و <i>«افزایش نرخ تعامل (Engagement) در سرور استریم»</i> را با دقت بالا استخراج نماید. "
    r"این بینش، بدونِ عملیات JOIN هرگز قابل کشف نبود!"
)
display_academic_text(insight_conclusion, "insight")
```

<li><b>ادغام بر اساس کلید مشترک:</b> با استفاده از تابع <b>pd.merge()</b> و تعیین <code>on='User_ID'</code>، دو منبع داده مجزا بر اساس شناسه مشترک کاربر ترکیب شدند.</li>
<li><b>مدیریت تعارض ساختار (Schema Matching):</b> با استفاده از متد <b>rename()</b>، نام ستون <b>'Region_Code'</b> به <b>'Country'</b> تغییر یافت تا با ساختار کل مجموعه داده یکنواخت شود.</li>
<li><b>مدیریت رکوردهای مفقود در ادغام (Left Join):</b> از الگوی <code>how='left'</code> استفاده شد تا تمام کاربران فعال پخش حفظ شوند و خانه‌های خالی ناشی از عدم وجود سابقه مالی در جدول دوم، با متد <b>fillna()</b> مقداردهی استاندارد گردند.</li>
<li><b>نتیجه خروجی:</b> یک دیتافریم یکپارچه، منسجم و جامع تولید شد که هم‌زمان اطلاعات رفتاری و مالی کاربران را برای تحلیل‌های هوش مصنوعی ارائه می‌دهد.</li>
