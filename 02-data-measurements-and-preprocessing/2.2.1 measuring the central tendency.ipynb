  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: تحلیل زمان ارسال و رفتار خریداران در «اسنپ‌فود» 📏
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در «اسنپ‌فود»، روزانه صدها هزار سفارش ثبت می‌شود. تیم داده‌کاوی برای پاسخ به سوالاتی مثل «زمان دقیق ارسال غذا به مشتری چقدر است؟» یا «مردم بیشتر چه ساعتی غذا سفارش می‌دهند؟» نیاز دارد میلیون‌ها داده را در چند عدد خلاصه کند. اینجاست که <b>معیارهای گرایش به مرکز</b> وارد میدان می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۱. میانگین (Mean) — سنجش عملکرد کلی در شرایط عادی</b><br>
    تیم الگوریتم، زمان ارسال ۵ سفارش یک پیک را ثبت می‌کند: ۲۵، ۳۰، ۲۸، ۳۲ و ۳۵ دقیقه.<br>
    حاصل‌جمع (۱۵۰) تقسیم بر تعداد (۵) برابر با <b>۳۰ دقیقه</b> می‌شود. میانگین برای داده‌های متقارن و بدون نوسان شدید، بهترین خلاصه عددی است.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۲. میانه (Median) — نجات از داده‌های پرت (Outliers) و ترافیک سنگین</b><br>
    فرض کنید ناگهان باران می‌بارد و یکی از پیک‌ها در ترافیک می‌ماند! زمان‌های جدید: ۲۰، ۲۲، ۲۵، ۲۸ و <b>۱۵۰ دقیقه</b>.<br>
    اگر میانگین بگیریم، عدد **۴۹ دقیقه** به دست می‌آید که کاملاً نادرست است و مشتری را نگران می‌کند! اما اگر داده‌ها را مرتب کنیم، عدد وسط یعنی <b>۲۵ دقیقه (میانه)</b> نشان‌دهنده زمان واقعی انتظار اکثر مشتریان است. میانه در برابر داده‌های پرت کاملاً مقاوم است.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۳. نما (Mode) — پیدا کردن محبوب‌ترین‌ها</b><br>
    اسنپ‌فود می‌خواهد بداند پرتقاضاترین نوع غذا در روزهای تعطیل چیست؟ مقادیر ثبت‌شده: [پیتزا، چلوکباب، پیتزا، سالاد، پیتزا، پاستا].<br>
    مقدار **«پیتزا»** با ۳ بار تکرار، <b>نمای (Mode)</b> داده‌هاست. نما تنها معیاری است که برای متغیرهای غیرعددی و اسمی (Nominal) هم استفاده می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> اسنپ‌فود برای تخمین زمان ارسال به مشتری از «میانه» استفاده می‌کند تا اثر ترافیک‌های ناگهانی را خنثی کند، و برای پیشنهادهای ویژه اپلیکیشن از «نما» بهره می‌برد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: انتخاب هوشمندانه معیار متناسب با نوع داده و حضور داده‌های پرت
    </div>
  </p>

```python
display_academic_text(
    "آمار توصیفی: ارزیابی معیارهای گرایش به مرکز در حضور داده‌های پرت (Outliers) 📉", 
    "main_title"
)

display_academic_text(
    "انتخاب معیار مرکزی به توزیع آماری بستگی دارد. در یک شبکه تحویل غذا، وقوع حوادث غیرمترقبه نظیر ترافیک یا شرایط جوی نامساعد، "
    "منجر به تولید <b>داده‌های پرت (Outliers)</b> می‌گردد. در این سناریو، مقاومت شاخص‌های <i>میانگین (Mean)</i> و <i>میانه (Median)</i> در مواجهه با نوسانات شدید زمانی، "
    "و همچنین کاربرد شاخص <i>نما (Mode)</i> برای تحلیل متغیرهای طبقه‌بندی‌شده مورد ارزیابی قرار می‌گیرد.", 
    "paragraph"
)

# ==========================================================
# SECTION 2: DATA INITIALIZATION & STATISTICAL COMPUTATION
# ==========================================================
display_academic_text(
    "مرحله نخست: بارگذاری داده‌های عملیاتی و استخراج شاخص‌های مرکزی", 
    "sub_title"
)

display_academic_text(
    "در این فاز پایگاه داده زمان‌های ارسال پیک استخراج می‌شود که حاوی یک <b>دادهٔ ناهنجار (۱۵۰ دقیقه)</b> است. سپس، توابع آماری پایه بر روی "
    "بردار ویژگی‌ها اعمال شده و خروجی‌ها در یک ماتریس تجمیعی جهت مقایسه در کنار داده‌های خام به نمایش در می‌آیند.", 
    "paragraph"
)

# Step 1: Initialize the mock database for delivery logistics.
# Note the intentional injection of an extreme outlier (150 minutes) to simulate severe traffic.
raw_data = {
    'Order_ID': [101, 102, 103, 104, 105],
    'Delivery_Time_Min': [20, 22, 25, 28, 150], 
    'Food_Category': ['Pizza', 'Kebab', 'Pizza', 'Burger', 'Pizza'] 
}
df_raw = pd.DataFrame(raw_data)

# Step 2: Compute the Arithmetic Mean (Highly sensitive to extreme values / outliers)
calculated_mean = df_raw['Delivery_Time_Min'].mean()

# Step 3: Compute the Median (Robust statistical measure, represents the exact 50th percentile)
calculated_median = df_raw['Delivery_Time_Min'].median()

# Step 4: Compute the Mode (The most frequent categorical occurrence)
# The [0] index extracts the first mode string from the returned pandas Series
calculated_mode = df_raw['Food_Category'].mode()[0]

# Step 5: Construct a summary dataframe to present the calculated metrics professionally
summary_data = {
    'Statistical Measure': [
        'Mean (Average)', 
        'Median (50th Percentile)', 
        'Mode (Highest Frequency)'
    ],
    'Computed Value': [
        f"{calculated_mean:.1f} Minutes", 
        f"{calculated_median:.1f} Minutes", 
        calculated_mode
    ],
    'Algorithmic Behavior': [
        'Heavily skewed by outlier (Misleading)', 
        'Robust to extreme anomalies (Accurate)', 
        'Identifies categorical peak demand'
    ]
}
df_summary = pd.DataFrame(summary_data)

# Render the raw transaction data and the extracted statistical insights side-by-side
display_side_by_side_datasets(
    [df_raw, df_summary],
    [
        "جدول ۱: رکوردهای ثبت‌شده زمان ارسال (حاوی داده پرت)", 
        "جدول ۲: تحلیل رفتار معیارهای گرایش به مرکز"
    ]
)

# ==========================================================
# SECTION 3: ANALYTICAL VISUALIZATION (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلیِ سوءگیری میانگین در برابر مقاومت میانه", 
    "sub_title"
)

display_academic_text(
    "نمودار سمت چپ به وضوح <b>انحراف شدید خط میانگین</b> به سمت دادهٔ پرت را به تصویر می‌کشد، در حالی که <b>خط میانه پایداری خود را حفظ کرده</b> و "
    "در هستهٔ اصلی توزیع داده‌ها باقی مانده است. نمودار سمت راست نیز بازنمایی گرافیکی استخراج شاخص نما (Mode) برای دسته‌بندی‌های اسمی است.", 
    "paragraph"
)

# Initialize a wide dual-axis canvas for visual comparative analytics
fig, axes = plt.subplots(1, 2, figsize=(19, 6.5), dpi=135)

# --- SUBPLOT 1: Continuous Data (Mean vs Median) ---
# A stripplot avoids overlapping dots, making the outlier visually distinct.
sns.stripplot(
    y=df_raw['Delivery_Time_Min'], 
    ax=axes[0], 
    color='#2c3e50', # Navy Blue
    size=12, 
    jitter=0.05, 
    edgecolor='white', 
    linewidth=1.5
)

# Overlay theoretical center lines: Mean vs Median
axes[0].axhline(
    calculated_mean, 
    color='#e74c3c', # Red for the misleading/skewed metric
    linestyle='--', 
    linewidth=3.0, 
    label=f'Mean (Skewed): {calculated_mean:.1f} Min'
)

axes[0].axhline(
    calculated_median, 
    color='#1abc9c', # Turquoise for the robust/accurate metric
    linestyle='-', 
    linewidth=3.0, 
    label=f'Median (Robust): {calculated_median:.1f} Min'
)

# Intelligent Annotations
axes[0].annotate(
    "The Outlier Pull:\nA single 150-minute delivery forcefully pulls\nthe mathematical average upward, creating\na false representation of normal operations.",
    xy=(0.0, calculated_mean), xytext=(0.15, calculated_mean + 15),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].annotate(
    "Robustness:\nThe median ignores the extreme value\nand correctly identifies that 50% of orders\narrive in 25 minutes or less.",
    xy=(0.0, calculated_median), xytext=(0.15, calculated_median - 20),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[0].set_title('Figure 1: Impact of Outliers on Central Tendency Measures', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[0].set_ylabel('Recorded Delivery Time (Minutes)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].legend(loc='center right', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=11)
axes[0].grid(axis='y', linestyle='--', alpha=0.5)

# --- SUBPLOT 2: Categorical Data (Mode) ---
# Countplot automatically aggregates and plots categorical distributions.
sns.countplot(
    x=df_raw['Food_Category'], 
    ax=axes[1], 
    palette=['#1abc9c', '#2980b9', '#2c3e50'], # Academic Blue Gradient
    edgecolor='#2c3e50',
    linewidth=1.5
)

# Formatting Subplot 2
axes[1].set_title(f'Figure 2: Nominal Attribute Distribution (Mode = {calculated_mode})', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[1].set_xlabel('Nominal Food Category', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_ylabel('Order Frequency Count', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].grid(axis='y', linestyle='--', alpha=0.5)

# Add data labels on top of bars
for p in axes[1].patches:
    axes[1].annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height()),
                     ha='center', va='center', fontsize=12, fontweight='bold', color='#2c3e50', xytext=(0, 10), textcoords='offset points')

# Remove the top and right spines from both axes for a cleaner academic look
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی عملیات (Strategic Implication):</b><br><br>"
    r"همان‌طور که در گرافِ اول ثابت شد، اتکا به <b>میانگین حسابی (Mean)</b> در پایگاه‌های داده‌ای که مستعدِ نوساناتِ محیطی و داده‌های پرت (مانند تأخیر ۱۵۰ دقیقه‌ای پیک) هستند، "
    r"به برآوردهای به‌شدت گمراه‌کننده منجر می‌شود. اگر سیستم به مشتری بگوید <i>«میانگین زمان ارسال ما ۴۹ دقیقه است»</i>، در واقع عملکردِ عالیِ خود در ۸۰٪ سفارش‌ها (زیر ۳۰ دقیقه) را زیر سؤال برده است! <br><br>"
    r"جایگزینی این شاخص با <b>میانه (Median = ۲۵ دقیقه)</b>، یک مکانیزم مقاوم‌سازیِ قدرتمند در برابر اختلالات ایجاد کرده و به تخمینِ واقعی‌تری از تجربهٔ بدنهٔ اصلیِ کاربران منتهی می‌گردد."
)
display_academic_text(insight_conclusion, "insight")
```
