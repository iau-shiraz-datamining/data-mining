<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: سیستم توصیه‌گر و خوشه‌بندی مشتریان در «آمازون» 🛒
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    سیستم هوش مصنوعی «آمازون» برای اینکه بفهمد کدام کاربران رفتاری شبیه به هم دارند و چه محصولاتی را باید به آن‌ها پیشنهاد دهد، داده‌های خریداران را در دو ساختار ریاضی کاملاً متفاوت اما مکمل هم ذخیره می‌کند: <b>ماتریس داده (Data Matrix)</b> و <b>ماتریس عدم تشابه (Dissimilarity Matrix)</b>.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۱. ماتریس داده (Data Matrix) — ساختار ساخت‌یافته‌ی اولیه ($n \times p$)</b><br>
    در این جدول، هر سطر نماینده یک **مشاهده/مشتری ($n$)** و هر ستون نماینده یک **ویژگی/صفت ($p$)** است:<br>
    &nbsp;&nbsp;&nbsp;• سطرها: خریدار A، خریدار B، خریدار C.<br>
    &nbsp;&nbsp;&nbsp;• ستون‌ها: سن، تعداد خرید ماهانه، میانگین مبلغ هر سفارش (دلار).<br>
    این ماتریس جدول اصلی داده‌هاست که مقدار خام هر ویژگی برای هر فرد را نگه می‌دارد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۲. ماتریس عدم تشابه (Dissimilarity Matrix) — محاسبه تفاوت‌های زوجی ($n \times n$)</b><br>
    برای اینکه آمازون بفهمد خریدار A چقدر با خریدار B تفاوت دارد، تمام ویژگی‌های آن‌ها را با یک فرمول فاصله (مثل فاصله اقلیدسی) مقایسه می‌کند و یک ماتریس مربعی متقارن می‌سازد:<br>
    &nbsp;&nbsp;&nbsp;• ابعاد ماتریس برابر با $n \times n$ است (مشتری در برابر مشتری).<br>
    &nbsp;&nbsp;&nbsp;• **قطر اصلی همیشه صفر است**؛ چون فاصله هر خریدار با خودش صفر است ($d(A,A) = 0$).<br>
    &nbsp;&nbsp;&nbsp;• هرچه عدد داخل ماتریس کوچک‌تر باشد، نشان‌دهنده **تشابه بیشتر** دو خریدار است.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۳. کاربرد عملی در الگوریتم‌های داده‌کاوی</b><br>
    الگوریتم‌های خوشه‌بندی (Clustering) برخلاف الگوریتم‌های ساده، مستقیماً از **ماتریس عدم تشابه** استفاده می‌کنند. وقتی آمازون می‌بیند فاصله بین خریدار A و خریدار C بسیار کوچک است (مثلاً ۰.۱۲)، خریدار C را هم‌گروه خریدار A قرار داده و محصولاتی که A خریده را به C پیشنهاد می‌کند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> ماتریس داده مشخصات خام هر عنصر را ذخیره می‌کند، اما ماتریس عدم تشابه تفاوت‌های نسبی عناصر را با هم می‌سنجد تا ورود به الگوریتم‌های هوش مصنوعی تسهیل شود.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: تبدیل جدول ویژگی‌های خام ($n \times p$) به جدول فاصله جفت‌به‌جفت عناصر ($n \times n$)
    </div>
  </p>

</div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.spatial.distance import pdist, squareform
from IPython.display import display, HTML

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})

plt.rcParams['figure.dpi'] = 135

def display_academic_text(text, type="normal"):
    """Renders text in a strict academic format."""
    styles = {
        "main_title": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 24px; "
            "font-weight: bold; text-align: center; border-bottom: 2px solid #34495e; "
            "padding-bottom: 15px; margin: 30px 0;"
        ),
        "sub_title": (
            "font-family: Tahoma, sans-serif; color: #34495e; font-size: 19px; "
            "font-weight: bold; border-right: 5px solid #2980b9; padding-right: 15px; "
            "margin: 40px 0px 20px 0px; background-color: #ecf0f1; padding: 12px; "
            "border-radius: 4px;"
        ),
        "paragraph": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 16px; "
            "margin-bottom: 20px; line-height: 1.8; text-align: justify;"
        ),
        "insight": (
            "font-family: Tahoma, sans-serif; color: #16a085; font-size: 16px; "
            "font-weight: bold; background-color: #e8f8f5; padding: 20px; "
            "border-left: 5px solid #1abc9c; margin: 30px 0; line-height: 1.8; "
            "text-align: justify;"
        )
    }
    display(HTML(f"<div dir='rtl' style='text-align: right; {styles[type]}'>{text}</div>"))

def get_protected_css():
    """Returns essential CSS for robust and clean table formatting."""
    return """
    <style>
    .academic-table-wrapper { 
        direction: ltr !important; 
        overflow-x: auto !important; 
        width: 100% !important; 
        background-color: #fff; 
        border: 1px solid #bdc3c7; 
        border-radius: 4px; 
        box-shadow: 0 3px 6px rgba(0,0,0,0.08); 
    }
    table.strict-tbl { 
        border-collapse: collapse !important; 
        width: 100% !important; 
        margin: 0 !important; 
        font-family: inherit !important;
        font-size: 14px !important; 
        table-layout: auto !important; 
        border: none !important; 
    }
    table.strict-tbl thead th { 
        background-color: #2c3e50 !important; 
        color: #ffffff !important; 
        padding: 12px 18px !important; 
        text-align: center !important; 
        border: 1px solid #34495e !important; 
        white-space: nowrap !important; 
        font-weight: bold !important; 
    }
    table.strict-tbl tbody td { 
        padding: 10px 18px !important; 
        text-align: center !important; 
        border: 1px solid #bdc3c7 !important; 
        white-space: nowrap !important; 
        color: #2c3e50 !important; 
    }
    table.strict-tbl tbody tr:nth-child(even) td { 
        background-color: #f8f9fa !important; 
    }
    table.strict-tbl tbody tr:hover td { 
        background-color: #e8ecef !important; 
    }
    </style>
    """

def display_side_by_side_datasets(dfs, titles):
    """Renders multiple pandas DataFrames side-by-side gracefully."""
    col_html_list = []
    for i, (df, title) in enumerate(zip(dfs, titles)):
        table_html = df.to_html(index=False, classes=['strict-tbl'], border=0, float_format=lambda x: f"{x:.1f}")
        bg_color = "#2c3e50" 
        col_html = f"""
        <div class="flex-column" style="flex: 1 1 0; min-width: 0; display: flex; flex-direction: column;">
            <div style="background-color: {bg_color}; color: white; padding: 12px; 
                        text-align: center; font-family: Tahoma, sans-serif; 
                        font-weight: bold; font-size: 15px; border-radius: 4px 4px 0 0;">
                {title}
            </div>
            <div class="academic-table-wrapper" style="margin-top: 0; border-top: none;">
                {table_html}
            </div>
        </div>
        """
        col_html_list.append(col_html)
    
    columns_joined = "".join(col_html_list)
    html = f"""
    <div style="display: flex; flex-direction: row; justify-content: space-between; 
                gap: 20px; width: 100%; max-width: 100%; margin: 20px 0 35px 0;
                box-sizing: border-box;">
        {columns_joined}
    </div>
    """
    display(HTML(get_protected_css() + html))

np.random.seed(42)
```

```python
display_academic_text(
    "خوشه‌بندی الگوهای رفتاری: گذر از ماتریس داده به ماتریس عدم تشابه (Dissimilarity Matrix) 🎯", 
    "main_title"
)

display_academic_text(
    "بر اساس اصول داده‌کاوی، الگوریتم‌های خوشه‌بندی به جای کار مستقیم بر روی مقادیرِ خامِ ویژگی‌ها، نیازمند ارزیابی فواصل جفت‌به‌جفت (Pairwise Distances) عناصر هستند. "
    "در این سناریو، سامانهٔ هوش مصنوعیِ یک پلتفرم فروشگاهی، ابتدا رکوردهای تراکنشی را در یک <b>«ماتریس داده دووجهی (Two-mode Data Matrix)»</b> تجمیع نموده "
    "و سپس آن را با استفاده از فواصل اقلیدسی به یک <b>«ماتریس عدم تشابه متقارن (One-mode Dissimilarity Matrix)»</b> تبدیل می‌نماید تا میزان نزدیکی رفتار کاربران را اندازه‌گیری کند.", 
    "paragraph"
)

# ==========================================================
# SECTION 2: MATRIX COMPUTATION AND TRANSFORMATION
# ==========================================================
display_academic_text(
    "مرحله نخست: مقداردهی ماتریس شیء-ویژگی و محاسبه فواصل اقلیدسی", 
    "sub_title"
)

display_academic_text(
    "در این فاز، <i>ماتریس داده</i> به عنوان نمایشگر خام ویژگی‌های مستقل تشکیل می‌گردد. سپس الگوریتمِ فاصله اقلیدسی (Euclidean Distance) بر روی تمامی بردارهای کاربری "
    "اعمال شده و خروجی در قالب یک ماتریس متقارن (Symmetric Matrix) با قطر اصلی صفر بازتولید می‌شود. این دو ساختار در جداول موازی زیر قابل مقایسه می‌باشند.", 
    "paragraph"
)

# Step 1: Initialize the two-mode Data Matrix (n objects x p attributes)
# In this scenario: n = 4 customers, p = 3 numerical attributes
data = {
    'Customer_ID': ['User_A', 'User_B', 'User_C', 'User_D'],
    'Age': [25, 26, 45, 50],
    'Monthly_Purchases': [10, 11, 2, 1],
    'Avg_Spend_USD': [120, 130, 450, 500]
}

# Set Customer_ID as the explicit index
data_matrix = pd.DataFrame(data).set_index('Customer_ID')

# Step 2: Compute Pairwise Euclidean Distance to extract the Dissimilarity Matrix.
# The pdist function calculates distances strictly between all n(n-1)/2 unique pairs.
distances_condensed = pdist(data_matrix.values, metric='euclidean')

# Step 3: Convert the condensed distance array into a symmetric square matrix (n x n).
# The squareform function ensures d(i,j) = d(j,i) and the main diagonal d(i,i) = 0.
dissimilarity_array = squareform(distances_condensed)

# Convert the derived array back into a structured pandas DataFrame
dissimilarity_matrix = pd.DataFrame(
    dissimilarity_array,
    index=data_matrix.index,  
    columns=data_matrix.index 
)

# Prepare cleanly formatted DataFrames for display (resetting indices to show names properly)
df_data_disp = data_matrix.reset_index()
df_dissim_disp = dissimilarity_matrix.reset_index().rename(columns={'Customer_ID': 'Target / Reference'})

# Render the two mathematically distinct structures side-by-side
display_side_by_side_datasets(
    [df_data_disp, df_dissim_disp],
    [
        "جدول ۱: ماتریس داده دووجهی (n=4 , p=3)", 
        "جدول ۲: ماتریس عدم تشابه یک‌وجهی متقارن (n=4 , n=4)"
    ]
)

# ==========================================================
# SECTION 3: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی فواصل و کشف بصری خوشه‌ها", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکیِ زیر، دلیلِ شکل‌گیریِ ماتریس عدم تشابه را اثبات می‌کند. در نمودار پراکندگی (سمت چپ)، جایگاه واقعی کاربران در فضای ویژگی‌ها مشخص است؛ "
    "و در نقشه حرارتی (سمت راست)، توپولوژیِ فواصلِ استخراج‌شده به تصویر کشیده می‌شود. قطر اصلی با <b>رنگ سرمه‌ای تیره (فاصله صفر)</b>، نشان‌دهندهٔ تطابقِ مطلقِ "
    "هر کاربر با خودش است. بلوک‌های تیره میان (User_A و User_B) اثبات‌کنندهٔ شباهت بالای این دو گروه می‌باشد.", 
    "paragraph"
)

# Initialize a dual-axis canvas for visual comparative analytics
fig, axes = plt.subplots(1, 2, figsize=(18, 6.5), dpi=135)

# --- SUBPLOT 1: Raw Data Scatter Representation ---
# Visualizing 'Age' vs 'Avg Spend' to show WHY the users cluster the way they do
sns.scatterplot(
    data=df_data_disp, 
    x='Age', 
    y='Avg_Spend_USD', 
    ax=axes[0],
    color='#1abc9c', # Turquoise
    s=400, 
    edgecolor='#2c3e50', 
    linewidth=2.5,
    zorder=5
)

# Annotate each point with the Customer ID
for idx, row in df_data_disp.iterrows():
    axes[0].text(row['Age'], row['Avg_Spend_USD'] + 20, row['Customer_ID'], 
                 ha='center', va='bottom', fontweight='bold', fontsize=12, color='#2c3e50')

# Draw circles around apparent clusters
from matplotlib.patches import Ellipse
axes[0].add_patch(Ellipse((25.5, 125), width=5, height=60, color='#2980b9', alpha=0.2, zorder=1))
axes[0].add_patch(Ellipse((47.5, 475), width=9, height=90, color='#e74c3c', alpha=0.2, zorder=1))

axes[0].set_title('Figure 1: Customer Position in Raw Feature Space', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[0].set_xlabel('Customer Age', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_ylabel('Average Monthly Spend (USD)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_xlim(20, 55)
axes[0].set_ylim(50, 550)
axes[0].grid(True, linestyle='--', alpha=0.5, zorder=0)

# --- SUBPLOT 2: Dissimilarity Matrix Heatmap ---
# Using an Academic Blue reversed colormap (Blues_r).
# Dark Blue = 0 distance (High Similarity/Match). White/Light Blue = High Distance (Dissimilarity).
sns.heatmap(
    dissimilarity_matrix, 
    annot=True, 
    cmap='Blues_r', 
    fmt='.1f', 
    linewidths=2.0, 
    linecolor='white',
    ax=axes[1],
    cbar_kws={'label': 'Calculated Euclidean Distance'},
    annot_kws={"size": 14, "weight": "bold"}
)

axes[1].set_title('Figure 2: Topology of Customer Dissimilarity Matrix', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[1].set_xlabel('Target Customer (Object j)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_ylabel('Reference Customer (Object i)', fontdict={'fontsize': 13, 'fontweight': 'bold'})

axes[1].set_xticklabels(axes[1].get_xticklabels(), fontweight='bold', fontsize=11)
axes[1].set_yticklabels(axes[1].get_yticklabels(), fontweight='bold', fontsize=11, rotation=0, va='center')

# Clean up top and right spines
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی عملیات (Strategic Implication):</b><br><br>"
    r"همان‌طور که در گراف‌ها مشاهده می‌کنید، اگرچه <b>ماتریس داده (جدول ۱)</b> حاوی اطلاعات مطلق و خام هر کاربر است، "
    r"اما الگوریتم‌های فاصله‌محور و خوشه‌بندی (نظیر <i>K-Means</i> یا <i>Hierarchical Clustering</i>) مستقیماً با <b>ماتریس عدم تشابه (جدول ۲)</b> کار می‌کنند. <br><br>"
    r"تحلیلِ گرافِ سمتِ راست نشان می‌دهد که مقادیرِ صِفر در <b>قطر اصلی (Main Diagonal)</b> و همچنین تقارنِ فواصل حولِ آن، این اطمینان ریاضیاتی را به سیستمِ هوشمند می‌دهد "
    r"که فاصلهٔ کاربر A تا B، با فاصلهٔ B تا A یکسان است. این مکانیزم، زیربنای اصلیِ سیستم‌های توصیه‌گر (Recommender Systems) است تا با شناسایی کوچک‌ترین فواصلِ ریاضی (نظیر فاصله ۱۰.۱ بین A و B)، محصولاتِ مشابه را میان این خوشه‌های همسایه به اشتراک بگذارند."
)
display_academic_text(insight_conclusion, "insight")
```

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیح مثال کد پایتون</span> 💻

<p style="font-size:16pt; text-align:right;">
در این مثال:
<ul>
<li>ما یک <b>ماتریس داده (Data Matrix)</b> اولیه برای ۴ کاربر آمازون ساختیم که دارای ۳ ویژگی عددی (سن، تعداد خرید ماهانه و میانگین مبلغ خرید) است. این ماتریس دارای ابعاد $4 \times 3$ می‌باشد.</li>
<li>از تابع <b>pdist()</b> از کتابخانه <code>scipy</code> با معیار فاصله اقلیدسی (Euclidean) استفاده کردیم تا فاصله تمام جفت‌کاربران ممکن را محاسبه کنیم.</li>
<li>با استفاده از تابع <b>squareform()</b> آرایه فواصل را به یک <b>ماتریس عدم تشابه (Dissimilarity Matrix)</b> مربعی با ابعاد $4 \times 4$ تبدیل کردیم.</li>
<li>همان‌طور که در خروجی مشخص است:
  <ul>
    <li>تمام عناصر روی <b>قطر اصلی صفر هستند</b>؛ زیرا فاصله هر کاربر از خودش صفر است ($d(A,A) = 0$).</li>
    <li>ماتریس <b>متقارن</b> است؛ یعنی فاصله کاربر A از B برابر با فاصله کاربر B از A است.</li>
    <li>عدد فاصله بین کاربر A و B بسیار کوچک است (حدود ۱۰.۰۵) که نشان‌دهنده تشابه رفتاری زیاد آن‌هاست؛ در حالی که فاصله بین A و D بسیار بزرگ است (حدود ۳۸۰.۸۸) که تفاوت الگوی خرید آن‌ها را اثبات می‌کند.</li>
