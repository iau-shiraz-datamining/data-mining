  <span style="font-size: 20pt;"></span> 2.5.2 Discretization
</div>

<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: دسته‌بندی زمان فیلم‌ها در «نتفلیکس» (Netflix) با سطل‌بندی عرض یکسان ✂️📊
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    الگوریتم‌های پیشنهاد فیلم در «نتفلیکس» با متغیرهای عددی پیوسته‌ای مانند <b>مدت زمان فیلم (Duration به دقیقه)</b> روبه‌رو هستند که می‌تواند هر عددی بین ۶۰ تا ۱۸۰ دقیقه باشد. برخی الگوریتم‌ها (مثل درخت تصمیم یا قوانین وابستگی) به جای اعداد پیوسته دقیق، با دسته‌های کیفی عملکرد بسیار بهتری دارند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>راه‌حل — گسسته‌سازی با سطل‌بندی عرض یکسان (Equal Width Binning):</b><br>
    در این روش، کل دامنه داده‌ها ($X_{\max} - X_{\min}$) به تعداد مشخصی بازه ($k$) با **عرض کاملاً برابر** تقسیم می‌شود:
  </p>

  <div style="text-align: center; margin: 15px 0;">
    $$\text{Bin Width} = \frac{X_{\max} - X_{\min}}{k}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>مثال عددی در نتفلیکس:</b><br>
    اگر کوتاه ترین فیلم ۶۰ دقیقه ($X_{\min}$) و طولانی‌ترین ۱۸۰ دقیقه ($X_{\max}$) باشد و بخواهیم ۳ دسته ($k=3$) بسازیم:<br>
    &nbsp;&nbsp;&nbsp;• <b>عرض هر سطل:</b> $(180 - 60) / 3 = 40$ دقیقه.<br>
    &nbsp;&nbsp;&nbsp;• <b>سطل اول (کوتاه / Short):</b> ۶۰ تا ۱۰۰ دقیقه.<br>
    &nbsp;&nbsp;&nbsp;• <b>سطل دوم (متوسط / Medium):</b> ۱۰۰ تا ۱۴۰ دقیقه.<br>
    &nbsp;&nbsp;&nbsp;• <b>سطل سوم (طولانی / Long):</b> ۱۴۰ تا ۱۸۰ دقیقه.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> با تبدیل داده پیوسته به ۳ دسته گسسته، سیستم نتفلیکس می‌تواند به‌راحتی ترجیحات کاربر (مثلاً «علاقه‌مند به فیلم‌های کوتاه») را استخراج کند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: تبدیل متغیر پیوسته به دسته‌های گسسته با فواصل عددی یکسان
    </div>
  </p>

</div>

```python
display_academic_text(
    "تبدیل داده‌ها: گسسته‌سازی متغیرهای پیوسته با سطل‌بندی عرض یکسان (Equal Width Binning) 🎬", 
    "main_title"
)

display_academic_text(
    "بر مبنای اصول پیش‌پردازش داده‌ها، بسیاری از الگوریتم‌ها (به ویژه مدل‌های مبتنی بر قوانین / Rule-based مانند Apriori یا Decision Trees) "
    "نیازمند دریافت ورودی‌های <b>دسته‌ای (Categorical)</b> هستند. در این سناریو، متغیر پیوستهٔ «مدت زمان فیلم» با بی‌نهایت حالتِ ممکن (اعشاری)، "
    "باید به دسته‌های معنایی تبدیل گردد. متد <b>سطل‌بندی با عرض یکسان (Equal Width Binning)</b>، دامنه نوسان داده‌ها را به $k$ بازه "
    "با <i>فواصلِ عددیِ کاملاً برابر</i> تقسیم می‌نماید.", 
    "paragraph"
)

display_academic_text(
    "معادله محاسبه عرض هر سطل ($W$) بر اساس دامنهٔ کل داده‌ها:", 
    "sub_title"
)

# Utilizing Raw Strings (r"") for correct LaTeX rendering of the Binning formula
display_academic_text(
    r"$$W = \frac{X_{max} - X_{min}}{k}$$", 
    "paragraph"
)


# ==========================================================
# SECTION 2: DATA INITIALIZATION AND DISCRETIZATION PIPELINE
# ==========================================================
display_academic_text(
    "مرحله نخست: تولید داده‌های پیوسته و استخراج فواصل گسسته", 
    "sub_title"
)

display_academic_text(
    "در این فاز، فهرستی از فیلم‌ها با زمان‌های تصادفی بین ۶۰ تا ۱۸۰ دقیقه شبیه‌سازی می‌گردد. "
    "سپس با بهره‌گیری از متد قدرتمند <code>pd.cut</code>، دامنهٔ فیزیکی (۱۲۰ دقیقه) به ۳ سطل با <b>عرضِ دقیقاً ۴۰ دقیقه</b> تقسیم شده "
    "و به هر رکورد، برچسب کیفی (Label) متناسب با سطل تعلق می‌گیرد.", 
    "paragraph"
)

# Step 1: Initialize the dataset with strictly continuous movie durations
n_movies = 150 # Increased sample size for better visual distribution
data = {
    'Movie_ID': [f"MOV_{i}" for i in range(1001, 1001 + n_movies)],
    # Generate uniform continuous durations between exactly 60 & 180 minutes
    'Duration_Min': np.random.uniform(60, 180, n_movies),  
}
df_raw = pd.DataFrame(data)

# Step 2: Apply Equal Width Binning using pd.cut()
k = 3
# Adding logical numerical ranges to labels for extreme clarity
labels = ['Short (60 - 100m)', 'Medium (100 - 140m)', 'Long (140 - 180m)']

# Perform Discretization and automatically extract mathematical bin boundaries (retbins=True)
df_binned = df_raw.copy()
df_binned['Duration_Category'], bin_edges = pd.cut(
    df_binned['Duration_Min'], 
    bins=k, 
    labels=labels, 
    retbins=True
)

# Step 3: Compute the frequency distribution (Number of movies in each bin)
df_bin_counts = df_binned['Duration_Category'].value_counts().sort_index().reset_index()
df_bin_counts.columns = ['Category_Label', 'Movie_Count']

# Render datasets cleanly
df_raw_disp = df_raw.head(8).reset_index(drop=True)
df_binned_disp = df_binned.head(8).reset_index(drop=True)

display_side_by_side_datasets(
    [df_raw_disp, df_binned_disp, df_bin_counts],
    [
        "جدول ۱: داده‌های پیوسته (نمونه خام)", 
        "جدول ۲: داده‌های گسسته‌شده (Discretized)",
        "جدول ۳: ماتریس توزیع فراوانی سطل‌ها"
    ]
)

# ==========================================================
# SECTION 3: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی - انتقال از پیوستگی به گسستگی", 
    "sub_title"
)

display_academic_text(
    "نمودار زیر ماهیت هندسیِ این انتقال را به تصویر می‌کشد. در <b>هیستوگرام سمت چپ</b>، توزیع پیوستهٔ فیلم‌ها مشاهده می‌شود که توسط "
    "خطوط‌چین قرمز (نقاط برش الگوریتم) به ۳ ناحیهٔ <i>کاملاً مساوی (از نظر عرض)</i> قطعه‌قطعه شده است. "
    "در <b>نمودار میله‌ای سمت راست</b>، نتیجه نهاییِ این تبدیل نمایش داده می‌شود: محور افقی دیگر از جنس <i>«زمانِ پیوسته»</i> نیست، "
    "بلکه به ۳ ماهیتِ <i>«متنی و کیفی»</i> تقلیل یافته است.", 
    "paragraph"
)

# Initialize a dual-axis canvas for visual comparative analytics
fig, axes = plt.subplots(1, 2, figsize=(19, 6.5), dpi=135)

# ----------------------------------------------------------
# Plot 1: Continuous Histogram with Explicit Bin Boundaries
# ----------------------------------------------------------
sns.histplot(
    data=df_raw, 
    x='Duration_Min', 
    bins=15, # Use higher resolution bins for the underlying continuous data
    color='#2c3e50', # Navy
    edgecolor='white', 
    linewidth=1.0,
    alpha=0.8, 
    ax=axes[0]
)

# Overlay vertical threshold lines to demonstrate the mathematical cut points
# We skip the very first (min) and last (max) edges to just show the internal cuts
for b in bin_edges[1:-1]:
    axes[0].axvline(
        x=b, 
        color='#e74c3c', # Red for the cut lines
        linestyle='--', 
        linewidth=3.0, 
        label=f'Algorithm Cut Threshold: {b:.1f}m' if b == bin_edges[1] else ""
    )

axes[0].annotate(
    "Equal Width Logic:\nRange = 180 - 60 = 120\nWidth (W) = 120 / 3 = 40 min.\nThe algorithm blindly slices the X-axis\nevery 40 minutes.",
    xy=(100, 15), xytext=(120, 15),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title('Figure 1: Continuous Topology & Equal Width Cut Points', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[0].set_xlabel('Exact Duration (Minutes)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_ylabel('Absolute Frequency', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_xlim(50, 190)
axes[0].legend(loc='upper right', frameon=True, fontsize=11)
axes[0].grid(True, linestyle='--', alpha=0.5)

# ----------------------------------------------------------
# Plot 2: Categorical Bar Chart (Discretized Results)
# ----------------------------------------------------------
# Using hue='Category_Label' and legend=False strictly to comply with seaborn v0.14.0 requirements
sns.barplot(
    data=df_bin_counts, 
    x='Category_Label', 
    y='Movie_Count', 
    hue='Category_Label', 
    legend=False, 
    palette=['#1abc9c', '#2980b9', '#2c3e50'], # Turquoise, Classic Blue, Navy
    edgecolor='#2c3e50', 
    linewidth=2.0,
    width=0.6,
    ax=axes[1]
)

axes[1].annotate(
    "Dimensionality Reduction:\nThe machine no longer cares if a movie\nis 101 or 139 minutes long. They are\nall now just treated as 'Medium'.",
    xy=(1.0, 50), xytext=(0.5, 65),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[1].set_title('Figure 2: Discretized Categorical Target (Final Output)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[1].set_xlabel('Semantic Duration Category (Discrete)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_ylabel('Number of Movies', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].grid(axis='y', linestyle='--', alpha=0.5)
axes[1].set_ylim(0, 80)
axes[1].set_xticklabels(axes[1].get_xticklabels(), fontweight='bold', fontsize=12)

# Add exact value annotations on top of the bars
for p in axes[1].patches:
    if p.get_height() > 0:
        axes[1].annotate(
            f"n={int(p.get_height())}", 
            (p.get_x() + p.get_width() / 2., p.get_height()), 
            ha='center', va='center', xytext=(0, 12), textcoords='offset points',
            fontweight='bold', fontsize=12, color='#2c3e50'
        )

# Clear top and right spines for a clean academic aesthetic
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی عملیات (Strategic Implication):</b><br>"
    r"گسسته‌سازی (Discretization)، ابزاری قدرتمند برای تقلیل پیچیدگی فضای ابعاد (Dimensionality Reduction) و مقاوم‌سازیِ مدل‌ها (Robustness) در برابر نویز است. "
    r"با تبدیل دامنهٔ پیوسته به مفاهیم کیفی، نویزهای جزئیِ سنسورها یا تفاوت‌های بی‌اهمیت (مثلاً تفاوتِ ارزشِ یک فیلم ۱۱۵ دقیقه‌ای با یک فیلم ۱۱۸ دقیقه‌ای) نادیده گرفته می‌شوند.<br>"
    r"در نتیجه، سیستم توصیه‌گر قادر می‌گردد تا الگوهای کلان رفتاری کاربران (علاقهٔ مطلق به فیلم‌های <i>کوتاه</i> یا <i>بلند</i>) را با سرعت پردازشی بسیار بالاتر، "
    r"و بدونِ درگیر شدن در محاسبات سنگینِ اعشاری، کشف و استخراج نماید."
)
display_academic_text(insight_conclusion, "insight")
```
