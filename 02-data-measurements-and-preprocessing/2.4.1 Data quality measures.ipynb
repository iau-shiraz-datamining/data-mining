  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: ارزیابی کیفیت داده و مدیریت مقادیر گمشده در «آمازون» (Amazon) ⚠️
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    پایگاه داده خریداران «آمازون» شامل میلیون‌ها رکورد است. پیش از اینکه الگوریتم‌های هوش مصنوعی وارد عمل شوند، تیم داده‌کاوی باید <b>کیفیت داده‌ها (Data Quality)</b> را بر اساس ۴ معیار اصلی بسنجد:
  </p>

  <ul style="text-align: right; margin: 0 0 10px; padding-right: 22px;">
    <li><b>کامل بودن (Completeness):</b> آیا تمام اطلاعات ضروری مشتریان ثبت شده است؟</li>
    <li><b>دقت (Accuracy):</b> آیا مقادیر ثبت‌شده (مثل آدرس و کد پستی) صحیح هستند؟</li>
    <li><b>سازگاری (Consistency):</b> آیا داده‌ها در بخش‌های مختلف سیستم با هم تناقض ندارند؟</li>
    <li><b>بهنگام بودن (Timeliness):</b> آیا اطلاعات خریدار به‌روز است؟</li>
  </ul>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام اول: چالش مقادیر گمشده (Missing Values / NaNs)</b><br>
    هنگام ثبت‌نام در آمازون، برخی کاربران سن، شماره تلفن یا کد پستی خود را وارد نمی‌کنند. وجود خانه خالی (Null / NaN) در ماتریس داده، باعث خطا در الگوریتم‌های داده‌کاوی (مثل شبکه‌های عصبی و رگرسیون) می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام دوم: سنجش درصد کامل بودن (Completeness Ratio)</b><br>
    ارزیابی کیفیت با محاسبه درصد خانه های خالی در هر ستون شروع می‌شود:<br>
    &nbsp;&nbsp;&nbsp;• اگر درصد مقادیر گمشده در یک ستون بسیار بالا باشد (مثلاً بالای ۷۰٪)، کل آن ستون حذف می‌شود.<br>
    &nbsp;&nbsp;&nbsp;• اگر درصد کم باشد (مثلاً زیر ۵٪)، رکوردهای دارای مقدار گمشده ترمیم یا بازسازی می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام سوم: راهکارهای هوشمند برخورد با داده‌های گمشده (Imputation)</b><br>
    &nbsp;&nbsp;&nbsp;• <b>برای ویژگی‌های عددی (مثل سن):</b> جایگذاری با میانه (Median) داده‌ها.<br>
    &nbsp;&nbsp;&nbsp;• <b>برای ویژگی‌های اسمی/کیفی (مثل شهر):</b> جایگذاری با پرکاربردترین مقدار (Mode).
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> ارزیابی و پاک‌سازی مقادیر گمشده باعث افزایش شفافیت، دقت محاسبات و جلوگیری از انحراف مدل‌های تحلیل رفتار خریداران می‌شود.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: سنجش درصد کامل بودن (Completeness) و ترمیم هوشمند مقادیر گمشده
    </div>
  </p>

</div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.impute import SimpleImputer 
from matplotlib.colors import ListedColormap

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})

plt.rcParams['figure.dpi'] = 135

np.random.seed(42)
```

```python
display_academic_text(
    "پیش‌پردازش داده‌ها: ارزیابی کیفیت و ترمیم مقادیر گمشده (Data Imputation) 🧩", 
    "main_title"
)

display_academic_text(
    "وجود <b>مقادیر گمشده (Missing Values / NaN)</b> یکی از رایج‌ترین چالش‌ها در پایگاه‌های دادهٔ عملیاتی نظیر پلتفرم‌های فروشگاهی است. "
    "الگوریتم‌های هوش مصنوعی نیازمند بردارهای اطلاعاتیِ کامل هستند و در صورت برخورد با مقدار <code>NaN</code> متوقف می‌شوند. پیش از استقرار مدل، "
    "شاخص «کامل بودن» (Completeness Ratio) محاسبه شده و سپس بر مبنای نوع متغیر، استراتژی‌های آماری <b>جایگذاری (Imputation)</b> اعمال می‌گردد.", 
    "paragraph"
)

display_academic_text("معادله محاسبه نرخ فقدان داده (Missing Ratio) در یک بردار ویژگی $x$ با طول $n$:", "sub_title")

# Utilizing Raw Strings (r"") for LaTeX rendering of the Missing Ratio formula
display_academic_text(
    r"$$Missing\_Ratio (\%) = \left( \frac{\sum_{i=1}^{n} \mathbb{I}(x_i = NaN)}{n} \right) \times 100$$", 
    "paragraph"
)

# ==========================================================
# SECTION 2: DATA QUALITY ASSESSMENT AND MISSING VALUE HANDLING
# ==========================================================
display_academic_text(
    "مرحله نخست: گزارش‌گیری کیفیت داده و اعمال استراتژی‌های ترمیم", 
    "sub_title"
)

display_academic_text(
    "در این فاز، پس از استخراج گزارش کیفیت، مقادیر گمشده در ویژگی‌های <i>کمّی</i> (سن و درآمد) با استفاده از شاخص مقاوم <b>«میانه (Median)»</b> "
    "بازسازی می‌شوند تا از انحراف ناشی از داده‌های پرت جلوگیری شود. برای ویژگی‌های <i>کیفی و دسته‌ای</i> (روش پرداخت) نیز از "
    "شاخص <b>«نما (Mode / Most Frequent)»</b> بهره‌برداری می‌گردد.", 
    "paragraph"
)

# Step 1: Initialize the Amazon customer dataset explicitly injecting NaN values
data = {
    'Customer_ID': [1001, 1002, 1003, 1004, 1005, 1006],
    'Age': [25, 34, np.nan, 45, 29, np.nan],  
    'Annual_Income_kUSD': [65, 80, 120, np.nan, 55, 95],  
    'Preferred_Payment': ['Credit Card', np.nan, 'PayPal', 'Credit Card', 'Credit Card', np.nan],  
}
df_raw = pd.DataFrame(data)

# Step 2: Evaluate Data Quality and Generate Completeness Report
missing_count = df_raw.isnull().sum()
missing_percentage = (df_raw.isnull().mean() * 100)

df_quality_report = pd.DataFrame({
    'Missing_Count': missing_count, 
    'Missing_Ratio (%)': missing_percentage
}).reset_index().rename(columns={'index': 'Feature'})

# Step 3: Handle Missing Values via Machine Learning Imputation (scikit-learn)
df_cleaned = df_raw.copy()

# 3.1 Impute Numeric columns with MEDIAN (Robust to outliers compared to Mean)
num_imputer = SimpleImputer(strategy='median')
df_cleaned[['Age', 'Annual_Income_kUSD']] = num_imputer.fit_transform(df_cleaned[['Age', 'Annual_Income_kUSD']])

# 3.2 Impute Categorical column with MODE (Most Frequent value)
cat_imputer = SimpleImputer(strategy='most_frequent')
df_cleaned[['Preferred_Payment']] = cat_imputer.fit_transform(df_cleaned[['Preferred_Payment']])

# Ensure floating values injected by Imputer for Age are converted back to integers for logical correctness
df_cleaned['Age'] = df_cleaned['Age'].astype(int)

# Render the datasets
display_side_by_side_datasets(
    [df_raw, df_quality_report, df_cleaned],
    [
        "جدول ۱: پایگاه داده خام (حاوی مقادیر NaN)", 
        "جدول ۲: گزارش کیفیت و نرخ فقدان",
        "جدول ۳: پایگاه داده ترمیم‌شده (Imputed)"
    ]
)

# ==========================================================
# SECTION 3: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی - نقشه فقدان داده و پروفایل کیفیت", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکی زیر توزیع فضایی نواقص را بازنمایی می‌کند. <b>نمودار سمت چپ (Nullity Heatmap)</b> دقیقاً نشان می‌دهد که کدام سطرها و ستون‌ها "
    "دچار خلأ اطلاعاتی هستند (رنگ سرمه‌ای = داده موجود، رنگ فیروزه‌ای = مقدار گمشده/NaN). <b>نمودار میله‌ای سمت راست</b> نیز با نمایش درصد فقدان، "
    "تیم داده‌کاوی را در خصوص حفظ و ترمیمِ یک ویژگی یا <i>حذف کامل (Drop)</i> آن در صورت عبور از آستانهٔ بحرانی راهنمایی می‌نماید.", 
    "paragraph"
)

# Initialize a dual-axis canvas for visual quality analytics
fig, axes = plt.subplots(1, 2, figsize=(19, 6.5), dpi=135)

# ----------------------------------------------------------
# Plot 1: Nullity Heatmap (Spatial distribution of missing values)
# ----------------------------------------------------------
# Academic Custom Colormap: Navy for valid data, Light Turquoise for missing data
cmap_missing = ListedColormap(['#2c3e50', '#1abc9c'])

# Visualizes exactly where the missing values are located across the DataFrame rows/columns
sns.heatmap(
    df_raw.isnull(), 
    cmap=cmap_missing, 
    cbar=False, 
    ax=axes[0], 
    yticklabels=df_raw['Customer_ID'],
    linewidths=2.0,
    linecolor='white'
)

axes[0].annotate(
    "Missing Data Blocks:\nThe light turquoise blocks represent\nmissing entries (NaN) that will crash\nany Machine Learning algorithm.",
    xy=(1.5, 2.5), xytext=(2.5, 4.5),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[0].set_title('Figure 1: Nullity Matrix Heatmap', fontdict={'fontsize': 15, 'fontweight': 'bold'}, pad=15)
axes[0].set_xlabel('Dataset Features', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_ylabel('Target Object (Customer ID)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].tick_params(axis='x', rotation=15)
axes[0].set_xticklabels(axes[0].get_xticklabels(), fontweight='bold', fontsize=11)
axes[0].set_yticklabels(axes[0].get_yticklabels(), fontweight='bold', fontsize=11)

# ----------------------------------------------------------
# Plot 2: Bar Plot of Completeness Ratio
# ----------------------------------------------------------
# Added hue='Feature' and legend=False to comply with seaborn v0.14.0+ requirements
sns.barplot(
    data=df_quality_report[df_quality_report['Feature'] != 'Customer_ID'], 
    x='Feature', 
    y='Missing_Ratio (%)', 
    hue='Feature',       
    legend=False,        
    ax=axes[1], 
    palette=['#2980b9', '#1abc9c', '#2c3e50'], 
    edgecolor='#2c3e50', 
    linewidth=1.5,
    width=0.6
)

# Plotting the threshold line for Drop vs Impute
axes[1].axhline(50, color='#e74c3c', linestyle='--', linewidth=3.0, label='Critical Threshold (50%)')

axes[1].annotate(
    "Imputation Strategy:\nBecause the missing ratio for all features\nis below 50%, we can safely use statistical\nimputation (Median/Mode) to save the data.",
    xy=(0.0, 33.3), xytext=(0.5, 60),
    fontsize=11, fontweight='bold', color='#2c3e50',
    arrowprops=dict(arrowstyle="->", color="#34495e", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#ecf0f1", ec="#bdc3c7", lw=1.5)
)

axes[1].set_title('Figure 2: Feature Missing Ratio Analysis', fontdict={'fontsize': 15, 'fontweight': 'bold'}, pad=15)
axes[1].set_xlabel('Dataset Features', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_ylabel('Missing Ratio (%)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_ylim(0, 100) # Set Y-axis strictly from 0% to 100%
axes[1].tick_params(axis='x', rotation=15)
axes[1].set_xticklabels(axes[1].get_xticklabels(), fontweight='bold', fontsize=11)
axes[1].grid(axis='y', linestyle='--', alpha=0.5)
axes[1].legend(loc='upper right', frameon=True, fontsize=11)

# Add value annotations on top of the bars for exact percentage readout
for p in axes[1].patches:
    axes[1].annotate(
        f"{p.get_height():.1f}%", 
        (p.get_x() + p.get_width() / 2., p.get_height()), 
        ha='center', va='center', xytext=(0, 12), textcoords='offset points',
        fontweight='bold', fontsize=12, color='#2c3e50'
    )

# Clear top and right spines for a clean academic aesthetic
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی عملیات (Strategic Implication):</b><br>"
    r"رها کردن مقادیر گمشده در پایگاه داده، به ایجاد سوگیری (Bias) و سقوطِ قطعیِ الگوریتم‌های پیش‌بینی (مثل XGBoost یا Neural Networks) منجر می‌شود. "
    r"در صورتی که نرخ فقدانِ یک ستون <i>بیشتر از ۵۰ درصد</i> باشد (خط قرمز در گراف ۲)، آن ویژگی دیگر قابل اعتماد نیست و باید به طور کامل از مدل <b>حذف (Drop)</b> شود.<br><br>"
    r"اما در این سناریو، به دلیل پایین بودن نرخ فقدان، <b>جایگذاریِ هوشمند (Imputation)</b> بر مبنای <i>میانه (برای متغیرهای پیوسته نظیر سن و درآمد)</i> و <i>نما (برای متغیرهای کیفی نظیر روش پرداخت)</i>، "
    r"ساختار توزیعِ احتمالاتیِ جامعه آماری را با کمترین میزانِ انحراف حفظ نموده و یکپارچگیِ اطلاعاتیِ لازم جهت اجرای مدل‌های هوش مصنوعی را تضمین نمود (جدول ۳)."
)
display_academic_text(insight_conclusion, "insight")
```
