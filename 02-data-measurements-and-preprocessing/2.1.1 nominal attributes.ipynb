    تیم داده‌کاوی فودباکس با ستونی از جنس متن روبه‌رو است:<br>
    &nbsp;&nbsp;&nbsp;• Diet_Type: کیتو، گیاه‌خواری، پروتئین بالا، عادی<br>
    مشکل اینجاست که الگوریتم‌های یادگیری ماشین، متن نمی‌فهمند و فقط با عدد کار می‌کنند. اگر بی‌گدار به آب بزنیم و مثلاً به کیتو عدد ۱، گیاه‌خواری عدد ۲ و پروتئین بالا عدد ۳ بدهیم، به مدل دروغ گفته‌ایم؛ چون این کار به مدل القا می‌کند که «پروتئین بالا» سه برابر «کیتو» ارزش دارد؛ فرضی کاملاً نادرست و گمراه‌کننده.

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام دوم: راه‌حل — One-Hot Encoding</b><br>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display, HTML

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})

plt.rcParams['figure.dpi'] = 120

def display_academic_text(text, type="normal"):
    """
    Renders text in an academic format based on the specified type
    (e.g., title, paragraph, insight).
    Uses strict typography and styling to avoid informal elements.
    """
    styles = {
        "main_title": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 24px; "
            "font-weight: bold; text-align: center; border-bottom: 2px solid #34495e; "
            "padding-bottom: 15px; margin: 30px 0;"
        ),
        "sub_title": (
            "font-family: Tahoma, sans-serif; color: #34495e; font-size: 19px; "
            "font-weight: bold; border-right: 5px solid #2980b9; padding-right: 15px; "
            "margin: 40px 0px 20px 0px; background-color: #ecf0f1; padding: 12px; "
            "border-radius: 4px;"
        ),
        "paragraph": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 16px; "
            "margin-bottom: 20px; line-height: 1.8; text-align: justify;"
        ),
        "insight": (
            "font-family: Tahoma, sans-serif; color: #16a085; font-size: 16px; "
            "font-weight: bold; background-color: #e8f8f5; padding: 20px; "
            "border-left: 5px solid #1abc9c; margin: 30px 0; line-height: 1.8; "
            "text-align: justify;"
        )
    }
    display(HTML(f"<div dir='rtl' style='text-align: right; {styles[type]}'>{text}</div>"))

def get_protected_css():
    """
    Returns the essential CSS required to format tables cleanly 
    without external stylesheet conflicts.
    """
    return """
    <style>
    .academic-table-wrapper { 
        direction: ltr !important; 
        overflow-x: auto !important; 
        width: 100% !important; 
        background-color: #fff; 
        border: 1px solid #bdc3c7; 
        border-radius: 4px; 
        box-shadow: 0 3px 6px rgba(0,0,0,0.08); 
    }
    table.strict-tbl { 
        border-collapse: collapse !important; 
        width: 100% !important; 
        margin: 0 !important; 
        font-family: inherit !important;
        font-size: 14px !important; 
        table-layout: auto !important; 
        border: none !important; 
    }
    table.strict-tbl thead th { 
        background-color: #2c3e50 !important; 
        color: #ffffff !important; 
        padding: 12px 18px !important; 
        text-align: center !important; 
        border: 1px solid #34495e !important; 
        white-space: nowrap !important; 
        font-weight: bold !important; 
    }
    table.strict-tbl tbody td { 
        padding: 10px 18px !important; 
        text-align: center !important; 
        border: 1px solid #bdc3c7 !important; 
        white-space: nowrap !important; 
        color: #2c3e50 !important; 
    }
    table.strict-tbl tbody tr:nth-child(even) td { 
        background-color: #f8f9fa !important; 
    }
    table.strict-tbl tbody tr:hover td { 
        background-color: #e8ecef !important; 
    }
    </style>
    """

def display_side_by_side_datasets(dfs, titles, *args):
    """
    Renders one or multiple pandas DataFrames.
    
    - If only one dataframe: displays it full-width.
    - If multiple dataframes: displays them side-by-side.
    
    Parameters:
    -----------
    dfs : list or DataFrame
        List of dataframes, or a single dataframe
    titles : list or str
        List of titles, or a single title string
    *args : (optional) for backward compatibility with old four-argument calls
    """
    if not isinstance(dfs, list) and not isinstance(titles, list):
        df1, title1, df2, title2 = dfs, titles, args[0], args[1]
        dfs = [df1, df2]
        titles = [title1, title2]
    else:
        if not isinstance(dfs, list) or not isinstance(titles, list):
            raise ValueError("دو آرگومان اول باید لیست باشند (یا چهار آرگومان جداگانه).")
        if len(dfs) != len(titles):
            raise ValueError("تعداد دیتافریم‌ها و عنوان‌ها باید برابر باشد.")
        if len(dfs) == 0:
            raise ValueError("حداقل یک دیتافریم برای نمایش نیاز است.")
    
    if len(dfs) == 1:
        df = dfs[0]
        title = titles[0]
        table_html = df.to_html(index=False, classes=['strict-tbl'], border=0)
        html = f"""
        <div style="width: 100%; margin: 20px 0 35px 0;">
            <div dir="rtl" style="background-color: #2980b9; color: white; padding: 12px; 
                                 text-align: center; font-family: Tahoma, sans-serif; 
                                 font-weight: bold; font-size: 15px; border-radius: 4px 4px 0 0;">
                {title}
            </div>
            <div class="academic-table-wrapper" style="margin-top: 0; border-top: none;">
                {table_html}
            </div>
        </div>
        """
        display(HTML(get_protected_css() + html))
        return
    
    num_cols = len(dfs)
    col_html_list = []
    for i, (df, title) in enumerate(zip(dfs, titles)):
        table_html = df.to_html(index=False, classes=['strict-tbl'], border=0)
        bg_color = "#2980b9" if i % 2 == 0 else "#27ae60"
        col_html = f"""
        <div class="flex-column" style="flex: 1 1 0; min-width: 0; display: flex; flex-direction: column;">
            <div style="background-color: {bg_color}; color: white; padding: 12px; 
                        text-align: center; font-family: Tahoma, sans-serif; 
                        font-weight: bold; font-size: 15px; border-radius: 4px 4px 0 0;">
                {title}
            </div>
            <div class="academic-table-wrapper" style="margin-top: 0; border-top: none;">
                {table_html}
            </div>
        </div>
        """
        col_html_list.append(col_html)
    
    columns_joined = "".join(col_html_list)
    html = f"""
    <div style="display: flex; flex-direction: row; justify-content: space-between; 
                gap: 20px; width: 100%; max-width: 100%; margin: 20px 0 35px 0;
                box-sizing: border-box;">
        {columns_joined}
    </div>
    """
    display(HTML(get_protected_css() + html))


display(HTML("""
<style>
@import url('https://cdn.jsdelivr.net/gh/rastikerdar/vazirmatn@v33.003/Vazirmatn-font-face.css');

* {
    font-family: 'Vazirmatn', 'Vazir', Tahoma, sans-serif !important;
}
div, p, h1, h2, h3, h4, td, th, span, li, .strict-tbl, .academic-table-wrapper {
    font-family: 'Vazirmatn', 'Vazir', Tahoma, sans-serif !important;
}
</style>
"""))



np.random.seed(42)
```

```python
display_academic_text(
    "پیش‌پردازش داده‌ها: مدیریت ویژگی‌های اسمی با استفاده از تکنیک رمزگذاری یک‌باره (One-Hot Encoding)", 
    "main_title"
)

display_academic_text(
    "بر اساس مبانی فصل دوم، ویژگی‌های اسمی (Nominal Features) نمایانگر مقادیر یا حالت‌های فاقد ترتیب هستند. "
    "در یک سیستم اشتراک غذای رژیمی، متغیر نوع رژیم غذایی هیچ‌گونه ساختار ترتیبی (Ordinal) ندارد. تیم کاوش دانش "
    "در مواجهه با این پایگاه داده متوجه می‌شود که تخصیص مقادیر عددی پیوسته (مانند ۱، ۲ و ۳) به این متغیرها، "
    "منجر به القای ارزش‌های نسبیِ کاذب (مثلاً فرض اینکه رژیم کتو دو برابرِ رژیم عادی ارزش دارد!) و بروز سوءگیری "
    "محاسباتی در الگوریتم‌های فاصله‌محورِ یادگیری ماشین خواهد شد.", 
    "paragraph"
)

# ==========================================================
# Section 2: Data Simulation & Preprocessing
# ==========================================================
display_academic_text(
    "بازنمایی معماری داده‌ها پیش و پس از اعمال تحولات ساختاری", 
    "sub_title"
)

display_academic_text(
    "جهت رفع ناهنجاری حاصل از مقادیر متنی، از مکانیزم <b>رمزگذاری یک‌باره (One-Hot Encoding)</b> بهره‌برداری می‌گردد. "
    "در این روش، به جای تخصیص یک شناسهٔ عددیِ گمراه‌کننده، به ازای هر دستهٔ مستقل، یک <i>بُعدِ باینریِ مجزا</i> در فضای ویژگی تولید می‌شود. "
    "جداول زیر ساختار خام و ساختارِ متعامدِ نهایی را در کنار یکدیگر مقایسه می‌کنند.", 
    "paragraph"
)

# Step 1: Simulate the raw dataset containing a nominal attribute (Diet_Type)
raw_data_dict = {
    'User_ID': [1, 2, 3, 4, 5],
    'Diet_Type': ['Keto', 'Vegetarian', 'High-Protein', 'Normal', 'Keto']
}
df_raw = pd.DataFrame(raw_data_dict)

# Step 2: Apply One-Hot Encoding to prevent ordinal bias.
# The 'get_dummies' function pivots the nominal categories into independent binary dimensions.
df_encoded = pd.get_dummies(df_raw, columns=['Diet_Type'], dtype=int)

# Clean up column names for better display
df_encoded.columns = [col.replace('Diet_Type_', '') if 'Diet_Type' in col else col for col in df_encoded.columns]

# Step 3: Render both raw and encoded DataFrames side-by-side using the list format
display_side_by_side_datasets(
    [df_raw, df_encoded], 
    [
        "وضعیت خام: ویژگی اسمی فاقد ترتیب (Nominal Attribute)", 
        "وضعیت نهایی: ویژگی‌های باینری و متعامد (One-Hot Encoded)"
    ]
)

# ==========================================================
# Section 3: Compact Visual Analytics
# ==========================================================
# Step 4: Isolate the binary feature matrix for visualization
heatmap_matrix = df_encoded.drop('User_ID', axis=1)

# Modify the index labels to represent actual users in the visualization
heatmap_matrix.index = [f"User Instance {i}" for i in df_encoded['User_ID']]

# Configure a customized matplotlib figure
plt.figure(figsize=(14, 5), dpi=135)

# Step 5: Render a heatmap to visually prove the orthogonality of the new features.
# We use a custom color map: White for 0, Navy/Academic Blue for 1
from matplotlib.colors import ListedColormap
cmap_binary = ListedColormap(['#f8f9fa', '#2c3e50'])

sns.heatmap(
    heatmap_matrix, 
    annot=True, 
    cmap=cmap_binary, 
    cbar=False, 
    linewidths=2.0, 
    linecolor='#bdc3c7', 
    annot_kws={"size": 16, "weight": "bold"}
)

# Add an explanatory annotation indicating the mutual exclusivity
plt.annotate(
    "Mutual Exclusivity (Orthogonality):\nNotice how every row contains exactly one '1'.\nThe machine no longer compares 'Keto' vs 'Normal' as numbers,\nbut treats them as independent dimensions!",
    xy=(4, 1.5), xytext=(5.5, 2.5),
    fontsize=11, fontweight='bold', color='#2c3e50',
    arrowprops=dict(arrowstyle="->", color="#34495e", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.5", fc="#ecf0f1", ec="#34495e", lw=1.5),
    clip_on=False
)

# Apply formal academic labeling
plt.title(
    'Figure 1: Orthogonal Representation of One-Hot Encoded Feature Space', 
    fontdict={'fontsize': 14, 'fontweight': 'bold'}, 
    pad=15
)
plt.ylabel('User Records', fontdict={'fontsize': 13, 'fontweight': 'bold'})
plt.xlabel('Binary Encoded Diet Dimensions (Features)', fontdict={'fontsize': 13, 'fontweight': 'bold'})

plt.tight_layout()
plt.show()

# ==========================================================
# Section 4: Final Strategic Conclusion
# ==========================================================
display_academic_text(
    r"<b>استنتاج راهبردیِ عملیات پیش‌پردازش:</b><br><br>"
    r"نمودار بالا (Figure 1) جادویِ ریاضیاتیِ One-Hot Encoding را نشان می‌دهد. اعمال این مکانیزمِ رمزگذاری، ماتریس ویژگی‌ها را به نحوی سازمان‌دهی می‌کند که الگوریتم‌های داده‌کاوی (به ویژه شبکه‌های عصبی و الگوریتم‌های فاصله‌محور مانند KNN) قادر خواهند بود "
    r"بدون مواجهه با سوءتفاهم‌های ناشی از «مقیاس‌گذاری کاذب»، الگوهای تغذیه‌ای را به صورت <b>عینی و متعامد (Orthogonal)</b> تفسیر نموده و پیشنهادات شخصی‌سازی‌شده را با بالاترین سطح دقت استخراج نمایند.", 
    "insight"
)
```
