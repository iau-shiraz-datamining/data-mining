<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: نرمال‌سازی ویژگی‌های محصول در «آمازون» (Amazon) 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    الگوریتم توصیه‌گر «آمازون» برای پیشنهاد محصولات مشابه به کاربران، ویژگی‌های مختلفی از هر کالا را بررسی می‌کند. این ویژگی‌ها مقیاس‌های کاملاً متفاوتی دارند:
  </p>

  <ul style="text-align: right; margin: 0 0 10px; padding-right: 22px;">
    <li><b>قیمت (Price):</b> از ۲۰ دلار تا ۱۰ geometric دلار.</li>
    <li><b>وزن (Weight):</b> از ۰.۱ کیلوگرم تا ۲۰ کیلوگرم.</li>
    <li><b>امتیاز کاربران (Rating):</b> از ۲ تا ۵ ستاره.</li>
  </ul>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>چالش بزرگ:</b> اگر ویژگی‌ها را به‌صورت خام وارد الگوریتم کنیم، ویژگی‌هایی با مقیاس بزرگ‌تر (مثل قیمت) به‌طور غیرعادلانه‌ای بر نتیجه تأثیر می‌گذارند و ویژگی‌های با مقیاس کوچک‌تر (مثل امتیاز) کاملاً سرکوب می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>راه‌حل — نرمال‌سازی Min-Max:</b><br>
    مقیاس‌بندی Min-Max تمام مقادیر هر ویژگی را به یک بازه ثابت (معمولاً <b>۰ تا ۱</b>) تبدیل می‌کند تا تمامی ویژگی‌ها شانس برابری در مدل داشته باشند.
  </p>

  <div style="text-align: center; margin: 15px 0;">
    $$\text{Normalized}(X) = \frac{X - X_{\min}}{X_{\max} - X_{\min}}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>بصری‌سازی با Matplotlib:</b><br>
    با رسم نمودار قبل و بعد از نرمال‌سازی، به‌وضوح مشاهده می‌کنیم که چطور نوسانات ناهمگن داده‌های اولیه (قیمت تا ۱۰۰۰ و وزن تا ۲۰)، پس از مقیاس‌بندی همگی در یک دامنه یکسان بین <b>۰ و ۱</b> هم‌تراز می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> پس از نرمال‌سازی، تمام ویژگی‌ها در یک مقیاس یکسان قرار می‌گیرند و الگوریتم‌های یادگیری ماشین (مثل K-Nearest Neighbors) می‌توانند به‌درستی شباهت‌ها را تشخیص دهند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: تبدیل ویژگی‌های با مقیاس‌های متفاوت به مقیاس واحد [۰, ۱] و رسم بصری مقایسه
    </div>
  </p>

</div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})

plt.rcParams['figure.dpi'] = 135

np.random.seed(42)
```

```python
display_academic_text(
    "پیش‌پردازش داده‌ها: هم‌ترازی ویژگی‌ها از طریق مقیاس‌بندی کمینه-بیشینه (Min-Max Normalization) 📏", 
    "main_title"
)

display_academic_text(
    "مقیاس‌های متفاوت ویژگی‌ها به شدت بر عملکرد الگوریتم‌های مبتنی بر فاصله (نظیر KNN، K-Means یا سیستم‌های توصیه‌گر) تأثیر می‌گذارند. "
    "در این سناریو، ویژگی «قیمت» (با نوسان بسیار بالا از ۲۰ تا ۱۰۰۰) می‌تواند تأثیر ویژگی «امتیاز» (با نوسان بسیار کم از ۲ تا ۵) "
    "را در فضای محاسباتی کاملاً خنثی نماید. استراتژی <b>نرمال‌سازی Min-Max</b> با انتقالِ هندسیِ تمامی مقادیر به یک بازه مشترک $X_{norm} \in [0, 1]$، "
    "تضمین می‌کند که هر ویژگی وزنِ محاسباتیِ عادلانه‌ای در مدل یادگیری ماشین داشته باشد.", 
    "paragraph"
)

display_academic_text("معادله نرمال‌سازی Min-Max (تغییر مقیاس خطی):", "sub_title")

# Utilizing Raw Strings (r"") for correct LaTeX rendering of the normalization formula
display_academic_text(
    r"$$X_{norm} = \frac{X - X_{min}}{X_{max} - X_{min}}$$", 
    "paragraph"
)


# ==========================================================
# SECTION 2: DATA INITIALIZATION AND NORMALIZATION TRANSFORMATION
# ==========================================================
display_academic_text(
    "مرحله نخست: بارگذاری داده‌های ناهمگن و اعمال تبدیل Min-Max", 
    "sub_title"
)

display_academic_text(
    "در این فاز، داده‌های خامِ محصولات با مقیاس‌های کاملاً متمایز بارگذاری می‌شوند. سپس با بهره‌گیری از کلاس <code>MinMaxScaler</code>، "
    "تمامی ویژگی‌های عددی به طور مستقل پردازش شده و به <b>بازه استاندارد ۰.۰ تا ۱.۰</b> منتقل می‌گردند. حفظِ تناسبِ مقادیر (عدم تخریب توزیع داخلی داده‌ها) "
    "از مهم‌ترین ویژگی‌های این تبدیلِ ریاضیاتیِ خطی است.", 
    "paragraph"
)

# Step 1: Initialize the Amazon product dataset with vastly different scaling magnitudes
data = {
    'Product_ID': ['P101', 'P102', 'P103', 'P104', 'P105'],
    'Price_USD': [50.0, 120.0, 800.0, 20.0, 1000.0],  # Massive scale: [20, 1000]
    'Weight_KG': [0.5, 1.2, 15.0, 0.1, 20.0],         # Medium scale: [0.1, 20]
    'Rating': [4.0, 3.0, 5.0, 2.0, 4.0],              # Micro scale: [2, 5]
}
df_raw = pd.DataFrame(data)

# Step 2: Isolate the numeric features that require mathematical scaling
numeric_columns = ['Price_USD', 'Weight_KG', 'Rating']

# Step 3: Initialize the scaler and apply the fit_transform pipeline
# MinMaxScaler subtracts the minimum and divides by the range (max - min) for each column independently
scaler = MinMaxScaler()
df_normalized = df_raw.copy()
df_normalized[numeric_columns] = scaler.fit_transform(df_raw[numeric_columns])

# Display preparation to prevent empty index columns in HTML
df_raw_disp = df_raw.copy()
df_norm_disp = df_normalized.copy()

# Render the raw heterogeneous dataset and the standardized dataset side-by-side
display_side_by_side_datasets(
    [df_raw_disp, df_norm_disp],
    [
        "جدول ۱: پایگاه داده خام محصولات (مقیاس‌های به‌شدت نامتجانس)", 
        "جدول ۲: داده‌های نرمال‌سازی شده (هم‌ترازی در بازه استاندارد ۰ تا ۱)"
    ]
)

# ==========================================================
# SECTION 3: ANALYTICAL VISUALIZATION (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی - مقایسه بصری توزیع داده‌ها پیش و پس از تغییر مقیاس", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکی زیر تأثیرِ چشمگیرِ نرمال‌سازی را به تصویر می‌کشد. در نمودار سمت چپ، سیطرهٔ مقادیرِ نجومیِ «قیمت» به حدی است که نوساناتِ متغیرهای "
    "«وزن» و «امتیاز» به صورت خطوطی کاملاً صاف در کفِ محور مختصات فشرده و محو شده‌اند (Scale Distortion).<br><br>"
    "در نمودار سمت راست، پس از انتقال هندسیِ داده‌ها به فضای $[0, 1]$، مشاهده می‌شود که تمامی ویژگی‌ها استقلال و پویاییِ نوسانیِ خود را بازیافته‌اند و "
    "الگوریتم‌های هوش مصنوعی اکنون قادرند الگوهای پنهان هر سه متغیر را به صورت هم‌زمان درک نمایند.", 
    "paragraph"
)

# Initialize a wide dual-axis canvas for visual comparative analytics
fig, axes = plt.subplots(1, 2, figsize=(19, 6.5), dpi=135)

# Define an explicit Academic Blue color palette for consistent feature tracking
colors = {'Price_USD': '#2c3e50', 'Weight_KG': '#2980b9', 'Rating': '#1abc9c'} # Navy, Classic Blue, Turquoise
markers = {'Price_USD': 'o', 'Weight_KG': 's', 'Rating': '^'}
line_styles = {'Price_USD': '-', 'Weight_KG': '--', 'Rating': ':'}

# ----------------------------------------------------------
# Plot 1: Original Data Scale (Feature Suppression)
# ----------------------------------------------------------
for col in numeric_columns:
    axes[0].plot(
        df_raw['Product_ID'], 
        df_raw[col], 
        marker=markers[col], 
        markersize=12, 
        linewidth=3.0, 
        color=colors[col], 
        linestyle=line_styles[col],
        label=col, 
        alpha=0.9
    )

axes[0].annotate(
    "Feature Suppression:\nThe Massive scale of 'Price' (up to $1000)\ncompletely crushes the 'Rating' line (2 to 5)\ninto a flat, indistinguishable line at the bottom.",
    xy=(2.0, 5.0), xytext=(0.5, 350),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title('Figure 1: Raw Data Topology (Severe Scale Distortion)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[0].set_xlabel('Product Identifiers', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_ylabel('Absolute Raw Values', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_xticklabels(df_raw['Product_ID'], fontweight='bold', fontsize=11)
axes[0].legend(title='Dataset Features', loc='upper left', frameon=True, fontsize=11)
axes[0].grid(True, linestyle='--', alpha=0.5)

# ----------------------------------------------------------
# Plot 2: Normalized Data Scale (Feature Alignment)
# ----------------------------------------------------------
for col in numeric_columns:
    axes[1].plot(
        df_normalized['Product_ID'], 
        df_normalized[col], 
        marker=markers[col], 
        markersize=12, 
        linewidth=3.0, 
        color=colors[col], 
        linestyle=line_styles[col],
        label=col, 
        alpha=0.9
    )

axes[1].annotate(
    "Hidden Dynamics Revealed:\nAfter Min-Max Scaling, we clearly see that\nP103 (Index 2) actually has the Highest Rating (1.0)\ndespite being the 2nd most expensive item.",
    xy=(2.0, 1.0), xytext=(0.1, 0.75),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[1].set_title('Figure 2: Normalized Topology (Equal Representation in [0, 1])', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[1].set_xlabel('Product Identifiers', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_ylabel('Scaled Mathematical Value (Min=0.0, Max=1.0)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_ylim(-0.05, 1.05) # Bound y-axis strictly around [0, 1]
axes[1].set_xticklabels(df_normalized['Product_ID'], fontweight='bold', fontsize=11)
axes[1].legend(title='Dataset Features', loc='center left', bbox_to_anchor=(1, 0.5), frameon=True, fontsize=11)
axes[1].grid(True, linestyle='--', alpha=0.5)

# Clear top and right spines for a clean academic aesthetic
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی عملیات (Strategic Implication):</b><br><br>"
    r"تغذیهٔ داده‌های خام با مقیاس‌های نامتجانس (مثلاً ترکیبِ قیمتِ ۱۰۰۰ دلاری با امتیازِ ۲) به الگوریتم‌های داده‌کاوی، <b>خطای تحلیلی شدیدی</b> به همراه دارد. "
    r"ماشین تصور می‌کند که «قیمت» ۵۰۰ برابر مهم‌تر از «امتیاز» است، در حالی که این صرفاً تفاوت در واحد اندازه‌گیری است!<br><br>"
    r"استراتژی <b>نرمال‌سازی Min-Max</b> با اجرای یک انتقال هندسیِ خطی، بدون تخریب توزیعِ درونی و نوساناتِ داده‌ها، تمام ویژگی‌ها را در یک جبههٔ تحلیلی هم‌سطح $X_{norm} \in [0, 1]$ قرار می‌دهد "
    r"(گراف ۲). این پیش‌پردازش، پیش‌نیازِ قطعیِ استقرار الگوریتم‌های توصیه‌گر (Recommender Systems)، شبکه‌های عصبی عمیق (Deep Learning) و روش‌های خوشه‌بندی مبتنی بر فاصله (نظیر K-Means) می‌باشد."
)
display_academic_text(insight_conclusion, "insight")
```
