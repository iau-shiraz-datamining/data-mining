  <span style="font-size: 20pt;"></span> 2.3.8 Measuring similar distributions: the Kullback-Leibler divergence
</div>

<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: کشف انحراف مدل و تغییر رفتار جستجو در «گوگل» 📉
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    تیم داده‌کاوی موتور جستجوی «گوگل» یک **توزیع احتمال مرجع ($Q$)** از موضوعات جستجوی کاربران در شرایط عادی دارد (مثلاً ۶۰٪ اخبار عمومی، ۳۰٪ سرگرمی، ۱۰٪ سلامت). اما ناگهان یک بحران جهانی (مثل یک پاندمی) رخ می‌دهد و **توزیع واقعی جستجوها ($P$)** به شکل شدیدی تغییر می‌کند (مثلاً ۷۰٪ سلامت، ۲۰٪ اخبار، ۱۰٪ سرگرمی). گوگل چگونه تفاوت این دو توزیع احتمال را اندازه‌گیری می‌کند؟ با **واگرایی کولبک-لایبلر (KL Divergence)**.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام اول: تعریف واگرایی KL</b><br>
    واگرایی KL میزان <b>اطلاعات تلف‌شده (Information Loss)</b> را در صورتی که توزیع واقعی $P$ را با توزیع پیش‌فرض $Q$ تقریب بزنیم، محاسبه می‌کند:
  </p>

  <div style="text-align: center; margin: 15px 0;">
    $$D_{\text{KL}}(P \parallel Q) = \sum_{x \in X} P(x) \log \left( \frac{P(x)}{Q(x)} \right)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام دوم: چرا واگرایی KL یک «فاصله ریاضی واقعی» نیست؟</b><br>
    واگرایی KL به دو دلیل مهم یک متریک فاصله اصیل نیست:
  </p>
  <ul style="text-align: right; margin: 0 0 10px; padding-right: 22px;">
    <li><b>نامتقارن بودن (Asymmetric):</b> واگرایی $P$ نسبت به $Q$ با واگرایی $Q$ نسبت به $P$ برابر نیست! ($D_{\text{KL}}(P \parallel Q) \neq D_{\text{KL}}(Q \parallel P)$).</li>
    <li><b>عدم پیروی از نامساوی مثلث:</b> فاصله مستقیم بین دو توزیع لزوماً کوتاه‌تر از مسیر واسط نیست.</li>
  </ul>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام سوم: تفسیر اعداد خروجی</b><br>
    &nbsp;&nbsp;&nbsp;• <b>$D_{\text{KL}} = 0$:</b> هر دو توزیع احتمال کاملاً یکسان هستند ($P = Q$).<br>
    &nbsp;&nbsp;&nbsp;• <b>$D_{\text{KL}} > 0$:</b> هرچه عدد بزرگ‌تر باشد، تفاوت و انحراف بین دو توزیع احتمال بیشتر است.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> وقتی گوگل ببیند $D_{\text{KL}}(P \parallel Q)$ از یک آستانه‌ای بالاتر رفته است، متوجه <b>Data Drift (انحراف داده‌ها)</b> می‌شود و بلافاصله مدل‌های هوش مصنوعی پیش‌بینی جستجو و تبلیغات خود را مجدداً آموزش می‌دهد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: سنجش عدم تشابه و میزان انحراف بین دو توزیع احتمال مستقل
    </div>
  </p>

</div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import entropy 

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})

plt.rcParams['figure.dpi'] = 135

np.random.seed(42)
```

```python
display_academic_text(
    "سنجش انحراف توزیع احتمالات: تشخیص Data Drift با واگرایی کولبک-لایبلر (Kullback-Leibler) 📉", 
    "main_title"
)

display_academic_text(
    "در تحلیل الگوهای کلانِ داده، توزیع ویژگی‌ها در طول زمان ثابت نمی‌ماند. در این سناریو، یک موتور جستجو نیازمند بررسی تغییرات ناگهانی (Data Drift) "
    "در رفتار کاربران در پی یک <i>بحران جهانی</i> است. <b>واگرایی کولبک-لایبلر (KL Divergence)</b> میزان اتلاف اطلاعات در صورتِ جایگزینیِ توزیع واقعی و جدید ($P$) "
    "با توزیع مرجع و پیش‌فرض ($Q$) را از منظرِ نظریه اطلاعات (Information Theory) مدل‌سازی می‌نماید.", 
    "paragraph"
)

display_academic_text(
    "معادله واگرایی KL برای توزیع‌های گسسته:", 
    "sub_title"
)

# Utilizing Raw Strings (r"") for exact LaTeX rendering in Jupyter
display_academic_text(
    r"$$D_{KL}(P \parallel Q) = \sum_{x \in \mathcal{X}} P(x) \log \left( \frac{P(x)}{Q(x)} \right)$$", 
    "paragraph"
)

display_academic_text(
    "اثبات نامتقارن بودن واگرایی KL (عدم برخورداری از خاصیتِ متریکِ هندسی مانند فاصله اقلیدسی):", 
    "sub_title"
)
display_academic_text(
    r"$$D_{KL}(P \parallel Q) \neq D_{KL}(Q \parallel P)$$", 
    "paragraph"
)

# ==========================================================
# SECTION 2: PROBABILITY VECTORS & KL COMPUTATION
# ==========================================================
display_academic_text(
    "مرحله نخست: بارگذاری بردارهای احتمال و سنجش میزان انحراف", 
    "sub_title"
)

display_academic_text(
    "در این فاز، سه توزیع احتمال مختلف تعریف می‌گردد. سپس، الگوریتم آنتروپی، واگراییِ توزیعِ بحرانی ($P$) نسبت به مرجع ($Q$) را محاسبه نموده "
    "و به منظور اثبات نامتقارن بودن این شاخص، <i>واگرایی معکوس ($Q$ نسبت به $P$)</i> نیز استخراج می‌شود.", 
    "paragraph"
)

# Step 1: Define the state space of categorical search topics
topics = ['News', 'Entertainment', 'Health']

# Step 2: Define probability distribution vectors
# The sum of elements in each vector must strictly equal 1.0 (Valid Probability Mass Functions)
# P: Actual observed distribution during a crisis (e.g., Pandemic)
P = np.array([0.20, 0.10, 0.70]) 
# Q: Baseline normal expected distribution
Q = np.array([0.60, 0.30, 0.10]) 
# R: Minor natural fluctuation from baseline
R = np.array([0.58, 0.32, 0.10]) 

# Construct a DataFrame to hold the distributions for reporting
df_distributions = pd.DataFrame(
    [P, Q, R], 
    index=['P (Actual Crisis)', 'Q (Baseline Normal)', 'R (Natural Fluctuation)'], 
    columns=topics
)
df_distributions.index.name = 'Distribution State'

# Step 3: Compute KL Divergences using scipy.stats.entropy
# Note: entropy(pk, qk) computes D_KL(pk || qk). It uses natural logarithm (Nats) by default.
kl_P_Q = entropy(P, Q)
kl_Q_P = entropy(Q, P)
kl_Q_R = entropy(Q, R)

# Step 4: Package the divergence results into a structured analytical dataframe
df_kl_results = pd.DataFrame({
    'Mathematical Notation': [
        'D_KL ( P || Q )', 
        'D_KL ( Q || P )', 
        'D_KL ( R || Q )'
    ],
    'Analytical Context': [
        'Crisis Shift vs Baseline (Drift!)', 
        'Baseline vs Crisis (Asymmetry Test)', 
        'Natural Fluctuation vs Baseline'
    ],
    'Divergence Score (Nats)': [kl_P_Q, kl_Q_P, kl_Q_R]
})

# Display preparation to prevent empty index columns in HTML
df_dist_disp = df_distributions.reset_index()

# Render the probability distributions and computed divergence scores side-by-side
display_side_by_side_datasets(
    [df_dist_disp, df_kl_results],
    [
        "جدول ۱: بردارهای توزیع احتمالِ جستجو (PMF)", 
        "جدول ۲: تحلیل واگرایی کولبک-لایبلر (Kullback-Leibler)"
    ]
)

# ==========================================================
# SECTION 3: ANALYTICAL VISUALIZATION (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلیِ شیفت احتمالات در فضای موضوعی", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکی زیر، تحول رفتاری کاربران را به تصویر می‌کشد. نمودار میله‌ای به وضوح نشان می‌دهد که توزیع مرجع ($Q$) و نوسان طبیعی ($R$) دارای ساختاری هم‌تراز هستند "
    "(واگرایی نزدیک به صفر). اما در توزیع بحرانی ($P$)، <b>تسلط ناگهانیِ جستجوهای مرتبط با «سلامت»</b> منجر به تغییر فاحش ساختار توزیع گردیده "
    "که مستقیماً به ثبت عدد واگراییِ هشداردهنده ($1.29$ Nats) منتهی شده است.", 
    "paragraph"
)

# Initialize a wide canvas optimized for comparing probability distributions
plt.figure(figsize=(14, 6.5), dpi=135)

# Prepare data for plotting by melting the dataframe into long format for seaborn
df_melted = df_distributions.reset_index().melt(
    id_vars='Distribution State', 
    var_name='Search Topic', 
    value_name='Probability Mass'
)

# Plot grouped bar chart to compare the three distributions side-by-side per topic
# Palette: Red for Crisis, Navy Blue for Baseline, Turquoise for Normal Fluctuation
sns.barplot(
    data=df_melted, 
    x='Search Topic', 
    y='Probability Mass', 
    hue='Distribution State',
    palette=['#e74c3c', '#2c3e50', '#1abc9c'], 
    edgecolor='black',
    linewidth=1.5
)

# Intelligent Annotations to explain the mathematics graphically
plt.annotate(
    "Massive Concept Drift:\nDuring the crisis, users abandoned News and Entertainment,\nfocusing 70% of their searches solely on 'Health'.\nThis creates a massive KL Divergence of 1.29!",
    xy=(2.0, 0.70), xytext=(0.5, 0.65),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

# Enhance plot labeling, titles, and structural configuration
plt.title('Figure 1: Comparison of Search Topic Probability Distributions (PMF)', fontdict={'fontsize': 15, 'fontweight': 'bold'}, pad=15)
plt.xlabel('Categorical Search Topic Space', fontdict={'fontsize': 13, 'fontweight': 'bold'})
plt.ylabel('Observed Probability (0.0 to 1.0)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
plt.ylim(0, 0.85)

# Add gridlines on the Y-axis for precise value reading
plt.grid(axis='y', linestyle='--', alpha=0.5, zorder=0)

# Make tick labels bold
plt.xticks(fontweight='bold', fontsize=12)

# Move the legend outside the plot to prevent obscuring the data bars
plt.legend(
    title='Distribution Model', 
    title_fontproperties={'weight': 'bold'},
    loc='upper right', 
    framealpha=1.0, 
    edgecolor='#bdc3c7',
    fontsize=11
)

# Clear top and right spines for a clean academic aesthetic
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی عملیات (Strategic Implication):</b><br>"
    r"۱. <b>نظارت بر سلامت مدل (MLOps):</b> محاسبات اثبات می‌نماید که واگرایی KL ابزاری قدرتمند برای نظارت بر سلامت مدل‌های یادگیری ماشین در مرحلهٔ Production است. "
    r"افزایش این شاخص فراتر از آستانه‌های مجاز (مانند جهش به ۱.۲۹ در زمان بحران)، به عنوان یک آژیر هشدار برای رخداد <b>انحراف داده‌ها (Data Drift / Concept Drift)</b> عمل می‌کند؛ "
    r"به این معنا که مدل هوش مصنوعیِ ما که بر اساس توزیع نرمال ($Q$) آموزش دیده بود، اکنون منقضی شده است و باید فوراً با داده‌های جدید ($P$) مجدداً آموزش (Retrain) ببیند.<br><br>"
    r"۲. <b>عدم تقارن (Asymmetry):</b> همچنین عدم تساوی مقادیر $D_{KL}(P \parallel Q) = 1.29$ و $D_{KL}(Q \parallel P) = 0.98$ از نظر ریاضی تأیید نمود که این شاخص یک «متریک فاصلهٔ هندسی» نبوده، "
    r"بلکه معیاری مبتنی بر نظریه اطلاعات برای سنجش <i>اتلافِ داده‌ها و میزان غافلگیری (Surprise)</i> است."
)
display_academic_text(insight_conclusion, "insight")
```
