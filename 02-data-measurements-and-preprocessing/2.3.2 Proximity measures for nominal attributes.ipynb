  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: سنجش شباهت سلیقه کاربران در «نتفلیکس» (Netflix) 🏷️
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    سیستم هوشمند «نتفلیکس» می‌خواهد ببیند سلیقه فیلم دیدن دو کاربر چقدر به هم نزدیک است. متغیرهای مورد بررسی، **ژانرها و برچسب‌های متنی (Nominal)** مانند <code>Action</code>، <code>Sci-Fi</code>، <code>Comedy</code> و <code>Drama</code> هستند. چون این ویژگی‌ها هیچ ترتیب ریاضی یا مقدار عددی ندارند، نمی‌توان از فاصله‌های معمولی (مثل اقلیدسی) استفاده کرد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام اول: چرا روش‌های عددی معمولی در اینجا شکست می‌خورند؟</b><br>
    اگر به ژانر اکشن عدد ۱، کمدی عدد ۲ و درام عدد ۳ بدهیم، به مدل هوش مصنوعی این اشتباه را القا کرده‌ایم که «درام» سه برابر «اکشن» ارزش دارد! برای ویژگی‌های اسمی، شباهت بر اساس **اشتراک و عدم اشتراک برچسب‌ها** سنجیده می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام دوم: فرمول شباهت جاکارد (Jaccard Similarity)</b><br>
    شباهت جاکارد میزان همپوشانی دو مجموعه برچسب را حساب می‌کند:
  </p>


  <div style="text-align: center; margin: 15px 0;">
    $$\text{Jaccard Similarity}(A, B) = \frac{|A \cap B|}{|A \cup B|} = \frac{\text{Intersection of Genres}}{\text{Union of All Genres}}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 10px 0;">
    &nbsp;&nbsp;&nbsp;• <b>کاربر A:</b> {اکشن، علمی-تخیلی، هیجان‌انگیز}<br>
    &nbsp;&nbsp;&nbsp;• <b>کاربر B:</b> {اکشن، علمی-تخیلی، کمدی، درام}<br>
    &nbsp;&nbsp;&nbsp;• <b>اشتراک (ژانرهای مشترک):</b> {اکشن، علمی-تخیلی} 👈 ۲ ژانر<br>
    &nbsp;&nbsp;&nbsp;• <b>اجتماع (کل ژانرها):</b> {اکشن، علمی-تخیلی، هیجان‌انگیز، کمدی، درام} 👈 ۵ ژانر<br>
    &nbsp;&nbsp;&nbsp;• <b>شباهت جاکارد:</b> $2 / 5 = 0.40$ (۴۰٪ شباهت سلیقه)
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام سوم: عدم تشابه جاکارد (Jaccard Distance)</b><br>
    برای تبدیل این شباهت به یک معیار فاصله (عدم تشابه)، کافی است آن را از عدد ۱ کم کنیم:<br>
    <div style="text-align:center; font-weight:bold; margin-top: 6px;">
      $$\text{Jaccard Distance} = 1 - 0.40 = 0.60$$
    </div>
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> هرچه شباهت جاکارد به ۱ نزدیک‌تر باشد (مثلاً ۰.۸۵)، نتفلیکس فیلم‌های محبوب کاربر A را به کاربر B پیشنهاد می‌دهد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: سنجش دقیق میزان همپوشانی ویژگی‌های اسمی بدون نیاز به فرضیات عددی کاذب
    </div>
  </p>

</div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.spatial.distance import pdist, squareform
from matplotlib.colors import ListedColormap

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})

plt.rcParams['figure.dpi'] = 135

# Fix random seed for exact reproducibility
np.random.seed(42)
```

```python
display_academic_text(
    "سنجش عدم تشابه در داده‌های اسمی: ضریب و فاصله جاکارد (Jaccard Index) 🎬", 
    "main_title"
)

display_academic_text(
    "مقایسه ویژگی‌های اسمی و مجموعه‌ای (مانند ژانرهای فیلم) از طریق توابع فاصله متریک (نظیر اقلیدسی) امکان‌پذیر نیست، "
    "چرا که هیچ ترتیب یا ارزش عددی ذاتی میان آن‌ها وجود ندارد. در این سیستم توصیه‌گر، سلیقه کاربران از طریق <b>«ضریب شباهت جاکارد»</b> سنجیده می‌شود "
    "که برابر است با نسبت اندازهٔ اشتراک (Intersection) به اندازهٔ اجتماع (Union) مجموعه‌ها:", 
    "paragraph"
)

# Strict LaTeX rendering for math formulas
display_academic_text(r"$$J(A, B)=\frac{|A \cap B|}{|A \cup B|}$$", "paragraph")

display_academic_text(
    "به منظور ورود این متغیرها به الگوریتم‌های فاصله‌محورِ هوش مصنوعی، ماتریس شباهت به <b>ماتریس فاصله (عدم تشابه) جاکارد</b> تبدیل می‌گردد:", 
    "paragraph"
)

display_academic_text(r"$$d_J(A, B)=1-J(A, B)$$", "paragraph")

# ==========================================================
# SECTION 2: DATA MATRIX CONSTRUCTION & JACCARD COMPUTATION
# ==========================================================
display_academic_text(
    "مرحله نخست: تولید ماتریس باینری و استخراج فواصل جاکارد", 
    "sub_title"
)

display_academic_text(
    "در این فاز، فهرست تماشای کاربران به یک <b>ماتریس نشانگر باینری (Binary Indicator Matrix)</b> تبدیل می‌شود (۱ به معنای مشاهده و ۰ به معنای عدم مشاهده). "
    "سپس، الگوریتم به صورت خودکار فاصله جاکارد را برای تمامی جفت کاربران محاسبه نموده و در نهایت، <i>ماتریس شباهت</i> به عنوان مکملِ ماتریس فاصله استخراج می‌گردد.", 
    "paragraph"
)

# Step 1: Initialize the binary indicator matrix representing nominal tags (Watched Genres)
data = {
    'Action':   [1, 1, 0, 0],
    'Sci-Fi':   [1, 1, 0, 0],
    'Thriller': [1, 0, 1, 0],
    'Comedy':   [0, 1, 1, 1],
    'Drama':    [0, 1, 0, 1]
}
users = ['User_A', 'User_B', 'User_C', 'User_D']
df_nominal = pd.DataFrame(data, index=users)
df_nominal.index.name = 'User_ID'

# Step 2: Compute Pairwise Jaccard Distance (Dissimilarity)
# pdist calculates the Jaccard distance based on the proportion of mismatched items to the union of observed items
jaccard_dist_condensed = pdist(df_nominal.values, metric='jaccard')

# Convert the condensed array into a square symmetric dissimilarity matrix
df_jaccard_dist = pd.DataFrame(
    squareform(jaccard_dist_condensed),
    index=users,
    columns=users
)
df_jaccard_dist.index.name = 'User_ID'

# Step 3: Derive the Jaccard Similarity Matrix (1 - Distance)
df_jaccard_sim = 1 - df_jaccard_dist

# Prepare clean copies for displaying without hiding the index
df_nom_disp = df_nominal.reset_index()
df_dist_disp = df_jaccard_dist.reset_index()
df_sim_disp = df_jaccard_sim.reset_index()

# Render the three data structures side-by-side
display_side_by_side_datasets(
    [df_nom_disp, df_dist_disp, df_sim_disp],
    [
        "جدول ۱: ماتریس باینری ویژگی‌های اسمی (ژانرها)", 
        "جدول ۲: ماتریس فاصله جاکارد (عدم تشابه)",
        "جدول ۳: ماتریس شباهت جاکارد (همپوشانی سلیقه)"
    ]
)

# ==========================================================
# SECTION 3: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی و کشف بصری شباهت‌ها (Visual Analytics)", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکی زیر، منشأ استخراج شباهت‌ها را به تصویر می‌کشد. در <b>نمودار سمت چپ</b>، رفتار خام کاربران ثبت شده است. "
    "در <b>نقشه حرارتی سمت راست</b>، ماتریس شباهت جاکارد رسم شده است. قطر اصلی (با شباهت ۱.۰) نشان‌دهندهٔ تطابق ۱۰۰ درصدی هر کاربر با خودش است. "
    "با بررسی طیف رنگی، مشخص می‌گردد که User_A و User_B دارای بیشترین همپوشانیِ سلیقه می‌باشند.", 
    "paragraph"
)

# Initialize a wide canvas for side-by-side structural comparison
fig, axes = plt.subplots(1, 2, figsize=(19, 6.5), dpi=135)

# --- CHART 1: Original Binary Matrix ---
# Custom binary colormap: Light gray for 0, Navy Blue for 1
cmap_binary = ListedColormap(['#ecf0f1', '#2c3e50'])

sns.heatmap(
    df_nominal, 
    annot=True, 
    cmap=cmap_binary, 
    fmt='d', 
    ax=axes[0], 
    linewidths=2.0, 
    linecolor='white',
    cbar=False,
    annot_kws={"size": 16, "weight": "bold"}
)

axes[0].annotate(
    "Intersection Logic:\nUser A and User B both watched\nAction and Sci-Fi. Their intersection\nsize is 2. Their union size is 5.",
    xy=(1.5, 0.5), xytext=(2.5, 2.5),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title('Figure 1: Raw Binary Feature Space (User vs. Genre)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[0].set_xlabel('Nominal Features (Movie Genres)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_ylabel('Target Customer', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_xticklabels(axes[0].get_xticklabels(), fontweight='bold', fontsize=11, rotation=0)
axes[0].set_yticklabels(axes[0].get_yticklabels(), fontweight='bold', fontsize=11, rotation=0, va='center')

# --- CHART 2: Jaccard Similarity Matrix ---
# Using Academic Blues palette to highlight similarities
sns.heatmap(
    df_jaccard_sim, 
    annot=True, 
    cmap='Blues', 
    fmt='.2f', 
    vmin=0, vmax=1, 
    ax=axes[1], 
    linewidths=2.0, 
    linecolor='white',
    cbar_kws={'label': 'Jaccard Similarity Coefficient'},
    annot_kws={"size": 14, "weight": "bold"}
)

axes[1].annotate(
    "Highest Similarity Match:\n2 / 5 = 0.40 (40% Overlap)\nUser A and B should be clustered together.",
    xy=(1.5, 0.5), xytext=(2.5, 2.5),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[1].set_title('Figure 2: Topology of User Jaccard Similarity Matrix', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[1].set_xlabel('Target User (Object j)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_ylabel('Reference User (Object i)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_xticklabels(axes[1].get_xticklabels(), fontweight='bold', fontsize=11, rotation=0)
axes[1].set_yticklabels(axes[1].get_yticklabels(), fontweight='bold', fontsize=11, rotation=0, va='center')

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی (Strategic Implication):</b><br>"
    r"تبدیل ویژگی‌های اسمی/متنی به بردارهای باینری و بهره‌گیری از <b>معیار جاکارد</b>، امکان سنجش دقیقِ شباهتِ سلیقه را در سیستم‌های بازیابی اطلاعات (Information Retrieval) فراهم می‌سازد. "
    r"در این ساختار محاسباتی، <i>«عدم وجود یک ژانر در سلیقه دو کاربر»</i> (صفرهای مشترک)، به عنوان وجه اشتراکِ آن‌ها لحاظ نمی‌گردد (ویژگی‌های دودویی نامتقارن). "
    r"این رویکرد، هستهٔ مرکزی و نقطهٔ تمایزِ الگوریتم‌های <b>توصیه‌گر مشارکتی (Collaborative Filtering)</b> در برابر روش‌های فاصله‌یابی کلاسیک (نظیر اقلیدسی) می‌باشد."
)
display_academic_text(insight_conclusion, "insight")
```
