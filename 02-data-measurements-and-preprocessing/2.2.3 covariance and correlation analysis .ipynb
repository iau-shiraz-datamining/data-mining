  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: تحلیل الگوی شنیداری و کشف روابط در «اسپاتیفای» (Spotify) 📈
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    تیم داده‌کاوی «اسپاتیفای» می‌خواهد بداند آیا بین «تعداد لایک کردن آهنگ‌ها» یا «نرخ رد کردن (Skip) آهنگ» با «کل ساعات استفاده از اپلیکیشن» رابطه‌ای وجود دارد یا خیر؟ برای پاسخ به این سوال و کشف وابستگی بین متغیرها، از کوواریانس و ضریب همبستگی استفاده می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۱. کوواریانس (Covariance) — جهت حرکت هم‌زمان متغیرها</b><br>
    کوواریانس نشان می‌دهد که آیا دو متغیر هم‌جهت تغییر می‌کنند یا در جهت مخالف:<br>
    &nbsp;&nbsp;&nbsp;• <b>کوواریانس مثبت:</b> با افزایش «تعداد لایک‌ها»، «ساعات گوش دادن» نیز افزایش می‌یابد.<br>
    &nbsp;&nbsp;&nbsp;• <b>کوواریانس منفی:</b> با افزایش «نرخ اسکپ/رد کردن آهنگ»، «ساعات گوش دادن» کاهش می‌یابد.<br>
    <i>ضعف کوواریانس:</i> عدد کوواریانس وابسته به واحد اندازه‌گیری است (مثلاً عدد ۶۵۰.۴ به دست می‌آید)؛ بنابراین نمی‌توان فهمید این رابطه چقدر قوی یا ضعیف است!
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۲. همبستگی پیرسون (Pearson Correlation) — مقیاس استاندارد و بی‌واحد (-1 تا +1)</b><br>
    برای حل مشکل مقیاس کوواریانس، ضریب همبستگی آن را نرمال‌سازی کرده و در بازه بین <b>۱- تا ۱+</b> قرار می‌دهد:<br>
    &nbsp;&nbsp;&nbsp;• <b>همبستگی ۰.۹۵+ (مثبت قوی):</b> رابطه مستقیم و فوق‌العاده شدید بین لایک کردن و زمان استفاده.<br>
    &nbsp;&nbsp;&nbsp;• <b>همبستگی ۰.۸۸- (منفی قوی):</b> رابطه معکوس شدید بین نرخ اسکپ بالا و کاهش زمان استفاده.<br>
    &nbsp;&nbsp;&nbsp;• <b>همبستگی ۰.۰۲ (نزدیک به صفر):</b> عدم وجود رابطه خطی (مثلاً بین قد کاربر و تعداد پادکست‌های گوش داده‌شده!).
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۳. کاربرد در تصمیم‌گیری استراتژیک</b><br>
    وقتی همبستگی مثبت شدید (۰.۹۵+) اثبات شد، اسپاتیفای دکمه «Heart/Like» را در همه‌جای رابط کاربری دم‌دست می‌گذارد تا با تشویق کاربر به لایک، زمان ماندگاری او را در اپلیکیشن بالا ببرد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> کوواریانس جهت تغییرات را مشخص می‌کند، اما ضریب همبستگی قدرت دقیق و استانداردشده‌ی این رابطه را برای الگوریتم‌های هوش مصنوعی شفاف می‌سازد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: سنجش جهت و شدت رابطه خطی بین متغیرها در بازه مقیاس‌شده [-1, +1]
    </div>
  </p>

</div>

```python
display_academic_text(
    "آمار توصیفی: تحلیل وابستگی متغیرها از طریق کوواریانس و ضریب همبستگی پیرسون 📊", 
    "main_title"
)

display_academic_text(
    "بر اساس اصول کاوش داده‌ها، شناسایی روابط خطی میان متغیرها امری حیاتی جهت استخراج الگوهای رفتاری است. "
    "در این سناریو، هدف ارزیابی وابستگی میان <i>نرخ تعامل کاربر (تعداد لایک‌ها و نرخ پرش)</i> با <i>میزان ماندگاری (ساعات شنیدن موسیقی)</i> "
    "در یک پلتفرم استریم است. در این راستا، محدودیت‌های مقیاسیِ شاخصِ <b>کوواریانس (Covariance)</b> تحلیل گردیده و سپس "
    "رویکرد استانداردسازی‌شدهٔ <b>همبستگی پیرسون (Pearson Correlation)</b> جهت کشف شدت و جهت دقیق این روابط پیاده‌سازی می‌شود.", 
    "paragraph"
)

# ==========================================================
# SECTION 2: DATA INITIALIZATION AND MATRIX COMPUTATION
# ==========================================================
display_academic_text(
    "مرحله نخست: بارگذاری داده‌های خام و محاسبه ماتریس‌های وابستگی", 
    "sub_title"
)

display_academic_text(
    "در این فاز، داده‌های فعالیت کاربران بارگذاری می‌شود. ستون شناسه کاربری (User_ID) جهت جلوگیری از ایجاد خطای محاسباتی "
    "از فضای پردازش حذف گردیده است. سپس <b>ماتریس کوواریانس</b> (حاوی مقادیر وابسته به مقیاس) و <b>ماتریس همبستگی پیرسون</b> "
    "(نرمال‌سازی شده در بازه استاندارد [-1, +1]) محاسبه و در کنار یکدیگر رندر می‌شوند.", 
    "paragraph"
)

# Step 1: Initialize the mock database for user activity on the streaming platform
raw_data = {
    'User_ID': [101, 102, 103, 104, 105],
    'Liked_Songs': [12, 45, 85, 110, 150],           
    'Listening_Hours': [5.5, 18.0, 32.0, 40.5, 55.0], 
    'Skip_Rate_Percent': [45, 30, 15, 10, 5]        
}
df_raw = pd.DataFrame(raw_data)

# Step 2: Isolate the analytical features by dropping irrelevant identifiers (User_ID)
df_analytical = df_raw.drop('User_ID', axis=1)

# Step 3: Compute the Covariance Matrix
# Reset index and name the column explicitly to guarantee clean HTML rendering
cov_matrix = df_analytical.cov().round(2)
cov_display = cov_matrix.reset_index().rename(columns={'index': 'Feature Attribute'})

# Step 4: Compute the Pearson Correlation Matrix
# Standardizes covariance by the product of standard deviations, yielding [-1, 1] range
corr_matrix = df_analytical.corr().round(3)
corr_display = corr_matrix.reset_index().rename(columns={'index': 'Feature Attribute'})

# Render the raw observation data alongside the extracted mathematical matrices
display_side_by_side_datasets(
    [df_raw, cov_display, corr_display],
    [
        "جدول ۱: رکوردهای ثبت‌شده فعالیت کاربران", 
        "جدول ۲: ماتریس کوواریانس (وابسته به مقیاس)",
        "جدول ۳: ماتریس همبستگی پیرسون (بازه نرمال)"
    ]
)

# ==========================================================
# SECTION 3: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی وابستگی‌های خطی و ساختار همبستگی", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکی زیر شامل سه پنل تحلیلی است. پنل نخست، ماتریس همبستگی را در قالب <b>نقشه حرارتی (Heatmap)</b> نشان می‌دهد. "
    "دو پنل بعدی، نمودارهای پراکندگی مجهز به <b>خط رگرسیون</b> هستند که شیب صعودی (همبستگی مثبت قوی بین لایک و زمان) "
    "و شیب نزولی (همبستگی منفی قوی بین پرش و زمان) را به صورت عینی اثبات می‌نمایند.", 
    "paragraph"
)

# Initialize a wide canvas with three distinct subplots for visual analytics
fig, axes = plt.subplots(1, 3, figsize=(20, 6), dpi=135)

# --- SUBPLOT 1: Pearson Correlation Heatmap ---
sns.heatmap(
    corr_matrix, 
    annot=True, 
    cmap='Blues', 
    vmin=-1, 
    vmax=1, 
    ax=axes[0], 
    square=True, 
    fmt='.2f', 
    linewidths=2.0, 
    linecolor='white',
    cbar_kws={"shrink": .8},
    annot_kws={"size": 13, "weight": "bold"}
)
axes[0].set_title('Figure 1: Pearson Correlation Heatmap', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=20, ha='right', fontweight='bold')
axes[0].set_yticklabels(axes[0].get_yticklabels(), fontweight='bold')

# --- SUBPLOT 2: Strong Positive Correlation (Likes vs Hours) ---
sns.regplot(
    data=df_analytical, 
    x='Liked_Songs', 
    y='Listening_Hours', 
    ax=axes[1], 
    color='#1abc9c', # Turquoise for positive engagement
    scatter_kws={'s': 160, 'edgecolor': '#2c3e50', 'linewidths': 1.5}, 
    line_kws={'color': '#2c3e50', 'linestyle': '--', 'linewidth': 2.5}
)

r_pos = corr_matrix.loc['Liked_Songs', 'Listening_Hours']
axes[1].annotate(
    f"Near-Perfect Positive Linearity:\nr = +{r_pos:.2f}\nMore likes strongly indicate\nhigher platform retention.",
    xy=(85, 32), xytext=(20, 42),
    fontsize=10.5, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[1].set_title('Figure 2: Positive Trend (Likes vs Hours)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[1].set_xlabel('Volume of Liked Songs (Count)', fontdict={'fontsize': 12, 'fontweight': 'bold'})
axes[1].set_ylabel('Total Listening Hours', fontdict={'fontsize': 12, 'fontweight': 'bold'})
axes[1].grid(True, linestyle='--', alpha=0.5)

# --- SUBPLOT 3: Strong Negative Correlation (Skips vs Hours) ---
sns.regplot(
    data=df_analytical, 
    x='Skip_Rate_Percent', 
    y='Listening_Hours', 
    ax=axes[2], 
    color='#e74c3c', # Red for churn / negative engagement
    scatter_kws={'s': 160, 'edgecolor': '#2c3e50', 'linewidths': 1.5},
    line_kws={'color': '#2c3e50', 'linestyle': '--', 'linewidth': 2.5}
)

r_neg = corr_matrix.loc['Skip_Rate_Percent', 'Listening_Hours']
axes[2].annotate(
    f"Near-Perfect Inverse Linearity:\nr = {r_neg:.2f}\nHigh skip rates lead directly\nto early churn and drop-off.",
    xy=(30, 18), xytext=(22, 38),
    fontsize=10.5, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[2].set_title('Figure 3: Negative Trend (Skips vs Hours)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[2].set_xlabel('Track Skip Rate (%)', fontdict={'fontsize': 12, 'fontweight': 'bold'})
axes[2].set_ylabel('Total Listening Hours', fontdict={'fontsize': 12, 'fontweight': 'bold'})
axes[2].grid(True, linestyle='--', alpha=0.5)

# Remove top and right spines from regression plots
for ax in [axes[1], axes[2]]:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی محاسبات (Strategic Implication):</b><br><br>"
    r"۱. <b>تلهٔ کوواریانس:</b> مقدار کوواریانس بین لایک‌ها و ساعات شنیدن برابر با <code>+1023.25</code> شد، در حالی که کوواریانس بین درصد پرش و ساعات شنیدن برابر با <code>-318.75</code> گردید. "
    r"عدد ۱۰۲۳ بزرگتر از ۳۱۸ است، اما آیا شدت رابطهٔ اول ۳ برابر قوی‌تر است؟ <b>خیر!</b> این اختلاف صرفاً به خاطر تفاوتِ مقیاسِ اعداد است و مقایسهٔ آن‌ها با کوواریانس از نظر علمی باطل است.<br><br>"
    r"۲. <b>قدرت پیرسون:</b> با تبدیل به ضریب همبستگی پیرسون (جدول ۳)، مشخص شد که هر دو رابطه دارای شدتی تقریباً یکسان و بسیار نزدیک به قطعی هستند "
    r"(<b>r = +0.999</b> برای لایک و <b>r = -0.996</b> برای پرش). <br><br>"
    r"این بینش آماری مستقیماً به الگوریتم پیشنهاددهنده (Recommender System) دیکته می‌کند که اگر کاربری ۲ بار پشت سر هم آهنگی را رد (Skip) کرد، "
    r"سیستم باید فوراً سبک موسیقی را تغییر دهد تا از خروج قطعی کاربر از برنامه جلوگیری کند."
)
display_academic_text(insight_conclusion, "insight")
```
