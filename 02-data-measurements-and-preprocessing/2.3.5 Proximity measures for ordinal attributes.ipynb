<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: تحلیل ارتقای شغلی و رضایت کارمندان در «لینکدین» (LinkedIn) 📊
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    تیم داده‌کاوی «لینکدین» می‌خواهد بداند آیا بین **سطح ارشدیت شغلی** (کارآموز، تازه‌کار، ارشد، مدیر) و **سطح رضایت شغلی** (کم، متوسط، زیاد، بسیار زیاد) یک رابطه یکنواخت و مستقیم وجود دارد یا خیر. چون این متغیرها دارای ترتیب هستند اما فواصل عددی آن‌ها یکسان نیست (مثلاً فاصله کارآموز تا تازه‌کار برابر با فاصله ارشد تا مدیر نیست)، از **ضریب همبستگی رتبه‌ای اسپیرمن** استفاده می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام اول: تبدیل مقادیر ترتیبی به رتبه (Rank Transformation)</b><br>
    در همبستگی اسپیرمن، به‌جای مقادیر کیفی واقعی، به هر دسته یک رتبه عددی (از ۱ تا $n$) داده می‌شود:<br>
    &nbsp;&nbsp;&nbsp;• **ارشدیت شغلی ($X$):** Intern 👈 ۱ | Junior 👈 ۲ | Senior 👈 ۳ | Executive 👈 ۴<br>
    &nbsp;&nbsp;&nbsp;• **رضایت شغلی ($Y$):** Low 👈 ۱ | Medium 👈 ۲ | High 👈 ۳ | Very High 👈 ۴
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام دوم: فرمول همبستگی رتبه‌ای اسپیرمن ($r_s$)</b><br>
    همبستگی اسپیرمن بر اساس اختلاف رتبه‌ها ($d_i = r_{xi} - r_{yi}$) برای هر فرد محاسبه می‌شود:
  </p>
  
  <div style="text-align: center; margin: 15px 0;">
    $$r_s = 1 - \frac{6 \sum d_i^2}{n(n^2 - 1)}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام سوم: تفسیر خروجی همبستگی یکنواخت (Monotonic Relationship)</b><br>
    مقدار ضریب همبستگی اسپیرمن همیشه بین **۱- تا ۱+** قرار دارد:<br>
    &nbsp;&nbsp;&nbsp;• <b>$r_s = +1$:</b> همبستگی یکنواخت مستقیم کامل (با افزایش رتبه شغلی، رضایت همواره افزایش می‌یابد).<br>
    &nbsp;&nbsp;&nbsp;• <b>$r_s = -1$:</b> همبستگی یکنواخت معکوس کامل.<br>
    &nbsp;&nbsp;&nbsp;• <b>$r_s = 0$:</b> عدم وجود هیچ‌گونه رابطه یکنواخت بین رتبه‌ها.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> همبستگی اسپیرمن برخلاف پیرسون نیازی به خطی بودن رابطه ندارد و تنها بر حفظ جهت و ترتیب رتبه‌ها تمرکز می‌کند؛ به همین دلیل بهترین معیار برای ویژگی‌های ترتیبی است.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: سنجش جهت و شدت رابطه بین ویژگی‌های رتبه‌دار در بازه مقیاس‌شده [-1, +1]
    </div>
  </p>

</div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import spearmanr # FIX: Added missing import

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})

plt.rcParams['figure.dpi'] = 135

np.random.seed(42)
```

```python
display_academic_text(
    "آمار توصیفی: تحلیل وابستگی ویژگی‌های ترتیبی با ضریب همبستگی رتبه‌ای اسپیرمن (Spearman Correlation) 📈", 
    "main_title"
)

display_academic_text(
    "متغیرهای ترتیبی (مانند رده‌های شغلی یا طیف رضایت) دارای فواصل عددی هم‌ارز نیستند. در این سناریو، به منظور کشف یک رابطهٔ "
    "<b>یکنواخت (Monotonic)</b> میان سطح ارشدیت و میزان رضایت، ابتدا مقادیر کیفی به رتبه‌های عددی (Ranks) تبدیل شده و سپس "
    "به جای ارزیابی خطی پیرسون، اختلاف رتبه‌ها با بهره‌گیری از ضریب اسپیرمن محاسبه می‌گردد.", 
    "paragraph"
)

display_academic_text("فرمول محاسبه اختلاف رتبه‌ها برای هر مشاهده:", "sub_title")
display_academic_text(r"$$d_i = r_{xi} - r_{yi}$$", "paragraph")

display_academic_text("معادله ضریب همبستگی رتبه‌ای اسپیرمن:", "sub_title")
display_academic_text(r"$$r_s = 1 - \frac{6 \sum d_i^2}{n(n^2 - 1)}$$", "paragraph")

# ==========================================================
# SECTION 2: DATA RANK TRANSFORMATION AND CORRELATION
# ==========================================================
display_academic_text(
    "مرحله نخست: اعمال نگاشت رتبه‌ای و استخراج ماتریس همبستگی", 
    "sub_title"
)

display_academic_text(
    "در این فاز، مقادیر متنی بر اساس جایگاه سلسله‌مراتبی خود به اعداد صحیح متوالی نگاشت می‌شوند. سپس ماتریس همبستگی با "
    "متد اسپیرمن محاسبه شده و برای تایید ارزش آماری، <b>مقدار پی (P-value)</b> از طریق کتابخانه SciPy استخراج می‌گردد.", 
    "paragraph"
)

# Step 1: Initialize the dataset with ordinal attributes (Text format)
data = {
    'Employee_ID': ['Emp_1', 'Emp_2', 'Emp_3', 'Emp_4', 'Emp_5'],
    'Seniority_Level': ['Intern', 'Junior', 'Senior', 'Senior', 'Executive'],
    'Job_Satisfaction': ['Low', 'Medium', 'High', 'Medium', 'Very High'],
}
df_raw = pd.DataFrame(data)

# Step 2: Define explicit order mappings to convert nominal strings into mathematical ranks
seniority_map = {'Intern': 1, 'Junior': 2, 'Senior': 3, 'Executive': 4}
satisfaction_map = {'Low': 1, 'Medium': 2, 'High': 3, 'Very High': 4}

df_ranked = df_raw.copy()
df_ranked['Seniority_Rank'] = df_ranked['Seniority_Level'].map(seniority_map)
df_ranked['Satisfaction_Rank'] = df_ranked['Job_Satisfaction'].map(satisfaction_map)

df_analytics = df_ranked[['Employee_ID', 'Seniority_Rank', 'Satisfaction_Rank']]

# Step 3: Calculate the Spearman Rank Correlation Matrix using pandas
spearman_matrix = df_ranked[['Seniority_Rank', 'Satisfaction_Rank']].corr(method='spearman')

# FIX: Reset the index and rename to explicitly assign a column name, preventing an empty left column in HTML
spearman_matrix_disp = spearman_matrix.reset_index().rename(columns={'index': 'Rank Variables'})

# Step 4: Calculate exact P-value utilizing scipy to confirm statistical significance
rho, p_value = spearmanr(df_ranked['Seniority_Rank'], df_ranked['Satisfaction_Rank'])

# Render the transformation pipeline side-by-side
display_side_by_side_datasets(
    [
        df_raw[['Employee_ID', 'Seniority_Level', 'Job_Satisfaction']], 
        df_analytics, 
        spearman_matrix_disp
    ],
    [
        "جدول ۱: داده‌های متنی (ویژگی‌های ترتیبی خام)", 
        "جدول ۲: ماتریس رتبه‌ها (Rank Transformation)",
        "جدول ۳: ماتریس همبستگی اسپیرمن"
    ]
)

display_academic_text(
    f"مقدار ضریب استخراج‌شده (rho) برابر با <b>{rho:.3f}</b> می‌باشد که نشان‌دهنده یک رابطه یکنواخت صعودی بسیار قوی است. "
    f"همچنین مقدار P-value معادل <b>{p_value:.3f}</b> می‌باشد.", 
    "paragraph"
)

# ==========================================================
# SECTION 3: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی وابستگی رتبه‌ای و جهت حرکت", 
    "sub_title"
)

display_academic_text(
    "در بوم گرافیکی زیر، نمودار پراکندگی رتبه‌ها (سمت راست) نشان می‌دهد که چگونه افزایش جایگاه شغلی، با افزایش پیوستهٔ سطح رضایت همراه است. "
    "از آنجا که این رابطه الزاماً به صورت یک خط راست ایده‌آل نیست، اسپیرمن با تمرکز بر حفظ «جهت» رتبه‌ها، ساختار یکنواخت و صعودی بازار کار را "
    "با موفقیت اثبات می‌نماید. نقشه حرارتی (سمت چپ) نیز تاییدیه بصری شدت این وابستگی است.", 
    "paragraph"
)

# FIX: Initialize a wide dual-axis canvas (19x6.5 inches) for proper visualization
fig, axes = plt.subplots(1, 2, figsize=(19, 6.5), dpi=135)

# --- SUBPLOT 1: Heatmap of the Spearman Rank Correlation Matrix ---
sns.heatmap(
    spearman_matrix, 
    annot=True, 
    cmap='Blues', 
    vmin=-1, vmax=1, 
    ax=axes[0], 
    square=True, 
    fmt='.2f', 
    linewidths=2.0, 
    linecolor='white',
    cbar_kws={'shrink': 0.8, 'label': 'Spearman Coefficient'},
    annot_kws={"size": 15, "weight": "bold"}
)
axes[0].set_title('Figure 1: Spearman Rank Correlation Heatmap', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=15, ha='right', fontweight='bold')
axes[0].set_yticklabels(axes[0].get_yticklabels(), fontweight='bold')

# --- SUBPLOT 2: Scatter plot with Monotonic Trendline ---
sns.regplot(
    data=df_ranked, 
    x='Seniority_Rank', 
    y='Satisfaction_Rank', 
    ax=axes[1], 
    color='#1abc9c', # Turquoise for scatter points
    scatter_kws={'s': 250, 'edgecolor': '#2c3e50', 'linewidths': 2.0}, 
    line_kws={'color': '#2c3e50', 'linestyle': '--', 'linewidth': 3.0} # Navy Blue for the trendline
)

# Intelligent Annotations
axes[1].annotate(
    "Monotonic Increase:\nAs Seniority Rank increases, Satisfaction Rank\nconsistently increases. The algorithm proves the\ndirection without assuming linear distance.",
    xy=(3, 3), xytext=(1.2, 3.5),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[1].set_title('Figure 2: Monotonic Relationship (Seniority vs Satisfaction)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[1].set_xlabel('Mapped Seniority Rank (1: Intern ➔ 4: Executive)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_ylabel('Mapped Satisfaction Rank (1: Low ➔ 4: Very High)', fontdict={'fontsize': 13, 'fontweight': 'bold'})

axes[1].set_xticks([1, 2, 3, 4])
axes[1].set_yticks([1, 2, 3, 4])
axes[1].grid(True, linestyle='--', alpha=0.5)

# Clean up spines
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی محاسبات (Strategic Implication):</b><br>"
    r"استفاده از متریک همبستگی خطی کلاسیک (Pearson) در متغیرهای ترتیبی، به دلیل <i>نابرابریِ فواصلِ مفهومیِ میان دسته‌ها</i> "
    r"(مثلاً تفاوت تجربه یک کارآموز با جونیور، لزوماً برابر با تفاوت سینیور و مدیر نیست) فاقد اعتبار ریاضی است.<br><br>"
    r"به‌کارگیری <b>ضریب همبستگی رتبه‌ای اسپیرمن (Spearman Rank Correlation)</b>، بدون پیش‌فرضِ خطی بودن فواصل، با موفقیت کشف نمود که "
    r"ارتقای ردهٔ سازمانی، یک نیروی محرکهٔ <b>یکنواخت (Monotonic)</b> و مستقیم در افزایش رضایت شغلی سرمایه‌های انسانی پلتفرم محسوب می‌گردد."
)
display_academic_text(insight_conclusion, "insight")
```
