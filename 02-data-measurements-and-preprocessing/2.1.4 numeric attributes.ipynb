  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: قیمت‌گذاری پویا و تحلیل سفرهای «اسنپ» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در الگوریتم‌های هوشمند «اسنپ»، برخلاف متغیرهای اسمی یا ترتیبی، با کمیت‌های کاملاً محاسباتی و عددی روبه‌رو هستیم. الگوریتم قیمت‌گذاری برای محاسبه کرایه عادلانه، دو نوع داده عددی متفاوت را ترکیب می‌کند: <b>ویژگی‌های عددی گسسته (Discrete)</b> و <b>ویژگی‌های عددی پیوسته (Continuous)</b>.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام اول: ویژگی‌های عددی گسسته (Discrete)</b><br>
    این مقادیر از جنس «شمارش» هستند، اعداد اعشاری نمی‌پذیرند و جدا از هم می‌باشند:<br>
    &nbsp;&nbsp;&nbsp;• <code>Passenger_Count</code> (تعداد مسافران): ۱، ۲ یا ۳ نفر (نمی‌توان ۱.۵ مسافر داشت!).<br>
    &nbsp;&nbsp;&nbsp;• <code>Daily_Completed_Trips</code> (تعداد سفرهای روزانه راننده): ۱۲ سفر.<br>
    &nbsp;&nbsp;&nbsp;• <code>Cancel_Count</code> (تعداد لغو سفر): ۰ یا ۲ بار.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام دوم: ویژگی‌های عددی پیوسته (Continuous)</b><br>
    این مقادیر از جنس «اندازه‌گیری» هستند و در یک بازه می‌توانند بی‌نهایت مقدار اعشاری بپذیرند:<br>
    &nbsp;&nbsp;&nbsp;• <code>Trip_Distance_KM</code> (مسافت سفر): ۱۲.۴۵ کیلومتر.<br>
    &nbsp;&nbsp;&nbsp;• <code>Trip_Duration_Min</code> (زمان سفر با ترافیک): ۳۴.۸ دقیقه.<br>
    &nbsp;&nbsp;&nbsp;• <code>Fare_Amount</code> (مبلغ کرایه): ۷۸,۵۰۰ تومان.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام سوم: چالش استانداردسازی (Feature Scaling)</b><br>
    تعداد مسافران بین ۱ تا ۴ است، اما کرایه بین ۴۰,۰۰۰ تا ۳۰۰,۰۰۰ تومان! اگر این مقادیر را مستقیم به مدل بدهیم، مدل به دلیل بزرگ بودن اعداد کرایه، ویژگی‌های کوچکی مثل تعداد مسافران را کاملاً نادیده می‌گیرد. راه‌حل، **مقیاس‌بندی (Standardization / Normalization)** اعداد در یک بازه مشخص (مثلاً ۰ تا ۱) است.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> الگوریتم قیمت‌گذاری با ترکیب مقادیر گسسته و پیوسته و هم‌مقیاس کردن آن‌ها، می‌تواند کرایه دقیق را بر اساس مسافت واقعی و میزان شلوغی محاسبه کند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: تبدیل متغیرهای عددی با مقیاس‌های متفاوت به داده‌های استاندارد برای مدل‌های پیش‌بینی
    </div>
  </p>

</div>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})

plt.rcParams['figure.dpi'] = 135

np.random.seed(42)
```

```python
display_academic_text(
    "پیش‌پردازش داده‌ها: استخراج و نرمال‌سازی ویژگی‌های عددیِ گسسته و پیوسته (Feature Scaling) 🧮", 
    "main_title"
)

display_academic_text(
    "بر اساس ساختار داده‌کاوی، ویژگی‌های عددی به دو دسته <b>گسسته (Discrete - قابل شمارش)</b> و <b>پیوسته (Continuous - قابل اندازه‌گیری)</b> تقسیم می‌گردند. "
    "در سیستم‌های قیمت‌گذاری هوشمند، تفاوت چشمگیر در دامنهٔ مقادیر این متغیرها می‌تواند به انحرافِ قطعیِ مدل‌های یادگیری ماشین (مانند K-Means یا شبکه‌های عصبی) منجر شود. "
    "در این سناریو، یک ساختار استانداردسازی <b>کمینه-بیشینه (Min-Max Scaler)</b> جهت هم‌ترازسازی فضای ابعاد عددی توسعه یافته است.", 
    "paragraph"
)

# ==========================================================
# SECTION 2: DATA INITIALIZATION AND FEATURE SCALING
# ==========================================================
display_academic_text(
    "مرحله نخست: بارگذاری داده‌های خام و اعمال الگوریتم مقیاس‌بندی (Min-Max Normalization)", 
    "sub_title"
)

display_academic_text(
    "در این فاز، پایگاه داده سفرها شامل ویژگیِ گسسته (تعداد مسافر) و ویژگی‌هایِ پیوسته (مسافت، زمان و کرایه) بارگذاری می‌گردد. "
    "سپس، الگوریتم نرمال‌سازی تمامی این مقادیرِ به شدت ناهمگون را به یک فضای مقیاس‌بندی‌شده‌ی <i>یکنواخت در بازه [0, 1]</i> انتقال می‌دهد.", 
    "paragraph"
)

# Step 1: Initialize the mock database for ride-hailing trips.
# Includes both discrete (e.g., Passenger_Count) and continuous (e.g., Trip_Distance_KM) variables.
raw_data = {
    'Trip_ID': [1001, 1002, 1003, 1004, 1005],
    'Passenger_Count': [1, 3, 2, 1, 4],            
    'Trip_Distance_KM': [3.2, 14.85, 8.1, 2.0, 22.4], 
    'Trip_Duration_Min': [12.0, 45.5, 25.2, 8.5, 62.0], 
    'Fare_Toman': [35000, 115000, 68000, 28000, 175000] 
}
df_raw = pd.DataFrame(raw_data)

# Step 2: Define the subset of numerical columns that require scale normalization. 
# Exclude identifiers like 'Trip_ID'.
numeric_cols = [
    'Passenger_Count', 
    'Trip_Distance_KM', 
    'Trip_Duration_Min', 
    'Fare_Toman'
]

# Step 3: Instantiate the Min-Max Scaler algorithm.
# This mathematical transformation scales all designated features strictly into a [0, 1] interval.
scaler = MinMaxScaler()

# Step 4: Create a deep copy of the original dataset to preserve data provenance.
df_scaled = df_raw.copy()

# Step 5: Execute the fit and transform operation on the selected numerical subspace.
df_scaled[numeric_cols] = scaler.fit_transform(df_raw[numeric_cols])

# Render the raw and transformed datasets side-by-side to visually demonstrate the scaling impact.
# Using the list format expected by the global function
display_side_by_side_datasets(
    [df_raw, df_scaled],
    [
        "جدول ۱: پایگاه داده عددی اولیه (ابعادِ به شدت ناهمگون)", 
        "جدول ۲: پایگاه داده نرمال‌سازی‌شده (همتراز شده در بازه یکنواخت)"
    ]
)

# ==========================================================
# SECTION 3: ANALYTICAL VISUALIZATION (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلیِ توزیع ویژگی‌ها پیش و پس از نرمال‌سازی", 
    "sub_title"
)

display_academic_text(
    "نمودار جعبه‌ایِ (Boxplot) زیر چالش بنیادین اختلافِ مقیاس را به تصویر می‌کشد. در فضای اصلی (سمت چپ)، مقادیرِ بزرگِ <code>Fare_Toman</code> "
    "موجب فشردگی و کاملاً نادیده گرفته شدنِ سایر ویژگی‌ها (تعداد مسافر یا مسافت) شده‌اند؛ در حالی که در فضای نرمال‌سازی‌شده (سمت راست)، "
    "تمامی متغیرها در یک مقیاس منسجم، قابلیت مقایسه و ارزش‌گذاریِ ریاضیاتیِ یکسانی را به دست آورده‌اند.", 
    "paragraph"
)

# Initialize a wide canvas to plot dual boxplots side-by-side for spatial comparison
fig, axes = plt.subplots(1, 2, figsize=(19, 6.5), dpi=135)

# --- CHART 1: Raw Feature Distribution (The Problem) ---
sns.boxplot(
    data=df_raw[numeric_cols], 
    orient='h', 
    ax=axes[0], 
    palette=['#bdc3c7', '#95a5a6', '#7f8c8d', '#2c3e50'], # Academic Grays with Navy for the dominator
    fliersize=8, 
    linewidth=2.0
)

axes[0].annotate(
    "The Scale Distortion Problem:\nThe massive values of 'Fare_Toman'\ncompletely squash the other features.\nA machine learning algorithm would falsely\nthink Fare is the only feature that matters!",
    xy=(40000, 2), xytext=(50000, 0.5),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title('Figure 1: Raw Feature Distribution (Severe Scale Distortion)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[0].set_xlabel('Absolute Value Measurement (Raw Scale)', fontdict={'fontsize': 12, 'fontweight': 'bold'})
axes[0].grid(True, linestyle='--', alpha=0.5)

# --- CHART 2: Scaled Feature Distribution (The Solution) ---
sns.boxplot(
    data=df_scaled[numeric_cols], 
    orient='h', 
    ax=axes[1], 
    palette=['#1abc9c', '#16a085', '#2980b9', '#2c3e50'], # Academic Blues & Teals
    fliersize=8, 
    linewidth=2.0
)

axes[1].annotate(
    "Harmonic Normalization:\nThanks to Min-Max Scaling, all features now\nlive perfectly between 0.0 and 1.0.\nThe Machine Learning model can now\nread and weigh them equally!",
    xy=(0.5, 3), xytext=(0.3, 0.5),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[1].set_title('Figure 2: Normalized Feature Distribution (Min-Max Scaled)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[1].set_xlabel('Standardized Value Range [0.0 , 1.0]', fontdict={'fontsize': 12, 'fontweight': 'bold'})
axes[1].grid(True, linestyle='--', alpha=0.5)

# Clean up top and right spines for both charts
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    # Make y-axis labels bold for readability
    ax.set_yticklabels(ax.get_yticklabels(), fontweight='bold')

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردیِ عملیات (Strategic Implication):</b><br><br>"
    r"تصویر اول نشان داد که در غیاب فرآیند استانداردسازی، الگوریتم‌های فاصله‌محور (نظیر <i>K-Nearest Neighbors</i>، <i>K-Means Clustering</i>) "
    r"یا الگوریتم‌های مبتنی بر گرادیان (نظیر <i>شبکه‌های عصبی عمیق</i>)، به دلیل تسلطِ عددیِ متغیرهای دارای مقیاس بزرگ (مانند ۱۷۵,۰۰۰ تومان کرایه)، "
    r"سایر عوامل تعیین‌کننده (مانند تعداد مسافر که بین ۱ تا ۴ متغیر است) را کاملاً نادیده خواهند گرفت. <br><br>"
    r"استفاده از <b>Min-Max Scaler</b> و انتقال تمامیِ داده‌ها به فضای <code>[0, 1]</code> (تصویر دوم)، این سوءگیری محاسباتی (Scale Bias) را به طور قطعی مرتفع می‌سازد و "
    r"تضمین می‌کند که ماشین، الگوها را بر اساسِ <i>تغییراتِ نسبیِ داده‌ها</i> یاد می‌گیرد، نه بزرگی و کوچکیِ اعداد خام."
)
display_academic_text(insight_conclusion, "insight")
```
