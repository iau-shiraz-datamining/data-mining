<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: پاکسازی داده‌های سنسورهای خودرو در «تسلا» (Tesla) 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    سیستم مرکزی «تسلا» (Tesla) روزانه میلیاردها داده لحظه‌ای از سنسورهای خودروها (دمای باتری، سرعت، میزان شارژ و پایش سلامت قطعات) دریافت می‌کند. این داده‌های خام به دلیل قطعی اینترنت، نویز سنسورها یا اختلالات نرم‌افزاری دچار خطا می‌شوند. سیستم پیش از تحلیل‌های داده‌کاوی، ۴ گام اصلی **پاکسازی داده (Data Cleaning)** را اجرا می‌کند:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۱. حذف رکوردهای تکراری (Duplicate Removal):</b><br>
    گاهی یک سنسور به‌دلیل ارسال مجدد سیگنال، یک سیگنال واحد را ۲ بار ذخیره می‌کند. رکوردهای تکراری باید شناسایی و حذف شوند تا وزن کاذب در محاسبات ایجاد نکنند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۲. اصلاح انواع داده‌ها (Data Type Correction):</b><br>
    گاهی دمای باتری به جای عدد به‌صورت متنی (مثلاً <code>"45.2"</code>) ذخیره می‌شود. برای انجام محاسبات ریاضی، این داده‌ها باید به نوع عددی (<code>Float</code>) تبدیل گردند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۳. مدیریت مقادیر گمشده (Missing Values Handling):</b><br>
    اگر سنسور در مقطعی قطع شده باشد:
  </p>
  <ul style="text-align: right; margin: 0 0 10px; padding-right: 22px;">
    <li><b>حذف (Removal):</b> اگر حجم داده‌های خالی کم باشد، آن سطر حذف می‌شود.</li>
    <li><b>جایگذاری (Imputation):</b> داده گمشده با میانگین/میانه مقادیر قبلی یا روش درونیابی (Interpolation) بازسازی می‌شود.</li>
  </ul>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۴. شناسایی و حذف داده‌های پرت و نویز (Outlier Removal):</b><br>
    اگر سنسور به‌دلیل نوسان ولتاژ، دمای باتری را ناگهان ۹۹۹ درجه سانتی‌گراد ثبت کند، این داده پرت و غیرممکن به عنوان نویز شناسایی و فیلتر می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> پاکسازی داده‌ها ورودی دقیق و قابل اعتمادی برای الگوریتم‌های هوش مصنوعی خودران تسلا فراهم می‌سازد تا از تصمیم‌گیری‌های اشتباه جلوگیری شود.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: تبدیل داده‌های خام و دارای نویز سنسورها به یک پایگاه داده یکدست، تمیز و استاندارد
    </div>
  </p>

</div>

```python
display_academic_text(
    "پیش‌پردازش داده‌ها: استراتژی‌های پاک‌سازی، حذف نویز و یکپارچه‌سازی پایگاه داده (Data Cleaning Pipeline) 🛠️", 
    "main_title"
)

display_academic_text(
    "اطلاعات مستخرج از شبکه‌های اینترنت اشیا (IoT) و سنسورهای فیزیکی به شدت مستعد خطا، افزونگی و داده‌های گمشده می‌باشند. "
    "در این سناریو، یک پایگاه داده خام از ناوگان خودروهای خودران تحت ۴ عملیات متوالی پاک‌سازی "
    "(<b>Deduplication, Type Correction, Outlier Masking, Imputation</b>) قرار می‌گیرد تا پایداری تحلیلیِ الگوریتم‌های هوش مصنوعی تضمین گردد.", 
    "paragraph"
)

display_academic_text("آستانه‌های فیزیکی مجاز جهت فیلترینگ داده‌های پرت (Outliers):", "sub_title")

# Utilizing Raw Strings (r"") for LaTeX rendering of boundary constraints
display_academic_text(
    r"$$Battery\_Temp \in [0, 100] \, ^\circ\text{C} \quad , \quad Speed \in [0, 300] \, \text{KM/H}$$", 
    "paragraph"
)

# ==========================================================
# SECTION 2: MULTI-STAGE DATA CLEANING PIPELINE
# ==========================================================
display_academic_text(
    "مرحله نخست: اعمال پایپ‌لاین پاک‌سازی بر روی پایگاه داده سنسورها", 
    "sub_title"
)

display_academic_text(
    "در این فاز، ابتدا رکوردهای <i>تکراری</i> حذف شده و نوع داده‌های متنی به عددی تبدیل می‌شوند. سپس مقادیرِ خارج از آستانه مجاز (مانند دمای ۹۹۹ درجه) "
    "به عنوان <i>نویزِ سخت‌افزاری</i> شناسایی شده و به مقادیر گمشده (NaN) تبدیل می‌گردند. در گام نهایی، تمامی مقادیر تهی با استفاده از شاخص آماری میانه بازسازی می‌شوند.", 
    "paragraph"
)

# Step 1: Initialize the raw, messy telemetry dataset mimicking real-world sensor noise
raw_data = {
    # Notice the exact Duplicate for TS-102
    'Vehicle_ID': ['TS-101', 'TS-102', 'TS-102', 'TS-103', 'TS-104', 'TS-105'], 
    # Notice String types ('35.5'), real NaN, and extreme physics outlier ('999.0')
    'Battery_Temp_C': ['35.5', '42.0', '42.0', np.nan, '999.0', '38.2'],        
    # Notice negative speed outlier (-50.0) which is impossible
    'Speed_KMH': [110.0, 85.0, 85.0, 95.0, 120.0, -50.0],                        
}
df_raw = pd.DataFrame(raw_data)

# Step 2: Clean Dataset Step-by-Step (The Preprocessing Pipeline)
# 2.1 Deduplication: Remove exact duplicate rows and create an isolated copy
df_clean = df_raw.drop_duplicates().copy()

# 2.2 Data Type Correction: Convert String to Numeric Float
# errors='coerce' gracefully converts unparseable strings to NaN
df_clean['Battery_Temp_C'] = pd.to_numeric(df_clean['Battery_Temp_C'], errors='coerce')

# 2.3 Outlier Masking: Filter out physically impossible values based on defined thresholds
valid_temp_mask = (df_clean['Battery_Temp_C'] >= 0) & (df_clean['Battery_Temp_C'] <= 100)
valid_speed_mask = (df_clean['Speed_KMH'] >= 0) & (df_clean['Speed_KMH'] <= 300)

# Replace outliers (where the mask is False) with NaN to prepare them for imputation
df_clean.loc[~valid_temp_mask, 'Battery_Temp_C'] = np.nan
df_clean.loc[~valid_speed_mask, 'Speed_KMH'] = np.nan

# 2.4 Missing Value Imputation: Fill NaN (both original and newly masked outliers) with Median
median_temp = df_clean['Battery_Temp_C'].median()
median_speed = df_clean['Speed_KMH'].median()

df_clean['Battery_Temp_C'] = df_clean['Battery_Temp_C'].fillna(median_temp)
df_clean['Speed_KMH'] = df_clean['Speed_KMH'].fillna(median_speed)

# Prepare cleanly formatted DataFrames for display
df_raw_disp = df_raw.copy()
df_clean_disp = df_clean.copy()

display_side_by_side_datasets(
    [df_raw_disp, df_clean_disp],
    [
        "جدول ۱: پایگاه داده خام سنسورها (دارای نویز و افزونگی)", 
        "جدول ۲: پایگاه داده پاک‌سازی شده نهایی (ایمن و یکپارچه)"
    ]
)

# ==========================================================
# SECTION 3: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی - توزیع داده‌ها پیش و پس از فیلترینگ", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکی زیر تأثیر شگرف فیلترینگ نویز و ترمیم داده‌ها را به تصویر می‌کشد. "
    "نمودارهای سمت چپ (قرمز رنگ) به شدت تحت تأثیر مقادیر فیزیکاً غیرممکن (۹۹۹ درجه و سرعت منفی) قرار گرفته‌اند که باعث فشردگی کاذبِ سایر رکوردهای سالم شده است. "
    "پس از اعمال پایپ‌لاین پاک‌سازی (نمودارهای سمت راست با پالت آبی آکادمیک)، توزیع مقادیر به صورت منطقی و در دامنه استاندارد بازسازی گردیده است.", 
    "paragraph"
)

# Initialize a 2x2 subplot canvas for comparative visual analytics
fig, axes = plt.subplots(2, 2, figsize=(18, 9), dpi=135)

# Pre-process raw data for visualization ONLY (Converting strings to numbers to allow plotting)
df_plot_raw = df_raw.copy()
df_plot_raw['Battery_Temp_C'] = pd.to_numeric(df_plot_raw['Battery_Temp_C'], errors='coerce')

# ----------------------------------------------------------
# ROW 1: Battery Temperature Distribution (Before & After)
# ----------------------------------------------------------
# Plot Raw Battery Temp (Showing the 999 Outlier)
sns.stripplot(
    data=df_plot_raw, y='Battery_Temp_C', 
    ax=axes[0,0], color='#e74c3c', size=12, jitter=0.05, edgecolor='black', linewidth=1.5
)

axes[0,0].annotate(
    "Sensor Failure (999°C):\nThis massive outlier squashes the real data\nat the bottom, making it unreadable.",
    xy=(0.05, 999), xytext=(0.15, 750),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0,0].set_title('Figure 1A: Raw Battery Temp (Sensor Noise)', fontdict={'fontsize': 13, 'fontweight': 'bold'}, pad=10)
axes[0,0].set_ylabel('Temperature (°C)', fontdict={'fontsize': 12, 'fontweight': 'bold'})
axes[0,0].grid(True, linestyle='--', alpha=0.5)


# Plot Cleaned Battery Temp
sns.stripplot(
    data=df_clean, y='Battery_Temp_C', 
    ax=axes[0,1], color='#1abc9c', size=12, jitter=0.05, edgecolor='black', linewidth=1.5
)

axes[0,1].set_title('Figure 1B: Cleaned Temp (Masked & Imputed)', fontdict={'fontsize': 13, 'fontweight': 'bold'}, pad=10)
axes[0,1].set_ylabel('Temperature (°C)', fontdict={'fontsize': 12, 'fontweight': 'bold'})
axes[0,1].set_ylim(34, 45) # Set reasonable physical limits for visualization
axes[0,1].grid(True, linestyle='--', alpha=0.5)

# ----------------------------------------------------------
# ROW 2: Vehicle Speed Distribution (Before & After)
# ----------------------------------------------------------
# Plot Raw Speed (Showing the -50 Outlier)
sns.stripplot(
    data=df_plot_raw, y='Speed_KMH', 
    ax=axes[1,0], color='#c0392b', size=12, jitter=0.05, edgecolor='black', linewidth=1.5
)

axes[1,0].annotate(
    "Physics Violation:\nSpeed cannot be -50 KM/H!\nA neural network would learn impossible physics.",
    xy=(0.05, -50), xytext=(0.15, -20),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#c0392b", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#c0392b", lw=1.5)
)

axes[1,0].set_title('Figure 2A: Raw Speed (Physics Violation)', fontdict={'fontsize': 13, 'fontweight': 'bold'}, pad=10)
axes[1,0].set_ylabel('Speed (KM/H)', fontdict={'fontsize': 12, 'fontweight': 'bold'})
axes[1,0].grid(True, linestyle='--', alpha=0.5)


# Plot Cleaned Speed
sns.stripplot(
    data=df_clean, y='Speed_KMH', 
    ax=axes[1,1], color='#2980b9', size=12, jitter=0.05, edgecolor='black', linewidth=1.5
)

axes[1,1].set_title('Figure 2B: Cleaned Speed (Standardized Range)', fontdict={'fontsize': 13, 'fontweight': 'bold'}, pad=10)
axes[1,1].set_ylabel('Speed (KM/H)', fontdict={'fontsize': 12, 'fontweight': 'bold'})
axes[1,1].set_ylim(70, 130) # Set reasonable physical limits for visualization
axes[1,1].grid(True, linestyle='--', alpha=0.5)

# Clear right and top spines for all subplots
for ax in axes.flat:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout(pad=3.0)
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی عملیات (Strategic Implication):</b><br>"
    r"تزریق مستقیم داده‌های خام و فیلترنشده به الگوریتم‌های هوش مصنوعی منجر به پدیدهٔ <b>GIGO (Garbage In, Garbage Out)</b> و فروپاشیِ قطعیِ محاسباتیِ مدل می‌گردد "
    r"(تغییرات شدیدِ گرادیان به دلیل دمای ۹۹۹ درجه!). <br><br>"
    r"پایپ‌لاین پیش‌پردازشِ ما با اجرای متوالی توابع: ۱) <i>جرم‌زدایی (Deduplication)</i>، ۲) <i>همگن‌سازی نوع داده‌ها (Casting)</i>، ۳) <i>اعمال آستانه‌های فیزیکی (Masking)</i> و ۴) <i>ترمیم داده‌های گمشده (Imputation)</i>، "
    r"یکپارچگی اطلاعاتی را تضمین نموده و بستر لازم جهت استقرارِ امن، قابل اعتماد و دقیقِ سیستم‌های خودران را فراهم آورد."
)
display_academic_text(insight_conclusion, "insight")
```

<li><b>شناسایی و حذف نویز/داده‌های پرت:</b> با اعمال فیلتر شرطی (دمای مجاز بین ۰ تا ۱۰۰ و سرعت مجاز بین ۰ تا ۳۰۰)، مقادیر پرت غیرمنطقی شناسایی و به مقدار خالی (NaN) تبدیل شدند.</li>
<li><b>ترمیم مقادیر گمشده:</b> در نهایت، تمام مقادیر خالی با استفاده از متد <b>fillna()</b> و جایگذاری با <b>میانه (Median)</b> ستون‌ها بازسازی شدند تا ساختار داده‌ها ۱۰۰٪ کامل و آماده ورود به مدل‌های یادگیری ماشین گردد.</li>
</ul>
