<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: سیستم تشخیص هوشمند بیماری در «اپل هلت» (Apple Health) 🏥
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    سیستم هوش مصنوعی «اپل هلت» برای گروه‌بندی بیماران و تشخیص بیماری‌های مشابه، علائم بالینی افراد را بررسی می‌کند. هر علامت یک <b>ویژگی دودویی (Binary Attribute)</b> است: عدد <b>۱</b> به معنای وجود علامت (مثلاً داشتن تب) و عدد <b>۰</b> به معنای عدم وجود آن است.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام اول: جدول توافقی ۲×۲ (Contingency Table)</b><br>
    برای مقایسه دو بیمار (مثلاً بیمار A و بیمار B) بر اساس تمام علائم پزشکی، ۴ حالت ممکن وجود دارد:
  </p>
  <ul style="text-align: right; margin: 0 0 10px; padding-right: 22px;">
    <li><b>q:</b> تعداد علائمی که در <b>هر دو بیمار وجود دارد (۱ و ۱)</b>.</li>
    <li><b>r:</b> علائمی که فقط در بیمار A وجود دارد (۱ و ۰).</li>
    <li><b>s:</b> علائمی که فقط در بیمار B وجود دارد (۰ و ۱).</li>
    <li><b>t:</b> تعداد علائمی که در <b>هیچ‌کدام وجود ندارد (۰ و ۰)</b>.</li>
  </ul>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام دوم: انتخاب معیار مناسب (متقارن در برابر نامتقارن)</b><br>
    بسته به نوع داده‌های دودویی، از دو فرمول مختلف برای محاسبه شباهت استفاده می‌شود:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 6px;">
    <b>۱. ضریب تطابق ساده (Symmetric Matching Coefficient - SMC):</b><br>
    برای متغیرهای <b>دودویی متقارن</b> (مثل جنسیت) استفاده می‌شود که در آن ۰ و ۱ ارزش یکسانی دارند ($t$ هم‌ارزش با $q$ است):
  </p>
  <div style="text-align: center; margin: 10px 0;">
    $$\text{SMC} = \frac{q + t}{q + r + s + t}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 10px 0 6px;">
    <b>۲. ضریب جاکارد دودویی (Asymmetric Jaccard Coefficient):</b><br>
    برای متغیرهای <b>دودویی نامتقارن</b> (مثل علائم بیماری) استفاده می‌شود. در پزشکی، نداشتن یک بیماری (حالت ۰-۰ یا $t$) نباید باعث شود دو فرد سالم شبیه به هم به نظر برسند! بنابراین $t$ از فرمول حذف می‌شود:
  </p>
  <div style="text-align: center; margin: 10px 0;">
    $$\text{Jaccard Similarity} = \frac{q}{q + r + s}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام سوم: محاسبه عدم تشابه (Distance)</b><br>
    برای به دست آوردن فاصله (عدم تشابه) دودویی نامتقارن، کافی است ضریب جاکارد را از ۱ کم کنیم:
  </p>
  <div style="text-align: center; margin: 10px 0;">
    $$\text{Jaccard Distance} = \frac{r + s}{q + r + s} = 1 - \text{Jaccard Similarity}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> اپل هلت با حذف حالت‌های ۰-۰ (سلامت کامل در یک علامت)، فقط بر علائم واقعی فعال تمرکز می‌کند تا بیماران مبتلا به ابتلاهای مشابه را دقیق‌تر شناسایی کند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: جداسازی فرمول‌های دودویی متقارن و نامتقارن برای جلوگیری از شباهت‌های کاذب
    </div>
  </p>

</div>

```python
display_academic_text(
    "پیش‌پردازش داده‌ها: تمایز شباهت‌های متقارن (SMC) و نامتقارن (Jaccard) در داده‌های دودویی بالینی ⚕️", 
    "main_title"
)

display_academic_text(
    "انتخاب معیار تشابه برای متغیرهای دودویی مستقیماً به <i>ماهیت ویژگی‌ها</i> بستگی دارد. در تحلیل علائم بالینی (Clinical Symptoms)، "
    "<b>نبودِ یک نشانه (حالت ۰-۰)</b> لزوماً بیانگرِ تشابهِ فیزیولوژیکِ دو فرد نیست (همهٔ افرادِ سالم شبیه به هم نیستند!). "
    "در این سناریو، عملکرد <b>ضریب تطابق ساده (Symmetric SMC)</b> در برابر <b>ضریب جاکارد (Asymmetric Jaccard)</b> جهت جلوگیری از ارزیابی شباهت‌های کاذب مورد بررسی قرار می‌گیرد.", 
    "paragraph"
)

# Use raw strings (r"") for correct LaTeX rendering
display_academic_text(
    r"فرمول Symmetric Matching Coefficient (SMC) - (شامل تطابق‌های ۰-۰ یا $t$):", 
    "sub_title"
)
display_academic_text(
    r"$$SMC = \frac{q + t}{q + r + s + t}$$", 
    "paragraph"
)

display_academic_text(
    r"فرمول Asymmetric Jaccard Coefficient - (حذف‌کنندهٔ تطابق‌های ۰-۰ یا $t$):", 
    "sub_title"
)
display_academic_text(
    r"$$Jaccard = \frac{q}{q + r + s}$$", 
    "paragraph"
)

# ==========================================================
# SECTION 2: DATA MATRIX CONSTRUCTION AND COEFFICIENT COMPUTATION
# ==========================================================
display_academic_text(
    "مرحله نخست: بارگذاری داده‌های خام و محاسبه ماتریس‌های شباهت", 
    "sub_title"
)

display_academic_text(
    "در این فاز، ماتریس باینری علائم بیماران (۱ به معنای وجود و ۰ به معنای عدم وجود علامت) تشکیل می‌گردد. "
    "سپس، الگوریتم به صورت خودکار هر دو ضریب SMC و جاکارد را برای تمامی جفتِ بیماران محاسبه نموده و نتایج جهت مقایسهٔ آنی در کنار یکدیگر رندر می‌شوند.", 
    "paragraph"
)

# Step 1: Initialize the binary indicator matrix representing clinical symptoms.
# 1 indicates symptom presence (important/rare state), 0 indicates absence.
data = {
    'Fever':      [1, 1, 0, 0],
    'Cough':      [1, 1, 0, 0],
    'Headache':   [1, 0, 1, 0],
    'Fatigue':    [0, 1, 1, 0],
    'Nausea':     [0, 0, 0, 1]
}
patients = ['Patient_A', 'Patient_B', 'Patient_C', 'Patient_D']
df_binary = pd.DataFrame(data, index=patients)
df_binary.index.name = 'Patient_ID'

# Step 2: Compute Pairwise Symmetric Binary Similarity (SMC)
# The 'matching' metric in scipy computes the proportion of mismatches. Similarity is derived as 1 - distance.
smc_dist_condensed = pdist(df_binary.values, metric='matching')
df_smc_sim = pd.DataFrame(
    1 - squareform(smc_dist_condensed),
    index=patients,
    columns=patients
)

# Step 3: Compute Pairwise Asymmetric Binary Similarity (Jaccard)
# The 'jaccard' metric in scipy actively ignores 0-0 matches. Similarity is derived as 1 - distance.
jaccard_dist_condensed = pdist(df_binary.values, metric='jaccard')
df_jaccard_sim = pd.DataFrame(
    1 - squareform(jaccard_dist_condensed),
    index=patients,
    columns=patients
)

# Render the three data structures side-by-side using the list format expected by the global function
df_bin_disp = df_binary.reset_index()
df_smc_disp = df_smc_sim.reset_index().rename(columns={'index': 'Target / Reference'})
df_jac_disp = df_jaccard_sim.reset_index().rename(columns={'index': 'Target / Reference'})

display_side_by_side_datasets(
    [df_bin_disp, df_smc_disp, df_jac_disp],
    [
        "جدول ۱: ماتریس باینری علائم بالینی بیماران", 
        "جدول ۲: ماتریس SMC (دودویی متقارن)",
        "جدول ۳: ماتریس جاکارد (دودویی نامتقارن)"
    ]
)

# ==========================================================
# SECTION 3: ANALYTICAL VISUALIZATION (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی و کشف خوشه‌های شباهت", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکیِ زیر، ساختار تشابه بیماران را به تصویر می‌کشد. نقشه حرارتی SMC (سمت چپ) به دلیل گنجاندن حالت‌های سلامت مشترک (۰-۰)، مقادیر شباهتِ بسیار بالایی را نمایش می‌دهد که <b>کاملاً گمراه‌کننده</b> است. "
    "در مقابل، نقشه حرارتی جاکارد (سمت راست) با تمرکزِ قطعی بر علائم فعال (۱-۱)، تفکیک واقع‌بینانه‌تری از خوشه‌های بیماران ارائه می‌دهد.", 
    "paragraph"
)

# Initialize a dual-axis canvas for visual comparative analytics
fig, axes = plt.subplots(1, 2, figsize=(19, 6.5), dpi=135)

# --- SUBPLOT 1: Symmetric SMC Heatmap ---
sns.heatmap(
    df_smc_sim, 
    annot=True, 
    cmap='Blues', 
    fmt='.2f', 
    vmin=0, vmax=1, 
    ax=axes[0],
    linewidths=2.0, 
    linecolor='white', 
    cbar_kws={'label': 'SMC Similarity (Includes 0-0)'},
    annot_kws={"size": 15, "weight": "bold"}
)

axes[0].annotate(
    "The 0-0 Illusion:\nPatients C and D share NO identical symptoms.\nYet, because they both 'do not have' Fever and Cough,\nthe SMC falsely gives them a 40% similarity score!",
    xy=(3.5, 2.5), xytext=(3.5, 1.2),
    fontsize=11, fontweight='bold', color='#c0392b', ha='center',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title('Figure 1: Symmetric SMC Heatmap (Includes 0-0 Matches)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[0].set_xlabel('Target Patient', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_ylabel('Reference Patient', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_xticklabels(axes[0].get_xticklabels(), fontweight='bold', fontsize=11, rotation=0)
axes[0].set_yticklabels(axes[0].get_yticklabels(), fontweight='bold', fontsize=11, rotation=0, va='center')

# --- SUBPLOT 2: Asymmetric Jaccard Heatmap ---
sns.heatmap(
    df_jaccard_sim, 
    annot=True, 
    cmap='YlGnBu', 
    fmt='.2f', 
    vmin=0, vmax=1, 
    ax=axes[1],
    linewidths=2.0, 
    linecolor='white', 
    cbar_kws={'label': 'Jaccard Similarity (Ignores 0-0)'},
    annot_kws={"size": 15, "weight": "bold"}
)

axes[1].annotate(
    "Realistic Filtering:\nJaccard ignores the shared 'healthy' traits (0-0).\nSince C and D have completely different illnesses,\ntheir true clinical similarity correctly falls to 0.00.",
    xy=(3.5, 2.5), xytext=(3.5, 1.2),
    fontsize=11, fontweight='bold', color='#16a085', ha='center',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[1].set_title('Figure 2: Asymmetric Jaccard Heatmap (Ignores 0-0 Matches)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[1].set_xlabel('Target Patient', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_ylabel('Reference Patient', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_xticklabels(axes[1].get_xticklabels(), fontweight='bold', fontsize=11, rotation=0)
axes[1].set_yticklabels(axes[1].get_yticklabels(), fontweight='bold', fontsize=11, rotation=0, va='center')

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی عملیات (Strategic Implication):</b><br>"
    r"این کد به زیبایی تفاوت‌های بنیادینِ محاسباتی را ثابت می‌کند. در <b>بیمارِ C</b> (دارای سردرد و خستگی) و <b>بیمارِ D</b> (دارای حالت تهوع)، ما با دو بیماریِ کاملاً متفاوت روبه‌رو هستیم.<br><br>"
    r"با این حال، به دلیل اینکه هیچ‌کدام از آن‌ها تب و سرفه نداشتند (حالت ۰-۰)، الگوریتمِ <code>SMC</code> فریب خورد و شباهتِ آن‌ها را به صورتِ کاذب <b>۰.۴۰ (۴۰٪)</b> تخمین زد (شکل ۱). این یعنی تجویزِ اشتباهِ دارو! <br><br>"
    r"در مقابل، اتکا به <b>ضریب جاکارد</b>، با حذف هوشمندانهٔ حالت‌های مشترکِ ۰-۰، تمرکز محاسباتی را <i>صرفاً بر روی پیوندهای ناشی از علائمِ فعال (۱-۱)</i> قرار داد و شباهت آن‌ها را به درستی <b>۰.۰۰</b> محاسبه کرد (شکل ۲). این امر منجر به شناسایی دقیق‌تر خوشه‌های بیماران و جلوگیری از خطای سیستم‌های بالینی می‌گردد."
)
display_academic_text(insight_conclusion, "insight")
```
