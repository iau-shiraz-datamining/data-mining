  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: مدیریت سلامت و امنیت غذایی در «فودباکس» 0️⃣1️⃣
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در استارتاپ «فودباکس»، حفظ سلامت کاربران در اولویت است. فرض کنید می‌خواهیم متغیرهایی را در پایگاه‌داده ذخیره کنیم که فقط دو حالت دارند: یا بیمار حساسیت دارد یا ندارد، یا اشتراک کاربر فعال است یا غیرفعال. این‌ها نمونه‌های بارز <b>ویژگی‌های دودویی (Binary Attributes)</b> هستند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام اول: تفاوت ظریف اما حیاتی (متقارن در مقابل نامتقارن)</b><br>
    در فودباکس ما با دو نوع متغیر دودویی کار می‌کنیم:<br>
    &nbsp;&nbsp;&nbsp;• <b>دودویی متقارن (Symmetric):</b> مثل جنسیت کاربر (زن / مرد). هر دو حالت ارزش اطلاعاتی یکسانی دارند و اولویت خاصی نسبت به هم ندارند.<br>
    &nbsp;&nbsp;&nbsp;• <b>دودویی نامتقارن (Asymmetric):</b> مثل حساسیت شدید به بادام‌زمینی (دارد / ندارد). در اینجا، حالت «داشتن حساسیت» (که معمولاً با ۱ نشان داده می‌شود) فوق‌العاده حیاتی و نادر است، در حالی که حالت «نداشتن حساسیت» (۰) حالت عادی است و ارزش اطلاعاتی کمتری دارد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام دوم: فرمول‌سازی داده‌ها برای آشپزخانه هوشمند</b><br>
    آشپزخانه خودکار فودباکس برای بسته‌بندی غذاها نیاز دارد تا این متغیرها را به فرمت عددی تبدیل کند. اگر متغیر متنیِ <code>Has_Allergy</code> مقدارش "Yes" باشد، به عدد ۱ تبدیل می‌شود تا بلافاصله سیستم هشدار حذف بادام‌زمینی را فعال کند. اگر "No" باشد، به ۰ تبدیل می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام سوم: مدل‌سازی عددی</b><br>
    برای تبدیل متغیرهای دو حالتی متنی (مانند Yes/No یا M/F) به کدهای صفر و یک، از نگاشت متناظر (Mapping) در پایتون استفاده می‌کنیم تا داده‌ها برای الگوریتم‌های هوش مصنوعی فودباکس قابل فهم شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> آشپزخانه هوشمند با خواندن آرایه عددی صفر و یک، بدون فوت وقت متوجه می‌شود کدام جعبه‌ها باید فاقد مواد حساسیت‌زا باشند و نرخ خطای ارسال به صفر می‌رسد.<br><br>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})

plt.rcParams['figure.dpi'] = 135

np.random.seed(42)
```

```python
# ==========================================================
# Section 1: Scenario Introduction
# ==========================================================
display_academic_text(
    "پیش‌پردازش داده‌ها: مدیریت ویژگی‌های دودویی متقارن و نامتقارن (Binary Attributes)", 
    "main_title"
)
display_academic_text(
    "ویژگی‌های دودویی (Binary) به دو دسته <b>متقارن (Symmetric)</b> و <b>نامتقارن (Asymmetric)</b> تقسیم می‌گردند. "
    "در این سناریو، متغیرِ <i>جنسیت</i> به عنوان یک ویژگی دودویی متقارن (هیچ‌یک از دو حالت بر دیگری ارجحیتِ اطلاعاتی ندارد) و "
    "متغیرِ <i>حساسیت به بادام زمینی</i> به عنوان یک ویژگی دودویی نامتقارن (حالتِ 'بله' نادرتر و از نظر اطلاعاتی بسیار باارزش‌تر است) "
    "مورد بررسی قرار می‌گیرند. به منظور درک الگوریتمیک توسط ماشین، این متغیرهای متنی می‌بایست "
    "از طریق نگاشت‌های ساختاریافته به کدهای عددی (۰ و ۱) تبدیل شوند.", 
    "paragraph"
)

# ==========================================================
# Section 2: Data Initialization and Processing
# ==========================================================
display_academic_text(
    "مرحله نخست: مقداردهی اولیه و استخراج نگاشت‌های عددی", 
    "sub_title"
)
display_academic_text(
    "در این فاز، پایگاه داده خامِ سیستمِ توصیه‌گرِ غذا تعریف شده و سپس دو رویکرد متفاوت "
    "نگاشت (Mapping) بر روی آن اعمال می‌گردد. نتایج عملیات، جهت ارزیابی تحلیلی به صورت موازی (سایدبای‌ساید) به نمایش درآمده‌اند.", 
    "paragraph"
)

# Step 1: Initialize the raw database containing both symmetric and asymmetric binary attributes
raw_data = {
    'User_ID': [101, 102, 103, 104, 105],
    'Gender (Symmetric)': ['F', 'M', 'F', 'F', 'M'],                
    'Nut_Allergy (Asymmetric)': ['Yes', 'No', 'No', 'Yes', 'No']      
}
df_raw = pd.DataFrame(raw_data)

# Step 2: Create a copy of the dataframe to store the processed numerical mappings
df_processed = df_raw.copy()

# Step 3: Map the Symmetric Binary Attribute (Gender) to numerical values.
# In symmetric binary attributes, both states carry equal weight. E.g., 'F':1, 'M':0
df_processed['Gender_Numeric'] = df_processed['Gender (Symmetric)'].map({'F': 1, 'M': 0})

# Step 4: Map the Asymmetric Binary Attribute (Nut_Allergy) to numerical values.
# In asymmetric binary attributes, the RARE or CRITICAL state (having the allergy) 
# MUST be conventionally mapped to 1.
df_processed['Nut_Allergy_Numeric'] = df_processed['Nut_Allergy (Asymmetric)'].map({'Yes': 1, 'No': 0})

# Extract only the processed numerical columns along with the identifier for final visualization
df_final = df_processed[['User_ID', 'Gender_Numeric', 'Nut_Allergy_Numeric']]

# Render the original and processed dataframes side-by-side
# Using the list format expected by the global function
display_side_by_side_datasets(
    [df_raw, df_final],
    [
        "جدول ۱: پایگاه داده خام (متغیرهای متنی)", 
        "جدول ۲: پایگاه داده پردازش‌شده (فضای عددی)"
    ]
)

# ==========================================================
# Section 3: Visual Analytics (Academic Blue)
# ==========================================================
display_academic_text("مرحله دوم: تفسیرِ بصریِ ارزشِ اطلاعاتی (Information Value)", "sub_title")

# Plotting the Information Value difference
fig, axes = plt.subplots(1, 2, figsize=(16, 5), dpi=135)

# Plot 1: Symmetric Distribution
sns.countplot(x='Gender_Numeric', data=df_final, palette=['#bdc3c7', '#2980b9'], ax=axes[0], edgecolor='#2c3e50', linewidth=1.5)
axes[0].set_title('Figure 1: Symmetric Binary (Gender)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=12)
axes[0].set_xticklabels(['Male (0)', 'Female (1)'], fontsize=12, fontweight='bold')
axes[0].set_ylabel('Observation Count', fontsize=12, fontweight='bold')
axes[0].set_xlabel('')
axes[0].annotate(
    "Equal Information Value:\nBoth states are equally important.\nSwapping 0 and 1 changes nothing.",
    xy=(0.5, 2.5), xytext=(0.5, 3.2),
    ha='center', fontsize=11, fontweight='bold', color='#2c3e50',
    bbox=dict(boxstyle="round,pad=0.4", fc="#f8f9fa", ec="#bdc3c7", lw=1.5)
)

# Plot 2: Asymmetric Distribution
sns.countplot(x='Nut_Allergy_Numeric', data=df_final, palette=['#bdc3c7', '#e74c3c'], ax=axes[1], edgecolor='#2c3e50', linewidth=1.5)
axes[1].set_title('Figure 2: Asymmetric Binary (Nut Allergy)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=12)
axes[1].set_xticklabels(['No Allergy (0)', 'Has Allergy (1)'], fontsize=12, fontweight='bold')
axes[1].set_ylabel('')
axes[1].set_xlabel('')
axes[1].annotate(
    "Critical Imbalance:\nThe '1' state is rare but extremely important!\nFailing to detect a '1' here could be fatal.",
    xy=(1, 2.0), xytext=(0.5, 2.8),
    ha='center', fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

for ax in axes:
    ax.grid(axis='y', linestyle='--', alpha=0.5)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 4: Final Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی عملیات پردازش:</b><br><br>"
    r"همان‌طور که در نمودارهای بالا قابل مشاهده است، تبدیل ویژگی‌های دودویی نامتقارن (مانند حساسیت غذایی) به فرمت عددیِ <code>0 و 1</code> تصادفی نیست. "
    r"در سیستم‌های تصمیم‌یار، رخدادهای <b>نادر اما حیاتی</b> باید لزوماً با کدِ <code>1</code> رمزگذاری شوند تا در محاسبهٔ فواصل (Distance Metrics) و ضرایب شباهت (مثل Jaccard Similarity)، "
    r"وزن محاسباتی به درستی اِعمال شود. این الگوی نگاشت، احتمال بروز خطای سیستمی را در محیط‌های دارای ریسک بالا (مثل پیشنهاد اشتباه غذای حاوی بادام‌زمینی) به صفر تقلیل می‌دهد."
)
display_academic_text(insight_conclusion, "insight")
```
