  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: مقایسه آگهی‌های خودرو در «دیوار» با فاصله گاور 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    سیستم هوش مصنوعی «دیوار» برای پیدا کردن آگهی‌های مشابه یا پیشنهاد خودروهای هم‌رده به خریداران، مجبور است آگهی‌هایی را مقایسه کند که ویژگی‌های آن‌ها هم‌زمان از ۴ نوع داده مختلف تشکیل شده است:
  </p>

  <ul style="text-align: right; margin: 0 0 10px; padding-right: 22px;">
    <li><b>ویژگی عددی (Numeric):</b> قیمت خودرو به میلیون تومان (مثلاً $500$ در برابر $1200$ میلیون تومان).</li>
    <li><b>ویژگی دودویی (Binary):</b> نوع گیربکس ($1$ برای دنده اتوماتیک، $0$ برای دنده دستی).</li>
    <li><b>ویژگی اسمی (Nominal):</b> برند و مدل خودرو (<code>Peugeot 207</code>، <code>Dena Plus</code>، <code>JAC S5</code>).</li>
    <li><b>ویژگی ترتیبی (Ordinal):</b> وضعیت سلامت بدنه (<code>تصادفی</code> 👈 ۱، <code>دوررنگ</code> 👈 ۲، <code>بی‌رنگ</code> 👈 ۳).</li>
  </ul>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>چالش بزرگ:</b> معیارهای فاصله معمولی (مثل فاصله اقلیدسی یا جاکارد) روی داده‌های ترکیبی شکست می‌خورند؛ چون نمی‌توان «قیمت به میلیون تومان» را با «مدل متنی خودرو» در یک فرمول ریاضی جمع کرد!
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>راه‌حل — فاصله گاور (Gower's Distance):</b><br>
    فاصله گاور عدم تشابه بین دو آگهی $i$ و $j$ را با نرمال‌سازی اختصاصی هر نوع ویژگی بین <b>۰ تا ۱</b> حساب کرده و سپس میانگین وزن‌دار آن‌ها را به دست می‌آورد:
  </p>

  <div style="text-align: center; margin: 15px 0;">
    $$\text{Gower Distance}(i, j) = 1 - \frac{\sum_{k=1}^{p} w_k \cdot s_{ijk}}{\sum_{k=1}^{p} w_k}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>نحوه برخورد فاصله گاور با انواع متغیرها ($s_{ijk}$):</b><br>
    &nbsp;&nbsp;&nbsp;• <b>ویژگی‌های عددی و ترتیبی:</b> اختلاف دو مقدار تقسیم بر دامنه (Range) کل آن ویژگی $1 - \frac{|x_{ik} - x_{jk}|}{R_k}$. با این کار، اعداد بزرگ قیمت نفوذ کاذب پیدا نمی‌کنند.<br>
    &nbsp;&nbsp;&nbsp;• <b>ویژگی‌های اسمی و دودویی:</b> اگر دو مقدار کاملاً یکسان باشند شباهت $1$ و در غیر این صورت $0$ منظور می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> خروجی فاصله گاور همواره یک عدد استاندارد بین <b>۰ (تطابق کامل)</b> تا <b>۱ (عدم تشابه مطلق)</b> است که به دیوار اجازه می‌دهد خودروها را با ترکیب ویژگی‌های ناهمگن به‌دقت گروه‌بندی کند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: محاسبه استاندارد عدم تشابه ترکیبی روی داده‌های عددی، اسمی، ترتیبی و دودویی
    </div>
  </p>

</div>

```python
display_academic_text(
    "معیارهای فاصله‌یابی: سنجش عدم تشابه در فضای ویژگی‌های ترکیبی با الگوریتم گاور (Gower's Distance) 🚙", 
    "main_title"
)

display_academic_text(
    "در پایگاه‌های داده‌ای که موجودیت‌ها توسط ترکیبی از ویژگی‌های <i>عددی، اسمی، ترتیبی و دودویی</i> توصیف می‌شوند (Heterogeneous Data)، "
    "توابع فاصله کلاسیک (نظیر اقلیدسی) فاقد اعتبار ریاضی هستند. در این سناریو، یک پلتفرم نیازمند محاسبه میزان شباهت آگهی‌های خودرو است. "
    "الگوریتم <b>فاصله گاور (Gower's Distance)</b> با اتخاذ رویکردی تفکیکی، میزان شباهت هر ویژگی را مستقلاً بر اساس نوع آن محاسبه نموده "
    "و سپس با تجمیعِ نرمال‌سازی‌شدهٔ آن‌ها، یک ماتریس فاصلهٔ بی‌واحد و یکپارچه در بازه $[0, 1]$ تولید می‌نماید.", 
    "paragraph"
)

display_academic_text("معادله اصلی فاصله گاور میان دو موجودیت $i$ و $j$:", "sub_title")
display_academic_text(
    r"$$d(i, j) = 1 - \frac{\sum_{k=1}^{p} w_k \cdot s_{ijk}}{\sum_{k=1}^{p} w_k}$$", 
    "paragraph"
)

display_academic_text("الگوریتم شباهت‌سنجی ($s_{ijk}$) متغیرهای عددی و ترتیبی مبتنی بر تفاضل مقادیر نسبت به دامنه ویژگی ($R_k$):", "sub_title")
display_academic_text(
    r"$$s_{ijk} = 1 - \frac{|x_{ik} - x_{jk}|}{R_k}$$", 
    "paragraph"
)

# ==========================================================
# SECTION 2: DATA INITIALIZATION AND CUSTOM ALGORITHM
# ==========================================================
display_academic_text(
    "مرحله نخست: بارگذاری داده‌های ناهمگن و پردازش ماتریس فاصله گاور", 
    "sub_title"
)

display_academic_text(
    "در این فاز، داده‌های خودرو با حفظ ماهیت ترکیبی در سیستم بارگذاری می‌شوند. سپس تابع سفارشیِ الگوریتم گاور، "
    "ضمن ارزیابی نوع هر ویژگی، مکانیزم ریاضیِ متناسب با آن را فراخوانی کرده و در نهایت <b>ماتریس مربعی متقارن عدم تشابه</b> را تولید می‌نماید.", 
    "paragraph"
)

# Step 1: Initialize the mixed-attribute database for car listings
# Attributes include Numeric (Price), Binary (Transmission), Nominal (Brand), and Ordinal (Body Condition)
data = {
    'Listing_ID': ['Car_1', 'Car_2', 'Car_3'],
    'Price_Million_Toman': [600, 650, 1400],           
    'Is_Automatic': [1, 1, 0],                         
    'Car_Brand': ['Peugeot 207', 'Peugeot 207', 'JAC S5'], 
    'Body_Condition': [3, 2, 1], # 3: Excellent, 2: Good, 1: Poor                       
}
df_mixed = pd.DataFrame(data).set_index('Listing_ID')

# Dictionary to store the independent feature similarities for Car_1 vs Car_2 visualization
feature_sims_1_2 = {}

# Step 2: Custom implementation of the Gower's Distance computation engine
def compute_gower_distance(df):
    """
    Computes a pairwise Gower distance matrix for mixed data types.
    Assumes equal weighting (w_k = 1) for all attributes.
    """
    n = len(df)
    gower_matrix = np.zeros((n, n))

    # Calculate absolute ranges (R_k) for numerical/ordinal scaling
    # Added max(..., 1e-9) to prevent mathematical ZeroDivisionError
    price_range = max(df['Price_Million_Toman'].max() - df['Price_Million_Toman'].min(), 1e-9)
    body_range = max(df['Body_Condition'].max() - df['Body_Condition'].min(), 1e-9)

    # O(n^2) loop to calculate pairwise dissimilarity across all combinations
    for i in range(n):
        for j in range(n):
            
            # 1. Similarity computation for Numeric attribute (Price)
            s_price = 1.0 - (abs(df.iloc[i]['Price_Million_Toman'] - df.iloc[j]['Price_Million_Toman']) / price_range)

            # 2. Similarity computation for Binary/Asymmetric attribute (Automatic Gearbox)
            s_binary = 1.0 if (df.iloc[i]['Is_Automatic'] == df.iloc[j]['Is_Automatic']) else 0.0

            # 3. Similarity computation for Nominal/Categorical attribute (Car Brand)
            s_nominal = 1.0 if (df.iloc[i]['Car_Brand'] == df.iloc[j]['Car_Brand']) else 0.0

            # 4. Similarity computation for Ordinal attribute (Body Condition)
            s_ordinal = 1.0 - (abs(df.iloc[i]['Body_Condition'] - df.iloc[j]['Body_Condition']) / body_range)

            # Aggregate similarity: Compute the unweighted average across all p=4 features
            avg_similarity = (s_price + s_binary + s_nominal + s_ordinal) / 4.0
            
            # Store isolated similarities for Car 1 vs Car 2 specifically for our plot
            if i == 0 and j == 1:
                feature_sims_1_2['Numeric (Price)'] = s_price
                feature_sims_1_2['Binary (Gearbox)'] = s_binary
                feature_sims_1_2['Nominal (Brand)'] = s_nominal
                feature_sims_1_2['Ordinal (Body)'] = s_ordinal

            # Convert final similarity score to distance (Distance = 1 - Similarity)
            gower_matrix[i, j] = 1.0 - avg_similarity

    return pd.DataFrame(gower_matrix, index=df.index, columns=df.index)

# Step 3: Execute the computation engine to derive the Gower Matrix
df_gower_dist = compute_gower_distance(df_mixed)

# Render the datasets cleanly (reset_index prevents hidden columns in HTML)
df_mixed_disp = df_mixed.reset_index()
df_gower_disp = df_gower_dist.reset_index().rename(columns={'Listing_ID': 'Target / Reference'})

display_side_by_side_datasets(
    [df_mixed_disp, df_gower_disp],
    [
        "جدول ۱: پایگاه داده آگهی‌ها (ویژگی‌های ناهمگن)", 
        "جدول ۲: ماتریس فاصله گاور (بازه استاندارد ۰ تا ۱)"
    ]
)

# ==========================================================
# SECTION 3: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی مکانیزم محاسبه و توپولوژی فواصل ترکیبی", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکی زیر معماری الگوریتم گاور را کالبدشکافی می‌کند. <b>نمودار سمت چپ</b> نشان می‌دهد که چگونه فاصلهٔ کلی بین آگهی ۱ و ۲ محاسبه شده است "
    "(تجمیع شباهت‌ها در ابعاد مختلف). <b>نقشه حرارتی سمت راست</b> خروجی نهایی الگوریتم را نمایش می‌دهد. تقاطع <code>(Car_1, Car_2)</code> با نمایش فاصله‌ای معادل <code>0.141</code>، "
    "نشان‌دهندهٔ شباهت بسیار بالا میان این دو آگهی است. در مقابل، <code>(Car_1, Car_3)</code> با فاصله حداکثری <code>1.000</code>، عدم تطابق مطلق در تمامی ویژگی‌ها را به تصویر می‌کشد.", 
    "paragraph"
)

# Initialize a dual-axis canvas for visual comparative analytics
fig, axes = plt.subplots(1, 2, figsize=(19, 6.5), dpi=135)

# --- SUBPLOT 1: Feature-by-Feature Similarity Decomposition ---
features = list(feature_sims_1_2.keys())
similarities = list(feature_sims_1_2.values())
average_sim = sum(similarities) / len(similarities)

# Bar chart of individual similarities
bars = axes[0].barh(features, similarities, color='#2980b9', edgecolor='#2c3e50', linewidth=1.5, height=0.6)

# Add data labels
for bar in bars:
    axes[0].text(
        bar.get_width() - 0.05 if bar.get_width() > 0.1 else bar.get_width() + 0.02, 
        bar.get_y() + bar.get_height()/2, 
        f'{bar.get_width():.3f}', 
        va='center', ha='right' if bar.get_width() > 0.1 else 'left', 
        fontweight='bold', fontsize=12, color='white' if bar.get_width() > 0.1 else '#2c3e50'
    )

# Add a dashed line representing the Final Average Similarity
axes[0].axvline(average_sim, color='#e74c3c', linestyle='--', linewidth=3.0, label=f'Avg Similarity (S) = {average_sim:.3f}')

axes[0].annotate(
    f"Gower Distance Calculation:\nDistance = 1 - Similarity\nd(1,2) = 1.000 - {average_sim:.3f} = {1 - average_sim:.3f}",
    xy=(average_sim, 3.5), xytext=(0.1, 3.5),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title('Figure 1: Similarity Decomposition (Car_1 vs Car_2)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[0].set_xlabel('Independent Feature Similarity ($s_{1,2,k}$)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_xlim(0, 1.05)
axes[0].legend(loc='lower left', frameon=True, fontsize=11)
axes[0].grid(axis='x', linestyle='--', alpha=0.5)

# --- SUBPLOT 2: Topology of Gower Distance Heatmap ---
# Using an Academic Blue colormap. Dark Blue = 0 distance (Identical), White = 1 distance (Totally Dissimilar)
sns.heatmap(
    df_gower_dist, 
    annot=True, 
    cmap='Blues_r', 
    fmt='.3f', 
    vmin=0, vmax=1, 
    ax=axes[1],
    linewidths=2.0, 
    linecolor='white',
    cbar_kws={'label': 'Gower Distance ($d$)'},
    annot_kws={"size": 15, "weight": "bold"}
)

axes[1].set_title('Figure 2: Full Topology of Gower Distance Matrix', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[1].set_xlabel('Target Listing (Object j)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_ylabel('Reference Listing (Object i)', fontdict={'fontsize': 13, 'fontweight': 'bold'})

axes[1].set_xticklabels(axes[1].get_xticklabels(), fontweight='bold', fontsize=12, rotation=0)
axes[1].set_yticklabels(axes[1].get_yticklabels(), fontweight='bold', fontsize=12, rotation=0, va='center')

# Remove top and right spines
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی عملیات (Strategic Implication):</b><br>"
    r"اتکا به معیارهای کلاسیک (نظیر فاصله اقلیدسی) در محیط‌های چندوجهی (Heterogeneous)، منجر به تخریبِ بارِ معناییِ ویژگی‌های اسمی و دودویی می‌گردد. "
    r"شما نمی‌توانید تفاوت کلمهٔ «پژو» با «جک» را به توان ۲ برسانید!<br><br>"
    r"الگوریتم <b>فاصله گاور (Gower)</b> با ایزوله‌سازی ریاضیاتی متغیرها (همان‌طور که در گراف ۱ تجزیه و اثبات شد) و اعمال نرمال‌سازیِ متناسب با ماهیتِ ذاتیِ هر یک از آن‌ها، "
    r"به سیستمِ توصیه‌گرِ پلتفرمِ خودرو این قابلیت را می‌بخشد تا آگهی‌های کاملاً ناهمگون را در یک <i>فضای متریکِ یکپارچه و استاندارد</i> ارزیابی نموده "
    r"و پیشنهادات مشابه (نظیر پیشنهاد Car_2 به بازدیدکنندهٔ Car_1) را با دقت مطلق اجرا نماید."
)
display_academic_text(insight_conclusion, "insight")
```
