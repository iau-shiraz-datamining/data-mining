<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: سنجش ثبات کیفیت پخش ویدیو در «نتفلیکس» (Netflix) 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    فرض کنید تیم فنی «نتفلیکس» می‌خواهد کیفیت دو سرور پخش آنلاین (CDN) مختلف را مقایسه کند. میانگین زمان معطلی (بافرینگ) فیلم در هر دو سرور دقیقاً <b>۵ ثانیه</b> است! اما کاربران سرور دوم مدام از قطعی و پرش تصویر شکایت می‌کنند. چرا؟ چون میانگین به تنهایی نوسانات را نشان نمی‌دهد؛ اینجاست که <b>معیارهای پراکندگی داده</b> راز قضیه را برملا می‌کنند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۱. دامنه (Range) — ساده‌ترین ابزار سنجش فاصله نوسان</b><br>
    زمان‌های بافرینگ در سرور اول: [۴، ۵، ۵، ۵، ۶] ثانیه 👈 دامنه = ۶ - ۴ = <b>۲ ثانیه</b>.<br>
    زمان‌های بافرینگ در سرور دوم: [۱، ۱، ۵، ۹، ۹] ثانیه 👈 دامنه = ۹ - ۱ = <b>۸ ثانیه</b>.<br>
    دامنه بزرگ‌تر در سرور دوم نشان می‌دهد که تفاوت بین بهترین و بدارین تجربه کاربر بسیار زیاد است.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۲. واریانس (Variance) — محاسبه شدت جریمه نوسانات</b><br>
    واریانس میانگین مجذور (توان دوم) فاصله اعداد از میانگین است. با مربعی کردن فاصله‌ها، نوسانات بزرگ جریمه سنگین‌تری می‌شوند.<br>
    &nbsp;&nbsp;&nbsp;• واریانس سرور اول = <b>۰.۴</b> (بسیار کم و پایدار)<br>
    &nbsp;&nbsp;&nbsp;• واریانس سرور دوم = <b>۱۲.۸</b> (بسیار بالا و ناپایدار)<br>
    <i>مشکل واریانس:</i> واحد آن «ثانیه به‌توان ۲» است که تفسیر ملموسی برای مدیر محصول ندارد!
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۳. انحراف معیار (Standard Deviation) — معیاری ملموس با واحد اصلی</b><br>
    با گرفتن جذر (ریشه دوم) از واریانس، واحد اندازه دوباره به «ثانیه» برمی‌گردد:<br>
    &nbsp;&nbsp;&nbsp;• انحراف معیار سرور اول = <b>۰.۶۳ ثانیه</b> (تجربه یکنواخت و قابل پیش‌بینی).<br>
    &nbsp;&nbsp;&nbsp;• انحراف معیار سرور دوم = <b>۳.۵۷ ثانیه</b> (تجربه غیرقابل پیش‌بینی و پر از شوک).
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> اگرچه میانگین هر دو سرور ۵ ثانیه بود، اما نتفلیکس سرور اول را انتخاب می‌کند؛ زیرا انحراف معیار پایین‌تر به معنای ثبات پخش و رضایت کاربر است.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: کشف نوسانات پنهان در داده‌ها با کمک انحراف معیار و واریانس
    </div>
  </p>

```python
display_academic_text(
    "آمار توصیفی: ارزیابی معیارهای پراکندگی داده (Data Dispersion) و کشف نوسانات پنهان 📉", 
    "main_title"
)

display_academic_text(
    "معیارهای گرایش به مرکز (مانند <i>میانگین</i>) به تنهایی تصویر کاملی از توزیع داده‌ها ارائه نمی‌دهند و می‌توانند <b>به شدت گمراه‌کننده</b> باشند. "
    "در این سناریو، دو سرور پخش آنلاین (CDN) با میانگینِ بافرینگِ کاملاً یکسان (۵ ثانیه) مورد بررسی قرار می‌گیرند. "
    "استفاده از معیارهای پراکندگی شامل <b>دامنه (Range)</b>، <b>واریانس (Variance)</b> و <b>انحراف معیار (Standard Deviation)</b>، "
    "جهت کشف تفاوت‌های ساختاری در پایداری (Stability) این دو سیستم به کار گرفته شده است.", 
    "paragraph"
)

# ==========================================================
# SECTION 2: DATA INITIALIZATION & DISPERSION COMPUTATION
# ==========================================================
display_academic_text(
    "مرحله نخست: بارگذاری داده‌های خام و استخراج شاخص‌های پراکندگی", 
    "sub_title"
)

display_academic_text(
    "در این فاز، زمان‌های معطلی دو سرور ثبت گردیده است. سپس شاخص‌های پراکندگی بر مبنای فرمول <b>واریانس جامعه (Population Variance)</b> استخراج شده "
    "و در قالب یک ماتریس تجمیعی جهت مقایسهٔ عملکرد، در کنار پایگاه داده خام رندر می‌شوند.", 
    "paragraph"
)

# Step 1: Initialize the mock database for Netflix server buffering times.
# Both servers intentionally share the exact same mean (5.0 seconds).
raw_data = {
    'Test_ID': [1, 2, 3, 4, 5],
    'Server_A_Sec': [4.0, 5.0, 5.0, 5.0, 6.0], 
    'Server_B_Sec': [1.0, 1.0, 5.0, 9.0, 9.0]  
}
df_raw = pd.DataFrame(raw_data)

# Step 2: Calculate the Range for both servers (Max - Min)
range_a = df_raw['Server_A_Sec'].max() - df_raw['Server_A_Sec'].min()
range_b = df_raw['Server_B_Sec'].max() - df_raw['Server_B_Sec'].min()

# Step 3: Calculate the Population Variance using ddof=0 (Delta Degrees of Freedom = 0).
# This penalizes larger deviations from the mean squarely.
var_a = df_raw['Server_A_Sec'].var(ddof=0)
var_b = df_raw['Server_B_Sec'].var(ddof=0)

# Step 4: Calculate the Population Standard Deviation (Square root of variance).
# This returns the dispersion metric back to the original unit (seconds).
std_a = df_raw['Server_A_Sec'].std(ddof=0)
std_b = df_raw['Server_B_Sec'].std(ddof=0)

# Step 5: Construct a summary dataframe to present the calculated dispersion metrics.
summary_data = {
    'Statistical Measure': [
        'Arithmetic Mean (Center)', 
        'Data Range (Max - Min)', 
        'Population Variance', 
        'Standard Deviation (Spread)'
    ],
    'Server A (Stable)': [
        f"{df_raw['Server_A_Sec'].mean():.1f} sec", 
        f"{range_a:.1f} sec", 
        f"{var_a:.2f}", 
        f"{std_a:.2f} sec"
    ],
    'Server B (Unstable)': [
        f"{df_raw['Server_B_Sec'].mean():.1f} sec", 
        f"{range_b:.1f} sec", 
        f"{var_b:.2f}", 
        f"{std_b:.2f} sec"
    ]
}
df_summary = pd.DataFrame(summary_data)

# Render the raw observation data and the extracted dispersion statistics side-by-side
# Using the list format expected by the global function
display_side_by_side_datasets(
    [df_raw, df_summary],
    [
        "جدول ۱: رکوردهای ثبت‌شده زمان بافرینگ (ثانیه)", 
        "جدول ۲: تحلیل مقایسه‌ای شاخص‌های پراکندگی"
    ]
)

# ==========================================================
# SECTION 3: ANALYTICAL VISUALIZATION (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلیِ نوسانات (Variance) و پایداری سیستم", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکی زیر ماهیت پنهانِ داده‌ها را آشکار می‌سازد. در نمودار پراکندگی (سمت چپ)، مشخص است که داده‌های <b>سرور A</b> به شدت در اطراف "
    "خطِ میانگین متمرکز هستند، در حالی که داده‌های <b>سرور B</b> دچار پراکندگی شدید و شوکِ زمانی می‌باشند. "
    "نمودار چگالی توزیع (سمت راست) نیز این تمرکز قطعی (قله بلند) را در برابر نوسانِ شدید (قله پهن) تأیید می‌نماید.", 
    "paragraph"
)

# Initialize a wide dual-axis canvas for visual comparative analytics
fig, axes = plt.subplots(1, 2, figsize=(19, 6.5), dpi=135)

# Restructure the dataframe into a long format using pandas 'melt' for seaborn compatibility
df_melted = df_raw.melt(
    id_vars='Test_ID', 
    value_vars=['Server_A_Sec', 'Server_B_Sec'], 
    var_name='Server_Type', 
    value_name='Buffering_Time'
)

# --- SUBPLOT 1: Absolute Data Dispersion (Strip Plot) ---
sns.stripplot(
    data=df_melted, 
    x='Server_Type', 
    y='Buffering_Time', 
    ax=axes[0],
    palette=['#1abc9c', '#e74c3c'], # Turquoise for Stable, Red for Unstable
    size=15, 
    jitter=0.05, 
    edgecolor='#2c3e50', 
    linewidth=1.5, 
    zorder=5
)

# Overlay the theoretical identical mean line for visual reference
axes[0].axhline(
    5.0, 
    color='#2c3e50', # Navy Blue
    linestyle='--', 
    linewidth=3.0, 
    zorder=1, 
    label='Identical Mean (5.0 sec)'
)

# Intelligent Annotations
axes[0].annotate(
    "High Variance (Unstable):\nUsers either experience instant loads (1s)\nor suffer severe lag (9s).\nThe 5s average is a pure illusion!",
    xy=(1.1, 9.0), xytext=(1.4, 6.5),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].annotate(
    "Low Variance (Stable):\nEvery user gets consistent performance.\nThe standard deviation is practically zero.",
    xy=(0.1, 4.0), xytext=(0.4, 1.5),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

# Formatting Subplot 1
axes[0].set_title('Figure 1: Absolute Data Dispersion around the Mean', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[0].set_xlabel('CDN Server Identification', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_ylabel('Buffering Duration (Seconds)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_xticklabels(['Server A (Stable)', 'Server B (Unstable)'], fontweight='bold')
axes[0].legend(loc='lower center', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=11)
axes[0].grid(True, linestyle='--', alpha=0.5, zorder=0)


# --- SUBPLOT 2: Probability Density Distribution (KDE) ---
sns.kdeplot(
    data=df_raw, 
    x='Server_A_Sec', 
    ax=axes[1], 
    fill=True, 
    color='#1abc9c', 
    label='Server A (Low Std Dev)', 
    linewidth=2.5
)
sns.kdeplot(
    data=df_raw, 
    x='Server_B_Sec', 
    ax=axes[1], 
    fill=True, 
    color='#e74c3c', 
    label='Server B (High Std Dev)', 
    linewidth=2.5
)

# Overlay the identical mean line on the density plot
axes[1].axvline(
    5.0, 
    color='#2c3e50', 
    linestyle='--', 
    linewidth=3.0, 
    label='Identical Mean (5.0 sec)'
)

# Formatting Subplot 2
axes[1].set_title('Figure 2: Probability Density Distribution (Shape of Variance)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[1].set_xlabel('Buffering Duration (Seconds)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_ylabel('Density (Probability Mass)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].legend(loc='upper right', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=11)
axes[1].grid(True, linestyle='--', alpha=0.5)

# Remove the top and right spines from both axes for a cleaner academic aesthetic
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی عملیات (Strategic Implication):</b><br><br>"
    r"ارزیابی کیفیت خدمات بر مبنای «میانگین» به تنهایی، خطای سیستمی به همراه خواهد داشت. سرور دوم (B) علیرغم داشتنِ میانگینِ کاملاً یکسان با سرور اول، "
    r"دارای <b>انحراف معیار بسیار بالایی (۳.۵۸ ثانیه)</b> است که مستقیماً به معنای تجربه کاربریِ غیرقابل‌پیش‌بینی، و همراه با شوک‌های ناشی از قطعیِ طولانیِ تصویر (۹ ثانیه) می‌باشد.<br><br>"
    r"در علم داده، <i>«واریانس، دشمنِ پیش‌بینی‌پذیری است»</i>. انتخاب سرور اول (A) بر مبنای انحراف معیارِ پایین‌تر (۰.۶۳ ثانیه)، تضمین‌کنندهٔ ثبات و یکنواختیِ جریان داده برای تمام کاربران است."
)
display_academic_text(insight_conclusion, "insight")
```

در این مثال:
<ul>
<li>ما یک دیتافریم شامل داده‌های بافرینگ دو سرور نتفلیکس (<b>Netflix_Server_A</b> و <b>Netflix_Server_B</b>) ساختیم که میانگین زمان معطلی در هر دو سرور برابر با ۵ ثانیه است.</li>
<li><b>دامنه (Range):</b> با تفریق مقدار کمینه از بیشینه (با متدهای <b>.max()</b> و <b>.min()</b>)، اختلاف پخش‌شدگی داده‌ها مشخص شد (۲ ثانیه در برابر ۸ ثانیه).</li>
<li><b>واریانس (Variance):</b> با تابع <b>.var(ddof=0)</b> میانگین مجذور انحرافات حساب شد؛ واریانس سرور B عدد ۱۶ را نشان داد که جریمه نوسانات شدید در آن سرور است.</li>
<li><b>انحراف معیار (Standard Deviation):</b> با تابع <b>.std(ddof=0)</b> جذر واریانس گرفته شد تا انحراف معیار با همان واحد اصلی داده‌ها (ثانیه) به دست آید.</li>
<li>نتیجه کد اثبات کرد که انحراف معیار سرور A (عدد ۰.۶۳ ثانیه) بسیار کمتر از سرور B (عدد ۴.۰۰ ثانیه) است، که نشان‌دهنده کیفیت و ثبات بالای سرور A برای کاربران است.</li>
</ul>
