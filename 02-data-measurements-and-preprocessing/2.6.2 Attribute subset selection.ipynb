  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: پیش‌بینی لغو اشتراک و حذف ویژگی‌های زائد در «آمازون» (Amazon) 🎯
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    تیم داده‌کاوی «آمازون» (Amazon) می‌خواهد مدل هوش مصنوعی طراحی کند که **لغو اشتراک کاربران (Churn)** را پیش‌بینی کند. در بانک اطلاعاتی، بیش از ۲۰ ویژگی مختلف برای هر کاربر وجود دارد؛ از «مبلغ کل خرید» و «تعداد تیکت‌های پشتیبانی» گرفته تا متغیرهای بی‌ارزشی مثل «ساعت ثبت‌نام»، «نوع مرورگر» یا «کد پستی».
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>چالش بزرگ — «نفرین ابعاد» (Curse of Dimensionality):</b><br>
    وارد کردن تمام ویژگی‌ها به مدل هوش مصنوعی نه تنها سرعت آموزش را بسیار کند می‌کند، بلکه وجود **ویژگی‌های نامرتبط و نویزدار** باعث کاهش دقت مدل و وقوع پدیده بیش‌برازش (Overfitting) می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>راه‌حل — انتخاب زیرمجموعه ویژگی‌ها (Feature Selection):</b><br>
    با استفاده از روش‌های آماری (مانند آزمون ANOVA F-value) یا الگوریتم‌های مبتنی بر مدل (مثل درخت تصمیم و Random Forest)، امتیاز اهمیت هر ویژگی محاسبه شده و فقط $K$ ویژگی برتر که بیشترین ارتباط را با متغیر هدف دارند، انتخاب می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>مزایای اصلی انتخاب ویژگی:</b><br>
    &nbsp;&nbsp;&nbsp;• <b>افزایش دقت:</b> حذف نویز و متغیرهای بی‌ارزش.<br>
    &nbsp;&nbsp;&nbsp;• <b>کاهش زمان آموزش:</b> سبک‌تر شدن محاسبات الگوریتم.<br>
    &nbsp;&nbsp;&nbsp;• <b>تفسیرپذیری بهتر:</b> فهم آسان‌تر دلایل اصلی لغو اشتراک برای مدیران.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> با انتخاب ۴ ویژگی کلیدی از بین ۸ ویژگی موجود، مدل آمازون با دو برابر سرعت بیشتر و بدون نویز، دقیق‌ترین پیش‌بینی را انجام می‌دهد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: جداسازی هوشمندانه ویژگی‌های کلیدی از متغیرهای نویزدار و بی‌ارزش
    </div>
  </p>

</div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_selection import SelectKBest, f_classif 

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})

plt.rcParams['figure.dpi'] = 135

np.random.seed(42)
```

```python
display_academic_text(
    "کاهش ابعاد: انتخاب ویژگی‌های کلیدی با آزمون آماری (ANOVA F-value / SelectKBest) 🎯", 
    "main_title"
)

display_academic_text(
    "در مسائل پیش‌بینی رفتار مشتریان، انباشت ویژگی‌های نامرتبط (Noise) موجب پیچیدگی محاسباتی و بروز پدیدهٔ <b>بیش‌برازش (Overfitting)</b> می‌گردد. "
    "فرآیند <b>«انتخاب ویژگی» (Feature Selection)</b> با استفاده از معیارهای ارزیابی آماری، متغیرهایی را که <i>بیشترین همبستگی و قدرت تفکیک</i> را "
    "با متغیر هدف (مثلاً ریزش مشتری یا Churn) دارند، شناسایی و استخراج می‌نماید.", 
    "paragraph"
)

# استفاده از رشته خام (r"") برای جلوگیری از خطای مفسر پایتون در پردازش کاراکترهای بک‌اسلش
display_academic_text(
    r"معادله شاخص آماری ANOVA F-Score (نسبت واریانسِ بین گروه‌ها به واریانسِ درون گروه‌ها):", 
    "sub_title"
)
display_academic_text(
    r"$$F = \frac{\text{Between-Group Variability}}{\text{Within-Group Variability}}$$", 
    "paragraph"
)


# ==========================================================
# SECTION 2: DATASET GENERATION & FEATURE SELECTION PIPELINE
# ==========================================================
display_academic_text(
    "مرحله نخست: تولید مجموعه داده ترکیبی و رتبه‌بندی اهمیت ویژگی‌ها", 
    "sub_title"
)

display_academic_text(
    "در این فاز، ۸ ویژگی برای خریداران شبیه‌سازی می‌شود که ۴ مورد آن <i>اطلاعات واقعی رفتاری (مهم)</i> و ۴ مورد دیگر <i>نویز تصادفی</i> هستند. "
    "سپس الگوریتم <code>SelectKBest</code> با بهره‌گیری از آزمون F-value، امتیاز ارتباط هر ویژگی را محاسبه نموده و <b>۴ ویژگیِ برتر</b> را غربال می‌نماید.", 
    "paragraph"
)

# Step 1: Initialize synthetic Amazon customer dataset (8 features: 4 relevant, 4 noise)
n_samples = 250

# Generate 4 informative features
total_spend = np.random.normal(300, 100, n_samples)
support_tickets = np.random.poisson(2, n_samples)
monthly_visits = np.random.normal(15, 5, n_samples)
account_age_days = np.random.uniform(30, 1000, n_samples)

# Generate 4 random noise features (irrelevant variables)
random_noise_1 = np.random.normal(0, 1, n_samples)
random_noise_2 = np.random.uniform(1, 100, n_samples)
signup_hour = np.random.randint(0, 24, n_samples)
browser_code = np.random.randint(1, 5, n_samples)

# Generate target variable (Churn: 0 = Stayed, 1 = Churned)
churn_score = (
    support_tickets * 0.8 - 
    (total_spend / 200) - 
    (monthly_visits / 10)
)
churn_prob = 1 / (1 + np.exp(-churn_score))
churn = (churn_prob > np.median(churn_prob)).astype(int)

# Package into a structured DataFrame
X_data = {
    'Total_Spend': total_spend,
    'Support_Tickets': support_tickets,
    'Monthly_Visits': monthly_visits,
    'Account_Age_Days': account_age_days,
    'Random_Noise_1': random_noise_1,
    'Random_Noise_2': random_noise_2,
    'Signup_Hour': signup_hour,
    'Browser_Code': browser_code,
}
X = pd.DataFrame(X_data)
y = churn

# Step 2: Apply Feature Selection using SelectKBest (ANOVA F-test)
k_best = 4
selector = SelectKBest(score_func=f_classif, k=k_best)
X_new = selector.fit_transform(X, y)

# Extract and sort feature importance scores
scores = selector.scores_
feature_names = X.columns

# Step 3: Map the Selection Status to sort and highlight exactly what the algorithm did
df_scores = pd.DataFrame({
    'Feature_Name': feature_names, 
    'ANOVA_F_Score': scores,
    'Selected': selector.get_support()
}).sort_values(by='ANOVA_F_Score', ascending=False).reset_index(drop=True)

df_scores['Selection_Status'] = ['Selected (Top K)' if m else 'Discarded (Noise)' for m in df_scores['Selected']]
df_scores = df_scores.drop(columns=['Selected'])

# Display preparation to prevent empty index columns in HTML
df_x_disp = X.head(6).reset_index(drop=True)

# Render the raw features preview and the ranked feature importance score matrix side-by-side
display_side_by_side_datasets(
    [df_x_disp, df_scores],
    [
        "جدول ۱: ویژگی‌های خام اولیه (۸ بُعدی - مخلوطی از سیگنال و نویز)", 
        "جدول ۲: رتبه‌بندی امتیازات اهمیت ویژگی‌ها (آزمون ANOVA)"
    ]
)

# ==========================================================
# SECTION 3: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی - تفکیک ویژگی‌های کلیدی از متغیرهای نویز", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکی زیر نتایج غربال‌گری آماری را به تصویر می‌کشد. میله‌های <b>فیروزه‌ای رنگ</b> نشان‌دهندهٔ ۴ ویژگیِ حیاتی هستند که بالاترین "
    "امتیاز F-Score را کسب کرده‌اند و مستقیماً با رفتار ریزش کاربر مرتبط‌اند (به خصوص تعداد تیکت‌های پشتیبانی). "
    "در مقابل، میله‌های <b>خاکستری‌رنگ</b> (نویزهای تصادفی و متغیرهای بی‌ارزشی مانند کد مرورگر) به دلیل امتیاز نزدیک به صفر، "
    "توسط سیستم حذف شده‌اند تا مدل یادگیری ماشین با تمرکز کامل آموزش ببیند.", 
    "paragraph"
)

# Initialize a plot canvas optimized for horizontal bar ranking
plt.figure(figsize=(12, 6.5), dpi=135)

# Define colors (Highlight selected k-best in Academic Blue/Turquoise, discarded noise in muted gray)
bar_colors = [
    '#1abc9c' if status == 'Selected (Top K)' else '#bdc3c7' 
    for status in df_scores['Selection_Status']
]

# Plot horizontal bar chart
bars = plt.barh(
    df_scores['Feature_Name'], 
    df_scores['ANOVA_F_Score'], 
    color=bar_colors, 
    edgecolor='#2c3e50',
    linewidth=1.5,
    height=0.6
)

# Place the highest scoring feature on top
plt.gca().invert_yaxis()  

# Intelligent Annotations
plt.annotate(
    "The Strongest Predictor:\nCustomers submitting a high number of\nSupport Tickets are highly likely to churn.\nThis is the most critical feature.",
    xy=(60, 0.2), xytext=(80, 2.0),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

plt.annotate(
    "Filtered Noise:\nThe algorithm correctly realized that the\nTime of Signup or the Browser Type have\nabsolutely nothing to do with churn.",
    xy=(3, 6), xytext=(40, 6.5),
    fontsize=11, fontweight='bold', color='#7f8c8d',
    arrowprops=dict(arrowstyle="->", color="#95a5a6", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#f8f9fa", ec="#bdc3c7", lw=1.5)
)

# Enhance plot labeling, titles, and structural configuration
plt.title(f'Figure 1: Feature Selection via SelectKBest (Top {k_best} Attributes)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
plt.xlabel('ANOVA F-Score (Higher Value = Stronger Relationship with Target)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
plt.ylabel('Dataset Feature Attributes', fontdict={'fontsize': 13, 'fontweight': 'bold'})
plt.grid(True, linestyle='--', alpha=0.5, axis='x')

# Make tick labels bold
plt.yticks(fontweight='bold', fontsize=11)
plt.xticks(fontweight='bold', fontsize=11)

# Add precise score text annotations on the bars
for bar in bars:
    width = bar.get_width()
    plt.text(
        width + 2.0, 
        bar.get_y() + bar.get_height() / 2, 
        f'{width:.1f}', 
        ha='left', 
        va='center', 
        fontsize=11,
        fontweight='bold',
        color='#2c3e50'
    )

# Clear top and right spines for a clean academic aesthetic
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

# Add an explicit legend explaining the colors
import matplotlib.patches as mpatches
selected_patch = mpatches.Patch(color='#1abc9c', label='Selected Feature (Signal)')
discarded_patch = mpatches.Patch(color='#bdc3c7', label='Discarded Feature (Noise)')
plt.legend(handles=[selected_patch, discarded_patch], loc='center right', frameon=True, fontsize=11)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی عملیات (Strategic Implication):</b><br>"
    r"انتخابِ هوشمندانهٔ ویژگی‌ها (Feature Selection)، سنگ بنای توسعهٔ مدل‌های پیش‌بینی پایدار است. "
    r"با حذف متغیرهای نویزدار و گمراه‌کننده (نظیر <i>ساعت ثبت‌نام</i> یا <i>کد مرورگر</i> که F-Score آن‌ها تقریباً صفر بود)، <b>حجم فضای فرضیاتِ مدل (Hypothesis Space) کاهش یافته</b>، "
    r"خطر بیش‌برازش (Overfitting) خنثی شده و سرعت آموزشِ الگوریتم به طور چشمگیری بهبود می‌یابد.<br>"
    r"این امر دقت تصمیم‌گیری مدیران تجاری را در مهار ریزش مشتریان به حداکثر می‌رساند، چرا که ماشین‌لرنینگ اکنون منحصراً بر روی سیگنال‌های واقعی (مثل <i>تعداد تیکت‌های پشتیبانی</i>) متمرکز است."
)
display_academic_text(insight_conclusion, "insight")
```
