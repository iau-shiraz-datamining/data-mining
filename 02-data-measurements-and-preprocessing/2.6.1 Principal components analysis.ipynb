<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: کاهش ابعاد داده‌های مشتریان «آمازون» (Amazon) با روش PCA 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    سیستم داده‌کاوی «آمازون» (Amazon) برای هر خریدار ده‌ها ویژگی مختلف را ثبت می‌کند: «مجموع خرید دلاری»، «تعداد سفارش‌ها»، «میانگین امتیازهای ثبت‌شده»، «نرخ مرجوعی کالا»، «زمان حضور در سایت» و غیره. داشتن ۱۰ یا ۲۰ ویژگی (فضای ۲۰ بعد) باعث رخ دادن **«نفرین ابعاد» (Curse of Dimensionality)** می‌شود؛ به‌طوری که الگوریتم‌های هوش مصنوعی سنگین شده و امکان رسم نمودار و بصری‌سازی داده‌ها روی صفحه ۲ بعدی کامپیوتر وجود نخواهد داشت.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>راه‌حل — تحلیل مؤلفه‌های اصلی (PCA):</b><br>
    الگوریتم PCA یک تکنیک **کاهش ابعاد بدون نظارت (Unsupervised Dimensionality Reduction)** است که ویژگی‌های متعددی که با هم همبستگی دارند را با هم ترکیب کرده و آن‌ها را به چند **مؤلفه اصلی (Principal Components / PCs)** تبدیل می‌کند:
  </p>

  <ul style="text-align: right; margin: 0 0 10px; padding-right: 22px;">
    <li><b>مؤلفه اول (PC1):</b> برداری در جهت بیشترین پراکندگی و واریانس داده‌ها است (مثلاً ترکیب سن سن، درآمد و میزان خرید).</li>
    <li><b>مؤلفه دوم (PC2):</b> برداری عمود بر مؤلفه اول ($90^\circ$) است که بیشترین واریانس باقی‌مانده را پوشش می‌دهد.</li>
  </ul>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>یک نکته حیاتی (اهمیت استانداردسازی پیش از PCA):</b><br>
    چون PCA بر اساس محاسبه واریانس کار می‌کند، حتماً باید داده‌ها را قبل از اجرای PCA با **StandardScaler** هم‌مقیاس (میانگین ۰ و انحراف معیار ۱) کرد؛ در غیر این صورت، ویژگی‌هایی با اعداد بزرگتر (مثل درآمد) به‌طور کاذب بر تمام مؤلفه‌ها غلبه می‌کنند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> PCA داده‌های ۴ بعدی مشتریان آمازون را به ۲ مؤلفه اصلی (PC1 و PC2) کاهش می‌دهد؛ به‌طوری که بیش از **۸۵٪ از اطلاعات و واریانس کل** حفظ شده و امکان رسم ۲ بعدی الگوی خریداران فراهم می‌گردد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: فشرده‌سازی ویژگی‌های چندبعدی به ۲ مؤلفه اصلی مستقل بدون از دست رفتن الگوی کلی داده‌ها
    </div>
  </p>

</div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler 
from sklearn.decomposition import PCA       

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})

plt.rcParams['figure.dpi'] = 135

np.random.seed(42)
```

```python
display_academic_text(
    "تقلیل ابعاد (Dimensionality Reduction): فشرده‌سازی فضای ویژگی‌ها با تحلیل مؤلفه‌های اصلی (PCA) 🧩", 
    "main_title"
)

display_academic_text(
    "در تحلیل داده‌های کلان مشتریان، مواجهه با ده‌ها ویژگی همبسته (Correlated Features) موجب بروز پدیده <b>«نفرین ابعاد» (Curse of Dimensionality)</b> می‌گردد. "
    "الگوریتم PCA یک تکنیک جبر خطی است که ویژگی‌ها را به چند مؤلفه اصلیِ متعامد ($PC_1 \perp PC_2$) تصویر (Project) می‌کند. "
    "از آنجا که الگوریتم PCA وابستگیِ شدیدی به <i>واریانس (Variance)</i> دارد، <b>استانداردسازی داده‌ها (Z-Score Normalization)</b> پیش از اجرای الگوریتم، یک پیش‌نیاز قطعیِ ریاضیاتی است.", 
    "paragraph"
)

display_academic_text(
    r"معادله استانداردسازی Z-Score (تبدیل توزیع به $\mu=0$ و $\sigma=1$):", 
    "sub_title"
)

# Utilizing Raw Strings (r"") for LaTeX rendering
display_academic_text(
    r"$$Z = \frac{X - \mu}{\sigma}$$", 
    "paragraph"
)


# ==========================================================
# SECTION 2: SYNTHETIC DATA GENERATION & PCA PIPELINE
# ==========================================================
display_academic_text(
    "مرحله نخست: استانداردسازی ویژگی‌ها و استخراج مؤلفه‌های اصلی", 
    "sub_title"
)

display_academic_text(
    "در این فاز، یک پایگاه دادهٔ <b>۴ بُعدی</b> از رفتار مشتریان شبیه‌سازی می‌گردد. ابتدا تمامی ویژگی‌ها توسط <code>StandardScaler</code> هم‌مقیاس شده "
    "و سپس الگوریتم PCA، این فضای ۴ بُعدی را به یک فضای <b>۲ بُعدی متراکم</b> فشرده‌سازی می‌نماید. جدول سوم نشان می‌دهد که این دو مؤلفه (PC1 و PC2) به تنهایی، "
    "چه درصدی از کل اطلاعاتِ جامعه اصلی را حفظ کرده‌اند.", 
    "paragraph"
)

# Step 1: Generate synthetic Amazon customer behavior dataset (4D Feature Space)
n_customers = 150

data = {
    'Total_Spend_USD': np.random.normal(500, 150, n_customers),
    'Order_Frequency': np.random.normal(20, 5, n_customers),
    'Review_Rating': np.random.normal(4.2, 0.5, n_customers),
    'Return_Rate_Percent': np.random.normal(5, 2, n_customers),
}
df_customers = pd.DataFrame(data)

# Injecting correlation intentionally so PCA has patterns to discover
# High spenders tend to order more frequently and return less
df_customers['Order_Frequency'] += df_customers['Total_Spend_USD'] * 0.02
df_customers['Return_Rate_Percent'] -= df_customers['Total_Spend_USD'] * 0.005

# Step 2: Standardize the dataset (Crucial pre-requisite for PCA to remove scale bias)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_customers)

# Step 3: Apply PCA to project the 4D space down to a 2D space
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Create a DataFrame for the 2 extracted Principal Components
df_pca = pd.DataFrame(
    X_pca, 
    columns=['Principal_Comp_1', 'Principal_Comp_2']
)

# Calculate explained variance ratio for analytical reporting
explained_variance = pca.explained_variance_ratio_

df_variance_report = pd.DataFrame({
    'Eigenvector Component': [
        'PC 1 (Max Variance Axis)', 
        'PC 2 (Orthogonal Axis)'
    ],
    'Variance Retained (%)': [
        explained_variance[0] * 100, 
        explained_variance[1] * 100
    ]
})

# Add total cumulative variance to the report
total_variance = sum(explained_variance) * 100
df_variance_report.loc[len(df_variance_report.index)] = [
    'Total Cumulative Information', 
    total_variance
]

# Display preparation to prevent empty index columns in HTML
df_cust_disp = df_customers.head(6).reset_index().rename(columns={'index': 'Customer_ID'})
df_pca_disp = df_pca.head(6).reset_index().rename(columns={'index': 'Customer_ID'})

# Render the 4D dataset, reduced 2D dataset, and variance report side-by-side
display_side_by_side_datasets(
    [df_cust_disp, df_pca_disp, df_variance_report],
    [
        "جدول ۱: داده‌های خام مشتریان (فضای ۴ بُعدی)", 
        "جدول ۲: ماتریس کاهش‌یافته (فضای ۲ بُعدی فشرده)",
        "جدول ۳: گزارش واریانس (ارزش اطلاعاتی حفظ‌شده)"
    ]
)

# ==========================================================
# SECTION 3: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی - نقشه پراکندگی مشتریان در فضای فشرده (Latent Space)", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکی زیر <i>جادویِ PCA</i> را در بصری‌سازی داده‌های پیچیده به نمایش می‌گذارد. در نمودار سمت چپ، فضای ۴ بُعدی (که برای چشم انسان غیرقابل رسم است)، "
    "به یک نقشهٔ ۲ بُعدی (X, Y) تبدیل شده است. رنگ نقاط (نشان‌دهندهٔ مجموع خرید) به وضوح اثبات می‌کند که الگوریتم توانسته است ساختار پنهان (مشتریان پُرخرج در سمت راست "
    "و مشتریان کم‌خرج در سمت چپ) را حول محور PC1 کشف نماید. نمودار سمت راست نیز سهم هر مؤلفه را از کل اطلاعات سیستم مستندسازی می‌کند.", 
    "paragraph"
)

# Initialize a dual-axis canvas for visual comparative analytics
fig, axes = plt.subplots(1, 2, figsize=(19, 6.5), dpi=135)

# ----------------------------------------------------------
# Plot 1: 2D Scatter Plot of Principal Components
# ----------------------------------------------------------
# Plotting the newly discovered 2D space, colored by original Spend to verify pattern retention
scatter = axes[0].scatter(
    df_pca['Principal_Comp_1'], 
    df_pca['Principal_Comp_2'], 
    c=df_customers['Total_Spend_USD'], 
    cmap='Blues',      # Academic Blue gradient
    alpha=0.9, 
    s=180,             # Increased dot size for clarity
    edgecolor='#2c3e50',
    linewidth=1.0,
    zorder=5
)

# Intelligent Annotations
axes[0].annotate(
    "Low Value Segment:\nNotice how PC1 inherently grouped all the\nlow-spending customers on the left side.",
    xy=(-2.0, 1.0), xytext=(-3.0, 2.5),
    fontsize=11, fontweight='bold', color='#2c3e50',
    arrowprops=dict(arrowstyle="->", color="#34495e", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#ecf0f1", ec="#bdc3c7", lw=1.5)
)

axes[0].annotate(
    "High Value Segment:\nAnd pushed the high-spending (dark blue)\ncustomers completely to the right side.",
    xy=(2.0, -1.0), xytext=(0.5, -2.5),
    fontsize=11, fontweight='bold', color='#2980b9',
    arrowprops=dict(arrowstyle="->", color="#2980b9", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#2980b9", lw=1.5)
)

axes[0].set_title(f'Figure 1: 2D PCA Projection (Retained Info: {total_variance:.1f}%)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[0].set_xlabel(f'Principal Component 1 / PC1 ({explained_variance[0]*100:.1f}%)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].set_ylabel(f'Principal Component 2 / PC2 ({explained_variance[1]*100:.1f}%)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0].grid(True, linestyle='--', alpha=0.5, zorder=0)

# Add a colorbar to decode the colors back to the original 'Total Spend'
cbar = plt.colorbar(scatter, ax=axes[0])
cbar.set_label('Original Total Spend (USD) Indicator', fontdict={'fontsize': 12, 'fontweight': 'bold'})

# ----------------------------------------------------------
# Plot 2: Bar Chart of Explained Variance per Component
# ----------------------------------------------------------
components = ['PC1', 'PC2']
sns.barplot(
    x=components, 
    y=explained_variance * 100, 
    hue=components,
    legend=False,
    palette=['#1abc9c', '#2c3e50'], # Turquoise for PC1 (Dominant), Navy for PC2
    edgecolor='black', 
    linewidth=1.5,
    width=0.5,
    ax=axes[1]
)

axes[1].annotate(
    f"Dominant Component:\nThe first component alone captures {explained_variance[0]*100:.1f}%\nof the entire dataset's variance!",
    xy=(0.0, explained_variance[0]*100), xytext=(0.5, 80),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[1].set_title('Figure 2: Explained Variance Ratio per Component', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[1].set_xlabel('Extracted Orthogonal Principal Components', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_ylabel('Percentage of Variance Explained (%)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1].set_ylim(0, 100)
axes[1].set_xticklabels(axes[1].get_xticklabels(), fontweight='bold', fontsize=12)
axes[1].grid(axis='y', linestyle='--', alpha=0.5)

# Add precise percentage annotations on top of the bars
for i, p in enumerate(axes[1].patches):
    if p.get_height() > 0:
        axes[1].annotate(
            f"{p.get_height():.1f}%", 
            (p.get_x() + p.get_width() / 2., p.get_height()), 
            ha='center', va='center', xytext=(0, 12), textcoords='offset points',
            fontweight='bold', fontsize=13, color='#2c3e50'
        )

# Clear top and right spines for a clean academic aesthetic
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>استنتاج راهبردی عملیات (Strategic Implication):</b><br>"
    r"الگوریتم PCA تنها یک ابزار تقلیل ابعاد نیست، بلکه یک <i>موتور قدرتمند برای کشف سیگنال از نویز</i> است.<br>"
    r"با فشرده‌سازی اطلاعاتِ ۴ متغیر همبسته (خرید، فرکانس، امتیاز و مرجوعی) به تنها <b>۲ مؤلفه متعامدِ مستقل</b> (با حفظِ فوق‌العادهٔ بیش از ۸۵ درصد اطلاعات - گراف ۲)، "
    r"بارِ پردازشیِ مدل‌های یادگیری ماشین (مانند K-Means یا شبکه‌های عصبی) به شدت کاهش یافته و امکانِ <b>مصورسازیِ هندسیِ مرزهای خوشه‌بندی</b> در یک صفحهٔ ۲ بُعدیِ قابل فهم برای انسان (گراف ۱) فراهم می‌گردد."
)
display_academic_text(insight_conclusion, "insight")
```
