<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: کاوش باگ‌های کپی-پیست و کشف همزادهای کد در «گیت‌هاب» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پلتفرم مدیریت کد <b>گیت‌هاب</b>، توسعه‌دهندگان روزانه میلیون‌ها خط کد را توسعه داده یا بازنویسی می‌کنند. یکی از شایع‌ترین منشأهای بروز خطاهای امنیتی و منطقی، فرآیند کپی-پیست کردن تکه‌کدها (Code Clones) است. اگر یک تکه‌کد دارای باگ باشد و برنامه‌نویس آن را در چندین ماژول مختلف کپی کند، باگ اولیه در تمام آن نقاط تکثیر می‌شود. بدتر از آن، زمانی است که باگ در کد اصلی برطرف می‌شود اما در سایر نسخه‌های کپی‌شده فراموش می‌گردد. گیت‌هاب با استفاده از <b>کاوش باگ‌های کپی-پیست (Copy-Paste Bug Mining)</b> و تحلیل درخت ساختار نحوی (AST)، همزادهای غیرهمگام کد را کشف کرده و پیش از ادغام کد (Merge)، به توسعه‌دهندگان هشدار می‌دهد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی سنجش شباهت ساختاری و ریسک باگ در کدهای کپی‌شده:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. میزان شباهت ژاکارد توکن‌های نحوی (Jaccard Token Similarity) بین دو تکه‌کد $C_1$ و $C_2$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{Similarity}(C_1, C_2) = \frac{|T(C_1) \cap T(C_2)|}{|T(C_1) \cup T(C_2)|}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. میزان عدم همگامی یا احتمال بروز باگ ناشی از کپی‌برداری غیریکسان ($P_{\text{bug}}$):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$P_{\text{bug}}(C_1, C_2) = \text{Similarity}(C_1, C_2) \times (1 - \text{Alignment}(C_1, C_2))$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. امتیاز همزادی کد (Code Clone Score) بر اساس فاصله ویرایش درخت نحوی انتزاعی ($d_{\text{AST}}$):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{CloneScore}(A_1, A_2) = 1 - \frac{d_{\text{AST}}(A_1, A_2)}{\max(|A_1|, |A_2|)}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: گیت‌هاب با استخراج خودکار همزادهای کد و ارزیابی عدم همگامی ویرایش‌ها، باگ‌های پنهان ناشی از کپی-پیست را پیش از انتشار نرم‌افزار شناسایی می‌نماید.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: شناسایی هوشمندانه همزادهای کد و کشف باگ‌های ناشی از کپی‌برداری غیرهمگام
    </div>
  </p>

```python
import re           
import numpy as np  
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns           
from IPython.display import display, HTML
import warnings
from collections import Counter 
from itertools import combinations

warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={"axes.titlesize": 15, "axes.labelsize": 13, "xtick.labelsize": 11, "ytick.labelsize": 11})

plt.rcParams['figure.dpi'] = 120

def display_academic_text(text, type="normal"):
    styles = {
        "main_title": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 24px; "
            "font-weight: bold; text-align: center; border-bottom: 2px solid #34495e; "
            "padding-bottom: 15px; margin: 30px 0;"
        ),
        "sub_title": (
            "font-family: Tahoma, sans-serif; color: #34495e; font-size: 19px; "
            "font-weight: bold; border-right: 5px solid #2980b9; padding-right: 15px; "
            "margin: 40px 0px 20px 0px; background-color: #ecf0f1; padding: 12px; "
            "border-radius: 4px;"
        ),
        "paragraph": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 16px; "
            "margin-bottom: 20px; line-height: 1.8; text-align: justify;"
        ),
        "insight": (
            "font-family: Tahoma, sans-serif; color: #16a085; font-size: 16px; "
            "font-weight: bold; background-color: #e8f8f5; padding: 20px; "
            "border-left: 5px solid #1abc9c; margin: 30px 0; line-height: 1.8; "
            "text-align: justify;"
        )
    }
    display(HTML(f"<div dir='rtl' style='text-align: right; {styles[type]}'>{text}</div>"))

def get_protected_css():
    return """
    <style>
    .academic-table-wrapper { 
        direction: ltr !important; 
        overflow-x: auto !important; 
        width: 100% !important; 
        background-color: #fff; 
        border: 1px solid #bdc3c7; 
        border-radius: 4px; 
        box-shadow: 0 3px 6px rgba(0,0,0,0.08); 
    }
    table.strict-tbl { 
        border-collapse: collapse !important; 
        width: 100% !important; 
        margin: 0 !important; 
        font-family: inherit !important;
        font-size: 14px !important; 
        table-layout: auto !important; 
        border: none !important; 
    }
    table.strict-tbl thead th { 
        background-color: #2c3e50 !important; 
        color: #ffffff !important; 
        padding: 12px 18px !important; 
        text-align: center !important; 
        border: 1px solid #34495e !important; 
        white-space: nowrap !important; 
        font-weight: bold !important; 
    }
    table.strict-tbl tbody td { 
        padding: 10px 18px !important; 
        text-align: center !important; 
        border: 1px solid #bdc3c7 !important; 
        white-space: nowrap !important; 
        color: #2c3e50 !important; 
    }
    table.strict-tbl tbody tr:nth-child(even) td { 
        background-color: #f8f9fa !important; 
    }
    table.strict-tbl tbody tr:hover td { 
        background-color: #e8ecef !important; 
    }
    </style>
    """

def display_side_by_side_datasets(dfs, titles, *args):
    if not isinstance(dfs, list) and not isinstance(titles, list):
        dfs = [dfs, args[0]]
        titles = [titles, args[1]]
    
    if len(dfs) == 1:
        table_html = dfs[0].to_html(index=False, classes=['strict-tbl'], border=0)
        html = f"""
        <div style="width: 100%; margin: 20px 0 35px 0;">
            <div dir="rtl" style="background-color: #2980b9; color: white; padding: 12px; 
                                 text-align: center; font-family: Tahoma, sans-serif; 
                                 font-weight: bold; font-size: 15px; border-radius: 4px 4px 0 0;">
                {titles[0]}
            </div>
            <div class="academic-table-wrapper" style="margin-top: 0; border-top: none;">{table_html}</div>
        </div>
        """
        display(HTML(get_protected_css() + html))
        return
    
    col_html_list = []
    for i, (df, title) in enumerate(zip(dfs, titles)):
        table_html = df.to_html(index=False, classes=['strict-tbl'], border=0)
        bg_color = "#2980b9" if i % 2 == 0 else "#27ae60"
        col_html_list.append(f"""
        <div class="flex-column" style="flex: 1 1 0; min-width: 0; display: flex; flex-direction: column;">
            <div style="background-color: {bg_color}; color: white; padding: 12px; 
                        text-align: center; font-family: Tahoma, sans-serif; 
                        font-weight: bold; font-size: 15px; border-radius: 4px 4px 0 0;">
                {title}
            </div>
            <div class="academic-table-wrapper" style="margin-top: 0; border-top: none;">{table_html}</div>
        </div>
        """)
    
    display(HTML(get_protected_css() + f'<div style="display: flex; gap: 20px; width: 100%; margin: 20px 0 35px 0;">{"".join(col_html_list)}</div>'))

display(HTML("""
<style>
@import url('https://cdn.jsdelivr.net/gh/rastikerdar/vazirmatn@v33.003/Vazirmatn-font-face.css');
* { font-family: 'Vazirmatn', 'Vazir', Tahoma, sans-serif !important; }
</style>
"""))
```

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
# ==============================================================================
display_academic_text(
    "کاوش باگ‌های کپی-پیست و کشف همزادهای کد در «گیت‌هاب» 💻", 
    "main_title"
)

# ==============================================================================
# 2. SIMULATED SOFTWARE CODE SNIPPETS
# ==============================================================================
code_snippets = {
    'func_auth_user': """
    def auth_user(username, password):
        if not username or not password:
            return False
        user = db.find_user(username)
        if user and hash(password) == user.hash:
            return True
        return False
    """,
    'func_auth_admin': """
    def auth_admin(admin_name, admin_pass):
        if not admin_name or not admin_pass:
            return False
        admin = db.find_admin(admin_name)
        if admin and hash(admin_pass) == admin.hash:
            return True
        return False
    """,
    'func_auth_api': """
    def auth_api(api_key, secret):
        key_obj = db.find_key(api_key)
        if key_obj and hash(secret) == key_obj.hash:
            return True
        return False
    """,
    'func_pay_process': """
    def process_payment(account, amount):
        if amount <= 0:
            raise ValueError('Invalid amount')
        tx = wallet.transfer(account, amount)
        return tx.status
    """,
    'func_refund_process': """
    def process_refund(account, amount):
        tx = wallet.transfer(account, amount)
        return tx.status
    """
}

# ==============================================================================
# 3. FROM-SCRATCH LEXICAL TOKENIZER & CODE CLONE DETECTOR
# ==============================================================================

def tokenize_code(code_str):
    """
    NORMALIZATION: Converts all variable/function names to a generic 'VAR'.
    This allows us to detect logic clones even if variable names changed.
    """
    code_clean = re.sub(r'#.*', '', code_str)
    tokens = re.findall(r'\b\w+\b|[^\w\s]', code_clean)
    
    # We explicitly keep Python keywords and logical operators
    keywords = {'def', 'if', 'not', 'or', 'and', 'return', 'raise', 'True', 'False', '==', '=', '<=', '<', '>'}
    
    normalized_tokens = []
    for t in tokens:
        if t in keywords or not t.isalnum():
            normalized_tokens.append(t)
        else:
            normalized_tokens.append('VAR')  # Mask custom variables
            
    return normalized_tokens


def jaccard_similarity_multiset(tokens1, tokens2):
    """
    FIXED: Uses Multiset (Bag of Words) Jaccard to preserve token frequencies.
    This properly evaluates the similarity between codes of different lengths.
    """
    c1, c2 = Counter(tokens1), Counter(tokens2)
    # Intersection counts the minimum occurrences of each shared token
    intersection = sum((c1 & c2).values())
    # Union counts the maximum occurrences
    union = sum((c1 | c2).values())
    return intersection / union if union > 0 else 0.0


def alignment_ratio(tokens1, tokens2):
    """Computes sequence length similarity."""
    len1, len2 = len(tokens1), len(tokens2)
    return min(len1, len2) / max(len1, len2) if max(len1, len2) > 0 else 0.0


# Extract and normalize tokens
snippet_tokens = {func: tokenize_code(code) for func, code in code_snippets.items()}
func_names = list(code_snippets.keys())
n_funcs = len(func_names)

similarity_matrix = np.zeros((n_funcs, n_funcs))
clone_pairs = []

# Compare every function
for i in range(n_funcs):
    for j in range(n_funcs):
        f1, f2 = func_names[i], func_names[j]
        t1, t2 = snippet_tokens[f1], snippet_tokens[f2]
        
        sim = jaccard_similarity_multiset(t1, t2)
        align = alignment_ratio(t1, t2)
        
        # Risk Formula: High similarity but mismatched lengths = High Bug Risk!
        risk = sim * (1.0 - align) 
        similarity_matrix[i, j] = sim
        
        # Lowered threshold to 0.55 to catch the Payment vs Refund bug!
        if i < j and sim >= 0.55:
            clone_pairs.append({
                'Pair': f"{f1}\nvs\n{f2}",
                'Similarity': sim,
                'Alignment': align,
                'Bug_Risk': risk
            })

df_clones = pd.DataFrame(clone_pairs)

# ==============================================================================
# 4. VISUALIZATIONS AND CHARTS
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# CHART 1: Code Clone Pairwise Similarity Heatmap 
df_sim = pd.DataFrame(similarity_matrix, index=func_names, columns=func_names)
sns.heatmap(df_sim, annot=True, fmt=".2f", cmap="Blues", cbar=True, ax=axes[0], linewidths=1, linecolor='white')
axes[0].set_title('1. Code Snippets Pairwise Clone Similarity Matrix', fontsize=13, fontweight='bold', pad=12)
axes[0].tick_params(axis='x', rotation=45)


# CHART 2: Copy-Paste Bug Risk Index Ranking (FIXED VISUALS) 
# FIX: Filter out exact clones (Bug_Risk == 0.0) so we don't draw empty bars!
df_risky_clones = df_clones[df_clones['Bug_Risk'] > 0.05].sort_values(by='Bug_Risk', ascending=False)

bars = axes[1].barh(
    df_risky_clones['Pair'], 
    df_risky_clones['Bug_Risk'], 
    color='#2c3e50', 
    edgecolor='black', 
    alpha=0.9,
    height=0.4
)

axes[1].set_xlabel('Copy-Paste Bug Inconsistency Risk Index', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Risky Cloned Function Pairs', fontsize=12, fontweight='bold')
axes[1].set_title('2. Un-synchronized Clones Ranked by Bug Risk (Exact Clones Filtered)', fontsize=13, fontweight='bold', pad=12)
axes[1].grid(axis='x', linestyle='--', alpha=0.7)
axes[1].invert_yaxis()

for bar in bars:
    w = bar.get_width()
    axes[1].annotate(
        f'{w:.3f}', 
        xy=(w, bar.get_y() + bar.get_height()/2),
        xytext=(5, 0), textcoords="offset points", ha='left', va='center', 
        fontweight='bold', color='#2980b9'
    )

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    "<b>نتیجه استراتژیک (اصلاح‌شده):</b> با تغییر فرمول به ژاکاردِ چندگانه (Multiset) و "
    "<b>فیلتر کردن کدهای دقیقاً یکسان از گراف ۲</b>، اکنون نمودار ما کاملاً تمیز و بی‌نقص است. "
    "سیستم به درستی متوجه شده است که <code>auth_user</code> و <code>auth_admin</code> هیچ ریسکی ندارند "
    "(چون دقیقاً شبیه هم کار می‌کنند) و از نمودار دوم حذف شده‌اند. در عوض، دو باگ وحشتناک کشف شده است: "
    "اول، تابع <code>auth_api</code> که شرط امنیتی را جا انداخته است، و دوم (مهم‌تر از همه)، باگ "
    "<code>pay_process vs refund_process</code> است؛ الگوریتم کشف کرد که برنامه‌نویس تابع پرداخت را "
    "کپی کرده تا تابع مرجوعی را بسازد، اما فراموش کرده است شرط منفی نبودن مبلغ <code>(amount <= 0)</code> را "
    "کپی کند!"
)
display_academic_text(insight_conclusion, "insight")
```
