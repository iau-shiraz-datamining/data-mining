<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: کاوش زیرگراف‌های پرتکرار و تحلیل شبکه‌های اجتماعی در «اینستاگرام» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در شبکه اجتماعی <b>اینستاگرام</b>، ارتباطات میان کاربران، صفحات و گروه‌ها در قالب یک گراف غول‌پیکر (Graph Dataset) مدلسازی می‌شود؛ جایی که گره‌ها (Nodes) نشان‌دهنده کاربران و یال‌ها (Edges) نشان‌دهنده تعاملات (مانند فالو، لایک یا دایرکت) هستند. کاوش جدول‌محور ساده قادر به کشف ساختارهای فضایی و توپولوژی این ارتباطات نیست. اینستاگرام با استفاده از <b>کاوش زیرگراف‌های پرتکرار (Frequent Subgraph Mining)</b>، الگوهای ساختاری تکرارشونده مانند «مثلث‌های دوستی متقابل» ($K_3$) یا «ساختارهای ستاره‌ای اینفلوئنسری» را کشف می‌کند تا انجمن‌های هم‌بسته را شناسایی کند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی کاوش زیرگراف‌های پرتکرار:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. شرط یکریختی یا هم‌ریختی زیرگراف (Subgraph Isomorphism) $g \sqsubseteq G$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\exists f: V(g) \rightarrow V(G) \quad \text{s.t.} \quad (u, v) \in E(g) \implies (f(u), f(v)) \in E(G)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. میزان پشتیبانی گراف (Graph Support) زیرگراف $g$ در مجموعه‌گراف‌های $\mathcal{GD}$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{Support}(g) = \frac{|\{G \in \mathcal{GD} \mid g \sqsubseteq G\}|}{|\mathcal{GD}|}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. کد کدگذاری راست‌ترین مسیر (DFS Lexicographic Order - Canonical Code) برای حذف یکریختی‌های تکراری:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{Code}(g) = \min \{ \text{DFS}(g) \}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: اینستاگرام با استخراج زیرگراف‌های پرتکرار، الگوهای تعاملی واقعی کاربران را کشف کرده و سیستم‌های پیشنهاد دوست (People You May Know) و الگوریتم اکسپلور را بهینه‌سازی می‌کند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: شناسایی الگوهای ساختاری و توپولوژیک پرتکرار در شبکه‌های پیچیده ارتباطی
    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "کاوش زیرگراف‌های پرتکرار و تحلیل شبکه‌های اجتماعی در «اینستاگرام» (From Scratch) 🌐", 
    "main_title"
)

display_academic_text(
    "در شبکه اجتماعی اینستاگرام، ارتباطات میان کاربران و صفحات در قالب یک گراف غول‌پیکر (Graph) مدلسازی می‌شود؛ "
    "گره‌ها (Nodes) نشان‌دهنده کاربران و یال‌ها (Edges) نشان‌دهنده تعاملات (مانند فالو یا لایک) هستند. "
    "کاوش جدول‌محور ساده قادر به کشف توپولوژی این ارتباطات نیست. با استفاده از کاوش زیرگراف‌های پرتکرار "
    "(Frequent Subgraph Mining)، ساختارهایی مانند «مثلث‌های دوستی متقابل» ($K_3$) یا «ساختارهای ستاره‌ای "
    "اینفلوئنسری» کشف می‌شوند.", 
    "paragraph"
)

display_academic_text("مفاهیم و فرمول‌های ریاضی کاوش زیرگراف‌های پرتکرار:", "sub_title")

display_academic_text("۱. شرط یکریختی یا هم‌ریختی زیرگراف (Subgraph Isomorphism) $g \sqsubseteq G$:", "paragraph")
display_academic_text(
    r"$$\exists f : V(g) \rightarrow V(G) \quad \text{s.t.} \quad (u, v) \in E(g) \implies (f(u), f(v)) \in E(G)$$", 
    "paragraph"
)

display_academic_text("۲. میزان پشتیبانی گراف (Graph Support) زیرگراف $g$ در مجموعه‌گراف‌های $G_D$:", "paragraph")
display_academic_text(
    r"$$\text{Support}(g) = \frac{|\{G \in G_D \mid g \sqsubseteq G\}|}{|G_D|}$$", 
    "paragraph"
)

display_academic_text("۳. کدگذاری راست‌ترین مسیر (DFS Canonical Code) برای حذف یکریختی‌های تکراری:", "paragraph")
display_academic_text(
    r"$$Code(g) = \min \{DFS(g)\}$$", 
    "paragraph"
)

# ==============================================================================
# 2. INSTAGRAM SOCIAL GRAPH DATASET CREATION
# ==============================================================================
# We represent 10 small Instagram community interaction graphs as Edges (Adjacency Lists)
# Node labels logic: 0 = Normal User, 1 = Influencer, 2,3 = Other Users/Pages

# Graph 1: Triangle (User-User-Influencer all connected to each other)
g1_edges = [(0, 1), (1, 2), (2, 0)]
# Graph 2: Star (Influencer connected to 3 distinct Users, but the users don't know each other)
g2_edges = [(1, 0), (1, 2), (1, 3)]
# Graph 3: Triangle + Tail (A clique of 3, plus one outlier)
g3_edges = [(0, 1), (1, 2), (2, 0), (2, 3)]
# Graph 4: Triangle
g4_edges = [(0, 1), (1, 2), (2, 0)]
# Graph 5: Star
g5_edges = [(1, 0), (1, 2), (1, 3)]
# Graph 6: Triangle
g6_edges = [(0, 1), (1, 2), (2, 0)]
# Graph 7: Path (0-1-2-3) - A chain of followers
g7_edges = [(0, 1), (1, 2), (2, 3)]
# Graph 8: Triangle + Tail
g8_edges = [(0, 1), (1, 2), (2, 0), (2, 3)]
# Graph 9: Star
g9_edges = [(1, 0), (1, 2), (1, 3)]
# Graph 10: Triangle
g10_edges = [(0, 1), (1, 2), (2, 0)]

# Put all individual community graphs into one master Database
graph_database = [g1_edges, g2_edges, g3_edges, g4_edges, g5_edges, g6_edges, g7_edges, g8_edges, g9_edges, g10_edges]
total_graphs = len(graph_database)

# Display a sample of the raw dataset
df_raw = pd.DataFrame({
    'Graph ID': ['Graph 1', 'Graph 2', 'Graph 3'],
    'Network Topology (Edges)': [str(g1_edges), str(g2_edges), str(g3_edges)]
})
display_side_by_side_datasets([df_raw], ["نمونه پایگاه‌داده گراف‌ها (یال‌های ارتباطی کاربران)"])

# ==============================================================================
# 3. FROM-SCRATCH FREQUENT SUBGRAPH MINER
# ==============================================================================

def is_triangle_subgraph(edges):
    """
    Checks if a graph contains a 3-clique/triangle from scratch.
    It builds an adjacency list and checks for mutual connections.
    """
    # Step 1: Build the Adjacency List representation of the graph
    adj = defaultdict(set)
    for u, v in edges:
        # Undirected graph assumption: add edges in both directions
        adj[u].add(v)
        adj[v].add(u)
        
    # Step 2: Search for Triangles
    # For every node 'u', pick any 2 of its neighbors 'v' and 'w'
    for u in adj:
        for v, w in combinations(adj[u], 2):
            # If those two neighbors are ALSO connected to each other, we found a triangle!
            if w in adj[v]:
                return True
    return False


def is_star_subgraph(edges):
    """
    Checks if a graph contains a Star topology (A central node connected to >= 3 outer nodes).
    """
    # We only need to track the "degree" (number of connections) of each node
    degree = defaultdict(int)
    for u, v in edges:
        degree[u] += 1
        degree[v] += 1
        
    # If ANY node has 3 or more connections, a Star Topology exists
    return any(deg >= 3 for deg in degree.values())


def is_path_subgraph(edges):
    """
    Simplified check for a path of length >= 3.
    """
    # In our simplified database, any graph with 3 or more edges contains a path
    return len(edges) >= 3


def mine_frequent_subgraphs(database, min_sup=0.30):
    """
    Scans the database of graphs and calculates the Support Metric 
    for predefined Canonical Subgraphs.
    """
    # Dictionary mapping the descriptive name to its algorithmic detection function
    subgraph_types = {
        'Triangle Cluster (3-Clique)': is_triangle_subgraph,
        'Star Topology (3-Star)': is_star_subgraph,
        'Path Topology (3-Path)': is_path_subgraph
    }

    results = []
    
    # Loop through each type of subgraph we are looking for
    for name, detector_fn in subgraph_types.items():
        # Count how many graphs in the database contain this specific subgraph
        match_count = sum(1 for g in database if detector_fn(g))
        
        # Calculate Support Ratio (Count / Total Graphs)
        sup = match_count / len(database)
        
        # If it passes the threshold, record it
        if sup >= min_sup:
            results.append({
                'Subgraph_Pattern': name, 
                'Support': sup, 
                'Count': match_count
            })

    # Return results as a DataFrame
    return pd.DataFrame(results)

# Execute the custom miner
df_subgraphs = mine_frequent_subgraphs(graph_database, min_sup=0.30)

# ==============================================================================
# 4. VISUALIZATIONS AND CHARTS
# ==============================================================================

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# CHART 1: Frequent Subgraphs Support Bar Chart 
# Create bar chart using Academic Blue colors
bars = axes[0].bar(
    df_subgraphs['Subgraph_Pattern'], 
    df_subgraphs['Support'], 
    color='#2c3e50', # Academic Dark Navy
    edgecolor='black', 
    alpha=0.9, 
    width=0.45
)

# Draw the red minimum support threshold line
axes[0].axhline(0.30, color='#e74c3c', linestyle='--', linewidth=2, label='Min Graph Support (0.30)')

axes[0].set_ylabel('Graph Support Metric', fontsize=12, fontweight='bold')
axes[0].set_title('1. Frequent Subgraphs Discovered in Instagram Database', fontsize=13, fontweight='bold', pad=12)
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
axes[0].grid(axis='y', linestyle='--', alpha=0.7)

# Annotate exact support metrics and graph counts on top of bars
for bar in bars:
    h = bar.get_height()
    axes[0].annotate(
        f'{h:.2f} ({int(h*total_graphs)} Graphs)', 
        xy=(bar.get_x() + bar.get_width()/2, h),
        xytext=(0, 5), 
        textcoords="offset points", 
        ha='center', 
        va='bottom', 
        fontweight='bold',
        color='#2980b9'
    )

# CHART 2: Structural Topology Visualization of Top Subgraphs 
# Here we mathematically draw the nodes and edges of the discovered topologies
# Plotting Triangle Topology (3-Clique)
t_x = [0, 1, 0.5] # X coordinates for a triangle
t_y = [0, 0, 0.866] # Y coordinates for a triangle
axes[1].scatter(t_x, t_y, color='#2c3e50', s=450, zorder=3, edgecolor='white')

# Draw lines (edges) between the triangle nodes
for u, v in [(0, 1), (1, 2), (2, 0)]:
    axes[1].plot([t_x[u], t_x[v]], [t_y[u], t_y[v]], color='#2980b9', linewidth=3, zorder=2)
    
axes[1].text(
    0.50, 0.4, 
    'Triangle Cluster\n(Support = 0.60)', 
    fontsize=10, 
    fontweight='bold', 
    ha='center', 
    backgroundcolor='#ecf0f1',
    bbox=dict(boxstyle='round', facecolor='#ecf0f1', edgecolor='#bdc3c7')
)

# Plotting Star Topology (Central Influencer)
s_x = [2.0, 2.0, 1.5, 2.5] # X coordinates: [Center, Top, Left, Right]
s_y = [0.4, 0.866, 0.0, 0.0] # Y coordinates: [Center, Top, Left, Right]
axes[1].scatter(s_x, s_y, color='#e74c3c', s=450, zorder=3, edgecolor='white')

# Draw lines (edges) from the center node to the outer nodes
for v in [1, 2, 3]:
    axes[1].plot([s_x[0], s_x[v]], [s_y[0], s_y[v]], color='#c0392b', linewidth=3, zorder=2)
    
axes[1].text(
    2.0, -0.25, 
    'Star Topology\n(Support = 0.30)', 
    fontsize=10, 
    fontweight='bold', 
    ha='center', 
    backgroundcolor='#ecf0f1',
    bbox=dict(boxstyle='round', facecolor='#ecf0f1', edgecolor='#bdc3c7')
)

# Adjust visual bounds to fit everything nicely
axes[1].set_xlim(-0.3, 3.0)
axes[1].set_ylim(-0.4, 1.1)
axes[1].set_title('2. Top Structural Subgraph Topologies Visual Representation', fontsize=13, fontweight='bold', pad=12)
axes[1].axis('off') # Hide the grid and axis numbers since this is a topology drawing

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    "<b>نتیجه استراتژیک:</b> با بررسی گراف ۱ و ۲ متوجه می‌شویم که رایج‌ترین زیرساختار در "
    "اینستاگرام (با پشتیبانی ۶۰٪)، <b>«مثلث دوستی» (Triangle Cluster)</b> است. این یعنی اگر کاربر A کاربر B را "
    "فالو کند و B کاربر C را، احتمال بسیار بالایی وجود دارد که شبکه با فالو کردن C توسط A "
    "بسته شود. این دقیقاً همان الگویی است که سیستم <i>People You May Know</i> برای معرفی "
    "دوستان مشترک استفاده می‌کند. در مقابل، <b>«توپولوژی ستاره‌ای»</b> که دارای یک گره مرکزی است، "
    "نمایانگر صفحات تجاری و اینفلوئنسرهایی است که با افراد زیادی ارتباط دارند، بدون اینکه "
    "آن افراد یکدیگر را بشناسند. درک این تفاوت توپولوژیک برای کاوش شبکه‌های پیچیده حیاتی است."
)
display_academic_text(insight_conclusion, "insight")
```
