<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: ارزیابی و انتخاب ویژگی‌های کلیدی درخواست سفر در «اوبر» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پلتفرم تاکسی آنلاین <b>اوبر</b>، هنگامی که یک درخواست سفر برای سفیران (رانندگان) ارسال می‌شود، سیستم باید پیش‌بینی کند که آیا راننده درخواست را می‌پذیرد یا خیر. الگوریتم‌های هوش مصنوعی اوبر با ویژگی‌های متعددی سرکار دارند (مانند ضریب افزایش قیمت یا Surge Multiplier، مسافت تا مسافر، زمان تخمینی رسیدن یا ETA، و امتیاز مسافر). برای ساخت مدل‌های سریع و دقیق، باید با استفاده از <b>معیارهای انتخاب ویژگی (Attribute Selection Measures)</b>، مهم‌ترین و تعیین‌کننده‌ترین متغیرها را شناسایی کرده و ویژگی‌های کم‌اثر را فیلتر نمود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی معیارهای اصلی انتخاب ویژگی:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. آنتروپی (Entropy) برای سنجش میزان بی‌نظمی در کلاس هدف $S$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{Entropy}(S) = -\sum_{i=1}^{c} p_i \log_2(p_i)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. بهره اطلاعاتی (Information Gain) برای ویژگی $A$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{Gain}(S, A) = \text{Entropy}(S) - \sum_{v \in \text{Values}(A)} \frac{|S_v|}{|S|} \text{Entropy}(S_v)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. نسبت بهره (Gain Ratio) جهت اصلاح تورش ویژگی‌های با مقادیر متعدد:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{GainRatio}(S, A) = \frac{\text{Gain}(S, A)}{\text{SplitInfo}_A(S)}, \quad \text{SplitInfo}_A(S) = -\sum_{v \in \text{Values}(A)} \frac{|S_v|}{|S|} \log_2\left(\frac{|S_v|}{|S|}\right)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. شاخص ناخالصی جینی (Gini Index):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{Gini}(S) = 1 - \sum_{i=1}^{c} p_i^2$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: اوبر با رتبه‌بندی ویژگی‌ها بر اساس نسبت بهره و شاخص جینی، متغیرهای بی‌پایداری مانند «امتیاز مسافر» را حذف کرده و مدل را بر روی «ضریب قیمت» و «زمان رسیدن» متمرکز می‌سازد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: جداسازی ویژگی‌های باقدرت تفکیک بالا و حذف متغیرهای نویزی جهت بهبود دقت و سرعت مدل
    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "ارزیابی و انتخاب ویژگی‌های کلیدی درخواست سفر در «اوبر» (From Scratch) 🚖", 
    "main_title"
)

display_academic_text(
    "در پلتفرم تاکسی آنلاین اوبر، هنگامی که یک درخواست سفر ارسال می‌شود، الگوریتم‌های هوش مصنوعی "
    "با ویژگی‌های متعددی سرکار دارند (مانند ضریب افزایش قیمت یا Surge، زمان رسیدن یا ETA، و امتیاز مسافر). "
    "برای ساخت مدل‌های سریع، باید مهم‌ترین متغیرها را شناسایی کرد. اما برخی معیارها مانند 'Information Gain' "
    "گول‌زننده هستند! در این تمرین از پایه کدنویسی می‌کنیم تا نشان دهیم چگونه محاسبه 'Gain Ratio' تورش "
    "الگوریتم‌ها را برطرف کرده و ویژگی‌های واقعاً ارزشمند را پیدا می‌کند.", 
    "paragraph"
)

display_academic_text("مفاهیم و فرمول‌های ریاضی معیارهای اصلی انتخاب ویژگی:", "sub_title")

display_academic_text("۱. آنتروپی (Entropy) برای سنجش میزان بی‌نظمی در کلاس هدف $S$:", "paragraph")
display_academic_text(r"$$Entropy(S) = - \sum_{i=1}^{c} p_i \log_2(p_i)$$", "paragraph")

display_academic_text("۲. بهره اطلاعاتی (Information Gain) برای ویژگی $A$:", "paragraph")
display_academic_text(r"$$Gain(S, A) = Entropy(S) - \sum_{v \in Values(A)} \frac{|S_v|}{|S|} Entropy(S_v)$$", "paragraph")

display_academic_text("۳. نسبت بهره (Gain Ratio) جهت اصلاح تورش ویژگی‌های دارای مقادیر متنوع:", "paragraph")
display_academic_text(
    r"$$GainRatio(S, A) = \frac{Gain(S, A)}{SplitInfo_A(S)}, \quad SplitInfo_A(S) = - \sum_{v \in Values(A)} \frac{|S_v|}{|S|} \log_2\left(\frac{|S_v|}{|S|}\right)$$", 
    "paragraph"
)

# ==============================================================================
# 2. UBER TRIP DISPATCH DATASET GENERATION
# ==============================================================================
np.random.seed(42)
n_samples = 180

# Simulate Driver Dispatch Features
surge_multiplier = np.random.choice([1.0, 1.2, 1.5, 2.0], size=n_samples, p=[0.4, 0.3, 0.2, 0.1])
eta_minutes = np.random.randint(1, 20, size=n_samples)
trip_distance = np.random.uniform(1.0, 25.0, size=n_samples)
# Rider rating has many unique categories, which usually fools simple Information Gain
rider_rating = np.random.choice([3.5, 3.8, 4.0, 4.2, 4.5, 4.8, 5.0], size=n_samples)

target = []
# Acceptance logic: Drivers heavily care about Surge Pricing and ETA. Rating rarely matters.
for surge, eta, dist, rating in zip(surge_multiplier, eta_minutes, trip_distance, rider_rating):
    # Base score determined mathematically
    score = (surge * 2.0) - (eta * 0.15) + np.random.normal(0, 0.3)
    target.append(1 if score > 1.8 else 0)

df_uber = pd.DataFrame({
    'Surge_Multiplier': surge_multiplier,
    'ETA_Minutes': eta_minutes,
    'Trip_Distance': trip_distance,
    'Rider_Rating': rider_rating,
    'Accepted': target
})

# OPTIMIZATION: Discretize continuous features safely using duplicates='drop'
df_uber['ETA_Bin'] = pd.qcut(df_uber['ETA_Minutes'], q=3, labels=['Fast', 'Medium', 'Slow'], duplicates='drop')
df_uber['Distance_Bin'] = pd.qcut(df_uber['Trip_Distance'], q=3, labels=['Short', 'Medium', 'Long'], duplicates='drop')

display_side_by_side_datasets(
    [df_uber[['Surge_Multiplier', 'ETA_Bin', 'Rider_Rating', 'Accepted']].head(6)],
    ["نمونه داده‌های اعزام سفر اوبر (ویژگی‌ها و تصمیم راننده)"]
)

# ==============================================================================
# 3. FROM-SCRATCH ATTRIBUTE SELECTION MEASURES (IG, GAIN RATIO, GINI)
# ==============================================================================

def calculate_entropy(y):
    """
    Calculates Shannon Entropy from scratch.
    """
    classes, counts = np.unique(y, return_counts=True)
    probs = counts / len(y)
    return -np.sum([p * np.log2(p) for p in probs if p > 0])

def calculate_gini(y):
    """
    Calculates Gini Impurity from scratch (Alternative to Entropy used in CART).
    """
    classes, counts = np.unique(y, return_counts=True)
    probs = counts / len(y)
    return 1.0 - np.sum(probs ** 2)

def calculate_feature_metrics(df, feature_col, target_col):
    """
    Calculates all three major Feature Selection Metrics mathematically from scratch.
    """
    y = df[target_col].values
    parent_entropy = calculate_entropy(y)
    parent_gini = calculate_gini(y)
    n_total = len(y)

    values = df[feature_col].unique()
    weighted_child_entropy = 0.0
    weighted_child_gini = 0.0
    split_info = 0.0 # Used to penalize features with too many unique categories

    # Loop through each unique category in the current feature
    for v in values:
        # Get subset of data that matches this category
        sub_y = df[df[feature_col] == v][target_col].values
        n_sub = len(sub_y)
        p_sub = n_sub / n_total

        # Accumulate weighted purity metrics
        weighted_child_entropy += p_sub * calculate_entropy(sub_y)
        weighted_child_gini += p_sub * calculate_gini(sub_y)
        
        # Calculate SplitInfo (Entropy of the feature itself)
        if p_sub > 0:
            split_info -= p_sub * np.log2(p_sub)

    # 1. Information Gain = Parent Purity - Child Purity
    info_gain = parent_entropy - weighted_child_entropy
    
    # 2. Gain Ratio = Information Gain / SplitInfo (To fix bias)
    gain_ratio = info_gain / split_info if split_info > 0 else 0.0
    
    # 3. Gini Reduction
    gini_reduction = parent_gini - weighted_child_gini

    return info_gain, gain_ratio, gini_reduction

# Evaluate the predictive power of our features
features_to_test = ['Surge_Multiplier', 'ETA_Bin', 'Distance_Bin', 'Rider_Rating']
metrics_summary = []

for feat in features_to_test:
    ig, gr, gini_red = calculate_feature_metrics(df_uber, feat, 'Accepted')
    metrics_summary.append({
        'Feature': feat,
        'Information_Gain': ig,
        'Gain_Ratio': gr,
        'Gini_Reduction': gini_red
    })

# Sort features by the unbiased metric (Gain Ratio)
df_metrics = pd.DataFrame(metrics_summary).sort_values(by='Gain_Ratio', ascending=False)

# ==============================================================================
# 4. VISUALIZATIONS AND CHARTS
# ==============================================================================

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# CHART 1: Comparison of Feature Selection Measures 
x = np.arange(len(df_metrics))
width = 0.35

# Plot Information Gain (Navy) vs Gain Ratio (Blue)
rects1 = axes[0].bar(x - width/2, df_metrics['Information_Gain'], width, label='Information Gain (Biased)', color='#2c3e50', alpha=0.9, edgecolor='black')
rects2 = axes[0].bar(x + width/2, df_metrics['Gain_Ratio'], width, label='Gain Ratio (Unbiased)', color='#2980b9', alpha=0.9, edgecolor='black')

axes[0].set_ylabel('Metric Evaluation Score', fontsize=12, fontweight='bold')
axes[0].set_title('1. Uber Feature Importance: Fixing Entropy Bias', fontsize=13, fontweight='bold', pad=12)
axes[0].set_xticks(x)
axes[0].set_xticklabels(df_metrics['Feature'], fontsize=11, fontweight='bold')
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
axes[0].grid(axis='y', linestyle='--', alpha=0.7)

# Annotate exact scores on the bars
for rect in rects1:
    h = rect.get_height()
    axes[0].annotate(f'{h:.3f}', xy=(rect.get_x() + rect.get_width()/2, h),
                     xytext=(0, 3), textcoords="offset points", ha='center', va='bottom', fontweight='bold', fontsize=9, color='#2c3e50')
for rect in rects2:
    h = rect.get_height()
    axes[0].annotate(f'{h:.3f}', xy=(rect.get_x() + rect.get_width()/2, h),
                     xytext=(0, 3), textcoords="offset points", ha='center', va='bottom', fontweight='bold', fontsize=9, color='#2980b9')

# CHART 2: Driver Acceptance Distribution across Top Feature 
# Academic styling for the countplot (Teal for Accept, Red/Orange for Reject)
sns.countplot(
    data=df_uber, 
    x='Surge_Multiplier', 
    hue='Accepted', 
    palette={1: '#27ae60', 0: '#e74c3c'}, 
    ax=axes[1], 
    edgecolor='black', 
    alpha=0.9
)

axes[1].set_title('2. Driver Acceptance Distribution by Top Feature (Surge)', fontsize=13, fontweight='bold', pad=12)
axes[1].set_xlabel('Surge Multiplier ($x$ times normal fare)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Driver Decision Count', fontsize=12, fontweight='bold')

# Customize Legend
handles, labels = axes[1].get_legend_handles_labels()
axes[1].legend(handles=handles, labels=['0 = Rejected', '1 = Accepted'], title='Final Decision', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
axes[1].grid(axis='y', linestyle='--', alpha=0.7)

# Clean up axes
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    "<b>نتیجه استراتژیک:</b> بررسی گراف ۱ یکی از معروف‌ترین تله‌های داده‌کاوی را آشکار می‌کند. "
    "اگر به میله‌های تیره (Information Gain) نگاه کنید، می‌بینید که ویژگی <code>Rider_Rating</code> امتیاز "
    "نسبتاً بالایی گرفته است. این یک تورش ریاضی است، زیرا تنوع بالای اعداد در این ستون، الگوریتم "
    "را فریب می‌دهد تا فکر کند شاخه‌بندی خوبی اتفاق افتاده است. اما وقتی از فرمول <b>Gain Ratio</b> "
    "(میله‌های آبی روشن) استفاده کردیم، این توهم از بین رفت و <code>Rider_Rating</code> به عنوان "
    "بی‌اثرترین متغیر شناخته شد. گراف ۲ تأیید می‌کند که رانندگان اوبر تصمیم نهایی خود را با تمرکز قطعی "
    "روی <code>Surge_Multiplier</code> (ضریب افزایش قیمت) می‌گیرند و متغیرهای دیگر نقش حاشیه‌ای دارند."
)
display_academic_text(insight_conclusion, "insight")
```
