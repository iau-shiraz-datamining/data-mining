  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: ارزیابی پایداری طبقه‌بند SVM با اعتبارسنجی متقابل در داده Iris
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در این مثال، هدف بررسی عملکرد و پایداری یک <b>طبقه‌بند خطی SVM</b> روی مجموعه‌داده معروف <b>Iris</b> است. از آنجا که پیاده‌سازی SVM در این مثال به صورت دودویی انجام شده است، کلاس‌های ۰ و ۱ در یک گروه و کلاس ۲ در گروه دیگر قرار می‌گیرند. مدل با استفاده از تابع زیان <b>Hinge Loss</b> و منظم‌سازی، از صفر پیاده‌سازی شده و پارامترهای آن با روش <b>Gradient Descent</b> به‌روزرسانی می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    برای جلوگیری از وابستگی نتیجه به یک تقسیم تصادفی خاص، از <b>اعتبارسنجی متقابل k-Fold</b> استفاده می‌شود. در این روش، داده‌ها به $k$ بخش تقسیم شده و در هر مرحله، یک بخش برای آزمون و $k-1$ بخش دیگر برای آموزش استفاده می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مجموعه داده $D$ به $k$ بخش مجزا تقسیم می‌شود:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$D = F_1 \cup F_2 \cup \cdots \cup F_k \quad , \quad F_i \cap F_j = \emptyset \quad \text{for } i \neq j$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در هر تکرار، مدل روی $k-1$ بخش آموزش داده شده و روی بخش باقی‌مانده ارزیابی می‌شود:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$M_i = \text{Accuracy}\left(\text{Train}(D \setminus F_i), F_i\right)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    پس از انجام تمام $k$ تکرار، میانگین Accuracy به عنوان برآورد کلی عملکرد مدل محاسبه می‌شود:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\bar{M}_{CV} = \frac{1}{k} \sum_{i=1}^{k} M_i$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    برای بررسی میزان تغییرپذیری عملکرد مدل در Foldهای مختلف، انحراف معیار Accuracy نیز محاسبه می‌شود:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$s_{CV} = \sqrt{\frac{1}{k-1} \sum_{i=1}^{k} (M_i - \bar{M}_{CV})^2}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    نتایج هر Fold به صورت جداگانه گزارش می‌شوند و سپس میانگین و انحراف معیار آن‌ها محاسبه می‌شود. همچنین برای درک بهتر پراکندگی عملکرد مدل، نتایج Foldها با نمودار میله‌ای و نمودار Box Plot نمایش داده می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: اگر میانگین Accuracy بالا و انحراف معیار پایین باشد، می‌توان نتیجه گرفت که مدل علاوه بر عملکرد مناسب، در برابر تغییر بخش آزمون نیز عملکرد نسبتاً پایداری دارد. در مقابل، انحراف معیار بالا نشان می‌دهد که عملکرد مدل به نمونه‌های آموزشی و آزمون هر Fold حساس است و ممکن است به داده بیشتر یا تنظیم مناسب‌تر پارامترهای مدل نیاز داشته باشد.<br><br>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "ارزیابی پایداری طبقه‌بند SVM با اعتبارسنجی متقابل در «اسپاتیفای» (From Scratch) 🎧", 
    "main_title"
)

display_academic_text(
    "در این مثال، هدف بررسی عملکرد یک طبقه‌بند خطی ماشین بردار پشتیبان (SVM) برای پیش‌بینی لغو اشتراک "
    "کاربران در اسپاتیفای است. مدل با استفاده از تابع زیان لولا (Hinge Loss) از پایه برنامه‌نویسی شده "
    "است. برای اطمینان از پایداری مدل در برابر تغییرات داده‌ها، از اعتبارسنجی متقابل (k-Fold CV) استفاده می‌شود.", 
    "paragraph"
)

# ==============================================================================
# 2. GENERATE SYNTHETIC SPOTIFY CHURN DATASET
# ==============================================================================
np.random.seed(42)
n_samples = 400

# Retained Users (Class -1 for SVM)
hours_retained = np.random.normal(50, 10, n_samples // 2)
skips_retained = np.random.normal(0.2, 0.1, n_samples // 2)
y_retained = np.full(n_samples // 2, -1)  # SVM requires labels -1 and 1

# Churned Users (Class +1 for SVM)
hours_churned = np.random.normal(30, 10, n_samples // 2)
skips_churned = np.random.normal(0.6, 0.1, n_samples // 2)
y_churned = np.full(n_samples // 2, 1)

X_data = np.vstack((np.column_stack((hours_retained, skips_retained)), 
                    np.column_stack((hours_churned, skips_churned))))
y_data = np.concatenate((y_retained, y_churned))

# Shuffle dataset
indices = np.random.permutation(n_samples)
X = X_data[indices]
y = y_data[indices]

# ==============================================================================
# 3. K-FOLD SPLIT & STANDARDIZATION FROM SCRATCH
# ==============================================================================
def kfold_split(n_samples, k=5, seed=42):
    rng = np.random.default_rng(seed)
    indices = np.arange(n_samples)
    rng.shuffle(indices)
    folds = np.array_split(indices, k)
    
    result = []
    for i in range(k):
        test_idx = folds[i]
        train_idx = np.concatenate([folds[j] for j in range(k) if j != i])
        result.append((train_idx, test_idx))
    return result

def standardize_train_test(X_train, X_test):
    """Standardizes using ONLY train statistics to prevent leakage."""
    mean = X_train.mean(axis=0)
    std = X_train.std(axis=0)
    std[std == 0] = 1.0
    return (X_train - mean) / std, (X_test - mean) / std

# ==============================================================================
# 4. BINARY LINEAR SVM FROM SCRATCH (HINGE LOSS)
# ==============================================================================
class BinarySVMScratch:
    """
    Binary Linear SVM using Hinge Loss + L2 Regularization (Soft Margin).
    Objective: J(w,b) = (lambda * ||w||^2) + (1/m) * sum(max(0, 1 - y_i * (w.x_i + b)))
    """
    def __init__(self, learning_rate=0.01, lambda_param=0.01, n_iters=1000):
        self.lr = learning_rate
        self.lambda_param = lambda_param
        self.n_iters = n_iters
        self.w = None
        self.b = None
        self.loss_history = []

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.w = np.zeros(n_features)
        self.b = 0.0
        self.loss_history = []

        for epoch in range(self.n_iters):
            scores = X @ self.w + self.b
            margins = y * scores
            
            # Mask identifying points that violate the margin (margin < 1)
            # These are our Support Vectors!
            mask = margins < 1

            # Subgradient of Hinge Loss + L2 Reg
            grad_w = (2 * self.lambda_param * self.w) - (X[mask].T @ y[mask]) / n_samples
            grad_b = -np.sum(y[mask]) / n_samples

            # Gradient descent update
            self.w -= self.lr * grad_w
            self.b -= self.lr * grad_b

            # Calculate Loss for tracking
            hinge_loss = np.maximum(0, 1 - y * (X @ self.w + self.b))
            loss = (self.lambda_param * np.sum(self.w ** 2)) + np.mean(hinge_loss)
            self.loss_history.append(loss)

        return self

    def predict(self, X):
        scores = X @ self.w + self.b
        return np.where(scores >= 0, 1, -1)

# ==============================================================================
# 5. EVALUATE SVM WITH 5-FOLD CROSS-VALIDATION
# ==============================================================================
k_folds = 5
fold_indices = kfold_split(len(y), k=k_folds, seed=42)
fold_scores = []
fold_results_table = []

for fold_num, (train_idx, test_idx) in enumerate(fold_indices):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    # Standardization must be INSIDE the fold loop
    X_train_std, X_test_std = standardize_train_test(X_train, X_test)

    svm = BinarySVMScratch(learning_rate=0.01, lambda_param=0.01, n_iters=1000)
    svm.fit(X_train_std, y_train)

    predictions = svm.predict(X_test_std)
    accuracy = np.mean(predictions == y_test)
    
    fold_scores.append(accuracy)
    fold_results_table.append({'Fold': f"Fold {fold_num + 1}", 'Test Accuracy': f"{accuracy:.4f}"})

mean_accuracy = np.mean(fold_scores)
std_accuracy = np.std(fold_scores, ddof=1) # Sample standard deviation

df_results = pd.DataFrame(fold_results_table)
df_results.loc[len(df_results)] = ['MEAN ± STD', f"{mean_accuracy:.4f} ± {std_accuracy:.4f}"]

display_side_by_side_datasets([df_results], [f"نتایج {k_folds}-Fold CV برای مدل SVM"])

# ==============================================================================
# 6. VISUALIZATIONS AND CHARTS (ACADEMIC BLUE PALETTE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

#  CHART 1: ACCURACY OF EACH FOLD 
fold_labels = [f"Fold {i + 1}" for i in range(len(fold_scores))]

bars = axes[0].bar(
    fold_labels, 
    fold_scores, 
    color="#2980b9", # Academic Blue
    edgecolor="black", 
    alpha=0.9, 
    width=0.6
)

# Mean accuracy line
axes[0].axhline(
    mean_accuracy, color="#c0392b", linewidth=2.5, linestyle="--", 
    label=f"Mean Accuracy = {mean_accuracy:.3f}"
)

# ±1 Standard deviation shading
axes[0].axhspan(
    mean_accuracy - std_accuracy, mean_accuracy + std_accuracy, 
    color="#e74c3c", alpha=0.15, label=f"±1 Std = {std_accuracy:.3f}"
)

axes[0].set_xlabel("Cross-Validation Fold", fontsize=12, fontweight="bold")
axes[0].set_ylabel("Test Accuracy", fontsize=12, fontweight="bold")
axes[0].set_title(f"{k_folds}-Fold Cross-Validation Accuracy (SVM)", fontsize=13, fontweight="bold", pad=12)
axes[0].legend(frameon=True, facecolor="#f8f9fa", edgecolor="#bdc3c7", fontsize=10, loc='lower right')
axes[0].grid(axis="y", linestyle="--", alpha=0.7)

# Dynamic y-axis for better visibility of variance
axes[0].set_ylim(min(fold_scores) - 0.05, 1.05)

# Annotate bars
for bar, value in zip(bars, fold_scores):
    axes[0].annotate(
        f"{value:.3f}", 
        xy=(bar.get_x() + bar.get_width() / 2, value),
        xytext=(0, 5), textcoords="offset points", 
        ha="center", fontsize=11, fontweight="bold", color="#2c3e50"
    )

#  CHART 2: BOX PLOT (VARIANCE ANALYSIS) 
axes[1].boxplot(
    fold_scores, 
    patch_artist=True, 
    widths=0.4,
    boxprops=dict(facecolor="#ecf0f1", color="#2c3e50", linewidth=2),
    medianprops=dict(color="#c0392b", linewidth=3),
    whiskerprops=dict(color="#2c3e50", linewidth=2),
    capprops=dict(color="#2c3e50", linewidth=2),
    flierprops=dict(marker="o", markerfacecolor="#e74c3c", markersize=8)
)

axes[1].set_xticks([1])
axes[1].set_xticklabels(["SVM Model Variance"], fontsize=12, fontweight="bold")
axes[1].set_ylabel("Accuracy Score", fontsize=12, fontweight="bold")
axes[1].set_title("Distribution of Accuracy Across Folds (Stability Analysis)", fontsize=13, fontweight="bold", pad=12)
axes[1].grid(axis="y", linestyle="--", alpha=0.7)
axes[1].set_ylim(axes[0].get_ylim())

# Mean marker on boxplot
axes[1].scatter([1], [mean_accuracy], color="#2980b9", s=100, zorder=5, edgecolor='black', label=f"Mean = {mean_accuracy:.3f}")
axes[1].legend(loc="lower right", fontsize=10, frameon=True, facecolor="#f8f9fa", edgecolor="#bdc3c7")

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 7: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    "<b>نتیجه استراتژیک:</b> بررسی گراف ۱ و گراف ۲ پایداری بسیار بالای این مدل SVM را تأیید می‌کند. "
    "انحراف معیار بسیار پایین مدل نشان می‌دهد که الگوریتم، نسبت به تغییر کردن بخش‌های دیتابیس "
    "(تغییر Training Set و Test Set در هر Fold) حساسیت یا واریانس بالایی ندارد و همواره دقتی نزدیک به "
    "میانگین را ثبت می‌کند. این یعنی مدل دچار پدیده بیش‌برازش (Overfitting) نشده و توانسته "
    "یک مرز تصمیم‌گیریِ ساختاری (مبتنی بر حاشیه ایمن یا Margin) ایجاد کند که در محیط Production "
    "می‌توان روی تعمیم‌پذیری آن به کاربران جدید اسپاتیفای کاملاً حساب باز کرد."
)
display_academic_text(insight_conclusion, "insight")
```

<li>
تفسیر خروجی و نمودارها:
