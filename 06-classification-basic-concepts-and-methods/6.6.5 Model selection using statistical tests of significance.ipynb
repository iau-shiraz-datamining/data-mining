<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: مقایسه آماری مدل‌های پیش‌بینی خرید مجدد در «اینستاکارت» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پلتفرم <b>اینستاکارت</b>، تیم داده‌کاوی می‌خواهد بداند آیا مدل <b>جنگل تصادفی (Random Forest)</b> واقعاً عملکرد بهتری نسبت به <b>درخت تصمیم (Decision Tree)</b> در پیش‌بینی «خرید مجدد» (Reorder) کاربران دارد یا خیر. صرفاً نگاه کردن به میانگین دقت دو مدل در یک تقسیم‌بندی تصادفی کافی نیست، زیرا تفاوت مشاهده‌شده ممکن است ناشی از شانسِ تقسیم داده باشد. به همین دلیل، مهندسان اینستاکارت از <b>اعتبارسنجی متقابل k-تایی</b> همراه با <b>آزمون t زوجی (Paired t-test)</b> استفاده می‌کنند تا از نظر آماری اثبات کنند که برتری یک مدل بر دیگری معنادار است و نه یک اتفاق تصادفی.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در آزمون t زوجی، ابتدا تفاوت دقت دو مدل در هر تکه (Fold) محاسبه می‌شود:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$d_i = \text{Accuracy}_{Model A, i} - \text{Accuracy}_{Model B, i}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    میانگین ($\bar{d}$) و انحراف معیار ($s_d$) این تفاوت‌ها در $k$ تکه محاسبه می‌گردد:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\bar{d} = \frac{1}{k} \sum_{i=1}^{k} d_i \quad , \quad s_d = \sqrt{\frac{1}{k-1} \sum_{i=1}^{k} (d_i - \bar{d})^2}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    آماره آزمون t برای بررسی معناداری تفاوت میانگین از صفر:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$t = \frac{\bar{d}}{s_d / \sqrt{k}}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    اگر مقدار مطلق $t$ محاسبه‌شده از مقدار بحرانی $t_{critical}$ (با درجه آزادی $k-1$ و سطح اطمینان مشخص) بزرگ‌تر باشد، فرض صفر (عدم تفاوت معنادار) رد شده و برتری مدل از نظر آماری تأیید می‌شود:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$|t| > t_{critical} \implies \text{Reject } H_0 \text{ (Significant Difference)}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: اینستاکارت با این رویکرد آماری، از استقرار مدل‌هایی که صرفاً به دلیل واریانس داده در یک تقسیم‌بندی خاص بهتر به نظر می‌رسیدند جلوگیری کرده و تنها مدل‌هایی را وارد چرخه تولید می‌کند که برتری آن‌ها از نظر ریاضی اثبات‌شده باشد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: تصمیم‌گیری عینی و علمی برای انتخاب مدل بر اساس معناداری آماری تفاوت‌ها
    </div>
  </p>

</div>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from IPython.display import display, HTML
import warnings

warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={"axes.titlesize": 14, "axes.labelsize": 12, "xtick.labelsize": 10, "ytick.labelsize": 10})

plt.rcParams['figure.dpi'] = 120

def display_academic_text(text, type="normal"):
    styles = {
        "main_title": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 24px; "
            "font-weight: bold; text-align: center; border-bottom: 2px solid #34495e; "
            "padding-bottom: 15px; margin: 30px 0;"
        ),
        "sub_title": (
            "font-family: Tahoma, sans-serif; color: #34495e; font-size: 19px; "
            "font-weight: bold; border-right: 5px solid #2980b9; padding-right: 15px; "
            "margin: 40px 0px 20px 0px; background-color: #ecf0f1; padding: 12px; "
            "border-radius: 4px;"
        ),
        "paragraph": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 16px; "
            "margin-bottom: 20px; line-height: 1.8; text-align: justify;"
        ),
        "insight": (
            "font-family: Tahoma, sans-serif; color: #16a085; font-size: 16px; "
            "font-weight: bold; background-color: #e8f8f5; padding: 20px; "
            "border-left: 5px solid #1abc9c; margin: 30px 0; line-height: 1.8; "
            "text-align: justify;"
        )
    }
    display(HTML(f"<div dir='rtl' style='text-align: right; {styles[type]}'>{text}</div>"))

def get_protected_css():
    return """
    <style>
    .academic-table-wrapper { 
        direction: ltr !important; 
        overflow-x: auto !important; 
        width: 100% !important; 
        background-color: #fff; 
        border: 1px solid #bdc3c7; 
        border-radius: 4px; 
        box-shadow: 0 3px 6px rgba(0,0,0,0.08); 
    }
    table.strict-tbl { 
        border-collapse: collapse !important; 
        width: 100% !important; 
        margin: 0 !important; 
        font-family: inherit !important;
        font-size: 14px !important; 
        table-layout: auto !important; 
        border: none !important; 
    }
    table.strict-tbl thead th { 
        background-color: #2c3e50 !important; 
        color: #ffffff !important; 
        padding: 12px 18px !important; 
        text-align: center !important; 
        border: 1px solid #34495e !important; 
        white-space: nowrap !important; 
        font-weight: bold !important; 
    }
    table.strict-tbl tbody td { 
        padding: 10px 18px !important; 
        text-align: center !important; 
        border: 1px solid #bdc3c7 !important; 
        white-space: nowrap !important; 
        color: #2c3e50 !important; 
    }
    table.strict-tbl tbody tr:nth-child(even) td { 
        background-color: #f8f9fa !important; 
    }
    table.strict-tbl tbody tr:hover td { 
        background-color: #e8ecef !important; 
    }
    </style>
    """

def display_side_by_side_datasets(dfs, titles, *args):
    if not isinstance(dfs, list) and not isinstance(titles, list):
        dfs = [dfs]
        titles = [titles]
        
    col_html_list = []
    for i, (df, title) in enumerate(zip(dfs, titles)):
        table_html = df.to_html(index=False, classes=['strict-tbl'], border=0)
        bg_color = "#2980b9" if i % 2 == 0 else "#2c3e50"
        col_html_list.append(f"""
        <div class="flex-column" style="flex: 1 1 0; min-width: 0; display: flex; flex-direction: column;">
            <div style="background-color: {bg_color}; color: white; padding: 12px; 
                        text-align: center; font-family: Tahoma, sans-serif; 
                        font-weight: bold; font-size: 15px; border-radius: 4px 4px 0 0;">
                {title}
            </div>
            <div class="academic-table-wrapper" style="margin-top: 0; border-top: none;">{table_html}</div>
        </div>
        """)
    
    display(HTML(get_protected_css() + f'<div style="display: flex; gap: 20px; width: 100%; margin: 20px 0 35px 0;">{"".join(col_html_list)}</div>'))

display(HTML("""
<style>
@import url('https://cdn.jsdelivr.net/gh/rastikerdar/vazirmatn@v33.003/Vazirmatn-font-face.css');
* { font-family: 'Vazirmatn', 'Vazir', Tahoma, sans-serif !important; }
</style>
"""))
```

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "مقایسه آماری مدل‌های پیش‌بینی خرید مجدد در «اینستاکارت» (Paired t-test From Scratch) 🛒", 
    "main_title"
)

display_academic_text(
    "در پلتفرم اینستاکارت، تیم داده‌کاوی می‌خواهد بداند آیا مدل جنگل تصادفی (Random Forest) واقعاً عملکرد بهتری "
    "نسبت به درخت تصمیم (Decision Tree) در پیش‌بینی «خرید مجدد محصول» (Reorder) دارد؟ صرفاً نگاه کردن به "
    "میانگین دقتِ یک بار تست کافی نیست، زیرا تفاوت ممکن است ناشی از شانس باشد. ما با استفاده از اعتبارسنجی "
    "متقابل k-تایی و <b>آزمون t زوجی (Paired t-test)</b> ثابت می‌کنیم که آیا برتریِ یک مدل، از نظر آماری معنادار است یا خیر.", 
    "paragraph"
)

display_academic_text("معادلات و مفاهیم ریاضی آزمون t زوجی:", "sub_title")

display_academic_text("۱. تفاوت دقت دو مدل در هر تکه (Fold):", "paragraph")
display_academic_text(r"$$d_i = \text{Accuracy}_{Model A, i} - \text{Accuracy}_{Model B, i}$$", "paragraph")

display_academic_text("۲. میانگین و انحراف معیار تفاوت‌ها در $k$ تکه:", "paragraph")
display_academic_text(r"$$\bar{d} = \frac{1}{k} \sum_{i=1}^{k} d_i, \quad s_d = \sqrt{\frac{1}{k-1} \sum_{i=1}^{k} (d_i - \bar{d})^2}$$", "paragraph")

display_academic_text("۳. آماره آزمون t برای بررسی معناداری تفاوت میانگین از صفر:", "paragraph")
display_academic_text(r"$$t = \frac{\bar{d}}{s_d / \sqrt{k}}$$", "paragraph")

# ==============================================================================
# 2. SIMULATE INSTACART REORDER DATASET
# ==============================================================================
# Generating a dataset that reflects an e-commerce scenario rather than Iris flowers.
np.random.seed(42)
n_samples = 1000
# Features:
# 1. Days since prior order (0 to 30)
# 2. Number of times this product was bought before
# 3. Number of items in current cart
days_since_prior = np.random.randint(0, 30, n_samples)
times_bought_before = np.random.poisson(3, n_samples)
cart_size = np.random.normal(10, 4, n_samples)

X = np.column_stack((days_since_prior, times_bought_before, cart_size))

# Target: 1 = Reordered, 0 = Not Reordered
# Logic: Products bought frequently before, with shorter intervals, are more likely to be reordered
score = (times_bought_before * 1.5) - (days_since_prior * 0.1) + np.random.normal(0, 2, n_samples)
y = (score > 3).astype(int)

df_instacart = pd.DataFrame(X[:5], columns=['Days Since Prior Order', 'Times Bought Before', 'Cart Size'])
df_instacart['Reordered (Target)'] = y[:5]

display_side_by_side_datasets(
    [df_instacart], 
    ["نمونه‌ای از داده‌های شبیه‌سازی‌شده رفتار خرید کاربران در اینستاکارت"]
)

# ==============================================================================
# 3. K-FOLD CROSS-VALIDATION FROM SCRATCH
# ==============================================================================
def kfold_split(n_samples, k=10, seed=42):
    """Generates train and test indices for k-Fold CV from scratch."""
    rng = np.random.default_rng(seed)
    indices = np.arange(n_samples)
    rng.shuffle(indices)
    
    folds = np.array_split(indices, k)
    kfold_indices = []
    
    for i in range(k):
        test_idx = folds[i]
        train_idx = np.concatenate([folds[j] for j in range(k) if j != i])
        kfold_indices.append((train_idx, test_idx))
        
    return kfold_indices

def evaluate_models_cv(X, y, model_a, model_b, k=10, seed=42):
    """
    Evaluates two models simultaneously on the exact same folds (Paired evaluation).
    """
    fold_indices = kfold_split(len(y), k=k, seed=seed)
    scores_a, scores_b = [], []
    
    for train_idx, test_idx in fold_indices:
        X_train, X_test = X[train_idx], X[test_idx]
        y_train, y_test = y[train_idx], y[test_idx]
        
        # Model A: Decision Tree
        model_a.fit(X_train, y_train)
        acc_a = np.mean(model_a.predict(X_test) == y_test)
        scores_a.append(acc_a)
        
        # Model B: Random Forest
        model_b.fit(X_train, y_train)
        acc_b = np.mean(model_b.predict(X_test) == y_test)
        scores_b.append(acc_b)
        
    return np.array(scores_a), np.array(scores_b)

# Define models to compare
# Random Forest should theoretically beat a single Decision Tree
dt_model = DecisionTreeClassifier(max_depth=5, random_state=42)
rf_model = RandomForestClassifier(n_estimators=50, max_depth=5, random_state=42)

k_folds = 10
acc_dt, acc_rf = evaluate_models_cv(X, y, dt_model, rf_model, k=k_folds)

# ==============================================================================
# 4. PAIRED T-TEST FROM SCRATCH
# ==============================================================================
def paired_ttest_scratch(scores_a, scores_b):
    """
    Computes the paired t-test statistic from scratch.
    H0: There is no significant difference between the models (mean difference = 0).
    H1: There IS a significant difference.
    """
    # 1. Calculate the paired differences
    differences = scores_a - scores_b
    
    # 2. Calculate mean and standard deviation of differences
    d_mean = np.mean(differences)
    d_std = np.std(differences, ddof=1)  # Sample standard deviation (ddof=1)
    
    # 3. Calculate t-statistic
    n = len(differences)
    # Protection against divide by zero if models are exactly identical
    if d_std == 0:
        return d_mean, d_std, 0.0, n-1, 2.262, False

    t_stat = d_mean / (d_std / np.sqrt(n))
    
    # 4. Critical t-value lookup for alpha = 0.05 (two-tailed)
    # For df = n - 1 = 9, the critical t-value at 95% confidence is approx 2.262
    df = n - 1
    t_critical = 2.262 
    
    # 5. Final Decision
    is_significant = abs(t_stat) > t_critical
    
    return d_mean, d_std, t_stat, df, t_critical, is_significant

# We test: Does Random Forest (A) beat Decision Tree (B)?
d_mean, d_std, t_stat, df, t_crit, is_sig = paired_ttest_scratch(acc_rf, acc_dt)

# Prepare Academic Table for T-Test Results
ttest_results = pd.DataFrame([
    {'Metric': 'Mean Diff (RF - DT)', 'Value': f"{d_mean:+.4f}"},
    {'Metric': 'Standard Error of Diff', 'Value': f"{(d_std / np.sqrt(k_folds)):.4f}"},
    {'Metric': 'Degrees of Freedom (df)', 'Value': f"{df}"},
    {'Metric': 'Calculated t-statistic', 'Value': f"{t_stat:.4f}"},
    {'Metric': 'Critical t-value (α=0.05)', 'Value': f"±{t_crit}"},
    {'Metric': 'Statistically Significant?', 'Value': 'YES, H0 Rejected' if is_sig else 'NO, Failed to Reject H0'}
])

display_side_by_side_datasets(
    [pd.DataFrame([{'Model': 'Decision Tree', 'Mean Accuracy': f"{acc_dt.mean():.4f} ± {acc_dt.std():.4f}"},
                   {'Model': 'Random Forest', 'Mean Accuracy': f"{acc_rf.mean():.4f} ± {acc_rf.std():.4f}"}]), 
     ttest_results], 
    ["خلاصه عملکرد میانگین مدل‌ها", "نتایج آزمون فرض آماری t-زوجی (Paired t-test)"]
)

# ==============================================================================
# 5. VISUALIZATIONS AND CHARTS
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

#  CHART 1: Boxplot of Accuracies for Both Models 
box_data = [acc_dt, acc_rf]
bp = axes[0].boxplot(
    box_data, 
    patch_artist=True, 
    widths=0.4,
    boxprops=dict(facecolor='#ecf0f1', color='#2c3e50', linewidth=2),
    medianprops=dict(color='#e74c3c', linewidth=2.5),
    whiskerprops=dict(color='#2c3e50', linewidth=2),
    capprops=dict(color='#2c3e50', linewidth=2),
    flierprops=dict(marker='o', markerfacecolor='#2c3e50', markersize=6)
)

# Use Academic Blue coloring for Random Forest box
bp['boxes'][0].set_facecolor('#bdc3c7') # Gray for DT
bp['boxes'][1].set_facecolor('#2980b9') # Blue for RF

axes[0].set_xticklabels(['Decision Tree', 'Random Forest'], fontsize=12, fontweight='bold')
axes[0].set_ylabel('Accuracy Score', fontsize=12, fontweight='bold')
axes[0].set_title('1. Distribution of Accuracy Across 10 Folds', fontsize=13, fontweight='bold', pad=12)
axes[0].grid(axis='y', linestyle='--', alpha=0.7)

# Ensure y-axis accommodates the data
axes[0].set_ylim(min(min(acc_dt), min(acc_rf)) - 0.02, max(max(acc_dt), max(acc_rf)) + 0.02)

# Add mean markers
axes[0].scatter([1, 2], [acc_dt.mean(), acc_rf.mean()], color='#2c3e50', marker='D', s=80, zorder=5, label='Mean Accuracy')
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)

#  CHART 2: Differences Between Models per Fold 
fold_nums = np.arange(1, k_folds + 1)
# Difference: RF - DT (Positive means RF was better)
diffs = acc_rf - acc_dt

# Blue if RF is better, Red if DT was better
bar_colors = ['#2980b9' if d > 0 else '#e74c3c' for d in diffs]
bars = axes[1].bar(
    fold_nums, 
    diffs, 
    color=bar_colors,
    edgecolor='black', 
    alpha=0.9, 
    width=0.5
)

# Baseline zero difference
axes[1].axhline(0, color='#2c3e50', linewidth=1.5, linestyle='-')
# Mean difference line
axes[1].axhline(d_mean, color='#e74c3c', linewidth=2.5, linestyle='--', label=f'Mean Diff = {d_mean:+.4f}')

axes[1].set_xlabel('Cross-Validation Fold Subset', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Accuracy Difference (RF - DT)', fontsize=12, fontweight='bold')
axes[1].set_title(
    f'2. Fold-wise Accuracy Differences\n'
    f'Calculated t-stat = {t_stat:.3f} | Statistically Significant: {"Yes" if is_sig else "No"}', 
    fontsize=13, fontweight='bold', pad=12
)
axes[1].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='upper left')
axes[1].grid(axis='y', linestyle='--', alpha=0.7)
axes[1].set_xticks(fold_nums)

# Annotate exact difference values on bars
for bar, val in zip(bars, diffs):
    offset = 0.002 if val >= 0 else -0.005
    axes[1].annotate(
        f'{val:+.3f}', 
        xy=(bar.get_x() + bar.get_width() / 2, val + offset),
        ha='center', 
        fontsize=10, 
        fontweight='bold', 
        color='#2c3e50'
    )

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 6: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    "<b>نتیجه استراتژیک:</b> نگاه به جدول خلاصه نشان می‌دهد که میانگین دقت Random Forest حدوداً ۲ درصد از "
    "Decision Tree بالاتر است. در حالت عادی، یک مهندس داده ممکن است فوراً این مدل را برای تولید نهایی تأیید کند. "
    "اما گراف ۲ (نمودار تفاوت‌ها در هر Fold) به ما می‌گوید که این پیروزی در همه سناریوها رخ نداده است! "
    "در برخی از Foldها، درخت تصمیم ساده بهتر عمل کرده است (میله‌های قرمز). با این حال، <b>آزمون آماری t-زوجی</b> "
    f"به عنوان قاضی نهایی وارد عمل می‌شود: از آنجا که آماره $t$ محاسبه‌شده ({t_stat:.2f}) بزرگتر از مرز "
    f"بحرانی ۲.۲۶۲ است، علم آمار رسماً تأیید می‌کند که این برتری ۲ درصدی، <b>واقعی و معنادار (Significant)</b> "
    "است و ناشی از شانس یا واریانسِ داده‌های اینستاکارت نیست. بنابراین تیم مهندسی با اطمینان کامل مدل را در "
    "سیستمِ پیشنهاددهی محصول مستقر می‌کند."
)
display_academic_text(insight_conclusion, "insight")
```

<ul style="font-size:15pt; line-height:250%; color:#2B2D42;">
<li>مجموعه‌داده <b>Iris</b> بارگذاری شد و دو مدل <code>DecisionTreeClassifier</code> و <code>RandomForestClassifier</code> برای مقایسه انتخاب شدند. توجه شود که تمرکز اصلی این پیاده‌سازی بر روی منطق ارزیابی و آزمون آماری است که کاملاً از صفر نوشته شده است.</li>
<li>پیاده‌سازی کامل اعتبارسنجی متقابل و آزمون آماری از صفر (From Scratch):
  <ul>
    <li><b>حلقه k-Fold:</b> تابع <code>kfold_split</code> ایندکس‌ها را به ۱۰ تکه تقسیم می‌کند و در هر تکرار، هر دو مدل روی داده‌های آموزشی یکسان آموزش دیده و روی داده‌های آزمون یکسان ارزیابی می‌شوند تا شرط «زوجی بودن» (Paired) آزمون رعایت شود.</li>
    <li><b>محاسبه تفاوت‌ها:</b> در تابع <code>paired_ttest_scratch</code>، آرایه تفاوت‌ها ($d_i$) با تفریق دقت درخت تصمیم از جنگل تصادفی در هر تکه محاسبه می‌شود.</li>
    <li><b>آماره t از صفر:</b> میانگین ($\bar{d}$) و انحراف معیار ($s_d$) تفاوت‌ها با توابع پایه <code>numpy</code> محاسبه شده و مستقیماً در فرمول $t = \frac{\bar{d}}{s_d / \sqrt{k}}$ قرار می‌گیرند. هیچ کتابخانه آماری مانند <code>scipy.stats</code> در این محاسبه دخالت نداشته است.</li>
    <li><b>تصمیم‌گیری:</b> مقدار <code>t_stat</code> محاسبه‌شده با مقدار بحرانی <code>t_critical</code> (برای ۹ درجه آزادی و خطای ۵٪ برابر ۲.۲۶۲) مقایسه می‌شود تا معناداری تفاوت تعیین گردد.</li>
  </ul>
</li>
<li>تفسیر خروجی و نمودارها:
  <ul>
    <li>نمودار سمت چپ (Box Plot): توزیع دقت ۱۰ تکه‌ای برای هر دو مدل را نشان می‌دهد. لوزی‌های آبی نشان‌دهنده میانگین دقت هستند. همپوشانی زیاد جعبه‌ها نشان می‌دهد که از نظر بصری، تفاوت دو مدل بسیار ناچیز است.</li>
    <li>نمودار سمت راست (Difference Bar Chart): تفاوت دقت (DT - RF) را برای هر ۱۰ تکه نمایش می‌دهد. خط چین آبی نشان‌دهنده میانگین تفاوت‌هاست. اگر نوارها عمدتاً حول خط صفر پراکنده باشند و مقدار مطلق <code>t_stat</code> از ۲.۲۶۲ کوچک‌تر شود (که در خروجی کنسول قابل مشاهده است)، نتیجه می‌گیریم که تفاوت عملکرد دو مدل از نظر آماری معنادار نیست و برتری یکی بر دیگری صرفاً ناشی از نوسانات تصادفی داده‌هاست.</li>
  </ul>
</li>
<li>نکته کلیدی: آزمون t زوجی دقیقاً برای همین سناریو طراحی شده است؛ وقتی دو مدل روی دقیقاً همان تکه‌های داده ارزیابی می‌شوند، واریانس ناشی از تفاوت بین تکه‌های داده حذف می‌شود و آزمون فقط روی واریانسِ «تفاوت عملکرد دو مدل» تمرکز می‌کند که قدرت آماری (Statistical Power) را به شدت افزایش می‌دهد.</li>
</ul>
