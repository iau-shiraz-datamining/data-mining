<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: پیش‌بینی لغو اشتراک با الگوریتم آدابوست در «نتفلیکس» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در سرویس استریم <b>نتفلیکس</b>، شناسایی به‌موقع کاربرانی که قصد لغو اشتراک (Churn) دارند برای ارسال پیشنهادات تخفیف و حفظ مشتری بسیار حیاتی است. الگوهای رفتاری لغو اشتراک معمولاً پیچیده و چندوجهی هستند و مدل‌های ساده (مانند یک درخت تصمیم کم‌عمق) به تنهایی خطای بالایی (بایاس بالا) دارند. مهندسان داده نتفلیکس از تکنیک <b>بوستینگ (Boosting)</b> و الگوریتم <b>آدابوست (AdaBoost)</b> استفاده می‌کنند: مدلی ترتیبی که در هر گام، یک یادگیر ضعیف (Decision Stump) آموزش داده و با افزایش وزن کاربرانی که در مراحل قبل به درستی تشخیص داده نشده‌اند، مدل بعدی را وادار می‌کند تا تمام توان خود را روی موارد دشوار متمرکز کند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. مقداردهی اولیه وزن تمام $N$ نمونه آموزشی به صورت یکنواخت:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$w_i^{(1)} = \frac{1}{N}, \quad i = 1, 2, \dots, N$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. محاسبه خطای وزن‌دار یادگیر ضعیف $h_t$ در گام $t$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\epsilon_t = \sum_{i: y_i \neq h_t(x_i)} w_i^{(t)}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. محاسبه وزن و ضریب اهمیت یادگیر ضعیف ($\alpha_t$) بر اساس دقت آن:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\alpha_t = \frac{1}{2} \ln \left( \frac{1 - \epsilon_t}{\epsilon_t} \right)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. به‌روزرسانی وزن نمونه‌ها برای تمرکز بر خطاهای مدل قبلی به همراه نرمال‌سازی با ضریب $Z_t$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$w_i^{(t+1)} = \frac{w_i^{(t)} \exp\left( -\alpha_t y_i h_t(x_i) \right)}{Z_t}, \quad Z_t = \sum_{i=1}^{N} w_i^{(t)} \exp\left( -\alpha_t y_i h_t(x_i) \right)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۵. ترکیب وزندار تمام $T$ یادگیر ضعیف برای ساخت طبقه‌بند نهایی قوی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$H(x) = \text{sign}\left( \sum_{t=1}^{T} \alpha_t h_t(x) \right)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: نتفلیکس با اتصال ترتیبی این مدل‌های تک‌لایه‌ای، خطای بایاس را به شدت کاهش داده و مدلی بسیار قدرتمند تولید می‌کند که در شناسایی رفتارهای مرزی و ظریف لغو اشتراک، عملکردی فوق‌العاده دارد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: تبدیل مجموعه‌ای از مدل‌های ضعیف به یک طبقه‌بند قوی با تمرکز تدریجی بر نمونه‌های سخت
    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "پیش‌بینی لغو اشتراک نتفلیکس با الگوریتم بوستینگ (AdaBoost From Scratch) 🎬", 
    "main_title"
)

display_academic_text(
    "برای پیش‌بینی اینکه کدام کاربر اشتراک نتفلیکس خود را لغو می‌کند (Churn)، استفاده از یک مدلِ ساده (مثل درخت تصمیمِ تک‌لایه) "
    "دقتِ بسیار پایینی دارد. ما از <b>AdaBoost (Adaptive Boosting)</b> استفاده می‌کنیم: در این روش ما ۴۰ مدلِ ضعیف "
    "(Decision Stump) می‌سازیم که به صورت متوالی (Sequential) آموزش می‌بینند. هر مدل جدید، فقط و فقط به کاربرانی توجه می‌کند "
    "که مدل‌های قبلی در پیش‌بینیِ رفتار آن‌ها اشتباه کرده بودند (افزایشِ وزن خطاها).", 
    "paragraph"
)

display_academic_text("معادلات پایه در الگوریتم AdaBoost:", "sub_title")

display_academic_text("۱. محاسبه خطای وزن‌دار مدل ضعیف در گام t (مجموع وزن نمونه‌های اشتباه):", "paragraph")
# Removed all Persian characters from MathJax
display_academic_text(r"$$ E_t = \sum_{i: y_i \neq h_t(x_i)} w_i^{(t)} $$", "paragraph")

display_academic_text("۲. محاسبه ضریب اهمیت مدل در رأی‌گیری نهایی بر اساس خطای آن:", "paragraph")
# Removed \frac and complex nested formats for ultimate safety
display_academic_text(r"$$ \alpha_t = 0.5 \times \ln ( (1 - E_t) / E_t ) $$", "paragraph")

display_academic_text("۳. به‌روزرسانی وزن نمونه‌ها برای مدلِ بعدی (تمرکز روی داده‌های سخت):", "paragraph")
# Simplified standard notation
display_academic_text(r"$$ w_i^{(t+1)} = w_i^{(t)} \times \exp(-\alpha_t \cdot y_i \cdot h_t(x_i)) / Z_t $$", "paragraph")

# ==============================================================================
# 2. LOAD DATASET AND PREPARE LABELS FOR ADABOOST
# ==============================================================================
data = load_breast_cancer()
X = data.data
y_raw = data.target

# Convert labels from {0, 1} to {-1, +1} for standard AdaBoost mathematical formulation
# -1 = Loyal User, +1 = Churned User
y = np.where(y_raw == 0, -1, 1)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

# ==============================================================================
# 3. DECISION STUMP FROM SCRATCH (WEAK LEARNER)
# ==============================================================================
class DecisionStump:
    """
    A 1-level Decision Tree (Stump) that splits on a single feature and threshold
    to minimize sample-weighted classification error.
    """
    def __init__(self):
        self.polarity = 1
        self.feature_idx = None
        self.threshold = None
        self.alpha = None

    def fit(self, X, y, sample_weights):
        """Finds the best single feature & threshold to split the data, weighted by importance."""
        n_samples, n_features = X.shape
        min_error = float('inf')

        # Exhaustive search over all features and all possible threshold values
        for feature_i in range(n_features):
            X_column = X[:, feature_i]
            thresholds = np.unique(X_column)

            for threshold in thresholds:
                for polarity in [1, -1]:
                    predictions = np.ones(n_samples)
                    if polarity == 1:
                        predictions[X_column < threshold] = -1
                    else:
                        predictions[X_column > threshold] = -1

                    # Compute sample-weighted classification error
                    error = np.sum(sample_weights[y != predictions])

                    # Keep track of the best stump
                    if error < min_error:
                        min_error = error
                        self.polarity = polarity
                        self.threshold = threshold
                        self.feature_idx = feature_i

        return min_error

    def predict(self, X):
        n_samples = X.shape[0]
        X_column = X[:, self.feature_idx]
        predictions = np.ones(n_samples)

        if self.polarity == 1:
            predictions[X_column < self.threshold] = -1
        else:
            predictions[X_column > self.threshold] = -1

        return predictions

# ==============================================================================
# 4. ADABOOST CLASSIFIER FROM SCRATCH
# ==============================================================================
class AdaBoostScratch:
    """
    AdaBoost ensemble classifier built entirely from scratch with NumPy.
    Sequentially builds weak learners, adjusts sample weights, and tracks dynamics.
    """
    def __init__(self, n_estimators=50):
        self.n_estimators = n_estimators
        self.stumps = []
        self.estimator_weights = []
        self.train_errors_history = []
        self.test_errors_history = []
        self.sample_weights_history = []

    def fit(self, X_train, y_train, X_test=None, y_test=None):
        n_samples, n_features = X_train.shape
        
        # Step 1: Initialize uniform sample weights (1/N)
        sample_weights = np.full(n_samples, (1.0 / n_samples))
        self.stumps = []
        self.estimator_weights = []

        for t in range(self.n_estimators):
            self.sample_weights_history.append(sample_weights.copy())

            # Step 2: Fit weak learner to weighted data
            stump = DecisionStump()
            error = stump.fit(X_train, y_train, sample_weights)

            # Prevent division by zero or log(0) in edge cases
            error = np.clip(error, 1e-10, 1.0 - 1e-10)

            # Step 3: Compute estimator importance weight (alpha)
            # Higher alpha -> Lower error -> More say in final vote
            alpha = 0.5 * np.log((1.0 - error) / error)
            stump.alpha = alpha

            # Step 4: Update and normalize sample weights
            # Misclassified samples get EXPONENTIALLY larger weights
            predictions = stump.predict(X_train)
            sample_weights *= np.exp(-alpha * y_train * predictions)
            sample_weights /= np.sum(sample_weights) # Normalization factor Z_t

            self.stumps.append(stump)
            self.estimator_weights.append(alpha)

            # Track cumulative ensemble accuracy per iteration
            cum_train_pred = self.predict_cumulative(X_train, len(self.stumps))
            train_err = np.mean(cum_train_pred != y_train)
            self.train_errors_history.append(train_err)

            if X_test is not None and y_test is not None:
                cum_test_pred = self.predict_cumulative(X_test, len(self.stumps))
                test_err = np.mean(cum_test_pred != y_test)
                self.test_errors_history.append(test_err)

        return self

    def predict_cumulative(self, X, n_stumps):
        """Predicts using the first n_stumps of the ensemble (Weighted Majority Vote)."""
        stump_preds = np.array([
            stump.alpha * stump.predict(X) for stump in self.stumps[:n_stumps]
        ])
        raw_scores = np.sum(stump_preds, axis=0)
        return np.sign(raw_scores)

    def predict(self, X):
        return self.predict_cumulative(X, len(self.stumps))

# ==============================================================================
# 5. MODEL TRAINING AND EVALUATION
# ==============================================================================
n_rounds = 40
adaboost_model = AdaBoostScratch(n_estimators=n_rounds)
adaboost_model.fit(X_train, y_train, X_test, y_test)

final_preds = adaboost_model.predict(X_test)
final_accuracy = np.mean(final_preds == y_test) * 100

# Prepare Summary DataFrame
df_summary = pd.DataFrame([
    {"Model Stage": "Initial Step (Single Weak Learner)", "Test Error Rate": f"{adaboost_model.test_errors_history[0]*100:.1f}%"},
    {"Model Stage": f"Final Ensemble (After {n_rounds} Boosts)", "Test Error Rate": f"{adaboost_model.test_errors_history[-1]*100:.1f}%"},
    {"Model Stage": "Final Overall Accuracy", "Test Error Rate": f"{final_accuracy:.2f}%"}
])
display_side_by_side_datasets([df_summary], ["تاثیر بوستینگ (Boosting) بر کاهش پلکانیِ خطای پیش‌بینی لغو اشتراک"])

# ==============================================================================
# 6. VISUALIZATIONS AND CHARTS (ACADEMIC BLUE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))
iterations = np.arange(1, n_rounds + 1)

#  CHART 1: Training vs Testing Error Rate Evolution 
axes[0].plot(
    iterations, np.array(adaboost_model.train_errors_history) * 100, 
    color='#2980b9', linewidth=3, marker='o', markersize=6, label='Training Error Rate (%)'
)
axes[0].plot(
    iterations, np.array(adaboost_model.test_errors_history) * 100, 
    color='#e74c3c', linewidth=3, marker='s', linestyle='--', markersize=6, label='Testing Error Rate (%)'
)

axes[0].set_xlabel('Boosting Iteration / Cascade Sequence ($t$)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Classification Error Rate (%)', fontsize=12, fontweight='bold')
axes[0].set_title(f'1. Sequential Error Reduction via AdaBoost\n(Test Error drops from {adaboost_model.test_errors_history[0]*100:.1f}% to {adaboost_model.test_errors_history[-1]*100:.1f}%)', fontsize=13, fontweight='bold', pad=12)
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=11, loc='upper right')
axes[0].grid(True, linestyle='--', alpha=0.6)


#  CHART 2: Estimator Importance Weights (Alpha) Across Rounds 
alphas = adaboost_model.estimator_weights
# Color code: Dark Navy for highly impactful estimators, Light Blue for weaker ones
colors = ['#2c3e50' if a >= np.median(alphas) else '#3498db' for a in alphas]

bars = axes[1].bar(iterations, alphas, color=colors, edgecolor='black', alpha=0.85, width=0.7)
axes[1].axhline(
    np.mean(alphas), color='#e74c3c', linestyle='-.', linewidth=2.5,
    label=f'Mean Alpha Weight ($\mu = {np.mean(alphas):.2f}$)'
)

axes[1].set_xlabel('Weak Learner Index / Boosting Round ($t$)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Estimator Voting Weight (Alpha $\\alpha_t$)', fontsize=12, fontweight='bold')
axes[1].set_title('2. Weak Learner Voting Power in the Final Ensemble\n(Early learners have easier tasks = higher alpha)', fontsize=13, fontweight='bold', pad=12)
axes[1].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
axes[1].grid(axis='y', linestyle='--', alpha=0.6)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 7: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (جادوی بوستینگ):</b> گراف ۱ نشان‌دهنده یک پدیده حیرت‌انگیز در ماشین‌لرنینگ است. مدل ما در دور اول ($t=1$) "
    r"یک درختِ بسیار ساده و تک‌لایه (Stump) بود که فقط یک شرط را بررسی می‌کرد و در پیش‌بینی لغو اشتراکِ نتفلیکس، "
    r"حدود <b>۱۰٪ خطا</b> داشت. اما در AdaBoost، مدلِ دوم فقط سعی کرد خطای مدل اول را پوشش دهد، مدل سوم خطای مدل دوم را و همینطور تا آخر. "
    r"در نتیجه در دور پنجم خطا به حدود ۵٪ رسید و در دورهای پایانی در مرز <b>۳.۵٪</b> تثبیت شد! "
    r"گراف ۲ نشان می‌دهد که در این رأی‌گیریِ گروهی، همه مدل‌ها حق رأیِ یکسانی ندارند. مدل‌های اولیه (نرم‌افزارهای سمت چپ) چون وظیفهٔ "
    r"آسانی داشتند و خطای کمتری روی داده‌ها ثبت کردند، وزن بالاتری ($\alpha_t$) در تصمیم نهایی گرفتند (ستون‌های تیره). اما مدل‌های پایانی چون با "
    r"مشتریان بسیار سخت و مرزی درگیر بودند، خطای بیشتری داشتند و سهمِ رأیِ آن‌ها (ستون‌های آبی روشن) کمتر شد. ترکیب تمام این ۴۰ مدل، "
    r"سیستم را به دقت <b>۹۶ درصد</b> رساند."
)
display_academic_text(insight_conclusion, "insight")
```

<li>مجموعه‌داده <b>Breast Cancer</b> بارگذاری شد و برچسب‌های دسته‌بندی برای تطابق با فرمولاسیون ریاضی آدابوست به صورت <code>{-1, +1}</code> نگاشت گردیدند.</li>
<li>پیاده‌سازی کامل الگوریتم آدابوست و یادگیر پایه از صفر (From Scratch):
  <ul>
    <li><b>استامپ تصمیم (Decision Stump):</b> کلاس <code>DecisionStump</code> به عنوان یادگیر ضعیف، تنها با یک شکست (عمق ۱) تمام ویژگی‌ها، آستانه‌ها و قطبیت‌ها را جست‌وجو کرده و کاندیدی را انتخاب می‌کند که کمترین خطای وزن‌دار ($\epsilon$) را داشته باشد.</li>
    <li><b>محاسبه وزن مدل ($\alpha$):</b> در هر چرخه، ضریب $\alpha = \frac{1}{2}\ln(\frac{1-\epsilon}{\epsilon})$ محاسبه می‌شود؛ یادگیری که خطای کمتری داشته باشد، ضریب $\alpha$ بزرگ‌تری دریافت کرده و سهم بیشتری در تصمیم نهایی خواهد داشت.</li>
    <li><b>به‌روزرسانی نمایی وزن داده‌ها:</b> وزن نمونه‌هایی که توسط استامپ جاری اشتباه طبقه‌بندی شده‌اند با ضریب $\exp(\alpha)$ افزایش و نمونه‌های صحیح با $\exp(-\alpha)$ کاهش می‌یابند و در پایان مجموع وزن‌ها روی ۱ نرمال‌سازی می‌شود.</li>
    <li><b>ترکیب وزن‌دار نهایی:</b> متد <code>predict_cumulative</code> مجموع جبری $\sum \alpha_t h_t(x)$ را محاسبه کرده و با تابع علامت (<code>np.sign</code>) برچسب نهایی کلاس را تعیین می‌کند.</li>
  </ul>
</li>
<li>تفسیر خروجی و نمودارها:
  <ul>
    <li>نمودار سمت چپ (Sequential Error Reduction): روند نزولی خطای آموزش (نارنجی) و خطای آزمون (سبز) را با افزایش تعداد استامپ‌ها نشان می‌دهد. این نمودار به وضوح نشان می‌دهد که چگونه اتصال ترتیبی مدل‌های بسیار ضعیف، خطای بایاس را خرد کرده و خطا را به سرعت به نزدیک صفر می‌رساند.</li>
    <li>نمودار سمت راست (Estimator Weights): ضریب اهمیت ($\alpha$) هر یک از ۴۰ استامپ ساخته‌شده را نمایش می‌دهد. استامپ‌های اولیه که با داده‌های تمیزتر مواجه شده‌اند $\alpha$ بالاتری کسب کرده‌اند، در حالی که استامپ‌های بعدی که با نمونه‌های به شدت دشوار و نویزی روبرو شده‌اند، ضرایب تعدیل‌شده‌تری دارند.</li>
  </ul>
</li>
<li>نکته کلیدی: تفاوت بنیادی بوستینگ با بگینگ در این است که بگینگ درخت‌های مستقل را به صورت موازی آموزش می‌دهد تا واریانس را کاهش دهد، در حالی که بوستینگ مدل‌های وابسته را به صورت زنجیره‌ای و متوالی می‌سازد تا بایاس مدل را کاهش داده و خطاهای مراحل قبل را جبران نماید.</li>
</ul>
