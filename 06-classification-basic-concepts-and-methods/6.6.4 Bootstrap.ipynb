<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: تخمین فاصله اطمینان میانگین زمان انتظار در «اوبر» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پلتفرم <b>اوبر</b>، تیم عملیات شهری نیاز دارد بداند میانگین زمان انتظار مسافر برای رسیدن راننده در یک منطقه خاص دقیقاً چقدر است. اما داده‌های موجود فقط یک نمونه محدود (مثلاً ۲۰۰ سفر) است و توزیع زمان انتظار به دلیل ترافیک و حوادث غیرمترقبه، به هیچ وجه نرمال نیست. در این شرایط، فرمول‌های کلاسیک آماری (مانند $t$-استیودنت) کارایی ندارند. مهندسان اوبر از <b>روش بوت‌استرپ (Bootstrap)</b> استفاده می‌کنند: با نمونه‌گیری مکرر <b>با جایگزینی</b> از داده موجود، هزاران نمونه مصنوعی می‌سازند و از توزیع آماره‌های محاسبه‌شده، فاصله اطمینان تجربی به دست می‌آورند — بدون هیچ فرض توزیعی.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در روش بوت‌استرپ، هر نمونه $b$-ام با انتخاب تصادفی $N$ نقطه از داده اصلی $X$ با جایگزینی ساخته می‌شود:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$X^{*(b)} = \{x_1^{*(b)}, x_2^{*(b)}, \dots, x_N^{*(b)}\}, \quad x_i^{*(b)} \overset{iid}{\sim} \text{Uniform}(X)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    برای هر نمونه بوت‌استرپ، آماره مورد نظر (مانند میانگین) محاسبه می‌شود:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\hat{\theta}^{*(b)} = s\left(X^{*(b)}\right), \quad b = 1, 2, \dots, B$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    میانگین بوت‌استرپ به عنوان برآورد نهایی آماره:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\bar{\theta}^* = \frac{1}{B} \sum_{b=1}^{B} \hat{\theta}^{*(b)}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    خطای استاندارد بوت‌استرپ که عدم قطعیت آماره را کمّی می‌کند:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$SE_{boot} = \sqrt{\frac{1}{B-1} \sum_{b=1}^{B} \left( \hat{\theta}^{*(b)} - \bar{\theta}^* \right)^2}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    فاصله اطمینان $(1-\alpha)$ درصدی با روش Percentile (با استفاده از کوانتایل‌های تجربی توزیع بوت‌استرپ):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$CI_{boot} = \left[ \hat{\theta}^*_{(\alpha/2)}, \; \hat{\theta}^*_{(1-\alpha/2)} \right]$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: اوبر با اجرای بوت‌استرپ روی داده زمان انتظار، بدون نیاز به فرض نرمال بودن، فاصله اطمینان ۹۵٪ واقعی را محاسبه می‌کند و بر اساس آن، الگوریتم قیمت‌گذاری پویا و تخصیص راننده را کالیبره می‌نماید.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: تخمین تجربی توزیع نمونه‌گیری آماره و محاسبه فاصله اطمینان بدون فرض پارامتری
    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "تخمین فاصله اطمینان میانگین زمان انتظار در «اوبر» (Bootstrap From Scratch) 🚕", 
    "main_title"
)

display_academic_text(
    "در پلتفرم اوبر، داده‌های زمان انتظار مسافران به دلیل ترافیک و حوادث، دارای چولگی شدید به راست (Right-Skewed) "
    "هستند و از توزیع نرمال پیروی نمی‌کنند. بنابراین استفاده از فرمول‌های کلاسیک آماری (t-استیودنت) برای "
    "یافتن فاصله اطمینان (Confidence Interval) اشتباه است. مهندسان اوبر از <b>روش بوت‌استرپ (Bootstrap)</b> "
    "استفاده می‌کنند: با نمونه‌گیری مکرر با جایگذاری، هزاران نمونه مصنوعی می‌سازند و فاصله اطمینان واقعی "
    "را به صورت تجربی محاسبه می‌کنند.", 
    "paragraph"
)

display_academic_text("معادلات و مفاهیم ریاضی بوت‌استرپ:", "sub_title")

display_academic_text("۱. هر نمونه $b$-ام با انتخاب تصادفی $N$ نقطه از داده اصلی $X$ با جایگذاری ساخته می‌شود:", "paragraph")
display_academic_text(r"$$X^{*(b)} = \{x_1^{*(b)}, x_2^{*(b)}, \dots, x_N^{*(b)}\}, \quad x_i^{*(b)} \overset{iid}{\sim} \text{Uniform}(X)$$", "paragraph")

display_academic_text("۲. میانگین بوت‌استرپ به عنوان برآورد نهایی آماره:", "paragraph")
display_academic_text(r"$$\bar{\theta}^* = \frac{1}{B} \sum_{b=1}^{B} \hat{\theta}^{*(b)}$$", "paragraph")

display_academic_text("۳. خطای استاندارد بوت‌استرپ که عدم قطعیت آماره را کمّی می‌کند:", "paragraph")
display_academic_text(r"$$SE_{boot} = \sqrt{\frac{1}{B-1} \sum_{b=1}^{B} \left( \hat{\theta}^{*(b)} - \bar{\theta}^* \right)^2}$$", "paragraph")

display_academic_text("۴. فاصله اطمینان با روش درصدی (Percentile):", "paragraph")
display_academic_text(r"$$CI_{boot} = \left[ \hat{\theta}^*_{(\alpha/2)}, \; \hat{\theta}^*_{(1-\alpha/2)} \right]$$", "paragraph")

# ==============================================================================
# 2. SIMULATED UBER WAIT-TIME DATASET
# ==============================================================================
np.random.seed(42)
N = 200  # Limited sample size (observed trips in a specific zone)
# Log-normal distribution: median ~ 4 min, heavy right tail to simulate traffic jams
wait_times = np.random.lognormal(mean=np.log(4), sigma=0.6, size=N)
wait_times = np.clip(wait_times, 0.5, None)  # Minimum logical wait time is 0.5 min

# Store stats for display
stats_df = pd.DataFrame([{
    'Observed Sample Size': N,
    'Sample Mean (min)': f"{wait_times.mean():.3f}",
    'Sample Median (min)': f"{np.median(wait_times):.3f}",
    'Sample Std Dev (min)': f"{wait_times.std():.3f}",
    'Skewness (Right-Tailed)': f"{((wait_times - wait_times.mean())**3).mean() / wait_times.std()**3:.3f}"
}])

# ==============================================================================
# 3. BOOTSTRAP IMPLEMENTATION FROM SCRATCH
# ==============================================================================
def run_bootstrap(data, statistic_fn, B=2000, seed=42):
    """
    Runs B bootstrap iterations, computing the given statistic on each resample.
    Uses 'Sampling with Replacement'.
    """
    rng = np.random.default_rng(seed)
    n = len(data)
    boot_stats = np.zeros(B)
    
    for b in range(B):
        # Sample N indices WITH REPLACEMENT
        indices = rng.integers(0, n, size=n)
        resample = data[indices]
        # Calculate and store the statistic (e.g., mean) for this synthetic sample
        boot_stats[b] = statistic_fn(resample)
    
    return boot_stats

def bootstrap_ci(boot_stats, alpha=0.05):
    """Computes the percentile-based (1-alpha) confidence interval."""
    lower = np.percentile(boot_stats, 100 * (alpha / 2))
    upper = np.percentile(boot_stats, 100 * (1 - alpha / 2))
    return lower, upper

def bootstrap_se(boot_stats):
    """Computes the bootstrap Standard Error (Std of the bootstrap distribution)."""
    return float(np.std(boot_stats, ddof=1))

def bootstrap_bias(boot_stats, original_stat):
    """Computes the bootstrap estimate of bias."""
    return float(np.mean(boot_stats) - original_stat)

# ==============================================================================
# 4. RUN BOOTSTRAP ANALYSIS
# ==============================================================================
B = 2000 # Number of Bootstrap iterations
# Run for Mean
boot_means = run_bootstrap(wait_times, statistic_fn=np.mean, B=B, seed=42)
orig_mean = np.mean(wait_times)
mean_se = bootstrap_se(boot_means)
mean_ci = bootstrap_ci(boot_means, alpha=0.05)
mean_bias = bootstrap_bias(boot_means, orig_mean)

# Run for Median
boot_medians = run_bootstrap(wait_times, statistic_fn=np.median, B=B, seed=42)
orig_median = np.median(wait_times)
median_se = bootstrap_se(boot_medians)
median_ci = bootstrap_ci(boot_medians, alpha=0.05)
median_bias = bootstrap_bias(boot_medians, orig_median)

# Format Results for Academic Display
results_df = pd.DataFrame([
    {'Statistic Target': 'MEAN Wait Time', 'Original Value': f"{orig_mean:.3f}", 'Bootstrap Average': f"{np.mean(boot_means):.3f}", 'Est. Bias': f"{mean_bias:+.4f}", 'Standard Error': f"{mean_se:.4f}", '95% Confidence Interval': f"[{mean_ci[0]:.3f}, {mean_ci[1]:.3f}]"},
    {'Statistic Target': 'MEDIAN Wait Time', 'Original Value': f"{orig_median:.3f}", 'Bootstrap Average': f"{np.mean(boot_medians):.3f}", 'Est. Bias': f"{median_bias:+.4f}", 'Standard Error': f"{median_se:.4f}", '95% Confidence Interval': f"[{median_ci[0]:.3f}, {median_ci[1]:.3f}]"}
])

display_side_by_side_datasets(
    [stats_df, results_df], 
    ["آمار توصیفی داده‌های اولیه اوبر (شدیداً چوله به راست)", f"نتایج تحلیل بوت‌استرپ با {B} نمونه‌گیری مصنوعی (محاسبه SE و CI)"]
)
# ==============================================================================
# 5. CONVERGENCE ANALYSIS
# ==============================================================================
B_values = [10, 50, 100, 250, 500, 1000, 1500, 2000]
se_evolution = []

for b in B_values:
    bs = run_bootstrap(wait_times, statistic_fn=np.mean, B=b, seed=42)
    se_evolution.append(bootstrap_se(bs))

# ==============================================================================
# 6. VISUALIZATIONS AND CHARTS
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

#  CHART 1: Bootstrap Distribution with Confidence Interval 
axes[0].hist(
    boot_means, 
    bins=40, 
    color='#2980b9', # Academic Blue
    edgecolor='black',
    alpha=0.85, 
    density=True, 
    label='Bootstrap Mean Distribution'
)

# Overlay Normal Approximation KDE
kde_x = np.linspace(boot_means.min(), boot_means.max(), 200)
kde_y = (1 / (mean_se * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((kde_x - np.mean(boot_means)) / mean_se) ** 2)
axes[0].plot(kde_x, kde_y, color='#2c3e50', linewidth=3, linestyle='-', label='Approximate Normal Fit')

# Mark original sample mean
axes[0].axvline(
    orig_mean, color='#e74c3c', linewidth=2.5, linestyle='-',
    label=f'Original Sample Mean = {orig_mean:.3f}'
)

# Mark and Shade 95% CI boundaries
axes[0].axvline(mean_ci[0], color='#c0392b', linewidth=2.5, linestyle='--', label=f'95% CI Lower Bound = {mean_ci[0]:.3f}')
axes[0].axvline(mean_ci[1], color='#c0392b', linewidth=2.5, linestyle='--', label=f'95% CI Upper Bound = {mean_ci[1]:.3f}')
axes[0].axvspan(mean_ci[0], mean_ci[1], alpha=0.1, color='#c0392b')

axes[0].set_xlabel('Bootstrap Sample Mean Evaluation (minutes)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Probability Density', fontsize=12, fontweight='bold')
axes[0].set_title(
    f'1. Non-parametric Bootstrap Distribution of Mean Wait Time\n'
    f'(Iterations B = {B} | SE = {mean_se:.3f} | 95% CI = [{mean_ci[0]:.2f}, {mean_ci[1]:.2f}])',
    fontsize=13, fontweight='bold', pad=12
)
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='upper right')
axes[0].grid(True, linestyle='--', alpha=0.6)

#  CHART 2: Standard Error (SE) Convergence 
axes[1].plot(
    B_values, 
    se_evolution, 
    marker='o', 
    color='#2c3e50',
    linewidth=3, 
    markersize=9, 
    markerfacecolor='#2980b9',
    label='Calculated Bootstrap SE'
)

# Reference line at final Converged SE
final_se = se_evolution[-1]
axes[1].axhline(
    final_se, color='#e74c3c', linewidth=2.5, linestyle='--',
    label=f'Converged SE line (B=2000) = {final_se:.4f}'
)

# Annotate points
for b_val, se_val in zip(B_values, se_evolution):
    # Skip some annotations to prevent crowding
    if b_val in [10, 50, 250, 1000, 2000]:
        axes[1].annotate(
            f'{se_val:.4f}', 
            xy=(b_val, se_val),
            xytext=(0, 12), textcoords='offset points',
            ha='center', fontsize=10, fontweight='bold', color='#2c3e50'
        )

axes[1].set_xlabel('Number of Bootstrap Iterations / Resamplings ($B$)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Bootstrap Standard Error ($SE_{boot}$)', fontsize=12, fontweight='bold')
axes[1].set_title(
    '2. Convergence of Bootstrap Standard Error\nAs computationally intensive resamplings increase',
    fontsize=13, fontweight='bold', pad=12
)
axes[1].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=11, loc='upper right')
axes[1].grid(True, linestyle='--', alpha=0.6)
axes[1].set_xscale('log')
axes[1].set_xticks(B_values)
axes[1].set_xticklabels([str(b) for b in B_values], fontweight='bold')

# Clean up axes
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 7: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    "<b>نتیجه استراتژیک:</b> بررسی گراف ۱ قدرت شگفت‌انگیز قضیه حد مرکزی (CLT) در ترکیب با بوت‌استرپ را نشان می‌دهد. "
    "با وجود اینکه داده‌های اولیه اوبر شدیداً چوله و غیرنرمال بودند (احتمال وجود ترافیک‌های ۳۰ دقیقه‌ای)، "
    "توزیع میانگینِ ۲۰۰۰ نمونه مصنوعی، کاملاً شکل زنگوله‌ای و نرمال (هیستوگرام آبی) به خود گرفته است. "
    "مدیران اوبر حالا می‌توانند با <b>۹۵ درصد اطمینان</b> بگویند که میانگین واقعی زمان انتظار مسافران در این شهر، "
    f"رقم دقیقی بین <b>{mean_ci[0]:.2f} و {mean_ci[1]:.2f} دقیقه</b> است. گراف ۲ نشان می‌دهد که برای رسیدن "
    "به این دقت، باید کامپیوتر را وادار کنیم حداقل ۵۰۰ الی ۱۰۰۰ بار نمونه‌گیری را تکرار کند تا خطای "
    "استاندارد (SE) به ثبات و همگرایی (خط‌چین قرمز) برسد."
)
display_academic_text(insight_conclusion, "insight")
```

<ul style="font-size:15pt; line-height:250%; color:#2B2D42;">
<li>یک نمونه شبیه‌سازی‌شده از زمان انتظار ۲۰۰ مسافر <b>اوبر</b> با توزیع <b>لاگ-نرمال</b> (چوله به راست) تولید شد؛ این توزیع به خوبی رفتار واقعی زمان انتظار را منعکس می‌کند که به هیچ وجه نرمال نیست و روش‌های پارامتری کلاسیک را بی‌اعتبار می‌سازد.</li>
<li>پیاده‌سازی کامل روش بوت‌استرپ از صفر (From Scratch):
  <ul>
    <li><b>نمونه‌گیری با جایگزینی:</b> تابع <code>bootstrap_sample</code> با استفاده از <code>rng.integers</code>، ایندکس‌هایی را از بازه $[0, N)$ انتخاب می‌کند که امکان تکرار یک داده در یک نمونه وجود دارد — این دقیقاً همان مفهوم «با جایگزینی» (With Replacement) است.</li>
    <li><b>حلقه بوت‌استرپ:</b> تابع <code>run_bootstrap</code> به تعداد $B=2000$ بار، یک نمونه جدید می‌سازد و آماره مورد نظر (میانگین یا میانه) را روی آن محاسبه می‌کند؛ خروجی آرایه‌ای از ۲۰۰۰ برآورد است.</li>
    <li><b>فاصله اطمینان Percentile:</b> تابع <code>bootstrap_ci</code> با استفاده از <code>np.percentile</code>، کوانتایل‌های ۲.۵٪ و ۹۷.۵٪ توزیع بوت‌استرپ را پیدا کرده و فاصله اطمینان ۹۵٪ را بدون هیچ فرض توزیعی می‌سازد.</li>
    <li><b>خطای استاندارد و سوگیری:</b> توابع <code>bootstrap_se</code> و <code>bootstrap_bias</code> به ترتیب انحراف معیار برآوردها و اختلاف میانگین بوت‌استرپ با آماره اصلی را محاسبه می‌کنند.</li>
  </ul>
</li>
<li>تفسیر خروجی و نمودارها:
  <ul>
    <li>نمودار سمت چپ (Bootstrap Distribution): هیستوگرام ۲۰۰۰ میانگین بوت‌استرپ را نشان می‌دهد. برخلاف داده اصلی که چوله به راست بود، توزیع میانگین بوت‌استرپ به شکل زنگوله‌ای و تقریباً نرمال است — این تجلی مستقیم <b>قضیه حد مرکزی</b> است. خط آبی عمودی میانگین نمونه اصلی، خطوط نارنجی کران‌های فاصله اطمینان ۹۵٪ و ناحیه سایه‌زده بازه اطمینان را نشان می‌دهند.</li>
    <li>نمودار سمت راست (SE Convergence): روند کاهش و همگرایی خطای استاندارد بوت‌استرپ را با افزایش تعداد تکرارها $B$ نمایش می‌دهد. در مقیاس لگاریتمی محور افقی، مشخص است که از حدود $B=500$ به بعد، خطای استاندارد تقریباً ثابت می‌ماند؛ این به مهندسان اوبر می‌گوید برای تخمین پایدار، $B=1000$ تا $2000$ کافی است و نیازی به تکرارهای بیشتر نیست.</li>

</li>
<li>نکته کلیدی: قدرت اصلی بوت‌استرپ در این است که بدون هیچ فرضی درباره شکل توزیع اصلی (نرمال، نمایی، دوجمله‌ای و...)، صرفاً با تکیه بر خود داده‌های مشاهده‌شده و قانون اعداد بزرگ، فاصله اطمینان معتبری ارائه می‌دهد. این ویژگی آن را برای داده‌های دنیای واقعی مانند زمان انتظار اوبر، درآمد رانندگان، یا تاخیر پروازها که هیچ‌کدام توزیع نرمال ندارند، بی‌نهایت ارزشمند می‌سازد.</li>
