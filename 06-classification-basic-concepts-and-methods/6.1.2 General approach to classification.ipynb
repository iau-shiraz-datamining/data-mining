  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: فرایند دومرحله‌ای تشخیص ایمیل هرزنامه در «میل‌چیمپ» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    سرویس بازاریابی ایمیلی <b>میل‌چیمپ</b> روزانه میلیاردها ایمیل را از طرف کسب‌وکارهای مختلف ارسال می‌کند و باید پیش از تحویل، هر پیام را در یکی از دو دستهٔ «هرزنامه» یا «سالم» قرار دهد. این فرایند دقیقاً از دو مرحلهٔ مجزا تبعیت می‌کند. در <b>مرحلهٔ یادگیری</b>، الگوریتم روی هزاران ایمیل قدیمی که برچسب واقعی آن‌ها (هرزنامه یا سالم) از قبل توسط کاربران یا تیم مدیریت مشخص شده، آموزش می‌بیند و یک مدل طبقه‌بند می‌سازد. چون برچسب واقعی هر رکورد از پیش در اختیار است، این فرایند نمونهٔ کامل یادگیری نظارت‌شده است؛ برخلاف خوشه‌بندی مشتریان که در آن نه برچسب و نه تعداد گروه‌ها از قبل معلوم نیست. در <b>مرحلهٔ طبقه‌بندی</b>، مدل ساخته‌شده روی ایمیل‌های جدید و ناشناخته اعمال می‌شود، اما پیش از آن، دقت مدل باید روی یک مجموعهٔ آزمون مستقل (که در ساخت مدل نقشی نداشته) سنجیده شود تا از خوش‌بینی کاذب ناشی از سنجش با داده‌های آموزشی جلوگیری شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی حاکم بر رویکرد کلی طبقه‌بندی:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. مجموعهٔ آموزشی با ویژگی برچسب کلاس اسمی (بدون ترتیب) از میان $k$ کلاس ممکن:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$D_{train} = \{(x_i, y_i)\}_{i=1}^{n}, \qquad y_i \in \{c_1, c_2, \dots, c_k\}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. مدل طبقه‌بند به عنوان تابعی که با کمینه‌سازی ریسک تجربی روی داده‌های آموزشی ساخته می‌شود:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$f^{*} = \arg\min_{f} \frac{1}{n} \sum_{i=1}^{n} \mathcal{L}\big(f(x_i), y_i\big)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. دقت طبقه‌بند بر پایهٔ مجموعهٔ آزمون مستقل $D_{test}$، با تابع نشانگر $\mathbb{1}[\cdot]$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$Accuracy = \frac{1}{|D_{test}|} \sum_{j=1}^{|D_{test}|} \mathbb{1}\big[f(x_j) = y_j\big]$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. شکاف تعمیم (Generalization Gap) به عنوان معیار خوش‌بینی کاذب ناشی از سنجش مدل با داده‌های آموزشی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$Gap = Accuracy_{train} - Accuracy_{test}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: میل‌چیمپ تنها زمانی مدل تشخیص هرزنامه را روی جریان زندهٔ ایمیل‌ها فعال می‌کند که $Accuracy_{test}$ در سطحی قابل‌قبول باشد و شکاف $Gap$ بین دقت آموزش و آزمون بیش از حد بزرگ نباشد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: تفکیک قابل‌اعتماد ایمیل‌های سالم از هرزنامه پیش از ورود به صندوق مشتریان
    </div>
  </p>

</div>

```python
# 1. SYNTHETIC MAILCHIMP EMAIL DATASET GENERATION (SPAM vs NOT-SPAM)
random_seed = 42
rng = np.random.default_rng(random_seed)

def generate_email_data(n_samples):
    word_freq_free = rng.exponential(scale=1.5, size=n_samples)
    num_links = rng.poisson(lam=2.0, size=n_samples)
    num_exclamations = rng.poisson(lam=1.5, size=n_samples)
    sender_reputation = rng.uniform(0.0, 1.0, n_samples)
    email_length = rng.normal(180, 60, n_samples).clip(20, 500)

    spam_signal = (1.1 * word_freq_free + 0.6 * num_links + 0.5 * num_exclamations - 3.0 * sender_reputation - 0.002 * email_length)
    prob_spam = 1 / (1 + np.exp(-(spam_signal - spam_signal.mean()) / spam_signal.std()))
    label = (rng.random(n_samples) < prob_spam).astype(int)

    return pd.DataFrame({
        'word_freq_free': word_freq_free, 'num_links': num_links, 
        'num_exclamations': num_exclamations, 'sender_reputation': sender_reputation, 
        'email_length': email_length, 'label': label
    })

feature_cols = ['word_freq_free', 'num_links', 'num_exclamations', 'sender_reputation', 'email_length']
email_df = generate_email_data(800)

feat_mean, feat_std = email_df[feature_cols].mean(), email_df[feature_cols].std()
email_df[feature_cols] = (email_df[feature_cols] - feat_mean) / feat_std

shuffle_idx = rng.permutation(len(email_df))
split_point = int(len(email_df) * 0.75)
train_df = email_df.iloc[shuffle_idx[:split_point]].reset_index(drop=True)
test_df = email_df.iloc[shuffle_idx[split_point:]].reset_index(drop=True)

X_train, y_train = train_df[feature_cols].values, train_df['label'].values
X_test, y_test = test_df[feature_cols].values, test_df['label'].values

# ==============================================================================
# 2. LEARNING PHASE: BUILD A K-NEAREST-NEIGHBORS CLASSIFIER FROM SCRATCH
# ==============================================================================
def euclidean_distance_matrix(A, B):
    A_sq = np.sum(A ** 2, axis=1).reshape(-1, 1)
    B_sq = np.sum(B ** 2, axis=1).reshape(1, -1)
    return np.sqrt(np.clip(A_sq + B_sq - 2 * (A @ B.T), a_min=0, a_max=None))

def knn_classify(X_query, X_reference, y_reference, k):
    distances = euclidean_distance_matrix(X_query, X_reference)
    nearest_idx = np.argsort(distances, axis=1)[:, :k]
    return np.array([np.bincount(y_reference[idx]).argmax() for idx in nearest_idx])

def compute_error_rate(y_true, y_pred):
    # Using ERROR RATE (1 - accuracy) for standard scientific plotting
    return np.mean(y_true != y_pred)

# ==============================================================================
# 3. EVALUATE LEARNING vs CLASSIFICATION ACROSS MODEL COMPLEXITY (k VALUES)
# ==============================================================================
k_values = list(range(1, 42, 2)) # 1, 3, 5 ... 41
train_errors, test_errors = [], []

for k in k_values:
    train_pred = knn_classify(X_train, X_train, y_train, k)
    test_pred = knn_classify(X_test, X_train, y_train, k)
    train_errors.append(compute_error_rate(y_train, train_pred))
    test_errors.append(compute_error_rate(y_test, test_pred))

# Best model minimizes TEST ERROR
best_k_idx = int(np.argmin(test_errors))
best_k = k_values[best_k_idx]
generalization_gap = test_errors[best_k_idx] - train_errors[best_k_idx]

final_test_pred = knn_classify(X_test, X_train, y_train, best_k)

def confusion_matrix_scratch(y_true, y_pred):
    tp = int(np.sum((y_true == 1) & (y_pred == 1)))
    tn = int(np.sum((y_true == 0) & (y_pred == 0)))
    fp = int(np.sum((y_true == 0) & (y_pred == 1)))
    fn = int(np.sum((y_true == 1) & (y_pred == 0)))
    return np.array([[tn, fp], [fn, tp]]), tp, tn, fp, fn

cm, tp, tn, fp, fn = confusion_matrix_scratch(y_test, final_test_pred)
# =============================================================================
# 4. LARGE, BALANCED SIDE-BY-SIDE VISUALIZATIONS (ACADEMIC BLUE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(19, 7.5), dpi=135)

#  CHART 1: The Classic Bias-Variance & Generalization Gap Curve 
# We plot Error Rate and INVERT the X-axis so complexity grows from Left to Right.

axes[0].plot(
    k_values, train_errors, 
    color='#2980b9', marker='o', linewidth=3.5, markersize=7, 
    label='Training Error (Memorization)'
)
axes[0].plot(
    k_values, test_errors, 
    color='#e74c3c', marker='s', linewidth=3.5, markersize=7, 
    label='Test Error (Real-World Generalization)'
)

# Fill the Generalization Gap specifically in the Overfitting zone (where test error > train error)
axes[0].fill_between(
    k_values, train_errors, test_errors, 
    where=(np.array(test_errors) > np.array(train_errors)),
    interpolate=True, color='#e74c3c', alpha=0.15, label='The Generalization Gap'
)

# Crucial: Invert X-axis so complexity increases to the right!
axes[0].invert_xaxis()

axes[0].axvline(best_k, color='#16a085', linestyle='--', linewidth=2.5, label=f'Optimal Capacity (Sweet Spot: k={best_k})')

# Academic Annotations for the 3 distinct zones
axes[0].annotate(
    "High Bias (Underfitting)\nModel is too simple to capture patterns.",
    xy=(35, test_errors[-3]), xytext=(38, 0.22),
    fontsize=11, fontweight='bold', color='#2c3e50',
    arrowprops=dict(arrowstyle="->", color="#34495e", lw=1.5),
    bbox=dict(boxstyle="round,pad=0.3", fc="#f8f9fa", ec="#34495e", lw=1.2)
)

axes[0].annotate(
    "High Variance (Overfitting)\nTrain Error drops to 0.0 (Perfect Memorization)\nbut Test Error violently spikes upward!",
    xy=(1, test_errors[0]), xytext=(9, 0.17),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title('1. Scientific Generalization Gap: Error Rate vs Model Complexity', fontsize=14, fontweight='bold', pad=12)
# Updated X-axis label to reflect the mathematical reality
axes[0].set_xlabel('Model Complexity Increases ⟶\n(Decreasing Number of Neighbors $k$)', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Error Rate (Lower is Better)', fontsize=13, fontweight='bold')
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='upper center')
axes[0].grid(True, linestyle='--', alpha=0.5)


#  CHART 2: Confusion Matrix Heatmap 
sns.heatmap(
    cm, annot=True, fmt='d', cmap='Blues', cbar=False, ax=axes[1],
    xticklabels=['Predicted: Safe Mail', 'Predicted: SPAM'],
    yticklabels=['Actual: Safe Mail', 'Actual: SPAM'],
    annot_kws={'fontsize': 18, 'fontweight': 'bold'}, linewidths=1.5, linecolor='white'
)

axes[1].annotate(
    f"Optimal Deployment (k={best_k}):\nBy selecting the 'Sweet Spot' from the left chart,\nthe algorithm correctly caught {tp} spam emails\nwith only {fp} False Positives.",
    xy=(1.5, 0.5), xytext=(1.8, 0.8),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[1].set_title('2. Final Classification Phase Matrix (On Unseen Test Data)', fontsize=14, fontweight='bold', pad=12)
axes[1].set_xlabel('AI Model Predictions', fontsize=13, fontweight='bold')
axes[1].set_ylabel('Actual Ground Truth', fontsize=13, fontweight='bold')

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (آناتومی یک گراف علمی دقیق):</b> با ارتقای گراف اول، ما اکنون یک تصویرِ کاملاً استاندارد و دانشگاهی از "
    r"<b>شکاف تعمیم (Generalization Gap)</b> داریم. در محور افقی، هرچه به سمت راست می‌رویم، پیچیدگیِ هوش مصنوعی افزایش می‌یابد (k کوچکتر می‌شود). "
    r"۱. در <b>سمتِ چپِ گراف</b>، مدل بسیار ساده‌لوح است (Underfitting). خطای آن هم روی داده‌های آموزشی و هم روی داده‌های تست بالاست. "
    r"۲. در <b>سمتِ راستِ گراف</b>، مدل بیش از حد پیچیده شده است (Overfitting). ماشین در این نقطه تمامِ داده‌های آموزشی را حفظ کرده است (خطِ آبی به خطای صفر رسیده است). "
    r"اما فاجعه در خطِ قرمز (داده‌های تست واقعی) نمایان می‌شود؛ خطا به شدت بالا رفته است و فضای سایه‌زدهٔ قرمز، میزانِ توهم و حفظ‌کردنِ ماشین را نشان می‌دهد. "
    r"۳. وظیفهٔ یک مهندسِ داده، یافتنِ <b>نقطهٔ بهینه (Sweet Spot)</b> است. خط‌چینِ عمودیِ فیروزه‌ای دقیقاً جایی را نشان می‌دهد که ماشین، کمترین خطای ممکن را در دنیای واقعی دارد (k=11)."
)
display_academic_text(insight_conclusion, "insight")
```

</p>
<ul style="font-size:15pt; line-height:250%; color:#2B2D42;">
<li>ابتدا با تابع <code>generate_email_data</code> پنج ویژگی توصیفی هر ایمیل (تکرار کلمه Free، تعداد لینک، تعداد علامت تعجب، اعتبار فرستنده و طول ایمیل) شبیه‌سازی شد و برچسب کلاس اسمی <code>label</code> (هرزنامه یا سالم) بدون هیچ ترتیب عددی میان دو مقدار تولید گردید.</li>
<li>تفکیک داده به مجموعهٔ آموزشی و مجموعهٔ آزمون کاملاً مستقل، با جابه‌جایی تصادفی اندیس‌ها و بدون هیچ همپوشانی انجام شد؛ این دقیقاً همان مفهومی است که در متن به عنوان شرط لازم برای سنجش صادقانهٔ دقت مدل ذکر شده است.</li>
<li>مرحلهٔ یادگیری (Learning Phase) با الگوریتم k نزدیک‌ترین همسایه (k-NN) از صفر پیاده‌سازی شد:
  <ul>
    <li><b>تابع <code>euclidean_distance_matrix</code>:</b> فاصلهٔ اقلیدسی میان تمام جفت نمونه‌ها را با عملیات ماتریسی بهینه محاسبه می‌کند.</li>
    <li><b>تابع <code>knn_classify</code>:</b> برای هر نمونهٔ پرسش، k همسایهٔ نزدیک را از داده‌های آموزشی می‌یابد و برچسب اکثریت را به عنوان پیش‌بینی نهایی برمی‌گرداند؛ این دقیقاً مرحلهٔ طبقه‌بندی (Classification Phase) است که مدل ساخته‌شده را روی داده جدید اعمال می‌کند.</li>
  </ul>
</li>
<li>تابع <code>compute_accuracy</code> عیناً همان فرمول ارائه‌شده در متن را پیاده‌سازی می‌کند: نسبت رکوردهایی که برچسب پیش‌بینی‌شده‌شان با برچسب واقعی مطابقت دارد.</li>
<li>برای اثبات ادعای مهم متن (خوش‌بینی کاذب ارزیابی با داده‌های آموزشی)، مدل با مقادیر مختلف $k$ ساخته شد و دقت آن هم روی داده آموزشی و هم روی داده آزمون محاسبه گردید. در $k=1$ هر نمونهٔ آموزشی نزدیک‌ترین همسایهٔ خودش است، بنابراین دقت آموزشی تقریباً به ۱۰۰٪ می‌رسد در حالی که این عدد هیچ تضمینی برای عملکرد واقعی مدل روی ایمیل‌های ناشناخته نیست.</li>
<li>بهترین مقدار $k$ صرفاً بر اساس دقت مجموعهٔ آزمون (نه آموزش) انتخاب شد و شکاف تعمیم (Generalization Gap) میان دقت آموزش و آزمون در همان نقطه محاسبه گردید.</li>
<li>تحلیل نمودارها:
  <ul>
    <li>نمودار اول (دقت آموزش در برابر آزمون بر حسب پیچیدگی مدل): به وضوح نشان می‌دهد دقت آموزشی در $k$ کوچک بسیار خوش‌بینانه و غیرقابل‌اتکاست، در حالی که دقت آزمون معیار واقعی و صادقانهٔ عملکرد مدل بر داده‌های جدید است؛ خط‌چین سبز نقطهٔ انتخاب نهایی $k$ را مشخص می‌کند.</li>
    <li>نمودار دوم (ماتریس درهم‌ریختگی): عملکرد نهایی مدل انتخاب‌شده را روی مجموعهٔ آزمون مستقل به تفکیک ایمیل‌های هرزنامه و سالم نمایش می‌دهد و مبنای تصمیم برای فعال‌سازی مدل روی جریان زندهٔ ایمیل‌ها را فراهم می‌کند.</li>
  </ul>
</li>
</ul>
