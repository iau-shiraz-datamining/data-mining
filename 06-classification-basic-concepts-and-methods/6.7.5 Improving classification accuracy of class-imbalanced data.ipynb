  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: مدیریت تراکنش‌های نامتعادل و تقلب‌های نادر در «استرایپ» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در درگاه پرداخت <b>استرایپ</b>، بیش از ۹۹.۸٪ تراکنش‌ها کاملاً سالم و قانونی هستند و کمتر از ۰.۲٪ موارد به کلاهبرداری و سرقت کارت تعلق دارند. اگر یک مدل استاندارد یادگیری ماشین بدون اصلاح روی این داده آموزش داده شود، با پیش‌بینی مداوم «تراکنش سالم» به دقت ظاهری ۹۹.۸٪ می‌رسد اما عملاً ۱۰۰٪ تقلب‌ها را از دست می‌دهد. برای حل این معضل، تیم امنیت داده استرایپ از تکنیک <b>تولید مصنوعی نمونه‌های اقلیت (SMOTE)</b> به همراه تنظیم وزن در تابع زیان استفاده می‌کند تا با ایجاد نقاط مصنوعی واقع‌گرایانه در امتداد بردارهای همسایگی، مرز تصمیم مدل را نسبت به کلاس اقلیت حساس و متوازن سازد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. نسبت عدم توازن کلاسی (Imbalance Ratio):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$IR = \frac{N_{majority}}{N_{minority}}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. فرمول درون‌یابی خطی الگوریتم SMOTE برای تولید نمونه‌های مصنوعی جدید:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$x_{new} = x_i + \lambda \cdot (x_{zi} - x_i), \quad \lambda \sim \text{Uniform}(0, 1)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    که در آن $x_i$ یک نمونه از کلاس اقلیت و $x_{zi}$ یکی از $k$ نزدیک‌ترین همسایه‌های آن در همان کلاس اقلیت است.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. تابع زیان آنتروپی متقاطع وزن‌دار برای جریمه سنگین‌تر خطای کلاس اقلیت:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\mathcal{L} = -\frac{1}{N} \sum_{i=1}^{N} \left[ w_1 y_i \log(\hat{p}_i) + w_0 (1 - y_i) \log(1 - \hat{p}_i) \right], \quad w_1 = \frac{N}{2 N_{minority}}, \; w_0 = \frac{N}{2 N_{majority}}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. معیار میانگین هندسی (Geometric Mean) برای ارزیابی همزمان هر دو کلاس فارغ از عدم توازن:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{G-Mean} = \sqrt{\text{Sensitivity} \times \text{Specificity}} = \sqrt{\frac{TP}{TP + FN} \times \frac{TN}{TN + FP}}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: استرایپ با متوازن‌سازی توزیع کلاس‌ها از طریق SMOTE، نرخ یادآوری (Recall) تقلب‌ها را از کمتر از ۲۰٪ به بیش از ۹۴٪ می‌رساند بدون آنکه نرخ هشدارهای اشتباه به شکل غیرقابل قبولی افزایش یابد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: متعادل‌سازی توزیع کلاس‌ها و بهبود چشمگیر توانایی مدل در کشف رخدادهای نادر
    </div>
  </p>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "مدیریت تقلب‌های نادر با الگوریتم SMOTE در «استرایپ» (100% From Scratch) 💳", 
    "main_title"
)

display_academic_text(
    "در درگاه‌های پرداخت، تقلب‌ها (Fraud) بسیار نادر هستند. اگر مدلی روی این داده‌های نامتعادل آموزش ببیند، "
    "برای بهینه‌سازیِ خطای خود یاد می‌گیرد که همیشه بگوید «تراکنش سالم است!» و در نتیجه هیچ تقلبی را شکار نمی‌کند. "
    "ما با استفاده از تکنیک <b>SMOTE (تولید مصنوعی نمونه‌های اقلیت)</b> از پایه، در فضای هندسیِ بینِ تقلب‌هایِ موجود، "
    "داده‌های جعلی اما واقع‌گرایانه تولید می‌کنیم تا کفه ترازو برای آموزش مدل متعادل شود.", 
    "paragraph"
)

display_academic_text("معادلات پایه در تولید نمونه‌های مصنوعی (SMOTE):", "sub_title")

display_academic_text("۱. نسبت عدم توازن کلاسی (Imbalance Ratio):", "paragraph")
display_academic_text(r"$$ IR = \frac{N_{\text{majority}}}{N_{\text{minority}}} $$", "paragraph")

display_academic_text("۲. فرمول درون‌یابی خطی الگوریتم SMOTE برای تولید نمونه مصنوعی جدید ($x_{new}$) بین نمونه $x_i$ و همسایه آن $x_{zi}$:", "paragraph")
display_academic_text(r"$$ x_{new} = x_i + \lambda \cdot (x_{zi} - x_i), \quad \lambda \sim \text{Uniform}(0, 1) $$", "paragraph")

display_academic_text("۳. معیار میانگین هندسی (G-Mean) برای ارزیابی عادلانه هر دو کلاس:", "paragraph")
display_academic_text(r"$$ \text{G-Mean} = \sqrt{\text{Sensitivity} \times \text{Specificity}} = \sqrt{\frac{TP}{TP + FN} \times \frac{TN}{TN + FP}} $$", "paragraph")

# ==============================================================================
# 2. CREATE AN ARTIFICIALLY HIGHLY IMBALANCED DATASET FROM SCRATCH
# ==============================================================================
np.random.seed(42)

# Generate Synthetic Imbalanced Dataset (2D for easy visualization)
# Majority Class: Safe Transactions (Label 0)
n_majority = 250
X_maj = np.random.multivariate_normal(mean=[3.0, 3.0], cov=[[1.5, 0.5], [0.5, 1.5]], size=n_majority)
y_maj = np.zeros(n_majority)

# Minority Class: Fraudulent Transactions (Label 1) - ONLY 10 SAMPLES!
n_minority = 10
X_min = np.random.multivariate_normal(mean=[6.0, 6.0], cov=[[0.8, -0.3], [-0.3, 0.8]], size=n_minority)
y_min = np.ones(n_minority)

X_all = np.vstack([X_maj, X_min])
y_all = np.concatenate([y_maj, y_min])

# Manual Train/Test Split (Stratified equivalent from scratch)
def stratified_split(X, y, test_size=0.3):
    indices_0 = np.where(y == 0)[0]
    indices_1 = np.where(y == 1)[0]
    
    np.random.shuffle(indices_0)
    np.random.shuffle(indices_1)
    
    test_len_0 = int(len(indices_0) * test_size)
    test_len_1 = int(len(indices_1) * test_size)
    
    test_idx = np.concatenate([indices_0[:test_len_0], indices_1[:test_len_1]])
    train_idx = np.concatenate([indices_0[test_len_0:], indices_1[test_len_1:]])
    
    return X[train_idx], X[test_idx], y[train_idx], y[test_idx]

X_train, X_test, y_train, y_test = stratified_split(X_all, y_all, test_size=0.3)

# ==============================================================================
# 3. SMOTE & LOGISTIC REGRESSION IMPLEMENTED FROM SCRATCH
# ==============================================================================
class SMOTEScratch:
    """Synthetic Minority Over-sampling Technique (SMOTE) implemented from scratch."""
    def __init__(self, k_neighbors=3, random_state=42):
        self.k_neighbors = k_neighbors
        self.rng = np.random.default_rng(random_state)

    def _euclidean_distance(self, a, b):
        return np.sqrt(np.sum((a - b) ** 2, axis=1))

    def fit_resample(self, X, y):
        minority_class = 1
        majority_class = 0
        
        X_min = X[y == minority_class]
        X_maj = X[y == majority_class]
        
        n_majority = len(X_maj)
        n_minority = len(X_min)
        # Calculate how many points we need to balance the dataset
        n_samples_to_generate = n_majority - n_minority

        if n_samples_to_generate <= 0 or n_minority < 2:
            return X.copy(), y.copy(), np.array([])

        k = min(self.k_neighbors, n_minority - 1)
        synthetic_samples = []

        for _ in range(n_samples_to_generate):
            # 1. Pick a random minority sample
            idx = self.rng.integers(0, n_minority)
            sample = X_min[idx]

            # 2. Find k-nearest neighbors within the minority class
            dists = self._euclidean_distance(X_min, sample)
            # Exclude the sample itself (index 0 in sorted argsort)
            neighbor_indices = np.argsort(dists)[1:k + 1]

            # 3. Randomly select one neighbor among the k-NN
            chosen_neighbor_idx = self.rng.choice(neighbor_indices)
            neighbor = X_min[chosen_neighbor_idx]

            # 4. Synthesize point via linear interpolation (The core of SMOTE)
            lam = self.rng.uniform(0.0, 1.0)
            synthetic_point = sample + lam * (neighbor - sample)
            synthetic_samples.append(synthetic_point)

        synthetic_samples = np.array(synthetic_samples)
        
        X_resampled = np.vstack([X, synthetic_samples])
        y_resampled = np.hstack([y, np.full(len(synthetic_samples), minority_class)])
        
        return X_resampled, y_resampled, synthetic_samples

class LogisticRegressionScratch:
    """Logistic Classifier trained via gradient descent from scratch."""
    def __init__(self, lr=0.1, n_epochs=500):
        self.lr = lr
        self.n_epochs = n_epochs
        self.weights = None
        self.bias = 0.0

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0.0

        for _ in range(self.n_epochs):
            z = np.dot(X, self.weights) + self.bias
            probs = 1.0 / (1.0 + np.exp(-np.clip(z, -25, 25)))
            
            dw = (1.0 / n_samples) * np.dot(X.T, (probs - y))
            db = (1.0 / n_samples) * np.sum(probs - y)
            
            self.weights -= self.lr * dw
            self.bias -= self.lr * db
        return self

    def predict(self, X):
        z = np.dot(X, self.weights) + self.bias
        probs = 1.0 / (1.0 + np.exp(-np.clip(z, -25, 25)))
        return (probs >= 0.5).astype(int)

# Run SMOTE to balance the dataset
smote = SMOTEScratch(k_neighbors=2, random_state=42)
X_train_res, y_train_res, syn_points = smote.fit_resample(X_train, y_train)

# ==============================================================================
# 4. TRAIN MODELS AND COMPUTE METRICS
# ==============================================================================
def evaluate_predictions_scratch(y_true, y_pred):
    """Calculates Confusion Matrix and Metrics from scratch."""
    TP = int(np.sum((y_pred == 1) & (y_true == 1)))
    TN = int(np.sum((y_pred == 0) & (y_true == 0)))
    FP = int(np.sum((y_pred == 1) & (y_true == 0)))
    FN = int(np.sum((y_pred == 0) & (y_true == 1)))

    accuracy = (TP + TN) / len(y_true)
    precision = TP / (TP + FP) if (TP + FP) > 0 else 0.0
    recall = TP / (TP + FN) if (TP + FN) > 0 else 0.0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0
    
    specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
    g_mean = np.sqrt(recall * specificity)

    return {'Recall': recall, 'Precision': precision, 'F1-Score': f1, 'G-Mean': g_mean, 'Accuracy': accuracy}

# Model 1: Trained on Raw Imbalanced Data
model_imbalanced = LogisticRegressionScratch(lr=0.1, n_epochs=1000)
model_imbalanced.fit(X_train, y_train)
preds_raw = model_imbalanced.predict(X_test)
metrics_raw = evaluate_predictions_scratch(y_test, preds_raw)

# Model 2: Trained on SMOTE-Balanced Data
model_smote = LogisticRegressionScratch(lr=0.1, n_epochs=1000)
model_smote.fit(X_train_res, y_train_res)
preds_smote = model_smote.predict(X_test)
metrics_smote = evaluate_predictions_scratch(y_test, preds_smote)

# Present the evaluation in a DataFrame using the Academic Blue palette
df_eval = pd.DataFrame([
    {
        'Metric': 'Recall (Sensitivity)', 
        'Imbalanced Model': f"{metrics_raw['Recall']*100:.1f}%", 
        'SMOTE Model': f"{metrics_smote['Recall']*100:.1f}%",
        'Improvement': f"+{(metrics_smote['Recall'] - metrics_raw['Recall'])*100:.1f}%"
    },
    {
        'Metric': 'F1-Score', 
        'Imbalanced Model': f"{metrics_raw['F1-Score']*100:.1f}%", 
        'SMOTE Model': f"{metrics_smote['F1-Score']*100:.1f}%",
        'Improvement': f"+{(metrics_smote['F1-Score'] - metrics_raw['F1-Score'])*100:.1f}%"
    },
    {
        'Metric': 'G-Mean', 
        'Imbalanced Model': f"{metrics_raw['G-Mean']*100:.1f}%", 
        'SMOTE Model': f"{metrics_smote['G-Mean']*100:.1f}%",
        'Improvement': f"+{(metrics_smote['G-Mean'] - metrics_raw['G-Mean'])*100:.1f}%"
    }
])

display_side_by_side_datasets(
    [df_eval], 
    ["ارزیابی توانایی مدل در کشف تقلب‌های نادر (قبل و بعد از تعادل‌سازی با SMOTE)"]
)

# ==============================================================================
# 5. VISUALIZATIONS AND CHARTS (ACADEMIC BLUE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

#  CHART 1: Geometric Feature Space & SMOTE Synthesized Points 
axes[0].scatter(
    X_train[y_train == 0, 0], X_train[y_train == 0, 1],
    c='#bdc3c7', s=55, edgecolors='black', label=f'Original Safe Trans. (n={np.sum(y_train == 0)})', alpha=0.8
)
axes[0].scatter(
    X_train[y_train == 1, 0], X_train[y_train == 1, 1],
    c='#e74c3c', s=130, edgecolors='black', marker='o', linewidth=1.5, label=f'Original Fraud Trans. (n={np.sum(y_train == 1)})', zorder=4
)
axes[0].scatter(
    syn_points[:, 0], syn_points[:, 1],
    c='#3498db', s=80, edgecolors='black', marker='^', alpha=0.9, label=f'SMOTE Synthetic Points (n={len(syn_points)})', zorder=5
)

# Draw decision boundary for Imbalanced Model
x_vals = np.linspace(X_all[:, 0].min() - 1, X_all[:, 0].max() + 1, 100)
y_line_imb = -(model_imbalanced.weights[0] * x_vals + model_imbalanced.bias) / model_imbalanced.weights[1]
axes[0].plot(x_vals, y_line_imb, color='#7f8c8d', linestyle=':', linewidth=3, label='Imbalanced Decision Boundary')

# Draw decision boundary for SMOTE Model
y_line_smote = -(model_smote.weights[0] * x_vals + model_smote.bias) / model_smote.weights[1]
axes[0].plot(x_vals, y_line_smote, color='#2c3e50', linestyle='-', linewidth=3, label='SMOTE Decision Boundary')

axes[0].set_ylim(X_all[:, 1].min() - 1, X_all[:, 1].max() + 1)
axes[0].set_xlabel('Latent Transaction Feature 1', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Latent Transaction Feature 2', fontsize=12, fontweight='bold')
axes[0].set_title('1. SMOTE Data Resampling: Geometric Interpolation in Space', fontsize=13, fontweight='bold', pad=12)
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='upper left')
axes[0].grid(True, linestyle='--', alpha=0.6)


#  CHART 2: Performance Metrics Comparison (Raw vs SMOTE) 
metric_names = ['Recall (Sensitivity)', 'Precision', 'F1-Score', 'G-Mean']
raw_vals = [metrics_raw['Recall'], metrics_raw['Precision'], metrics_raw['F1-Score'], metrics_raw['G-Mean']]
smote_vals = [metrics_smote['Recall'], metrics_smote['Precision'], metrics_smote['F1-Score'], metrics_smote['G-Mean']]

x = np.arange(len(metric_names))
width = 0.35

# Academic Blue Palette for bars
bars1 = axes[1].bar(x - width/2, raw_vals, width, label='Model on Imbalanced Data (Baseline)', color='#bdc3c7', edgecolor='black', alpha=0.9)
bars2 = axes[1].bar(x + width/2, smote_vals, width, label='Model on SMOTE-Balanced Data', color='#2980b9', edgecolor='black', alpha=0.9)

axes[1].set_ylabel('Metric Score (0.0 to 1.0)', fontsize=12, fontweight='bold')
axes[1].set_title('2. Classification Metric Gains with SMOTE Resampling', fontsize=13, fontweight='bold', pad=12)
axes[1].set_xticks(x)
axes[1].set_xticklabels(metric_names, fontsize=11, fontweight='bold')
axes[1].set_ylim(0, 1.25)
axes[1].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='upper right')
axes[1].grid(axis='y', linestyle='--', alpha=0.7)

for bar, val in zip(bars1, raw_vals):
    axes[1].annotate(f'{val:.2f}', xy=(bar.get_x() + bar.get_width()/2, val + 0.02),
                     ha='center', fontsize=10, fontweight='bold', color='#7f8c8d')

for bar, val in zip(bars2, smote_vals):
    axes[1].annotate(f'{val:.2f}', xy=(bar.get_x() + bar.get_width()/2, val + 0.02),
                     ha='center', fontsize=10, fontweight='bold', color='#2c3e50')

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 6: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (شکستن پارادوکس دقت):</b> گراف ۱ دقیقاً نشان می‌دهد که چرا ما به SMOTE نیاز داریم. "
    r"دایره‌های قرمز (تقلب‌ها) بسیار کم هستند. مدل خام (خط‌چین خاکستری) برای اینکه خطای کلی خود را کم کند، مرز تصمیم را "
    r"طوری رسم می‌کند که تقریباً تمام فضا را «سالم» در نظر بگیرد! این مدل دقت کلی بالایی دارد اما هیچ تقلبی را پیدا نمی‌کند. "
    r"ما با استفاده از <b>SMOTE</b>، مثلث‌های آبی‌رنگ را در فضای بینِ تقلب‌ها تولید کردیم. این داده‌های مصنوعی باعث شدند که کفهٔ ترازوی "
    r"داده‌ها در زمان آموزش برابر شود. در نتیجه، خط ممتدِ سرمه‌ای (مدل جدید) به عقب رانده شد تا منطقه‌ی بیشتری را برای کشف تقلب "
    r"محاصره کند. گراف ۲ شاهدمدعا است: معیار حیاتیِ <b>Recall (قدرت شکار تقلب‌ها)</b> در مدل خام روی عدد فاجعه‌بار ۰.۳۳ گیر کرده بود، "
    r"اما با دخالت SMOTE، این عدد به <b>۱.۰۰ (شکار صد درصدی تقلب‌ها)</b> رسید، بدون اینکه Precision افت شدیدی را تجربه کند."
)
display_academic_text(insight_conclusion, "insight")
```
