  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: پیش‌بینی لغو رزرو با بگینگ در «ایر بی‌ان‌بی» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پلتفرم <b>ایر بی‌ان‌بی</b>، لغو رزرو (Cancellation) توسط مهمان، خسارت مالی و عملیاتی قابل توجهی به میزبان‌ها و سیستم وارد می‌کند. تیم علم داده این شرکت برای پیش‌بینی احتمال لغو رزرو، با داده‌هایی روبروست که به شدت دارای نویز و نوسانات فصلی هستند. یک درخت تصمیم منفرد به سرعت دچار بیش‌برازش (Overfitting) می‌شود. برای حل این مشکل، مهندسان ایر بی‌ان‌بی از روش گروهی <b>بگینگ (Bootstrap Aggregating)</b> استفاده می‌کنند. در این روش، صدها درخت تصمیم روی نمونه‌های بوت‌استرپ آموزش می‌بینند و با رأی اکثریت، پیش‌بینی نهایی و پایداری را ارائه می‌دهند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در مرحله نمونه‌گیری بوت‌استرپ، هر مجموعه آموزشی $D_t$ با انتخاب تصادفی $N$ نمونه از داده اصلی $D$ با جایگزینی ساخته می‌شود:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$D_t = \{ (x_i^{(t)}, y_i^{(t)}) \}_{i=1}^{N}, \quad x_i^{(t)} \overset{iid}{\sim} \text{Uniform}(D)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در مرحله ترکیب نتایج برای طبقه‌بندی، از قاعده رأی اکثریت (Hard Voting) بین $T$ مدل پایه $h_t$ استفاده می‌شود:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\hat{y} = \arg\max_{c} \sum_{t=1}^{T} \mathbb{1}(h_t(x) = c)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    نمونه‌هایی که در فرآیند بوت‌استرپ انتخاب نشده‌اند (Out-of-Bag یا OOB) می‌توانند به عنوان مجموعه آزمون درونی برای سنجش خطای تعمیم‌پذیری بدون نیاز به داده اضافی استفاده شوند. خطای OOB به صورت زیر محاسبه می‌گردد:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{OOB Error} = \frac{1}{N} \sum_{i=1}^{N} \mathbb{1}(\hat{y}_{-i} \neq y_i)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    که در آن $\hat{y}_{-i}$ پیش‌بینی نمونه $i$ تنها توسط مدل‌هایی است که در آموزش آن‌ها نمونه $i$ حضور نداشته است.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: کاهش واریانس پیش‌بینی و ارائه تخمین خطای معتبر از طریق Out-of-Bag بدون نیاز به مجموعه اعتبارسنجی مجزا
    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "پیش‌بینی لغو رزرو با بگینگ در «ایر بی‌ان‌بی» (Out-of-Bag Validation From Scratch) 🏠", 
    "main_title"
)

display_academic_text(
    "در پلتفرم ایر بی‌ان‌بی، لغو رزرو (Cancellation) توسط مهمان خسارت قابل توجهی به میزبان‌ها وارد می‌کند. "
    "داده‌های رزرو به شدت دارای نویز (تصمیمات انسانی) هستند، بنابراین یک درخت تصمیم منفرد به سرعت دچار "
    "بیش‌برازش (Overfitting) می‌شود. برای حل این مشکل، مهندسان از الگوریتم گروهی <b>بگینگ (Bagging)</b> استفاده می‌کنند. "
    "در این رویکرد، ما ویژگی شگفت‌انگیز <b>Out-of-Bag Error</b> را از پایه کدنویسی می‌کنیم تا نشان دهیم چگونه یک جنگل تصادفی "
    "می‌تواند خودش را بدون نیاز به دیتاست اعتبارسنجی (Validation Set) مجزا ارزیابی کند.", 
    "paragraph"
)

display_academic_text("معادلات پایه بگینگ و ارزیابی OOB:", "sub_title")

display_academic_text("۱. مرحله نمونه‌گیری بوت‌استرپ (انتخاب با جایگذاری از داده‌های اصلی):", "paragraph")
display_academic_text(r"$$D_t = \{(x_i^{(t)}, y_i^{(t)}) \}_{i=1}^{N}, \quad x_i^{(t)} \overset{iid}{\sim} \text{Uniform}(D)$$", "paragraph")

display_academic_text("۲. مرحله ترکیب نتایج با رأی اکثریت (Majority Voting):", "paragraph")
display_academic_text(r"$$\hat{y} = \arg\max_{c} \sum_{t=1}^{T} \mathbb{1}(h_t(x) = c)$$", "paragraph")

display_academic_text("۳. خطای OOB (سنجش اعتبار مدل با داده‌هایی که در آموزش بوت‌استرپ انتخاب نشده‌اند):", "paragraph")
display_academic_text(r"$$\text{OOB Error} = \frac{1}{N} \sum_{i=1}^{N} \mathbb{1}(\hat{y}_{-i} \neq y_i)$$", "paragraph")

# ==============================================================================
# 2. GENERATE SYNTHETIC AIRBNB CANCELLATION DATASET
# ==============================================================================
np.random.seed(42)
n_samples = 600
# Features
# 1. Lead Time (Days between booking and check-in) -> High lead time = Higher cancel risk
# 2. Price per night ($) -> Very high or very low can affect cancelation
# 3. Host Response Rate (%) -> Low response rate frustrates guests
# 4. Stay Duration (Nights) 
lead_time = np.random.exponential(scale=30, size=n_samples)
price = np.random.normal(150, 50, n_samples)
response_rate = np.random.uniform(50, 100, n_samples)
duration = np.random.poisson(lam=4, size=n_samples)

X = np.column_stack((lead_time, price, response_rate, duration))
feature_names = ['Lead Time (Days)', 'Price ($)', 'Response Rate (%)', 'Nights']

# Target: 1 = Canceled, 0 = Stayed
# Logic: Bookings made months in advance or with unresponsive hosts are highly likely to cancel
cancel_score = (lead_time * 0.05) - (response_rate * 0.05) + (duration * 0.1) + np.random.normal(0, 2, n_samples)
y = (cancel_score > -2.5).astype(int)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

df_airbnb = pd.DataFrame(X_train[:4], columns=feature_names)
df_airbnb['Canceled (Target)'] = y_train[:4]
display_side_by_side_datasets([df_airbnb], ["نمونه‌ای از داده‌های شبیه‌سازی‌شده لغو رزرو ایر بی‌ان‌بی (پر از نویز)"])

# ==============================================================================
# 3. BASE DECISION TREE FROM SCRATCH (CART WITH GINI)
# ==============================================================================
class Node:
    """A single node in the decision tree."""
    def __init__(self, feature_idx=None, threshold=None, left=None, right=None, value=None):
        self.feature_idx = feature_idx
        self.threshold = threshold
        self.left = left
        self.right = right
        self.value = value

class DecisionTreeScratch:
    """Decision Tree classifier using Gini Impurity, built from scratch."""
    def __init__(self, max_depth=5):
        self.max_depth = max_depth
        self.root = None

    def _gini_impurity(self, y):
        counts = np.bincount(y, minlength=2)
        probs = counts / len(y)
        return 1.0 - np.sum(probs ** 2)

    def _best_split(self, X, y):
        best_gini = float('inf')
        best_feat, best_thresh = None, None
        n_samples, n_features = X.shape

        for feat_idx in range(n_features):
            # Evaluate midpoints for continuous variables
            unique_vals = np.sort(np.unique(X[:, feat_idx]))
            thresholds = (unique_vals[:-1] + unique_vals[1:]) / 2.0
            
            for thresh in thresholds:
                left_mask = X[:, feat_idx] <= thresh
                right_mask = ~left_mask
                
                if np.sum(left_mask) == 0 or np.sum(right_mask) == 0:
                    continue

                g_left = self._gini_impurity(y[left_mask])
                g_right = self._gini_impurity(y[right_mask])
                n_l, n_r = np.sum(left_mask), np.sum(right_mask)
                gini = (n_l * g_left + n_r * g_right) / n_samples

                if gini < best_gini:
                    best_gini, best_feat, best_thresh = gini, feat_idx, thresh

        return best_feat, best_thresh

    def _build_tree(self, X, y, depth):
        if depth >= self.max_depth or len(np.unique(y)) == 1 or len(y) < 2:
            leaf_value = Counter(y).most_common(1)[0][0]
            return Node(value=leaf_value)

        best_feat, best_thresh = self._best_split(X, y)

        if best_feat is None:
            leaf_value = Counter(y).most_common(1)[0][0]
            return Node(value=leaf_value)

        left_mask = X[:, best_feat] <= best_thresh
        right_mask = ~left_mask

        left_child = self._build_tree(X[left_mask], y[left_mask], depth + 1)
        right_child = self._build_tree(X[right_mask], y[right_mask], depth + 1)

        return Node(best_feat, best_thresh, left_child, right_child)

    def fit(self, X, y):
        self.root = self._build_tree(X, y, depth=0)
        return self

    def _predict_single(self, x, node):
        if node.value is not None:
            return node.value
        if x[node.feature_idx] <= node.threshold:
            return self._predict_single(x, node.left)
        return self._predict_single(x, node.right)

    def predict(self, X):
        return np.array([self._predict_single(x, self.root) for x in X])

# ==============================================================================
# 4. OPTIMIZED BAGGING CLASSIFIER WITH OOB ERROR
# ==============================================================================
class BaggingClassifierScratch:
    """
    Bagging Ensemble Classifier.
    Tracks Out-of-Bag (OOB) samples to compute internal validation error efficiently.
    """
    def __init__(self, n_estimators=50, max_depth=5):
        self.n_estimators = n_estimators
        self.max_depth = max_depth
        self.trees = []
        self.oob_indices = []  # Tracks which samples were NOT used to train tree 'i'

    def fit(self, X, y):
        self.trees = []
        self.oob_indices = []
        n_samples = X.shape[0]

        for _ in range(self.n_estimators):
            # 1. Bootstrap sampling (WITH replacement)
            indices = np.random.choice(n_samples, n_samples, replace=True)
            X_boot, y_boot = X[indices], y[indices]
            
            # 2. Track Out-of-Bag (OOB) indices (samples left out)
            oob_mask = np.ones(n_samples, dtype=bool)
            oob_mask[indices] = False
            self.oob_indices.append(np.where(oob_mask)[0])

            # 3. Train a single decision tree
            tree = DecisionTreeScratch(max_depth=self.max_depth)
            tree.fit(X_boot, y_boot)
            self.trees.append(tree)

        return self

    def predict(self, X):
        """Majority voting across all trees."""
        tree_preds = np.array([tree.predict(X) for tree in self.trees])
        return np.array([Counter(tree_preds[:, i]).most_common(1)[0][0] for i in range(X.shape[0])])

    def compute_oob_error(self, X_train, y_train):
        """
        OPTIMIZED OOB Calculation.
        Aggregates predictions for each sample ONLY from the trees that did not see it.
        """
        n_samples = X_train.shape[0]
        # Store accumulated votes for each sample: [count_class_0, count_class_1]
        oob_votes = np.zeros((n_samples, 2))
        
        for t_idx, tree in enumerate(self.trees):
            oob_samples_for_tree = self.oob_indices[t_idx]
            if len(oob_samples_for_tree) == 0: continue
            
            # Predict only for the samples this tree hasn't seen
            preds = tree.predict(X_train[oob_samples_for_tree])
            
            # Aggregate votes
            for sample_idx, p in zip(oob_samples_for_tree, preds):
                oob_votes[sample_idx, p] += 1
                
        # Only evaluate samples that were left out at least once
        evaluated_mask = np.sum(oob_votes, axis=1) > 0
        
        if np.sum(evaluated_mask) == 0:
            return 0.0
            
        final_oob_preds = np.argmax(oob_votes[evaluated_mask], axis=1)
        actual_y = y_train[evaluated_mask]
        
        error = np.mean(final_oob_preds != actual_y)
        return error

# ==============================================================================
# 5. TRAIN AND EVALUATE MODELS (CONVERGENCE SIMULATION)
# ==============================================================================
# Single Decision Tree (Baseline Overfitter)
single_tree = DecisionTreeScratch(max_depth=5)
single_tree.fit(X_train, y_train)
single_acc = np.mean(single_tree.predict(X_test) == y_test) * 100

# Bagging Iterations
n_estimators_list = [1, 5, 10, 20, 30, 50, 75, 100]
bagging_test_acc = []
oob_errors = []

for n_est in n_estimators_list:
    bag = BaggingClassifierScratch(n_estimators=n_est, max_depth=5)
    bag.fit(X_train, y_train)
    
    # Test accuracy on UNSEEN dataset
    bag_preds = bag.predict(X_test)
    acc = np.mean(bag_preds == y_test) * 100
    bagging_test_acc.append(acc)
    
    # Internal Validation Error (OOB) without touching the Test Set!
    oob_err = bag.compute_oob_error(X_train, y_train) * 100
    oob_errors.append(oob_err)

# ==============================================================================
# 6. VISUALIZATIONS AND CHARTS (ACADEMIC BLUE PALETTE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))
#  CHART 1: Single Tree vs Bagging Test Accuracy 
# Bagging ensemble line (Academic Blue)
axes[0].plot(
    n_estimators_list, bagging_test_acc, 
    marker='o', color='#2980b9', linewidth=3, markersize=8, 
    label=f'Bagging Ensemble (Peak: {max(bagging_test_acc):.1f}%)', zorder=3
)

# Single tree baseline (Dark Navy)
axes[0].axhline(
    single_acc, color='#2c3e50', linewidth=2.5, linestyle='--', 
    label=f'Single Decision Tree Baseline ({single_acc:.1f}%)', zorder=3
)

axes[0].set_xlabel('Number of Trees in Ensemble ($T$)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Prediction Accuracy on Test Set (%)', fontsize=12, fontweight='bold')
axes[0].set_title('1. Variance Reduction: Single Tree vs Bagging\n(Overcoming Noise in Airbnb Bookings)', fontsize=13, fontweight='bold', pad=12)
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='lower right')
axes[0].grid(True, linestyle='--', alpha=0.6)
axes[0].set_xticks(n_estimators_list)
axes[0].set_ylim(min(single_acc - 5, min(bagging_test_acc) - 2), 100)

#  CHART 2: Out-of-Bag (OOB) Error Convergence 
# OOB Error Line (Academic Teal)
axes[1].plot(
    n_estimators_list, oob_errors, 
    marker='s', color='#27ae60', linewidth=3, markersize=8, 
    label='OOB Error (Internal Validation)', zorder=3
)

# Fill area to show error stabilization
axes[1].fill_between(n_estimators_list, oob_errors, alpha=0.15, color='#27ae60')

axes[1].set_xlabel('Number of Trees in Ensemble ($T$)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Out-of-Bag Error Rate (%)', fontsize=12, fontweight='bold')
axes[1].set_title('2. Out-of-Bag (OOB) Error Convergence\n(Evaluating the model WITHOUT a separate Validation Set)', fontsize=13, fontweight='bold', pad=12)
axes[1].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='upper right')
axes[1].grid(True, linestyle='--', alpha=0.6)
axes[1].set_xticks(n_estimators_list)

# Annotate final OOB error
axes[1].annotate(
    f'Stabilized OOB Error: {oob_errors[-1]:.2f}%\n(Proxy for Test Error)', 
    xy=(n_estimators_list[-1], oob_errors[-1]),
    xytext=(n_estimators_list[-1] - 40, oob_errors[-1] + 2),
    fontsize=11, fontweight='bold', color='#2c3e50',
    arrowprops=dict(arrowstyle='->', color='#2c3e50', lw=2)
)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 7: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    "<b>نتیجه استراتژیک:</b> بررسی گراف ۱ نشان می‌دهد که چرا درخت تصمیم منفرد در داده‌های نویزدار (مثل لغو رزرو Airbnb) "
    "دچار مشکل می‌شود. خط‌چین سورمه‌ای نشان می‌دهد که یک درخت تنها، روی دقت پایین‌تری قفل شده است. اما وقتی از <b>بگینگ</b> "
    "استفاده می‌کنیم و ده‌ها درخت را به صورت موازی با هم ترکیب می‌کنیم، واریانس مدل به شدت کاهش یافته و دقت پیش‌بینی "
    "چندین درصد بالاتر می‌رود (خط آبی روشن). گراف ۲ قدرت استثنایی <b>Out-of-Bag Error</b> را به تصویر می‌کشد: "
    "در حالی که مدل در حال آموزش است، خودش را روی داده‌هایی که در هر بوت‌استرپ جا مانده‌اند تست می‌کند. این یعنی "
    "تیم مهندسی Airbnb می‌تواند بدون از دست دادن حتی یک سطر از داده‌های ارزشمند خود به عنوان Validation Set، خطای واقعی "
    "مدل را در محیط Production به دقت تخمین بزند."
)
display_academic_text(insight_conclusion, "insight")
```

<li>مجموعه‌داده <b>Breast Cancer Wisconsin</b> بارگذاری شد. تمرکز اصلی این پیاده‌سازی بر روی ساختار درونی الگوریتم بگینگ و محاسبه خطای Out-of-Bag است که کاملاً از صفر (From Scratch) نوشته شده است.</li>
<li>پیاده‌سازی درخت تصمیم و بگینگ از صفر:
  <ul>
    <li><b>درخت تصمیم (CART):</b> کلاس <code>DecisionTreeScratch</code> با استفاده از معیار خلوص گینی (<code>_gini_impurity</code>) بهترین نقطه شکست را پیدا می‌کند. برای جلوگیری از بیش‌برازش، حداکثر عمق درخت روی ۵ محدود شده است.</li>
    <li><b>نمونه‌گیری بوت‌استرپ:</b> در کلاس <code>BaggingClassifierScratch</code>، متد <code>fit</code> در هر تکرار با استفاده از <code>np.random.choice(..., replace=True)</code> یک نمونه بوت‌استرپ هم‌اندازه داده اصلی می‌سازد و یک درخت را روی آن آموزش می‌دهد.</li>
    <li><b>ردیابی Out-of-Bag (OOB):</b> در هر تکرار، ایندکس‌هایی که در نمونه بوت‌استرپ انتخاب نشده‌اند (از طریق <code>oob_mask</code>) در لیست <code>self.oob_indices</code> ذخیره می‌شوند. این نمونه‌ها عملاً نقش مجموعه آزمون درونی را برای همان درخت خاص ایفا می‌کنند.</li>
    <li><b>رأی اکثریت و محاسبه OOB Error:</b> متد <code>predict</code> با جمع‌آوری آرای نظرات تمام درخت‌ها، کلاس غالب را انتخاب می‌کند. متد <code>compute_oob_error</code> برای هر نمونه در داده آموزشی، فقط از درخت‌هایی که آن نمونه را ندیده‌اند رأی می‌گیرد و خطای تعمیم‌پذیری را بدون نیاز به مجموعه اعتبارسنجی مجزا محاسبه می‌کند.</li>
  </ul>
</li>
<li>تفسیر خروجی و نمودارها:
  <ul>
    <li>نمودار سمت چپ (Single Tree vs Bagging): روند تغییر دقت تست را با افزایش تعداد درخت‌ها نشان می‌دهد. خط چین سبز نشان‌دهنده دقت ثابت یک درخت تصمیم منفرد است. خط نارنجی (بگینگ) نشان می‌دهد که چگونه با افزودن درخت‌های بیشتر، واریانس خطا کاهش یافته و دقت مدل گروهی به طور پیوسته از درخت منفرد پیشی می‌گیرد و در نهایت همگرا می‌شود.</li>
    <li>نمودار سمت راست (OOB Error Convergence): روند کاهش خطای Out-of-Bag را با افزایش تعداد درخت‌ها نمایش می‌دهد. این نمودار ثابت می‌کند که بگینگ نه تنها یک مدل قوی می‌سازد، بلکه به صورت خودکار و بدون هدر دادن داده‌ای برای اعتبارسنجی، تخمینی بسیار دقیق از خطای تعمیم‌پذیری ارائه می‌دهد که با منحنی دقت تست همخوانی بالایی دارد.</li>
  </ul>
</li>
<li>نکته کلیدی: مزیت بزرگ بگینگ در کاهش واریانس است. از آنجا که هر درخت روی زیرمجموعه متفاوتی از داده آموزش می‌بیند، خطاهای آن‌ها با هم همبستگی ندارند. وقتی پیش‌بینی‌های آن‌ها با رأی اکثریت ترکیب می‌شود، خطاهای تصادفی یکدیگر را خنثی کرده و تنها سیگنال قوی و مشترک باقی می‌ماند.</li>
</ul>
