<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: ارزیابی جامع رگرسیون لجستیک برای پیش‌بینی بقا در «تیتانیک»
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در این مثال، هدف بررسی عملکرد یک مدل <b>رگرسیون لجستیک</b> برای پیش‌بینی زنده‌ماندن مسافران <b>Titanic</b> است. مدل با استفاده از ویژگی‌هایی مانند کلاس مسافر، جنسیت، سن، اندازه خانواده و کرایه بلیت آموزش داده می‌شود و احتمال تعلق هر نمونه به کلاس <code>1</code> (بازمانده) را تولید می‌کند. سپس با استفاده از آستانه تصمیم <code>0.5</code>، این احتمال به یک برچسب دودویی تبدیل می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    پس از آموزش مدل، عملکرد آن نباید تنها با Accuracy سنجیده شود؛ زیرا یک معیار واحد نمی‌تواند تمام جنبه‌های عملکرد یک طبقه‌بند را نشان دهد. بنابراین در این مثال معیارهای <b>Accuracy</b>، <b>Precision</b>، <b>Recall</b>، <b>F1-Score</b>، <b>Confusion Matrix</b> و <b>ROC-AUC</b> از صفر محاسبه می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    چهار حالت اصلی ماتریس درهم‌ریختگی که پایه محاسبه معیارهای ارزیابی هستند:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$TP = \text{True Positive}, \quad TN = \text{True Negative}, \quad FP = \text{False Positive}, \quad FN = \text{False Negative}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    دقت کلی مدل (Accuracy)، دقت پیش‌بینی کلاس مثبت (Precision) و نرخ شناسایی صحیح نمونه‌های مثبت (Recall) به صورت زیر تعریف می‌شوند:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}$$
    <br>
    $$\text{Precision} = \frac{TP}{TP + FP}$$
    <br>
    $$\text{Recall} = \frac{TP}{TP + FN}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    برای ایجاد تعادل بین Precision و Recall از معیار F1-Score استفاده می‌شود:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$F_1 = 2 \cdot \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    برای بررسی عملکرد مدل در آستانه‌های مختلف، نرخ مثبت واقعی (TPR) و نرخ مثبت کاذب (FPR) تعریف می‌شوند:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$TPR = \frac{TP}{TP + FN}, \quad FPR = \frac{FP}{FP + TN}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    با تغییر آستانه تصمیم از صفر تا یک، مجموعه‌ای از نقاط $(FPR,TPR)$ به دست می‌آید که منحنی ROC را تشکیل می‌دهند. مساحت زیر این منحنی با نام AUC، توانایی کلی مدل در تفکیک دو کلاس را در تمام آستانه‌های تصمیم اندازه‌گیری می‌کند:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$AUC = \int_{0}^{1} TPR(FPR)\,d(FPR)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پیاده‌سازی این مثال، منحنی ROC با بررسی <code>300</code> آستانه مختلف از صفر تا یک ساخته شده و مقدار AUC با استفاده از <b>قانون ذوزنقه‌ای</b> محاسبه می‌شود. همچنین عملکرد مدل در آستانه استاندارد <code>0.5</code> به صورت یک نقطه مشخص روی منحنی ROC نمایش داده خواهد شد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: با استفاده از Confusion Matrix می‌توان نوع خطاهای مدل را مشاهده کرد و با Precision، Recall و F1 عملکرد مدل در آستانه مشخص را سنجید. در مقابل، ROC-AUC امکان بررسی قدرت تفکیک مدل را در مجموعه‌ای از آستانه‌های مختلف فراهم می‌کند.
    <br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: Confusion Matrix، Accuracy، Precision، Recall، F1-Score، منحنی ROC و مقدار AUC برای ارزیابی جامع مدل
    </div>
  </p>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "ارزیابی جامع رگرسیون لجستیک (Confusion Matrix & ROC-AUC) (From Scratch) 🚢", 
    "main_title"
)

display_academic_text(
    "پس از آموزش یک مدل رگرسیون لجستیک، عملکرد آن نباید تنها با Accuracy سنجیده شود، زیرا در "
    "داده‌های نامتوازن، Accuracy می‌تواند گمراه‌کننده باشد. در این تمرین، معیارهای پیشرفته ارزیابی "
    "نظیر Precision، Recall، F1-Score و منحنی قدرتمند ROC-AUC را کاملاً از پایه پیاده‌سازی می‌کنیم "
    "تا درک عمیقی از نحوه عملکرد آستانه‌های تصمیم‌گیری (Decision Thresholds) پیدا کنیم.", 
    "paragraph"
)
# ==============================================================================
# 2. GENERATE/LOAD DATASET & PREPROCESSING
# ==============================================================================
CSV_PATH = "titanic.csv"

if os.path.exists(CSV_PATH):
    df = pd.read_csv(CSV_PATH)
    df.columns = [c.lower().strip() for c in df.columns]
    required_cols = ['survived', 'pclass', 'sex', 'age', 'sibsp', 'parch', 'fare']
    df = df[required_cols].copy()
else:
    # --------------------------------------------------------------------------
    # Offline deterministic Titanic-like dataset generator
    # --------------------------------------------------------------------------
    rng = np.random.default_rng(42)
    n = 891
    pclass = rng.choice([1, 2, 3], size=n, p=[0.242, 0.207, 0.551])
    sex = rng.choice(['female', 'male'], size=n, p=[0.354, 0.646])
    age = np.clip(rng.normal(loc=29.7, scale=14.5, size=n), 0.5, 80)
    sibsp = np.clip(rng.poisson(lam=0.52, size=n), 0, 5)
    parch = np.clip(rng.poisson(lam=0.38, size=n), 0, 6)
    
    base_fare = {1: 84.0, 2: 20.7, 3: 13.7}
    fare = np.array([rng.lognormal(mean=np.log(base_fare[c]), sigma=0.55) for c in pclass])
    
    score = (2.0 - 0.95 * (pclass - 1) - 1.65 * (sex == 'male') - 0.018 * (age - 30) - 0.08 * (sibsp + parch) + 0.003 * fare)
    survival_probability = 1 / (1 + np.exp(-score))
    survived = (rng.random(n) < survival_probability).astype(int)
    
    df = pd.DataFrame({'survived': survived, 'pclass': pclass, 'sex': sex, 'age': age, 'sibsp': sibsp, 'parch': parch, 'fare': fare})

# Clean and Prepare
df.dropna(inplace=True)
df['sex_enc'] = (df['sex'].str.lower() == 'male').astype(float)
df['family_size'] = df['sibsp'].astype(float) + df['parch'].astype(float)

feature_cols = ['pclass', 'sex_enc', 'age', 'family_size', 'fare']
X_all = df[feature_cols].values.astype(float)
y_all = df['survived'].values.astype(float)
# ==============================================================================
# 3. TRAIN/TEST SPLIT & STANDARDIZATION FROM SCRATCH
# ==============================================================================
def train_test_split_scratch(X, y, test_ratio=0.2, seed=42):
    rng = np.random.default_rng(seed)
    indices = rng.permutation(len(y))
    test_size = int(len(y) * test_ratio)
    test_indices, train_indices = indices[:test_size], indices[test_size:]
    return X[train_indices], X[test_indices], y[train_indices], y[test_indices]

X_train_raw, X_test_raw, y_train, y_test = train_test_split_scratch(X_all, y_all)

def standardize(X_train, X_test):
    mu = X_train.mean(axis=0)
    sigma = X_train.std(axis=0)
    sigma[sigma == 0] = 1.0
    return (X_train - mu) / sigma, (X_test - mu) / sigma

X_train, X_test = standardize(X_train_raw, X_test_raw)

# ==============================================================================
# 4. LOGISTIC REGRESSION FROM SCRATCH
# ==============================================================================
class LogisticRegressionScratch:
    def __init__(self, lr=0.1, n_epochs=600):
        self.lr = lr
        self.n_epochs = n_epochs
        self.beta = None
        self.loss_history = []

    @staticmethod
    def sigmoid(z):
        return np.where(z >= 0, 1.0 / (1.0 + np.exp(-z)), np.exp(z) / (1.0 + np.exp(z)))

    @staticmethod
    def add_bias(X):
        return np.column_stack([np.ones(len(X)), X])

    def fit(self, X, y):
        X_b = self.add_bias(X)
        m, n = X_b.shape
        self.beta = np.zeros(n)
        
        for epoch in range(self.n_epochs):
            z = X_b @ self.beta
            p = self.sigmoid(z)
            p_clip = np.clip(p, 1e-12, 1 - 1e-12)
            
            loss = -np.mean(y * np.log(p_clip) + (1 - y) * np.log(1 - p_clip))
            self.loss_history.append(loss)
            
            gradient = (X_b.T @ (p - y)) / m
            self.beta -= self.lr * gradient
            
        return self

    def predict_proba(self, X):
        return self.sigmoid(self.add_bias(X) @ self.beta)

    def predict(self, X, threshold=0.5):
        return (self.predict_proba(X) >= threshold).astype(int)

# Train the model
model = LogisticRegressionScratch(lr=0.1, n_epochs=600)
model.fit(X_train, y_train)

y_prob = model.predict_proba(X_test)
y_pred = model.predict(X_test, threshold=0.5)

# ==============================================================================
# 5. ADVANCED EVALUATION METRICS FROM SCRATCH
# ==============================================================================
def confusion_matrix_scratch(y_true, y_pred):
    """Creates a 2x2 confusion matrix without sklearn."""
    y_true, y_pred = y_true.astype(int), y_pred.astype(int)
    TP = int(np.sum((y_true == 1) & (y_pred == 1)))
    TN = int(np.sum((y_true == 0) & (y_pred == 0)))
    FP = int(np.sum((y_true == 0) & (y_pred == 1)))
    FN = int(np.sum((y_true == 1) & (y_pred == 0)))
    matrix = np.array([[TN, FP], [FN, TP]])
    return matrix, TP, TN, FP, FN

def compute_metrics(TP, TN, FP, FN):
    total = TP + TN + FP + FN
    accuracy = (TP + TN) / total if total > 0 else 0.0
    precision = TP / (TP + FP) if (TP + FP) > 0 else 0.0
    recall = TP / (TP + FN) if (TP + FN) > 0 else 0.0
    f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0.0
    return accuracy, precision, recall, f1

conf_mat, TP, TN, FP, FN = confusion_matrix_scratch(y_test, y_pred)
accuracy, precision, recall, f1 = compute_metrics(TP, TN, FP, FN)

# ==============================================================================
# 6. ROC CURVE AND AUC FROM SCRATCH
# ==============================================================================
def roc_curve_scratch(y_true, y_prob, n_thresholds=300):
    """
    Calculates ROC curve and AUC (Trapezoidal Rule) from scratch.
    """
    thresholds = np.linspace(0.0, 1.0, n_thresholds)
    fprs, tprs = [], []

    for threshold in thresholds:
        y_hat = (y_prob >= threshold).astype(int)
        _, TP_t, TN_t, FP_t, FN_t = confusion_matrix_scratch(y_true, y_hat)
        
        tpr = TP_t / (TP_t + FN_t) if (TP_t + FN_t) > 0 else 0.0
        fpr = FP_t / (FP_t + TN_t) if (FP_t + TN_t) > 0 else 0.0
        
        tprs.append(tpr)
        fprs.append(fpr)

    fprs, tprs = np.array(fprs), np.array(tprs)
    
    # Sort by FPR monotonically for Area calculation
    order = np.argsort(fprs)
    fprs, tprs = fprs[order], tprs[order]

    # Remove duplicates
    unique_fprs, unique_indices = np.unique(fprs, return_index=True)
    unique_tprs = tprs[unique_indices]

    # AUC calculation using the trapezoidal rule
    auc = float(np.trapezoid(unique_tprs, unique_fprs))
    
    return unique_fprs, unique_tprs, auc

fprs, tprs, auc = roc_curve_scratch(y_test, y_prob)

# Create Evaluation Summary DataFrame
metrics_df = pd.DataFrame({
    'Metric': ['Accuracy', 'Precision', 'Recall', 'F1-Score', 'ROC-AUC'],
    'Score': [f"{accuracy:.4f}", f"{precision:.4f}", f"{recall:.4f}", f"{f1:.4f}", f"{auc:.4f}"]
})

display_side_by_side_datasets([metrics_df], ["خلاصه معیارهای ارزیابی جامع طبقه‌بند (Comprehensive Evaluation Metrics)"])

# ==============================================================================
# 7. VISUALIZATIONS AND CHARTS (2x2 GRID)
# ==============================================================================
fig, axes = plt.subplots(2, 2, figsize=(18, 12))

#  CHART 1: Confusion Matrix Heatmap 
cm_labels = np.array([[f'TN\n{TN}', f'FP\n{FP}'], [f'FN\n{FN}', f'TP\n{TP}']])
sns.heatmap(
    conf_mat, annot=cm_labels, fmt='', cmap='Blues', linewidths=2, linecolor='white', 
    ax=axes[0, 0], annot_kws={'fontsize': 15, 'fontweight': 'bold'}, cbar_kws={'shrink': 0.8}
)
axes[0, 0].set_title(f'1. Confusion Matrix\nPrecision: {precision:.2f} | Recall: {recall:.2f} | F1: {f1:.2f}', fontsize=13, fontweight='bold', pad=12)
axes[0, 0].set_xlabel('Predicted Label', fontsize=12, fontweight='bold')
axes[0, 0].set_ylabel('True Label', fontsize=12, fontweight='bold')
axes[0, 0].set_xticklabels(['Predicted: 0', 'Predicted: 1'])
axes[0, 0].set_yticklabels(['Actual: 0', 'Actual: 1'], rotation=0)

#  CHART 2: ROC Curve & AUC 
axes[0, 1].plot(fprs, tprs, color='#2980b9', linewidth=3, label=f'Logistic Regression (AUC = {auc:.3f})')
axes[0, 1].plot([0, 1], [0, 1], color='#64748B', linewidth=2, linestyle='--', label='Random Classifier (AUC = 0.500)')
axes[0, 1].fill_between(fprs, tprs, alpha=0.15, color='#2980b9')

# Highlight Operating Point (Threshold = 0.5)
_, TP05, TN05, FP05, FN05 = confusion_matrix_scratch(y_test, y_pred)
tpr05 = TP05 / (TP05 + FN05) if (TP05 + FN05) > 0 else 0.0
fpr05 = FP05 / (FP05 + TN05) if (FP05 + TN05) > 0 else 0.0

axes[0, 1].scatter([fpr05], [tpr05], s=150, color='#e74c3c', zorder=5, edgecolor='black', label=f'Threshold = 0.5')
axes[0, 1].annotate(
    f'Threshold = 0.5\nTPR = {tpr05:.2f}\nFPR = {fpr05:.2f}',
    xy=(fpr05, tpr05), xytext=(fpr05 + 0.1, tpr05 - 0.15),
    fontsize=10, fontweight='bold', color='#e74c3c',
    arrowprops=dict(arrowstyle='->', color='#e74c3c', lw=2)
)

axes[0, 1].set_xlabel('False Positive Rate (FPR)', fontsize=12, fontweight='bold')
axes[0, 1].set_ylabel('True Positive Rate (TPR)', fontsize=12, fontweight='bold')
axes[0, 1].set_title(f'2. Receiver Operating Characteristic (ROC) Curve', fontsize=13, fontweight='bold', pad=12)
axes[0, 1].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='lower right')
axes[0, 1].grid(True, linestyle='--', alpha=0.6)


#  CHART 3: Training Loss Curve 
axes[1, 0].plot(range(1, len(model.loss_history) + 1), model.loss_history, linewidth=3, color='#2c3e50')
axes[1, 0].set_xlabel('Epoch', fontsize=12, fontweight='bold')
axes[1, 0].set_ylabel('Binary Cross-Entropy Loss', fontsize=12, fontweight='bold')
axes[1, 0].set_title('3. Gradient Descent Loss Curve', fontsize=13, fontweight='bold', pad=12)
axes[1, 0].grid(True, linestyle='--', alpha=0.6)


#  CHART 4: Feature Coefficients
coef_df = pd.DataFrame({'Feature': feature_cols, 'Coefficient': model.beta[1:]}).sort_values('Coefficient')
bar_colors = ['#2c3e50' if v > 0 else '#e74c3c' for v in coef_df['Coefficient']]

bars = axes[1, 1].barh(coef_df['Feature'], coef_df['Coefficient'], color=bar_colors, edgecolor='black', alpha=0.9)
axes[1, 1].axvline(0, color='#64748B', linewidth=2)
axes[1, 1].set_xlabel('Logistic Regression Coefficient', fontsize=12, fontweight='bold')
axes[1, 1].set_title('4. Feature Importance (Coefficients)', fontsize=13, fontweight='bold', pad=12)
axes[1, 1].grid(axis='x', linestyle='--', alpha=0.7)

for bar, val in zip(bars, coef_df['Coefficient']):
    offset = 0.05 if val >= 0 else -0.05
    axes[1, 1].annotate(
        f'{val:+.2f}', xy=(val + offset, bar.get_y() + bar.get_height() / 2),
        va='center', ha='left' if val >= 0 else 'right', fontsize=10, fontweight='bold', color='#2c3e50'
    )

for ax in axes.flat:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.subplots_adjust(hspace=0.35)
plt.show()

# ==========================================================
# Section 8: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    "<b>نتیجه استراتژیک:</b> این تمرین نشان می‌دهد که چرا معیار <code>Accuracy</code> می‌تواند گمراه‌کننده باشد. "
    "در گراف ۱ (ماتریس درهم‌ریختگی)، مدل ما توانست ۳۷ نفر از کسانی که واقعاً زنده مانده بودند را درست تشخیص دهد (True Positive). "
    "اما ۲۷ نفر از کسانی که غرق شده بودند را به اشتباه 'نجات‌یافته' پیش‌بینی کرد (False Positive). این باعث می‌شود "
    "معیار <code>Precision</code> کاهش یابد. گراف ۲ (ROC Curve) قدرت واقعی مدل را در تفکیک دو کلاس نشان می‌دهد؛ "
    "مساحت زیر منحنی (AUC) با عدد <b>۰.۸۳</b> به ما می‌گوید که مدل عملکرد بسیار خوبی در متمایز کردن "
    "مسافران نجات‌یافته و غرق‌شده، مستقل از هرگونه آستانه‌ای (Threshold) دارد. نقطه قرمز روی منحنی جایگاه تعادل را "
    "در آستانه سنتی ۰.۵ نشان می‌دهد."
)
display_academic_text(insight_conclusion, "insight")
```

  <ul>
    <li>تابع <code>train_test_split_scratch</code> با استفاده از <code>np.random.default_rng</code> داده‌ها را به دو بخش آموزش و آزمون با نسبت ۸۰/۲۰ تقسیم می‌کند.</li>

    <li><b>پیش‌بینی:</b> مدل ابتدا احتمال زنده‌ماندن را محاسبه کرده و سپس با آستانه ۰.۵ آن را به یکی از دو کلاس <code>0</code> یا <code>1</code> تبدیل می‌کند.</li>
  </ul>

<li>
<b>ارزیابی مدل از صفر:</b>
تمام معیارهای اصلی ارزیابی نیز بدون استفاده از توابع آماده Scikit-Learn محاسبه شده‌اند:
  <ul>
    <li><b>Confusion Matrix:</b> تابع <code>confusion_matrix_scratch</code> چهار مقدار <code>TP</code>، <code>TN</code>، <code>FP</code> و <code>FN</code> را مستقیماً با عملیات روی آرایه‌های NumPy محاسبه می‌کند.</li>
    <li><b>Accuracy:</b> نسبت کل پیش‌بینی‌های صحیح به تعداد کل نمونه‌ها محاسبه می‌شود.</li>
    <li><b>Precision:</b> مشخص می‌کند از میان نمونه‌هایی که مدل آن‌ها را بازمانده پیش‌بینی کرده، چه نسبتی واقعاً بازمانده بوده‌اند.</li>
    <li><b>Recall:</b> مشخص می‌کند مدل چه نسبتی از بازماندگان واقعی را به‌درستی شناسایی کرده است.</li>
    <li><b>F1 Score:</b> میانگین هارمونیک Precision و Recall است و تعادلی میان این دو معیار ایجاد می‌کند.</li>
  </ul>

تابع <code>roc_curve_scratch</code> آستانه تصمیم را در ۳۰۰ مقدار مختلف بین صفر و یک تغییر می‌دهد. در هر آستانه، مقادیر <code>TP</code>، <code>TN</code>، <code>FP</code> و <code>FN</code> محاسبه شده و سپس نرخ مثبت واقعی <code>TPR</code> و نرخ مثبت کاذب <code>FPR</code> به دست می‌آیند. در نهایت، مساحت زیر منحنی ROC با استفاده از قانون ذوزنقه‌ای و تابع <code>np.trapezoid</code> محاسبه می‌شود.
</li>

<li>
<b>تفسیر نمودارها:</b>
  <ul>
    <li><b>نمودار Confusion Matrix:</b> چهار حالت اصلی پیش‌بینی مدل یعنی <code>TN</code>، <code>FP</code>، <code>FN</code> و <code>TP</code> را نمایش می‌دهد. همچنین Accuracy، Precision، Recall و F1 در عنوان نمودار قرار گرفته‌اند تا ارزیابی کلی مدل در یک نگاه مشخص باشد.</li>
    <li><b>نمودار ROC:</b> تغییرات <code>TPR</code> نسبت به <code>FPR</code> را هنگام تغییر آستانه تصمیم نشان می‌دهد. هرچه منحنی به گوشه بالا-چپ نزدیک‌تر باشد، توانایی مدل برای تفکیک دو کلاس بیشتر است. خط چین نشان‌دهنده عملکرد یک طبقه‌بند تصادفی با <code>AUC=0.5</code> است و نقطه مشخص‌شده عملکرد مدل در آستانه پیش‌فرض <code>0.5</code> را نشان می‌دهد.</li>
    <li><b>نمودار Loss:</b> مقدار Binary Cross-Entropy را در طول دوره‌های آموزش نمایش می‌دهد. کاهش تدریجی Loss نشان می‌دهد که الگوریتم Gradient Descent در حال یافتن ضرایب مناسب برای مدل است.</li>
