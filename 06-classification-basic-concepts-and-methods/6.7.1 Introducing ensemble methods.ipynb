<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: پیش‌بینی زمان رسیدن و قیمت پویا در «اوبر» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پلتفرم <b>اوبر</b>، مهندسان داده با دو چالش کاملاً متفاوت روبرو هستند: پیش‌بینی زمان رسیدن راننده (ETA) که به شدت تحت تأثیر نویزهای لحظه‌ای (مانند اختلال GPS یا ترافیک‌های تصادفی) است، و پیش‌بینی قیمت پویا (Surge Pricing) که نیازمند درک الگوهای پیچیده و غیرخطی بین تقاضا، عرضه و شرایط جوی است. برای حل این دو مسئله، اوبر از دو رویکرد گروهی متفاوت استفاده می‌کند: <b>بگینگ (Bagging)</b> برای کاهش واریانس و خطاهای تصادفی در ETA، و <b>بوستینگ (Boosting)</b> برای کاهش بایاس و مدل‌سازی روابط پیچیده در قیمت‌گذاری.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. بگینگ (Bagging): در این روش، مدل‌ها به صورت موازی و مستقل آموزش می‌بینند. هر مدل روی یک نمونه بوت‌استرپ از داده‌ها ساخته می‌شود تا واریانس خطا کاهش یابد. پیش‌بینی نهایی از طریق رأی اکثریت (در طبقه‌بندی) یا میانگین‌گیری (در رگرسیون) به دست می‌آید:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\hat{f}_{bag}(x) = \frac{1}{T} \sum_{t=1}^{T} f_t(x) \quad \text{(Regression)} \quad , \quad \hat{y} = \arg\max_{c} \sum_{t=1}^{T} \mathbb{1}(f_t(x) = c) \quad \text{(Classification)}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. بوستینگ (Boosting - AdaBoost): در این روش، مدل‌ها به صورت ترتیبی آموزش می‌بینند. هر مدل جدید (یادگیر ضعیف) سعی می‌کند خطاهای مدل‌های قبلی را اصلاح کند. وزن نمونه‌های اشتباه طبقه‌بندی‌شده افزایش می‌یابد تا مدل بعدی روی آن‌ها تمرکز کند:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\epsilon_t = \sum_{i: y_i \neq h_t(x_i)} w_i^{(t)} \quad , \quad \alpha_t = \frac{1}{2} \ln \left( \frac{1 - \epsilon_t}{\epsilon_t} \right)$$
  </div>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$w_i^{(t+1)} = w_i^{(t)} \exp\left(-\alpha_t y_i h_t(x_i)\right) \quad \text{(Normalized)}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: اوبر با ترکیب این دو فلسفه، سیستمی ساخته است که در آن بگینگ، پایداری و مقاومت در برابر نویزهای سنسورها را تضمین می‌کند، و بوستینگ، دقت محاسباتی را برای بهینه‌سازی درآمد و تعادل بازار بالا می‌برد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: کاهش واریانس با بگینگ (موازی) و کاهش بایاس با بوستینگ (ترتیبی) برای سناریوهای متفاوت
    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "پیش‌بینی زمان رسیدن و قیمت پویا در «اوبر» (Bagging vs Boosting From Scratch) 🚕", 
    "main_title"
)

display_academic_text(
    "در پلتفرم اوبر، مهندسان از دو رویکرد گروهی (Ensemble) متفاوت برای چالش‌های مختلف استفاده می‌کنند: "
    "از <b>بگینگ (Bagging)</b> برای کاهش واریانس در سیستم‌هایی که نویز تصادفی بالایی دارند (مانند اختلال GPS در زمان رسیدن)، "
    "و از <b>بوستینگ (Boosting - AdaBoost)</b> برای کاهش بایاس و یادگیری الگوهای پیچیده (مانند قیمت‌گذاری پویا بر اساس تقاضا). "
    "در این تمرین، هر دو الگوریتم به همراه مدل‌های پایه (Decision Trees) کاملاً از صفر پیاده‌سازی می‌شوند.", 
    "paragraph"
)

display_academic_text("۱. فرمول بگینگ (Bagging): کاهش واریانس از طریق آموزش موازی و رأی‌گیری اکثریت", "sub_title")
display_academic_text(r"$$\hat{y} = \arg\max_{c} \sum_{t=1}^{T} \mathbb{1}(f_t(x) = c)$$", "paragraph")

display_academic_text("۲. فرمول بوستینگ (AdaBoost): آموزش ترتیبی، محاسبه خطای وزن‌دار و آپدیت نمایی اوزان", "sub_title")
display_academic_text(r"$$\epsilon_t = \sum_{i: y_i \neq h_t(x_i)} w_i^{(t)}, \quad \alpha_t = \frac{1}{2} \ln \left( \frac{1 - \epsilon_t}{\epsilon_t} \right), \quad w_i^{(t+1)} = w_i^{(t)} \exp\left(-\alpha_t y_i h_t(x_i)\right)$$", "paragraph")

# ==============================================================================
# 2. SIMULATE UBER DYNAMIC PRICING (SURGE) DATASET
# ==============================================================================
np.random.seed(42)
n_samples = 600
# Features
# 1. Available Drivers in Zone
# 2. Active Ride Requests
# 3. Weather Severity (0=Clear, 10=Heavy Storm)
# 4. Traffic Congestion Index (0=Empty, 10=Gridlock)
drivers = np.random.poisson(lam=15, size=n_samples)
requests = np.random.poisson(lam=25, size=n_samples)
weather = np.random.uniform(0, 10, n_samples)
traffic = np.random.uniform(0, 10, n_samples)

X = np.column_stack((drivers, requests, weather, traffic))
feature_names = ['Available Drivers', 'Ride Requests', 'Weather Severity', 'Traffic Index']

# Target: Is Surge Pricing Active? (1=Yes, 0=No)
# Logic: Surge happens when Demand > Supply, heavily amplified by Bad Weather and Traffic
surge_score = (requests * 1.5) - (drivers * 2.0) + (weather * 1.2) + (traffic * 1.0) + np.random.normal(0, 3, n_samples)
y_raw = (surge_score > 15).astype(int)

# AdaBoost mathematically requires targets to be {-1, +1}
y_adaboost = np.where(y_raw == 0, -1, 1)

# Bagging uses standard binary labels {0, 1}
y_bagging = y_raw.copy()

X_train, X_test, y_train_ab, y_test_ab, y_train_bg, y_test_bg = train_test_split(
    X, y_adaboost, y_bagging, test_size=0.25, random_state=42, stratify=y_raw
)

df_uber = pd.DataFrame(X_train[:4], columns=feature_names)
df_uber['Surge Active?'] = y_train_bg[:4]
display_side_by_side_datasets([df_uber], ["نمونه‌ای از داده‌های شبیه‌سازی‌شده تقاضا و قیمت‌پویا (Surge) در اوبر"])

# ==============================================================================
# 3. BASE LEARNERS FROM SCRATCH (DECISION TREE & STUMP)
# ==============================================================================
class Node:
    def __init__(self, feature_idx=None, threshold=None, left=None, right=None, value=None):
        self.feature_idx = feature_idx
        self.threshold = threshold
        self.left = left
        self.right = right
        self.value = value

class DecisionTreeScratch:
    """Decision Tree implemented from scratch using Gini Impurity."""
    def __init__(self, max_depth=3, max_features=None):
        self.max_depth = max_depth
        self.max_features = max_features
        self.root = None
        self.feature_splits = Counter()

    def _gini(self, y, weights=None):
        if len(y) == 0: return 0.0
        
        if weights is None:
            counts = np.bincount(y, minlength=2)
            probs = counts / len(y)
        else:
            total_weight = np.sum(weights)
            if total_weight == 0: return 0.0
            probs = np.array([
                np.sum(weights[y == 0]) / total_weight,
                np.sum(weights[y == 1]) / total_weight
            ])
            
        return 1.0 - np.sum(probs ** 2)

    def _best_split(self, X, y, feature_indices, weights=None):
        best_gini = float('inf')
        best_feat, best_thresh = None, None

        for feat_idx in feature_indices:
            thresholds = np.unique(X[:, feat_idx])
            for thresh in thresholds:
                left_mask = X[:, feat_idx] <= thresh
                right_mask = ~left_mask

                if np.sum(left_mask) == 0 or np.sum(right_mask) == 0:
                    continue

                if weights is None:
                    g_left, g_right = self._gini(y[left_mask]), self._gini(y[right_mask])
                    gini = (np.sum(left_mask) * g_left + np.sum(right_mask) * g_right) / len(y)
                else:
                    w_left, w_right = weights[left_mask], weights[right_mask]
                    g_left = self._gini(y[left_mask], w_left)
                    g_right = self._gini(y[right_mask], w_right)
                    total_weight = np.sum(weights)
                    if total_weight == 0: continue
                    gini = (np.sum(w_left) * g_left + np.sum(w_right) * g_right) / total_weight

                if gini < best_gini:
                    best_gini, best_feat, best_thresh = gini, feat_idx, thresh

        return best_feat, best_thresh

    def _build_tree(self, X, y, depth, weights=None):
        n_samples, n_features = X.shape

        # Stopping conditions
        if depth >= self.max_depth or len(np.unique(y)) == 1 or n_samples < 2:
            if weights is None:
                val = Counter(y).most_common(1)[0][0]
            else:
                weight_class_0, weight_class_1 = np.sum(weights[y == 0]), np.sum(weights[y == 1])
                val = 1 if weight_class_1 > weight_class_0 else 0
            return Node(value=val)

        feat_indices = np.arange(n_features)
        if self.max_features is not None and self.max_features < n_features:
            feat_indices = np.random.choice(n_features, self.max_features, replace=False)

        best_feat, best_thresh = self._best_split(X, y, feat_indices, weights)

        if best_feat is None:
            val = Counter(y).most_common(1)[0][0] if weights is None else (1 if np.sum(weights[y == 1]) > np.sum(weights[y == 0]) else 0)
            return Node(value=val)

        self.feature_splits[best_feat] += 1
        left_mask = X[:, best_feat] <= best_thresh
        right_mask = ~left_mask

        left_weights = weights[left_mask] if weights is not None else None
        right_weights = weights[right_mask] if weights is not None else None

        left_node = self._build_tree(X[left_mask], y[left_mask], depth + 1, left_weights)
        right_node = self._build_tree(X[right_mask], y[right_mask], depth + 1, right_weights)

        return Node(feature_idx=best_feat, threshold=best_thresh, left=left_node, right=right_node)

    def fit(self, X, y, weights=None):
        self.feature_splits = Counter()
        self.root = self._build_tree(X, y, depth=0, weights=weights)
        return self

    def _predict_single(self, x, node):
        if node.value is not None:
            return node.value
        if x[node.feature_idx] <= node.threshold:
            return self._predict_single(x, node.left)
        return self._predict_single(x, node.right)

    def predict(self, X):
        return np.array([self._predict_single(x, self.root) for x in X])


class DecisionStump(DecisionTreeScratch):
    """Decision Tree with max_depth=1. Used as the weak learner for AdaBoost."""
    def __init__(self):
        super().__init__(max_depth=1)

# ==============================================================================
# 4. BAGGING ENSEMBLE FROM SCRATCH
# ==============================================================================
class BaggingClassifierScratch:
    """Each tree is trained independently on a bootstrap sample. Final prediction via majority voting."""
    def __init__(self, n_estimators=50, max_depth=3, max_features='sqrt'):
        self.n_estimators = n_estimators
        self.max_depth = max_depth
        self.max_features = max_features
        self.trees = []
        self.feature_importances_ = None

    def fit(self, X, y):
        self.trees = []
        n_samples, n_features = X.shape
        max_feat = max(1, int(np.sqrt(n_features))) if self.max_features == 'sqrt' else n_features
        total_splits = Counter()

        for _ in range(self.n_estimators):
            # Sampling WITH REPLACEMENT (Bootstrap)
            indices = np.random.choice(n_samples, n_samples, replace=True)
            X_boot, y_boot = X[indices], y[indices]

            tree = DecisionTreeScratch(max_depth=self.max_depth, max_features=max_feat)
            tree.fit(X_boot, y_boot)
            self.trees.append(tree)
            total_splits.update(tree.feature_splits)

        total = sum(total_splits.values())
        self.feature_importances_ = np.array([total_splits.get(i, 0) / total for i in range(n_features)]) if total > 0 else np.zeros(n_features)
        return self

    def predict(self, X):
        preds = np.array([tree.predict(X) for tree in self.trees])
        return np.array([Counter(preds[:, i]).most_common(1)[0][0] for i in range(X.shape[0])])

# ==============================================================================
# 5. ADABOOST ENSEMBLE FROM SCRATCH
# ==============================================================================
class AdaBoostClassifierScratch:
    """Trains stumps sequentially. Misclassified samples get higher weights for the next stump."""
    def __init__(self, n_estimators=50):
        self.n_estimators = n_estimators
        self.stumps = []
        self.alphas = []
        self.feature_importances_ = None

    def fit(self, X, y):
        self.stumps, self.alphas = [], []
        n_samples, n_features = X.shape

        # Initial Uniform Weights
        weights = np.full(n_samples, 1.0 / n_samples)
        total_feat_importance = np.zeros(n_features)

        for _ in range(self.n_estimators):
            stump = DecisionStump()
            
            # Trees need 0,1 internally for Gini calculations
            y_tree = np.where(y == 1, 1, 0)
            stump.fit(X, y_tree, weights=weights)
            preds_tree = stump.predict(X)
            
            # Convert internal {0,1} predictions back to AdaBoost {-1,+1}
            preds = np.where(preds_tree == 1, 1, -1)

            # 1. Calculate Weighted Error
            incorrect = (preds != y)
            err = np.sum(weights * incorrect)
            err = np.clip(err, 1e-10, 1 - 1e-10) # Prevent zero division

            # 2. Calculate Alpha (Amount of Say/Voting Power)
            alpha = 0.5 * np.log((1.0 - err) / err)

            # 3. Update Sample Weights (Misclassified ones grow heavier)
            weights *= np.exp(-alpha * y * preds)
            weight_sum = np.sum(weights)
            if weight_sum == 0: break
            weights /= weight_sum # Normalize

            self.stumps.append(stump)
            self.alphas.append(alpha)

            if stump.root.feature_idx is not None:
                total_feat_importance[stump.root.feature_idx] += abs(alpha)

        total_importance = np.sum(total_feat_importance)
        self.feature_importances_ = total_feat_importance / total_importance if total_importance > 0 else np.zeros(n_features)
        return self

    def predict(self, X):
        scores = np.zeros(X.shape[0])
        for stump, alpha in zip(self.stumps, self.alphas):
            tree_pred = stump.predict(X)
            pred = np.where(tree_pred == 1, 1, -1)
            # Add weighted vote
            scores += alpha * pred
        return np.where(scores >= 0, 1, -1)

# ==============================================================================
# 6. TRAIN AND EVALUATE MODELS (CONVERGENCE ANALYSIS)
# ==============================================================================
n_estimators_list = [5, 10, 20, 30, 50]
bag_cum_acc, ab_cum_acc = [], []
# Train Final Full Models
bag_model = BaggingClassifierScratch(n_estimators=50, max_depth=3, max_features='sqrt')
bag_model.fit(X_train, y_train_bg)
bag_acc = np.mean(bag_model.predict(X_test) == y_test_bg) * 100

ab_model = AdaBoostClassifierScratch(n_estimators=50)
ab_model.fit(X_train, y_train_ab)
ab_acc = np.mean(ab_model.predict(X_test) == y_test_ab) * 100

# Compute cumulative accuracy for chart
for n in n_estimators_list:
    # Bagging Cum
    t_bag = BaggingClassifierScratch(n_estimators=n, max_depth=3, max_features='sqrt')
    t_bag.fit(X_train, y_train_bg)
    bag_cum_acc.append(np.mean(t_bag.predict(X_test) == y_test_bg) * 100)
    
    # AdaBoost Cum (Re-using trained stumps up to n)
    t_ab_scores = np.zeros(X_test.shape[0])
    for i in range(min(n, len(ab_model.stumps))):
        tree_pred = ab_model.stumps[i].predict(X_test)
        pred = np.where(tree_pred == 1, 1, -1)
        t_ab_scores += ab_model.alphas[i] * pred
        
    t_ab_preds = np.where(t_ab_scores >= 0, 1, -1)
    ab_cum_acc.append(np.mean(t_ab_preds == y_test_ab) * 100)

# ==============================================================================
# 7. VISUALIZATIONS AND CHARTS (ACADEMIC BLUE PALETTE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

#  CHART 1: ENSEMBLE CONVERGENCE 
axes[0].plot(
    n_estimators_list, bag_cum_acc, 
    marker='o', color='#2c3e50', linewidth=3, markersize=8, 
    label=f'Bagging (Parallel Voting) | Final: {bag_acc:.1f}%'
)
axes[0].plot(
    n_estimators_list, ab_cum_acc, 
    marker='s', color='#2980b9', linewidth=3, markersize=8, 
    label=f'AdaBoost (Sequential Boosting) | Final: {ab_acc:.1f}%'
)

axes[0].set_xlabel('Number of Ensemble Iterations / Estimators ($T$)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Test Accuracy (%)', fontsize=12, fontweight='bold')
axes[0].set_title('1. Ensemble Convergence: Bagging vs Boosting\n(Stabilization of Prediction Accuracy)', fontsize=13, fontweight='bold', pad=12)
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='lower right')
axes[0].grid(True, linestyle='--', alpha=0.6)
axes[0].set_xticks(n_estimators_list)

#  CHART 2: FEATURE IMPORTANCE 
top_k = 4
bag_top_idx = np.argsort(bag_model.feature_importances_)[-top_k:][::-1]
plot_features = [feature_names[i] for i in bag_top_idx]
x = np.arange(len(plot_features))
width = 0.35

bars1 = axes[1].bar(
    x - width / 2, bag_model.feature_importances_[bag_top_idx], 
    width, label='Bagging (Gini Split Frequency)', color='#2c3e50', edgecolor='black', alpha=0.9
)
bars2 = axes[1].bar(
    x + width / 2, ab_model.feature_importances_[bag_top_idx], 
    width, label='AdaBoost (Weighted Alpha Contribution)', color='#2980b9', edgecolor='black', alpha=0.9
)

axes[1].set_ylabel('Normalized Feature Importance', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Uber Environmental / Demand Features', fontsize=12, fontweight='bold')
axes[1].set_title('2. Feature Importance Extraction\n(How each Ensemble values data differently)', fontsize=13, fontweight='bold', pad=12)
axes[1].set_xticks(x)
axes[1].set_xticklabels(plot_features, fontsize=11, fontweight='bold')
axes[1].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
axes[1].grid(axis='y', linestyle='--', alpha=0.7)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 8: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    "<b>نتیجه استراتژیک:</b> بررسی گراف ۱ نشان می‌دهد که هر دو متد گروهی، دقت اولیه مدل‌ها را به میزان قابل "
    "توجهی بهبود داده‌اند. در حالی که AdaBoost با تمرکز پی‌درپی روی داده‌های سخت‌تر، دقت بالاتری را نسبت به "
    "Bagging ثبت کرده است، اما گراف ۲ تفاوت عمیق این دو فلسفه را نشان می‌دهد: مدل Bagging با رأی‌گیری "
    "موازی متوجه شده است که <code>Ride Requests</code> (تقاضا) مهم‌ترین عامل تغییر قیمت است. اما "
    "مدل AdaBoost که روی تصحیح خطاهای حاشیه‌ای (Bias) تمرکز دارد، وزن بسیار بالایی به <code>Weather Severity</code> "
    "(آب و هوا) داده است. این یعنی مدل بوستینگ دریافته است که در حالت‌های عادی، تقاضا قیمت را تعیین می‌کند، "
    "اما در لبه‌های پیچیده و غیرخطی تصمیم‌گیری (مثلاً وقتی تقاضا متوسط است اما باران شدید می‌بارد)، این "
    "وضعیت جوی است که باعث شروع Surge Pricing می‌شود. اوبر از این ترکیب برای ساخت مدل‌های مقاوم و در عین "
    "حال به شدت دقیق استفاده می‌کند."
)
display_academic_text(insight_conclusion, "insight")
```

<li>پیاده‌سازی کامل یادگیرهای پایه و روش‌های گروهی از صفر (From Scratch):
  <ul>
    <li><b>درخت تصمیم و استامپ:</b> کلاس <code>DecisionTreeScratch</code> با معیار گینی پیاده‌سازی شد. برای بوستینگ، از کلاس مشتق‌شده <code>DecisionStump</code> (درختی با عمق ۱) به عنوان یادگیر ضعیف استفاده شد. همچنین قابلیت محاسبه گینی وزن‌دار (<code>weights</code>) به درخت اضافه شد تا AdaBoost بتواند روی نمونه‌های سخت‌تر تمرکز کند.</li>
    <li><b>بگینگ (Bagging):</b> کلاس <code>BaggingClassifierScratch</code> در هر تکرار با <code>np.random.choice(..., replace=True)</code> یک نمونه بوت‌استرپ می‌سازد و یک درخت با عمق ۳ روی آن آموزش می‌دهد. اهمیت ویژگی‌ها بر اساس تعداد دفعات استفاده برای شکست گره‌ها محاسبه می‌شود.</li>
    <li><b>بوستینگ (AdaBoost):</b> کلاس <code>AdaBoostClassifierScratch</code> الگوریتم را به صورت ترتیبی پیاده‌سازی می‌کند. در هر گام، خطای وزن‌دار ($\epsilon$) محاسبه شده، وزن مدل ($\alpha$) تعیین می‌گردد و وزن نمونه‌های اشتباه با فرمول نمایی افزایش می‌یابد. اهمیت ویژگی‌ها در اینجا بر اساس مجموع $\alpha$های مدل‌هایی است که از آن ویژگی برای شکست استفاده کرده‌اند.</li>
  </ul>
</li>
<li>تفسیر خروجی و نمودارها:
  <ul>
    <li>نمودار سمت چپ (Convergence Curve): روند افزایش دقت را با افزودن مدل‌ها نشان می‌دهد. بگینگ (نارنجی) معمولاً سریع‌تر به سقف دقت خود می‌رسد و ثابت می‌ماند (کاهش واریانس). اما AdaBoost (سبز) ممکن است در ابتدا کندتر رشد کند اما با تمرکز تدریجی روی نمونه‌های سخت، در نهایت به دقت بسیار بالایی دست می‌یابد (کاهش بایاس).</li>
    <li>نمودار سمت راست (Feature Importance): مقایسه دیدگاه دو الگوریتم نسبت به ویژگی‌ها. بگینگ ویژگی‌هایی را مهم‌تر می‌داند که بیشترین سهم را در شکست‌های درخت‌های عمیق‌تر داشته‌اند، در حالی که AdaBoost ویژگی‌هایی را برجسته می‌کند که در حل نمونه‌های پیچیده و اشتباه‌شده توسط مدل‌های قبلی (دارای $\alpha$ بالا) نقش کلیدی داشته‌اند.</li>
  </ul>
</li>
