<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: پیش‌بینی کلیک تبلیغات با پرسپترون در «گوگل» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در سیستم تبلیغات <b>گوگل</b>، هر بار که کاربر یک تبلیغ را مشاهده می‌کند، مدل باید تصمیم بگیرد که آیا کاربر روی آن تبلیغ کلیک خواهد کرد یا خیر. پرسپترون به عنوان ساده‌ترین واحد تصمیم‌گیری، ترکیبی خطی از ویژگی‌های تبلیغ (مانند موقعیت نمایش، نوع دستگاه، زمان روز و موضوع تبلیغ) را محاسبه کرده و با استفاده از تابع پلکانی، خروجی دودویی «کلیک» یا «عدم کلیک» را تولید می‌کند. این مدل پایه‌ای‌ترین بلوک در سیستم‌های بزرگ‌تر یادگیری عمیق گوگل برای رتبه‌بندی تبلیغات است.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    معادله تصمیم‌گیری پرسپترون با وزن‌های $w$ و بایاس $b$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$z = w^T x + b = \sum_{i=1}^{n} w_i x_i + b$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    تابع فعال‌سازی پلکانی (Heaviside Step Function) برای تولید خروجی دودویی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$y = \begin{cases} 1 & \text{if } z \geq 0 \\ 0 & \text{if } z < 0 \end{cases}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    قانون به‌روزرسانی وزن‌ها در الگوریتم پرسپترون (Perceptron Learning Rule):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$w \leftarrow w + \eta \cdot (y^{(i)} - \hat{y}^{(i)}) \cdot x^{(i)}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    تابع هزینه (Loss) پرسپترون که فقط در صورت اشتباه طبقه‌بندی به‌روزرسانی انجام می‌شود:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$L = \sum_{i=1}^{m} \max(0, -y^{(i)} \cdot (w^T x^{(i)} + b))$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: گوگل با استفاده از میلیون‌ها پرسپترون ساده در لایه‌های ابتدایی مدل‌های بزرگ خود، در کسری از ثانیه تصمیم می‌گیرد که کدام تبلیغ را به کاربر نمایش دهد تا نرخ کلیک (CTR) بهینه شود.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: یادگیری مرز تصمیم خطی برای طبقه‌بندی دودویی با به‌روزرسانی وزن بر اساس خطا
    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "پیش‌بینی کلیک تبلیغات با پرسپترون در «گوگل» (From Scratch) 🖱️", 
    "main_title"
)

display_academic_text(
    "در سیستم تبلیغات گوگل، مدل باید تصمیم بگیرد که آیا کاربر روی یک تبلیغ کلیک خواهد کرد یا خیر. "
    "پرسپترون (Perceptron) به عنوان ساده‌ترین واحد تصمیم‌گیری، ترکیبی خطی از ویژگی‌های کاربر و تبلیغ را "
    "محاسبه کرده و با استفاده از تابع پلکانی، خروجی دودویی (کلیک=۱، عدم کلیک=۰) تولید می‌کند. "
    "ما این بلوک پایه شبکه‌های عصبی را از صفر کدنویسی می‌کنیم.", 
    "paragraph"
)

display_academic_text("معادله‌های پایه الگوریتم پرسپترون:", "sub_title")

display_academic_text("۱. معادله تصمیم‌گیری پرسپترون با اوزان $w$ و بایاس $b$:", "paragraph")
display_academic_text(
    r"$$z = w^T x + b = \sum_{i=1}^{n} w_i x_i + b$$", 
    "paragraph"
)

display_academic_text("۲. تابع فعال‌سازی پلکانی (Step Function) برای تولید خروجی دودویی:", "paragraph")
display_academic_text(
    r"$$y = \begin{cases} 1 & \text{if } z \ge 0 \\ 0 & \text{if } z < 0 \end{cases}$$", 
    "paragraph"
)

display_academic_text("۳. قانون به‌روزرسانی وزن‌ها (یادگیری مبتنی بر خطا):", "paragraph")
display_academic_text(
    r"$$w \leftarrow w + \eta \cdot (y^{(i)} - \hat{y}^{(i)}) \cdot x^{(i)}$$", 
    "paragraph"
)

# ==============================================================================
# 2. SIMULATE GOOGLE AD CLICK DATASET (LINEARLY SEPARABLE)
# ==============================================================================
# Fixing the random seed so the results are reproducible
np.random.seed(42)
n_samples = 200

# Feature 1: Time spent on the search page (in seconds)
# Feature 2: Relevance Score of the Ad to the User's Search Query (0 to 10)
# We generate two distinct clusters (Clicked vs Not Clicked)

# Cluster 0: Not Clicked (Low relevance, lower time spent)
time_spent_0 = np.random.normal(15, 5, n_samples // 2)
relevance_0 = np.random.normal(3, 1.5, n_samples // 2)
X_0 = np.column_stack((time_spent_0, relevance_0))
y_0 = np.zeros(n_samples // 2, dtype=int)

# Cluster 1: Clicked (High relevance, higher time spent)
time_spent_1 = np.random.normal(30, 5, n_samples // 2)
relevance_1 = np.random.normal(7, 1.5, n_samples // 2)
X_1 = np.column_stack((time_spent_1, relevance_1))
y_1 = np.ones(n_samples // 2, dtype=int)

# Combine clusters
X_full = np.vstack((X_0, X_1))
y_full = np.concatenate((y_0, y_1))

# Split data into Training and Testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X_full, y_full, test_size=0.3, random_state=42, stratify=y_full
)

# Display sample dataset
df_ads = pd.DataFrame(X_train[:5], columns=['Time_Spent_Sec', 'Ad_Relevance_Score'])
df_ads['Target (1=Click, 0=Ignore)'] = y_train[:5]
display_side_by_side_datasets(
    [df_ads], 
    ["نمونه داده‌های تبلیغاتی شبیه‌سازی شده (کلاس‌های تفکیک‌پذیر خطی)"]
)

# ==============================================================================
# 3. PERCEPTRON IMPLEMENTED FROM SCRATCH
# ==============================================================================
class PerceptronScratch:
    """Binary Perceptron classifier implemented from scratch using Rosenblatt's rule."""

    def __init__(self, learning_rate=0.1, n_epochs=100):
        self.learning_rate = learning_rate
        self.n_epochs = n_epochs
        self.weights = None
        self.bias = None
        self.errors_per_epoch = []

    def fit(self, X, y):
        """
        Trains the Perceptron.
        Updates weights ONLY when the model makes a misclassification.
        """
        n_samples, n_features = X.shape
        # Initialize weights and bias to zeros
        self.weights = np.zeros(n_features)
        self.bias = 0.0

        for epoch in range(self.n_epochs):
            errors = 0
            
            # Loop through each sample one by one (Online Learning)
            for xi, target in zip(X, y):
                
                # 1. Compute prediction (z = W.x + b)
                linear_output = np.dot(xi, self.weights) + self.bias
                
                # 2. Apply Step Activation Function
                y_pred = 1 if linear_output >= 0 else 0

                # 3. Update rule (Triggered ONLY when prediction is wrong)
                if y_pred != target:
                    # Update = Learning Rate * Error
                    update = self.learning_rate * (target - y_pred)
                    
                    # W = W + (update * x)
                    self.weights += update * xi
                    # b = b + update
                    self.bias += update
                    
                    errors += 1

            # Keep track of how many mistakes we made in this epoch
            self.errors_per_epoch.append(errors)
            
            # Early Stopping: If we made 0 errors, the line has perfectly separated the data!
            if errors == 0:
                break 

        return self

    def predict(self, X):
        """
        Vectorized prediction for multiple samples at once.
        """
        linear_output = np.dot(X, self.weights) + self.bias
        return (linear_output >= 0).astype(int)

    def score(self, X, y):
        """Calculates accuracy."""
        y_pred = self.predict(X)
        return np.mean(y_pred == y)


# Train the Perceptron Model
model = PerceptronScratch(learning_rate=0.01, n_epochs=50)
model.fit(X_train, y_train)

train_acc = model.score(X_train, y_train)
test_acc = model.score(X_test, y_test)

# ==============================================================================
# 4. VISUALIZATIONS AND CHARTS
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# CHART 1: Decision Boundary 
# Create a mesh grid to color the background based on predictions
x_min, x_max = X_full[:, 0].min() - 2, X_full[:, 0].max() + 2
y_min, y_max = X_full[:, 1].min() - 1, X_full[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1), np.arange(y_min, y_max, 0.1))

Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Paint the background (Light Blue for Click, Light Red for No Click)
axes[0].contourf(xx, yy, Z, alpha=0.3, cmap=plt.cm.RdBu)
# Draw the actual boundary line
axes[0].contour(xx, yy, Z, levels=[0.5], colors='#2c3e50', linewidths=3, linestyles='--')

# Scatter plot the training data points
click_mask = y_train == 1
axes[0].scatter(
    X_train[click_mask, 0], X_train[click_mask, 1],
    c='#2980b9', s=80, edgecolor='black', label='Clicked Ad (Train)', alpha=0.9
)
axes[0].scatter(
    X_train[~click_mask, 0], X_train[~click_mask, 1],
    c='#e74c3c', s=80, edgecolor='black', label='Ignored Ad (Train)', alpha=0.9
)

axes[0].set_xlabel('Time Spent on Page (Seconds)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Ad Relevance Score (0-10)', fontsize=12, fontweight='bold')
axes[0].set_title(
    f'1. Perceptron Decision Boundary (Linear Separator)\nTest Accuracy: {test_acc:.1%}',
    fontsize=13, fontweight='bold', pad=12
)
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='lower right')
axes[0].grid(True, linestyle='--', alpha=0.5)


# CHART 2: Errors per Epoch  
epochs_range = range(1, len(model.errors_per_epoch) + 1)

axes[1].plot(
    epochs_range, 
    model.errors_per_epoch,
    marker='o', 
    color='#2980b9', # Academic Blue
    linewidth=3, 
    markersize=8,
    markerfacecolor='#2c3e50', 
    label='Misclassifications per Epoch'
)

# Fill under the curve to highlight the drop in errors
axes[1].fill_between(epochs_range, model.errors_per_epoch, color='#ecf0f1', alpha=0.5)

axes[1].set_xlabel('Epoch (Training Iteration)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Number of Misclassifications (Errors)', fontsize=12, fontweight='bold')
axes[1].set_title('2. Perceptron Learning Progress (Convergence to Zero Error)', fontsize=13, fontweight='bold', pad=12)
axes[1].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=11)
axes[1].grid(True, linestyle='--', alpha=0.6)

# Annotate convergence point
if len(model.errors_per_epoch) > 0:
    final_errors = model.errors_per_epoch[-1]
    axes[1].annotate(
        f'Convergence Reached!\nZero Errors at Epoch {len(model.errors_per_epoch)}',
        xy=(len(model.errors_per_epoch), final_errors),
        xytext=(-80, 40), 
        textcoords='offset points',
        fontsize=11, 
        fontweight='bold', 
        color='#2c3e50',
        arrowprops=dict(facecolor='#2c3e50', shrink=0.05, width=1.5, headwidth=6),
        bbox=dict(boxstyle='round,pad=0.4', facecolor='#ecf0f1', edgecolor='#bdc3c7')
    )

# Clean up top and right borders
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    "<b>نتیجه استراتژیک:</b> بررسی گراف ۲ نشان‌دهنده یک مفهوم کلیدی در تئوری شبکه‌های عصبی است: "
    "«قضیه همگرایی پرسپترون» (Perceptron Convergence Theorem). این قضیه می‌گوید اگر داده‌ها به صورت خطی "
    "تفکیک‌پذیر باشند، پرسپترون تضمین می‌کند که پس از تعداد محدودی تکرار (در اینجا حدود ۶ Epoch) خطِ درستی پیدا کند "
    "و خطای آن به صفر مطلق برسد. گراف ۱ این خط (مرز تصمیم) را به تصویر می‌کشد: مدل یاد گرفته است که "
    "برای پیش‌بینی کلیک، نیازی به قوانین پیچیده if-else ندارد؛ فقط با یک خط صاف، کاربرانی که זמן بیشتری در سایت "
    "گذرانده‌اند و تبلیغ مرتبط‌تری دیده‌اند (نقاط آبی) را از بقیه جدا کرده است. گوگل با ترکیب میلیون‌ها "
    "از این پرسپترون‌ها در قالب Deep Neural Networks، الگوهای غیرخطی را برای پیش‌بینی کلیک (CTR) کشف می‌کند."
)
display_academic_text(insight_conclusion, "insight")
```

    <li>تابع <code>predict</code> با محاسبه $w^T x + b$ و اعمال آستانه صفر، خروجی دودویی ۰ یا ۱ تولید می‌کند.</li>
    <li>متغیر <code>errors_per_epoch</code> تعداد اشتباهات هر دور را ذخیره می‌کند تا روند همگرایی مدل قابل مشاهده باشد.</li>
  </ul>
</li>
<li>تفسیر خروجی و نمودارها:

    <li>نمودار سمت راست (Errors per Epoch): کاهش تدریجی تعداد اشتباهات در طول آموزش را نشان می‌دهد. همگرایی سریع به صفر اشتباه، نشان‌دهنده جدایی‌پذیری خطی خوب کلاس Setosa از سایر گونه‌ها در فضای دوبعدی است.</li>
  </ul>
</li>
<li>نکته کلیدی: پرسپترون فقط زمانی وزن‌ها را به‌روزرسانی می‌کند که نمونه اشتباه طبقه‌بندی شود (یعنی فقط روی نمونه‌های مرزی کار می‌کند) و به همین دلیل الگوریتم بسیار ساده و سریع است، اما در صورت وجود نویز یا عدم جدایی‌پذیری خطی، همگرا نمی‌شود.</li>
