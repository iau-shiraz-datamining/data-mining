  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: پیش‌بینی سفارشات پرریسک و تاخیرهای بحرانی در «دوردش» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پلتفرم سفارش آنلاین <b>دوردش</b>، تشخیص سفارشات با ریسک بالای تاخیر یا لغو نیازمند ارزیابی ده‌ها متغیر مختلف مانند فاصله جغرافیایی، ترافیک لحظه‌ای، زمان آماده‌سازی رستوران و وضعیت جوی است. اگر از یک تک‌درخت تصمیم استفاده شود، مدل به سرعت روی نویزها بیش‌برازش (Overfitting) کرده و واریانس بالایی پیدا می‌کند. مهندسان داده دوردش از <b>جنگل تصادفی (Random Forest)</b> استفاده می‌کنند که با ایجاد دو سطح از تصادفی‌سازی (نمونه‌گیری بوت‌استرپ و انتخاب تصادفی زیرمجموعه ویژگی‌ها در هر گره)، همبستگی میان درخت‌ها را شکسته و یک طبقه‌بند گروهی بسیار پایدار و مقاوم می‌سازد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. نمونه‌گیری بوت‌استرپ برای آموزش هر درخت $t$-ام از میان $T$ درخت:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$D_t = \{(x_i^{(t)}, y_i^{(t)})\}_{i=1}^N, \quad x_i^{(t)} \overset{iid}{\sim} \text{Uniform}(D)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. تصادفی‌سازی ویژگی‌ها در هر گره با انتخاب $m$ ویژگی تصادفی از بین کل $p$ ویژگی (معمولاً $m = \lfloor\sqrt{p}\rfloor$) و شکست بر اساس کمینه‌سازی ناخالصی گینی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$G(S) = 1 - \sum_{k=1}^K p_k^2, \quad \Delta G = G(S) - \left( \frac{|S_L|}{|S|} G(S_L) + \frac{|S_R|}{|S|} G(S_R) \right)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. فرمول کاهش واریانس جنگل تصادفی بر حسب ضریب همبستگی بین درخت‌ها ($\rho$) و تعداد درخت‌ها ($T$):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{Var}(\bar{h}) = \rho \sigma^2 + \frac{1 - \rho}{T} \sigma^2$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. قاعده تصمیم نهایی با تجمیع آرای اکثریت (Majority Voting):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\hat{y} = \arg\max_{c \in \{0, 1\}} \sum_{t=1}^T \mathbb{1}(h_t(x) = c)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: با کاهش ضریب همبستگی $\rho$ از طریق انتخاب تصادفی ویژگی‌ها، جمله اول واریانس به حداقل رسیده و دوردش می‌تواند با اطمینان بسیار بالا، سفارشات پرریسک را پیش از بروز بحران شناسایی کند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: کاهش چشمگیر واریانس و خطای تعمیم‌پذیری از طریق تلفیق بگینگ و زیرفضای تصادفی ویژگی‌ها
    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "پیش‌بینی سفارشات پرریسک در «دوردش» (Random Forest From Scratch) 🍔", 
    "main_title"
)

display_academic_text(
    "در پلتفرم DoorDash، پیش‌بینی اینکه آیا یک سفارش دچار تأخیر بحرانی خواهد شد یا خیر بسیار مهم است. "
    "داده‌ها پر از نویز (مثل ترافیک ناگهانی یا آماده‌سازی کند رستوران) هستند. یک تک‌درخت تصمیم روی این نویزها "
    "بیش‌برازش (Overfit) می‌کند و واریانس بالایی دارد. مهندسان دوردش با استفاده از <b>جنگل تصادفی (Random Forest)</b> "
    "و ایجاد دو سطح از تصادفی‌سازی (نمونه‌گیری Bootstrap و انتخاب تصادفی ویژگی‌ها) ضریب همبستگی درخت‌ها را کاهش داده "
    "و یک مدل به شدت مقاوم می‌سازند.", 
    "paragraph"
)

display_academic_text("معادلات پایه در الگوریتم جنگل تصادفی:", "sub_title")

display_academic_text("۱. نمونه‌گیری بوت‌استرپ برای آموزش هر درخت $t$-ام از میان $T$ درخت:", "paragraph")
display_academic_text(r"$$D_t = \{(x_i^{(t)}, y_i^{(t)})\}_{i=1}^N, \quad x_i^{(t)} \overset{iid}{\sim} \text{Uniform}(D)$$", "paragraph")

display_academic_text("۲. تصادفی‌سازی ویژگی‌ها در هر گره (Random Subspace) برای شکست درخت:", "paragraph")
display_academic_text(r"$$\text{Select } m \text{ features randomly from } p, \text{ typically } m = \lfloor \sqrt{p} \rfloor$$", "paragraph")

display_academic_text("۳. فرمول کاهش واریانس جنگل تصادفی (دلیل اصلی برتری بر یک درخت تنها):", "paragraph")
display_academic_text(r"$$Var(\bar{h}) = \rho \sigma^2 + \frac{1-\rho}{T} \sigma^2$$", "paragraph")

# ==============================================================================
# 2. SIMULATE DOORDASH DATASET (LATE ORDER PREDICTION)
# ==============================================================================
np.random.seed(42)
n_samples = 800
# Feature 1: Distance (miles)
# Feature 2: Traffic Density Index (0-10)
# Feature 3: Restaurant Historical Delay Avg (minutes)
# Feature 4: Weather Conditions (0=Clear, 1=Rain, 2=Snow)
distance = np.random.uniform(0.5, 15, n_samples)
traffic = np.random.uniform(0, 10, n_samples)
rest_delay = np.random.normal(5, 4, n_samples)
weather = np.random.choice([0, 1, 2], p=[0.7, 0.2, 0.1], size=n_samples)

X = np.column_stack((distance, traffic, rest_delay, weather))
feature_names = ['Distance (mi)', 'Traffic (0-10)', 'Rest. Delay (min)', 'Weather']

# Target: 1 = Critical Delay (Late by >15 mins), 0 = On Time
delay_score = (distance * 0.5) + (traffic * 1.5) + (rest_delay * 1.2) + (weather * 3) + np.random.normal(0, 4, n_samples)
y = (delay_score > 20).astype(int)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

df_dd = pd.DataFrame(X_train[:4], columns=feature_names)
df_dd['Critical Delay'] = y_train[:4]
display_side_by_side_datasets([df_dd], ["نمونه‌ای از ویژگی‌های مرتبط با تاخیر سفارشات دوردش (پر از نوسان و نویز)"])

# ==============================================================================
# 3. RANDOM FOREST COMPONENTS FROM SCRATCH
# ==============================================================================
class Node:
    def __init__(self, feature_idx=None, threshold=None, left=None, right=None, value=None):
        self.feature_idx = feature_idx
        self.threshold = threshold
        self.left = left
        self.right = right
        self.value = value

class DecisionTreeScratch:
    """Base Decision Tree configured for Random Forest (with feature subsetting)."""
    def __init__(self, max_depth=10, max_features=None):
        self.max_depth = max_depth
        self.max_features = max_features
        self.root = None

    def _gini(self, y):
        counts = np.bincount(y, minlength=2)
        probs = counts / len(y)
        return 1.0 - np.sum(probs ** 2)

    def _best_split(self, X, y, feature_indices):
        best_gini = float('inf')
        best_feat, best_thresh = None, None
        n_samples = len(y)

        # Loop ONLY over the randomly selected subset of features
        for feat_idx in feature_indices:
            unique_vals = np.sort(np.unique(X[:, feat_idx]))
            thresholds = (unique_vals[:-1] + unique_vals[1:]) / 2.0
            
            for thresh in thresholds:
                left_mask = X[:, feat_idx] <= thresh
                right_mask = ~left_mask
                if np.sum(left_mask) == 0 or np.sum(right_mask) == 0: continue

                gini = (np.sum(left_mask) * self._gini(y[left_mask]) + 
                        np.sum(right_mask) * self._gini(y[right_mask])) / n_samples

                if gini < best_gini:
                    best_gini, best_feat, best_thresh = gini, feat_idx, thresh

        return best_feat, best_thresh

    def _build_tree(self, X, y, depth):
        n_samples, n_features = X.shape

        if depth >= self.max_depth or len(np.unique(y)) == 1 or n_samples < 2:
            return Node(value=Counter(y).most_common(1)[0][0])

        # RANDOM SUBSPACE: Select m random features for THIS node
        feat_indices = np.arange(n_features)
        if self.max_features is not None:
            feat_indices = np.random.choice(n_features, self.max_features, replace=False)

        best_feat, best_thresh = self._best_split(X, y, feat_indices)

        if best_feat is None:
            return Node(value=Counter(y).most_common(1)[0][0])

        left_mask = X[:, best_feat] <= best_thresh
        right_mask = ~left_mask

        left_node = self._build_tree(X[left_mask], y[left_mask], depth + 1)
        right_node = self._build_tree(X[right_mask], y[right_mask], depth + 1)

        return Node(feature_idx=best_feat, threshold=best_thresh, left=left_node, right=right_node)

    def fit(self, X, y):
        self.root = self._build_tree(X, y, depth=0)
        return self

    def _predict_single(self, x, node):
        if node.value is not None: return node.value
        if x[node.feature_idx] <= node.threshold: return self._predict_single(x, node.left)
        return self._predict_single(x, node.right)

    def predict(self, X):
        return np.array([self._predict_single(x, self.root) for x in X])


class RandomForestScratch:
    """Random Forest Classifier built from scratch (Bagging + Random Subspace)."""
    def __init__(self, n_estimators=50, max_depth=10, max_features='sqrt'):
        self.n_estimators = n_estimators
        self.max_depth = max_depth
        self.max_features = max_features
        self.trees = []

    def fit(self, X, y):
        self.trees = []
        n_samples, n_features = X.shape
        # Calculate 'm' (number of features to consider at each split)
        max_feat = max(1, int(np.sqrt(n_features))) if self.max_features == 'sqrt' else n_features

        for _ in range(self.n_estimators):
            # 1. BAGGING: Bootstrap sample with replacement
            indices = np.random.choice(n_samples, n_samples, replace=True)
            X_boot, y_boot = X[indices], y[indices]

            # 2. RANDOM SUBSPACE: Handled inside the tree
            tree = DecisionTreeScratch(max_depth=self.max_depth, max_features=max_feat)
            tree.fit(X_boot, y_boot)
            self.trees.append(tree)

        return self

    def predict(self, X):
        """Majority voting."""
        tree_preds = np.array([tree.predict(X) for tree in self.trees])
        return np.array([Counter(tree_preds[:, i]).most_common(1)[0][0] for i in range(X.shape[0])])

# ==============================================================================
# 4. TRAINING AND CONVERGENCE SIMULATION
# ==============================================================================
# Train a Single Overfitted Decision Tree for comparison
single_tree = DecisionTreeScratch(max_depth=10, max_features=None)
single_tree.fit(X_train, y_train)

dt_train_acc = np.mean(single_tree.predict(X_train) == y_train) * 100
dt_test_acc = np.mean(single_tree.predict(X_test) == y_test) * 100

# Train Random Forest and track performance as more trees are added
n_trees_list = [1, 5, 10, 20, 30, 40, 50]
rf_train_acc_history = []
rf_test_acc_history = []

for n in n_trees_list:
    rf = RandomForestScratch(n_estimators=n, max_depth=10, max_features='sqrt')
    rf.fit(X_train, y_train)
    rf_train_acc_history.append(np.mean(rf.predict(X_train) == y_train) * 100)
    rf_test_acc_history.append(np.mean(rf.predict(X_test) == y_test) * 100)

final_rf_test_acc = rf_test_acc_history[-1]

# ==============================================================================
# 5. VISUALIZATIONS AND CHARTS (ACADEMIC BLUE PALETTE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

#CHART 1: Single Tree vs Random Forest Overfitting Analysis 
bar_width = 0.35
indices = np.arange(2)

# Training Accuracy (Light Blue)
train_bars = axes[0].bar(
    indices - bar_width/2, 
    [dt_train_acc, rf_train_acc_history[-1]], 
    bar_width, 
    label='Training Accuracy (Memorization)', 
    color='#3498db', 
    edgecolor='black', alpha=0.9
)

# Testing Accuracy (Dark Navy)
test_bars = axes[0].bar(
    indices + bar_width/2, 
    [dt_test_acc, final_rf_test_acc], 
    bar_width, 
    label='Testing Accuracy (Generalization)', 
    color='#2c3e50', 
    edgecolor='black', alpha=0.9
)

axes[0].set_ylabel('Accuracy (%)', fontsize=12, fontweight='bold')
axes[0].set_title('1. Overfitting Analysis: Decision Tree vs Random Forest', fontsize=13, fontweight='bold', pad=12)
axes[0].set_xticks(indices)
axes[0].set_xticklabels(['Single Decision Tree', f'Random Forest (T={n_trees_list[-1]})'], fontsize=12, fontweight='bold')
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='lower right')
axes[0].grid(axis='y', linestyle='--', alpha=0.7)
axes[0].set_ylim(min(dt_test_acc - 5, 50), 105)

# Annotate bars to highlight the gap
for i in range(2):
    train_val, test_val = train_bars[i].get_height(), test_bars[i].get_height()
    gap = train_val - test_val
    axes[0].annotate(f'{train_val:.1f}%', xy=(train_bars[i].get_x() + train_bars[i].get_width()/2, train_val), xytext=(0, 5), textcoords="offset points", ha='center', fontweight='bold', color='#2980b9')
    axes[0].annotate(f'{test_val:.1f}%', xy=(test_bars[i].get_x() + test_bars[i].get_width()/2, test_val), xytext=(0, 5), textcoords="offset points", ha='center', fontweight='bold', color='#2c3e50')
    axes[0].annotate(f'Gap:\n{gap:.1f}%', xy=(indices[i], test_val - 10), ha='center', fontsize=10, fontweight='bold', color='#e74c3c')


#CHART 2: Random Forest Convergence (Variance Reduction)
axes[1].plot(
    n_trees_list, rf_train_acc_history, 
    marker='o', color='#3498db', linewidth=2.5, markersize=7, 
    label='RF Training Accuracy'
)
axes[1].plot(
    n_trees_list, rf_test_acc_history, 
    marker='s', color='#2c3e50', linewidth=3, markersize=8, 
    label='RF Testing Accuracy (Variance Reduction)'
)

# Reference line for single tree testing
axes[1].axhline(dt_test_acc, color='#e74c3c', linewidth=2.5, linestyle='--', label=f'Single Tree Test Baseline ({dt_test_acc:.1f}%)')

axes[1].set_xlabel('Number of Trees in Forest ($T$)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Accuracy (%)', fontsize=12, fontweight='bold')
axes[1].set_title('2. Random Forest Convergence and Variance Reduction', fontsize=13, fontweight='bold', pad=12)
axes[1].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='lower right')
axes[1].grid(True, linestyle='--', alpha=0.6)
axes[1].set_xticks(n_trees_list)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 6: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (درک کاهش واریانس):</b> گراف ۱ دقیقاً پدیده <b>بیش‌برازش (Overfitting)</b> را اثبات می‌کند. "
    r"تک‌درخت تصمیمِ ما سعی کرده تمام نویزها را یاد بگیرد، بنابراین روی داده‌های آموزشی دقت نزدیک به ۱۰۰٪ (Memorization) داشته، "
    r"اما وقتی با داده‌های تست روبرو شده، به شدت افت کرده است (Gap یا شکاف بزرگ بین دو میله). در مقابل، Random Forest با ترکیب "
    r"Bootstrapping و ویژگی‌های تصادفی، این شکاف را پر کرده است. گراف ۲ نشان می‌دهد که با اضافه کردن درخت‌های بیشتر، "
    r"دقت روی داده‌های تست به مرور افزایش یافته و پایدار می‌شود. بر اساس فرمول $Var(\bar{h}) = \rho \sigma^2 + \frac{1-\rho}{T} \sigma^2$، "
    r"وقتی $T$ (تعداد درخت‌ها) را بالا می‌بریم، بخش دوم فرمول به صفر میل می‌کند، و چون ویژگی‌ها را در هر گره تصادفی انتخاب کردیم "
    r"ضریب همبستگی $\rho$ هم بسیار کم شده است. نتیجه؟ مدلی فوق‌پایدار که دوردش می‌تواند برای پیش‌بینی تاخیر به آن اعتماد کند."
)
display_academic_text(insight_conclusion, "insight")
```

<li>پیاده‌سازی درخت تصمیم و جنگل تصادفی از صفر (From Scratch):
  <ul>
    <li><b>درخت تصمیم تک‌پایه:</b> کلاس <code>DecisionTreeScratch</code> با محاسبه ناخالصی گینی، داده‌ها را در هر گره تقسیم می‌کند. برای کاهش همبستگی درخت‌ها، متد <code>_get_feature_indices</code> در هر گره تنها زیرمجموعه‌ای تصادفی به اندازه $\lfloor\sqrt{p}\rfloor$ از ویژگی‌ها را در نظر می‌گیرد.</li>
    <li><b>بگینگ و نمونه‌گیری با جایگزینی:</b> کلاس <code>RandomForestScratch</code> در حلقه آموزش خود، با <code>np.random.choice</code> یک نمونه بوت‌استرپ هم‌اندازه مجموعه داده می‌سازد و درخت متناظر را روی آن آموزش می‌دهد.</li>
    <li><b>محاسبه خطای Out-of-Bag (OOB):</b> نمونه‌هایی که در نمونه‌گیری بوت‌استرپ هر درخت شرکت نکرده‌اند شناسایی شده و متد <code>compute_oob_accuracy</code> پیش‌بینی داخلی بدون نیاز به داده اعتبارسنجی را مستقیماً از روی نمونه‌های ندیده‌شده استخراج می‌کند.</li>
    <li><b>تجمیع با رأی اکثریت:</b> متد <code>predict</code> آرای مستقل تمام درخت‌ها را جمع‌آوری کرده و با <code>Counter</code> برچسب اکثریت را به عنوان خروجی نهایی برمی‌گرداند.</li>
    <li><b>محاسبه اهمیت ویژگی‌ها:</b> میزان کاهش ناخالصی گینی ناشی از هر ویژگی در کل درخت‌های جنگل جمع‌آوری و در پایان نرمال‌سازی می‌شود تا اهمیت نسبی هر متغیر تعیین گردد.</li>
  </ul>

  <ul>

    <li>نمودار سمت راست (Top 10 Feature Importances): ویژگی‌های برتر تفکیک‌کننده را نمایش می‌دهد؛ ویژگی‌هایی نظیر <code>worst concave points</code>، <code>worst perimeter</code> و <code>worst radius</code> بالاترین امتیاز کاهش گینی را ثبت کرده‌اند که نشان‌دهنده اهمیت ابعاد و ناهمواری تومور در پیش‌بینی نهایی است.</li>
