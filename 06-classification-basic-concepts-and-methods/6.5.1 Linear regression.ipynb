<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: پیش‌بینی قیمت خانه با رگرسیون خطی چندگانه در «زیلو» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پلتفرم ملکی <b>زیلو</b>، موتور ارزیابی قیمت (Zestimate) با استفاده از <b>رگرسیون خطی چندگانه (Multiple Linear Regression)</b> روی ویژگی‌هایی مانند میانگین درآمد منطقه، تعداد اتاق‌ها، سن بنا و فاصله تا مراکز شغلی، قیمت تخمینی هر خانه را محاسبه می‌کند. مدل با یافتن بهترین ضرایب خطی، رابطه‌ای ریاضی بین این ویژگی‌ها و قیمت نهایی ملک برقرار می‌کند که به خریداران و فروشندگان کمک می‌کند ارزش واقعی بازار را بشناسند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    معادله رگرسیون خطی چندگانه برای پیش‌بینی قیمت خانه با $n$ ویژگی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\hat{y} = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \cdots + \beta_n x_n = \beta_0 + \sum_{j=1}^{n} \beta_j x_j$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    تابع هزینه (Cost Function) مجموع مربعات خطا که مدل با کمینه کردن آن بهترین ضرایب را می‌یابد:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$J(\beta) = \frac{1}{2m} \sum_{i=1}^{m} \left( \hat{y}^{(i)} - y^{(i)} \right)^2 = \frac{1}{2m} \| X\beta - y \|^2$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    حل تحلیلی بسته (Normal Equation) برای یافتن مستقیم بهترین ضرایب $\beta$ بدون نیاز به تکرار:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\hat{\beta} = (X^T X)^{-1} X^T y$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ضریب تعیین $R^2$ برای سنجش میزان واریانس توضیح‌داده‌شده توسط مدل:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$R^2 = 1 - \frac{\sum_{i=1}^{m}(y^{(i)} - \hat{y}^{(i)})^2}{\sum_{i=1}^{m}(y^{(i)} - \bar{y})^2} = 1 - \frac{SS_{res}}{SS_{tot}}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    معیار ریشه میانگین مربعات خطا (RMSE) برای ارزیابی دقت پیش‌بینی در واحد اصلی قیمت:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$RMSE = \sqrt{\frac{1}{m} \sum_{i=1}^{m} \left( \hat{y}^{(i)} - y^{(i)} \right)^2}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: زیلو با اجرای این مدل روی میلیون‌ها ملک در سراسر آمریکا، تخمین قیمت لحظه‌ای ارائه می‌دهد که میانگین خطای آن به کمتر از ۲٪ از قیمت واقعی فروش رسیده است.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: برازش بهترین هایپرپلان خطی در فضای ویژگی‌های ملکی و پیش‌بینی دقیق قیمت بازار
    </div>
  </p>

</div>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from IPython.display import display, HTML
import warnings

warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={"axes.titlesize": 15, "axes.labelsize": 13, "xtick.labelsize": 11, "ytick.labelsize": 11})

plt.rcParams['figure.dpi'] = 120

def display_academic_text(text, type="normal"):
    styles = {
        "main_title": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 24px; "
            "font-weight: bold; text-align: center; border-bottom: 2px solid #34495e; "
            "padding-bottom: 15px; margin: 30px 0;"
        ),
        "sub_title": (
            "font-family: Tahoma, sans-serif; color: #34495e; font-size: 19px; "
            "font-weight: bold; border-right: 5px solid #2980b9; padding-right: 15px; "
            "margin: 40px 0px 20px 0px; background-color: #ecf0f1; padding: 12px; "
            "border-radius: 4px;"
        ),
        "paragraph": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 16px; "
            "margin-bottom: 20px; line-height: 1.8; text-align: justify;"
        ),
        "insight": (
            "font-family: Tahoma, sans-serif; color: #16a085; font-size: 16px; "
            "font-weight: bold; background-color: #e8f8f5; padding: 20px; "
            "border-left: 5px solid #1abc9c; margin: 30px 0; line-height: 1.8; "
            "text-align: justify;"
        )
    }
    display(HTML(f"<div dir='rtl' style='text-align: right; {styles[type]}'>{text}</div>"))

def get_protected_css():
    return """
    <style>
    .academic-table-wrapper { 
        direction: ltr !important; 
        overflow-x: auto !important; 
        width: 100% !important; 
        background-color: #fff; 
        border: 1px solid #bdc3c7; 
        border-radius: 4px; 
        box-shadow: 0 3px 6px rgba(0,0,0,0.08); 
    }
    table.strict-tbl { 
        border-collapse: collapse !important; 
        width: 100% !important; 
        margin: 0 !important; 
        font-family: inherit !important;
        font-size: 14px !important; 
        table-layout: auto !important; 
        border: none !important; 
    }
    table.strict-tbl thead th { 
        background-color: #2c3e50 !important; 
        color: #ffffff !important; 
        padding: 12px 18px !important; 
        text-align: center !important; 
        border: 1px solid #34495e !important; 
        white-space: nowrap !important; 
        font-weight: bold !important; 
    }
    table.strict-tbl tbody td { 
        padding: 10px 18px !important; 
        text-align: center !important; 
        border: 1px solid #bdc3c7 !important; 
        white-space: nowrap !important; 
        color: #2c3e50 !important; 
    }
    table.strict-tbl tbody tr:nth-child(even) td { 
        background-color: #f8f9fa !important; 
    }
    table.strict-tbl tbody tr:hover td { 
        background-color: #e8ecef !important; 
    }
    </style>
    """

def display_side_by_side_datasets(dfs, titles, *args):
    if not isinstance(dfs, list) and not isinstance(titles, list):
        dfs = [dfs]
        titles = [titles]
        
    col_html_list = []
    for i, (df, title) in enumerate(zip(dfs, titles)):
        table_html = df.to_html(index=False, classes=['strict-tbl'], border=0)
        bg_color = "#2980b9" if i % 2 == 0 else "#27ae60"
        col_html_list.append(f"""
        <div class="flex-column" style="flex: 1 1 0; min-width: 0; display: flex; flex-direction: column;">
            <div style="background-color: {bg_color}; color: white; padding: 12px; 
                        text-align: center; font-family: Tahoma, sans-serif; 
                        font-weight: bold; font-size: 15px; border-radius: 4px 4px 0 0;">
                {title}
            </div>
            <div class="academic-table-wrapper" style="margin-top: 0; border-top: none;">{table_html}</div>
        </div>
        """)
    
    display(HTML(get_protected_css() + f'<div style="display: flex; gap: 20px; width: 100%; margin: 20px 0 35px 0;">{"".join(col_html_list)}</div>'))

display(HTML("""
<style>
@import url('https://cdn.jsdelivr.net/gh/rastikerdar/vazirmatn@v33.003/Vazirmatn-font-face.css');
* { font-family: 'Vazirmatn', 'Vazir', Tahoma, sans-serif !important; }
</style>
"""))
```

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "پیش‌بینی قیمت خانه با رگرسیون خطی چندگانه در «زیلو» (From Scratch) 🏡", 
    "main_title"
)

display_academic_text(
    "در پلتفرم ملکی زیلو (Zillow)، موتور ارزیابی قیمت یا Zestimate با استفاده از رگرسیون خطی چندگانه "
    "قیمت تخمینی هر خانه را محاسبه می‌کند. در این تمرین ما به جای استفاده از متدهای آماده و تکرارشونده، "
    "معادله نرمال (Normal Equation) را پیاده‌سازی می‌کنیم تا بهترین ضرایب را مستقیماً با جبر خطی کشف کنیم.", 
    "paragraph"
)
# ==============================================================================
# 2. LOAD CALIFORNIA HOUSING DATASET
# ==============================================================================
housing = fetch_california_housing(as_frame=True)
df = housing.frame.copy()

selected_features = ['MedInc', 'AveRooms', 'HouseAge', 'AveOccup']
X_raw = df[selected_features].values
y_raw = df['MedHouseVal'].values

# ==============================================================================
# 3. FEATURE SCALING FROM SCRATCH (Z-SCORE STANDARDIZATION)
# ==============================================================================
def standardize(X_train, X_test):
    """Applies Z-score standardization strictly using training set parameters."""
    mean = X_train.mean(axis=0)
    std = X_train.std(axis=0)
    std[std == 0] = 1.0  
    
    return (X_train - mean) / std, (X_test - mean) / std, mean, std

X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X_raw, y_raw, test_size=0.2, random_state=42
)

X_train, X_test, feat_mean, feat_std = standardize(X_train_raw, X_test_raw)

# ==============================================================================
# 4. MULTIPLE LINEAR REGRESSION FROM SCRATCH (NORMAL EQUATION)
# ==============================================================================
class LinearRegressionScratch:
    """Multiple Linear Regression solver using the closed-form Normal Equation."""

    def __init__(self):
        self.beta = None
        self.feature_names = None

    def fit(self, X, y, feature_names=None):
        m = X.shape[0]
        # Add bias column for the intercept
        X_b = np.column_stack([np.ones(m), X])

        # Normal Equation: beta = (X^T X)^{-1} X^T y
        self.beta = np.linalg.pinv(X_b.T @ X_b) @ X_b.T @ y
        
        self.feature_names = ['Intercept'] + (feature_names or [f'x{i}' for i in range(X.shape[1])])
        return self

    def predict(self, X):
        m = X.shape[0]
        X_b = np.column_stack([np.ones(m), X])
        return X_b @ self.beta

    def score_r2(self, X, y):
        y_hat = self.predict(X)
        ss_res = np.sum((y - y_hat) ** 2)
        ss_tot = np.sum((y - y.mean()) ** 2)
        return 1.0 - ss_res / ss_tot

    def rmse(self, X, y):
        y_hat = self.predict(X)
        return float(np.sqrt(np.mean((y - y_hat) ** 2)))

# ==============================================================================
# 5. TRAIN AND EVALUATE
# ==============================================================================
model = LinearRegressionScratch()
model.fit(X_train, y_train, feature_names=selected_features)

y_pred = model.predict(X_test)
r2_test  = model.score_r2(X_test, y_test)
rmse_test = model.rmse(X_test, y_test)

df_results = pd.DataFrame({
    'Actual Price ($100K)': y_test[:5],
    'Predicted Price ($100K)': y_pred[:5],
    'Error (Residual)': np.abs(y_test[:5] - y_pred[:5])
})

display_side_by_side_datasets(
    [df_results],
    [f"نمونه پیش‌بینی‌های مدل Zestimate روی داده‌های تست (R² = {r2_test:.3f})"]
)

# ==============================================================================
# 6. VISUALIZATIONS AND CHARTS
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 6.5))

# CHART 1: Predicted vs Actual Scatter Plot 
axes[0].scatter(
    y_test, y_pred, alpha=0.4, s=15,
    color='#2980b9', edgecolors='none', label='Model Predictions'
)

lim_min = min(y_test.min(), y_pred.min()) - 0.1
lim_max = max(y_test.max(), y_pred.max()) + 0.1
axes[0].plot(
    [lim_min, lim_max], [lim_min, lim_max],
    color='#e74c3c', linewidth=2.5, linestyle='--', label='Perfect Prediction'
)

axes[0].set_xlabel('Actual House Value ($100K units)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Predicted House Value ($100K units)', fontsize=12, fontweight='bold')
axes[0].set_title(
    f'1. Predicted vs Actual — California Housing\n'
    f'Test R² = {r2_test:.3f}  |  RMSE = {rmse_test:.3f} ($100K)',
    fontsize=13, fontweight='bold', pad=12
)
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
axes[0].grid(True, linestyle='--', alpha=0.6)

# CHART 2: Feature Coefficients Bar Chart 
coef_names = model.feature_names[1:]
coef_vals  = model.beta[1:]

colors = ['#2c3e50' if v > 0 else '#e74c3c' for v in coef_vals]
bars = axes[1].bar(
    coef_names, coef_vals, color=colors,
    edgecolor='black', alpha=0.9, width=0.5
)

axes[1].axhline(0, color='#2c3e50', linewidth=2, linestyle='-')
axes[1].set_xlabel('Housing Feature', fontsize=12, fontweight='bold')

axes[1].set_ylabel(r'Standardized Coefficient Value ($\beta$)', fontsize=12, fontweight='bold')

axes[1].set_title(
    '2. Linear Regression Feature Importance (Coefficients)\n'
    'Standardized Data ensures fair comparison',
    fontsize=13, fontweight='bold', pad=12
)
axes[1].grid(axis='y', linestyle='--', alpha=0.7)

for bar, val in zip(bars, coef_vals):
    offset = 0.02 if val >= 0 else -0.05
    axes[1].annotate(
        f'{val:+.3f}',
        xy=(bar.get_x() + bar.get_width() / 2, val + offset),
        ha='center', fontsize=11, fontweight='bold',
        color='#2c3e50' if val > 0 else '#c0392b'
    )

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 7: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    f"<b>نتیجه استراتژیک:</b> با رفع خطای رندرینگ و تولید گراف‌های نهایی، تاثیر استانداردسازی نمایان‌تر "
    f"می‌شود. در فرمول معادله نرمال، اگر داده‌ها استاندارد نشده باشند، ویژگی‌ای که ارقام بزرگتری دارد "
    f"(مانند درآمد) ضریب $\\beta$ نامتوازنی دریافت می‌کند. اما در این کد استانداردشده، گراف ۲ دقیقاً ارزش "
    f"حقیقی ویژگی‌ها را در پلتفرم زیلو نشان می‌دهد. <code>MedInc</code> (درآمد) قدرتمندترین عامل در "
    f"تخمین قیمت است، و مدل ما با دقت {r2_test:.2f} توانست الگوهای مسکن کالیفرنیا را شبیه‌سازی کند."
)
display_academic_text(insight_conclusion, "insight")
```

    <li><b>معیارهای ارزیابی:</b> متدهای <code>score_r2</code>، <code>rmse</code> و <code>mae</code> هر سه معیار را مستقیماً از فرمول‌های تعریفی پیاده‌سازی کرده‌اند بدون اتکا به هیچ کتابخانه آماده‌ای.</li>
