<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: قضیه بِیز و فیلترینگ هوشمند ایمیل‌های اسپم در «جیمیل» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در سرویس ایمیل <b>جیمیل</b>، روزانه میلیاردها پیام دریافت می‌شود که بخش قابل توجهی از آن‌ها اسپم یا تبلیغاتی هستند. موتور فیلترینگ جیمیل با استفاده از <b>قضیه بِیز (Bayes' Theorem)</b>، باور اولیه خود درباره اسپم بودن یک ایمیل را با مشاهده شواهد جدید (مانند حضور کلماتی نظیر «رایگان»، «برنده»، یا «تخفیف ویژه») به‌روزرسانی می‌کند. این رویکرد احتمالاتی به سیستم اجازه می‌دهد تا با هر ایمیل جدیدی که کاربر به عنوان اسپم علامت‌گذاری می‌کند، هوشمندتر شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی طبقه‌بندی بِیزی و نایو بِیز:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. فرمول اصلی قضیه بِیز برای محاسبه احتمال پسین (Posterior Probability):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$P(C \mid X) = \frac{P(X \mid C) \cdot P(C)}{P(X)}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. فرض استقلال شرطی در نایو بِیز برای بردار ویژگی‌های $X = (x_1, x_2, \dots, x_n)$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$P(X \mid C) = \prod_{i=1}^{n} P(x_i \mid C)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. تخمین احتمال با هموارسازی لاپلاس (Laplace Smoothing) برای جلوگیری از احتمال صفر:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$P(x_i \mid C) = \frac{\text{Count}(x_i, C) + 1}{\text{Count}(C) + |V|}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. قاعده تصمیم‌گیری نهایی طبقه‌بند بِیزی (Maximum A Posteriori):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\hat{C} = \arg\max_{C_k} \left[ \log P(C_k) + \sum_{i=1}^{n} \log P(x_i \mid C_k) \right]$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: جیمیل با محاسبه احتمال پسین هر ایمیل و مقایسه آن با آستانه تصمیم، پیام‌های مخرب و تبلیغاتی را با دقتی نزدیک به ۹۹٪ به پوشه اسپم منتقل می‌کند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: به‌روزرسانی احتمالاتی باورها بر اساس شواهد متنی و طبقه‌بندی دقیق ایمیل‌های اسپم
    </div>
  </p>

</div>

```python
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display, HTML
import warnings
from collections import defaultdict

warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={"axes.titlesize": 15, "axes.labelsize": 13, "xtick.labelsize": 11, "ytick.labelsize": 11})

plt.rcParams['figure.dpi'] = 120

def display_academic_text(text, type="normal"):
    styles = {
        "main_title": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 24px; "
            "font-weight: bold; text-align: center; border-bottom: 2px solid #34495e; "
            "padding-bottom: 15px; margin: 30px 0;"
        ),
        "sub_title": (
            "font-family: Tahoma, sans-serif; color: #34495e; font-size: 19px; "
            "font-weight: bold; border-right: 5px solid #2980b9; padding-right: 15px; "
            "margin: 40px 0px 20px 0px; background-color: #ecf0f1; padding: 12px; "
            "border-radius: 4px;"
        ),
        "paragraph": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 16px; "
            "margin-bottom: 20px; line-height: 1.8; text-align: justify;"
        ),
        "insight": (
            "font-family: Tahoma, sans-serif; color: #16a085; font-size: 16px; "
            "font-weight: bold; background-color: #e8f8f5; padding: 20px; "
            "border-left: 5px solid #1abc9c; margin: 30px 0; line-height: 1.8; "
            "text-align: justify;"
        )
    }
    display(HTML(f"<div dir='rtl' style='text-align: right; {styles[type]}'>{text}</div>"))

def get_protected_css():
    return """
    <style>
    .academic-table-wrapper { 
        direction: ltr !important; 
        overflow-x: auto !important; 
        width: 100% !important; 
        background-color: #fff; 
        border: 1px solid #bdc3c7; 
        border-radius: 4px; 
        box-shadow: 0 3px 6px rgba(0,0,0,0.08); 
    }
    table.strict-tbl { 
        border-collapse: collapse !important; 
        width: 100% !important; 
        margin: 0 !important; 
        font-family: inherit !important;
        font-size: 14px !important; 
        table-layout: auto !important; 
        border: none !important; 
    }
    table.strict-tbl thead th { 
        background-color: #2c3e50 !important; 
        color: #ffffff !important; 
        padding: 12px 18px !important; 
        text-align: center !important; 
        border: 1px solid #34495e !important; 
        white-space: nowrap !important; 
        font-weight: bold !important; 
    }
    table.strict-tbl tbody td { 
        padding: 10px 18px !important; 
        text-align: center !important; 
        border: 1px solid #bdc3c7 !important; 
        white-space: nowrap !important; 
        color: #2c3e50 !important; 
    }
    table.strict-tbl tbody tr:nth-child(even) td { 
        background-color: #f8f9fa !important; 
    }
    table.strict-tbl tbody tr:hover td { 
        background-color: #e8ecef !important; 
    }
    </style>
    """

def display_side_by_side_datasets(dfs, titles, *args):
    if not isinstance(dfs, list) and not isinstance(titles, list):
        dfs = [dfs]
        titles = [titles]
        
    col_html_list = []
    for i, (df, title) in enumerate(zip(dfs, titles)):
        table_html = df.to_html(index=False, classes=['strict-tbl'], border=0)
        bg_color = "#2980b9" if i % 2 == 0 else "#27ae60"
        col_html_list.append(f"""
        <div class="flex-column" style="flex: 1 1 0; min-width: 0; display: flex; flex-direction: column;">
            <div style="background-color: {bg_color}; color: white; padding: 12px; 
                        text-align: center; font-family: Tahoma, sans-serif; 
                        font-weight: bold; font-size: 15px; border-radius: 4px 4px 0 0;">
                {title}
            </div>
            <div class="academic-table-wrapper" style="margin-top: 0; border-top: none;">{table_html}</div>
        </div>
        """)
    
    display(HTML(get_protected_css() + f'<div style="display: flex; gap: 20px; width: 100%; margin: 20px 0 35px 0;">{"".join(col_html_list)}</div>'))

display(HTML("""
<style>
@import url('https://cdn.jsdelivr.net/gh/rastikerdar/vazirmatn@v33.003/Vazirmatn-font-face.css');
* { font-family: 'Vazirmatn', 'Vazir', Tahoma, sans-serif !important; }
</style>
"""))
```

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
# ==============================================================================
display_academic_text(
    "قضیه بِیز و فیلترینگ هوشمند ایمیل‌های اسپم در «جیمیل» (From Scratch) 📧", 
    "main_title"
)

display_academic_text(
    "در سرویس ایمیل جیمیل، روزانه میلیاردها پیام دریافت می‌شود. موتور فیلترینگ جیمیل با استفاده از "
    "قضیه بِیز (Bayes' Theorem)، باور اولیه خود درباره اسپم بودن یک ایمیل را با مشاهده شواهد جدید "
    "(مانند حضور کلماتی نظیر «رایگان» یا «برنده») به‌روزرسانی می‌کند. در این تمرین، الگوریتم Naive Bayes را "
    "بدون استفاده از هیچ کتابخانه آماده‌ای، با استفاده از محاسبه لاپلاس و در فضای لگاریتمی توسعه می‌دهیم.", 
    "paragraph"
)

display_academic_text("مفاهیم و فرمول‌های ریاضی طبقه‌بندی بِیزی و نایو بِیز:", "sub_title")

display_academic_text("۱. فرمول اصلی قضیه بِیز برای محاسبه احتمال پسین (Posterior Probability):", "paragraph")
display_academic_text(r"$$P(C \mid X) = \frac{P(X \mid C) \cdot P(C)}{P(X)}$$", "paragraph")

display_academic_text("۲. فرض استقلال شرطی در نایو بِیز برای بردار ویژگی‌های متنی:", "paragraph")
display_academic_text(r"$$P(X \mid C) = \prod_{i=1}^{n} P(x_i \mid C)$$", "paragraph")

display_academic_text("۳. تخمین احتمال کلمات با هموارسازی لاپلاس (برای جلوگیری از احتمال ضرب‌در‌صفر):", "paragraph")
display_academic_text(r"$$P(x_i \mid C) = \frac{\text{Count}(x_i, C) + 1}{\text{Count}(C) + |V|}$$", "paragraph")

display_academic_text("۴. قاعده تصمیم‌گیری در فضای لگاریتمی (برای جلوگیری از خطای Underflow):", "paragraph")
display_academic_text(r"$$\hat{C} = \arg\max_{C_k} \left[ \log P(C_k) + \sum_{i=1}^{n} \log P(x_i \mid C_k) \right]$$", "paragraph")


# ==============================================================================
# 2. GMAIL EMAIL DATASET (SPAM VS HAM)
# ==============================================================================

# Training corpus: (email_text, label) where 1 = Spam, 0 = Ham (Normal Email)
train_emails = [
    ("congratulations you win a free prize claim now", 1),
    ("free money offer limited time win big cash", 1),
    ("claim your free gift card now urgent offer", 1),
    ("win lottery cash prize click link immediately", 1),
    ("urgent free discount offer expires today", 1),
    ("exclusive offer win free vacation package now", 1),
    ("meeting scheduled tomorrow morning with team", 0),
    ("please review the attached project report", 0),
    ("lunch meeting at the office cafeteria today", 0),
    ("your monthly invoice statement is attached", 0),
    ("team project deadline moved to next friday", 0),
    ("please confirm your attendance for the meeting", 0),
    ("annual performance review scheduled next week", 0),
    ("the quarterly report has been shared with team", 0)
]

# Test emails for prediction
test_emails = [
    ("win free cash prize now", 1),
    ("meeting report attached for review", 0),
    ("urgent claim your free offer today", 1),
    ("project deadline scheduled next week", 0),
    ("free gift click link win money", 1),
    ("team lunch meeting tomorrow office", 0)
]

# Displaying a sample of the training data
df_train = pd.DataFrame(train_emails, columns=['Email Text', 'Label (1=Spam, 0=Ham)'])
display_side_by_side_datasets([df_train.head(6)], ["نمونه‌ای از دادگان آموزش فیلتر اسپم جیمیل"])


# ==============================================================================
# 3. NAIVE BAYES CLASSIFIER FROM SCRATCH (WITH LAPLACE & LOG-SPACE)
# ==============================================================================

def tokenize(text):
    """
    Tokenizes email text into cleaned lowercase words from scratch.
    Uses regex to extract only alphabet words.
    """
    return re.findall(r'\b[a-z]+\b', text.lower())

class NaiveBayesSpamFilter:
    """Multinomial Naive Bayes with Laplace Smoothing implemented from scratch."""
    
    def __init__(self):
        self.class_priors = {}
        # Stores count of each word per class (0 or 1)
        self.word_counts = {0: defaultdict(int), 1: defaultdict(int)}
        # Total number of words in each class
        self.total_words = {0: 0, 1: 0}
        # The unique set of all words across the entire training dataset
        self.vocabulary = set()

    def fit(self, dataset):
        """Trains the classifier by computing P(C) priors and Likelihood counts."""
        class_doc_counts = {0: 0, 1: 0}
        total_docs = len(dataset)

        for text, label in dataset:
            class_doc_counts[label] += 1
            tokens = tokenize(text)
            
            for token in tokens:
                self.word_counts[label][token] += 1
                self.total_words[label] += 1
                self.vocabulary.add(token)

        # Compute prior probabilities: P(C)
        for label in [0, 1]:
            self.class_priors[label] = class_doc_counts[label] / total_docs

    def compute_likelihood(self, word, label):
        """
        Computes P(word | class) using Laplace Smoothing (+1 logic).
        This prevents a probability of 0 if a word was never seen in training.
        """
        numerator = self.word_counts[label][word] + 1
        denominator = self.total_words[label] + len(self.vocabulary)
        return numerator / denominator

    def predict_proba(self, text):
        """
        Computes posterior probability using Log-Space.
        Multiplying tiny probabilities (e.g., 0.001 * 0.005 * ...) causes underflow.
        Log-Space solves this: log(a * b) = log(a) + log(b).
        """
        tokens = tokenize(text)
        log_scores = {}

        for label in [0, 1]:
            # Start with log prior: log P(C)
            log_prob = np.log(self.class_priors[label])
            
            # Add log likelihoods: sum log P(x_i | C)
            for token in tokens:
                # We only consider words that exist in our trained vocabulary
                # OOV (Out Of Vocabulary) words are ignored in standard Naive Bayes
                if token in self.vocabulary:
                    log_prob += np.log(self.compute_likelihood(token, label))
            
            log_scores[label] = log_prob

        # Convert log scores back into normalized probabilities (0 to 1 range)
        max_log = max(log_scores.values())
        # Subtract max_log to prevent overflow during np.exp()
        exp_scores = {k: np.exp(v - max_log) for k, v in log_scores.items()}
        total = sum(exp_scores.values())
        
        return {k: v / total for k, v in exp_scores.items()}

    def predict(self, text):
        """Predicts class 1 if Spam Probability >= 50%"""
        probs = self.predict_proba(text)
        return 1 if probs[1] >= 0.5 else 0


# --- Train and Execute the Model ---
nb_model = NaiveBayesSpamFilter()
nb_model.fit(train_emails)

# Predict test emails
results = []
for text, true_label in test_emails:
    probs = nb_model.predict_proba(text)
    pred = nb_model.predict(text)
    results.append({
        'Email': text[:32] + '...',
        'Spam_Probability': probs[1],
        'Ham_Probability': probs[0],
        'Predicted': pred,
        'Actual': true_label
    })

df_results = pd.DataFrame(results)
accuracy = np.mean(df_results['Predicted'] == df_results['Actual']) * 100

# Compute Discriminative Power of top words (Likelihood Ratio)
word_power = []
for word in nb_model.vocabulary:
    p_spam = nb_model.compute_likelihood(word, 1)
    p_ham = nb_model.compute_likelihood(word, 0)
    # How much more likely is this word to appear in Spam vs Ham?
    likelihood_ratio = p_spam / p_ham
    word_power.append({'Word': word, 'P_Spam': p_spam, 'P_Ham': p_ham, 'Ratio': likelihood_ratio})

df_words = pd.DataFrame(word_power).sort_values(by='Ratio', ascending=False)
top_spam_words = df_words.head(8)


# ==============================================================================
# 4. VISUALIZATIONS AND CHARTS
# ==============================================================================

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# --- CHART 1: Posterior Spam Probability for Test Emails ---
# Red/Orange for Spam (>0.5), Academic Blue for Ham (<0.5)
colors = ['#e74c3c' if p >= 0.5 else '#2980b9' for p in df_results['Spam_Probability']]
bars = axes[0].barh(
    df_results['Email'], 
    df_results['Spam_Probability'], 
    color=colors, 
    edgecolor='black', 
    alpha=0.9
)

axes[0].axvline(0.5, color='#2c3e50', linestyle='--', linewidth=2, label='Decision Threshold (0.5)')
axes[0].set_xlabel('Posterior Spam Probability P(Spam | Email)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Test Email Content Snippet', fontsize=12, fontweight='bold')
axes[0].set_title(f'1. Bayesian Spam Classification Results (Acc: {accuracy:.1f}%)', fontsize=13, fontweight='bold', pad=12)
axes[0].set_xlim(0, 1.15)
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
axes[0].grid(axis='x', linestyle='--', alpha=0.7)
axes[0].invert_yaxis()

for bar in bars:
    w = bar.get_width()
    axes[0].annotate(
        f'{w:.3f}', 
        xy=(w, bar.get_y() + bar.get_height()/2),
        xytext=(5, 0), textcoords="offset points", ha='left', va='center', 
        fontweight='bold', fontsize=10, color='#2c3e50'
    )

#  CHART 2: Word Likelihood Comparison P(word|Spam) vs P(word|Ham) 
x = np.arange(len(top_spam_words))
width = 0.35

rects1 = axes[1].bar(
    x - width/2, 
    top_spam_words['P_Spam'], 
    width, 
    label='P(word | Spam)', 
    color='#e74c3c', # Red for Spam Likelihood
    alpha=0.9, 
    edgecolor='black'
)
rects2 = axes[1].bar(
    x + width/2, 
    top_spam_words['P_Ham'], 
    width, 
    label='P(word | Ham)', 
    color='#2980b9', # Blue for Ham Likelihood
    alpha=0.9, 
    edgecolor='black'
)

axes[1].set_ylabel('Conditional Likelihood Probability', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Most Discriminative Keywords', fontsize=12, fontweight='bold')
axes[1].set_title('2. Word Likelihood Evidence Strength in Bayesian Filter', fontsize=13, fontweight='bold', pad=12)
axes[1].set_xticks(x)
axes[1].set_xticklabels(top_spam_words['Word'], fontsize=11, fontweight='bold', rotation=30)
axes[1].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
axes[1].grid(axis='y', linestyle='--', alpha=0.7)

# Clean up axes
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    "<b>نتیجه استراتژیک:</b> بررسی گراف ۱ نشان می‌دهد که فیلتر بِیزی ما با استفاده از دانشِ پیشین (Priors) "
    "و احتمال شواهد جدید، توانست با دقت ۱۰۰٪ ایمیل‌های تست را طبقه‌بندی کند. زیبایی این الگوریتم در گراف ۲ "
    "مشخص می‌شود: سیستم با شمارش کلمات یاد گرفته است که کلماتی مانند <code>win</code> و <code>free</code> "
    "در کلاس اسپم بسیار پررنگ‌تر (Likelihood بالاتر) از کلاس هم (Ham) هستند. بنابراین، وقتی یک ایمیل جدید "
    "حاوی این کلمات وارد می‌شود، احتمال پسینِ (Posterior) اسپم بودن آن به شدت تقویت شده و سیستم "
    "بدون نیاز به قوانین سخت و شرطیِ سنتی، پیام را به صورت هوشمند و احتمالاتی مسدود می‌کند."
)
display_academic_text(insight_conclusion, "insight")
```
