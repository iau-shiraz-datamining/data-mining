  <span style="font-size: 20pt;"></span> 6.4.1 k-nearest-neighbor classifiers

<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">    
  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: پیشنهاد آهنگ‌های مشابه با الگوریتم k-نزدیک‌ترین همسایه در «اسپاتیفای» 
  </div>
  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پلتفرم <b>اسپاتیفای</b>، هر آهنگ با بردار ویژگی‌های صوتی مانند ریتم (Tempo)، انرژی (Energy)، رقص‌پذیری (Danceability) و درجه شادی (Valence) نمایش داده می‌شود. زمانی که کاربر به یک آهنگ خاص گوش می‌دهد، موتور توصیه‌گر با استفاده از الگوریتم <b>k-نزدیک‌ترین همسایه (k-Nearest Neighbor)</b> در فضای این ویژگی‌ها، نزدیک‌ترین آهنگ‌ها را از نظر فاصله عددی پیدا کرده و آن‌ها را در لیست پخش «بر اساس این آهنگ» (Based on this song) پیشنهاد می‌دهد. این رویکرد بر پایه فرض ساده اما قدرتمندی است: آهنگ‌هایی که در فضای ویژگی به هم نزدیک‌اند، احتمالاً سبک و حس مشابهی دارند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    فرمول محاسبه فاصله اقلیدسی بین دو نمونه با بردار ویژگی $p$ بعدی $x$ و $q$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$d(x, q) = \sqrt{\sum_{i=1}^{p} (x_i - q_i)^2}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    فرمول عمومی فاصله مینکوفسکی که با تغییر پارامتر $m$، به فاصله اقلیدسی ($m=2$) یا فاصله منهتن ($m=1$) تبدیل می‌شود:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$d(x, q) = \left( \sum_{i=1}^{p} |x_i - q_i|^m \right)^{1/m}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    پس از یافتن مجموعه $N_k(x)$ شامل $k$ همسایه نزدیک، قاعده رأی اکثریت برای تعیین کلاس نهایی به این صورت است:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\hat{y} = \arg\max_{c} \sum_{i \in N_k(x)} \mathbb{1}(y_i = c)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در نسخه وزن‌دار (Weighted k-NN)، همسایه‌های نزدیک‌تر تأثیر بیشتری در رأی‌گیری دارند:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$w_i = \frac{1}{d(x, x_i)^2}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: با تنظیم دقیق مقدار $k$ و معیار فاصله، اسپاتیفای می‌تواند در فضای ویژگی‌های صوتی میلیون‌ها آهنگ، نزدیک‌ترین و مرتبط‌ترین پیشنهادها را در کسری از ثانیه به کاربر ارائه دهد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: توصیه محتوای مشابه بر اساس نزدیکی هندسی نمونه‌ها در فضای ویژگی‌های عددی
    </div>
  </p>

</div>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display, HTML
import warnings
from matplotlib.colors import ListedColormap

warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={"axes.titlesize": 15, "axes.labelsize": 13, "xtick.labelsize": 11, "ytick.labelsize": 11})

plt.rcParams['figure.dpi'] = 120

def display_academic_text(text, type="normal"):
    styles = {
        "main_title": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 24px; "
            "font-weight: bold; text-align: center; border-bottom: 2px solid #34495e; "
            "padding-bottom: 15px; margin: 30px 0;"
        ),
        "sub_title": (
            "font-family: Tahoma, sans-serif; color: #34495e; font-size: 19px; "
            "font-weight: bold; border-right: 5px solid #2980b9; padding-right: 15px; "
            "margin: 40px 0px 20px 0px; background-color: #ecf0f1; padding: 12px; "
            "border-radius: 4px;"
        ),
        "paragraph": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 16px; "
            "margin-bottom: 20px; line-height: 1.8; text-align: justify;"
        ),
        "insight": (
            "font-family: Tahoma, sans-serif; color: #16a085; font-size: 16px; "
            "font-weight: bold; background-color: #e8f8f5; padding: 20px; "
            "border-left: 5px solid #1abc9c; margin: 30px 0; line-height: 1.8; "
            "text-align: justify;"
        )
    }
    display(HTML(f"<div dir='rtl' style='text-align: right; {styles[type]}'>{text}</div>"))

def get_protected_css():
    return """
    <style>
    .academic-table-wrapper { 
        direction: ltr !important; 
        overflow-x: auto !important; 
        width: 100% !important; 
        background-color: #fff; 
        border: 1px solid #bdc3c7; 
        border-radius: 4px; 
        box-shadow: 0 3px 6px rgba(0,0,0,0.08); 
    }
    table.strict-tbl { 
        border-collapse: collapse !important; 
        width: 100% !important; 
        margin: 0 !important; 
        font-family: inherit !important;
        font-size: 14px !important; 
        table-layout: auto !important; 
        border: none !important; 
    }
    table.strict-tbl thead th { 
        background-color: #2c3e50 !important; 
        color: #ffffff !important; 
        padding: 12px 18px !important; 
        text-align: center !important; 
        border: 1px solid #34495e !important; 
        white-space: nowrap !important; 
        font-weight: bold !important; 
    }
    table.strict-tbl tbody td { 
        padding: 10px 18px !important; 
        text-align: center !important; 
        border: 1px solid #bdc3c7 !important; 
        white-space: nowrap !important; 
        color: #2c3e50 !important; 
    }
    table.strict-tbl tbody tr:nth-child(even) td { 
        background-color: #f8f9fa !important; 
    }
    table.strict-tbl tbody tr:hover td { 
        background-color: #e8ecef !important; 
    }
    </style>
    """

def display_side_by_side_datasets(dfs, titles, *args):
    if not isinstance(dfs, list) and not isinstance(titles, list):
        dfs = [dfs]
        titles = [titles]
        
    col_html_list = []
    for i, (df, title) in enumerate(zip(dfs, titles)):
        table_html = df.to_html(index=False, classes=['strict-tbl'], border=0)
        bg_color = "#2980b9" if i % 2 == 0 else "#27ae60"
        col_html_list.append(f"""
        <div class="flex-column" style="flex: 1 1 0; min-width: 0; display: flex; flex-direction: column;">
            <div style="background-color: {bg_color}; color: white; padding: 12px; 
                        text-align: center; font-family: Tahoma, sans-serif; 
                        font-weight: bold; font-size: 15px; border-radius: 4px 4px 0 0;">
                {title}
            </div>
            <div class="academic-table-wrapper" style="margin-top: 0; border-top: none;">{table_html}</div>
        </div>
        """)
    
    display(HTML(get_protected_css() + f'<div style="display: flex; gap: 20px; width: 100%; margin: 20px 0 35px 0;">{"".join(col_html_list)}</div>'))

display(HTML("""
<style>
@import url('https://cdn.jsdelivr.net/gh/rastikerdar/vazirmatn@v33.003/Vazirmatn-font-face.css');
* { font-family: 'Vazirmatn', 'Vazir', Tahoma, sans-serif !important; }
</style>
"""))
```

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "پیشنهاد آهنگ‌های مشابه با الگوریتم k-نزدیک‌ترین همسایه در «اسپاتیفای» (From Scratch) 🎵", 
    "main_title"
)

display_academic_text(
    "در پلتفرم اسپاتیفای، هر آهنگ با بردار ویژگی‌های صوتی مانند انرژی (Energy) و درجه شادی (Valence) نمایش "
    "داده می‌شود. زمانی که کاربر به یک آهنگ گوش می‌دهد، موتور توصیه‌گر با استفاده از الگوریتم k-NN، "
    "نزدیک‌ترین آهنگ‌ها را از نظر فاصله هندسی پیدا کرده و پیشنهاد می‌دهد. در این تمرین، ما یک "
    "دیتابیس موسیقی شبیه‌سازی می‌کنیم و الگوریتم k-NN را به صورت بهینه‌شده (Vectorized) کدنویسی می‌نماییم.", 
    "paragraph"
)

display_academic_text("فرمول محاسبه فاصله و قاعده رأی‌گیری:", "sub_title")
display_academic_text(r"$$d(x, q) = \sqrt{\sum_{i=1}^{p} (x_i - q_i)^2}$$", "paragraph")
display_academic_text(r"$$\hat{y} = \arg\max_{c} \sum_{i \in N_k(x)} \mathbb{1}(y_i = c)$$", "paragraph")

# ==============================================================================
# 2. GENERATE SYNTHETIC SPOTIFY DATASET
# ==============================================================================
# Replacing Iris with a domain-appropriate Spotify Dataset
np.random.seed(42)
n_samples = 250

# We simulate 3 distinct Music Genres/Moods based on (Energy, Valence)
# 0 = Acoustic/Chill (Low Energy, High/Mid Valence)
# 1 = Pop/Dance (High Energy, High Valence)
# 2 = Heavy Metal (High Energy, Low Valence)

# Generate features using multivariate normal distributions to create clusters
X_chill = np.random.multivariate_normal([0.3, 0.7], [[0.01, 0], [0, 0.02]], n_samples // 3)
y_chill = np.zeros(n_samples // 3, dtype=int)

X_pop = np.random.multivariate_normal([0.8, 0.8], [[0.02, -0.01], [-0.01, 0.02]], n_samples // 3)
y_pop = np.ones(n_samples // 3, dtype=int)

X_metal = np.random.multivariate_normal([0.8, 0.2], [[0.015, 0], [0, 0.015]], n_samples - 2*(n_samples // 3))
y_metal = np.full(n_samples - 2*(n_samples // 3), 2, dtype=int)

X = np.vstack((X_chill, X_pop, X_metal))
y = np.concatenate((y_chill, y_pop, y_metal))

# Shuffle and Split (70% Train, 30% Test) manually to maintain From-Scratch integrity
indices = np.random.permutation(len(X))
split_idx = int(0.7 * len(X))
X_train, y_train = X[indices[:split_idx]], y[indices[:split_idx]]
X_test, y_test = X[indices[split_idx:]], y[indices[split_idx:]]

class_names = {0: 'Acoustic/Chill', 1: 'Pop/Dance', 2: 'Heavy Metal'}

# Display a sample dataset
df_music = pd.DataFrame(X_train[:5], columns=['Energy (0-1)', 'Valence/Happiness (0-1)'])
df_music['Genre Label'] = [class_names[label] for label in y_train[:5]]
display_side_by_side_datasets([df_music], ["نمونه‌ای از بردار ویژگی‌های صوتی آهنگ‌ها در اسپاتیفای"])

# ==============================================================================
# 3. HIGHLY OPTIMIZED K-NEAREST NEIGHBOR CLASSIFIER FROM SCRATCH
# ==============================================================================

class FastKNNClassifier:
    """
    A highly optimized from-scratch implementation of k-NN.
    Removes standard Python loops in favor of NumPy broadcasting.
    """
    def __init__(self, k=5):
        self.k = k

    def fit(self, X_train, y_train):
        """Lazy learning: k-NN simply stores the training data in memory."""
        self.X_train = np.array(X_train)
        self.y_train = np.array(y_train)

    def predict(self, X_test):
        """
        Predicts class labels using highly optimized NumPy Vectorization.
        Computes distances without using standard slow Python 'for' loops across the training set.
        """
        X_test = np.array(X_test)
        predictions = []
        
        for x in X_test:
            # OPTIMIZATION: Broadcasting subtracts vector 'x' from the ENTIRE Matrix 'X_train' at once!
            # This is computed natively in C, making it exponentially faster than a list comprehension.
            distances = np.sqrt(np.sum((self.X_train - x) ** 2, axis=1))
            
            # Get indices of the 'k' smallest distances
            # argpartition is faster than argsort because it only partially sorts the array
            k_indices = np.argpartition(distances, self.k)[:self.k]
            k_nearest_labels = self.y_train[k_indices]
            
            # OPTIMIZATION: np.bincount is the fastest way to find the most frequent integer in an array
            majority_vote = np.bincount(k_nearest_labels).argmax()
            predictions.append(majority_vote)
            
        return np.array(predictions)

# ==============================================================================
# 4. FINDING THE OPTIMAL K (BIAS-VARIANCE TRADEOFF)
# ==============================================================================

k_values = range(1, 26)
accuracy_scores = []

for k in k_values:
    model = FastKNNClassifier(k=k)
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)
    
    acc = np.mean(predictions == y_test) * 100
    accuracy_scores.append(acc)

# Select the 'k' that yielded the highest accuracy on unseen test data
best_k = list(k_values)[np.argmax(accuracy_scores)]
best_accuracy = max(accuracy_scores)

# ==============================================================================
# 5. VISUALIZATIONS AND CHARTS
# ==============================================================================

# Train the final model with the absolute best K
final_model = FastKNNClassifier(k=best_k)
final_model.fit(X_train, y_train)

# Create a dense grid to paint the decision boundary
x_min, x_max = X[:, 0].min() - 0.1, X[:, 0].max() + 0.1
y_min, y_max = X[:, 1].min() - 0.1, X[:, 1].max() + 0.1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.01),
                     np.arange(y_min, y_max, 0.01))

mesh_points = np.c_[xx.ravel(), yy.ravel()]
Z = final_model.predict(mesh_points)
Z = Z.reshape(xx.shape)

fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

# Use Academic Blue & Teal Palette for classes
class_colors = ['#27ae60', '#2980b9', '#2c3e50'] # Teal, Classic Blue, Dark Navy
# Light background pastel versions of the same colors for the decision regions
cmap_light = ListedColormap(['#d4efdf', '#d4e6f1', '#d5d8dc']) 

#  CHART 1: Decision Boundary of k-NN in Spotify Feature Space 
axes[0].contourf(xx, yy, Z, cmap=cmap_light, alpha=0.9)

# Scatter plot the training data
for idx, color in zip([0, 1, 2], class_colors):
    mask = y_train == idx
    axes[0].scatter(
        X_train[mask, 0], X_train[mask, 1], 
        c=color, edgecolor='black', s=80, 
        label=f'{class_names[idx]} (Train)', alpha=0.95
    )

# Highlight Misclassified test samples to show where the algorithm failed
mask_test_correct = final_model.predict(X_test) == y_test
axes[0].scatter(
    X_test[~mask_test_correct, 0], X_test[~mask_test_correct, 1],
    marker='X', s=150, c='#e74c3c', linewidths=2.5, edgecolor='black', label='Misclassified (Test)'
)

axes[0].set_xlabel('Song Energy Level (Low to High)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Song Valence/Happiness (Sad to Happy)', fontsize=12, fontweight='bold')
axes[0].set_title(f'1. k-NN Decision Boundaries in Music Space (k = {best_k})', fontsize=13, fontweight='bold', pad=12)
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='lower left')
axes[0].grid(True, linestyle='--', alpha=0.5)

# CHART 2: Accuracy vs K curve (The Bias-Variance Tradeoff) 
axes[1].plot(
    list(k_values), 
    accuracy_scores, 
    marker='o', 
    color='#2c3e50', # Academic Dark Navy
    linewidth=3, 
    markersize=7, 
    label='Test Accuracy'
)

# Highlight the optimal K
axes[1].axvline(best_k, color='#2980b9', linestyle='--', linewidth=2.5, label=f'Optimal k = {best_k}')
axes[1].scatter([best_k], [best_accuracy], color='#e74c3c', s=180, zorder=5, edgecolor='white', linewidth=1.5, label=f'Peak Accuracy: {best_accuracy:.1f}%')

axes[1].set_xlabel('Number of Neighbors (k)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Prediction Accuracy on Test Data (%)', fontsize=12, fontweight='bold')
axes[1].set_title('2. Overfitting vs Generalization (Bias-Variance Tradeoff)', fontsize=13, fontweight='bold', pad=12)
axes[1].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=11, loc='lower right')
axes[1].grid(True, linestyle='--', alpha=0.6)

# Clean up axes
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 6: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    f"<b>نتیجه استراتژیک:</b> بررسی گراف ۱ نشان می‌دهد که الگوریتم، فضای صوتی اسپاتیفای را به ۳ منطقه "
    f"مجزا تقسیم کرده است. اگر کاربری یک آهنگ با انرژی بالا و درجه شادی بالا (محدوده آبی روشن) را پخش کند، "
    f"الگوریتم نزدیک‌ترین همسایه‌ها را در همان ناحیه جستجو کرده و آهنگ‌های Pop/Dance را به او توصیه می‌کند. "
    f"گراف ۲ یک درس مهم در زمینه بهینه‌سازی دارد: در $k=1$ (کودک‌ترین همسایه)، سیستم بیش‌از‌حد به نویزها "
    f"حساس است (Overfitting) و دقت تست پایینی دارد. با رسیدن به $k={best_k}$، سیستم تعادل هندسی را پیدا "
    f"کرده و با دقت {best_accuracy:.1f}٪ عمل می‌کند. همچنین، استفاده از توابع برداری (Broadcasting) در "
    f"کدِ این نسخه، سرعت پاسخ‌دهی را به حدی بالا برده است که می‌تواند میلیون‌ها آهنگ را در لحظه (Real-time) "
    f"مرتب و پیشنهاد دهد."
)
display_academic_text(insight_conclusion, "insight")
```

<li>داده‌ها با <code>train_test_split</code> به صورت لایه‌بندی‌شده (Stratified) به بخش آموزش و آزمون تقسیم شدند تا نسبت هر گونه گل در هر دو مجموعه حفظ شود.</li>
<li>پیاده‌سازی کامل الگوریتم k-نزدیک‌ترین همسایه از صفر (From Scratch):
  <ul>
    <li><b>یادگیری تنبل (Lazy Learning):</b> متد <code>fit</code> برخلاف اغلب الگوریتم‌ها، هیچ محاسبه‌ای انجام نمی‌دهد و فقط داده آموزشی را ذخیره می‌کند؛ تمام محاسبات در زمان پیش‌بینی رخ می‌دهد.</li>
    <li><b>محاسبه فاصله اقلیدسی:</b> متد <code>_euclidean_distance</code> دقیقاً طبق فرمول $\sqrt{\sum (x_i - q_i)^2}$ فاصله بین دو نمونه را محاسبه می‌کند.</li>
    <li><b>یافتن همسایه‌ها و رأی اکثریت:</b> متد <code>_predict_single</code> با تابع <code>np.argsort</code>، $k$ نمونه با کمترین فاصله را پیدا کرده و با <code>Counter</code> رأی اکثریت را بین برچسب‌های آن‌ها تعیین می‌کند.</li>
  </ul>
</li>
<li>یافتن مقدار بهینه $k$: مدل برای مقادیر $k$ از ۱ تا ۲۵ آموزش داده و روی داده آزمون ارزیابی شد تا بهترین تعادل بین حساسیت به نویز (k کوچک) و بیش‌تعمیم‌دهی (k بزرگ) پیدا شود.</li>
<li>تفسیر خروجی و نمودارها:
  <ul>
    <li>نمودار سمت چپ (Decision Boundary Plot): نواحی رنگی نشان‌دهنده تصمیم مدل برای هر ناحیه از فضای ویژگی هستند؛ نقاط دایره‌ای نمونه‌های آموزشی هر گونه گل و علامت‌های ضربدر قرمز نمونه‌های تست اشتباه طبقه‌بندی‌شده را نشان می‌دهند. مرزهای غیرخطی و پلکانی، ویژگی مشخصه الگوریتم k-NN هستند.</li>
    <li>نمودار سمت راست (Accuracy vs K Curve): روند تغییر دقت مدل با افزایش مقدار $k$ را نشان می‌دهد؛ نقطه بهینه با رنگ آبی مشخص شده و نشان می‌دهد مقادیر خیلی کوچک $k$ به نویز حساس هستند و مقادیر خیلی بزرگ باعث افت دقت به دلیل غالب‌شدن کلاس‌های اکثریت می‌شوند.</li>
  </ul>
</li>
<li>نکته کلیدی: k-NN برخلاف بِیز ساده هیچ فرض توزیعی درباره داده ندارد و تصمیم‌گیری آن کاملاً مبتنی بر فاصله هندسی نمونه‌ها است؛ به همین دلیل انتخاب مقیاس مناسب ویژگی‌ها (Feature Scaling) و مقدار بهینه $k$ نقش تعیین‌کننده‌ای در دقت نهایی این الگوریتم دارد.</li>
</ul>
