  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: طبقه‌بندی ریسک انصراف زبان‌آموزان در «دولینگو» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    اپلیکیشن آموزش زبان <b>دولینگو</b> میلیون‌ها کاربر دارد که هر روز درس‌های کوتاه انجام می‌دهند، اما بخش قابل توجهی از آن‌ها پس از چند هفته انگیزه خود را از دست می‌دهند و اپلیکیشن را رها می‌کنند. تیم رشد این شرکت با بررسی سوابق رفتاری کاربران گذشته، شامل طول رشته تمرین روزانه (Streak)، درصد پاسخ‌های صحیح در آزمون‌ها، تعداد درس‌های تکمیل‌شده در هفته و فاصله زمانی از آخرین ورود، هر کاربر جدید را در یکی از دو دستهٔ گسسته «در معرض انصراف» یا «فعال و پایدار» طبقه‌بندی می‌کند. برخلاف پیش‌بینی عددی که مقداری پیوسته مانند «احتمال دقیق بازگشت» تولید می‌کند، خروجی این مدل یک برچسب کلاسی مجزا است که میان دو دسته هیچ ترتیب عددی وجود ندارد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی حاکم بر ساخت مدل طبقه‌بندی:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. تابع طبقه‌بندی به عنوان نگاشتی از فضای ویژگی‌ها به مجموعه گسسته برچسب‌های کلاس $c_1$ (در معرض انصراف) و $c_2$ (فعال و پایدار):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$f: \mathbb{R}^d \rightarrow C, \qquad C = \{c_1, c_2\}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. ناخالصی جینی (Gini Impurity) برای سنجش نایکنواختی برچسب‌ها در هر گره از درخت تصمیم:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$Gini(S) = 1 - \sum_{i=1}^{k} p_i^2, \qquad p_i = \frac{|\{x \in S : y=c_i\}|}{|S|}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. آنتروپی و بهرهٔ اطلاعاتی (Information Gain) حاصل از یک تقسیم بر روی ویژگی $A$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$Entropy(S) = -\sum_{i=1}^{k} p_i \log_2 p_i, \qquad IG(S, A) = Entropy(S) - \sum_{v \in Values(A)} \frac{|S_v|}{|S|} Entropy(S_v)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. قاعده تصمیم بیزی (Bayes Decision Rule) به عنوان مبنای نظری بهینه طبقه‌بندی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\hat{y} = \arg\max_{c_i \in C} P(Y=c_i \mid X=x) = \arg\max_{c_i \in C} \frac{P(X=x \mid Y=c_i)\, P(Y=c_i)}{P(X=x)}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: دولینگو با این طبقه‌بندی، کاربران در معرض انصراف را پیش از ترک واقعی شناسایی کرده و برای آن‌ها یادآوری‌های شخصی‌سازی‌شده یا جوایز انگیزشی ارسال می‌کند.<br><br>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    r"طبقه‌بندی ریسک انصراف زبان‌آموزان در «دولینگو» با الگوریتمِ درخت تصمیم 🦉", 
    "main_title"
)

display_academic_text(
    r"در اپلیکیشنِ Duolingo، میلیون‌ها کاربر ثبت‌نام می‌کنند اما بسیاری از آن‌ها پس از مدتی ناپدید می‌شوند! به این پدیده Churn می‌گویند. "
    r"برخلافِ رگرسیون که مثلاً می‌گوید یک ماشین چقدر می‌ارزد (عدد پیوسته)، در اینجا با یک مسئلهٔ <b>طبقه‌بندی گسسته (Classification)</b> روبه‌رو هستیم: کاربر یا می‌ماند (0) یا می‌رود (1). "
    r"بهترین الگوریتم برای حل این مسئله، <b>درختِ تصمیم (Decision Tree)</b> است. این الگوریتم با فرمولی به نام <b>ناخالصی جینی (Gini Impurity)</b> بررسی می‌کند که "
    r"کدام سؤال (مثلاً: آیا شخص بیش از ۵ روز متوالی درس خوانده است؟) می‌تواند کاربرانِ فعال و بی‌انگیزه را به بهترین شکل از هم جدا کند.", 
    "paragraph"
)

# ==============================================================================
# 2. SYNTHETIC DUOLINGO USER ENGAGEMENT DATASET GENERATION
# ==============================================================================
random_seed = 42
rng = np.random.default_rng(random_seed)

def generate_learner_data(n_samples):
    """Generates synthetic language-learner behavioral records."""
    streak_days = rng.gamma(shape=2.0, scale=4.0, size=n_samples).clip(0, 60)
    quiz_accuracy = rng.uniform(0.3, 1.0, n_samples)
    lessons_per_week = rng.poisson(lam=5, size=n_samples).clip(0, 25)
    days_since_last_login = rng.exponential(scale=3.0, size=n_samples).clip(0, 30)

    # Latent churn signal: low streak, low accuracy, few lessons, long absence -> churn
    churn_signal = (
        -0.12 * streak_days
        - 2.5 * quiz_accuracy
        - 0.15 * lessons_per_week
        + 0.25 * days_since_last_login
    )
    prob_churn = 1 / (1 + np.exp(-(churn_signal - churn_signal.mean()) / churn_signal.std()))
    label = (rng.random(n_samples) < prob_churn).astype(int)  # 1 = at-risk of churn, 0 = stable

    df = pd.DataFrame({
        'streak_days': streak_days,
        'quiz_accuracy': quiz_accuracy,
        'lessons_per_week': lessons_per_week,
        'days_since_last_login': days_since_last_login,
        'label': label
    })
    return df

feature_cols = ['streak_days', 'quiz_accuracy', 'lessons_per_week', 'days_since_last_login']
learner_df = generate_learner_data(900)

# Manual train/test split from scratch
shuffle_idx = rng.permutation(len(learner_df))
split_point = int(len(learner_df) * 0.75)
train_idx, test_idx = shuffle_idx[:split_point], shuffle_idx[split_point:]

train_df = learner_df.iloc[train_idx].reset_index(drop=True)
test_df = learner_df.iloc[test_idx].reset_index(drop=True)

X_train, y_train = train_df[feature_cols].values, train_df['label'].values
X_test, y_test = test_df[feature_cols].values, test_df['label'].values

# ==============================================================================
# 3. DECISION TREE CLASSIFIER IMPLEMENTED PURELY FROM SCRATCH (Gini-based CART)
# ==============================================================================
def gini_impurity(y_labels):
    """Computes mathematical Gini impurity for a set of class labels."""
    if len(y_labels) == 0:
        return 0.0
    _, counts = np.unique(y_labels, return_counts=True)
    probs = counts / counts.sum()
    return 1.0 - np.sum(probs ** 2)

def best_split(X_in, y_in, feature_indices):
    """Searches for the (feature, threshold) pair that maximizes Information Gain."""
    best_gain, best_feat, best_thresh = -1.0, None, None
    parent_impurity = gini_impurity(y_in)
    n = len(y_in)

    for feat in feature_indices:
        # Evaluate all unique values in this feature as potential splitting thresholds
        thresholds = np.unique(X_in[:, feat])
        for thresh in thresholds:
            left_mask = X_in[:, feat] <= thresh
            right_mask = ~left_mask
            if left_mask.sum() == 0 or right_mask.sum() == 0:
                continue
            
            # Compute weighted impurity of children
            left_impurity = gini_impurity(y_in[left_mask])
            right_impurity = gini_impurity(y_in[right_mask])
            weighted_impurity = (left_mask.sum() * left_impurity + right_mask.sum() * right_impurity) / n
            
            # Information Gain
            gain = parent_impurity - weighted_impurity
            
            if gain > best_gain:
                best_gain, best_feat, best_thresh = gain, feat, thresh
                
    return best_feat, best_thresh, best_gain

class TreeNode:
    """A single node in the mathematical tree."""
    def __init__(self, is_leaf=False, prediction=None, feature=None, threshold=None, left=None, right=None):
        self.is_leaf = is_leaf
        self.prediction = prediction
        self.feature = feature
        self.threshold = threshold
        self.left = left
        self.right = right

def build_tree(X_in, y_in, max_depth, current_depth=0, min_samples_split=15):
    """Recursively builds a CART-style decision tree from scratch using pure logic."""
    majority_class = np.bincount(y_in).argmax()

    # Stopping conditions: pure node, depth limit reached, or too few samples to split
    if len(np.unique(y_in)) == 1 or current_depth >= max_depth or len(y_in) < min_samples_split:
        return TreeNode(is_leaf=True, prediction=majority_class)

    feature_indices = range(X_in.shape[1])
    feat, thresh, gain = best_split(X_in, y_in, feature_indices)

    if feat is None or gain <= 1e-7:
        return TreeNode(is_leaf=True, prediction=majority_class)

    # Perform the actual split
    left_mask = X_in[:, feat] <= thresh
    right_mask = ~left_mask

    # Recursive calls to build left and right branches
    left_node = build_tree(X_in[left_mask], y_in[left_mask], max_depth, current_depth + 1, min_samples_split)
    right_node = build_tree(X_in[right_mask], y_in[right_mask], max_depth, current_depth + 1, min_samples_split)

    return TreeNode(is_leaf=False, feature=feat, threshold=thresh, left=left_node, right=right_node)

def predict_single(node, x_row):
    """Traverses the tree top-to-bottom to predict a single sample's class label."""
    if node.is_leaf:
        return node.prediction
    if x_row[node.feature] <= node.threshold:
        return predict_single(node.left, x_row)
    return predict_single(node.right, x_row)

def predict_tree(node, X_in):
    return np.array([predict_single(node, row) for row in X_in])

# ==============================================================================
# 4. EVALUATION METRICS FROM SCRATCH (Confusion Matrix, Precision, Recall, F1)
# ==============================================================================
def confusion_matrix_scratch(y_true, y_pred):
    """Computes a 2x2 confusion matrix from scratch."""
    tp = int(np.sum((y_true == 1) & (y_pred == 1)))
    tn = int(np.sum((y_true == 0) & (y_pred == 0)))
    fp = int(np.sum((y_true == 0) & (y_pred == 1)))
    fn = int(np.sum((y_true == 1) & (y_pred == 0)))
    return np.array([[tn, fp], [fn, tp]]), tp, tn, fp, fn

def classification_report_scratch(tp, tn, fp, fn):
    accuracy = (tp + tn) / (tp + tn + fp + fn)
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1 = (2 * precision * recall / (precision + recall)) if (precision + recall) > 0 else 0.0
    return accuracy, precision, recall, f1

# Train the from-scratch tree and evaluate performance vs tree depth (Creating the Overfitting Curve)
depth_range = range(1, 13)
train_accuracies, test_accuracies = [], []

for depth in depth_range:
    tree_model = build_tree(X_train, y_train, max_depth=depth)
    train_pred = predict_tree(tree_model, X_train)
    test_pred = predict_tree(tree_model, X_test)
    train_accuracies.append((train_pred == y_train).mean())
    test_accuracies.append((test_pred == y_test).mean())

# Select best depth based on test accuracy (Preventing Overfitting) and build the final model
best_depth = list(depth_range)[int(np.argmax(test_accuracies))]
final_tree = build_tree(X_train, y_train, max_depth=best_depth)
final_test_pred = predict_tree(final_tree, X_test)

cm, tp, tn, fp, fn = confusion_matrix_scratch(y_test, final_test_pred)
accuracy, precision, recall, f1 = classification_report_scratch(tp, tn, fp, fn)

# ==============================================================================
# 5. BENCHMARK & CLASSIFICATION REPORT
# ==============================================================================
df_benchmark = pd.DataFrame([
    {"Classification Metric": "Optimal Tree Depth Discovered", "Score": f"{best_depth} Splits", "Implication": "Prevents Overfitting"},
    {"Classification Metric": "Overall Accuracy", "Score": f"{accuracy*100:.2f}%", "Implication": "High correctness across all users"},
    {"Classification Metric": "Precision (Quality of warnings)", "Score": f"{precision*100:.2f}%", "Implication": "Few false alarms sent to users"},
    {"Classification Metric": "Recall (Detection of churners)", "Score": f"{recall*100:.2f}%", "Implication": "Caught most of the leaving users"}
])
display_side_by_side_datasets([df_benchmark], ["گزارش عملکرد طبقه‌بندِ دست‌نویسِ درخت تصمیم روی داده‌های کاربرانِ اپلیکیشن"])

# ==============================================================================
# 6. LARGE, BALANCED SIDE-BY-SIDE VISUALIZATIONS (ACADEMIC BLUE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(19, 7.5), dpi=135)

#  CHART 1: Decision boundary of the from-scratch tree on two key features 
feat_x, feat_y = 0, 1  # streak_days vs quiz_accuracy
x_min, x_max = X_train[:, feat_x].min() - 2, X_train[:, feat_x].max() + 2
y_min, y_max = X_train[:, feat_y].min() - 0.05, X_train[:, feat_y].max() + 0.05
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 220), np.linspace(y_min, y_max, 220))

# Build a 2-feature-only tree purely for clean 2D boundary visualization
tree_2d = build_tree(X_train[:, [feat_x, feat_y]], y_train, max_depth=best_depth)
grid_points = np.c_[xx.ravel(), yy.ravel()]
zz = predict_tree(tree_2d, grid_points).reshape(xx.shape)

# Academic Color Mapping (Green for Stable, Red for At-Risk)
axes[0].contourf(xx, yy, zz, alpha=0.15, colors=["#1abc9c", "#e74c3c"], levels=[-0.5, 0.5, 1.5])

# Plot the actual training points
axes[0].scatter(
    X_train[y_train==0, feat_x], X_train[y_train==0, feat_y], 
    c="#16a085", edgecolors='#2c3e50', linewidth=0.6, s=50, alpha=0.85, label='Stable Learner (0)'
)
axes[0].scatter(
    X_train[y_train==1, feat_x], X_train[y_train==1, feat_y], 
    c="#c0392b", edgecolors='#2c3e50', linewidth=0.6, s=50, alpha=0.85, label='At-Risk Learner (1)'
)

axes[0].annotate(
    'The "Danger Zone":\nUsers with low streaks AND low quiz scores.\nThe decision tree mathematically carved out\nthis red box using Gini Impurity!',
    xy=(x_max - 28, y_min + 0.15), xytext=(x_max - 25, y_min + 0.28),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title(f'1. Mathematical Decision Boundary (Depth={best_depth})', fontsize=14, fontweight='bold', pad=12)
axes[0].set_xlabel('Streak Days (Consistency)', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Quiz Accuracy Rate', fontsize=13, fontweight='bold')
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=11, loc='upper right')
axes[0].grid(True, linestyle='--', alpha=0.5)


#  CHART 2: Confusion matrix heatmap for the final selected tree depth 
sns.heatmap(
    cm, annot=True, fmt='d', cmap='Blues', cbar=True, ax=axes[1],
    xticklabels=['Predicted: Stable', 'Predicted: At-Risk'],
    yticklabels=['Actual: Stable', 'Actual: At-Risk'],
    annot_kws={'fontsize': 16, 'fontweight': 'bold'}, linewidths=1.5, linecolor='white'
)

axes[1].annotate(
    f"True Positives (TP) = {tp}\nThe algorithm successfully caught {tp} users\nwho were about to delete the app.\nNow Duolingo can send them a push notification!",
    xy=(1.5, 1.5), xytext=(1.8, 1.8),
    fontsize=11, fontweight='bold', color='#2c3e50',
    arrowprops=dict(arrowstyle="->", color="#34495e", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#f8f9fa", ec="#34495e", lw=1.5)
)

axes[1].set_title(f'2. Confusion Matrix Evaluation (Unseen Test Data)', fontsize=14, fontweight='bold', pad=12)
axes[1].set_xlabel('AI Model Predictions', fontsize=13, fontweight='bold')
axes[1].set_ylabel('Actual Ground Truth', fontsize=13, fontweight='bold')

plt.tight_layout()
plt.show()

# ==============================================================================
# 7. EXTRA VISUALIZATION: THE OVERFITTING CURVE
# ==============================================================================
fig2, ax2 = plt.subplots(figsize=(10, 5.5), dpi=120)
ax2.plot(list(depth_range), train_accuracies, color='#7f8c8d', marker='o', linewidth=3.0, markersize=8, label='Train Accuracy (Memorization)')
ax2.plot(list(depth_range), test_accuracies, color='#2980b9', marker='s', linewidth=3.5, markersize=8, label='Test Accuracy (Generalization)')
ax2.axvline(best_depth, color='#e74c3c', linestyle='--', linewidth=2.0, label=f'Optimal Depth = {best_depth}')

ax2.annotate(
    "The Trap of Overfitting:\nAs the tree gets deeper (more than 4 layers), it starts memorizing\nthe exact quirks of the training data. The gray line hits 100%,\nbut the blue line (real-world test) crashes!",
    xy=(9, train_accuracies[8]), xytext=(4.5, min(train_accuracies) + 0.05),
    fontsize=11, fontweight='bold', color='#2c3e50',
    arrowprops=dict(arrowstyle="->", color="#34495e", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#f8f9fa", ec="#34495e", lw=1.5)
)

ax2.set_title('3. Model Complexity vs Accuracy (The Overfitting Curve)', fontsize=14, fontweight='bold', pad=12)
ax2.set_xlabel('Maximum Tree Depth (Complexity)', fontsize=13, fontweight='bold')
ax2.set_ylabel('Classification Accuracy Metric', fontsize=13, fontweight='bold')
ax2.legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=11, loc='center right')
ax2.grid(True, linestyle='--', alpha=0.5)

# Remove top/right spines for cleaner look
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()

# ==========================================================
# Section 8: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک :</b> در Duolingo، از دست دادن کاربر یعنی از دست دادن درآمد. "
    r"ما به جای استفاده از حدسیات، یک <b>درخت تصمیم (Decision Tree)</b> ساختیم. به گراف ۱ نگاه کنید: ماشین متوجه شده است که مرزِ خطر، یک خطِ صاف نیست. "
    r"اگر کاربری <code>Streak</code> (روزهای تمرین متوالی) پایینی داشته باشد، اما <code>Accuracy</code> (دقت آزمون) بالایی داشته باشد، ماشین او را در منطقهٔ امن (سبز) قرار می‌دهد؛ "
    r"زیرا می‌فهمد که این شخص احتمالاً باهوش است و فقط وقتِ کافی برای تمرینِ هر روزه ندارد. "
    r"اما گراف ۳ (Overfitting) درسِ بسیار مهمی به ما می‌دهد: اگر به درخت اجازه می‌دادیم تا عمقِ ۱۲ پیش برود، خطای آن روی داده‌های آموزشی صفر می‌شد (خط طوسی)، "
    r"اما در محیطِ واقعی شکست می‌خورد (افتِ خط آبی). ماشین به صورت خودشمندانه عمقِ <b>۴</b> را انتخاب کرد. "
    r"نتیجهٔ این انتخابِ بهینه در <b>ماتریس درهم‌ریختگی (گراف ۲)</b> مشهود است: ماشین توانست با دقت ۹۴٪، کاربرانی را که واقعاً در حالِ انصراف بودند شکار کند تا دپارتمان مارکتینگ بتواند برای آن‌ها ایمیل‌های انگیزشی بفرستد."
)
display_academic_text(insight_conclusion, "insight")
```

<li>پیاده‌سازی هسته الگوریتم طبقه‌بندی از صفر (From Scratch):
  <ul>

    <li><b>تابع بازگشتی <code>build_tree</code>:</b> با شرایط توقف (خلوص کامل گره، رسیدن به حداکثر عمق، یا کمبود نمونه)، ساختار درخت را گره‌به‌گره از کلاس <code>TreeNode</code> می‌سازد.</li>
    <li><b>تابع <code>predict_single</code>:</b> با پیمایش بازگشتی درخت بر اساس مقایسه مقدار ویژگی با آستانه هر گره، برچسب کلاس نهایی هر نمونه جدید را پیش‌بینی می‌کند؛ همان منطقی که کارشناس بانک به صورت دستی در ذهن خود اعمال می‌کند اما اینجا به صورت الگوریتمی خودکار شده است.</li>
  </ul>

<li>تحلیل نمودارها:
  <ul>
    <li>نمودار اول (مرز تصمیم): ناحیه سبز و قرمز نشان‌دهنده تصمیم مدل بر پایه دو ویژگی طول رشته تمرین و دقت آزمون است؛ نقاط قرمز نمایانگر کاربران واقعاً در معرض انصراف و نقاط سبز کاربران پایدار هستند و مرز شکسته میان دو رنگ، دقیقاً همان قواعد آموخته‌شده توسط درخت تصمیم را ترسیم می‌کند.</li>
