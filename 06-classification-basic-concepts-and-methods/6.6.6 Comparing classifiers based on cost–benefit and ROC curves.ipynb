  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: مقایسه مدل‌های تشخیص تقلب با ROC و تحلیل هزینه-سود در «استرایپ» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پلتفرم پرداخت <b>استرایپ</b>، تیم امنیت مالی با یک چالش حیاتی روبروست: انتخاب بین دو مدل تشخیص تراکنش تقلبی — <b>رگرسیون لجستیک</b> (سریع و قابل تفسیر) و <b>جنگل تصادفی</b> (پیچیده‌تر اما دقیق‌تر). اما معیار «دقت» به تنهایی کافی نیست؛ زیرا هزینه‌های چهار نتیجه ممکن کاملاً نامتقارن‌اند: یک <b>مثبت کاذب</b> (رد تراکنش واقعی مشتری) باعث نارضایتی و از دست رفتن درآمد می‌شود، در حالی که یک <b>منفی کاذب</b> (قبول تراکنش تقلبی) می‌تواند ده‌ها هزار دلار خسارت مستقیم به بار آورد. مهندسان استرایپ با ترکیب <b>منحنی ROC</b> و <b>تحلیل هزینه-سود</b>، مدلی را انتخاب می‌کنند که نه تنها AUC بالاتری دارد، بلکه <b>سود خالص انتظاری</b> آن در سناریوی واقعی کسب‌وکار بیشینه است.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    نرخ مثبت واقعی (TPR) و نرخ مثبت کاذب (FPR) در هر آستانه $\theta$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$TPR(\theta) = \frac{TP}{TP + FN}, \quad FPR(\theta) = \frac{FP}{FP + TN}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مساحت زیر منحنی ROC (AUC) با انتگرال‌گیری عددی ذوزنقه‌ای:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$AUC = \int_{0}^{1} TPR \, d(FPR) \approx \sum_{i=1}^{n-1} \frac{(FPR_{i+1} - FPR_i)(TPR_{i+1} + TPR_i)}{2}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    تابع سود خالص (Net Benefit) با وزن‌دهی اقتصادی به چهار خانه ماتریس درهم‌ریختگی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{NetBenefit}(\theta) = \frac{TP \cdot B_{TP} + TN \cdot B_{TN} - FP \cdot C_{FP} - FN \cdot C_{FN}}{N}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    آستانه بهینه اقتصادی $\theta^*$ که سود خالص را بیشینه می‌کند:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\theta^* = \arg\max_{\theta} \; \text{NetBenefit}(\theta)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: استرایپ با این رویکرد دوگانه، هم از نظر آماری (AUC) و هم از نظر اقتصادی (سود خالص) مدل برتر را شناسایی می‌کند؛ ممکن است مدلی با AUC پایین‌تر، به دلیل هزینه کمتر خطاهای گران‌قیمت، در عمل سودآورتر باشد.<br><br>

      خروجی: انتخاب مدل برتر با ترکیب معیار آماری (AUC) و معیار اقتصادی (Net Benefit)
    </div>
  </p>

</div>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from IPython.display import display, HTML
import warnings

warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={"axes.titlesize": 14, "axes.labelsize": 12, "xtick.labelsize": 10, "ytick.labelsize": 10})

plt.rcParams['figure.dpi'] = 120

def display_academic_text(text, type="normal"):
    styles = {
        "main_title": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 24px; "
            "font-weight: bold; text-align: center; border-bottom: 2px solid #34495e; "
            "padding-bottom: 15px; margin: 30px 0;"
        ),
        "sub_title": (
            "font-family: Tahoma, sans-serif; color: #34495e; font-size: 19px; "
            "font-weight: bold; border-right: 5px solid #2980b9; padding-right: 15px; "
            "margin: 40px 0px 20px 0px; background-color: #ecf0f1; padding: 12px; "
            "border-radius: 4px;"
        ),
        "paragraph": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 16px; "
            "margin-bottom: 20px; line-height: 1.8; text-align: justify;"
        ),
        "insight": (
            "font-family: Tahoma, sans-serif; color: #16a085; font-size: 16px; "
            "font-weight: bold; background-color: #e8f8f5; padding: 20px; "
            "border-left: 5px solid #1abc9c; margin: 30px 0; line-height: 1.8; "
            "text-align: justify;"
        )
    }
    display(HTML(f"<div dir='rtl' style='text-align: right; {styles[type]}'>{text}</div>"))

def get_protected_css():
    return """
    <style>
    .academic-table-wrapper { 
        direction: ltr !important; 
        overflow-x: auto !important; 
        width: 100% !important; 
        background-color: #fff; 
        border: 1px solid #bdc3c7; 
        border-radius: 4px; 
        box-shadow: 0 3px 6px rgba(0,0,0,0.08); 
    }
    table.strict-tbl { 
        border-collapse: collapse !important; 
        width: 100% !important; 
        margin: 0 !important; 
        font-family: inherit !important;
        font-size: 14px !important; 
        table-layout: auto !important; 
        border: none !important; 
    }
    table.strict-tbl thead th { 
        background-color: #2c3e50 !important; 
        color: #ffffff !important; 
        padding: 12px 18px !important; 
        text-align: center !important; 
        border: 1px solid #34495e !important; 
        white-space: nowrap !important; 
        font-weight: bold !important; 
    }
    table.strict-tbl tbody td { 
        padding: 10px 18px !important; 
        text-align: center !important; 
        border: 1px solid #bdc3c7 !important; 
        white-space: nowrap !important; 
        color: #2c3e50 !important; 
    }
    table.strict-tbl tbody tr:nth-child(even) td { 
        background-color: #f8f9fa !important; 
    }
    table.strict-tbl tbody tr:hover td { 
        background-color: #e8ecef !important; 
    }
    </style>
    """

def display_side_by_side_datasets(dfs, titles, *args):
    if not isinstance(dfs, list) and not isinstance(titles, list):
        dfs = [dfs]
        titles = [titles]
        
    col_html_list = []
    for i, (df, title) in enumerate(zip(dfs, titles)):
        table_html = df.to_html(index=False, classes=['strict-tbl'], border=0)
        bg_color = "#2980b9" if i % 2 == 0 else "#2c3e50"
        col_html_list.append(f"""
        <div class="flex-column" style="flex: 1 1 0; min-width: 0; display: flex; flex-direction: column;">
            <div style="background-color: {bg_color}; color: white; padding: 12px; 
                        text-align: center; font-family: Tahoma, sans-serif; 
                        font-weight: bold; font-size: 15px; border-radius: 4px 4px 0 0;">
                {title}
            </div>
            <div class="academic-table-wrapper" style="margin-top: 0; border-top: none;">{table_html}</div>
        </div>
        """)
    
    display(HTML(get_protected_css() + f'<div style="display: flex; gap: 20px; width: 100%; margin: 20px 0 35px 0;">{"".join(col_html_list)}</div>'))

display(HTML("""
<style>
@import url('https://cdn.jsdelivr.net/gh/rastikerdar/vazirmatn@v33.003/Vazirmatn-font-face.css');
* { font-family: 'Vazirmatn', 'Vazir', Tahoma, sans-serif !important; }
</style>
"""))
```

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "مقایسه مدل‌های تشخیص تقلب با ROC و تحلیل هزینه-سود در «استرایپ» (From Scratch) 💳", 
    "main_title"
)

display_academic_text(
    "در پلتفرم پرداخت استرایپ، تیم امنیت مالی با چالش انتخاب مدل مواجه است. معیار «دقت» کافی نیست؛ "
    "زیرا خطاهای مثبت کاذب (بلاک کردن مشتری واقعی) باعث ریزش درآمد، و خطاهای منفی کاذب (قبول تراکنش کلاهبردار) "
    "باعث ضررهای هنگفت هزاران دلاری (Chargebacks) می‌شود. در این تمرین، با ترکیب <b>منحنی ROC</b> و <b>تحلیل سود خالص (Net Benefit)</b> "
    "نشان می‌دهیم که چگونه آستانه تصمیم‌گیری (Threshold) بر اساس اقتصادِ کسب‌وکار تعیین می‌شود.", 
    "paragraph"
)

# ==============================================================================
# 2. SIMULATE STRIPE FRAUD DATASET (HIGHLY IMBALANCED)
# ==============================================================================
np.random.seed(42)
n_samples = 2000

# Feature 1: Transaction Amount (Normalized)
# Feature 2: Distance from last IP Address login
# Feature 3: Time since account creation
# Feature 4: Number of failed attempts

# 95% Legitimate Transactions (Class 0)
n_legit = int(n_samples * 0.95)
X_legit = np.column_stack((
    np.random.normal(50, 20, n_legit),
    np.random.normal(10, 5, n_legit),
    np.random.normal(365, 100, n_legit),
    np.random.poisson(0.5, n_legit)
))
y_legit = np.zeros(n_legit)

# 5% Fraudulent Transactions (Class 1) - Often higher amounts, fast IP jumps, new accounts
n_fraud = n_samples - n_legit
X_fraud = np.column_stack((
    np.random.normal(250, 80, n_fraud),
    np.random.normal(150, 50, n_fraud),
    np.random.normal(5, 2, n_fraud),
    np.random.poisson(3.5, n_fraud)
))
y_fraud = np.ones(n_fraud)

X_data = np.vstack((X_legit, X_fraud))
y_data = np.concatenate((y_legit, y_fraud))

# Shuffle and Split
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X_data, y_data, test_size=0.3, random_state=42, stratify=y_data
)

# Standardize
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train_raw)
X_test_s = scaler.transform(X_test_raw)

df_fraud = pd.DataFrame(X_train_raw[:4], columns=['Amount ($)', 'IP Distance (km)', 'Account Age (Days)', 'Failed Logins'])
df_fraud['Is Fraud?'] = y_train[:4].astype(int)
display_side_by_side_datasets([df_fraud], ["نمونه‌ای از داده‌های شبیه‌سازی‌شده تراکنش‌های استرایپ (کلاهبرداری بسیار نادر است)"])

# ==============================================================================
# 3. LOGISTIC REGRESSION FROM SCRATCH (BATCH GRADIENT DESCENT)
# ==============================================================================
class LogisticRegressionScratch:
    """Binary logistic regression with sigmoid + cross-entropy loss built from scratch."""
    def __init__(self, lr=0.1, n_epochs=1000):
        self.lr = lr
        self.n_epochs = n_epochs
        self.beta = None

    @staticmethod
    def sigmoid(z):
        return np.where(z >= 0, 1/(1+np.exp(-z)), np.exp(z)/(1+np.exp(z)))

    def fit(self, X, y):
        X_b = np.column_stack([np.ones(len(X)), X])
        self.beta = np.zeros(X_b.shape[1])
        for _ in range(self.n_epochs):
            p = self.sigmoid(X_b @ self.beta)
            grad = (X_b.T @ (p - y)) / len(y)
            self.beta -= self.lr * grad
        return self

    def predict_proba(self, X):
        X_b = np.column_stack([np.ones(len(X)), X])
        return self.sigmoid(X_b @ self.beta)

# Train Logistic Regression (Linear)
logreg = LogisticRegressionScratch(lr=0.1, n_epochs=1000)
logreg.fit(X_train_s, y_train)
prob_logreg = logreg.predict_proba(X_test_s)

# Train Random Forest (Non-Linear, usually better at detecting fraud patterns)
rf = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42)
rf.fit(X_train_s, y_train)
prob_rf = rf.predict_proba(X_test_s)[:, 1]

# ==============================================================================
# 4. ROC CURVE AND AUC FROM SCRATCH
# ==============================================================================
def roc_curve_scratch(y_true, y_prob, n_thresholds=300):
    """Computes ROC curve by sweeping decision thresholds from scratch."""
    thresholds = np.linspace(0.0, 1.0, n_thresholds)
    fprs, tprs = [], []

    for thresh in thresholds:
        y_pred = (y_prob >= thresh).astype(int)
        # Calculate Confusion Matrix Elements
        TP = np.sum((y_pred == 1) & (y_true == 1))
        TN = np.sum((y_pred == 0) & (y_true == 0))
        FP = np.sum((y_pred == 1) & (y_true == 0))
        FN = np.sum((y_pred == 0) & (y_true == 1))

        # TPR (Recall/Sensitivity) and FPR
        tpr = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0.0
        tprs.append(tpr)
        fprs.append(fpr)

    fprs, tprs = np.array(fprs), np.array(tprs)
    # Sort for area calculation
    order = np.argsort(fprs)
    return fprs[order], tprs[order], thresholds[order]

def auc_trapezoidal(fprs, tprs):
    """Computes AUC using the trapezoidal rule from scratch."""
    auc = 0.0
    for i in range(len(fprs) - 1):
        auc += (fprs[i+1] - fprs[i]) * (tprs[i+1] + tprs[i]) / 2
    return float(auc)

fpr_lr, tpr_lr, _ = roc_curve_scratch(y_test, prob_logreg)
fpr_rf, tpr_rf, _ = roc_curve_scratch(y_test, prob_rf)

auc_lr = auc_trapezoidal(fpr_lr, tpr_lr)
auc_rf = auc_trapezoidal(fpr_rf, tpr_rf)

# ==============================================================================
# 5. COST-BENEFIT ANALYSIS FROM SCRATCH (THE BUSINESS LOGIC)
# ==============================================================================
def cost_benefit_analysis(y_true, y_prob, thresholds, cost_matrix):
    """
    Translates ML metrics into Dollar values.
    Net Benefit = (Profits from Correct Decisions - Losses from Mistakes) / Total Transactions
    """
    N = len(y_true)
    net_benefits = []

    for thresh in thresholds:
        y_pred = (y_prob >= thresh).astype(int)
        TP = np.sum((y_pred == 1) & (y_true == 1))
        TN = np.sum((y_pred == 0) & (y_true == 0))
        FP = np.sum((y_pred == 1) & (y_true == 0))
        FN = np.sum((y_pred == 0) & (y_true == 1))

        # Financial Equation
        benefit = (
            TP * cost_matrix['TP'] +
            TN * cost_matrix['TN'] -
            FP * cost_matrix['FP'] -
            FN * cost_matrix['FN']
        ) / N
        net_benefits.append(benefit)

    return np.array(net_benefits)

# Define a hyper-realistic FinTech Cost Matrix
cost_matrix = {
    'TP': 150,   # Profit from correctly stopping fraud (Saved chargeback fee + blocked loss)
    'TN': 2,     # Small processing fee earned on a successful normal transaction
    'FP': 25,    # Cost of a False Alarm: Customer gets angry, card declined, loss of future lifetime value
    'FN': 600    # FATAL ERROR: Letting a fraudster through costs the merchant $600 in stolen goods + Stripe penalties
}

thresholds_eval = np.linspace(0.01, 0.99, 200)
nb_logreg = cost_benefit_analysis(y_test, prob_logreg, thresholds_eval, cost_matrix)
nb_rf = cost_benefit_analysis(y_test, prob_rf, thresholds_eval, cost_matrix)

# Find the threshold that makes the company the MOST money
opt_thresh_lr = thresholds_eval[np.argmax(nb_logreg)]
opt_thresh_rf = thresholds_eval[np.argmax(nb_rf)]
opt_nb_lr = np.max(nb_logreg)
opt_nb_rf = np.max(nb_rf)

# ==============================================================================
# 6. VISUALIZATIONS AND CHARTS (ACADEMIC BLUE PALETTE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

#  CHART 1: ROC Curves Comparison 
# Academic Blue for RF, Dark Navy for LR
axes[0].plot(fpr_rf, tpr_rf, color='#2980b9', linewidth=3, label=f'Random Forest (AUC = {auc_rf:.3f})')
axes[0].plot(fpr_lr, tpr_lr, color='#2c3e50', linewidth=3, label=f'Logistic Regression (AUC = {auc_lr:.3f})')
axes[0].plot([0, 1], [0, 1], color='#e74c3c', linewidth=2, linestyle='--', label='Random Guessing (AUC = 0.5)')

axes[0].fill_between(fpr_rf, tpr_rf, alpha=0.15, color='#2980b9')

# Calculate coordinates for the "Academic Standard" Threshold = 0.5
y_pred_rf_05 = (prob_rf >= 0.5).astype(int)
TP_rf, FN_rf = np.sum((y_pred_rf_05 == 1) & (y_test == 1)), np.sum((y_pred_rf_05 == 0) & (y_test == 1))
FP_rf, TN_rf = np.sum((y_pred_rf_05 == 1) & (y_test == 0)), np.sum((y_pred_rf_05 == 0) & (y_test == 0))
tpr_rf_05, fpr_rf_05 = TP_rf / (TP_rf + FN_rf), FP_rf / (FP_rf + TN_rf)

axes[0].scatter([fpr_rf_05], [tpr_rf_05], s=150, color='#e74c3c', zorder=5, edgecolor='black', label=f'RF at Default θ=0.50')

axes[0].set_xlabel('False Positive Rate (FPR)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('True Positive Rate (TPR / Sensitivity)', fontsize=12, fontweight='bold')
axes[0].set_title('1. ROC Curves — Fraud Detection Models\n(Higher AUC = Better General Discrimination)', fontsize=13, fontweight='bold', pad=12)
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='lower right')
axes[0].grid(True, linestyle='--', alpha=0.6)

#  CHART 2: Financial Net Benefit vs Threshold 
axes[1].plot(
    thresholds_eval, nb_rf, color='#2980b9', linewidth=3,
    label=f'Random Forest (Max Profit = ${opt_nb_rf:.2f} per txn)'
)
axes[1].plot(
    thresholds_eval, nb_logreg, color='#2c3e50', linewidth=3,
    label=f'Logistic Regression (Max Profit = ${opt_nb_lr:.2f} per txn)'
)

# Mark the financially optimal points with Red Stars
axes[1].scatter([opt_thresh_rf], [opt_nb_rf], s=250, color='#e74c3c', zorder=5, edgecolor='black', marker='*')
axes[1].axvline(opt_thresh_rf, color='#e74c3c', linestyle=':', linewidth=2)

axes[1].annotate(
    f'Optimal Business Threshold\nθ = {opt_thresh_rf:.2f}',
    xy=(opt_thresh_rf, opt_nb_rf),
    xytext=(opt_thresh_rf + 0.1, opt_nb_rf - 5),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle='->', color='#c0392b', lw=2)
)
# Zero line (Break-even where the model loses as much as it makes)
axes[1].axhline(0, color='#64748B', linewidth=2, linestyle='-', label='Break-even (Zero Net Benefit)')

axes[1].set_xlabel('Decision Threshold (θ)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Expected Net Benefit per Transaction ($)', fontsize=12, fontweight='bold')
axes[1].set_title(
    f'2. Cost-Benefit Analysis (Financial Impact)\n'
    f'Fatal Error (FN) Cost: ${cost_matrix["FN"]} | False Alarm (FP) Cost: ${cost_matrix["FP"]}',
    fontsize=13, fontweight='bold', pad=12
)
axes[1].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='upper right')
axes[1].grid(True, linestyle='--', alpha=0.6)
axes[1].set_xlim(0, 1)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 7: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    "<b>نتیجه استراتژیک (ترجمه آمار به پول):</b> اگر در چارت ۱ به آستانه پیش‌فرض ۰.۵۰ (نقطه قرمز) نگاه کنید، "
    "می‌بینید که مدل در آن نقطه رفتار محافظه‌کارانه‌ای دارد. اما وقتی گراف ۲ (محاسبات مالی) را رسم می‌کنیم، "
    "واقعیت تلخ کسب‌وکار آشکار می‌شود: چون هزینه از دست دادن یک تراکنش کلاهبردارانه (FN) برابر با <b>۶۰۰ دلار</b> خسارت است "
    "در حالی که بلاک کردن اشتباه یک مشتری (FP) فقط <b>۲۵ دلار</b> هزینه دارد، استرایپ نمی‌تواند روی آستانه ۰.۵۰ ریسک کند! "
    "نمودار ۲ نشان می‌دهد که برای <b>بیشینه کردن سود خالص</b>، آستانه باید روی یک عدد بسیار پایین (حدود <b>" f"{opt_thresh_rf:.2f}" "</b>) تنظیم شود. "
    "این یعنی سیستم به حدی سخت‌گیر می‌شود که حتی با کوچکترین شک (احتمال ۱۰٪)، تراکنش را مسدود می‌کند. در این حالت، "
    "با وجود افزایش خطاهای FP، از ضررهای هنگفت کلاهبرداری جلوگیری شده و <b>Random Forest</b> به عنوان سودآورترین مدل انتخاب می‌شود."
)
display_academic_text(insight_conclusion, "insight")
```
