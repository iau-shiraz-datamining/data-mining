  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">مثال عملی: فرآیند کشف دانش در عمل 🧩</div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در این مثال، فرآیند کشف دانش را با استفاده از یک سناریو واقعی در صنعت تحلیل می‌کنیم. فرض کنید یک شرکت فروشگاهی می‌خواهد از رفتار مشتریان خود برای پیش‌بینی خرید بعدی استفاده کند. این فرآیند به شرح زیر انجام می‌شود:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>1. آماده‌سازی داده:</b><br>
    &nbsp;&nbsp;&nbsp;- داده‌های فروش از چندین فروشگاه و سیستم‌های آنلاین جمع‌آوری می‌شوند.<br>
    &nbsp;&nbsp;&nbsp;- داده‌ها پاک‌سازی می‌شوند: مثلاً رکوردهای با اطلاعات ناقص یا تاریخ‌های غیرمنطقی حذف می‌شوند.<br>
    &nbsp;&nbsp;&nbsp;- داده‌ها تجمیع و تبدیل می‌شوند: مثلاً تراکنش‌ها به دسته‌های فصلی و محصولی خلاصه می‌شوند.<br>
    &nbsp;&nbsp;&nbsp;- تنها داده‌های مرتبط با تحلیل (مثلاً خریدهای اخیر) انتخاب می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>2. داده‌کاوی:</b><br>
    &nbsp;&nbsp;&nbsp;- از الگوریتم‌هایی مانند Apriori یا FP-Growth برای کشف الگوهای خرید (مانند "خرید لپ‌تاپ → خرید کیف حمل") استفاده می‌شود.<br>
    &nbsp;&nbsp;&nbsp;- از روش‌هایی مانند درخت تصمیم یا شبکه عصبی برای پیش‌بینی احتمال خرید محصولات خاص استفاده می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>3. ارزیابی الگو:</b><br>
    &nbsp;&nbsp;&nbsp;- الگوهای کشف‌شده بر اساس معیارهایی مانند حمایت (support)، اطمینان (confidence) و جالبیت (interestingness) ارزیابی می‌شوند.<br>
    &nbsp;&nbsp;&nbsp;- مثلاً الگوی "خرید لپ‌تاپ → خرید کیف حمل" با حمایت بالا و اطمینان قابل توجه انتخاب می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>4. ارائه دانش:</b><br>
    &nbsp;&nbsp;&nbsp;- نتایج به مدیران با استفاده از گرافیک‌های تعاملی و گزارش‌های خلاصه ارائه می‌شود.<br>
    &nbsp;&nbsp;&nbsp;- این دانش به عنوان یک استراتژی بازاریابی جدید (مثلاً پیشنهاد کیف حمل هنگام خرید لپ‌تاپ) در سیستم اجرایی ذخیره می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    این مثال نشان می‌دهد که چگونه داده‌کاوی به عنوان مرحله‌ای از فرآیند کشف دانش، در ایجاد ارزش از داده‌های عظیم نقش محوری ایفا می‌کند و چگونه از یک مجموعه بزرگ داده به دانش کاربردی تبدیل می‌شود.
  </p>

```python
# این قسمت اول کده. ما کتابخونه‌های لازم رو وارد می‌کنیم تا بتونیم فرآیند کشف دانش (KDD) رو به صورت عملیاتی نشون بدیم.
# 'pandas' برای مدیریت داده، 'mlxtend' برای داده‌کاوی و 'matplotlib' برای بصری‌سازی نتایج استفاده می‌شه.
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori
import matplotlib.pyplot as plt

# این بخش، یک مجموعه داده فرضی از **تراکنش‌های خرید مشتریان** در یک شبکه فروشگاهی رو شبیه‌سازی می‌کنه.
# این داده‌ها نشون‌دهندهٔ ورودی اولیه‌ست که در فرآیند KDD استفاده می‌شه.
# هر سطر یک تراکنش و ستون "Items" شامل لیست آیتم‌های خریداری‌شده‌ست.
raw_data = {
    'TransactionID': [1, 2, 3, 4, 5, 6, 7, 8],
    'Items': [
        ['Laptop', 'Mouse', 'Bag'],
        ['Mouse', 'Keyboard'],
        ['Laptop', 'Bag', 'Charger'],
        ['Mouse', 'Bag'],
        ['Laptop', 'Keyboard'],
        ['Bag', 'Charger'],
        ['Laptop', 'Mouse', 'Keyboard'],
        ['Laptop', 'Bag']
    ],
    'Date': ['2023-10-01', '2023-10-02', '2023-10-03', '2023-10-04', '2023-10-05', '2023-10-06', '2023-10-07', '2023-10-08']
}
df = pd.DataFrame(raw_data)

# این بخش، **آماده‌سازی داده** رو انجام می‌ده — اولین مرحله از فرآیند KDD.
# 1. پاک‌سازی داده: حذف تراکنش‌های با اطلاعات ناقص (در اینجا فرض می‌کنیم داده تمیزه)
# 2. تجمیع داده: داده‌ها از چندین فروشگاه جمع‌آوری شدن (شبیه‌سازی شده با یک دیتافریم واحد)
# 3. تبدیل داده: تراکنش‌ها به فرمت مناسب برای داده‌کاوی (یک‌دسته‌ای) تبدیل می‌شن
# 4. انتخاب داده: فقط داده‌های مربوط به آخرین هفته انتخاب می‌شن (شبیه‌سازی شده)
df['Date'] = pd.to_datetime(df['Date'])
prepared_df = df[df['Date'] >= '2023-10-02'].copy()  # انتخاب داده‌های اخیر

# تبدیل داده به فرمت یک‌دسته‌ای برای الگوریتم‌های داده‌کاوی
te = TransactionEncoder()
te_ary = te.fit(prepared_df['Items']).transform(prepared_df['Items'])
basket_df = pd.DataFrame(te_ary, columns=te.columns_)

# این بخش، مرحله **داده‌کاوی** رو انجام می‌ده.
# ما از الگوریتم Apriori برای کشف الگوهای پرتکرار در خرید مشتریان استفاده می‌کنیم.
# این دقیقاً همون کاریه که در مثال "خرید لپ‌تاپ → خرید کیف حمل" اشاره شد.
# هدف کشف قوانین انجمنی (Association Rules) برای پیشنهاد محصوله.
frequent_itemsets = apriori(basket_df, min_support=0.25, use_colnames=True)

# این بخش، **ارزیابی الگو** رو شبیه‌سازی می‌کنه.
# ما فقط الگوهایی رو نگه می‌داریم که شامل "Laptop" و "Bag" باشن و پشتیبانی کافی داشته باشن.
# این فیلتر نشون‌دهندهٔ ارزیابی الگوها بر اساس جالبیت و معیارهای حمایت و اطمینانه.
relevant_patterns = frequent_itemsets[frequent_itemsets['itemsets'].apply(lambda x: {'Laptop', 'Bag'}.issubset(x))]

# این بخش، **ارائه دانش** رو شبیه‌سازی می‌کنه.
# نتایج به صورت بصری و قابل فهم برای مدیران ارائه می‌شه.
print("فرآیند کشف دانش (KDD) در عمل:")
print("\n1. داده خام:")
print(df[['TransactionID', 'Items']].head())
print("...")
print("\n2. داده پردازش‌شده (فقط ۷ روز اخیر):")
print(prepared_df[['TransactionID', 'Items']].reset_index(drop=True))
print("\n3. الگوهای پرتکرار کشف‌شده:")
print(frequent_itemsets)
print("\n4. الگوی جالب و ارزشمند (Laptop + Bag):")
print(relevant_patterns)

# نمایش بصری الگوی اصلی
plt.figure(figsize=(8, 4))
if not relevant_patterns.empty:
    plt.barh(['Laptop & Bag'], relevant_patterns.iloc[0]['support'], 
             color='skyblue', edgecolor='black')
else:
    plt.text(0.5, 0.5, 'No Pattern Found', ha='center', va='center', fontsize=12)
plt.xlim(0, 1)
plt.title('Valuable pattern discovered: Laptop and Bag co-purchase')
plt.xlabel('Support')
plt.grid(axis='x', alpha=0.3)
plt.show()
```

<p style="font-size:16pt; text-align:right;">
در این مثال، فرآیند کامل <b>کشف دانش از داده</b> (KDD) به صورت عملیاتی شبیه‌سازی شده است:
<ul>
<li>ابتدا کتابخانه‌های لازم شامل <b>pandas</b> برای مدیریت داده، <b>mlxtend</b> برای انجام داده‌کاوی و <b>matplotlib</b> برای بصری‌سازی نتایج وارد شده‌اند.</li>
<li>یک مجموعه داده فرضی از <b>تراکنش‌های خرید مشتریان</b> در یک شبکه فروشگاهی ایجاد شده که شامل شناسه تراکنش، لیست آیتم‌های خریداری‌شده و تاریخ است.</li>
<li>در مرحله <b>آماده‌سازی داده</b>، داده‌ها پاک‌سازی شده (در اینجا فرض تمیز بودن داده)، تجمیع شده (شبیه‌سازی داده از چندین فروشگاه) و با تبدیل به فرمت یک‌دسته‌ای (one-hot encoding) برای داده‌کاوی آماده می‌شوند. همچنین فقط داده‌های هفته اخیر برای تحلیل انتخاب شده‌اند.</li>
<li>در مرحله <b>داده‌کاوی</b>، از الگوریتم <b>Apriori</b> برای کشف الگوهای پرتکرار در خرید مشتریان استفاده شده است. این الگوریتم به دنبال مجموعه‌ای از آیتم‌هاست که به‌طور مکرر با هم خریداری می‌شوند (مانند "لپ‌تاپ و کیف حمل").</li>
<li>در مرحله <b>ارزیابی الگو</b>، الگوهای کشف‌شده بر اساس معیارهایی مانند <b>حمایت</b> (Support) فیلتر شده‌اند و فقط الگوهایی که شامل هم‌خرید لپ‌تاپ و کیف باشند، به عنوان الگوهای ارزشمند انتخاب شده‌اند.</li>
<li>در نهایت، در مرحله <b>ارائه دانش</b>، نتایج به صورت متنی و گرافیکی به کاربر (مثلاً مدیر فروش) ارائه شده‌اند. نمودار نشان می‌دهد که چگونه یک دانش عملیاتی ("مشتریانی که لپ‌تاپ می‌خرند، اغلب کیف نیز می‌خرند") از داده‌های خام استخراج شده است.</li>
</ul>
این مثال به‌وضوح نشان می‌دهد که چگونه داده‌کاوی به عنوان یکی از مراحل فرآیند KDD، می‌تواند به تبدیل داده‌های خام به <b>دانش کاربردی</b> برای تصمیم‌گیری‌های تجاری کمک کند.
