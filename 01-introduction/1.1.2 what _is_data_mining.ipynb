  <div style="font-size: 16pt;text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;"> مثال عملی: چطور داده‌کاوی «اکسپلور» و «ریلز» اینستاگرام را می‌سازد؟</div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>مثال ۱.۱: داده‌کاوی الگوهای پنهان رفتاری کاربران را کشف می‌کند.</b><br>
    شبکه اجتماعی اینستاگرام (Meta) روزانه با بیش از ۲ میلیارد کاربر فعال، حجم عظیمی از داده‌های رفتاری (پتابایت‌ها داده در روز) تولید می‌کند. هیچ انسان یا تیم پشتیبانی نمی‌تواند روزانه میلیاردها پست، ریلز و رفتار کاربران را به‌صورت دستی بررسی کند تا ببیند چه چیزی را به چه کسی نشان دهد. اینجاست که داده‌کاوی وارد میدان می‌شود تا از دل این سنگ‌ها و خاک‌های داده، «طلای دانش» را استخراج کند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>مراحل تبدیل داده خام به دانش هوشمند:</b>
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۱) داده‌های ورودی (Massive Input Data):</b><br>
    &nbsp;&nbsp;&nbsp;• <b>داده‌های صریح:</b> لایک‌ها، کامنت‌ها، سیو (Save) و شیر (Share) کردن‌ها.<br>
    &nbsp;&nbsp;&nbsp;• <b>داده‌های پنهان:</b> میزان زمانی که چشم کاربر روی یک پست مکث کرده (Dwell Time)، سرعت رد کردن (Skip) یک ریلز، زمان‌های آنلاینی، و حتی صدای فعال روی ویدیوها.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۲) فرایند داده‌کاوی (Data Mining Process):</b><br>
    الگوریتم داده‌کاوی در حجم عظیم داده‌ها می‌کاود و <b>الگوی پنهانی</b> کشف می‌کند. برای مثال: «۸۵٪ کاربرانی که ویدیوهای آموزشی ۳۰ ثانیه‌ای پایتون را تا انتها می‌بینند و آن را سیو می‌کنند، معمولاً روی ویدیوهای مربوط به امنیت سایبری، ابزارهای لینوکس و فناوری‌های هوش مصنوعی هم مکث بالای ۵ ثانیه دارند.»
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۳) دانش استخراج‌شده و خروجی هوشمند (Knowledge &amp; Value):</b><br>
    سیستم بدون اینکه شما کلمه‌ای سرچ کرده باشید یا پیج خاصی را فالو داشته باشید، الگوی علاقه‌مندی واقعی شما را کشف می‌کند.<br>
    &nbsp;&nbsp;&nbsp;• <b>ارزش برای کسب‌وکار (متا):</b> افزایش زمان حضور کاربر در اپلیکیشن (Engagement) و نمایش تبلیغات هدفمندتر.<br>
    &nbsp;&nbsp;&nbsp;• <b>ارزش برای کاربر:</b> دریافت محتوای کاملاً شخصی‌سازی‌شده و متناسب با نیاز در صفحه Explore.
  </p>

</div>

<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #0369A1; font-size: 15pt; line-height: 2.2; background: #F0F9FF; color: #0C4A6E; box-sizing: border-box;">
  <div style="font-size: 16pt; font-weight: 700; margin-bottom: 8px; color: #0369A1;">💡 نکته مهم</div>
  اینستاگرام بدون داده‌کاوی، فقط یک "انبار عظیم و شلوغ از ویدیو و تصویر" است که کاربر در آن گم می‌شود (Data Rich).<br><br>
  اما با به‌کارگیری تکنیک‌های داده‌کاوی، این توده عظیم داده به "دانش رفتار کاربر" تبدیل می‌شود که می‌تواند حدس بزند شما در ۵ دقیقه آینده دوست دارید چه چیزی ببینید!
</div>

```python
# این قسمت اول کده. کتابخونه‌های لازم برای شبیه‌سازی داده، تحلیل الگو و رسم نمودار رو وارد می‌کنیم.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)  # برای تولید نتایج تکرارپذیر

# ==========================================================
# ۱) شبیه‌سازی داده‌های ورودی عظیم (Massive Input Data)
# ==========================================================
# فرض می‌کنیم اینستاگرام رفتار ۳۰۰۰ کاربر رو روی ۷ دسته محتوایی مختلف ثبت کرده.
# دسته‌ها به دو گروه پنهان تقسیم می‌شن: "فناوری" و "سبک زندگی" — اما این گروه‌بندی
# رو خود سیستم نمی‌دونه؛ فقط باید از دل داده کشفش کنه (دقیقاً مثل دنیای واقعی).
n_users = 3000
categories = ['Python_Tutorial', 'Cybersecurity', 'Linux', 'AI',
              'Cooking', 'Sports', 'Fashion']

# هر کاربر یک گرایش پنهان (Latent Affinity) به "فناوری" و "سبک زندگی" داره.
# این چیزیه که در دنیای واقعی هیچ‌کس مستقیم اعلام نمی‌کنه، ولی در رفتارش نمایان میشه.
tech_affinity = np.random.beta(2, 5, n_users)      # اکثر کاربرا علاقه کم تا متوسط دارن
lifestyle_affinity = np.random.beta(2, 5, n_users)

data = {'UserID': np.arange(1, n_users + 1)}

for cat in categories:
    if cat in ['Python_Tutorial', 'Cybersecurity', 'Linux', 'AI']:
        base_affinity = tech_affinity
    else:
        base_affinity = lifestyle_affinity

    # داده‌های صریح: لایک، کامنت، سیو، شیر (وابسته به علاقه‌مندی پنهان + نویز)
    like = np.random.binomial(1, np.clip(base_affinity + np.random.normal(0, 0.1, n_users), 0, 1))
    save = np.random.binomial(1, np.clip(base_affinity * 0.7 + np.random.normal(0, 0.1, n_users), 0, 1))

    # داده‌های پنهان: مدت‌زمان مکث (Dwell Time) روی هر ویدیو (بر حسب ثانیه از یک ویدیوی ۳۰ ثانیه‌ای)
    dwell_time = np.clip(base_affinity * 30 + np.random.normal(0, 5, n_users), 0, 30)

    # سرعت رد کردن (Skip): هرچقدر علاقه کمتر باشه، سرعت رد کردن بیشتره
    skip = np.clip(1 - base_affinity + np.random.normal(0, 0.1, n_users), 0, 1)

    data[f'{cat}_Like'] = like
    data[f'{cat}_Save'] = save
    data[f'{cat}_DwellTime'] = dwell_time.round(1)
    data[f'{cat}_SkipRate'] = skip.round(2)

behavior_data = pd.DataFrame(data)

print("نمونه‌ای از داده خام رفتاری کاربران (شبیه‌سازی‌شده):")
print(behavior_data.head(5))

# ==========================================================
# ۲) فرایند داده‌کاوی (Data Mining Process)
# ==========================================================
# هدف: کشف اینکه آیا کاربرانی که ویدیوهای Python رو تا انتها دیده و سیو کرده‌اند،
# در دسته‌های دیگه (Cybersecurity, Linux, AI) هم رفتار مشابهی (مکث بالا) دارن یا نه.

python_savers = behavior_data[
    (behavior_data['Python_Tutorial_Save'] == 1) &
    (behavior_data['Python_Tutorial_DwellTime'] >= 25)  # تقریباً "تا انتها دیده"
]

print(f"\nتعداد کاربرانی که ویدیوی پایتون رو کامل دیده و سیو کرده‌اند: {len(python_savers)}")

# حالا بررسی می‌کنیم چند درصد از این کاربرها روی محتوای مرتبط با فناوری هم مکث بالا (>۵ ثانیه) داشته‌اند
related_tech_categories = ['Cybersecurity', 'Linux', 'AI']
threshold_dwell = 5

for cat in related_tech_categories:
    percentage = (python_savers[f'{cat}_DwellTime'] > threshold_dwell).mean() * 100
    print(f"➡️ {percentage:.1f}% از این کاربران روی محتوای «{cat}» هم مکث بالای {threshold_dwell} ثانیه داشته‌اند.")

# همچنین همین بررسی رو برای دسته‌های نامرتبط (سبک زندگی) انجام می‌دیم تا تضاد الگو مشخص بشه
unrelated_categories = ['Cooking', 'Sports', 'Fashion']
print("\nبرای مقایسه، همین بررسی روی دسته‌های نامرتبط:")
for cat in unrelated_categories:
    percentage = (python_savers[f'{cat}_DwellTime'] > threshold_dwell).mean() * 100
    print(f"➡️ {percentage:.1f}% از این کاربران روی محتوای «{cat}» مکث بالای {threshold_dwell} ثانیه داشته‌اند.")

# ==========================================================
# ۳) دانش استخراج‌شده و خروجی هوشمند (Knowledge & Value)
# ==========================================================
# با محاسبه ماتریس همبستگی بین دسته‌ها، الگوی پنهان "خوشه فناوری" در برابر "خوشه سبک زندگی"
# رو به‌صورت بصری کشف می‌کنیم — دقیقاً کاری که سیستم توصیه‌گر اینستاگرام انجام می‌ده.
dwell_columns = [f'{cat}_DwellTime' for cat in categories]
corr_matrix = behavior_data[dwell_columns].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f',
            xticklabels=categories, yticklabels=categories)
plt.title('Hidden pattern discoverd by Data Mining\n(correlation between categories dewll time)')
plt.tight_layout()
plt.show()

# ==========================================================
# ۴) شبیه‌سازی خروجی صفحه Explore برای یک کاربر خاص
# ==========================================================
# فرض کنید می‌خوایم صفحه Explore رو برای یک کاربر که علاقه بالا به فناوری داره بسازیم.
sample_user = behavior_data.iloc[python_savers.index[0]]

interest_scores = {}
for cat in categories:
    # امتیاز علاقه‌مندی ترکیبی از سیو کردن و مدت‌زمان مکث محاسبه می‌شه
    score = (sample_user[f'{cat}_Save'] * 40) + (sample_user[f'{cat}_DwellTime'] / 30 * 60)
    interest_scores[cat] = round(score, 1)

interest_series = pd.Series(interest_scores).sort_values(ascending=False)

print(f"\nصفحه Explore پیشنهادی برای کاربر شماره {int(sample_user['UserID'])} "
      f"بر اساس دانش استخراج‌شده (رتبه‌بندی محتوا):")
print(interest_series)

plt.figure(figsize=(8, 5))
interest_series.plot(kind='bar', color=['red' if c in related_tech_categories + ['Python_Tutorial']
                                         else 'gray' for c in interest_series.index])
plt.title(f'Explore Page (personalized) - user {int(sample_user["UserID"])}')
plt.ylabel('Interest Score')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

در این مثال (شبیه‌سازی اکسپلور و ریلز اینستاگرام):
<ul>
<li>ابتدا کتابخانه‌های مورد نیاز شامل <b>pandas</b>، <b>numpy</b>، <b>matplotlib</b> و <b>seaborn</b> وارد شده‌اند تا امکان ساخت داده‌های رفتاری عظیم، کشف الگو و نمایش بصری فراهم شود.</li>
<li>یک مجموعه داده فرضی از رفتار <b>۳۰۰۰ کاربر</b> روی ۷ دسته محتوایی مختلف (مانند آموزش پایتون، امنیت سایبری، لینوکس، هوش مصنوعی، آشپزی و ورزش) شبیه‌سازی شده است. داده‌ها به دو لایه تقسیم می‌شوند: <b>داده‌های صریح</b> (لایک، سیو، شیر) و <b>داده‌های پنهان</b> (مدت‌زمان مکث چشم روی ویدیو یا Dwell Time و سرعت رد کردن یا Skip Rate).</li>
<li>هر کاربر دارای یک <b>گرایش پنهان</b> (Latent Affinity) به فناوری یا سبک زندگی است که مستقیماً ثبت نشده، اما در رفتارهای ظریف او (مانند مدت‌زمان تماشای ویدیو) خود را نشان می‌دهد — دقیقاً همان چیزی که الگوریتم اینستاگرام بدون پرسش از کاربر کشف می‌کند.</li>
<li>سپس فرآیند <b>داده‌کاوی</b> اجرا می‌شود: کاربرانی که ویدیوهای آموزشی پایتون را تا انتها دیده و سیو کرده‌اند انتخاب می‌شوند و بررسی می‌شود که آیا روی محتوای مرتبط (امنیت سایبری، لینوکس، هوش مصنوعی) نیز مکث بالایی داشته‌اند یا خیر. همچنین با رسم <b>نقشه حرارتی همبستگی</b> (Heatmap)، خوشه‌ی پنهان فناوری در مقابل خوشه‌ی سبک زندگی آشکار می‌شود.</li>
<li>در نهایت، <b>دانش استخراج‌شده</b> به یک خروجی هوشمند تبدیل می‌شود: برای یک کاربر نمونه، امتیاز علاقه‌مندی ترکیبی (بر اساس سیو و زمان مکث) محاسبه شده و یک صفحه‌ی <b>Explore شخصی‌سازی‌شده</b> ساخته می‌شود که محتوا را بر اساس الگوی پنهان رفتار کاربر رتبه‌بندی می‌کند.</li>
</ul>
این مثال نشان می‌دهد که چگونه داده‌کاوی می‌تواند از میان پتابایت‌ها داده‌ی خام و به‌ظاهر تصادفی (که هیچ انسانی قادر به خواندن تک‌تک آن‌ها نیست)، الگوهای رفتاری پنهان را کشف کند و آن‌ها را به <b>دانشی کاربردی</b> برای شخصی‌سازی محتوا و افزایش تعامل کاربر تبدیل کند — دقیقاً همان کاری که سیستم‌های ریلز و اکسپلور متا روزانه برای میلیاردها کاربر انجام می‌دهند.
