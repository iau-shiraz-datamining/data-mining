  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    دیجی‌کالا به‌عنوان بزرگ‌ترین فروشگاه آنلاین ایران، روزانه پتابایت‌ها داده از رفتارهای میلیون‌ها کاربر جمع‌آوری می‌کند. اگر این داده‌ها را کالبدشکافی کنیم، دقیقاً تمام دسته‌بندی‌های متن کتاب را در آن می‌بینیم:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۱. داده‌های ساختاریافته (Structured Data)</b><br><br>
    <b>چیست؟</b> داده‌های جدولی (سطر و ستونی) در پایگاه‌های داده رابطه‌ای که طرح (Schema) کاملاً مشخصی دارند.<br><br>
    <b>در دیجی‌کالا:</b><br>
    &nbsp;&nbsp;&nbsp;• جدول مشخصات فنی کالاها (نام کالا، برند، میزان حافظه، قیمت).<br>
    &nbsp;&nbsp;&nbsp;• جدول پروفایل مشتریان (شناسه کاربر، تاریخ ثبت‌نام، شهر سکونت).<br><br>
    <b>کاربرد داده‌کاوی:</b> محاسبه میانگین قیمت‌ها، رگرسیون برای پیش‌بینی فروش و دسته‌بندی مشتریان بر اساس منطقه جغرافیایی.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۲. داده‌های نیمه‌ساختاریافته، تراکنشی و گراف (Semi-Structured / Transactional / Graph)</b><br><br>
    <b>چیست؟</b> داده‌هایی که جدول سخت‌گیرانه SQL را ندارند، اما برچسب‌ها (Tags) یا روابط ساختاریافته در آن‌ها مشهود است.<br><br>
    <b>در دیجی‌کالا:</b><br>
    &nbsp;&nbsp;&nbsp;• داده تراکنشی (Basket Data): فاکتورهای خرید؛ مثلاً سبد خریدی شامل [کاور گوشی، محافظ صفحه، هندزفری].<br>
    &nbsp;&nbsp;&nbsp;• داده توالی‌ای (Sequential / Clickstream): ترتیبی که کاربر کلیک کرده است؛ [جستجوی لپ‌تاپ ➔ کلیک روی برند ایسوس ➔ مقایسه با لنوو ➔ افزودن به سبد].<br>
    &nbsp;&nbsp;&nbsp;• داده گراف و شبکه (Graph Data): گراف ارتباطی بین «کاربران متصل» یا «کالاهای مشابه» (کدام کاربران الگوی خرید یکسانی دارند).<br><br>
    <b>کاربرد داده‌کاوی:</b> اجرای قواعد وابستگی (Market Basket Analysis) برای سیستم پیشنهاددهنده (مثلاً: "خریداران این کالا، این موارد را هم خریده‌اند").
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۳. داده‌های بدون ساختار (Unstructured Data)</b><br><br>
    <b>چیست؟</b> متون آزاد، تصاویر و ویدیوهایی که ساختار سطری/ستونی ندارند اما حاوی دانشی ارزشمند هستند.<br><br>
    <b>در دیجی‌کالا:</b><br>
    &nbsp;&nbsp;&nbsp;• متن: هزاران کامنت و نقد و بررسی متنی کاربران ذیل هر محصول.<br>
    &nbsp;&nbsp;&nbsp;• تصویر و ویدیو: عکس‌های خریداران از کالای تحویل‌گرفته‌شده یا ویدیوهای آنباکسینگ (Unboxing).<br>
    &nbsp;&nbsp;&nbsp;• صدا: فایل‌های ضبط‌شده تماس‌های مشتریان با مرکز پشتیبانی.<br><br>
    <b>کاربرد داده‌کاوی:</b> به‌کارگیری پردازش زبان طبیعی (NLP) برای تحلیل حس خریداران (Sentiment Analysis) جهت شناسایی کالاهای باکیفیت یا مرجوعی، و استفاده از بینایی کامپیوتر (Computer Vision) برای حذف عکس‌های غیراخلاقی یا نامرتبط.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۴. داده‌های مکانی-زمانی (Spatio-Temporal Data)</b><br><br>
    <b>چیست؟</b> داده‌هایی که مختصات جغرافیایی (Location) و زمان (Time) در آن‌ها نقش کلیدی دارد.<br><br>
    <b>در دیجی‌کالا:</b> لوکیشن تحویل سفارش‌ها + زمان ثبت خرید (مثلاً خریدهای شبانه در مناطق خاص شهری).<br><br>
    <b>کاربرد داده‌کاوی:</b> داده‌کاوی مکانی برای بهینه‌سازی مسیر ناوگان توزیع (دیجی‌اکسپرس) و جانمایی هوشمند کالاها در انبارهای منطقه‌ای.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>۵. داده‌های جریانی (Stream Data)</b><br><br>
    <b>چیست؟</b> جریان نامحدود و مداومی از داده‌ها که به‌صورت آنی (Real-Time) وارد سیستم می‌شوند.<br><br>
    <b>در دیجی‌کالا:</b> جریان چندصدهزار کلیک هم‌زمان در حراجی‌های بزرگ (مثل بلک فرایدی یا حراج شگفت‌انگیز).<br><br>
    <b>کاربرد داده‌کاوی:</b> کشف زنده ناهنجاری (Anomaly Detection) جهت مسدود کردن ربات‌های خریدار یا جلوگیری از از دسترس خارج شدن سرورها.
  </p>

</div>

```python
# ==========================================================
#کتابخانه‌های مورد نیاز 
# ==========================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import networkx as nx
from datetime import datetime, timedelta
from collections import Counter
from itertools import combinations

np.random.seed(42)

print("="*70)
print("۱) داده‌های ساختاریافته (Structured Data)")
print("="*70)

# ==========================================================
# ۱) داده‌های ساختاریافته: جدول کالاها و جدول مشتریان
# ==========================================================
n_products = 200
brands = ['Samsung', 'Apple', 'Xiaomi', 'Asus', 'Lenovo']

products = pd.DataFrame({
    'ProductID': range(1, n_products + 1),
    'Brand': np.random.choice(brands, n_products),
    'RAM_GB': np.random.choice([4, 8, 16, 32], n_products),
    'Price': np.random.randint(3_000_000, 60_000_000, n_products)
})

cities = ['Tehran', 'Mashhad', 'Isfahan', 'Shiraz', 'Tabriz']
n_customers = 1000
customers = pd.DataFrame({
    'CustomerID': range(1, n_customers + 1),
    'City': np.random.choice(cities, n_customers, p=[0.4, 0.15, 0.15, 0.15, 0.15]),
    'JoinDate': [datetime(2022,1,1) + timedelta(days=int(x)) for x in np.random.randint(0, 900, n_customers)]
})

print("\nنمونه جدول محصولات:")
print(products.head(3))
print("\nنمونه جدول مشتریان:")
print(customers.head(3))

# --- داده‌کاوی: میانگین قیمت به تفکیک برند + رگرسیون ساده قیمت بر اساس رم ---
avg_price_by_brand = products.groupby('Brand')['Price'].mean().sort_values(ascending=False)
print("\n📊 دانش استخراج‌شده - میانگین قیمت به تفکیک برند:")
print(avg_price_by_brand.round(0))

# رگرسیون خطی ساده (بدون sklearn) برای رابطه رم و قیمت
slope, intercept = np.polyfit(products['RAM_GB'], products['Price'], 1)
print(f"\n📈 مدل رگرسیون کشف‌شده: Price ≈ {slope:,.0f} * RAM_GB + {intercept:,.0f}")

# دسته‌بندی مشتریان بر اساس شهر (تعداد مشتری در هر شهر)
customer_segments = customers['City'].value_counts()
print("\n👥 دانش استخراج‌شده - دسته‌بندی مشتریان بر اساس شهر:")
print(customer_segments)


print("\n" + "="*70)
print("۲) داده‌های نیمه‌ساختاریافته / تراکنشی / گراف")
print("="*70)

# ==========================================================
# ۲-الف) داده تراکنشی (Basket Data) → Market Basket Analysis
# ==========================================================
item_pool = ['CoverPhone', 'ScreenProtector', 'Handsfree', 'PowerBank',
             'Laptop', 'Mouse', 'LaptopBag', 'USB_Hub']

n_baskets = 500
baskets = []
for _ in range(n_baskets):
    if np.random.rand() < 0.5:  # الگوی پرتکرار: خریداران گوشی، لوازم جانبی هم می‌خرن
        basket = np.random.choice(['CoverPhone', 'ScreenProtector', 'Handsfree', 'PowerBank'],
                                   size=np.random.randint(2, 4), replace=False).tolist()
    else:  # الگوی دیگر: خریداران لپ‌تاپ
        basket = np.random.choice(['Laptop', 'Mouse', 'LaptopBag', 'USB_Hub'],
                                   size=np.random.randint(2, 4), replace=False).tolist()
    baskets.append(basket)

print(f"\nنمونه ۳ سبد خرید از {n_baskets} سبد شبیه‌سازی‌شده:")
for b in baskets[:3]:
    print(b)

# --- داده‌کاوی ساده Market Basket: شمارش هم‌رخدادی جفت‌کالاها ---
pair_counts = Counter()
for basket in baskets:
    for pair in combinations(sorted(basket), 2):
        pair_counts[pair] += 1

top_pairs = pair_counts.most_common(5)
print("\n🛒 دانش استخراج‌شده - پرتکرارترین جفت‌کالاهای خریداری‌شده با هم (قانون وابستگی):")
for pair, count in top_pairs:
    support = count / n_baskets
    print(f"  {pair[0]} + {pair[1]}  →  support={support:.2f}  (تعداد={count})")

# ==========================================================
# ۲-ب) داده توالی‌ای (Clickstream)
# ==========================================================
click_events = ['Search_Laptop', 'Click_Asus', 'Click_Lenovo', 'Compare',
                 'AddToCart', 'ViewReviews', 'Checkout']

n_sessions = 300
sessions = []
for _ in range(n_sessions):
    length = np.random.randint(3, 6)
    # الگوی رایج: جستجو -> کلیک برند -> مقایسه -> افزودن به سبد
    if np.random.rand() < 0.6:
        seq = ['Search_Laptop', np.random.choice(['Click_Asus', 'Click_Lenovo']),
               'Compare', 'AddToCart']
    else:
        seq = list(np.random.choice(click_events, size=length, replace=True))
    sessions.append(seq)

print(f"\nنمونه ۳ توالی کلیک از {n_sessions} سشن کاربر:")
for s in sessions[:3]:
    print(" ➔ ".join(s))

# --- داده‌کاوی: کشف رایج‌ترین الگوی توالی ---
sequence_counts = Counter([tuple(s) for s in sessions])
most_common_sequence = sequence_counts.most_common(1)[0]
print(f"\n🔎 دانش استخراج‌شده - رایج‌ترین مسیر کلیک کاربران:")
print(f"  {' ➔ '.join(most_common_sequence[0])}  (تکرار در {most_common_sequence[1]} سشن)")

# ==========================================================
# ۲-ج) داده گراف (Graph Data) → کاربران با الگوی خرید مشابه
# ==========================================================
n_graph_users = 15
G = nx.Graph()
G.add_nodes_from(range(1, n_graph_users + 1))

# فرض می‌کنیم کاربرانی که سبد خرید مشابه دارن، به هم متصل می‌شن
for i in range(1, n_graph_users + 1):
    for j in range(i + 1, n_graph_users + 1):
        if np.random.rand() < 0.15:  # احتمال شباهت الگوی خرید
            G.add_edge(i, j)

# --- داده‌کاوی گرافی: پیدا کردن پرارتباط‌ترین کاربر (Centrality) ---
centrality = nx.degree_centrality(G)
top_user = max(centrality, key=centrality.get)
print(f"\n🕸️ دانش استخراج‌شده - کاربر شماره {top_user} بیشترین شباهت خرید را با دیگر کاربران دارد "
      f"(Degree Centrality = {centrality[top_user]:.2f})")

plt.figure(figsize=(6, 6))
nx.draw(G, with_labels=True, node_color='lightblue', node_size=600,
        font_weight='bold', edge_color='gray')
plt.title("Graph of similar-purchase-pattern customers")
plt.show()


print("\n" + "="*70)
print("۳) داده‌های بدون ساختار (Unstructured Data)")
print("="*70)

# ==========================================================
# ۳) متن: تحلیل احساسات ساده کامنت‌های کاربران
# ==========================================================
comments = [
    "کیفیت این محصول عالی بود و خیلی راضی هستم",
    "بسته‌بندی خوب بود اما ارسال دیر شد",
    "اصلا راضی نبودم، محصول معیوب بود",
    "کیفیت ساخت خوب است ولی قیمت بالاست",
    "عالی بود، پیشنهاد میکنم حتما بخرید",
    "بد بود و خراب به دستم رسید، ناراضی هستم",
    "خیلی خوب و باکیفیت، ممنون از دیجی‌کالا",
    "متوسط بود، نه خیلی خوب نه خیلی بد",
]

positive_words = ['عالی', 'خوب', 'راضی', 'باکیفیت', 'پیشنهاد']
negative_words = ['بد', 'ناراضی', 'معیوب', 'خراب', 'دیر']

def simple_sentiment(text):
    score = 0
    for w in positive_words:
        if w in text:
            score += 1
    for w in negative_words:
        if w in text:
            score -= 1
    return 'Positive' if score > 0 else ('Negative' if score < 0 else 'Neutral')

reviews_df = pd.DataFrame({'Comment': comments})
reviews_df['Sentiment'] = reviews_df['Comment'].apply(simple_sentiment)

print("\n💬 نتیجه تحلیل احساسات (Sentiment Analysis) روی نظرات کاربران:")
print(reviews_df)

sentiment_summary = reviews_df['Sentiment'].value_counts()
print("\n📊 دانش استخراج‌شده - خلاصه احساسات کاربران نسبت به کالا:")
print(sentiment_summary)

# شبیه‌سازی ساده تصویر/ویدیو: فقط متادیتای تعداد فایل‌های آپلودی و برچسب کیفیت
media_data = pd.DataFrame({
    'MediaType': ['Image']*40 + ['Video']*10,
    'IsInappropriate': np.random.choice([0, 1], 50, p=[0.92, 0.08])
})
flagged = media_data['IsInappropriate'].sum()
print(f"\n🖼️ داده‌کاوی روی تصاویر/ویدیوها (شبیه‌سازی Computer Vision): "
      f"{flagged} فایل از ۵۰ فایل به‌عنوان نامناسب/نامرتبط شناسایی و حذف شدند.")


print("\n" + "="*70)
print("۴) داده‌های مکانی-زمانی (Spatio-Temporal Data)")
print("="*70)

# ==========================================================
# ۴) داده مکانی-زمانی: زمان و شهر ثبت سفارش‌ها
# ==========================================================
n_orders = 2000
orders = pd.DataFrame({
    'OrderID': range(1, n_orders + 1),
    'City': np.random.choice(cities, n_orders, p=[0.4, 0.15, 0.15, 0.15, 0.15]),
    'Hour': np.random.choice(range(24), n_orders,
                              p=np.array([1,1,1,1,1,1,2,3,4,5,5,6,6,6,5,5,6,7,8,9,10,8,4,2]) / 100 * (100/ sum([1,1,1,1,1,1,2,3,4,5,5,6,6,6,5,5,6,7,8,9,10,8,4,2])))
})

# --- داده‌کاوی: کشف الگوی خرید شبانه در هر شهر ---
night_orders = orders[(orders['Hour'] >= 23) | (orders['Hour'] <= 4)]
night_pattern = night_orders['City'].value_counts()
print("\n🌙 دانش استخراج‌شده - تعداد خریدهای شبانه (۲۳ تا ۴ صبح) به تفکیک شهر:")
print(night_pattern)

pivot = orders.pivot_table(index='City', columns='Hour', values='OrderID', aggfunc='count', fill_value=0)
plt.figure(figsize=(12, 5))
sns.heatmap(pivot, cmap='YlOrRd')
plt.title("Spatio-Temporal pattern of orders (city vs hour)")
plt.xlabel("Hour of day")
plt.ylabel("City")
plt.tight_layout()
plt.show()


print("\n" + "="*70)
print("۵) داده‌های جریانی (Stream Data)")
print("="*70)

# ==========================================================
# ۵) داده جریانی: شبیه‌سازی کلیک‌های آنی در حراج و تشخیص ناهنجاری (ربات)
# ==========================================================
n_seconds = 200
normal_clicks = np.random.poisson(lam=50, size=n_seconds)  # ترافیک عادی کاربران

# تزریق یک حمله ربات (افزایش ناگهانی کلیک در چند ثانیه خاص)
bot_attack_seconds = [80, 81, 82, 120, 121]
for s in bot_attack_seconds:
    normal_clicks[s] += np.random.randint(300, 500)

stream_df = pd.DataFrame({'Second': range(n_seconds), 'Clicks': normal_clicks})

# --- داده‌کاوی جریانی: تشخیص ناهنجاری با روش Z-Score متحرک ---
window = 10
stream_df['RollingMean'] = stream_df['Clicks'].rolling(window).mean()
stream_df['RollingStd'] = stream_df['Clicks'].rolling(window).std()
stream_df['ZScore'] = (stream_df['Clicks'] - stream_df['RollingMean']) / stream_df['RollingStd']

anomalies = stream_df[stream_df['ZScore'] > 3]
print("\n🚨 دانش استخراج‌شده - ثانیه‌هایی که رفتار غیرعادی (احتمال حمله ربات) شناسایی شد:")
print(anomalies[['Second', 'Clicks', 'ZScore']].round(2))

plt.figure(figsize=(12, 5))
plt.plot(stream_df['Second'], stream_df['Clicks'], label='Real-time clicks', color='steelblue')
plt.scatter(anomalies['Second'], anomalies['Clicks'], color='red', s=80, label='Detected anomaly (bot)', zorder=5)
plt.xlabel('Second')
plt.ylabel('Number of clicks')
plt.title('Real-time Stream Data - Anomaly Detection during flash sale')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال:
<ul>
<li>ابتدا کتابخانه‌های مورد نیاز شامل <b>pandas</b> و <b>numpy</b> برای مدیریت و شبیه‌سازی داده، <b>matplotlib</b> و <b>seaborn</b> برای نمایش بصری، و <b>networkx</b> برای ساخت و تحلیل گراف وارد شده‌اند.</li>
<li><b>داده‌های ساختاریافته</b> شامل دو جدول <b>محصولات</b> (برند، رم، قیمت) و <b>مشتریان</b> (شهر، تاریخ عضویت) شبیه‌سازی شده است. سپس با استفاده از <code>groupby</code> میانگین قیمت به تفکیک برند محاسبه و با یک <b>مدل رگرسیون خطی ساده</b> رابطه بین حجم رم و قیمت کالا کشف شده است.</li>
<li>برای <b>داده‌های نیمه‌ساختاریافته</b>، سه نوع داده جداگانه شبیه‌سازی شده‌اند:
  <ul style="margin-top:10px;">
    <li><b>داده تراکنشی (سبد خرید):</b> با شمارش هم‌رخدادی کالاها در سبدهای مختلف، پرتکرارترین جفت‌کالاهای خریداری‌شده با هم کشف شده‌اند — پایه‌ای ساده برای الگوریتم <b>Market Basket Analysis</b>.</li>
    <li><b>داده توالی‌ای (Clickstream):</b> مسیر کلیک کاربران قبل از خرید (جستجو ➔ کلیک برند ➔ مقایسه ➔ افزودن به سبد) شبیه‌سازی و رایج‌ترین الگوی توالی استخراج شده است.</li>
    <li><b>داده گراف:</b> با کتابخانه <code>networkx</code> شبکه‌ای از کاربران با الگوی خرید مشابه ساخته شده و با معیار <b>Degree Centrality</b>، پرارتباط‌ترین کاربر شناسایی شده است.</li>
  </ul>
</li>
<li>برای <b>داده‌های بدون ساختار</b>، نظرات متنی کاربران با یک روش ساده مبتنی بر واژگان مثبت و منفی تحلیل شده‌اند (<b>Sentiment Analysis</b>) و همچنین متادیتای تصاویر/ویدیوهای آپلودی برای شبیه‌سازی فیلتر محتوای نامناسب با <b>بینایی کامپیوتر</b> بررسی شده است.</li>
<li>برای <b>داده‌های مکانی-زمانی</b>، ساعت و شهر ثبت سفارش‌ها شبیه‌سازی شده و با یک نمودار <b>Heatmap</b>، الگوی خریدهای شبانه در شهرهای مختلف به‌صورت بصری کشف شده است.</li>
<li>در نهایت، برای <b>داده‌های جریانی</b>، جریان کلیک‌های آنی کاربران در یک حراج زنده شبیه‌سازی شده و با روش <b>Z-Score متحرک</b>، ثانیه‌هایی که رفتار غیرعادی (مثل حمله ربات) رخ داده، به‌صورت آنی شناسایی و روی نمودار مشخص شده‌اند.</li>
</ul>
این مثال به‌خوبی نشان می‌دهد که چگونه یک پلتفرم بزرگ مانند دیجی‌کالا با ترکیب چندین نوع داده متفاوت (ساختاریافته، نیمه‌ساختاریافته، بدون ساختار، مکانی-زمانی و جریانی) و به‌کارگیری تکنیک‌های ساده داده‌کاوی، می‌تواند از دل داده‌های خام و پراکنده، <b>دانشی ارزشمند</b> برای بهبود سیستم پیشنهاددهنده، تحلیل رضایت مشتریان، بهینه‌سازی توزیع و حفظ امنیت سیستم در لحظات حساس (مانند حراجی‌های بزرگ) استخراج کند.
</p>
