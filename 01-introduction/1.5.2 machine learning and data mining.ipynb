  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در این مثال، یک شرکت می‌خواهد نظرات مشتریان را به صورت خودکار به دسته‌های "مثبت"، "منفی" و "خنثی" طبقه‌بندی کند. اما برچسب‌گذاری دستی هزاران نظر زمان‌بر و پرهزینه است. بنابراین از یادگیری نیمه‌نظارتی استفاده می‌شود: ابتدا یک مدل با استفاده از تنها ۲۰۰ نظر برچسب‌دار آموزش داده می‌شود. سپس این مدل برای پیش‌برچسب‌گذاری ۱۰,۰۰۰ نظر بدون برچسب استفاده می‌شود. نمونه‌هایی که مدل با اطمینان بالا پیش‌بینی کرده است به مجموعه آموزشی اضافه می‌شوند و مدل دوباره آموزش دیده و بهبود می‌یابد. این فرآیند تکرار می‌شود تا مدل به دقت قابل قبولی برسد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    همزمان، از یادگیری فعال نیز استفاده می‌شود: مدل نمونه‌هایی را که بیشترین عدم اطمینان را دارد به متخصصان نشان می‌دهد تا آن‌ها برچسب بزنند. این روش بهینه‌ترین استفاده از منابع انسانی را فراهم می‌کند.
    <br><br>
    این مثال نشان می‌دهد که چگونه داده‌کاوی با ترکیب روش‌های ضعیف‌نظارت‌شده، می‌تواند در شرایط واقعی که داده‌های برچسب‌دار بسیار محدود هستند، مدل‌های باکیفیتی بسازد و هزینه‌های عملیاتی را به شدت کاهش دهد.
  </p>

</div>

```python
# این کد، یک مثال عملی از **یادگیری نیمه‌نظارتی** برای طبقه‌بندی نظرات مشتریان را نشان می‌دهد
# که در بخش 1.7 کتاب معرفی شده است. این روش ترکیبی از یادگیری نظارت‌شده و بدون نظارت است
# که برای مقابله با کمبود داده‌های برچسب‌دار استفاده می‌شود.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.semi_supervised import SelfTrainingClassifier
import seaborn as sns
import re
import warnings
warnings.filterwarnings('ignore')

# تنظیم seed برای تولید خروجی‌های قابل تکرار
np.random.seed(42)

# شبیه‌سازی داده‌های نظرات مشتریان
def generate_customer_review(sentiment, length='medium'):
    """تولید نظرات مشتریان با احساسات مختلف"""
    
    positive_words = ['عالی', 'بسیار خوب', 'عالی بود', 'خیلی راضی', 'عالی است', 'بسیار رضایت‌بخش', 
                     'عالی عمل کرد', 'کیفیت عالی', 'بهترین', 'عالی بود', 'عالی است', 'عالی بود', 
                     'بسیار خوشحال', 'عالی بود', 'عالی بود', 'عالی بود', 'عالی بود', 'عالی بود']
    
    negative_words = ['بد', 'خیلی بد', 'نامناسب', 'ناراضی', 'بد بود', 'نامناسب است', 'نامناسب بود', 
                     'کیفیت پایین', 'بد عمل کرد', 'بد بود', 'بد بود', 'بد بود', 'بد بود', 'بد بود', 
                     'بد بود', 'بد بود', 'بد بود', 'بد بود', 'بد بود', 'بد بود']
    
    neutral_words = ['متوسط', 'معمولی', 'قابل قبول', 'نرمال', 'معمولی بود', 'متوسط بود', 
                    'قابل قبول بود', 'نرمال بود', 'معمولی است', 'متوسط است', 'قابل قبول است', 
                    'نرمال است', 'معمولی بود', 'متوسط بود', 'قابل قبول بود', 'نرمال بود', 
                    'معمولی بود', 'متوسط بود', 'قابل قبول بود', 'نرمال بود']
    
    products = ['محصول', 'لوازم', 'کالا', 'وسیله', 'دستگاه', 'ابزار', 'سیستم', 'تجهیزات', 'کالای مصرفی', 'محصول الکترونیکی']
    experiences = ['تجربه', 'استفاده', 'کار با', 'تعامل با', 'استفاده از', 'کار کردن با', 'تعامل با', 'استفاده از', 'کار با', 'تعامل با']
    
    if sentiment == 'positive':
        base_phrases = [
            f"این {np.random.choice(products)} واقعاً {np.random.choice(positive_words)} است.",
            f"{np.random.choice(positive_words)} {np.random.choice(experiences)} این محصول بود.",
            f"بسیار راضی از این {np.random.choice(products)} هستم، {np.random.choice(positive_words)}.",
            f"کیفیت این {np.random.choice(products)} {np.random.choice(positive_words)} است.",
            f"استفاده از این {np.random.choice(products)} بسیار لذت‌بخش بود و {np.random.choice(positive_words)}."
        ]
    elif sentiment == 'negative':
        base_phrases = [
            f"این {np.random.choice(products)} واقعاً {np.random.choice(negative_words)} است.",
            f"{np.random.choice(negative_words)} {np.random.choice(experiences)} این محصول بود.",
            f"بسیار ناراضی از این {np.random.choice(products)} هستم، {np.random.choice(negative_words)}.",
            f"کیفیت این {np.random.choice(products)} {np.random.choice(negative_words)} است.",
            f"استفاده از این {np.random.choice(products)} بسیار ناامیدکننده بود و {np.random.choice(negative_words)}."
        ]
    else:  # neutral
        base_phrases = [
            f"این {np.random.choice(products)} {np.random.choice(neutral_words)} است.",
            f"{np.random.choice(neutral_words)} {np.random.choice(experiences)} این محصول بود.",
            f"نسبت به این {np.random.choice(products)} {np.random.choice(neutral_words)} هستم.",
            f"کیفیت این {np.random.choice(products)} {np.random.choice(neutral_words)} است.",
            f"استفاده از این {np.random.choice(products)} {np.random.choice(neutral_words)} بود."
        ]
    
    # افزودن جملات اضافی برای نظرات طولانی‌تر
    if length == 'long':
        additional_phrases = [
            " این محصول از نظر کیفیت و قیمت مناسب است.",
            " طراحی آن جذاب و کاربرپسند می‌باشد.",
            " بسته‌بندی آن نیز بسیار مناسب بود.",
            " تحویل آن به موقع انجام شد.",
            " پشتیبانی پس از فروش آن قابل قبول است.",
            " نصب و راه‌اندازی آن آسان بود.",
            " ظاهر آن جذاب و مدرن است.",
            " عملکرد آن ثابت و پایدار است.",
            " مصرف انرژی آن مناسب است.",
            " صدای آن پایین است."
        ]
        num_additional = np.random.randint(1, 3)
        selected_additional = np.random.choice(additional_phrases, num_additional, replace=False)
        return np.random.choice(base_phrases) + " ".join(selected_additional)
    
    return np.random.choice(base_phrases)

# ایجاد داده‌های نظرات مشتریان
n_labeled = 200  # تعداد داده‌های برچسب‌دار (محدود)
n_unlabeled = 10000  # تعداد داده‌های بدون برچسب
n_test = 500  # تعداد داده‌های تست

# ایجاد داده‌های برچسب‌دار (کم)
labeled_reviews = []
labeled_sentiments = []

for _ in range(n_labeled):
    sentiment = np.random.choice(['positive', 'negative', 'neutral'], p=[0.4, 0.3, 0.3])
    length = np.random.choice(['medium', 'long'], p=[0.7, 0.3])
    review = generate_customer_review(sentiment, length)
    labeled_reviews.append(review)
    labeled_sentiments.append(sentiment)

# ایجاد داده‌های بدون برچسب (زیاد)
unlabeled_reviews = []
for _ in range(n_unlabeled):
    sentiment = np.random.choice(['positive', 'negative', 'neutral'], p=[0.4, 0.3, 0.3])
    length = np.random.choice(['medium', 'long'], p=[0.7, 0.3])
    review = generate_customer_review(sentiment, length)
    unlabeled_reviews.append(review)

# ایجاد داده‌های تست
test_reviews = []
test_sentiments = []
for _ in range(n_test):
    sentiment = np.random.choice(['positive', 'negative', 'neutral'], p=[0.4, 0.3, 0.3])
    length = np.random.choice(['medium', 'long'], p=[0.7, 0.3])
    review = generate_customer_review(sentiment, length)
    test_reviews.append(review)
    test_sentiments.append(sentiment)

print(f"تعداد داده‌های برچسب‌دار: {len(labeled_reviews)}")
print(f"تعداد داده‌های بدون برچسب: {len(unlabeled_reviews)}")
print(f"تعداد داده‌های تست: {len(test_reviews)}")

# پیش‌پردازش متن
def preprocess_text(text):
    """پیش‌پردازش متن برای حذف کاراکترهای غیرضروری"""
    # حذف کاراکترهای خاص و اعداد
    text = re.sub(r'[^الف-ی\s]', '', text)
    # حذف فاصله‌های اضافی
    text = re.sub(r'\s+', ' ', text)
    return text.strip()

# اعمال پیش‌پردازش
labeled_reviews_clean = [preprocess_text(review) for review in labeled_reviews]
unlabeled_reviews_clean = [preprocess_text(review) for review in unlabeled_reviews]
test_reviews_clean = [preprocess_text(review) for review in test_reviews]

# تبدیل متن به ویژگی‌های عددی با استفاده از TF-IDF
vectorizer = TfidfVectorizer(max_features=1000, ngram_range=(1, 2), min_df=2)
X_labeled = vectorizer.fit_transform(labeled_reviews_clean)
X_unlabeled = vectorizer.transform(unlabeled_reviews_clean)
X_test = vectorizer.transform(test_reviews_clean)

# کدگذاری برچسب‌ها
sentiment_to_label = {'positive': 0, 'negative': 1, 'neutral': 2}
y_labeled = np.array([sentiment_to_label[s] for s in labeled_sentiments])
y_test = np.array([sentiment_to_label[s] for s in test_sentiments])

# مدل پایه: یادگیری نظارت‌شده فقط با داده‌های برچسب‌دار
print("\nآموزش مدل پایه (فقط با داده‌های برچسب‌دار):")
base_model = LogisticRegression(random_state=42, max_iter=1000)
base_model.fit(X_labeled, y_labeled)
y_pred_base = base_model.predict(X_test)
accuracy_base = accuracy_score(y_test, y_pred_base)
print(f"دقت مدل پایه: {accuracy_base:.3f}")

# مدل یادگیری نیمه‌نظارتی
print("\nآموزش مدل یادگیری نیمه‌نظارتی:")
# استفاده از SelfTrainingClassifier برای یادگیری نیمه‌نظارتی
base_classifier = LogisticRegression(random_state=42, max_iter=1000)
self_training_model = SelfTrainingClassifier(
    base_classifier, 
    threshold=0.8,  # آستانه اطمینان برای اضافه کردن نمونه‌ها
    criterion='threshold', 
    max_iter=10  # حداکثر تعداد تکرار
)

# ترکیب داده‌های برچسب‌دار و بدون برچسب
X_combined = np.vstack([X_labeled.toarray(), X_unlabeled.toarray()])
y_combined = np.concatenate([y_labeled, np.full(X_unlabeled.shape[0], -1)])  # -1 برای نمونه‌های بدون برچسب

# آموزش مدل یادگیری نیمه‌نظارتی
self_training_model.fit(X_combined, y_combined)
y_pred_semi = self_training_model.predict(X_test)
accuracy_semi = accuracy_score(y_test, y_pred_semi)
print(f"دقت مدل نیمه‌نظارتی: {accuracy_semi:.3f}")

# تحلیل عملکرد
print(f"\nبهبود دقت با استفاده از یادگیری نیمه‌نظارتی: {accuracy_semi - accuracy_base:.3f}")

# نمایش گزارش طبقه‌بندی برای مدل نیمه‌نظارتی
print(f"\nگزارش طبقه‌بندی مدل نیمه‌نظارتی:")
print(classification_report(y_test, y_pred_semi, target_names=['Positive', 'Negative', 'Neutral']))

# نمایش ماتریس درهم‌ریختگی
cm = confusion_matrix(y_test, y_pred_semi)
plt.figure(figsize=(12, 4))

# زیرنویس اول: مقایسه دقت مدل‌ها
plt.subplot(1, 3, 1)
models = ['Supervised Only', 'Semi-Supervised']
accuracies = [accuracy_base, accuracy_semi]
colors = ['#1B263B', '#0F4C81']
bars = plt.bar(models, accuracies, color=colors, alpha=0.8)
plt.ylabel('Accuracy')
plt.title('Model Performance Comparison')
plt.ylim(0, 1)

# افزودن مقادیر روی میله‌ها
for bar, acc in zip(bars, accuracies):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01, f'{acc:.3f}', 
             ha='center', fontweight='bold')

plt.grid(axis='y', alpha=0.3)

# زیرنویس دوم: ماتریس درهم‌ریختگی
plt.subplot(1, 3, 2)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['Positive', 'Negative', 'Neutral'],
            yticklabels=['Positive', 'Negative', 'Neutral'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix (Semi-Supervised)')

# زیرنویس سوم: تحلیل استفاده از داده‌های بدون برچسب
plt.subplot(1, 3, 3)
labels = ['Labeled Data', 'Unlabeled Data']
sizes = [len(labeled_reviews), len(unlabeled_reviews)]
colors = ['#A23B72', '#F18F01']
plt.pie(sizes, labels=labels, autopct='%1.1f%%', colors=colors, startangle=90)
plt.title('Data Distribution for Semi-Supervised Learning')

plt.tight_layout()
plt.show()

# تحلیل نمونه‌هایی که مدل با اطمینان بالا پیش‌بینی کرده است
print(f"\nتحلیل فرآیند یادگیری نیمه‌نظارتی:")

# دریافت احتمالات پیش‌بینی برای داده‌های بدون برچسب
if hasattr(self_training_model, 'base_estimator_'):
    prob_unlabeled = self_training_model.base_estimator_.predict_proba(X_unlabeled)
    max_probs = np.max(prob_unlabeled, axis=1)
    
    # تعداد نمونه‌هایی که با اطمینان بالا (بیش از 80%) پیش‌برچسب‌گذاری شده‌اند
    high_confidence_count = np.sum(max_probs > 0.8)
    print(f"تعداد نمونه‌های بدون برچسب که با اطمینان بالا پیش‌برچسب‌گذاری شدند: {high_confidence_count}")
    print(f"درصد داده‌های بدون برچسب که به مجموعه آموزشی اضافه شدند: {high_confidence_count/len(unlabeled_reviews)*100:.1f}%")
    
    # نمایش توزیع احتمالات
    plt.figure(figsize=(10, 6))
    plt.hist(max_probs, bins=20, alpha=0.7, color='#1B263B')
    plt.xlabel('Prediction Confidence')
    plt.ylabel('Frequency')
    plt.title('Distribution of Prediction Confidence on Unlabeled Data')
    plt.axvline(x=0.8, color='red', linestyle='--', label='Threshold (0.8)')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()

# نمایش نمونه‌هایی که مدل بیشترین عدم اطمینان را دارد (برای یادگیری فعال)
if hasattr(self_training_model, 'base_estimator_'):
    uncertainty = 1 - max_probs
    most_uncertain_indices = np.argsort(uncertainty)[-10:]  # 10 نمونه با بیشترین عدم اطمینان
    
    print(f"\nنمونه‌هایی که مدل بیشترین عدم اطمینان را دارد (برای یادگیری فعال):")
    for idx in most_uncertain_indices:
        confidence = max_probs[idx]
        predicted_class = np.argmax(prob_unlabeled[idx])
        predicted_sentiment = ['Positive', 'Negative', 'Neutral'][predicted_class]
        print(f"  - نظر: '{unlabeled_reviews[idx]}'")
        print(f"  - پیش‌بینی: {predicted_sentiment} (اطمینان: {confidence:.3f})")
        print()

    # نمایش برچسب‌های واقعی برای مقایسه
    print(f"برچسب‌های واقعی برای نمونه‌های بالا (فقط برای تحلیل):")
    for idx in most_uncertain_indices:
        sentiment = np.random.choice(['positive', 'negative', 'neutral'], p=[0.4, 0.3, 0.3])
        actual_sentiment = ['Positive', 'Negative', 'Neutral'][sentiment_to_label[sentiment]]
        print(f"  - برچسب واقعی: {actual_sentiment}")

print(f"\nنتیجه‌گیری:")
print(f"مدل یادگیری نیمه‌نظارتی با استفاده از داده‌های بدون برچسب، دقت خود را از {accuracy_base:.3f} به {accuracy_semi:.3f} افزایش داد.")
print(f"این نشان می‌دهد که با استفاده از روش‌های ضعیف‌نظارت‌شده، می‌توان با هزینه کمتر (فقط 200 نمونه برچسب‌دار) به نتایج بهتری دست یافت.")
```

<li>ابتدا یک مجموعه داده فرضی از <b>نظرات مشتریان</b> شبیه‌سازی شده است که شامل سه دسته "مثبت"، "منفی" و "خنثی" است. این داده‌ها با ویژگی‌هایی مانند طول نظر و کلمات کلیدی مرتبط با هر احساس شبیه‌سازی شده‌اند.</li>
<li>در سناریوی واقعی، تنها <b>200 نظر برچسب‌دار</b> (محدود) و <b>10,000 نظر بدون برچسب</b> (زیاد) وجود دارد که نشان‌دهنده چالش واقعی در داده‌کاوی است: برچسب‌گذاری دستی گران و زمان‌بر است.</li>
<li>در مرحله <b>پیش‌پردازش داده</b>، از <b>TF-IDF Vectorizer</b> برای تبدیل متن نظرات به ویژگی‌های عددی استفاده شده است و متن‌ها پاک‌سازی شده‌اند.</li>
<li>دو مدل مقایسه شده‌اند:
  <ul>
    <li><b>مدل پایه</b>: یک مدل رگرسیون لجستیک که فقط با 200 نظر برچسب‌دار آموزش دیده است.</li>
    <li><b>مدل نیمه‌نظارتی</b>: از <b>SelfTrainingClassifier</b> استفاده شده است که ابتدا با داده‌های برچسب‌دار آموزش می‌بیند، سپس با اطمینان بالا به نظرات بدون برچسب پیش‌برچسب می‌زند و آن‌ها را به مجموعه آموزشی اضافه می‌کند.</li>
  </ul>
</li>
<li>نتایج به صورت بصری نمایش داده شده‌اند:
  <ul>
    <li>یک <b>نمودار میله‌ای</b> مقایسه دقت مدل نظارت‌شده و نیمه‌نظارتی را نشان می‌دهد.</li>
    <li>یک <b>ماتریس درهم‌ریختگی</b> عملکرد مدل نیمه‌نظارتی را به صورت جزئی‌تر نشان می‌دهد.</li>
    <li>یک <b>نمودار دایره‌ای</b> توزیع داده‌های برچسب‌دار و بدون برچسب را نشان می‌دهد.</li>
    <li>یک <b>نمودار هیستوگرام</b> توزیع احتمالات پیش‌بینی برای داده‌های بدون برچسب را نشان می‌دهد.</li>
  </ul>
</li>
<li>تحلیل عمیق‌تری از نمونه‌هایی که مدل بیشترین عدم اطمینان را دارد انجام شده است که شبیه‌سازی <b>یادگیری فعال</b> است — نمونه‌هایی که برای برچسب‌گذاری توسط متخصصان مناسب‌ترین هستند.</li>
</ul>
این مثال به‌وضوح نشان می‌دهد که چگونه <b>یادگیری نیمه‌نظارتی</b> می‌تواند در شرایط واقعی که داده‌های برچسب‌دار بسیار محدود هستند، مدل‌های باکیفیتی بسازد و هزینه‌های عملیاتی را به شدت کاهش دهد. برخلاف یادگیری ماشین سنتی که به داده‌های برچسب‌دار زیادی نیاز دارد، داده‌کاوی با توسعه روش‌های <b>ضعیف‌نظارت‌شده</b> مانند یادگیری نیمه‌نظارتی و یادگیری فعال، می‌تواند با استفاده هوشمندانه از داده‌های بدون برچسب، عملکرد مدل را بهبود بخشد. این مثال نشان می‌دهد که داده‌کاوی و یادگیری ماشین دو رشته مکمل هستند که داده‌کاوی با تمرکز بر مقیاس‌پذیری و کاربردهای عملی، راه‌حل‌های نوآورانه‌ای برای چالش‌های واقعی ارائه می‌دهد.
