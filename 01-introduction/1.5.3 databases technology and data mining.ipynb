    مثال عملی: تحلیل فروش با استفاده از مکعب داده و داده‌کاوی 📊
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در این مثال، یک شرکت چندملیتی داده‌های فروش خود را از ۱۰ کشور مختلف در یک انبار داده یکپارچه کرده است. داده‌ها در ابعاد "زمان"، "محصول"، "منطقه" و "فروشنده" سازمان‌دهی شده و به صورت یک مکعب داده چندبعدی ذخیره شده‌اند. مدیران می‌توانند با استفاده از عملیات drill-down (مثلاً از سال به فصل) و roll-up (مثلاً از شهر به کشور)، به سرعت تحلیل‌های مختلفی انجام دهند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    علاوه بر تحلیل‌های توصیفی، از داده‌کاوی برای کشف الگوهای پنهان استفاده می‌شود. مثلاً با اعمال الگوریتم Apriori روی داده‌های فروش در بعد "محصول"، قوانین انجمنی مانند "خرید گوشی → خرید محافظ صفحه نمایش" کشف می‌شود. همچنین با ترکیب بعد "زمان" و "منطقه"، الگوهای فصلی در تقاضا شناسایی می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    این مثال نشان می‌دهد که چگونه تلفیق فناوری پایگاه داده (مکعب داده، OLAP) و داده‌کاوی می‌تواند به تولید بینش‌های استراتژیک منجر شود که نه‌تنها عملکرد گذشته را خلاصه می‌کند، بلکه فرصت‌های آینده را نیز پیش‌بینی می‌کند.
  </p>

```python
# این کد، یک مثال عملی از **تحلیل فروش با استفاده از مکعب داده و داده‌کاوی** را نشان می‌دهد
# که در بخش 1.8 کتاب معرفی شده است. این مثال نشان می‌دهد که چگونه فناوری پایگاه داده
# و داده‌کاوی می‌توانند ترکیب شوند تا بینش‌های استراتژیک تولید کنند.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
import seaborn as sns
from itertools import product

# تنظیم seed برای تولید خروجی‌های قابل تکرار
np.random.seed(42)

# شبیه‌سازی داده‌های فروش برای یک شرکت چندملیتی
# این بخش، **پیش‌پردازش داده** و **یکپارچه‌سازی داده** از منابع مختلف را شبیه‌سازی می‌کند.

# تعریف ابعاد مکعب داده
countries = ['USA', 'Germany', 'Japan', 'UK', 'France', 'Canada', 'Australia', 'Italy', 'Spain', 'Sweden']
products = ['Smartphone', 'Screen Protector', 'Phone Case', 'Laptop', 'Headphones', 'Tablet', 'Charger', 'Smartwatch']
quarters = ['Q1', 'Q2', 'Q3', 'Q4']
sellers = ['Amazon', 'BestBuy', 'ElectronicsStore', 'OnlineRetailer', 'TechMart']

# ایجاد داده‌های فروش برای 2 سال (8 فصل)
n_records = 5000
data = []

for _ in range(n_records):
    country = np.random.choice(countries)
    product = np.random.choice(products)
    quarter = np.random.choice(quarters)
    seller = np.random.choice(sellers)
    
    # شبیه‌سازی حجم فروش با توجه به عوامل مختلف
    base_sales = np.random.poisson(5)  # میانگین 5 واحد در هر تراکنش
    
    # افزایش فروش در فصل‌های خاص
    if quarter == 'Q4':  # تعطیلات سال
        base_sales *= 2
    
    # افزایش فروش در کشورهای خاص
    if country in ['USA', 'Germany', 'Japan']:
        base_sales *= 1.5
    
    # افزایش فروش برای فروشندگان بزرگ
    if seller in ['Amazon', 'BestBuy']:
        base_sales *= 1.8
    
    # شبیه‌سازی قیمت با توجه به محصول و کشور
    base_price = {
        'Smartphone': 800, 'Screen Protector': 20, 'Phone Case': 30, 
        'Laptop': 1200, 'Headphones': 150, 'Tablet': 500, 
        'Charger': 25, 'Smartwatch': 300
    }
    
    # تغییرات قیمت بر اساس کشور
    country_multiplier = {
        'USA': 1.0, 'Germany': 1.1, 'Japan': 1.05, 'UK': 1.15, 
        'France': 1.1, 'Canada': 0.95, 'Australia': 1.2, 
        'Italy': 1.05, 'Spain': 1.0, 'Sweden': 1.15
    }
    
    price = base_price[product] * country_multiplier[country] * np.random.uniform(0.9, 1.1)
    revenue = base_sales * price
    
    data.append({
        'Country': country,
        'Product': product,
        'Quarter': quarter,
        'Seller': seller,
        'Sales_Volume': base_sales,
        'Price': price,
        'Revenue': revenue
    })

# ایجاد دیتافریم
df = pd.DataFrame(data)

# مرحله 1: تحلیل OLAP و مکعب داده
print("مرحله 1: تحلیل OLAP و ساخت مکعب داده")

# ساخت مکعب داده چندبعدی با استفاده از pivot_table
# مکعب داده بر اساس کشور و فصل
cube_country_quarter = pd.pivot_table(
    df, 
    values='Revenue', 
    index='Country', 
    columns='Quarter', 
    aggfunc='sum', 
    fill_value=0,
    margins=True  # اضافه کردن ردیف و ستون کل
)

# مکعب داده بر اساس محصول و فصل
cube_product_quarter = pd.pivot_table(
    df, 
    values='Revenue', 
    index='Product', 
    columns='Quarter', 
    aggfunc='sum', 
    fill_value=0,
    margins=True
)

# مکعب داده بر اساس کشور و محصول
cube_country_product = pd.pivot_table(
    df, 
    values='Revenue', 
    index='Country', 
    columns='Product', 
    aggfunc='sum', 
    fill_value=0,
    margins=True
)

print(f"مکعب داده کشور-فصل (نمونه 5 کشور):")
print(cube_country_quarter.head())

print(f"\nمکعب داده محصول-فصل:")
print(cube_product_quarter)

# نمایش نتایج
plt.figure(figsize=(20, 6))

# زیرنویس اول: تحلیل فروش بر اساس کشور
plt.subplot(1, 3, 1)
country_revenue = df.groupby('Country')['Revenue'].sum().sort_values(ascending=False)
plt.bar(country_revenue.index, country_revenue.values, color='#1B263B', alpha=0.8)
plt.xlabel('Country')
plt.ylabel('Total Revenue')
plt.title('Sales Revenue by Country')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

# زیرنویس دوم: تحلیل فروش بر اساس فصل
plt.subplot(1, 3, 2)
quarter_revenue = df.groupby('Quarter')['Revenue'].sum()
quarters_ordered = ['Q1', 'Q2', 'Q3', 'Q4']
plt.bar(quarters_ordered, [quarter_revenue[q] for q in quarters_ordered], color='#0F4C81', alpha=0.8)
plt.xlabel('Quarter')
plt.ylabel('Total Revenue')
plt.title('Seasonal Sales Pattern')
plt.grid(True, alpha=0.3)

# زیرنویس سوم: تحلیل فروش بر اساس محصول
plt.subplot(1, 3, 3)
product_revenue = df.groupby('Product')['Revenue'].sum().sort_values(ascending=False)
plt.bar(product_revenue.index, product_revenue.values, color='#A23B72', alpha=0.8)
plt.xlabel('Product')
plt.ylabel('Total Revenue')
plt.title('Revenue by Product Category')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# عملیات OLAP: Drill-down و Roll-up
print("\nعملیات OLAP:")
print("Drill-down: تحلیل فروش گوشی هوشمند در کشورهای مختلف:")

smartphone_data = df[df['Product'] == 'Smartphone']
smartphone_country = smartphone_data.groupby('Country')['Revenue'].sum().sort_values(ascending=False)
print(smartphone_country)

print("\nRoll-up: تحلیل کلی فروش در سطح جهانی:")
total_revenue = df['Revenue'].sum()
print(f"کل درآمد: ${total_revenue:,.2f}")
print(f"میانگین درآمد در هر فصل: ${total_revenue/8:,.2f}")

# مرحله 2: داده‌کاوی بر روی مکعب داده
print("\nمرحله 2: داده‌کاوی بر روی داده‌های فروش")

# آماده‌سازی داده‌ها برای تحلیل انجمنی
# گروه‌بندی تراکنش‌ها بر اساس کشور و فصل
df['Country_Quarter'] = df['Country'] + '_' + df['Quarter']
transactions = []

# شبیه‌سازی سبد خرید بر اساس فروشندگان و زمان
for (seller, country_quarter), group in df.groupby(['Seller', 'Country_Quarter']):
    if len(group) > 1:
        basket = group['Product'].tolist()
        transactions.append(basket)

# تبدیل داده‌های تراکنشی به فرمت مناسب برای تحلیل انجمنی
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df_transactions = pd.DataFrame(te_ary, columns=te.columns_)

# این بخش، **داده‌کاوی** را شبیه‌سازی می‌کند.
# استفاده از الگوریتم Apriori برای یافتن مجموعه‌های پرتکرار
frequent_itemsets = apriori(df_transactions, min_support=0.02, use_colnames=True)

# این بخش، **ارزیابی الگو** را شبیه‌سازی می‌کند.
# تولید قوانین انجمنی با حداقل اطمینان 0.3
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.3)

# فیلتر قوانین برای نمایش قوانین جالب‌تر
interesting_rules = rules[
    (rules['lift'] > 1.2) & 
    (rules['confidence'] > 0.4)
].sort_values('lift', ascending=False)

print(f"\nقوانین انجمنی جالب کشف شده ({len(interesting_rules)} قانون):")
if len(interesting_rules) > 0:
    print(interesting_rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head(10))
else:
    print("هیچ قانون جالبی با معیارهای تعیین‌شده پیدا نشد.")

# کاوش الگوهای فصلی
print("\nکاوش الگوهای فصلی در تقاضا:")

# تحلیل فروش محصولات خاص در فصول مختلف
seasonal_analysis = df.groupby(['Product', 'Quarter'])['Sales_Volume'].mean().unstack()
print(seasonal_analysis.round(2))

# شناسایی محصولات با الگوی فصلی قوی
strong_seasonal_products = []
for product in products:
    sales_by_quarter = seasonal_analysis.loc[product]
    seasonal_variation = sales_by_quarter.std() / sales_by_quarter.mean()
    if seasonal_variation > 0.3:  # ضریب تغییرات بالا
        strong_seasonal_products.append(product)
        print(f"{product}: تغییرات فصلی قوی (ضریب تغییرات: {seasonal_variation:.2f})")

# نمایش الگوهای فصلی
plt.figure(figsize=(15, 5))

# زیرنویس اول: الگوهای فصلی در فروش
plt.subplot(1, 2, 1)
for product in strong_seasonal_products[:4]:  # فقط 4 محصول برای خوانایی
    sales_by_quarter = seasonal_analysis.loc[product]
    plt.plot(sales_by_quarter.index, sales_by_quarter.values, marker='o', label=product, linewidth=2)

plt.xlabel('Quarter')
plt.ylabel('Average Sales Volume')
plt.title('Seasonal Sales Patterns for Key Products')
plt.legend()
plt.grid(True, alpha=0.3)

# زیرنویس دوم: قوانین انجمنی کشف شده
plt.subplot(1, 2, 2)
if len(interesting_rules) > 0:
    top_rules = interesting_rules.head(6)
    rule_labels = [f"{list(rule['antecedents'])[0]} → {list(rule['consequents'])[0]}" for _, rule in top_rules.iterrows()]
    lift_values = top_rules['lift'].values
    
    plt.barh(rule_labels, lift_values, color='#1B263B', alpha=0.8)
    plt.xlabel('Lift')
    plt.title('Top Association Rules (Lift)')
    plt.grid(True, alpha=0.3)
else:
    plt.text(0.5, 0.5, 'No strong association rules found', horizontalalignment='center', 
             verticalalignment='center', transform=plt.gca().transAxes, fontsize=12)

plt.tight_layout()
plt.show()

# تحلیل ترکیبی: داده‌کاوی چندبعدی
print("\nتحلیل ترکیبی: داده‌کاوی چندبعدی")

# تحلیل قوانین انجمنی در کشورهای خاص
us_data = df[df['Country'] == 'USA']
te_us = TransactionEncoder()
te_ary_us = te_us.fit(us_data[['Product']]).transform(us_data[['Product']])
df_us = pd.DataFrame(te_ary_us, columns=te_us.columns_)

frequent_itemsets_us = apriori(df_us, min_support=0.05, use_colnames=True)
rules_us = pd.DataFrame()  # خالی
if not frequent_itemsets_us.empty:
    rules_us = association_rules(frequent_itemsets_us, metric="confidence", min_threshold=0.4)
    print(f"\nقوانین انجمنی در ایالات متحده ({len(rules_us)} قانون):")
    if len(rules_us) > 0:
        print(rules_us[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head())
    else:
        print("هیچ قانونی در ایالات متحده پیدا نشد.")
else:
    print("\n⚠️ هیچ مجموعه پرتکراری در داده‌های ایالات متحده پیدا نشد (min_support خیلی بالاست).")

print(f"\nقوانین انجمنی در ایالات متحده ({len(rules_us)} قانون):")
if len(rules_us) > 0:
    print(rules_us[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head())
else:
    print("هیچ قانونی در ایالات متحده پیدا نشد.")

# نتیجه‌گیری
print(f"\nنتیجه‌گیری:")
print(f"ترکیب فناوری پایگاه داده (مکعب داده، OLAP) و داده‌کاوی (تحلیل انجمنی) منجر به تولید بینش‌های استراتژیک شده است.")
print(f"تحلیل OLAP نشان داد که {country_revenue.index[0]} بیشترین درآمد را دارد و فصل {quarter_revenue.idxmax()} اوج فروش است.")
if not interesting_rules.empty:
    first_rule = interesting_rules.iloc[0]
    print(f"داده‌کاوی قوانین انجمنی مانند {list(first_rule['antecedents'])} → {list(first_rule['consequents'])} را کشف کرد.")
else:
    print("هیچ قانون انجمنی جالبی کشف نشد.")
print(f"تحلیل ترکیبی امکان شناسایی الگوهای خاص کشور و فصل را فراهم کرد.")
```

در این مثال عملی، فرآیند <b>تحلیل فروش با استفاده از مکعب داده و داده‌کاوی</b> به صورت کامل و در یک سناریوی واقعی شرکت چندملیتی پیاده‌سازی شده است:
<ul>
<li>ابتدا یک مجموعه داده فرضی از <b>فروش محصولات الکترونیکی</b> در 10 کشور مختلف برای 2 سال شبیه‌سازی شده است که شامل ابعاد "کشور"، "محصول"، "فصل"، "فروشنده"، "حجم فروش"، "قیمت" و "درآمد" است.</li>
<li>در مرحله <b>ساخت مکعب داده</b>، از تکنیک‌های <b>pivot_table</b> برای ساخت مکعب‌های داده چندبعدی استفاده شده است که داده‌ها را در ابعاد مختلف (کشور-فصل، محصول-فصل، کشور-محصول) تجمیع می‌کند.</li>
<li>در مرحله <b>تحلیل OLAP</b>، از عملیات <b>Drill-down</b> (مثلاً تحلیل فروش گوشی هوشمند در کشورهای مختلف) و <b>Roll-up</b> (مثلاً تحلیل کلی فروش در سطح جهانی) استفاده شده است.</li>
<li>در مرحله <b>داده‌کاوی</b>، از الگوریتم <b>Apriori</b> برای کشف قوانین انجمنی استفاده شده است که روابط بین محصولات خریداری شده را کشف می‌کند.</li>
<li>در مرحله <b>کاوش الگوهای فصلی</b>، تحلیل فروش محصولات مختلف در فصول مختلف انجام شده و محصولات با الگوی فصلی قوی شناسایی شده‌اند.</li>
<li>در مرحله <b>داده‌کاوی چندبعدی</b>، تحلیل ترکیبی انجام شده است که امکان شناسایی الگوهای خاص کشور و فصل را فراهم می‌کند.</li>
<li>نتایج به صورت بصری نمایش داده شده‌اند:
  <ul>
    <li>سه <b>نمودار میله‌ای</b> تحلیل فروش بر اساس کشور، فصل و محصول را نشان می‌دهند.</li>
    <li>یک <b>نمودار خطی</b> الگوهای فصلی در فروش محصولات کلیدی را نشان می‌دهد.</li>
    <li>یک <b>نمودار میله‌ای افقی</b> قوانین انجمنی برتر را بر اساس معیار Lift نشان می‌دهد.</li>
  </ul>
</li>
</ul>
این مثال به‌وضوح نشان می‌دهد که چگونه <b>تلفیق فناوری پایگاه داده و داده‌کاوی</b> می‌تواند به تولید بینش‌های استراتژیک منجر شود. فناوری پایگاه داده با ایجاد <b>انبار داده</b> و <b>مکعب داده چندبعدی</b>، امکان تحلیل‌های توصیفی سریع و انعطاف‌پذیر (OLAP) را فراهم می‌کند. در عین حال، داده‌کاوی با اعمال الگوریتم‌های هوشمند بر روی این داده‌های یکپارچه، قادر است الگوهای پنهان و روابط غیربدیهی را کشف کند. این ترکیب نه‌تنها عملکرد گذشته را خلاصه می‌کند، بلکه فرصت‌های آینده را نیز پیش‌بینی می‌کند و به مدیران کمک می‌کند تا تصمیم‌های استراتژیک بهتری بگیرند. این رویکرد ترکیبی نشان می‌دهد که داده‌کاوی و فناوری پایگاه داده دو رشته مکمل هستند که با کاربرد همزمان، ارزش بسیار بیشتری نسبت به استفاده جداگانه ایجاد می‌کنند.
