  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>مرحله ۱: فیلتر عینی (Statistical Filtering)</b><br>
    ابتدا قوانینی که حداقل حمایت ۵٪ و حداقل اطمینان ۶۰٪ ندارند حذف می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>مرحله ۲: فیلتر ذهنی و کیفی (Business Evaluation)</b><br>
    از بین قوانین باقی‌مانده، مدیر بازاریابی آمازون آن‌ها را تحلیل می‌کند:
  </p>

  <p style="font-size: 15pt; text-align: center; margin: 0 0 10px;">
    <b>قانون اول (بدیهی و غیرجالب):</b><br><br>
    <code>خرید کتاب‌خوان Kindle ⇒ خرید کابل شارژر USB (Confidence = 92%)</code>
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    تحلیل: اگرچه اطمینان بسیار بالا دارد، اما این یک الگوی بدیهی (Trivial) است. همه می‌دانند خریدار دستگاه الکترونیکی کابل شارژ هم می‌خواهد. این قانون هیچ دانش جدیدی ایجاد نمی‌کند و حذف می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: center; margin: 0 0 10px;">
    <b>قانون دوم (غیرمنتظره و قابل اقدام - جالب):</b><br><br>
    <code>خرید هدست گیمینگ ⇒ خرید قطره چشم ضدخستگی (Confidence = 68%)</code>
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    تحلیل: این الگو اصلاً بدیهی نیست! داده‌کاوی نشان داده گیمرهایی که ساعات طولانی بازی می‌کنند دچار خشکی چشم می‌شوند. این قانون هم غیرمنتظره (Novel) است و هم قابل اقدام (Actionable)؛ آمازون می‌تواند این قطره را به عنوان پیشنهاد خریدی کنار هدست‌های گیمینگ نمایش دهد (Cross-selling).

```python
# ==========================================================
# فیلتر کردن الگوهای بدیهی در آمازون
# با استفاده از الگوریتم واقعی Apriori (کتابخانه mlxtend)
# ==========================================================

# نصب کتابخانه در صورت نیاز:
# pip install mlxtend

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import networkx as nx
from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing import TransactionEncoder

np.random.seed(42)

# ==========================================================
# گام اول: شبیه‌سازی فاکتورهای خرید کاربران
# ==========================================================
products = [
    'Kindle',
    'USB_Charging_Cable',
    'Gaming_Headset',
    'Eye_Drops',
    'Gaming_Mouse',
    'Laptop',
    'Laptop_Bag',
    'Coffee_Maker',
    'Coffee_Filters'
]

n_transactions = 5000
transactions = []

for _ in range(n_transactions):
    basket = set()

    # ------------------------------------------------------
    # الگوی بدیهی: خریدار Kindle معمولاً کابل شارژر هم می‌خرد
    # ------------------------------------------------------
    if np.random.rand() < 0.18:
        basket.add('Kindle')
        if np.random.rand() < 0.92:
            basket.add('USB_Charging_Cable')

    # ------------------------------------------------------
    # الگوی غیرمنتظره: خریدار هدست گیمینگ ممکن است قطره چشم بخرد
    # ------------------------------------------------------
    if np.random.rand() < 0.20:
        basket.add('Gaming_Headset')
        if np.random.rand() < 0.68:
            basket.add('Eye_Drops')

    # ------------------------------------------------------
    # الگوهای عمومی دیگر برای واقعی‌تر شدن داده‌ها
    # ------------------------------------------------------
    if np.random.rand() < 0.25:
        basket.add('Laptop')
        if np.random.rand() < 0.55:
            basket.add('Laptop_Bag')

    if np.random.rand() < 0.20:
        basket.add('Gaming_Mouse')
        if np.random.rand() < 0.45:
            basket.add('Gaming_Headset')

    if np.random.rand() < 0.15:
        basket.add('Coffee_Maker')
        if np.random.rand() < 0.70:
            basket.add('Coffee_Filters')

    if basket:
        transactions.append(list(basket))

print(f"تعداد کل فاکتورهای شبیه‌سازی‌شده: {len(transactions)}")
print("\nنمونه‌ای از فاکتورهای خرید:")
for t in transactions[:5]:
    print(t)


# ==========================================================
# گام دوم: تبدیل فاکتورها به فرمت مناسب Apriori (One-Hot Encoding)
# ==========================================================
encoder = TransactionEncoder()
encoded_array = encoder.fit(transactions).transform(transactions)

basket_df = pd.DataFrame(encoded_array, columns=encoder.columns_)

print("\nنمونه‌ای از داده رمزگذاری‌شده (One-Hot):")
print(basket_df.head())


# ==========================================================
# گام سوم: استخراج مجموعه‌های پرتکرار با الگوریتم Apriori
# ==========================================================
frequent_itemsets = apriori(
    basket_df,
    min_support=0.05,   # حداقل حمایت مطابق متن مثال
    use_colnames=True
)

print("\n" + "=" * 70)
print("مجموعه‌های پرتکرار استخراج‌شده (Support >= 5%)")
print("=" * 70)
print(frequent_itemsets.sort_values(by='support', ascending=False).round(3))


# ==========================================================
# گام چهارم: استخراج قوانین وابستگی از مجموعه‌های پرتکرار
# ==========================================================
rules_all = association_rules(
    frequent_itemsets,
    metric='confidence',
    min_threshold=0.60   # حداقل اطمینان مطابق متن مثال
)

# تعداد واقعی قوانین استخراج‌شده پس از فیلتر آماری (support/confidence)
total_rules_after_statistical_filter = len(rules_all)

# فقط قوانین با یک آیتم در سمت چپ و راست برای خوانایی بهتر
rules = rules_all[
    (rules_all['antecedents'].apply(lambda x: len(x) == 1)) &
    (rules_all['consequents'].apply(lambda x: len(x) == 1))
].copy()

# تبدیل frozenset به رشته متنی برای نمایش بهتر
rules['Antecedent'] = rules['antecedents'].apply(lambda x: list(x)[0])
rules['Consequent'] = rules['consequents'].apply(lambda x: list(x)[0])

statistical_rules = rules[
    ['Antecedent', 'Consequent', 'support', 'confidence', 'lift']
].sort_values(by='confidence', ascending=False).reset_index(drop=True)

print("\n" + "=" * 70)
print("مرحله ۱ - قوانین باقی‌مانده پس از فیلتر آماری (Statistical Filtering)")
print("=" * 70)
print(statistical_rules.round(3).to_string(index=False))


# ==========================================================
# گام پنجم: فیلتر ذهنی و کیفی (Business Evaluation)
# ==========================================================
# قوانینی که از نظر کسب‌وکار بدیهی هستند و دانش جدیدی ایجاد نمی‌کنند
trivial_rules = {
    ('Kindle', 'USB_Charging_Cable'),
    ('USB_Charging_Cable', 'Kindle')
}

# قوانینی که غیرمنتظره و قابل اقدام هستند
interesting_rules = {
    ('Gaming_Headset', 'Eye_Drops'),
    ('Eye_Drops', 'Gaming_Headset')
}

def evaluate_rule(row):
    pair = (row['Antecedent'], row['Consequent'])

    if pair in trivial_rules:
        return 'Trivial - بدیهی و حذف‌شدنی'
    elif pair in interesting_rules:
        return 'Novel & Actionable - غیرمنتظره و قابل اقدام'
    else:
        return 'Needs Business Review - نیازمند بررسی مدیر'


statistical_rules['BusinessEvaluation'] = statistical_rules.apply(
    evaluate_rule, axis=1
)

print("\n" + "=" * 70)
print("مرحله ۲ - نتیجه فیلتر ذهنی و کیفی مدیر بازاریابی")
print("=" * 70)
print(
    statistical_rules[
        ['Antecedent', 'Consequent', 'confidence', 'BusinessEvaluation']
    ].round(3).to_string(index=False)
)


# ==========================================================
# گام ششم: جداسازی قوانین بدیهی از قوانین نهایی
# ==========================================================
removed_rules = statistical_rules[
    statistical_rules['BusinessEvaluation'] == 'Trivial - بدیهی و حذف‌شدنی'
]

final_rules = statistical_rules[
    statistical_rules['BusinessEvaluation'] != 'Trivial - بدیهی و حذف‌شدنی'
]

print("\n" + "=" * 70)
print("قوانین بدیهی حذف‌شده توسط مدیر بازاریابی")
print("=" * 70)

for _, row in removed_rules.iterrows():
    print(f"\n❌ {row['Antecedent']} ⇒ {row['Consequent']}")
    print(f"   Confidence = {row['confidence']:.0%}")
    print("   تحلیل: این الگو بدیهی است؛ خریدار دستگاه الکترونیکی")
    print("   به‌طور طبیعی به کابل شارژر نیاز دارد. دانش جدیدی تولید نمی‌شود.")


print("\n" + "=" * 70)
print("قوانین نهایی تأییدشده برای سیستم پیشنهاددهنده آمازون")
print("=" * 70)

for _, row in final_rules.iterrows():
    print(f"\n✅ {row['Antecedent']} ⇒ {row['Consequent']}")
    print(f"   Support    = {row['support']:.2%}")
    print(f"   Confidence = {row['confidence']:.0%}")
    print(f"   Lift       = {row['lift']:.2f}")
    print(f"   ارزیابی    = {row['BusinessEvaluation']}")

    if row['Antecedent'] == 'Gaming_Headset' and row['Consequent'] == 'Eye_Drops':
        print("   💡 تحلیل: گیمرهایی که ساعات طولانی بازی می‌کنند دچار")
        print("      خشکی چشم می‌شوند. این رابطه غیرمنتظره و قابل اقدام است.")
        print("   📌 استراتژی: نمایش قطره چشم به‌عنوان پیشنهاد Cross-selling")
        print("      در کنار صفحه محصول هدست گیمینگ.")

sns.set_style('whitegrid')

# ==========================================================
# نمودار ۱: پراکندگی Support در برابر Confidence همه قوانین
# ==========================================================
plt.figure(figsize=(9, 7))

scatter = plt.scatter(
    statistical_rules['support'],
    statistical_rules['confidence'],
    c=statistical_rules['lift'],
    s=statistical_rules['lift'] * 150,
    cmap='coolwarm',
    edgecolor='black',
    alpha=0.8
)

# خط آستانه‌های فیلتر آماری
plt.axhline(0.60, color='gray', linestyle='--', linewidth=1, label='Min Confidence = 60%')
plt.axvline(0.05, color='gray', linestyle=':', linewidth=1, label='Min Support = 5%')

# برچسب‌گذاری دو قانون کلیدی مثال
for _, row in statistical_rules.iterrows():
    pair = (row['Antecedent'], row['Consequent'])
    if pair in trivial_rules or pair in interesting_rules:
        plt.annotate(
            f"{row['Antecedent']} → {row['Consequent']}",
            (row['support'], row['confidence']),
            textcoords="offset points", xytext=(8, 8),
            fontsize=9, fontweight='bold'
        )

plt.colorbar(scatter, label='Lift')
plt.xlabel('Support')
plt.ylabel('Confidence')
plt.title('Association Rules: Support vs Confidence (color/size = Lift)')
plt.legend()
plt.tight_layout()
plt.show()


# ==========================================================
# نمودار ۲: مقایسه ستونی قوانین بدیهی در برابر قوانین جالب
# ==========================================================
comparison_rules = statistical_rules[
    statistical_rules['BusinessEvaluation'] != 'Needs Business Review - نیازمند بررسی مدیر'
].copy()

comparison_rules['RuleLabel'] = (
    comparison_rules['Antecedent'] + ' ==> ' + comparison_rules['Consequent']
)

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

metrics = ['support', 'confidence', 'lift']
titles = ['Support', 'Confidence', 'Lift']
colors = comparison_rules['BusinessEvaluation'].map({
    'Trivial - بدیهی و حذف‌شدنی': 'lightgray',
    'Novel & Actionable - غیرمنتظره و قابل اقدام': 'seagreen'
})

for ax, metric, title in zip(axes, metrics, titles):
    ax.bar(comparison_rules['RuleLabel'], comparison_rules[metric], color=colors)
    ax.set_title(title)
    ax.tick_params(axis='x', rotation=20)

plt.suptitle('Trivial Rule (gray) vs Novel & Actionable Rule (green)', fontsize=13)
plt.tight_layout()
plt.show()


# ==========================================================
# نمودار ۳: قیف فیلترینگ (Funnel) نشان‌دهنده مراحل کاهش قوانین
# ==========================================================
all_possible_rules_count = total_rules_after_statistical_filter
after_statistical_count = len(statistical_rules)   # پس از فیلتر خوانایی (۱ به ۱)
after_business_count = len(final_rules)

funnel_stages = [
    'کل قوانین استخراج‌شده\n(قبل از فیلتر خوانایی)',
    'قوانین ۱ به ۱ خوانا\n(پس از فیلتر خوانایی)',
    'پس از فیلتر کیفی\n(حذف قوانین بدیهی)'
]
funnel_values = [all_possible_rules_count, after_statistical_count, after_business_count]

plt.figure(figsize=(8, 5))
bars = plt.bar(funnel_stages, funnel_values, color=['#B0C4DE', '#4682B4', '#2E8B57'])

for bar, value in zip(bars, funnel_values):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.3,
             str(value), ha='center', fontweight='bold')

plt.ylabel('تعداد قوانین')
plt.title('Funnel of Rule Filtering Process')
plt.tight_layout()
plt.show()


# ==========================================================
# نمودار ۴: گراف شبکه‌ای روابط بین کالاها (Network Graph)
# ==========================================================
G = nx.DiGraph()

for _, row in statistical_rules.iterrows():
    G.add_edge(
        row['Antecedent'],
        row['Consequent'],
        weight=row['confidence'],
        evaluation=row['BusinessEvaluation']
    )

edge_colors = []
for u, v, data in G.edges(data=True):
    if data['evaluation'] == 'Trivial - بدیهی و حذف‌شدنی':
        edge_colors.append('lightgray')
    elif data['evaluation'] == 'Novel & Actionable - غیرمنتظره و قابل اقدام':
        edge_colors.append('seagreen')
    else:
        edge_colors.append('steelblue')

edge_widths = [G[u][v]['weight'] * 6 for u, v in G.edges()]

plt.figure(figsize=(10, 8))
pos = nx.spring_layout(G, seed=42, k=1.2)

nx.draw_networkx_nodes(G, pos, node_color='lightyellow', node_size=1800, edgecolors='black')
nx.draw_networkx_labels(G, pos, font_size=9, font_weight='bold')
nx.draw_networkx_edges(
    G, pos, edge_color=edge_colors, width=edge_widths,
    arrowstyle='-|>', arrowsize=20, connectionstyle='arc3,rad=0.1'
)

edge_labels = {(u, v): f"{d['weight']:.0%}" for u, v, d in G.edges(data=True)}
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=8)

plt.title('Product Association Network\n(gray = trivial, green = actionable, blue = needs review)')
plt.axis('off')
plt.tight_layout()
plt.show()


# ==========================================================
# نمودار ۵: نمودار Lift برای نشان دادن قدرت واقعی رابطه
# ==========================================================
plt.figure(figsize=(8, 5))

lift_data = comparison_rules.sort_values('lift')
colors_lift = lift_data['BusinessEvaluation'].map({
    'Trivial - بدیهی و حذف‌شدنی': 'lightgray',
    'Novel & Actionable - غیرمنتظره و قابل اقدام': 'seagreen'
})

plt.barh(lift_data['RuleLabel'], lift_data['lift'], color=colors_lift)
plt.axvline(1, color='red', linestyle='--', label='Lift = 1 (no real relationship)')
plt.xlabel('Lift')
plt.title('Lift Comparison: Trivial vs Actionable Rules')
plt.legend()
plt.tight_layout()
plt.show()


# ==========================================================
# گام هفتم: خروجی نهایی برای تیم بازاریابی
# ==========================================================
marketing_opportunities = final_rules[
    final_rules['BusinessEvaluation'] == 'Novel & Actionable - غیرمنتظره و قابل اقدام'
]

print("\n" + "=" * 70)
print("خلاصه فرصت‌های بازاریابی کشف‌شده توسط داده‌کاوی")
print("=" * 70)

for _, row in marketing_opportunities.iterrows():
    print(
        f"وقتی کاربر «{row['Antecedent']}» را خریداری می‌کند، "
        f"پیشنهاد «{row['Consequent']}» در صفحه محصول نمایش داده شود."
    )
```

در این مثال:
<ul>
<li>ابتدا با استفاده از کتابخانه‌های <b>pandas</b> و <b>numpy</b>، هزاران فاکتور خرید فرضی شبیه‌سازی شده است که در دل خود دو نوع رابطه پنهان دارند: یک رابطه <b>بدیهی</b> (خرید Kindle و کابل شارژر) و یک رابطه <b>غیرمنتظره</b> (خرید هدست گیمینگ و قطره چشم ضدخستگی).</li>
<li>با استفاده از کتابخانه استاندارد <b>mlxtend</b> و الگوریتم واقعی <b>Apriori</b>، مجموعه‌های پرتکرار کالاها استخراج شده و از دل آن‌ها، قوانین وابستگی (Association Rules) به همراه سه معیار کلیدی <b>Support</b>، <b>Confidence</b> و <b>Lift</b> محاسبه شده‌اند.</li>
<li>در <b>مرحله فیلتر عینی (آماری)</b>، قوانینی که حداقل ۵٪ حمایت و ۶۰٪ اطمینان را ندارند حذف می‌شوند تا فقط الگوهای قابل اتکا باقی بمانند.</li>
<li>در <b>مرحله فیلتر ذهنی (کیفی)</b>، با تعریف یک تابع ارزیابی کسب‌وکاری، قانون Kindle و کابل شارژر به‌عنوان الگویی <b>بدیهی (Trivial)</b> شناسایی و حذف می‌شود، در حالی‌که قانون هدست گیمینگ و قطره چشم به‌عنوان الگویی <b>غیرمنتظره و قابل اقدام (Novel & Actionable)</b> برای استفاده در سیستم پیشنهاددهنده حفظ می‌شود.</li>
<li>در بخش <b>ویژوالایز کردن نتایج</b>، پنج نمودار مکمل ترسیم شده‌اند:
  <ul style="margin-top:10px;">
    <li><b>نمودار پراکندگی Support-Confidence:</b> موقعیت تمام قوانین را نسبت به خطوط آستانه فیلتر آماری نشان می‌دهد، درحالی‌که رنگ و اندازه نقاط بر اساس مقدار Lift تنظیم شده است.</li>
    <li><b>نمودار مقایسه ستونی:</b> تفاوت مستقیم بین قانون بدیهی (خاکستری) و قانون جالب (سبز) را در سه معیار اصلی به‌صورت کنار هم نمایش می‌دهد.</li>
    <li><b>نمودار قیف فیلترینگ (Funnel):</b> کاهش تدریجی تعداد قوانین را از مرحله اولیه تا فیلتر آماری و سپس فیلتر کیفی به تصویر می‌کشد.</li>
    <li><b>گراف شبکه‌ای محصولات:</b> با استفاده از کتابخانه <b>networkx</b>، رابطه بین کالاها را به‌صورت یک شبکه جهت‌دار ترسیم می‌کند که ضخامت هر یال متناسب با میزان Confidence آن قانون است.</li>
    <li><b>نمودار مقایسه Lift:</b> نشان می‌دهد چرا معیار Lift اهمیت بالایی دارد؛ قانون بدیهی با وجود Confidence بالا، Lift نزدیک به ۱ (رابطه ضعیف) دارد، اما قانون جالب Lift به‌مراتب بالاتری (رابطه واقعی و معنادار) نشان می‌دهد.</li>
  </ul>
</li>
</ul>
این مثال به‌خوبی نشان می‌دهد که فرایند واقعی داده‌کاوی در شرکت‌هایی مانند آمازون، صرفاً به اجرای یک الگوریتم آماری محدود نمی‌شود؛ بلکه ترکیبی از <b>فیلتر عینی (آماری)</b> و <b>فیلتر ذهنی (تخصص و قضاوت انسانی)</b> است که در نهایت، از میان هزاران قانون احتمالی، تنها دانشی <b>غیرمنتظره، معنادار و قابل تبدیل به ارزش تجاری</b> (مانند فروش مکمل یا Cross-selling) استخراج و به کار گرفته می‌شود.
