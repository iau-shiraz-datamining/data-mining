  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    📦 مثال عملی: بخش‌بندی مشتریان آمازون (Amazon Customer Segmentation)
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    شرکت آمازون با داشتن مکرر بیش از ۳۰۰ میلیون کاربر فعال، نمی‌تواند برای هر کاربر برچسب دستی بزند. تیم بازاریابی آمازون از الگوریتم‌های خوشه‌بندی روی مدل RFM (تازگی خرید، فراوانی خرید، ارزش مالی خرید) استفاده می‌کند تا ساختارهای پنهان رفتار خریداران را بدون داشتن هیچ الگوی قبلی کشف کند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>روند کشف خوشه‌ها در آمازون:</b><br><br>
    ویژگی‌های ورودی بدون برچسب (مثل: تعداد خرید در ماه، میانگین مبلغ هر سبد خرید، استفاده از سرویس Prime و نرخ مرجوعی کالا) به مدل داده می‌شود. الگوریتم خوشه‌بندی، مشتریان را به چند گروه متمایز تفکیک می‌کند:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>خوشه ۱ (خریداران وفادار و پرمصرف / Amazon Prime Loyalists):</b><br>
    &nbsp;&nbsp;&nbsp;• ویژگی: فرکانس خرید بسیار بالا، میانگین خریدهای گران‌قیمت، عضو اشتراک پرایم.<br>
    &nbsp;&nbsp;&nbsp;• استراتژی آمازون: ارائه دسترسی زودتر به تخفیف‌های Black Friday و پیشنهاد سرویس‌های جدید مثل Amazon Fresh.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>خوشه ۲ (شکارچیان تخفیف و خریداران فصلی / Bargain Hunters):</b><br>
    &nbsp;&nbsp;&nbsp;• ویژگی: خرید فقط در زمان حراجی‌های بزرگ (Prime Day)، فرکانس خرید پایین در طول سال، sensitivity بالا به قیمت.<br>
    &nbsp;&nbsp;&nbsp;• استراتژی آمازون: ارسال کوپن‌های تخفیف و ایمیل‌های پیشنهاد شگفت‌انگیز برای ترغیب آن‌ها به خرید.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>خوشه ۳ (خریداران تک-منظوره یا گذری / One-Time Buyers):</b><br>
    &nbsp;&nbsp;&nbsp;• ویژگی: ثبت‌نام کرده‌اند، فقط یک کالای خاص خریده‌اند و ماه‌هاست برنگشته‌اند.<br>
    &nbsp;&nbsp;&nbsp;• استراتژی آمازون: ارسال ایمیل‌های یادآوری ("بقیه خریداران این کالا، این موارد را هم پسندیده‌اند").
  </p>

</div>

```python
# ==========================================================
# کتابخانه‌های مورد نیاز
# ==========================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

np.random.seed(42)

# ==========================================================
# گام اول: شبیه‌سازی داده خام و بدون برچسب مشتریان آمازون
# ==========================================================
# فرض می‌کنیم ۲۰۰۰ مشتری داریم که هرکدام به‌صورت پنهان به یکی از ۳ رفتار خرید
# (بدون اینکه سیستم از قبل این برچسب رو بدونه) گرایش دارند.
n_customers = 2000

# --- گروه پنهان ۱: خریداران وفادار و پرمصرف (Prime Loyalists) ---
n1 = 700
purchase_freq_1 = np.random.normal(18, 3, n1)          # تعداد خرید در ماه: بالا
avg_basket_1 = np.random.normal(2_500_000, 400_000, n1) # میانگین مبلغ سبد: بالا
recency_1 = np.random.normal(3, 2, n1)                  # روزهای سپری‌شده از آخرین خرید: کم (تازه)
prime_member_1 = np.random.binomial(1, 0.95, n1)        # عضویت پرایم: تقریباً همه
return_rate_1 = np.random.normal(0.03, 0.02, n1)        # نرخ مرجوعی: کم

# --- گروه پنهان ۲: شکارچیان تخفیف (Bargain Hunters) ---
n2 = 800
purchase_freq_2 = np.random.normal(2, 1, n2)
avg_basket_2 = np.random.normal(600_000, 150_000, n2)
recency_2 = np.random.normal(45, 20, n2)
prime_member_2 = np.random.binomial(1, 0.25, n2)
return_rate_2 = np.random.normal(0.08, 0.03, n2)

# --- گروه پنهان ۳: خریداران تک‌منظوره یا گذری (One-Time Buyers) ---
n3 = n_customers - n1 - n2
purchase_freq_3 = np.random.normal(1, 0.3, n3)
avg_basket_3 = np.random.normal(300_000, 100_000, n3)
recency_3 = np.random.normal(150, 40, n3)
prime_member_3 = np.random.binomial(1, 0.05, n3)
return_rate_3 = np.random.normal(0.05, 0.03, n3)

# --- ترکیب همه گروه‌ها در یک دیتافریم واحد (بدون برچسب گروه!) ---
customer_data = pd.DataFrame({
    'CustomerID': range(1, n_customers + 1),
    'MonthlyPurchaseFrequency': np.concatenate([purchase_freq_1, purchase_freq_2, purchase_freq_3]),
    'AvgBasketValue': np.concatenate([avg_basket_1, avg_basket_2, avg_basket_3]),
    'RecencyDays': np.concatenate([recency_1, recency_2, recency_3]),
    'PrimeMember': np.concatenate([prime_member_1, prime_member_2, prime_member_3]),
    'ReturnRate': np.concatenate([return_rate_1, return_rate_2, return_rate_3])
})

# پاکسازی مقادیر منفی/غیرمنطقی که ممکنه از نویز نرمال ایجاد بشه
customer_data['MonthlyPurchaseFrequency'] = customer_data['MonthlyPurchaseFrequency'].clip(lower=0)
customer_data['AvgBasketValue'] = customer_data['AvgBasketValue'].clip(lower=50_000)
customer_data['RecencyDays'] = customer_data['RecencyDays'].clip(lower=0)
customer_data['ReturnRate'] = customer_data['ReturnRate'].clip(lower=0, upper=1)

# شافل کردن ردیف‌ها تا ترتیب گروه‌ها لو نره (چون در دنیای واقعی داده‌ها بی‌نظمن)
customer_data = customer_data.sample(frac=1, random_state=42).reset_index(drop=True)

print("نمونه‌ای از داده خام و بدون برچسب مشتریان (شبیه‌سازی مدل RFM):")
print(customer_data.head(6).round(2))


# ==========================================================
# گام دوم: پیش‌پردازش و استانداردسازی ویژگی‌ها
# ==========================================================
features = ['MonthlyPurchaseFrequency', 'AvgBasketValue', 'RecencyDays', 'PrimeMember', 'ReturnRate']
X = customer_data[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("\n✅ ویژگی‌ها استاندارد شدند تا مقیاس متفاوت آن‌ها (مثلاً تومان در برابر روز) اثر نامتوازن روی خوشه‌بندی نگذارد.")


# ==========================================================
# گام سوم: تعیین تعداد بهینه خوشه‌ها (Elbow Method + Silhouette Score)
# ==========================================================
inertia_values = []
silhouette_values = []
k_range = range(2, 8)

for k in k_range:
    kmeans_temp = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels_temp = kmeans_temp.fit_predict(X_scaled)
    inertia_values.append(kmeans_temp.inertia_)
    silhouette_values.append(silhouette_score(X_scaled, labels_temp))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(k_range, inertia_values, marker='o', color='steelblue')
axes[0].set_xlabel('Number of Clusters (k)')
axes[0].set_ylabel('Inertia (WCSS)')
axes[0].set_title('Elbow Method')

axes[1].plot(k_range, silhouette_values, marker='o', color='darkorange')
axes[1].set_xlabel('Number of Clusters (k)')
axes[1].set_ylabel('Silhouette Score')
axes[1].set_title('Silhouette Analysis')

plt.tight_layout()
plt.show()

print("\n📊 با توجه به نمودار Elbow و بالاترین Silhouette Score، عدد k=3 به‌عنوان تعداد بهینه خوشه انتخاب می‌شود")
print("   (دقیقاً منطبق با ۳ رفتار خرید متمایزی که در متن توضیح داده شده است).")


# ==========================================================
# گام چهارم: اجرای نهایی الگوریتم K-Means با k=3
# ==========================================================
optimal_k = 3
kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
customer_data['Cluster'] = kmeans.fit_predict(X_scaled)

print(f"\n✅ خوشه‌بندی نهایی با {optimal_k} خوشه انجام شد. تعداد مشتریان در هر خوشه:")
print(customer_data['Cluster'].value_counts().sort_index())


# ==========================================================
# گام پنجم: پروفایل‌سازی خوشه‌ها (Cluster Profiling)
# ==========================================================
cluster_profile = customer_data.groupby('Cluster')[features].mean().round(2)
print("\n📋 پروفایل میانگین ویژگی‌های هر خوشه (دانش استخراج‌شده از داده‌کاوی):")
print(cluster_profile)

# --- نگاشت خودکار خوشه‌ها به برچسب‌های تجاری بر اساس فرکانس خرید ---
cluster_labels_map = cluster_profile['MonthlyPurchaseFrequency'].sort_values(ascending=False)
ranked_clusters = cluster_labels_map.index.tolist()

business_names = {
    ranked_clusters[0]: " Amazon Prime Loyalists (خریداران وفادار و پرمصرف)",
    ranked_clusters[1]: " Bargain Hunters (شکارچیان تخفیف)",
    ranked_clusters[2]: " One-Time Buyers (خریداران تک‌منظوره)"
}

strategy_map = {
    ranked_clusters[0]: "ارائه دسترسی زودتر به تخفیف‌های Black Friday و پیشنهاد Amazon Fresh",
    ranked_clusters[1]: "ارسال کوپن تخفیف و ایمیل‌های پیشنهاد شگفت‌انگیز نزدیک Prime Day",
    ranked_clusters[2]: "ارسال ایمیل یادآوری مبتنی بر کالاهای مشابه خریداران دیگر"
}

print("\n" + "="*70)
print("🔍 نتیجه نهایی: شناسایی و نام‌گذاری خودکار خوشه‌ها")
print("="*70)
for cluster_id in ranked_clusters:
    print(f"\nخوشه شماره {cluster_id} → {business_names[cluster_id]}")
    print(cluster_profile.loc[cluster_id])
    print(f"💡 استراتژی پیشنهادی آمازون: {strategy_map[cluster_id]}")

customer_data['SegmentName'] = customer_data['Cluster'].map(business_names)


# ==========================================================
# گام ششم: تجسم بصری خوشه‌ها با کاهش بعد (PCA)
# ==========================================================
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
customer_data['PCA1'] = X_pca[:, 0]
customer_data['PCA2'] = X_pca[:, 1]

plt.figure(figsize=(9, 7))
palette = {business_names[ranked_clusters[0]]: 'green',
           business_names[ranked_clusters[1]]: 'orange',
           business_names[ranked_clusters[2]]: 'gray'}

sns.scatterplot(data=customer_data, x='PCA1', y='PCA2', hue='SegmentName',
                 palette=palette, alpha=0.6, s=50)

# نمایش مرکز خوشه‌ها روی نمودار
centers_pca = pca.transform(kmeans.cluster_centers_)
plt.scatter(centers_pca[:, 0], centers_pca[:, 1], c='red', marker='X', s=250,
            label='Cluster Centers', edgecolor='black')

plt.title('Amazon Customer Segmentation (K-Means Clustering on RFM Features)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()


# ==========================================================
# گام هفتم: نمودار مقایسه‌ای ویژگی‌های کلیدی بین خوشه‌ها
# ==========================================================
profile = cluster_profile.reset_index()

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

sns.barplot(
    data=profile,
    x="Cluster",
    y="MonthlyPurchaseFrequency",
    hue="Cluster",
    palette="viridis",
    legend=False,
    ax=axes[0]
)
axes[0].set_title("Purchase Frequency by Cluster")

sns.barplot(
    data=profile,
    x="Cluster",
    y="AvgBasketValue",
    hue="Cluster",
    palette="viridis",
    legend=False,
    ax=axes[1]
)
axes[1].set_title("Avg Basket Value by Cluster")

sns.barplot(
    data=profile,
    x="Cluster",
    y="RecencyDays",
    hue="Cluster",
    palette="viridis",
    legend=False,
    ax=axes[2]
)
axes[2].set_title("Recency (Days Since Last Purchase)")

plt.tight_layout()
plt.show()
```
