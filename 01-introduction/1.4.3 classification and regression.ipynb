  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی : اعتبارسنجی و مدیریت فروش اقساطی در «دیجی‌پِی»
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    فرض کنید تیم داده‌کاوی دیجی‌پِی می‌خواهد سیستم خرید اقساطی (الان بخر، بعداً پرداخت کن - BNPL) را برای خریداران دیجی‌کالا هوشمند کند. در این کاربرد واقعی، هر دو روش طبقه‌بندی و رگرسیون در کنار هم استفاده می‌شوند:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام اول: انتخاب ویژگی (Pre-processing)</b><br>
    دیجی‌پِی بیش از ۱۰۰ ویژگی از هر کاربر دارد (از سن و شهر گرفته تا رنگ قاب گوشی خریده‌شده و ساعت آنلاینی).<br>
    مرحله انتخاب ویژگی، موارد نامرتبط مثل «رنگ قاب گوشی» یا «ساعت آنلاین شدن در سایت» را حذف می‌کند و فقط متغیرهای کلیدی مثل «درآمد ماهانه»، «سابقه خرید» و «سوابق چک بازگشتی» را به عنوان ورودی مدل نگه می‌دارد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>گام دوم: طبقه‌بندی (پیش‌بینی کلاس گسسته)</b><br>
    سوال سیستم: «آیا این کاربر واجد شرایط دریافت اعتبار BNPL هست یا خیر؟»<br>
    عملکرد مدل: مدل درخت تصمیم با بررسی داده‌های آموزشی، خروجی را در قالب یکی از ۳ کلاس زیر تعیین می‌کند:<br>
    &nbsp;&nbsp;&nbsp;• کلاس A (کم‌ریسک): اعطای فوری اعتبار بدون نیاز به ضامن.<br>
    &nbsp;&nbsp;&nbsp;• کلاس B (ریسک متوسط): اعطای اعتبار مشروط به ارائه سفته یا چک.<br>
    &nbsp;&nbsp;&nbsp;• کلاس C (پرریسک): رد درخواست اعتبار.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>گام سوم: رگرسیون (پیش‌بینی عدد پیوسته)</b><br>
    سوال سیستم: «حداکثر سقف ریالی اعتباری که می‌توان به این کاربر اعطا کرد چقدر است؟»<br>
    عملکرد مدل: مدل رگرسیونی بر اساس متغیرهایی مثل درآمد و میانگین گردش مالی، یک فرمول ریاضی تشکیل می‌دهد و یک مقدار عددی دقیق خروجی می‌دهد؛ مثلاً:<br><br>
    <div style="text-align:center; font-weight:bold;">
      سقف اعتبار پیش‌بینی‌شده = ۳۵,۴۰۰,۰۰۰ تومان
    </div>
  </p>

</div>

```python
# ==========================================================
# کتابخانه‌های مورد نیاز
# ==========================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestRegressor
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.metrics import classification_report, mean_absolute_error, r2_score

np.random.seed(42)

# ==========================================================
# گام صفر: شبیه‌سازی داده خام کاربران دیجی‌پی (۱۰۰+ ویژگی)
# ==========================================================
n_users = 3000

# --- ویژگی‌های واقعاً مرتبط (Meaningful Features) ---
monthly_income = np.random.gamma(shape=5, scale=1_500_000, size=n_users)          # درآمد ماهانه (تومان)
purchase_history = np.random.poisson(lam=8, size=n_users)                          # تعداد خرید قبلی موفق
bounced_checks = np.random.poisson(lam=0.3, size=n_users)                          # تعداد چک/قسط برگشتی
avg_cashflow = monthly_income * np.random.uniform(0.5, 1.5, n_users)               # میانگین گردش مالی

# --- ویژگی‌های نامرتبط (Noise Features) که باید در انتخاب ویژگی حذف بشن ---
phone_case_color = np.random.choice(['Black', 'Blue', 'Red', 'White'], n_users)
online_hour = np.random.randint(0, 24, n_users)

# برای شبیه‌سازی واقعی‌تر «بیش از ۱۰۰ ویژگی»، ۱۵ ستون نویز کاملاً تصادفی هم اضافه می‌کنیم
noise_features = {f'RandomFeature_{i}': np.random.normal(0, 1, n_users) for i in range(1, 16)}

raw_data = pd.DataFrame({
    'UserID': range(1, n_users + 1),
    'MonthlyIncome': monthly_income.round(0),
    'PurchaseHistory': purchase_history,
    'BouncedChecks': bounced_checks,
    'AvgCashflow': avg_cashflow.round(0),
    'PhoneCaseColor': phone_case_color,   # نامرتبط
    'OnlineHour': online_hour,            # نامرتبط
    **noise_features                      # نامرتبط
})

print(f"تعداد کل ویژگی‌های خام هر کاربر: {raw_data.shape[1] - 1} ستون (شبیه‌سازی محیط واقعی دیجی‌پی)")
print("\nنمونه‌ای از داده خام کاربران:")
print(raw_data[['UserID', 'MonthlyIncome', 'PurchaseHistory', 'BouncedChecks', 'PhoneCaseColor']].head(5))

# ==========================================================
# ساخت متغیرهای هدف (Target) بر اساس منطق واقعی کسب‌وکار
# ==========================================================
# امتیاز ریسک ترکیبی (فقط برای تولید داده آموزشی واقع‌گرایانه - در دنیای واقعی این برچسب از تاریخچه واقعی می‌آید)
risk_score = (
    (raw_data['MonthlyIncome'] / 1_000_000) * 0.5 +
    raw_data['PurchaseHistory'] * 1.2 -
    raw_data['BouncedChecks'] * 8 +
    np.random.normal(0, 2, n_users)
)

# تعیین کلاس ریسک (A: کم‌ریسک, B: متوسط, C: پرریسک)
def assign_class(score):
    if score > 10:
        return 'A'
    elif score > 4:
        return 'B'
    else:
        return 'C'

raw_data['RiskClass'] = risk_score.apply(assign_class) if hasattr(risk_score, 'apply') else [assign_class(s) for s in risk_score]

# سقف اعتبار (برای رگرسیون) — فقط برای کاربرانی معنا داره که رد نشدن (کلاس C سقف صفر می‌گیره)
credit_limit = (
    raw_data['MonthlyIncome'] * 2.5 +
    raw_data['AvgCashflow'] * 0.8 +
    raw_data['PurchaseHistory'] * 300_000 -
    raw_data['BouncedChecks'] * 5_000_000 +
    np.random.normal(0, 1_000_000, n_users)
)
credit_limit = np.clip(credit_limit, 0, None)
raw_data['CreditLimit'] = np.where(raw_data['RiskClass'] == 'C', 0, credit_limit).round(-4)

print("\nنمونه‌ای از داده با برچسب‌های نهایی (کلاس ریسک و سقف اعتبار):")
print(raw_data[['UserID', 'MonthlyIncome', 'PurchaseHistory', 'BouncedChecks', 'RiskClass', 'CreditLimit']].head(5))


# ==========================================================
# گام اول: انتخاب ویژگی (Feature Selection)
# ==========================================================
print("\n" + "="*70)
print("گام اول: انتخاب ویژگی (Feature Selection)")
print("="*70)

# تبدیل ستون رنگ (که قراره حذف بشه) به عدد فقط برای اینکه بتونیم در آزمون آماری وارد کنیم
raw_data['PhoneCaseColor_encoded'] = raw_data['PhoneCaseColor'].astype('category').cat.codes

feature_candidates = ['MonthlyIncome', 'PurchaseHistory', 'BouncedChecks', 'AvgCashflow',
                       'PhoneCaseColor_encoded', 'OnlineHour'] + list(noise_features.keys())

X_candidates = raw_data[feature_candidates]
y_class = raw_data['RiskClass']

selector = SelectKBest(score_func=f_classif, k=4)
selector.fit(X_candidates, y_class)

feature_scores = pd.Series(selector.scores_, index=feature_candidates).sort_values(ascending=False)
print("\n📊 امتیاز اهمیت هر ویژگی (هرچه بالاتر، مرتبط‌تر با پیش‌بینی ریسک):")
print(feature_scores.head(10).round(2))

selected_features = feature_scores.head(4).index.tolist()
print(f"\n✅ ویژگی‌های نهایی انتخاب‌شده توسط سیستم: {selected_features}")
print("❌ ویژگی‌های نامرتبط مثل 'رنگ قاب گوشی' و 'ساعت آنلاین' و ویژگی‌های تصادفی حذف شدند.")

plt.figure(figsize=(10, 5))
feature_scores.head(10).plot(kind='bar', color='teal')
plt.title('Feature Importance Score (Feature Selection)')
plt.ylabel('ANOVA F-Score')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()


# ==========================================================
# گام دوم: طبقه‌بندی (Classification) - تعیین کلاس اعتباری A/B/C
# ==========================================================
print("\n" + "="*70)
print("گام دوم: طبقه‌بندی با درخت تصمیم (Decision Tree Classifier)")
print("="*70)

X = raw_data[selected_features]
y = raw_data['RiskClass']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

clf = DecisionTreeClassifier(max_depth=4, random_state=42)
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)

print("\n📋 گزارش عملکرد مدل طبقه‌بندی روی داده آزمون:")
print(classification_report(y_test, y_pred))

plt.figure(figsize=(16, 8))
plot_tree(clf, feature_names=selected_features, class_names=clf.classes_,
          filled=True, rounded=True, fontsize=9)
plt.title('Decision Tree - BNPL Credit Risk Classification')
plt.show()


# ==========================================================
# گام سوم: رگرسیون (Regression) - پیش‌بینی سقف اعتبار ریالی
# ==========================================================
print("\n" + "="*70)
print("گام سوم: رگرسیون برای پیش‌بینی سقف اعتبار (Random Forest Regressor)")
print("="*70)

# فقط روی کاربرانی که رد نشدن (کلاس A و B) مدل رگرسیون معنا داره
eligible = raw_data[raw_data['RiskClass'] != 'C']

X_reg = eligible[selected_features]
y_reg = eligible['CreditLimit']

X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X_reg, y_reg, test_size=0.25, random_state=42)

reg_model = RandomForestRegressor(n_estimators=200, max_depth=6, random_state=42)
reg_model.fit(X_train_r, y_train_r)
y_pred_r = reg_model.predict(X_test_r)

mae = mean_absolute_error(y_test_r, y_pred_r)
r2 = r2_score(y_test_r, y_pred_r)
print(f"\n📈 عملکرد مدل رگرسیون روی داده آزمون:")
print(f"   میانگین خطای مطلق (MAE): {mae:,.0f} تومان")
print(f"   ضریب تعیین (R²): {r2:.3f}")

plt.figure(figsize=(7, 6))
plt.scatter(y_test_r, y_pred_r, alpha=0.4, color='darkorange')
plt.plot([y_test_r.min(), y_test_r.max()], [y_test_r.min(), y_test_r.max()], 'k--', lw=2)
plt.xlabel('Actual Credit Limit (Toman)')
plt.ylabel('Predicted Credit Limit (Toman)')
plt.title('Regression Model - Predicted vs Actual Credit Limit')
plt.tight_layout()
plt.show()


# ==========================================================
# سناریوی نهایی: بررسی یک کاربر جدید (شبیه‌سازی یک درخواست واقعی BNPL)
# ==========================================================
print("\n" + "="*70)
print("سناریوی عملیاتی: بررسی درخواست اعتبار یک کاربر جدید")
print("="*70)

new_user = pd.DataFrame({
    'MonthlyIncome': [18_000_000],
    'PurchaseHistory': [12],
    'BouncedChecks': [0],
    'AvgCashflow': [20_000_000],
    'PhoneCaseColor_encoded': [2],   # این ویژگی در مدل نهایی استفاده نمی‌شه چون حذف شده
    'OnlineHour': [23]
})

new_user_selected = new_user[selected_features]

predicted_class = clf.predict(new_user_selected)[0]
print(f"\n🔎 نتیجه گام طبقه‌بندی: کاربر جدید در کلاس ریسک «{predicted_class}» قرار گرفت.")

if predicted_class == 'C':
    print("❌ نتیجه نهایی: درخواست اعتبار رد شد (پرریسک).")
else:
    predicted_limit = reg_model.predict(new_user_selected)[0]
    guarantee_note = "بدون نیاز به ضامن" if predicted_class == 'A' else "مشروط به ارائه چک/سفته"
    print(f"✅ نتیجه گام رگرسیون: سقف اعتبار پیش‌بینی‌شده = {predicted_limit:,.0f} تومان")
    print(f"📌 شرایط اعطای اعتبار: {guarantee_note}")
```

در این مثال:
<ul>
<li>ابتدا کتابخانه‌های مورد نیاز شامل <b>pandas</b> و <b>numpy</b> برای مدیریت داده‌ها، ابزارهای رسم نمودار (<b>matplotlib</b> و <b>seaborn</b>) و ماژول‌های هوش مصنوعی و یادگیری ماشین از کتابخانه قدرتمند <b>scikit-learn</b> وارد شده‌اند.</li>
<li>یک مجموعه داده فرضی از کاربران دیجی‌پِی با بیش از ۲۰ ویژگی شبیه‌سازی شده است که شامل متغیرهای کلیدی (مانند درآمد ماهانه، تعداد خرید قبلی و چک‌های برگشتی) در کنار ویژگی‌های نامرتبط و نویزی (مانند رنگ قاب گوشی و ساعت آنلاین) است تا محیط واقعی داده‌ها شبیه‌سازی شود.</li>
<li>در <b>گام اول (انتخاب ویژگی - Feature Selection)</b>، با استفاده از متد <b>SelectKBest</b> و آزمون آماری ANOVA، ویژگی‌های نامرتبط حذف شده و صرفاً متغیرهای اثرگذار برای مدل‌سازی حفظ می‌شوند.</li>
<li>در <b>گام دوم (طبقه‌بندی - Classification)</b>، یک مدل <b>درخت تصمیم (Decision Tree)</b> برای پیش‌بینی کلاس گسسته ریسک کاربر (کلاس A: کم‌ریسک بدون ضامن، کلاس B: ریسک متوسط مشروط به چک، و کلاس C: پرریسک و رد درخواست) آموزش داده شده و ساختار تصمیم‌گیری آن رسم شده است.</li>
<li>در <b>گام سوم (رگرسیون - Regression)</b>، یک مدل <b>رگرسیون جنگل تصادفی (Random Forest Regressor)</b> برای کاربران واجد شرایط (کلاس‌های A و B) پیاده‌سازی شده تا یک عدد پیوسته دقیق یعنی <b>سقف اعتبار ریالی</b> (بر حسب تومان) پیش‌بینی شود.</li>
<li>در نهایت، در بخش <b>سناریوی عملیاتی</b>، درخواست یک کاربر جدید از طریق خط لوله مدل‌سازی بررسی می‌شود تا نشان دهد چگونه ترکیب هم‌زمان طبقه‌بندی و رگرسیون می‌تواند تصمیمات اعتباری سیستم BNPL دیجی‌پِی را اتوماسیون کند.</li>
</ul>
این مثال به‌وضوح نشان می‌دهد که چگونه یک سامانه داده‌کاوی پیشرفته می‌تواند از میان انبوهی از داده‌های خام و پرنویز، ویژگی‌های کلیدی را استخراج کرده و با به‌کارگیری هم‌زمان تکنیک‌های <b>طبقه‌بندی</b> و <b>رگرسیون</b>، ارزش کسب‌وکاری فوق‌العاده‌ای (مدیریت ریسک مالی و اعطای هوشمند اعتبار) خلق کند.
