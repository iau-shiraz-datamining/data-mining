  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در این مثال، یک بانک مرکزی از داده‌کاوی برای پیش‌بینی بحران‌های اقتصادی احتمالی استفاده می‌کند. داده‌های کلان اقتصادی، رفتار بانکی شهروندان، فعالیت‌های بازار سهام و الگوهای مصرف جمع‌آوری می‌شوند. با این حال، تمام داده‌های شخصی به‌صورت رمزگذاری شده و جمع‌آوری شده (aggregated) استفاده می‌شوند تا هویت افراد محفوظ بماند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مدل داده‌کاوی با استفاده از تکنیک‌های تشخیص ناهنجاری و رگرسیون چندمتغیره، الگوهای غیرمعمول در رفتار اقتصادی را شناسایی می‌کند. مثلاً کاهش ناگهانی و گسترده در وام‌های مسکن یا افزایش سریع در برداشت‌های نقدی ممکن است نشانه‌ای از بی‌اعتمادی عمومی باشد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتایج به صورت گزارش‌های سطح کلان و بدون اشاره به افراد خاص، به سیاست‌گذاران ارائه می‌شود. این یافته‌ها به دولت کمک می‌کند تا قبل از وقوع بحران، اقدامات پیشگیرانه مانند تعدیل نرخ بهره یا افزایش تأمین مالی بخش‌های حیاتی را آغاز کند.
    <br><br>
    این مثال نشان می‌دهد که چگونه داده‌کاوی می‌تواند به‌صورت مسئولیت‌پذیر و اخلاقی برای منافع عمومی به کار گرفته شود، بدون اینکه به حریم خصوصی افراد آسیب برسد.
  </p>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
import seaborn as sns

# تنظیم seed برای تولید خروجی‌های قابل تکرار
np.random.seed(42)

# شبیه‌سازی داده‌های اقتصادی برای یک بانک مرکزی
# این بخش، **جمع‌آوری داده‌های کلان اقتصادی و رفتار بانکی** را شبیه‌سازی می‌کند
# با تأکید بر حفاظت از حریم خصوصی (داده‌ها از ابتدا به صورت تجمیع‌شده و رمزگذاری‌شده در نظر گرفته می‌شوند)

n_months = 120  # داده‌های 10 ساله
months = pd.date_range('2013-01-01', periods=n_months, freq='ME')

# شبیه‌سازی شاخص‌های اقتصادی کلان

# ستون‌های پایه
gdp_growth = 0.5 + np.random.normal(0, 0.8, n_months)
inflation_rate = 3 + 2 * np.sin(2 * np.pi * np.arange(n_months) / 24) + np.random.normal(0, 1, n_months)
unemployment_rate = 6 - 0.5 * gdp_growth + np.random.normal(0, 0.5, n_months)
stock_index = 1000 + np.cumsum(np.random.normal(0.5, 2, n_months))
interest_rate = 5 + np.sin(2 * np.pi * np.arange(n_months) / 36) + np.random.normal(0, 0.3, n_months)

# ساخت دیکشنری
data = {
    'Month': months,
    'GDP_Growth': gdp_growth,
    'Inflation_Rate': inflation_rate,
    'Unemployment_Rate': unemployment_rate,
    'Stock_Index': stock_index,
    'Interest_Rate': interest_rate
}


# تبدیل به دیتافریم
df = pd.DataFrame(data)

# حفظ محدوده معقول برای شاخص‌ها
df['Inflation_Rate'] = df['Inflation_Rate'].clip(lower=0, upper=20)
df['Unemployment_Rate'] = df['Unemployment_Rate'].clip(lower=2, upper=15)
df['Interest_Rate'] = df['Interest_Rate'].clip(lower=1, upper=10)

# شبیه‌سازی شاخص‌های رفتار مصرف و وام
# این داده‌ها به صورت تجمیع‌شده و بدون اشاره به افراد خاص شبیه‌سازی شده‌اند
df['Consumer_Confidence'] = (
    100 - 5 * df['Inflation_Rate'] + 
    2 * df['GDP_Growth'] - 
    3 * df['Unemployment_Rate'] + 
    np.random.normal(0, 5, n_months)
)

df['Housing_Loan_Applications'] = (
    50000 - 2000 * df['Interest_Rate'] + 
    500 * df['GDP_Growth'] - 
    1000 * df['Unemployment_Rate'] + 
    np.random.normal(0, 3000, n_months)
)

df['Credit_Card_Spending'] = (
    800 + 10 * df['GDP_Growth'] - 
    5 * df['Unemployment_Rate'] + 
    np.random.normal(0, 50, n_months)
)

df['Cash_Withdrawals'] = (
    1200 + 50 * df['Inflation_Rate'] + 
    20 * df['Unemployment_Rate'] + 
    np.random.normal(0, 80, n_months)
)

# شبیه‌سازی بحران اقتصادی (کاهش ناگهانی اعتماد عمومی)
crisis_start = 80  # ماه 80
crisis_duration = 12

# اعمال تغییرات برای شبیه‌سازی بحران
df.loc[crisis_start:crisis_start+crisis_duration, 'Consumer_Confidence'] -= np.linspace(20, 0, crisis_duration+1)
df.loc[crisis_start:crisis_start+crisis_duration, 'Housing_Loan_Applications'] -= np.linspace(15000, 0, crisis_duration+1)
df.loc[crisis_start:crisis_start+crisis_duration, 'Credit_Card_Spending'] -= np.linspace(150, 0, crisis_duration+1)
df.loc[crisis_start:crisis_start+crisis_duration, 'Cash_Withdrawals'] += np.linspace(300, 0, crisis_duration+1)
df.loc[crisis_start:crisis_start+crisis_duration, 'Stock_Index'] -= np.cumsum(np.random.normal(5, 3, crisis_duration+1))

# مرحله 1: پیش‌پردازش داده با تأکید بر حفاظت از حریم خصوصی
print("مرحله 1: پیش‌پردازش داده با تأکید بر حفاظت از حریم خصوصی")

# تأیید اینکه داده‌ها از ابتدا به صورت تجمیع‌شده هستند
print(f"تعداد رکوردها: {len(df)}")
print(f"شاخص‌های اقتصادی کلان: {list(df.columns[1:])}")
print("داده‌های شخصی فردی وجود ندارد. تمام داده‌ها به صورت تجمیع‌شده و برای کل جامعه هستند.")

# استانداردسازی داده‌ها برای تحلیل
scaler = StandardScaler()
economic_indicators = [
    'GDP_Growth', 'Inflation_Rate', 'Unemployment_Rate', 'Stock_Index', 
    'Interest_Rate', 'Consumer_Confidence', 'Housing_Loan_Applications',
    'Credit_Card_Spending', 'Cash_Withdrawals'
]

X_scaled = scaler.fit_transform(df[economic_indicators])

# مرحله 2: تشخیص ناهنجاری برای شناسایی علائم بحران
print("\nمرحله 2: تشخیص ناهنجاری برای شناسایی علائم بحران")

# استفاده از Isolation Forest برای شناسایی ناهنجاری‌ها
isolation_forest = IsolationForest(contamination=0.1, random_state=42, n_estimators=100)
anomaly_scores = isolation_forest.fit_predict(X_scaled)
anomaly_scores_binary = (anomaly_scores == -1).astype(int)

# افزودن نتایج به دیتافریم
df['Anomaly_Score'] = anomaly_scores_binary

print(f"تعداد ماه‌های شناسایی شده به عنوان ناهنجاری: {sum(anomaly_scores_binary)}")

# مرحله 3: مدل رگرسیون چندمتغیره برای پیش‌بینی شاخص‌های کلیدی
print("\nمرحله 3: مدل رگرسیون چندمتغیره برای پیش‌بینی شاخص‌های کلیدی")

# ایجاد ویژگی‌های تأخیری برای پیش‌بینی
for indicator in economic_indicators:
    df[f'{indicator}_Lag1'] = df[indicator].shift(1)
    df[f'{indicator}_Lag2'] = df[indicator].shift(2)

# حذف ردیف‌هایی که ویژگی‌های تأخیری آن‌ها NaN است
df_model = df.dropna()

# انتخاب ویژگی‌ها و هدف
feature_columns = [col for col in df_model.columns if col.endswith('_Lag1') or col.endswith('_Lag2')]
target_column = 'Consumer_Confidence'

X = df_model[feature_columns]
y = df_model[target_column]

# تقسیم داده‌ها
split_idx = int(0.8 * len(X))
X_train, X_test = X[:split_idx], X[split_idx:]
y_train, y_test = y[:split_idx], y[split_idx:]

# ساخت و آموزش مدل
regression_model = LinearRegression()
regression_model.fit(X_train, y_train)

# پیش‌بینی
y_pred = regression_model.predict(X_test)

# ارزیابی مدل
mse = mean_squared_error(y_test, y_pred)
print(f"MSE: {mse:.2f}")

# مرحله 4: تولید گزارش‌های سطح کلان برای سیاست‌گذاران
print("\nمرحله 4: تولید گزارش‌های سطح کلان برای سیاست‌گذاران")

# شناسایی ماه‌های بحرانی
crisis_months = df[df['Anomaly_Score'] == 1]['Month']
if len(crisis_months) > 0:
    print(f"ماه‌های شناسایی شده به عنوان بحران احتمالی:")
    for month in crisis_months:
        print(f"  - {month.strftime('%B %Y')}")
else:
    print("هیچ ماه بحرانی شناسایی نشد.")

# تحلیل عوامل مؤثر در ناهنجاری
print(f"\nتحلیل عوامل مؤثر در ناهنجاری‌ها:")

# محاسبه میانگین شاخص‌ها در ماه‌های ناهنجار در مقایسه با ماه‌های عادی
anomaly_data = df[df['Anomaly_Score'] == 1]
normal_data = df[df['Anomaly_Score'] == 0]

for indicator in economic_indicators:
    anomaly_mean = anomaly_data[indicator].mean()
    normal_mean = normal_data[indicator].mean()
    difference = anomaly_mean - normal_mean
    
    if abs(difference) > df[indicator].std() * 0.5:  # تفاوت معنادار
        direction = "افزایش" if difference > 0 else "کاهش"
        print(f"  - {indicator}: {direction} قابل توجه در ماه‌های ناهنجار ({anomaly_mean:.1f} vs {normal_mean:.1f})")

# تنظیمات کلی برای نمایش بهتر
plt.rcParams.update({'font.size': 10})
sns.set_style("whitegrid")

# نمایش نتایج
fig = plt.figure(figsize=(20, 15))

# زیرنویس اول: روند شاخص اعتماد مصرف‌کننده
plt.subplot(3, 3, 1)
plt.plot(df['Month'], df['Consumer_Confidence'], label='Consumer Confidence', color='#1B263B', linewidth=2)
plt.axvspan(df['Month'][crisis_start], df['Month'][crisis_start + crisis_duration], alpha=0.3, color='red', label='Crisis Period')
plt.xlabel('Month', fontsize=10)
plt.ylabel('Index', fontsize=10)
plt.title('Consumer Confidence Trend', fontsize=12, fontweight='bold')
plt.legend(fontsize=9)
plt.grid(True, alpha=0.3)

# زیرنویس دوم: روند درخواست وام مسکن
plt.subplot(3, 3, 2)
plt.plot(df['Month'], df['Housing_Loan_Applications'], label='Housing Loan Applications', color='#0F4C81', linewidth=2)
plt.axvspan(df['Month'][crisis_start], df['Month'][crisis_start + crisis_duration], alpha=0.3, color='red')
plt.xlabel('Month', fontsize=10)
plt.ylabel('Number of Applications', fontsize=10)
plt.title('Housing Loan Applications Trend', fontsize=12, fontweight='bold')
plt.grid(True, alpha=0.3)

# زیرنویس سوم: روند برداشت نقدی
plt.subplot(3, 3, 3)
plt.plot(df['Month'], df['Cash_Withdrawals'], label='Cash Withdrawals', color='#A23B72', linewidth=2)
plt.axvspan(df['Month'][crisis_start], df['Month'][crisis_start + crisis_duration], alpha=0.3, color='red')
plt.xlabel('Month', fontsize=10)
plt.ylabel('Amount', fontsize=10)
plt.title('Cash Withdrawals Trend', fontsize=12, fontweight='bold')
plt.grid(True, alpha=0.3)

# زیرنویس چهارم: ناهنجاری‌ها در رفتار اقتصادی
plt.subplot(3, 3, 4)
plt.plot(df['Month'], df['Anomaly_Score'], 'ro-', label='Anomaly Detected', markersize=5, linewidth=1)
plt.xlabel('Month', fontsize=10)
plt.ylabel('Anomaly Score', fontsize=10)
plt.title('Detected Anomalies in Economic Behavior', fontsize=12, fontweight='bold')
plt.yticks([0, 1], ['Normal', 'Anomaly'])
plt.grid(True, alpha=0.3)

# زیرنویس پنجم: پیش‌بینی اعتماد مصرف‌کننده
plt.subplot(3, 3, 5)
test_months = df['Month'][split_idx+2:]
plt.plot(test_months, y_test.values, label='Actual', color='#1B263B', linewidth=2)
plt.plot(test_months, y_pred, label='Predicted', color='#F18F01', linestyle='--', linewidth=2)
plt.xlabel('Month', fontsize=10)
plt.ylabel('Consumer Confidence', fontsize=10)
plt.title('Consumer Confidence: Actual vs Predicted', fontsize=12, fontweight='bold')
plt.legend(fontsize=9)
plt.grid(True, alpha=0.3)

# زیرنویس ششم: همبستگی شاخص‌ها
plt.subplot(3, 3, 6)
correlation_matrix = df[economic_indicators].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, square=True, fmt='.2f', cbar_kws={"shrink": .8})
plt.title('Correlation Matrix of Economic Indicators', fontsize=12, fontweight='bold')

# زیرنویس هفتم: توزیع نرخ تورم
plt.subplot(3, 3, 7)
plt.hist(df['Inflation_Rate'], bins=15, alpha=0.7, color='#1B263B', edgecolor='white', linewidth=0.5)
plt.xlabel('Inflation Rate (%)', fontsize=10)
plt.ylabel('Frequency', fontsize=10)
plt.title('Distribution of Inflation Rate', fontsize=12, fontweight='bold')
plt.grid(True, alpha=0.3)

# زیرنویس هشتم: روند بازار سهام
plt.subplot(3, 3, 8)
plt.plot(df['Month'], df['Stock_Index'], label='Stock Index', color='#0F4C81', linewidth=2)
plt.axvspan(df['Month'][crisis_start], df['Month'][crisis_start + crisis_duration], alpha=0.3, color='red')
plt.xlabel('Month', fontsize=10)
plt.ylabel('Index Value', fontsize=10)
plt.title('Stock Market Index Trend', fontsize=12, fontweight='bold')
plt.grid(True, alpha=0.3)

# زیرنویس نهم: ترکیب نتایج برای سیاست‌گذاران
plt.subplot(3, 3, 9)
summary_data = [len(crisis_months), sum(df['Anomaly_Score']), int(mse), int(regression_model.score(X_test, y_test)*100)]
summary_labels = ['Crisis\nMonths', 'Anomalies\nDetected', 'MSE', 'Model\nR² (%)']
colors = ['#1B263B', '#0F4C81', '#A23B72', '#F18F01']
bars = plt.bar(summary_labels, summary_data, color=colors, alpha=0.85, edgecolor='black', linewidth=0.5)
plt.ylabel('Value', fontsize=10)
plt.title('Executive Summary for Policymakers', fontsize=12, fontweight='bold')
plt.xticks(rotation=0, fontsize=10)  # حذف چرخش برای خوانایی بهتر
plt.grid(axis='y', alpha=0.3)

# افزودن مقادیر روی میله‌ها
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 0.5, '%d' % int(height),
             ha='center', va='bottom', fontweight='bold', fontsize=9)

# تنظیم فاصله‌ها به صورت هوشمند
plt.tight_layout(pad=3.0)  # افزایش padding برای جلوگیری از همپوشانی

# نمایش نهایی
plt.show()

# --------------------------
# گزارش نهایی برای سیاست‌گذاران
print("\n" + "="*70)
print("📊 گزارش نهایی برای سیاست‌گذاران")
print("="*70)
print(f"🔹 سیستم داده‌کاوی مسئولیت‌پذیر بانک مرکزی {sum(anomaly_scores_binary)} ماه بالقوه بحرانی را شناسایی کرده است.")
print(f"🔹 علائم کلیدی بحران شامل کاهش ناگهانی در اعتماد مصرف‌کننده، کاهش درخواست وام مسکن و افزایش برداشت نقدی است.")
print(f"🔹 مدل پیش‌بینی با دقت قابل قبولی (R² = {regression_model.score(X_test, y_test):.3f}) قادر به پیش‌بینی شاخص‌های کلیدی است.")
print(f"🔹 این یافته‌ها به صورت گزارش‌های سطح کلان و بدون اشاره به افراد خاص ارائه شده‌اند.")
print(f"🔹 پیشنهاد: بررسی اقدامات پیشگیرانه مانند تعدیل نرخ بهره یا افزایش تأمین مالی بخش‌های حیاتی.")
print("="*70)
```
