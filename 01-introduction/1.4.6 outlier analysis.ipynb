  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: تشخیص تقلب در استفاده از کارت‌های اعتباری 💳
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    در این مثال، تحلیل داده‌های پرت برای شناسایی استفاده‌های تقلبی از کارت‌های اعتباری استفاده شده است. سیستم با بررسی الگوهای خرید یک حساب، خریدهایی را که از نظر مقدار، مکان، نوع کالا یا فرکانس خرید، انحراف غیرعادی نسبت به رفتار معمول دارند، به عنوان ناهنجاری علامت‌گذاری می‌کند. برای مثال، خریدی به مبلغ ۱۰ برابر میانگین خریدهای قبلی، در یک کشور خارجی و در فاصله زمانی کوتاه با چند خرید دیگر، به‌راحتی توسط مدل پرت‌یابی شناسایی می‌شود. این سیستم می‌تواند به‌صورت خودکار هشدار ایجاد کند و اقدامات امنیتی را فعال کند. این مثال نشان می‌دهد که چگونه داده‌های پرت که اغلب به عنوان نویز کنار گذاشته می‌شوند، می‌توانند حامل دانش بسیار ارزشمندی در کاربردهای حیاتی باشند.
  </p>

</div>

```python
# این کد، یک مثال عملی از **تحلیل داده‌های پرت** برای تشخیص تقلب در کارت‌های اعتباری را نشان می‌دهد
# که در بخش 1.4.6 کتاب معرفی شده است. این تکنیک از مهم‌ترین روش‌های داده‌کاوی برای
# شناسایی رویدادهای نادر و غیرعادی است که می‌توانند بسیار ارزشمند باشند.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report
import seaborn as sns

# تنظیم seed برای تولید خروجی‌های قابل تکرار
np.random.seed(42)

# شبیه‌سازی داده‌های تراکنش‌های کارت اعتباری
n_normal_transactions = 1000
n_fraud_transactions = 50

# ایجاد داده‌های تراکنش‌های عادی
normal_data = {
    'Amount': np.random.lognormal(3, 1, n_normal_transactions),  # توزیع لگ‌نرمال برای مبالغ
    'Time_of_Day': np.random.normal(14, 5, n_normal_transactions),  # زمان خرید (میانگین بعدازظهر)
    'Merchant_Category': np.random.choice(['Grocery', 'Restaurant', 'Gas', 'Retail'], 
                                        n_normal_transactions, p=[0.4, 0.3, 0.2, 0.1]),
    'Transaction_Frequency': np.random.poisson(2, n_normal_transactions),  # فرکانس خرید
    'Location_Score': np.random.beta(2, 5, n_normal_transactions),  # امتیاز مکان (نرمال)
    'Is_Fraud': 0  # برچسب غیرتقلبی
}

df_normal = pd.DataFrame(normal_data)

# ایجاد داده‌های تراکنش‌های تقلبی (داده‌های پرت)
fraud_data = {
    'Amount': np.random.lognormal(5, 1.5, n_fraud_transactions),  # مبالغ بسیار بزرگتر
    'Time_of_Day': np.random.choice([2, 3, 4, 5], n_fraud_transactions),  # خریدهای نیمه‌شب
    'Merchant_Category': np.random.choice(['Electronics', 'Jewelry', 'Online_Gaming'], 
                                        n_fraud_transactions, p=[0.4, 0.3, 0.3]),
    'Transaction_Frequency': np.random.poisson(8, n_fraud_transactions),  # خریدهای پیاپی سریع
    'Location_Score': np.random.beta(8, 2, n_fraud_transactions),  # مکان‌های غیرمعمول
    'Is_Fraud': 1  # برچسب تقلبی
}

df_fraud = pd.DataFrame(fraud_data)

# ترکیب داده‌های عادی و تقلبی
df = pd.concat([df_normal, df_fraud], ignore_index=True)

# اختلال در داده‌ها برای شبیه‌سازی واقعی‌تر
df['Time_of_Day'] = np.clip(df['Time_of_Day'], 0, 23)
df['Location_Score'] = np.clip(df['Location_Score'], 0, 1)

# آماده‌سازی ویژگی‌ها برای مدل‌های تشخیص پرت
feature_columns = ['Amount', 'Time_of_Day', 'Transaction_Frequency', 'Location_Score']
X = df[feature_columns]

# استانداردسازی ویژگی‌ها
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# روش 1: Isolation Forest - یک روش مبتنی بر درخت برای تشخیص پرت
isolation_forest = IsolationForest(n_estimators=100, contamination=0.05, random_state=42)
df['Anomaly_Score_IF'] = isolation_forest.fit_predict(X_scaled)
df['Anomaly_Score_IF'] = df['Anomaly_Score_IF'].map({1: 0, -1: 1})  # 0: نرمال، 1: پرت

# روش 2: Local Outlier Factor - یک روش مبتنی بر چگالی
lof = LocalOutlierFactor(n_neighbors=20, contamination=0.05)
df['Anomaly_Score_LOF'] = lof.fit_predict(X_scaled)
df['Anomaly_Score_LOF'] = df['Anomaly_Score_LOF'].map({1: 0, -1: 1})  # 0: نرمال، 1: پرت

# محاسبه امتیاز پرت ترکیبی
df['Composite_Anomaly_Score'] = (df['Anomaly_Score_IF'] + df['Anomaly_Score_LOF']) / 2

# تحلیل نتایج
print("خلاصه داده‌ها:")
print(f"تعداد کل تراکنش‌ها: {len(df)}")
print(f"تعداد تراکنش‌های تقلبی (واقعی): {df['Is_Fraud'].sum()}")
print(f"نرخ تقلب واقعی: {df['Is_Fraud'].mean():.3f}")

# ارزیابی عملکرد مدل Isolation Forest
print(f"\nنتایج Isolation Forest:")
print(classification_report(df['Is_Fraud'], df['Anomaly_Score_IF'], 
                          target_names=['Normal', 'Fraud']))

# ارزیابی عملکرد مدل Local Outlier Factor
print(f"\nنتایج Local Outlier Factor:")
print(classification_report(df['Is_Fraud'], df['Anomaly_Score_LOF'], 
                          target_names=['Normal', 'Fraud']))

# رسم نتایج
plt.figure(figsize=(18, 6))

# زیرنویس اول: توزیع مبالغ تراکنش
plt.subplot(1, 3, 1)
plt.hist(df[df['Is_Fraud'] == 0]['Amount'], bins=30, alpha=0.7, label='Normal', color='blue')
plt.hist(df[df['Is_Fraud'] == 1]['Amount'], bins=30, alpha=0.7, label='Fraud', color='red')
plt.xlabel('Transaction Amount')
plt.ylabel('Frequency')
plt.title('Distribution of Transaction Amounts')
plt.legend()
plt.grid(True, alpha=0.3)

# زیرنویس دوم: تشخیص پرت با Isolation Forest
plt.subplot(1, 3, 2)
scatter = plt.scatter(df['Amount'], df['Time_of_Day'], 
                     c=df['Anomaly_Score_IF'], cmap='coolwarm', alpha=0.7, s=50)
plt.colorbar(scatter)
plt.xlabel('Transaction Amount')
plt.ylabel('Time of Day (hours)')
plt.title('Anomaly Detection: Isolation Forest')
plt.grid(True, alpha=0.3)

# زیرنویس سوم: تشخیص پرت با Local Outlier Factor
plt.subplot(1, 3, 3)
scatter = plt.scatter(df['Amount'], df['Transaction_Frequency'], 
                     c=df['Anomaly_Score_LOF'], cmap='coolwarm', alpha=0.7, s=50)
plt.colorbar(scatter)
plt.xlabel('Transaction Amount')
plt.ylabel('Transaction Frequency')
plt.title('Anomaly Detection: Local Outlier Factor')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# تحلیل عمیق‌تر تراکنش‌های شناسایی شده به عنوان پرت
print("\nتحلیل تراکنش‌های شناسایی شده به عنوان پرت:")
anomalies_if = df[df['Anomaly_Score_IF'] == 1]
anomalies_lof = df[df['Anomaly_Score_LOF'] == 1]

print(f"تعداد تراکنش‌های شناسایی شده به عنوان پرت توسط Isolation Forest: {len(anomalies_if)}")
print(f"درصد تراکنش‌های تقلبی واقعی که تشخیص داده شده‌اند: {anomalies_if['Is_Fraud'].mean():.3f}")

print(f"\nتعداد تراکنش‌های شناسایی شده به عنوان پرت توسط LOF: {len(anomalies_lof)}")
print(f"درصد تراکنش‌های تقلبی واقعی که تشخیص داده شده‌اند: {anomalies_lof['Is_Fraud'].mean():.3f}")

# نمایش ویژگی‌های متمایز تراکنش‌های تقلبی
print("\nمقایسه میانگین ویژگی‌ها برای تراکنش‌های عادی و تقلبی:")
comparison = df.groupby('Is_Fraud')[feature_columns].mean().round(2)
print(comparison)

# تحلیل خطاها (False Negatives)
false_negatives = df[(df['Is_Fraud'] == 1) & (df['Anomaly_Score_IF'] == 0)]
if len(false_negatives) > 0:
    print(f"\nتعداد تراکنش‌های تقلبی که به اشتباه به عنوان عادی طبقه‌بندی شده‌اند: {len(false_negatives)}")
    print("ویژگی‌های این تراکنش‌ها:")
    print(false_negatives[feature_columns].mean().round(2))
```

در این مثال عملی، فرآیند <b>تحلیل داده‌های پرت</b> به صورت کامل و در یک سناریوی واقعی تشخیص تقلب کارت اعتباری پیاده‌سازی شده است:
<ul style="font-size:16pt;">
<li>ابتدا یک مجموعه داده فرضی از <b>تراکنش‌های کارت اعتباری</b> شبیه‌سازی شده است که شامل دو دسته داده است: تراکنش‌های عادی (نرمال) و تراکنش‌های تقلبی (پرت). تراکنش‌های تقلبی با ویژگی‌هایی مانند مبالغ بسیار بزرگتر، زمان خرید در نیمه‌شب، دسته‌های تجاری غیرمعمول و فرکانس خرید بالا شبیه‌سازی شده‌اند.</li>
<li>در مرحله <b>پیش‌پردازش داده</b>، از <b>StandardScaler</b> برای استانداردسازی ویژگی‌ها استفاده شده است تا همه ویژگی‌ها در مقیاس مشابهی باشند.</li>
<li>دو روش پیشرفته برای تشخیص پرت استفاده شده است:<br>
  <ul>
    <li><b>Isolation Forest</b>: یک روش مبتنی بر درخت که با ایجاد درختان تصادفی، نقاطی را که به راحتی قابل جداسازی هستند به عنوان پرت شناسایی می‌کند.</li>
    <li><b>Local Outlier Factor (LOF)</b>: یک روش مبتنی بر چگالی که نقاط را بر اساس چگالی محلی آن‌ها نسبت به همسایگانشان به عنوان پرت شناسایی می‌کند.</li>
  </ul><br>
</li>
<li>نتایج به صورت بصری نمایش داده شده‌اند:
  <ul style="font-size:16pt;">
    <li>یک <b>نمودار هیستوگرام</b> توزیع مبالغ تراکنش‌ها را برای تراکنش‌های عادی و تقلبی نشان می‌دهد.</li>
    <li>دو <b>نمودار پراکندگی</b> نتایج تشخیص پرت توسط Isolation Forest و Local Outlier Factor را نشان می‌دهند که در آن نقاط قرمز (پرت) به وضوح از خوشه اصلی داده‌ها جدا شده‌اند.</li>
  </ul>
</li>
<li>عملکرد مدل‌ها با استفاده از <b>گزارش طبقه‌بندی</b> ارزیابی شده است که شامل معیارهای دقت، یادآوری و F1-Score برای هر دو کلاس است.</li>
<li>تحلیل عمیق‌تری از تراکنش‌های شناسایی شده به عنوان پرت انجام شده است که نشان می‌دهد چند درصد از تراکنش‌های تقلبی واقعی توسط مدل‌ها تشخیص داده شده‌اند.</li>
<li>در نهایت، مقایسه میانگین ویژگی‌ها برای تراکنش‌های عادی و تقلبی نشان می‌دهد که چگونه ویژگی‌های مختلف می‌توانند به تشخیص تقلب کمک کنند.</li>
</ul>
<ul style="font-size:16pt;">این مثال به‌وضوح نشان می‌دهد که چگونه <b>تحلیل داده‌های پرت</b> می‌تواند در کاربردهای حیاتی مانند تشخیص تقلب، ارزش بسیار بالاتری نسبت به تحلیل داده‌های عادی داشته باشد. برخلاف رویکردهای سنتی که داده‌های پرت را به عنوان نویز حذف می‌کنند، این مثال نشان می‌دهد که چگونه این داده‌های غیرعادی می‌توانند حامل اطلاعات بسیار ارزشمندی باشند. سیستم‌های تشخیص تقلب می‌توانند به‌صورت خودکار هشدار ایجاد کنند و اقدامات امنیتی را فعال کنند، که این امر می‌تواند از خسارات مالی قابل توجهی جلوگیری کند.</ul>
