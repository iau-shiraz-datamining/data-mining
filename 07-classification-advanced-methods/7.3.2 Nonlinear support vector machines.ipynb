<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: تفکیک الگوهای غیرخطی سیگنال‌های صوتی و ژانر موسیقی در «اسپاتیفای» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در سیستم تحلیل محتوای صوتی <b>اسپاتیفای</b>، قطعات موسیقی بر مبنای شاخص‌های غیرخطی (نظیر انرژی، آکوستیک بودن، رقص‌پذیری و تمپو) مدل‌سازی می‌شوند. مرز میان پلی‌لیست‌های متمرکز (مانند موسیقی مطالعه یا مدیتیشن) و قطعات پرانرژی (مانند پاپ و الکترونیک) در فضای ویژگی‌های دوبعدی به شکل حلقه‌های تودرتو یا الگوهای غیرخطی پیچیده ظاهر می‌شود که با هیچ خط یا ابرصفحه مستقیمی تفکیک‌پذیر نیست. مهندسان اسپاتیفای با استفاده از <b>ماشین بردار پشتیبان غیرخطی (Nonlinear SVM)</b> و به‌کارگیری «ترفند هسته» (Kernel Trick)، بردارها را به صورت ضمنی به فضایی با ابعاد بی‌نهایت منتقل کرده و مرزهای تصمیم منحنی‌شکلِ بهینه را استخراج می‌نمایند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی ماشین بردار پشتیبان غیرخطی و هسته RBF:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. تابع هسته گوسی یا پایه شعاعی (Radial Basis Function - RBF):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$K(\mathbf{x}_i, \mathbf{x}_j) = \exp\left( -\gamma \|\mathbf{x}_i - \mathbf{x}_j\|_2^2 \right) = \langle \phi(\mathbf{x}_i), \phi(\mathbf{x}_j) \rangle, \quad \gamma > 0$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. فرم دوگان بهینه‌سازی لاگرانژ با ترفند هسته (Dual Formulation):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\max_{\boldsymbol{\alpha}} \sum_{i=1}^{N} \alpha_i - \frac{1}{2} \sum_{i=1}^{N} \sum_{j=1}^{N} \alpha_i \alpha_j y_i y_j K(\mathbf{x}_i, \mathbf{x}_j) \quad \text{subject to} \quad 0 \le \alpha_i \le C, \quad \sum_{i=1}^{N} \alpha_i y_i = 0$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. تابع مرز تصمیم‌گیری نهایی غیرخطی بر مبنای ضرایب دوگان ($\alpha_i$):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$f(\mathbf{x}) = \text{sign}\left( \sum_{i \in \text{SV}} \alpha_i y_i K(\mathbf{x}_i, \mathbf{x}) + b \right)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. محاسبه عبارت بایاس ($b$) با استفاده از بردارهای پشتیبان روی حاشیه ($0 < \alpha_k < C$):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$b = \frac{1}{|\mathcal{S}|} \sum_{k \in \mathcal{S}} \left( y_k - \sum_{i \in \text{SV}} \alpha_i y_i K(\mathbf{x}_i, \mathbf{x}_k) \right)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: اسپاتیفای بدون نیاز به نگاشت صریح و پرهزینه به ابعاد بالا، مرزهای غیرخطی بسیار دقیقی برای دسته‌بندی خودکار حالات خلقی (Moods) و ژانرهای ترکیبی تولید می‌کند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: کشف مرزهای تصمیم‌گیری منعطف و غیرخطی با استفاده از ضرب داخلی در فضای هیلبرت
    </div>
  </p>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "تفکیک الگوهای غیرخطی موسیقی در «اسپاتیفای» (Nonlinear RBF SVM From Scratch) 🎵", 
    "main_title"
)

display_academic_text(
    "در سیستم تحلیل محتوای اسپاتیفای، مرز میان سبک‌های مختلف موسیقی در فضای ویژگی‌ها (مثل تمپو و انرژی) "
    "معمولاً به شکل حلقه‌های تودرتو یا الگوهای پیچیده است که با هیچ خط مستقیمی تفکیک‌پذیر نیست. ما با استفاده "
    "از <b>ترفند هسته (Kernel Trick)</b> و الگوریتم پیچیده <b>SMO (Sequential Minimal Optimization)</b> "
    "که از صفر نوشته شده است، داده‌ها را به فضای ابعاد بی‌نهایت (فضای هیلبرت) برده و مرزهای منحنی‌شکل استخراج می‌کنیم.", 
    "paragraph"
)

display_academic_text("فرمول‌های پایه ماشین بردار پشتیبان غیرخطی:", "sub_title")

display_academic_text("۱. تابع هسته گوسی یا پایه شعاعی (Radial Basis Function - RBF):", "paragraph")
display_academic_text(r"$$K(x_i, x_j) = \exp\left(-\gamma \|x_i - x_j\|_2^2\right) = \langle \phi(x_i), \phi(x_j) \rangle$$", "paragraph")

display_academic_text("۲. فرم دوگان بهینه‌سازی لاگرانژ برای یافتن آلفاها (Dual Formulation):", "paragraph")
display_academic_text(r"$$\max_{\boldsymbol{\alpha}} \sum_{i=1}^{N} \alpha_i - \frac{1}{2} \sum_{i=1}^{N} \sum_{j=1}^{N} \alpha_i \alpha_j y_i y_j K(\mathbf{x}_i, \mathbf{x}_j)$$", "paragraph")

display_academic_text("۳. تابع مرز تصمیم‌گیری نهایی (Nonlinear Boundary) بر مبنای ضرب نقطه‌ای کرنل با بردارهای پشتیبان:", "paragraph")
display_academic_text(r"$$f(x) = \text{sign}\left(\sum_{i \in \text{SV}} \alpha_i y_i K(x_i, x) + b\right)$$", "paragraph")

# ==============================================================================
# 2. SPOTIFY AUDIO FEATURES DATASET (CONCENTRIC NON-LINEAR PATTERNS)
# ==============================================================================
np.random.seed(42)
n_samples_per_class = 75

# Class +1: High Danceability / Electronic Tracks (Outer Ring of the Spotify Feature Space)
r_outer = np.random.uniform(3.0, 4.8, n_samples_per_class)
theta_outer = np.random.uniform(0, 2 * np.pi, n_samples_per_class)
X_outer = np.column_stack([r_outer * np.cos(theta_outer), r_outer * np.sin(theta_outer)])
y_outer = np.ones(n_samples_per_class)

# Class -1: Ambient / Deep Focus Acoustic Tracks (Inner Core of the Spotify Feature Space)
r_inner = np.random.uniform(0.1, 1.8, n_samples_per_class)
theta_inner = np.random.uniform(0, 2 * np.pi, n_samples_per_class)
X_inner = np.column_stack([r_inner * np.cos(theta_inner), r_inner * np.sin(theta_inner)])
y_inner = -np.ones(n_samples_per_class)

# Combine and Shuffle
X = np.vstack((X_outer, X_inner))
y = np.hstack((y_outer, y_inner))

perm = np.random.permutation(len(y))
X = X[perm]
y = y[perm]

# ==============================================================================
# 3. NONLINEAR RBF KERNEL SVM (SIMPLIFIED SMO ALGORITHM) FROM SCRATCH
# ==============================================================================
class NonLinearKernelSVMScratch:
    """Nonlinear SVM with RBF Kernel trained via Simplified Sequential Minimal Optimization (SMO)."""

    def __init__(self, C=1.0, gamma=0.35, max_passes=50, tol=1e-3):
        self.C = C
        self.gamma = gamma
        self.max_passes = max_passes
        self.tol = tol
        self.alphas = None
        self.b = 0.0
        self.X_train = None
        self.y_train = None
        self.support_vectors_idx = None
        self.gram_matrix = None

    def _rbf_kernel(self, x1, x2):
        """
        Computes RBF Kernel between two vector matrices efficiently.
        K(x, y) = exp(-gamma * ||x - y||^2)
        """
        if x1.ndim == 1:
            x1 = x1.reshape(1, -1)
        if x2.ndim == 1:
            x2 = x2.reshape(1, -1)
        
        # Fast Vectorized Euclidean squared distance: ||x1 - x2||^2
        dist_sq = np.sum(x1**2, axis=1)[:, np.newaxis] + np.sum(x2**2, axis=1) - 2 * np.dot(x1, x2.T)
        return np.exp(-self.gamma * dist_sq)

    def fit(self, X, y):
        n_samples, _ = X.shape
        self.X_train = X
        self.y_train = y
        self.alphas = np.zeros(n_samples) # Lagrange Multipliers
        self.b = 0.0

        # Precompute the Kernel Gram Matrix: K_ij = K(x_i, x_j) for massive speedup
        self.gram_matrix = self._rbf_kernel(X, X)

        passes = 0
        # SMO Algorithm Loop
        while passes < self.max_passes:
            num_changed_alphas = 0

            for i in range(n_samples):
                # Calculate prediction error for sample i: E_i = f(x_i) - y_i
                f_xi = np.sum(self.alphas * self.y_train * self.gram_matrix[:, i]) + self.b
                E_i = f_xi - self.y_train[i]

                # Check if this alpha violates the Karush-Kuhn-Tucker (KKT) conditions
                if (self.y_train[i] * E_i < -self.tol and self.alphas[i] < self.C) or \
                   (self.y_train[i] * E_i > self.tol and self.alphas[i] > 0):
                    
                    # Select random index j to pair with i for optimization
                    j = np.random.choice([idx for idx in range(n_samples) if idx != i])

                    # Prediction error for sample j
                    f_xj = np.sum(self.alphas * self.y_train * self.gram_matrix[:, j]) + self.b
                    E_j = f_xj - self.y_train[j]

                    alpha_i_old = self.alphas[i]
                    alpha_j_old = self.alphas[j]

                    # Compute lower (L) and upper (H) clipping bounds for alpha_j
                    if self.y_train[i] != self.y_train[j]:
                        L = max(0.0, self.alphas[j] - self.alphas[i])
                        H = min(self.C, self.C + self.alphas[j] - self.alphas[i])
                    else:
                        L = max(0.0, self.alphas[i] + self.alphas[j] - self.C)
                        H = min(self.C, self.alphas[i] + self.alphas[j])

                    if L == H:
                        continue

                    # Compute eta: 2*K(x_i, x_j) - K(x_i, x_i) - K(x_j, x_j)
                    eta = 2.0 * self.gram_matrix[i, j] - self.gram_matrix[i, i] - self.gram_matrix[j, j]
                    if eta >= 0:
                        continue

                    # Update alpha_j and clip it inside the [L, H] box constraint
                    self.alphas[j] -= (self.y_train[j] * (E_i - E_j)) / eta
                    self.alphas[j] = np.clip(self.alphas[j], L, H)

                    # If alpha_j didn't change enough, move to the next pair
                    if abs(self.alphas[j] - alpha_j_old) < 1e-5:
                        continue

                    # Update alpha_i based on alpha_j's change (maintaining the sum constraint)
                    self.alphas[i] += self.y_train[i] * self.y_train[j] * (alpha_j_old - self.alphas[j])

                    # Update the threshold bias b
                    b1 = self.b - E_i - self.y_train[i] * (self.alphas[i] - alpha_i_old) * self.gram_matrix[i, i] \
                         - self.y_train[j] * (self.alphas[j] - alpha_j_old) * self.gram_matrix[i, j]
                    
                    b2 = self.b - E_j - self.y_train[i] * (self.alphas[i] - alpha_i_old) * self.gram_matrix[i, j] \
                         - self.y_train[j] * (self.alphas[j] - alpha_j_old) * self.gram_matrix[j, j]

                    if 0 < self.alphas[i] < self.C:
                        self.b = b1
                    elif 0 < self.alphas[j] < self.C:
                        self.b = b2
                    else:
                        self.b = (b1 + b2) / 2.0

                    num_changed_alphas += 1

            # If no alphas violated KKT in a full pass, we are closer to convergence
            if num_changed_alphas == 0:
                passes += 1
            else:
                passes = 0

        # Extract Support Vector indices (Points where alpha > 0)
        self.support_vectors_idx = np.where(self.alphas > 1e-4)[0]
        return self

    def decision_function(self, X_new):
        """Computes nonlinear decision scores for arbitrary points in the original space."""
        K_new = self._rbf_kernel(X_new, self.X_train)
        return np.dot(K_new, self.alphas * self.y_train) + self.b

    def predict(self, X_new):
        return np.sign(self.decision_function(X_new))


# Train Nonlinear RBF Kernel SVM from scratch
rbf_svm = NonLinearKernelSVMScratch(C=2.0, gamma=0.45, max_passes=30)
rbf_svm.fit(X, y)

# ==============================================================================
# 4. VISUALIZATIONS AND NON-LINEAR DECISION SURFACE (ACADEMIC BLUE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

#  CHART 1: RBF Kernel Decision Surface and Non-Linear Contours 
grid_range_x = np.linspace(X[:, 0].min() - 0.8, X[:, 0].max() + 0.8, 300)
grid_range_y = np.linspace(X[:, 1].min() - 0.8, X[:, 1].max() + 0.8, 300)
xx, yy = np.meshgrid(grid_range_x, grid_range_y)
grid_pts = np.c_[xx.ravel(), yy.ravel()]

# Calculate decision scores for the entire space
Z = rbf_svm.decision_function(grid_pts).reshape(xx.shape)

# Filled confidence contours (Blue = Electronic, Light Teal = Acoustic)
axes[0].contourf(xx, yy, Z, levels=np.linspace(Z.min(), Z.max(), 30), cmap='Blues', alpha=0.35)

# Margin and Decision Boundaries (Teal and Navy)
axes[0].contour(
    xx, yy, Z, levels=[-1.0, 0.0, 1.0], linestyles=['--', '-', '--'],
    colors=['#16a085', '#2c3e50', '#2980b9'], linewidths=[2, 3, 2]
)

# Plot Data Points
axes[0].scatter(
    X[y == 1, 0], X[y == 1, 1], 
    color='#2980b9', s=60, edgecolors='black', label='Electronic / High Energy (+1)', zorder=3
)
axes[0].scatter(
    X[y == -1, 0], X[y == -1, 1], 
    color='#16a085', s=60, edgecolors='black', label='Ambient / Acoustic (-1)', zorder=3
)

# Highlight Support Vectors (Red Circles)
svs = rbf_svm.support_vectors_idx
axes[0].scatter(
    X[svs, 0], X[svs, 1], 
    s=170, facecolors='none', edgecolors='#e74c3c', linewidths=2.0, 
    label=f'Support Vectors (Count: {len(svs)})', zorder=4
)

axes[0].set_title(f'1. Spotify Nonlinear RBF SVM Decision Surface ($\gamma=0.45, C=2.0$)', fontsize=13, fontweight='bold', pad=12)
axes[0].set_xlabel('Acoustic Resonance Component ($x_1$)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Rhythm Danceability Component ($x_2$)', fontsize=12, fontweight='bold')
axes[0].legend(loc='upper right', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
axes[0].grid(True, linestyle='--', alpha=0.6)


#  CHART 2: Lagrange Multipliers (Alpha) Spectrum and Sparsity Analysis 
sorted_alpha_indices = np.argsort(rbf_svm.alphas)[::-1]
sorted_alphas = rbf_svm.alphas[sorted_alpha_indices]
sv_threshold = 1e-4

# Color active Support Vectors Dark Navy, the rest (Zeros) light gray
bar_colors = ['#2c3e50' if a > sv_threshold else '#bdc3c7' for a in sorted_alphas]

axes[1].bar(range(len(sorted_alphas)), sorted_alphas, color=bar_colors, width=0.8, edgecolor='black', linewidth=0.5)

axes[1].axhline(rbf_svm.C, color='#e74c3c', linestyle='--', linewidth=2, label=f'Upper Box Constraint ($C = {rbf_svm.C}$)')

axes[1].axhline(sv_threshold, color='black', linestyle=':', linewidth=1.5, label=r'Support Vector Threshold (Non-Zero $\alpha$)')

axes[1].set_title(r'2. Dual Multipliers Sparsity Spectrum ($\alpha_i$ Distribution from SMO)', fontsize=13, fontweight='bold', pad=12)
axes[1].set_xlabel('Training Samples (Ranked by Dual Weight)', fontsize=12, fontweight='bold')
axes[1].set_ylabel(r'Lagrange Multiplier Weight ($\alpha_i$)', fontsize=12, fontweight='bold')
axes[1].set_xlim(-2, 50) # Zoomed in to see the active vectors
axes[1].legend(loc='upper right', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
axes[1].grid(axis='y', linestyle='--', alpha=0.7)

# Annotate sparsity percentage
axes[1].annotate(
    f'Active Support Vectors: {len(svs)} / {len(y)}\n({(len(svs)/len(y))*100:.1f}% Sparsity Rate)',
    xy=(5, sorted_alphas[5]),
    xytext=(15, rbf_svm.C * 0.6),
    arrowprops=dict(arrowstyle="->", color='#e74c3c', lw=2),
    bbox=dict(boxstyle='round,pad=0.5', facecolor='#ecf0f1', edgecolor='#c0392b', alpha=0.9),
    fontsize=11, fontweight='bold', color='#2c3e50'
)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (رسم مرزهای پیچیده بدون تغییر ابعاد):</b> در گراف ۱ می‌بینیم که گروه موسیقیِ اکوستیک "
    r"(نقاط فیروزه‌ای) کاملاً توسط گروه موسیقی الکترونیک (نقاط آبی) محاصره شده است! اگر از رگرسیون لجستیک یا یک SVM خطیِ معمولی "
    r"استفاده می‌کردیم، سیستم هرگز نمی‌توانست مرزی برای جداسازی این دو سبک موسیقی پیدا کند. اما به لطف <b>تابع کرنل RBF</b> "
    r"الگوریتم ما توانست فاصله نقطه‌ای داده‌ها را محاسبه کرده و در فضایی با بی‌نهایت بُعد، مرز تصمیمی رسم کند که در این فضای "
    r"دوبعدی، شبیه به یک دایره (حلقه مرزی تیره) دیده می‌شود! همچنین گراف ۲ قدرت «پراکنده‌سازی» (Sparsity) مدل را نشان می‌دهد: "
    r"بیش از ۷۰٪ از ترانه‌ها دارای وزنِ صفر ($\alpha = 0$) هستند (میله‌های خاکستری زیر خط)، به این معنا که مدل اسپاتیفای توانسته "
    r"این مرز دایره‌ایِ پیچیده را تنها با نگاه کردن به تعداد معدودی از آهنگ‌های مرزی (بردارهای پشتیبان با آلفای بزرگ) ترسیم کند."
)
display_academic_text(insight_conclusion, "insight")
```

<li>مجموعه داده‌ای متشکل از ۱۵۰ قطعه صوتی در اسپاتیفای با الگوی توزیع غیرخطی (حلقه‌های هم‌مرکز شامل موسیقی‌های پرانرژی در حلقه بیرونی و موسیقی‌های آکوستیک آرام در هسته مرکزی) شبیه‌سازی گردید که تفکیک خطی آن‌ها غیرممکن است.</li>
<li>پیاده‌سازی کامل ماشین بردار پشتیبان غیرخطی با هسته RBF و الگوریتم بهینه‌سازی دوگان SMO از صفر (From Scratch):
  <ul>
    <li><b>ماتریس گِرم هسته RBF:</b> در متد <code>_rbf_kernel</code>، فاصله اقلیدسی دوبه‌دوی نمونه‌ها محاسبه شده و ماتریس شباهت هسته گوسی با پارامتر پهنای باند ($\gamma = 0.45$) تشکیل شد.</li>
    <li><b>الگوریتم بهینه‌سازی حداقل‌سازی ترتیبی (SMO):</b> در متد <code>fit</code>، در هر تکرار دو ضریب لاگرانژ ($\alpha_i, \alpha_j$) به صورت تحلیلی و مقید به بازه $[L, H]$ به‌روزرسانی شده و نقض شرایط بهینگی KKT پایش گردید.</li>
    <li><b>به‌روزرسانی تحلیلی بایاس ($b$) و تابع پیش‌بینی:</b> در متد <code>decision_function</code>، محاسبه پیش‌بینی نقاط جدید با ترکیب هسته‌ای نمونه‌ها و ضرایب دوگان استخراج‌شده انجام شد.</li>
  </ul>
</li>
<li>تفسیر خروجی و نمودارهای تحلیلی:
  <ul>
    <li><b>نمودار رویه تصمیم غیرخطی (سمت چپ):</b> نشان می‌دهد که چگونه هسته RBF بدون نیاز به نگاشت صریح، یک مرز تصمیم کاملاً حلقوی و منطبق بر هندسه داده‌ها (خط پررنگ میانی) همراه با مرزهای حاشیه تشکیل داده است؛ نقاط مشخص‌شده با دایره‌های قرمز، تنها نقاط فعال (Support Vectors) هستند که شکل این مرز را کنترل می‌کنند.</li>
    <li><b>نمودار طیف ضرایب دوگان و پراکندگی (سمت راست):</b> اصل «پراکندگی» (Sparsity) در SVM را اثبات می‌کند؛ بخش عمده داده‌های درون خوشه‌ها دارای ضریب $\alpha_i = 0$ هستند و تنها زیرمجموعه محدودی از داده‌های بحرانی نزدیک مرز، مقدار ضریب مثبت دریافت کرده‌اند که این امر سرعت طبقه‌بندی بلادرنگ را تضمین می‌نماید.</li>
  </ul>
</li>
