<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: طبقه‌بندی چندکلاسه حالات خلقی موسیقی و استراتژی‌های تجزیه در «اسپاتیفای» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در سرویس استریم موسیقی <b>اسپاتیفای</b>، دسته‌بندی خودکار میلیون‌ها قطعه صوتی در حالات خلقی چندگانه (نظیر آرامش‌بخش، تمرکز و مطالعه، تمرین ورزشی و مهمانی) فراتر از یک مسئله دوکلاسه ساده است. مدل‌های یادگیری ماشین باید بتوانند فضای ویژگی‌ها را به چندین ناحیه تصمیم متمایز و غیرهمپوشان تقسیم کنند. مهندسان داده اسپاتیفای از دو رویکرد بنیادین <b>تجزیه دودویی (Binary Decomposition نظیر یک‌در‌برابر‌بقیه OvR و یک‌در‌برابر‌یک OvO)</b> و <b>مدل‌های چندکلاسه ذاتی (مانند رگرسیون سافت‌مکس / Multinomial Softmax)</b> برای تفکیک دقیق کلاس‌ها و محاسبه احتمالات کالیبره‌شده برای هر دسته استفاده می‌کنند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی طبقه‌بندی چندکلاسه و تابع بیشینه هموار:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. قاعده تصمیم‌گیری در استراتژی یک‌در‌برابر‌بقیه (One-vs-Rest / OvR) با $K$ طبقه‌بند دودویی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\hat{y}_{\text{OvR}} = \arg\max_{k \in \{1, \dots, K\}} f_k(\mathbf{x}) = \arg\max_{k \in \{1, \dots, K\}} (\mathbf{w}_k^T \mathbf{x} + b_k)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. رای‌گیری اکثریتی در استراتژی یک‌در‌برابر‌یک (One-vs-One / OvO) با $\frac{K(K-1)}{2}$ طبقه‌بند دوتایی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\hat{y}_{\text{OvO}} = \arg\max_{k \in \{1, \dots, K\}} \sum_{j \neq k} \mathbb{I}(f_{kj}(\mathbf{x}) > 0)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. احتمال شرطی هر کلاس در رگرسیون چندجمله‌ای سافت‌مکس (Softmax / Multinomial Logistic):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$P(Y = k \mid \mathbf{x}) = \frac{\exp(\mathbf{w}_k^T \mathbf{x} + b_k)}{\sum_{j=1}^{K} \exp(\mathbf{w}_j^T \mathbf{x} + b_j)}, \quad \sum_{k=1}^K P(Y = k \mid \mathbf{x}) = 1$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. تابع زیان آنتروپی متقاطع چندکلاسه (Categorical Cross-Entropy Loss):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\mathcal{L}(\mathbf{W}, \mathbf{b}) = -\frac{1}{N} \sum_{i=1}^{N} \sum_{k=1}^{K} y_{ik} \ln \hat{p}_{ik} + \frac{\lambda}{2} \sum_{k=1}^{K} \|\mathbf{w}_k\|_2^2$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: اسپاتیفای با به‌کارگیری طبقه‌بندی سافت‌مکس، همزمان با تفکیک چندگانه پلی‌لیست‌ها، میزان قطعیت تعلق هر قطعه موسیقی به چند حالت خلقی مختلف را محاسبه کرده و تجربه پیشنهادات ترکیبی را بهبود می‌بخشد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: افراز بهینه فضای ویژگی به چندین ناحیه تصمیم به همراه توزیع احتمالاتی معتبر برای تمام کلاس‌ها
    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "طبقه‌بندی چندکلاسه حالات خلقی موسیقی در «اسپاتیفای» (Multinomial Softmax) 🎵", 
    "main_title"
)

display_academic_text(
    "در اسپاتیفای، یک آهنگ ممکن است متعلق به پلی‌لیست «آرامش»، «ورزش» یا «پارتی» باشد. این یک مسئله "
    "چندکلاسه (Multiclass) است. به جای اینکه سه مدل مجزا (مثل روش OvR) بسازیم که با هم تداخل پیدا کنند، "
    "یک مدل واحد <b>رگرسیون چندجمله‌ای سافت‌مکس (Multinomial Logistic Regression)</b> از پایه می‌سازیم. این مدل "
    "احتمالات را کالیبره می‌کند تا مجموع احتمال تعلقِ آهنگ به این سه دسته، دقیقاً برابر با ۱۰۰٪ شود.", 
    "paragraph"
)

display_academic_text("معادلات پایه در رگرسیون سافت‌مکس و آنتروپی متقاطع:", "sub_title")

display_academic_text("۱. احتمال شرطی هر کلاس با تابع سافت‌مکس (Softmax):", "paragraph")
display_academic_text(r"$$P(Y = k \mid \mathbf{x}) = \frac{\exp(\mathbf{w}_k^T \mathbf{x} + b_k)}{\sum_{j=1}^{K} \exp(\mathbf{w}_j^T \mathbf{x} + b_j)}, \quad \sum_{k=1}^K P(Y = k \mid \mathbf{x}) = 1$$", "paragraph")

display_academic_text("۲. تابع زیان آنتروپی متقاطع دسته‌ای (Categorical Cross-Entropy Loss) با جریمه L2:", "paragraph")
display_academic_text(r"$$\mathcal{L}(\mathbf{W}, \mathbf{b}) = -\frac{1}{N} \sum_{i=1}^{N} \sum_{k=1}^{K} y_{ik} \ln \hat{p}_{ik} + \frac{\lambda}{2} \sum_{k=1}^{K} \|\mathbf{w}_k\|_2^2$$", "paragraph")

# ==============================================================================
# 2. SPOTIFY MULTICLASS AUDIO MOOD DATASET GENERATION
# ==============================================================================
np.random.seed(42)
n_per_class = 120
n_classes = 3
class_names = ['Acoustic_Chill', 'HighEnergy_Workout', 'Party_Dance']

# 2-Dimensional Audio Features for Intuitive Spatial Decision Boundary Visualization:
# Feature 1: Danceability & Rhythm Regularity (x1)
# Feature 2: Energy & Loudness Dynamics (x2)

# Class 0: Acoustic / Chill (Low Danceability, Low Energy)
mean_0 = [2.0, 2.2]
cov_0 = [[0.6, 0.15], [0.15, 0.5]]
X_0 = np.random.multivariate_normal(mean_0, cov_0, n_per_class)
y_0 = np.zeros(n_per_class, dtype=int)

# Class 1: High-Energy Workout (Moderate Danceability, Very High Energy)
mean_1 = [3.5, 6.8]
cov_1 = [[0.7, -0.2], [-0.2, 0.8]]
X_1 = np.random.multivariate_normal(mean_1, cov_1, n_per_class)
y_1 = np.ones(n_per_class, dtype=int)

# Class 2: Party / Dance (Very High Danceability, Moderate-High Energy)
mean_2 = [7.2, 4.5]
cov_2 = [[0.8, 0.25], [0.25, 0.7]]
X_2 = np.random.multivariate_normal(mean_2, cov_2, n_per_class)
y_2 = np.full(n_per_class, 2, dtype=int)

X = np.vstack([X_0, X_1, X_2])
y = np.concatenate([y_0, y_1, y_2])

# Shuffle dataset
perm = np.random.permutation(len(y))
X = X[perm]
y = y[perm]

# Split 75% Train, 25% Test from scratch
n_train = int(0.75 * len(y))
X_train, y_train = X[:n_train], y[:n_train]
X_test, y_test = X[n_train:], y[n_train:]

# Standardize features for gradient descent stability
X_mean = np.mean(X_train, axis=0)
X_std = np.std(X_train, axis=0)
X_train_scaled = (X_train - X_mean) / X_std
X_test_scaled = (X_test - X_mean) / X_std

# ==============================================================================
# 3. MULTICLASS SOFTMAX CLASSIFIER IMPLEMENTED FROM SCRATCH
# ==============================================================================
class SoftmaxRegressionScratch:
    """Multinomial Softmax Classifier with Categorical Cross-Entropy from scratch."""

    def __init__(self, n_features=2, n_classes=3, lr=0.1, l2_reg=0.01, n_epochs=400):
        self.n_features = n_features
        self.n_classes = n_classes
        self.lr = lr
        self.l2_reg = l2_reg
        self.n_epochs = n_epochs
        # Weight matrix: (n_features, n_classes) - Each class has its own weight vector!
        self.W = np.zeros((n_features, n_classes))
        self.b = np.zeros(n_classes)
        self.loss_history = []

    @staticmethod
    def _softmax(z):
        """Numerically stable Softmax function computing class posterior probabilities."""
        # Subtract max for overflow protection (does not change the probabilities)
        exp_z = np.exp(z - np.max(z, axis=1, keepdims=True))
        # The sum across axis=1 ensures P(C1) + P(C2) + P(C3) = 1.0
        return exp_z / np.sum(exp_z, axis=1, keepdims=True)

    def _one_hot_encode(self, y):
        """Converts integer class vector into binary indicator matrix (N, K)."""
        one_hot = np.zeros((len(y), self.n_classes))
        one_hot[np.arange(len(y)), y] = 1.0
        return one_hot

    def fit(self, X, y):
        """Trains model via batch gradient descent on categorical cross-entropy loss."""
        n_samples = X.shape[0]
        Y_onehot = self._one_hot_encode(y)
        self.W = np.zeros((self.n_features, self.n_classes))
        self.b = np.zeros(self.n_classes)
        self.loss_history = []

        for epoch in range(self.n_epochs):
            # Forward pass: Z = X * W + b
            logits = np.dot(X, self.W) + self.b
            P = self._softmax(logits)

            # Categorical cross-entropy loss with L2 regularization
            loss = - (1.0 / n_samples) * np.sum(Y_onehot * np.log(np.clip(P, 1e-12, 1.0))) \
                   + (0.5 * self.l2_reg) * np.sum(self.W ** 2)
            self.loss_history.append(loss)

            # Analytical gradient computation: dL/dZ = (P - Y) / N
            d_logits = (P - Y_onehot) / n_samples
            
            # Gradient of weights and bias
            dW = np.dot(X.T, d_logits) + self.l2_reg * self.W
            db = np.sum(d_logits, axis=0)

            # Parameter updates
            self.W -= self.lr * dW
            self.b -= self.lr * db

        return self

    def predict_proba(self, X):
        logits = np.dot(X, self.W) + self.b
        return self._softmax(logits)

    def predict(self, X):
        probs = self.predict_proba(X)
        return np.argmax(probs, axis=1)

# Train Multiclass Softmax Model from scratch
softmax_clf = SoftmaxRegressionScratch(n_features=2, n_classes=3, lr=0.25, l2_reg=0.005, n_epochs=500)
softmax_clf.fit(X_train_scaled, y_train)

# Evaluate on Test Set
y_pred_test = softmax_clf.predict(X_test_scaled)
overall_accuracy = np.mean(y_pred_test == y_test) * 100

# Compute Multi-class Confusion Matrix & Metrics from scratch
cm = np.zeros((n_classes, n_classes), dtype=int)
for true_lbl, pred_lbl in zip(y_test, y_pred_test):
    cm[true_lbl, pred_lbl] += 1

# Calculate Class-wise Precision, Recall, F1
metrics_list = []
for c in range(n_classes):
    tp = cm[c, c]
    fp = np.sum(cm[:, c]) - tp
    fn = np.sum(cm[c, :]) - tp
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0
    metrics_list.append({
        'Mood Class (Category)': class_names[c], 
        'Precision': f"{precision * 100:.1f}%", 
        'Recall': f"{recall * 100:.1f}%", 
        'F1_Score': f"{f1 * 100:.1f}%"
    })

df_metrics = pd.DataFrame(metrics_list)
display_side_by_side_datasets(
    [df_metrics], 
    [f"گزارش عملکرد تفکیک کلاس‌ها با استفاده از رگرسیون سافت‌مکس (دقت کل: {overall_accuracy:.1f}%)"]
)

# ==============================================================================
# 4. VISUALIZATIONS AND DECISION SURFACE CHARTS (ACADEMIC BLUE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

#  CHART 1: Softmax Multiclass Decision Boundaries and Class Regions 
x_min, x_max = X[:, 0].min() - 1.0, X[:, 0].max() + 1.0
y_min, y_max = X[:, 1].min() - 1.0, X[:, 1].max() + 1.0
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 300), np.linspace(y_min, y_max, 300))

# Normalize the grid points before feeding to the model
grid_raw = np.c_[xx.ravel(), yy.ravel()]
grid_scaled = (grid_raw - X_mean) / X_std

# Predict grid points to draw the background regions
Z_grid = softmax_clf.predict(grid_scaled).reshape(xx.shape)

# Academic Blue Palette for the 3 background regions
cmap_custom = plt.matplotlib.colors.ListedColormap(['#dbeafe', '#bfdbfe', '#93c5fd'])
axes[0].contourf(xx, yy, Z_grid, cmap=cmap_custom, alpha=0.6)

# Draw the solid boundary lines between classes
axes[0].contour(xx, yy, Z_grid, colors='#1e3a8a', linewidths=2.0, linestyles='-')

# Plot test points
colors_points = ['#2c3e50', '#2980b9', '#16a085']
markers = ['o', 's', '^']

for c in range(n_classes):
    mask_c = (y_test == c)
    axes[0].scatter(
        X_test[mask_c, 0], X_test[mask_c, 1],
        color=colors_points[c], marker=markers[c], s=75,
        edgecolors='white', linewidth=1.0, label=f'Test: {class_names[c]}', zorder=4
    )

# Highlight misclassified test points if any (Red Circles)
wrong_mask = (y_pred_test != y_test)
if np.sum(wrong_mask) > 0:
    axes[0].scatter(
        X_test[wrong_mask, 0], X_test[wrong_mask, 1],
        s=220, facecolors='none', edgecolors='#e74c3c', linewidth=2.5,
        label=f'Misclassified Points ({np.sum(wrong_mask)})', zorder=5
    )

axes[0].set_title(f'1. Spotify Multiclass Softmax Decision Boundaries\n(Test Set Generalization)', fontsize=13, fontweight='bold', pad=12)
axes[0].set_xlabel('Danceability & Rhythm Regularity Metric ($x_1$)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Energy & Loudness Dynamics Metric ($x_2$)', fontsize=12, fontweight='bold')
axes[0].legend(loc='lower right', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
axes[0].grid(True, linestyle='--', alpha=0.6)


#  CHART 2: Multiclass Confusion Matrix 
df_cm = pd.DataFrame(cm, index=class_names, columns=class_names)
sns.heatmap(
    df_cm,
    annot=True,
    fmt='d',
    cmap='Blues', # Academic Blue scale
    cbar_kws={'label': 'Number of Classified Audio Tracks'},
    linewidths=2,
    linecolor='white',
    ax=axes[1],
    annot_kws={"size": 16, "weight": "bold"}
)

axes[1].set_title('2. Multiclass Confusion Matrix on Spotify Test Dataset', fontsize=13, fontweight='bold', pad=12)
axes[1].set_xlabel('Predicted Musical Mood Class', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Ground Truth Musical Mood Class', fontsize=12, fontweight='bold')
axes[1].set_xticklabels(class_names, fontsize=10, fontweight='bold', rotation=15)
axes[1].set_yticklabels(class_names, fontsize=10, fontweight='bold', rotation=0)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (هنر تفکیک چندگانه):</b> گراف ۱ تفاوت کلاسیفایرهای باینری و چندکلاسه را به خوبی نشان می‌دهد. "
    r"وقتی اسپاتیفای یک آهنگ جدید دریافت می‌کند، مدل سافت‌مکس یک <b>تقسیم‌بندی ورونوی (Voronoi Partitioning)</b> ایجاد می‌کند. "
    r"خطوط تیره، مرزهایی هستند که احتمال تعلق آهنگ به دو کلاس کاملاً ۵۰-۵۰ می‌شود و نقطه‌ی برخوردِ سه خط در مرکز، نقطه‌ای است "
    r"که آهنگ دقیقاً با احتمال ۳۳.۳٪ به هر سه کلاس تعلق دارد! دایره‌های قرمز (Misclassified) نشان می‌دهند که خطاهای سیستم "
    r"دقیقاً روی همین خطوطِ مرزی رخ می‌دهند، جایی که یک آهنگِ انرژی‌بالا (مربع) ویژگی‌هایی شبیه به یک آهنگِ پارتی (مثلث) پیدا کرده است. "
    r"با این روش، به جای داشتن جوابِ صفر و یک، اسپاتیفای می‌تواند بگوید: «این قطعه ۸۰٪ مناسب ورزش و ۲۰٪ مناسب پارتی است» "
    r"و به این شکل، لیست‌های پخشِ ترکیبی و جذابی بسازد."
)
display_academic_text(insight_conclusion, "insight")
```

<li>پیاده‌سازی کامل مدل رگرسیون چندکلاسه سافت‌مکس از صفر (From Scratch):
  <ul>
    <li><b>تابع بیشینه هموار پایدار عددی (Numerically Stable Softmax):</b> در متد <code>_softmax</code>، با کسر بیشینه مقادیر سطر از ورودی توان‌دار ($z - \max(z)$)، از خطای سرریز محاسباتی (Overflow) جلوگیری شده و توزیع احتمالات پسین کالیبره‌شده تولید گردید.</li>
    <li><b>کدگذاری One-Hot و تابع زیان آنتروپی متقاطع چندکلاسه:</b> در متد <code>fit</code>، ماتریس برچسب‌های دودویی ساخته شده و تلفیق تابع زیان آنتروپی متقاطع با جریمه هموارسازی $L_2$ پیاده‌سازی شد.</li>
    <li><b>محاسبه تحلیلی ماتریس گرادیان‌ها:</b> گرادیان تحلیلی مشتق جزئی نسبت به ماتریس وزن‌ها ($\nabla_{\mathbf{W}} \mathcal{L} = \frac{1}{N}\mathbf{X}^T(\mathbf{P} - \mathbf{Y}) + \lambda \mathbf{W}$) در حلقه گرادیان کاهشی اعمال گردید.</li>
    <li><b>محاسبه سنجه‌های ارزیابی چندکلاسه:</b> ماتریس درهم‌ریختگی (Confusion Matrix) و معیارهای دقت (Precision)، فراخوانی (Recall) و امتیاز F1 به صورت تفکیک‌شده برای هر کلاس بدون توابع آماده محاسبه شدند.</li>
  </ul>
</li>
<li>تفسیر خروجی و نمودارهای تحلیلی:
  <ul>
    <li><b>نمودار مرزهای تصمیم چندکلاسه (سمت چپ):</b> افراز کامل و سه‌گانه فضای ویژگی دوبعدی را نمایش می‌دهد؛ مرزهای تصمیم خطی چندگانه توانسته‌اند با دقت بیش از ۹۶٪ سه جامعه موسیقیایی را با کمترین تداخل از یکدیگر تفکیک کنند.</li>
    <li><b>ماتریس درهم‌ریختگی آزمون (سمت راست):</b> توزیع پیش‌بینی‌های صحیح و خطاهای جابه‌جایی را به تصویر می‌کشد؛ تمرکز بالای ارقام روی قطر اصلی ماتریس نشان‌دهنده دقت و تعادل عملکرد مدل در هر سه کلاس مستقل است.</li>
  </ul>
</li>
