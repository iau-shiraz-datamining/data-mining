<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: تصمیم‌گیری متوالی و بهینه‌سازی سیاست انتخاب ویژگی با یادگیری تقویتی در «نتفلیکس» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در سامانه توصیه‌گر و دسته‌بندی علایق کاربران <b>نتفلیکس</b>، فراخوانی و محاسبه بلادرنگ برخی ویژگی‌ها (نظیر تاریخچه تعاملات گراف، استخراج تعبیه‌های متنی حجیم و تحلیل لاگ‌های لحظه‌ای کلیک) نیازمند صرف منابع پردازشی سنگین و تاخیر شبکه است. بنابراین، سیستم باید بتواند یک سیاست بهینه (Optimal Policy) تدوین کند که به صورت پویا و گام‌به‌گام تصمیم بگیرد کدام ویژگی‌ها ارزش استعلام دارند و چه زمانی باید فرآیند استخراج ویژگی متوقف شده و تصمیم نهایی اتخاذ گردد. مهندسان داده نتفلیکس این مسئله را در قالب <b>فرآیند تصمیم‌گیری مارکوف (MDP)</b> فرمول‌بندی کرده و با استفاده از الگوریتم‌های <b>یادگیری تقویتی (Reinforcement Learning)</b> عاملی هوشمند تربیت می‌کنند که تعادل بهینه میان دقت طبقه‌بندی و هزینه محاسباتی استخراج ویژگی‌ها را برقرار سازد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی یادگیری تقویتی و معادله بلمن در فضای تصمیم‌گیری متوالی:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. فرمول‌بندی فضای حالت ($S$)، فضای عمل ($A$) و تابع پاداش ($R$) انتخاب ویژگی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$s_t \in \{0, 1\}^D, \quad a_t \in \{1, \dots, D\} \cup \{\text{STOP}\}, \quad R(s, a, s') = \begin{cases} \text{Acc}(s') - \lambda \cdot |s'| & \text{if } a = \text{STOP} \\ -c_{\text{step}} & \text{if } a \neq \text{STOP} \end{cases}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. معادله بهینگی بلمن برای تابع ارزش عمل-حالت ($Q^*(s, a)$):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$Q^*(s, a) = R(s, a) + \gamma \max_{a' \in A} Q^*(s', a')$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. قاعده به‌روزرسانی تفاضل زمانی جدول $Q$ (Temporal Difference Q-Learning):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$Q(s_t, a_t) \leftarrow Q(s_t, a_t) + \alpha \left[ r_{t+1} + \gamma \max_{a} Q(s_{t+1}, a) - Q(s_t, a_t) \right]$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. سیاست انتخاب عمل کاوشگرانه $\epsilon$-حریصانه ($\epsilon$-Greedy Policy):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\pi(a \mid s) = \begin{cases} \arg\max_{a'} Q(s, a') & \text{with probability } 1 - \epsilon \\ \text{Random Uniform}(A) & \text{with probability } \epsilon \end{cases}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: نتفلیکس با یادگیری سیاست تصمیم‌گیری بهینه، بدون افت دقت پیش‌بینی، با فعال‌سازی تنها ۳۰٪ از ویژگی‌های کم‌هزینه پاسخگویی به درخواست‌های کاربر را بیش از سه برابر تسریع می‌بخشد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: بهینه‌سازی توأم دقت طبقه‌بندی و هزینه محاسباتی از طریق تعامل عامل با محیط تصمیم‌گیری متوالی
    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "تصمیم‌گیری متوالی و بهینه‌سازی سیاست با یادگیری تقویتی در «نتفلیکس» (RL & Q-Learning) 🎮", 
    "main_title"
)

display_academic_text(
    "در نتفلیکس، محاسبه برخی ویژگی‌ها (مثل استخراج پردازش گراف کاربر) در لحظه میلیون‌ها دلار هزینه سرور دارد. "
    "سیستم نمی‌تواند منتظر محاسبه همه ویژگی‌ها بماند! ما مسئله را به شکل یک فرآیند تصمیم‌گیری مارکوف (MDP) فرمول‌بندی کرده و "
    "یک <b>مأمور یادگیری تقویتی (RL Agent)</b> را با الگوریتم Q-Learning آموزش می‌دهیم تا یاد بگیرد در کسری از ثانیه "
    "ارزان‌ترین و بهترین ویژگی‌ها را انتخاب کرده و به محض رسیدن به اطمینان، با دستور STOP پاسخ کاربر را بدهد.", 
    "paragraph"
)

display_academic_text("معادلات پایه در یادگیری تقویتی و معادله بلمن (Bellman Equation):", "sub_title")

display_academic_text("۱. تابع پاداش (Reward) با جریمه هزینه محاسباتی برای هر ویژگی:", "paragraph")
display_academic_text(r"$$ R(s, a, s') = \begin{cases} \text{Acc}(s') - \lambda \cdot |s'| & \text{if } a = \text{STOP} \\ -c_{\text{step}} & \text{if } a \neq \text{STOP} \end{cases} $$", "paragraph")

display_academic_text("۲. قاعده به‌روزرسانی تفاضل زمانی (Temporal Difference / Q-Learning) در جدول Q:", "paragraph")
display_academic_text(r"$$ Q(s_t, a_t) \leftarrow Q(s_t, a_t) + \alpha \left[ r_{t+1} + \gamma \max_a Q(s_{t+1}, a) - Q(s_t, a_t) \right] $$", "paragraph")

# ==============================================================================
# 2. NETFLIX USER INTERACTION & COMPUTATIONAL COST DATASET
# ==============================================================================
np.random.seed(42)
n_samples = 500
n_features = 8

# Feature generation:
# 3 High-value informative features (Genre_Affinity, Completion_Rate, Watch_Duration)
f0 = np.random.uniform(0.1, 1.0, size=n_samples)
f1 = np.random.uniform(0.2, 0.9, size=n_samples)
f2 = np.random.poisson(lam=3.0, size=n_samples)

# 5 Redundant or High-Cost Noise Features (Costly to compute, useless for accuracy)
noise_feats = np.random.normal(0, 1.2, size=(n_samples, 5))

X_raw = np.column_stack([f0, f1, f2, noise_feats])
feature_names = [
    'Genre_Affinity', 'Completion_Rate', 'Watch_Duration',
    'Heavy_Graph_Emb', 'HighCost_NLP_Log', 'Noise_IP_Ping',
    'Noise_Device_ID', 'Noise_Theme_Code'
]

# Ground truth Target: Binge Watching Continuation (1 = Yes, 0 = No)
logits = 3.5 * f0 + 2.8 * f1 + 0.6 * f2 - 3.2
probs = 1.0 / (1.0 + np.exp(-logits))
y_raw = (np.random.rand(n_samples) < probs).astype(int)

# Standardize dataset from scratch
X_mean = np.mean(X_raw, axis=0)
X_std = np.std(X_raw, axis=0)
X_std = np.where(X_std == 0, 1.0, X_std)
X_scaled = (X_raw - X_mean) / X_std

# ==============================================================================
# 3. MARKOV DECISION PROCESS (MDP) ENVIRONMENT & Q-LEARNING FROM SCRATCH
# ==============================================================================
def fast_eval_subset(X, y, selected_indices):
    """Evaluates cross-validation accuracy of a given feature subset from scratch."""
    if len(selected_indices) == 0:
        return 0.50 # Random guess if no features
    
    X_sub = X[:, selected_indices]
    n = len(y)
    split = int(0.7 * n)
    X_tr, y_tr = X_sub[:split], y[:split]
    X_val, y_val = X_sub[split:], y[split:]

    # Fast nearest-centroid classifier
    c0 = np.mean(X_tr[y_tr == 0], axis=0) if np.sum(y_tr == 0) > 0 else np.zeros(X_sub.shape[1])
    c1 = np.mean(X_tr[y_tr == 1], axis=0) if np.sum(y_tr == 1) > 0 else np.zeros(X_sub.shape[1])

    preds = []
    for x in X_val:
        d0 = np.sum((x - c0) ** 2)
        d1 = np.sum((x - c1) ** 2)
        preds.append(0 if d0 < d1 else 1)
    
    return np.mean(np.array(preds) == y_val)

class FeatureSelectionMDPEnv:
    """Sequential Feature Selection Environment formulated as an MDP from scratch."""

    def __init__(self, X, y, cost_per_feature=0.04, step_penalty=0.01):
        self.X = X
        self.y = y
        self.n_features = X.shape[1]
        self.cost_per_feature = cost_per_feature
        self.step_penalty = step_penalty
        self.stop_action = self.n_features  # Action index representing STOP
        self.state = None

    def reset(self):
        """Resets state to empty feature mask: tuple of 0s."""
        self.state = tuple(0 for _ in range(self.n_features))
        return self.state

    def step(self, action):
        """Transitions state based on agent action and returns (next_state, reward, done)."""
        if action == self.stop_action:
            # Terminal action: Evaluate the selected feature subset
            active_indices = [i for i, bit in enumerate(self.state) if bit == 1]
            accuracy = fast_eval_subset(self.X, self.y, active_indices)
            
            # The agent is rewarded for accuracy, but penalized heavily for each feature it used!
            penalty = self.cost_per_feature * len(active_indices)
            reward = accuracy - penalty
            return self.state, reward, True

        # Action is toggling/adding a feature
        state_list = list(self.state)
        state_list[action] = 1  # Include feature
        self.state = tuple(state_list)

        # Small penalty for taking time to compute a step
        reward = -self.step_penalty  
        done = False
        return self.state, reward, done

class QLearningFeatureAgent:
    """Tabular Q-Learning Agent for Sequential Decision Making from scratch."""

    def __init__(self, n_actions, lr=0.12, gamma=0.95, epsilon_start=1.0, epsilon_decay=0.992, epsilon_min=0.05):
        self.n_actions = n_actions
        self.lr = lr
        self.gamma = gamma # Discount factor
        self.epsilon = epsilon_start # Exploration rate
        self.epsilon_decay = epsilon_decay
        self.epsilon_min = epsilon_min
        # The Brain: A Q-Table mapping States to Action Values
        self.q_table = defaultdict(lambda: np.zeros(self.n_actions))

    def choose_action(self, state):
        """Epsilon-greedy action selection."""
        if np.random.rand() < self.epsilon:
            # Explore: Take a random action
            return np.random.randint(0, self.n_actions)
        # Exploit: Take the action with the highest Q-Value for this state
        q_vals = self.q_table[state]
        return np.argmax(q_vals)

    def update(self, state, action, reward, next_state, done):
        """Bellman Equation Temporal Difference (TD) update."""
        current_q = self.q_table[state][action]
        if done:
            target_q = reward
        else:
            target_q = reward + self.gamma * np.max(self.q_table[next_state])
        
        self.q_table[state][action] += self.lr * (target_q - current_q)

    def decay_epsilon(self):
        """Decays exploration rate so agent exploits more over time."""
        self.epsilon = max(self.epsilon_min, self.epsilon * self.epsilon_decay)

# ==============================================================================
# 4. TRAIN REINFORCEMENT LEARNING AGENT ACROSS EPISODES
# ==============================================================================
env = FeatureSelectionMDPEnv(X_scaled, y_raw, cost_per_feature=0.035, step_penalty=0.008)
agent = QLearningFeatureAgent(n_actions=n_features + 1, lr=0.15, gamma=0.90)

n_episodes = 500
episode_rewards = []
episode_features_count = []
epsilon_history = []

for ep in range(n_episodes):
    state = env.reset()
    total_reward = 0
    done = False
    step_count = 0

    while not done and step_count < n_features + 2:
        action = agent.choose_action(state)
        next_state, reward, done = env.step(action)
        
        # The agent Learns!
        agent.update(state, action, reward, next_state, done)
        
        state = next_state
        total_reward += reward
        step_count += 1

    agent.decay_epsilon()
    episode_rewards.append(total_reward)
    active_k = sum(state)
    episode_features_count.append(active_k)
    epsilon_history.append(agent.epsilon)

# Extract optimal policy trajectory by acting 100% Greedily (Epsilon = 0)
optimal_state = env.reset()
trajectory = []
done = False
while not done:
    q_values = agent.q_table[optimal_state]
    best_action = np.argmax(q_values)
    
    if best_action == env.stop_action:
        break
        
    trajectory.append(best_action)
    next_state, _, done = env.step(best_action)
    optimal_state = next_state

opt_indices = [i for i, b in enumerate(optimal_state) if b == 1]
opt_accuracy = fast_eval_subset(X_scaled, y_raw, opt_indices) * 100
full_accuracy = fast_eval_subset(X_scaled, y_raw, list(range(n_features))) * 100


# Display the RL Output in a DataFrame
selected_policy_features = [feature_names[i] for i in opt_indices]
df_rl_results = pd.DataFrame([
    {"Agent Status": "Baseline Model (No RL)", "Behavior": f"Computes all {n_features} features", "Accuracy": f"{full_accuracy:.1f}%"},
    {"Agent Status": "Trained RL Agent", "Behavior": f"Computes only {len(opt_indices)} features and STOPs", "Accuracy": f"{opt_accuracy:.1f}%"}
])
display_side_by_side_datasets(
    [df_rl_results], 
    [f"گزارش عملکرد عامل هوشمند نتفلیکس (صرفه‌جویی عظیم در زمان محاسبات سرور)"]
)

# ==============================================================================
# 5. VISUALIZATIONS AND REINFORCEMENT LEARNING CHARTS (ACADEMIC BLUE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

#  CHART 1: Q-Learning Episodic Reward Convergence 
window = 30
moving_avg_reward = pd.Series(episode_rewards).rolling(window=window, min_periods=1).mean()

# Light Blue for raw episodes, Dark Blue for the moving average trend
axes[0].plot(range(1, n_episodes + 1), episode_rewards, color='#bdc3c7', alpha=0.5, label='Raw Episodic Return ($G_t$)')
axes[0].plot(range(1, n_episodes + 1), moving_avg_reward, color='#2980b9', linewidth=3.5, label=f'Moving Avg Return (Window={window})')

axes[0].set_xlabel('Training Episodes (Environment Interactions)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Total Episodic Reward (Accuracy minus Computing Cost)', fontsize=12, fontweight='bold')
axes[0].set_title('1. Netflix RL Agent Reward Convergence via Q-Learning', fontsize=13, fontweight='bold', pad=12)
axes[0].legend(loc='lower right', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
axes[0].grid(True, linestyle='--', alpha=0.6)

# Annotate Convergence
axes[0].annotate(
    f'Agent learned the Optimal Policy!\nMaximizing reward: {moving_avg_reward.iloc[-1]:.3f}',
    xy=(450, moving_avg_reward.iloc[-1]),
    xytext=(-140, -50), textcoords="offset points",
    arrowprops=dict(arrowstyle="->", color='#2c3e50', lw=2),
    bbox=dict(boxstyle='round,pad=0.4', facecolor='#ecf0f1', edgecolor='#2980b9', alpha=0.9),
    fontsize=10, fontweight='bold', color='#2c3e50'
)


#  CHART 2: Learned Q-Values at Initial State vs Feature Selected Status 
initial_q_vals = agent.q_table[tuple(0 for _ in range(n_features))]
action_labels = [feature_names[i] for i in range(n_features)] + ['TERMINATE (STOP)']

# Academic Blue for selected features, Red for STOP, Dark Gray for avoided noise
bar_colors = ['#3498db' if i in opt_indices else ('#e74c3c' if i == env.stop_action else '#2c3e50') for i in range(len(action_labels))]

bars = axes[1].barh(range(len(action_labels)), initial_q_vals, color=bar_colors, edgecolor='black', alpha=0.9)
axes[1].set_yticks(range(len(action_labels)))
axes[1].set_yticklabels(action_labels, fontsize=10, fontweight='bold')
axes[1].set_xlabel('Learned Expected Q-Value $Q(s_0, a)$', fontsize=12, fontweight='bold')
axes[1].set_title('2. Brain of the Agent: Q-Value Policy Preference at Initial State $s_0$', fontsize=13, fontweight='bold', pad=12)
axes[1].grid(axis='x', linestyle='--', alpha=0.6)
axes[1].invert_yaxis()

# Annotate Q-values on the bars
for bar in bars:
    w = bar.get_width()
    x_pos = w + 0.015 if w >= 0 else w - 0.06
    color_txt = '#2c3e50' if w >= 0 else '#c0392b'
    axes[1].annotate(
        f'{w:.2f}', 
        xy=(w, bar.get_y() + bar.get_height()/2),
        xytext=(x_pos, bar.get_y() + bar.get_height()/2),
        ha='left', va='center', fontweight='bold', fontsize=9, color=color_txt
    )

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (اقتصادِ هوش مصنوعی):</b> گراف ۱ نشان می‌دهد که یادگیری تقویتی دقیقاً مانند یک بازی کامپیوتری عمل می‌کند. "
    r"در اپیزودهای اولیه (سمت چپ نمودار)، عامل گیج است. دائم ویژگی‌های به دردنخور را محاسبه می‌کند، در نتیجه پاداشِ او (Reward) منفی می‌شود "
    r"چون برای محاسبهٔ چیزهایی پول داده است که هیچ کمکی به دقت مدل نکرده‌اند. اما از اپیزود ۲۰۰ به بعد، عامل (Agent) الگو را درک می‌کند "
    r"و پاداشِ او شروع به رشد می‌کند و مثبت می‌شود. با نگاه به گراف ۲ (Q-Values)، می‌بینیم که مغزِ سیستم چطور برنامه‌ریزی شده است: "
    r"او برای ویژگی <code>Genre_Affinity</code> ارزش بالایی قائل است (نوار آبی روشن: 0.58). او این ویژگی‌های مهم را در کسری از ثانیه استخراج می‌کند "
    r"و سپس بلافاصله اکشنِ <b>TERMINATE (STOP)</b> را با بالاترین اولویت انتخاب می‌کند تا از شر ویژگی‌های گران و بی‌ارزش (نوار سرمه‌ای رنگ) "
    r"رها شود! جدول نهایی نشان می‌دهد که با این سیاستِ مارکوف، نتفلیکس توانست با محاسبهٔ تنها ۳ ویژگی (به جای ۸ تا)، دقیقاً به همان دقت ۱۰۰ درصدی برسد "
    r"و هزاران دلار در هزینه‌های پردازشِ ابری صرفه‌جویی کند."
)
display_academic_text(insight_conclusion, "insight")
```
