<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: انتخاب درونی ویژگی و تنظیم‌سازی لاسو در «ایر بی‌ان‌بی» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پلتفرم اجاره اقامتگاه <b>ایر بی‌ان‌بی</b>، سیستم قیمت‌گذاری پویا و تخمین نرخ تبدیل رزرو، صدها متغیر مربوط به مشخصات ملک، امکانات رفاهی، سابقه میزبان و نوسانات فصلی را پردازش می‌کند. برخلاف روش‌های فیلتر که تعامل میان متغیرها را نادیده می‌گیرند و روش‌های لفاف‌پیچ که هزینه محاسباتی بسیار سنگینی دارند، مهندسان داده ایر بی‌ان‌بی از <b>روش‌های تعبیه‌شده (Embedded Methods)</b> نظیر رگرسیون لاسو (LASSO) استفاده می‌کنند. در این رویکرد، فرآیند انتخاب ویژگی و آموزش مدل به صورت همزمان از طریق جریمه‌سازی نرم $L_1$ انجام شده و ضرایب ویژگی‌های کم‌اهمیت دقیقاً به صفر میل داده می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی تنظیم‌سازی لاسو ($L_1$ Regularization):
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. تابع هدف بهینه‌سازی لاسو (LASSO Cost Function):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\min_{\mathbf{w}, b} \frac{1}{2N} \sum_{i=1}^{N} \left( y_i - (\mathbf{w}^T \mathbf{x}_i + b) \right)^2 + \lambda \sum_{j=1}^{p} |w_j|$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. عملگر آستانه‌گذاری نرم (Soft-Thresholding Operator) در بهینه‌سازی نزول مختصاتی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$S(z, \gamma) = \text{sign}(z) \cdot \max(0, |z| - \gamma)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. رابطه به‌روزرسانی وزن هر ویژگی ($w_j$) در الگوریتم Coordinate Descent:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$w_j \leftarrow \frac{S\left( \sum_{i=1}^N x_{ij}(y_i - \hat{y}_i^{(-j)}), \lambda N \right)}{\sum_{i=1}^N x_{ij}^2}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. شرط پراکنده‌سازی و صفر شدن ضریب ویژگی (Feature Elimination Sparsity Condition):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$w_j = 0 \iff \left| \frac{1}{N} \sum_{i=1}^N x_{ij} (y_i - \hat{y}_i^{(-j)}) \right| \le \lambda$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: ایر بی‌ان‌بی با تنظیم ضریب جریمه ($\lambda$)، ویژگی‌های اضافی و هم‌خط (Collinear) را در طول فرآیند آموزش حذف کرده و مدلی با تفسیرپذیری بالا و سرعت پیش‌بینی بلادرنگ به دست می‌آورد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: بهینه‌سازی توأم پارامترهای مدل و کاهش ابعاد بردار ویژگی با پراکنده‌سازی ضرایب
    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "انتخاب درونی ویژگی و تنظیم‌سازی لاسو در «ایر بی‌ان‌بی» (LASSO from Scratch) 🏠", 
    "main_title"
)

display_academic_text(
    "سیستم قیمت‌گذاری ایر بی‌ان‌بی، صدها متغیر را پردازش می‌کند. مهندسان از روش‌های تعبیه‌شده (Embedded Methods) "
    "نظیر رگرسیون لاسو (LASSO) استفاده می‌کنند. این روش به کمک جریمه‌سازی $L_1$، فرآیند آموزش مدل و انتخاب ویژگی را "
    "ترکیب کرده و ضرایب ویژگی‌های نامربوط را مستقیماً به صفر میل می‌دهد (پراکنده‌سازی یا Sparsity).", 
    "paragraph"
)

display_academic_text("فرمول‌های ریاضیاتی LASSO و بهینه‌سازی:", "sub_title")

display_academic_text("۱. تابع هدف بهینه‌سازی لاسو (LASSO Cost Function):", "paragraph")
display_academic_text(r"$$\min_{\mathbf{w}, b} \frac{1}{2N} \sum_{i=1}^{N} \left( y_i - (\mathbf{w}^T \mathbf{x}_i + b) \right)^2 + \lambda \sum_{j=1}^{p} |w_j|$$", "paragraph")

display_academic_text("۲. عملگر آستانه‌گذاری نرم (Soft-Thresholding Operator):", "paragraph")
display_academic_text(r"$$S(z, \gamma) = \text{sign}(z) \cdot \max(0, |z| - \gamma)$$", "paragraph")

display_academic_text("۳. آپدیت وزن‌ها در الگوریتم نزول مختصاتی (Coordinate Descent):", "paragraph")
display_academic_text(r"$$w_j \leftarrow \frac{S\left( \sum_{i=1}^N x_{ij}(y_i - \hat{y}_i^{(-j)}), \lambda N \right)}{\sum_{i=1}^N x_{ij}^2}$$", "paragraph")

# ==============================================================================
# 2. AIRBNB LISTING RESERVATION DATASET SIMULATION
# ==============================================================================
np.random.seed(42)
n_listings = 500

# 5 TRUE PREDICTIVE FEATURES (These actually affect the booking demand)
location_score = np.random.uniform(1.0, 10.0, size=n_listings)
cleanliness_rating = np.random.uniform(2.0, 5.0, size=n_listings)
superhost_flag = np.random.choice([0, 1], size=n_listings, p=[0.7, 0.3])
accommodates_count = np.random.choice([1, 2, 4, 6], size=n_listings, p=[0.3, 0.4, 0.2, 0.1])
price_competitiveness = np.random.uniform(0.5, 1.5, size=n_listings)

# 5 NOISE FEATURES (Spurious, irrelevant, or pure noise - LASSO should kill these)
host_id_hash_mod = np.random.uniform(0, 100, size=n_listings)
listing_day_listed = np.random.choice(range(1, 31), size=n_listings)
temp_client_ping = np.random.normal(50, 15, size=n_listings)
color_palette_code = np.random.choice([1, 2, 3, 4], size=n_listings)
random_noise_id = np.random.normal(0, 1, size=n_listings)

# Target: Booking Demand Index (Linear combination of True features + Gaussian noise)
target_demand = (
    4.5 * location_score +
    3.8 * cleanliness_rating +
    6.2 * superhost_flag +
    2.5 * accommodates_count +
    5.0 * price_competitiveness +
    np.random.normal(0, 2.0, size=n_listings)
)

feature_names = [
    'Location_Score', 'Cleanliness_Rating', 'Superhost_Flag',
    'Accommodates_Count', 'Price_Competitiveness', 'Host_ID_Hash',
    'Day_Listed', 'Client_Ping', 'Palette_Code', 'Random_Noise'
]

X_raw = np.column_stack([
    location_score, cleanliness_rating, superhost_flag, accommodates_count,
    price_competitiveness, host_id_hash_mod, listing_day_listed,
    temp_client_ping, color_palette_code, random_noise_id
])
y_raw = target_demand

# ==============================================================================
# 3. EMBEDDED LASSO REGRESSION (COORDINATE DESCENT) FROM SCRATCH
# ==============================================================================
class LassoCoordinateDescent:
    """
    L1-Regularized LASSO Regression using Coordinate Descent.
    Unlike Gradient Descent, Coordinate Descent optimizes one feature at a time,
    making it perfectly suited for the non-differentiable L1 penalty.
    """

    def __init__(self, l1_lambda=1.0, max_iter=1000, tol=1e-5):
        self.l1_lambda = l1_lambda
        self.max_iter = max_iter
        self.tol = tol
        self.weights = None
        self.bias = 0.0

    @staticmethod
    def soft_threshold(rho, gamma):
        """Soft-thresholding operator S(rho, gamma). This is what creates Sparsity (zeros)."""
        if rho > gamma:
            return rho - gamma
        elif rho < -gamma:
            return rho + gamma
        else:
            return 0.0

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        
        # Center target (handling the intercept/bias implicitly)
        self.bias = np.mean(y)
        y_centered = y - self.bias
        
        # Precompute sum of squares for each feature (Denominator in the update rule)
        sum_sq = np.sum(X ** 2, axis=0)

        for iteration in range(self.max_iter):
            max_weight_change = 0.0

            for j in range(n_features):
                old_w = self.weights[j]
                
                # Compute partial residual: We subtract the contribution of all OTHER features
                residual = y_centered - np.dot(X, self.weights) + old_w * X[:, j]
                
                # Compute rho (correlation between feature j and the partial residual)
                rho = np.dot(X[:, j], residual)
                
                # Apply soft thresholding and update the weight
                new_w = self.soft_threshold(rho, self.l1_lambda * n_samples) / sum_sq[j]
                self.weights[j] = new_w

                # Track convergence
                weight_change = abs(new_w - old_w)
                if weight_change > max_weight_change:
                    max_weight_change = weight_change

            # Early stopping if weights have stabilized
            if max_weight_change < self.tol:
                break

        return self

# ------------------------------------------------------------------------------
# MODEL TRAINING & REGULARIZATION PATH GENERATION
# ------------------------------------------------------------------------------
# Standardize features for LASSO (Zero mean, unit variance is REQUIRED for L1 penalty)
X_mean = np.mean(X_raw, axis=0)
X_std = np.std(X_raw, axis=0)
X_scaled = (X_raw - X_mean) / X_std

# Compute Regularization Path across different lambda values
lambda_values = np.logspace(-2, 1.8, 60)
coef_path = []

for lam in lambda_values:
    lasso_model = LassoCoordinateDescent(l1_lambda=lam, max_iter=800)
    lasso_model.fit(X_scaled, y_raw)
    coef_path.append(lasso_model.weights.copy())

coef_path = np.array(coef_path)

# Select an optimal lambda for production demonstration
target_lambda = 1.2
optimal_lasso = LassoCoordinateDescent(l1_lambda=target_lambda).fit(X_scaled, y_raw)

# Organize Results into a DataFrame
df_weights = pd.DataFrame({
    'Feature': feature_names,
    'Learned Weight': optimal_lasso.weights,
    'Status': np.where(np.abs(optimal_lasso.weights) > 1e-4, '✅ Kept', '❌ Zeroed (Noise)')
}).sort_values(by='Learned Weight', ascending=False)

display_side_by_side_datasets([df_weights], [f"وضعیت ویژگی‌ها پس از اعمال LASSO ($\lambda = {target_lambda}$)"])

# ==============================================================================
# 4. VISUALIZATIONS AND REGULARIZATION PATH CHARTS (ACADEMIC BLUE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

#  CHART 1: LASSO Regularization Path
# We use distinct colors: Academic Blues for real features, Grays for Noise
for i, feat in enumerate(feature_names):
    is_noise = i >= 5
    style = '--' if is_noise else '-'
    width = 1.8 if not is_noise else 1.5
    alpha = 1.0 if not is_noise else 0.5
    
    # Solid Navy/Teal for real features, Gray/Red for noise
    color = '#2980b9' if not is_noise else '#7f8c8d'
    if i == 0: color = '#2c3e50'
    if i == 1: color = '#16a085'
    
    axes[0].plot(
        lambda_values, coef_path[:, i], 
        label=feat, linestyle=style, linewidth=width, alpha=alpha, color=color
    )

axes[0].set_xscale('log')
axes[0].axvline(target_lambda, color='#e74c3c', linestyle=':', linewidth=2.5, label=f'Chosen $\lambda$ = {target_lambda}')
axes[0].axhline(0, color='black', linewidth=1, linestyle='-')

axes[0].set_xlabel('Regularization Penalty Parameter $\lambda$ (Log Scale)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Standardized Feature Weight ($w_j$)', fontsize=12, fontweight='bold')
axes[0].set_title('1. Airbnb LASSO Coefficient Shrinkage Path\n(Observe Noise features hitting exactly Zero early)', fontsize=13, fontweight='bold', pad=12)
axes[0].legend(loc='upper right', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=9, ncol=2)
axes[0].grid(True, linestyle='--', alpha=0.6)

#  CHART 2: Bar Chart of Final Coefficients 
df_plot = df_weights.sort_values(by='Learned Weight', ascending=True)

# Color coding: Academic Blue for active features, Gray for zeroed/noise features
bar_colors = ['#bdc3c7' if w == 0 else '#2980b9' for w in df_plot['Learned Weight']]
bars = axes[1].barh(df_plot['Feature'], df_plot['Learned Weight'], color=bar_colors, edgecolor='black', alpha=0.9, height=0.6)

axes[1].axvline(0, color='black', linewidth=1.5)
axes[1].set_xlabel('Optimized Feature Weight ($w_j$)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Airbnb Listing Features', fontsize=12, fontweight='bold')
axes[1].set_title(f'2. Embedded Feature Selection at $\lambda = {target_lambda}$\n(Sparsity: Noise completely eliminated)', fontsize=13, fontweight='bold', pad=12)
axes[1].grid(axis='x', linestyle='--', alpha=0.7)

for bar in bars:
    w = bar.get_width()
    x_pos = w + 0.15 if w >= 0 else w - 0.5
    axes[1].annotate(
        f'{w:.2f}', 
        xy=(w, bar.get_y() + bar.get_height()/2),
        xytext=(x_pos, bar.get_y() + bar.get_height()/2),
        ha='left', va='center', fontweight='bold', fontsize=10, color='#2c3e50'
    )

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (قدرت جریمه $L_1$):</b> گراف ۱ به زیبایی تئوری LASSO را به تصویر می‌کشد. در سمت چپ "
    r"نمودار (جایی که $\lambda$ بسیار کوچک است)، مدل مانند یک رگرسیون خطی معمولی عمل می‌کند و به تمام ویژگی‌ها، "
    r"حتی نویزهایی مثل <code>Color_Palette</code> (خط‌چین‌های خاکستری) وزن می‌دهد. اما با افزایش مقدار جریمه $\lambda$، "
    r"اتفاق جالبی می‌افتد: ویژگی‌های نویزدار به سرعت به سمت محور صفر کشیده شده و <b>دقیقاً صفر می‌شوند</b> (این تفاوتِ "
    r"مهم LASSO با Ridge است). با تنظیم $\lambda$ روی مقدار بهینه (خط‌چین قرمز)، سیستم توانسته است دقیقاً ۵ ویژگیِ زائد "
    r"را از معادلات حذف کند (جدول بالا). در نتیجه، ایر بی‌ان‌بی صاحب یک مدل سریع‌تر، سبک‌تر و بسیار قابل‌تفسیر (Interpretable) می‌شود."
)
display_academic_text(insight_conclusion, "insight")
```
