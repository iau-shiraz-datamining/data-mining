  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: تبدیل ساختار سلسله‌مراتبی درخت تصمیم به قواعد شفاف در «نتفلیکس» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در سیستم مدیریت پهنای باند و تنظیم خودکار کیفیت پخش ویدیو در <b>نتفلیکس</b>، مدل‌های درخت تصمیم برای تعیین کیفیت بهینه استریم (SD، HD یا 4K) بر اساس شاخص‌های شبکه و دستگاه کاربر آموزش داده می‌شوند. با وجود کارایی درخت تصمیم، ساختارهای تودرتو و عمیق برای مهندسان سیستم‌های توزیع‌شده خوانایی پایینی دارند و تعبیه آن‌ها در موتورهای قوانین برخط (Rule Engines) به صورت مستقیم پرهزینه است. مهندسان داده نتفلیکس با استفاده از تکنیک‌های <b>استخراج قواعد (Rule Extraction)</b>، تمام مسیرهای منتهی از ریشه به برگ‌ها را به مجموعه‌ای از قواعد صریح «اگر-آنگاه» (IF-THEN) مستقل تبدیل و ساده‌سازی می‌کنند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی استخراج و ارزیابی قواعد از درخت تصمیم:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. معیار ناخالصی جینی (Gini Impurity) در هر گره تصمیم درخت:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{Gini}(D) = 1 - \sum_{k=1}^{C} p_k^2$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. کاهش ناخالصی حاصل از تفکیک گره با ویژگی $A$ و آستانه $\theta$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\Delta \text{Gini}(D, A, \theta) = \text{Gini}(D) - \left( \frac{|D_L|}{|D|} \text{Gini}(D_L) + \frac{|D_R|}{|D|} \text{Gini}(D_R) \right)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. فرمول استخراج منطقی قاعده از مسیر ریشه به گره برگ $l_j$ با عمق $d_j$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$R_j: \text{IF } \left( \bigwedge_{m=1}^{d_j} (x_{a_m} \bowtie_m \theta_m) \right) \implies \text{THEN } y = c_j, \quad \bowtie_m \in \{\le, >\}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. محاسبه نرخ خطای تعمیم بدبینانه (Pessimistic Error Pruning) برای هرس شروط اضافه قاعده:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{Error}_{\text{pessimistic}}(R) = \frac{e(R) + \frac{1}{2}}{n(R)}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: نتفلیکس با استخراج قواعد مستقل و هرس شروط زائد، منطق تصمیم‌گیری درخت را به صورت کدهای شرطی بسیار سبک در میکروکنترلرها و اپلیکیشن‌های تلویزیون هوشمند پیاده‌سازی می‌نماید.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: تبدیل درخت تصمیم به قواعد منطقی مستقل، موازی‌ساز و کاملاً تفسیرپذیر

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "تبدیل ساختار سلسله‌مراتبی درخت تصمیم به قواعد شفاف در «نتفلیکس» (Rule Extraction From Scratch) 🎬", 
    "main_title"
)

display_academic_text(
    "در سیستم نتفلیکس، درخت‌های تصمیم کیفیت استریم (SD، HD، 4K) را تعیین می‌کنند. اما پیاده‌سازیِ مستقیمِ "
    "یک درختِ تودرتو و عمیق در موتورهای قوانین برخط (Rule Engines) یا اپلیکیشنِ تلویزیون، کُند و ناخوانا است. "
    "ما با استفاده از تکنیک <b>استخراج قواعد (Rule Extraction)</b> تمام مسیرهای گرافِ درخت را شخم زده و به یک لیست "
    "مسطح، موازی و شفاف از شروط «IF-THEN» تبدیل می‌کنیم.", 
    "paragraph"
)

display_academic_text("معادلات پایه در ارزیابی و استخراج قواعد:", "sub_title")

display_academic_text("۱. فرمول استخراج منطقی قاعده از مسیر ریشه به گره برگ $l_j$ با عمق $d_j$:", "paragraph")
display_academic_text(r"$$R_j: \text{IF } \left( \bigwedge_{m=1}^{d_j} (x_{a_m} \bowtie_m \theta_m) \right) \implies \text{THEN } y = c_j$$", "paragraph")

display_academic_text("۲. محاسبه نرخ خطای تعمیم بدبینانه (Pessimistic Error Pruning) برای هرس شروط:", "paragraph")
display_academic_text(r"$$\text{Error}_{\text{pessimistic}}(R) = \frac{e(R) + 0.5}{n(R)}$$", "paragraph")

# ==============================================================================
# 2. NETFLIX STREAMING QUALITY DATASET SIMULATION
# ==============================================================================
np.random.seed(42)
n_sessions = 800

# Telemetry features for video stream quality adaptation
bandwidth = np.random.uniform(1.0, 50.0, size=n_sessions)
ping = np.random.uniform(10.0, 150.0, size=n_sessions)
screen_size = np.random.choice([5.5, 13.3, 27.0, 55.0, 65.0], size=n_sessions)
buffer_drops = np.random.poisson(lam=1.5, size=n_sessions)

# Target Stream Quality Tier: 0 = SD (Low), 1 = HD (Medium), 2 = 4K (Ultra HD)
quality_tier = np.zeros(n_sessions, dtype=int)
for i in range(n_sessions):
    if bandwidth[i] >= 25.0 and ping[i] < 60.0 and buffer_drops[i] <= 1:
        quality_tier[i] = 2  # 4K
    elif bandwidth[i] >= 8.0 and buffer_drops[i] <= 3:
        quality_tier[i] = 1  # HD
    else:
        quality_tier[i] = 0  # SD

# Introduce some noise to make the tree work harder
noise_idx = np.random.choice(n_sessions, size=int(n_sessions * 0.05), replace=False)
quality_tier[noise_idx] = np.random.choice([0, 1, 2], size=len(noise_idx))

df_netflix = pd.DataFrame({
    'Bandwidth': bandwidth,
    'Ping': ping,
    'Screen_Size': screen_size,
    'Buffer_Drops': buffer_drops,
    'Quality_Tier': quality_tier
})

feature_names = ['Bandwidth', 'Ping', 'Screen_Size', 'Buffer_Drops']
X_data = df_netflix[feature_names].values
y_data = df_netflix['Quality_Tier'].values

# ==============================================================================
# 3. DECISION TREE & RULE EXTRACTION ENGINE FROM SCRATCH
# ==============================================================================
class TreeNode:
    """Node structure for decision tree representation."""
    def __init__(self, is_leaf=False, prediction=None, feature=None, threshold=None, left=None, right=None, samples=0, gini=0.0):
        self.is_leaf = is_leaf
        self.prediction = prediction
        self.feature = feature
        self.threshold = threshold
        self.left = left
        self.right = right
        self.samples = samples
        self.gini = gini

class DecisionTreeScratch:
    """Decision Tree Classifier built from scratch with Gini Impurity."""
    def __init__(self, max_depth=3, min_samples_split=10):
        self.max_depth = max_depth
        self.min_samples_split = min_samples_split
        self.root = None

    def _gini(self, y):
        if len(y) == 0: return 0.0
        _, counts = np.unique(y, return_counts=True)
        probs = counts / len(y)
        return 1.0 - np.sum(probs ** 2)

    def _best_split(self, X, y):
        n_samples, n_features = X.shape
        best_gain = -1.0
        best_feat, best_thresh = None, None
        current_gini = self._gini(y)

        for feat_idx in range(n_features):
            thresholds = np.unique(X[:, feat_idx])
            for thresh in thresholds:
                left_mask = X[:, feat_idx] <= thresh
                right_mask = ~left_mask

                if np.sum(left_mask) == 0 or np.sum(right_mask) == 0: continue

                y_l, y_r = y[left_mask], y[right_mask]
                gain = current_gini - (len(y_l)/n_samples * self._gini(y_l) + len(y_r)/n_samples * self._gini(y_r))

                if gain > best_gain:
                    best_gain = gain
                    best_feat = feat_idx
                    best_thresh = thresh

        return best_feat, best_thresh, best_gain

    def _build_tree(self, X, y, depth=0):
        n_samples = len(y)
        classes, counts = np.unique(y, return_counts=True)
        majority_class = classes[np.argmax(counts)]
        current_gini = self._gini(y)

        if depth >= self.max_depth or n_samples < self.min_samples_split or len(classes) == 1:
            return TreeNode(is_leaf=True, prediction=majority_class, samples=n_samples, gini=current_gini)

        best_feat, best_thresh, best_gain = self._best_split(X, y)

        if best_feat is None or best_gain <= 1e-6:
            return TreeNode(is_leaf=True, prediction=majority_class, samples=n_samples, gini=current_gini)

        left_mask = X[:, best_feat] <= best_thresh
        right_mask = ~left_mask

        left_child = self._build_tree(X[left_mask], y[left_mask], depth + 1)
        right_child = self._build_tree(X[right_mask], y[right_mask], depth + 1)

        return TreeNode(
            is_leaf=False, feature=best_feat, threshold=best_thresh,
            left=left_child, right=right_child, samples=n_samples, gini=current_gini
        )

    def fit(self, X, y):
        self.root = self._build_tree(X, y)
        return self


class RuleExtractor:
    """Traverses a decision tree (DFS) and extracts human-readable IF-THEN rules from scratch."""
    def __init__(self, tree_model, feature_names, class_names):
        self.tree = tree_model
        self.feature_names = feature_names
        self.class_names = class_names
        self.extracted_rules = []

    def _traverse_path(self, node, current_conditions):
        """Recursive DFS traversal from root to leaf to formulate path conjunctions."""
        if node.is_leaf:
            self.extracted_rules.append({
                'Conditions': list(current_conditions),
                'Prediction': node.prediction,
                'Class_Label': self.class_names[node.prediction],
                'Leaf_Samples': node.samples,
                'Leaf_Gini': node.gini,
                'Rule_Length': len(current_conditions)
            })
            return

        feat_name = self.feature_names[node.feature]
        thresh_val = node.threshold

        # Traverse Left (<= threshold)
        self._traverse_path(node.left, current_conditions + [f"{feat_name} <= {thresh_val:.1f}"])
        # Traverse Right (> threshold)
        self._traverse_path(node.right, current_conditions + [f"{feat_name} > {thresh_val:.1f}"])

    def extract(self, X, y):
        """Extracts all rules and computes coverage and empirical accuracy from scratch."""
        self.extracted_rules = []
        self._traverse_path(self.tree.root, [])

        results = []
        total_samples = len(y)

        for idx, rule in enumerate(self.extracted_rules):
            # Parse and evaluate rule against dataset
            mask = np.ones(total_samples, dtype=bool)
            for cond in rule['Conditions']:
                if "<=" in cond:
                    feat, val = cond.split(" <= ")
                    mask = mask & (X[:, self.feature_names.index(feat)] <= float(val))
                elif ">" in cond:
                    feat, val = cond.split(" > ")
                    mask = mask & (X[:, self.feature_names.index(feat)] > float(val))

            n_covered = np.sum(mask)
            n_correct = np.sum(y[mask] == rule['Prediction']) if n_covered > 0 else 0

            coverage = n_covered / total_samples
            accuracy = n_correct / n_covered if n_covered > 0 else 0.0
            
            # Pessimistic Error Pruning Formulation: (e + 0.5) / n
            errors = n_covered - n_correct
            pessimistic_err = (errors + 0.5) / n_covered if n_covered > 0 else 1.0

            cond_str = " AND ".join(rule['Conditions'])
            rule_str = f"IF {cond_str} \u2794 THEN {rule['Class_Label']}"

            results.append({
                'Rule_ID': f"R{idx+1}",
                'Extracted Flattened Logic Rule': rule_str,
                'Prediction': rule['Class_Label'],
                'Length': rule['Rule_Length'],
                'Coverage': coverage,
                'Accuracy': accuracy,
                'Pessimistic_Error': pessimistic_err,
                'Covered_Samples': n_covered
            })

        return pd.DataFrame(results).sort_values(by='Accuracy', ascending=False).reset_index(drop=True)

# Train decision tree
dt = DecisionTreeScratch(max_depth=3, min_samples_split=15)
dt.fit(X_data, y_data)

# Extract Rules
extractor = RuleExtractor(dt, feature_names, class_names=['SD Quality', 'HD Quality', '4K Ultra HD'])
df_rules = extractor.extract(X_data, y_data)

# Reassign IDs after sorting
df_rules['Rule_ID'] = [f"R{i+1}" for i in range(len(df_rules))]

# Display the rules as a clean, dashboard-like table
display_cols = ['Rule_ID', 'Extracted Flattened Logic Rule', 'Covered_Samples', 'Accuracy', 'Pessimistic_Error']

df_display = df_rules[display_cols].copy()
df_display['Accuracy'] = df_display['Accuracy'].apply(lambda x: f"{x*100:.1f}%")
df_display['Pessimistic_Error'] = df_display['Pessimistic_Error'].apply(lambda x: f"{x*100:.1f}%")

display_side_by_side_datasets(
    [df_display], 
    ["فهرست شروط مسطح (Flattened Rules) مستخرج از درخت تصمیم نتفلیکس"]
)

# ==============================================================================
# 4. VISUALIZATIONS AND RULE QUALITY ANALYSIS CHARTS (ACADEMIC BLUE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

#  CHART 1: Extracted Rules Accuracy & Coverage Performance 
y_pos = np.arange(len(df_rules))
bar_height = 0.35

# Academic Blue Palette for Quality Tiers
tier_colors = {'SD Quality': '#bdc3c7', 'HD Quality': '#3498db', '4K Ultra HD': '#2c3e50'}
colors_acc = [tier_colors[p] for p in df_rules['Prediction']]

rects1 = axes[0].barh(
    y_pos - bar_height/2, df_rules['Accuracy'] * 100, height=bar_height,
    color=colors_acc, edgecolor='black', alpha=0.9, label='Empirical Rule Accuracy (%)'
)

rects2 = axes[0].barh(
    y_pos + bar_height/2, df_rules['Coverage'] * 100, height=bar_height,
    color='#ecf0f1', edgecolor='black', alpha=0.9, hatch='////', label='Rule Data Coverage (%)'
)

axes[0].set_yticks(y_pos)
axes[0].set_yticklabels(df_rules['Rule_ID'] + " (" + df_rules['Prediction'] + ")", fontsize=10, fontweight='bold')
axes[0].set_xlabel('Percentage Value (%)', fontsize=12, fontweight='bold')
axes[0].set_title('1. Extracted Rules Profile: Accuracy vs Coverage by Stream Tier', fontsize=13, fontweight='bold', pad=12)
axes[0].legend(loc='lower right', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
axes[0].grid(axis='x', linestyle='--', alpha=0.6)
axes[0].invert_yaxis()

# Annotate accuracy percentages
for bar in rects1:
    w = bar.get_width()
    axes[0].annotate(
        f'{w:.1f}%', 
        xy=(w, bar.get_y() + bar.get_height()/2),
        xytext=(4, 0), textcoords="offset points", ha='left', va='center', fontweight='bold', fontsize=9, color='#2c3e50'
    )


#  CHART 2: Rule Complexity (Path Depth) vs Accuracy & Pessimistic Error 
axes[1].scatter(
    df_rules['Length'],
    df_rules['Accuracy'] * 100,
    s=df_rules['Covered_Samples'] * 4.5, # Bubble size = coverage
    c=colors_acc,
    edgecolors='black',
    linewidths=1.5,
    alpha=0.85,
    zorder=3
)

for _, row in df_rules.iterrows():
    axes[1].annotate(
        f"{row['Rule_ID']}",
        xy=(row['Length'], row['Accuracy'] * 100),
        xytext=(0, 15), textcoords='offset points',
        fontsize=10, fontweight='bold', color='#2c3e50', ha='center',
        bbox=dict(boxstyle='round,pad=0.2', facecolor='#f8f9fa', edgecolor='#bdc3c7', alpha=0.8)
    )

axes[1].set_xlabel('Rule Length / Logical Complexity (Number of AND Conditions)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Empirical Accuracy (%)', fontsize=12, fontweight='bold')
axes[1].set_title('2. Rule Complexity vs Accuracy (Bubble Sized by Sample Coverage)', fontsize=13, fontweight='bold', pad=12)
axes[1].set_xticks([1, 2, 3])
axes[1].grid(True, linestyle='--', alpha=0.6)

# Adding a dummy legend for bubble size explanation
axes[1].scatter([], [], s=100, color='gray', edgecolors='black', label='Bubble Size = No. of Covered Users')
axes[1].legend(loc='lower left', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (از گره به قانون):</b> جدول قوانین خروجی به خوبی نشان می‌دهد که چگونه یک ساختار درختی و در‌هم‌تنیده، "
    r"به گزاره‌های خطی و مستقل تبدیل شده است. قانون <b>R1</b> به تنهایی بیان می‌کند که: «اگر پهنای باند کمتر از ۲۴ باشد اما قطعی "
    r"بافر نداشته باشیم، کیفیت HD پخش کن». این یک دستور یک‌خطی است که مهندس نتفلیکس می‌تواند آن را به راحتی در کدهای جاوااسکریپت "
    r"سمت کلاینت قرار دهد. در گراف ۲، قانون R1 با یک دایره فیروزه‌ای نسبتاً بزرگ نشان داده شده است (چون کاربران زیادی را پوشش می‌دهد). "
    r"نکته جالب این است که قوانین با طول ۳ (کمی پیچیده‌تر) با وجود داشتن دقت بالاتر، دایره‌های بسیار کوچکتری دارند، یعنی در محیط واقعی، "
    r"موارد بسیار نادری را پوشش می‌دهند. این تحلیل بصری به نتفلیکس کمک می‌کند که آیا حفظِ این قوانینِ طولانی اما کم‌کاربرد، "
    r"ارزش افزودن پیچیدگی به موتور پردازشی (Rule Engine) را دارد یا خیر."
)
display_academic_text(insight_conclusion, "insight")
```

<li>پیاده‌سازی کامل درخت تصمیم و موتور استخراج قواعد از صفر (From Scratch):
  <ul>
    <li><b>الگوریتم درخت تصمیم (CART):</b> در کلاس <code>DecisionTreeScratch</code>، تابع محاسبه ناخالصی جینی (<code>_gini</code>) و جستجوی بهینه‌ترین ویژگی و آستانه تفکیک (<code>_best_split</code>) پیاده‌سازی شده و ساختار درختی با محدودیت عمق ساخته شد.</li>
    <li><b>استخراج قواعد با پیمایش DFS:</b> در کلاس <code>RuleExtractor</code> و متد <code>_traverse_path</code>، تمام مسیرهای از ریشه به برگ پیمایش شده و گزاره‌های شرطی هر یال با عملگرهای منطقی عطف (AND) به یک گزاره شرطی مستقل تبدیل شدند.</li>
    <li><b>سنجش معیارهای کیفی و خطای بدبینانه:</b> در متد <code>extract</code>، پوشش، دقت تجربی و خطای بدبینانه هر قاعده روی کل داده‌ها به دست آمد.</li>
  </ul>
</li>
<li>تفسیر خروجی و نمودارهای تحلیلی:
  <ul>
    <li><b>نمودار افقی چپ (Accuracy vs Coverage):</b> مقایسه دقت و پوشش هر قاعده استخراج‌شده را نشان می‌دهد؛ قواعد مرتبط با کیفیت 4K و SD به دقتی بالای ۹۰٪ دست یافته‌اند در حالی که هر قاعده بخش مشخصی از سشن‌های استریم را پوشش می‌دهد.</li>
    <li><b>نمودار حبابی راست (Complexity vs Accuracy):</b> رابطه میان طول قاعده (تعداد شروط عطف) و دقت نهایی را نمایش می‌دهد؛ اندازه حباب‌ها نشان‌دهنده تعداد نمونه‌های پشتیبانی‌کننده است و نشان می‌دهد که افزودن شروط هوشمندانه عمق درخت چگونه بدون افت شدید پوشش، دقت تصمیم را ارتقا می‌دهد.</li>
  </ul>
</li>
</ul>
