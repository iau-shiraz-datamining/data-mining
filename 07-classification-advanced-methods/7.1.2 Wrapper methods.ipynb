<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: بهینه‌سازی زیرمجموعه ویژگی‌ها در «نتفلیکس» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در سرویس استریم فیلم و سریال <b>نتفلیکس</b>، پیش‌بینی تمدید اشتراک یا ریزش کاربران (Churn Prediction) از اهمیت حیاتی برخوردار است. رفتار تعاملی هر کاربر با ده‌ها شاخص مانند نرخ تکمیل قسمت‌ها، تعداد توقف و بازپخش، ساعات تماشا در آخر هفته، تنوع دستگاه‌ها و جستجوهای ناموفق سنجیده می‌شود. از آنجا که اثر این ویژگی‌ها مستقل از یکدیگر نیست و تعاملات غیرخطی پیچیده‌ای با یکدیگر دارند، مهندسان نتفلیکس از <b>Wrapper Methods</b> برای ارزیابی جامع ترکیب‌های مختلف ویژگی‌ها و انتخاب بهینه‌ترین زیرمجموعه بر مبنای عملکرد واقعی الگوریتم طبقه‌بندی استفاده می‌کنند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی حاکم بر روش‌های لفاف‌پیچ و الگوریتم‌های جستجو:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. تابع هدف بهینه‌سازی برای یافتن زیرمجموعه بهینه ویژگی‌ها ($\mathcal{S}^*$):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\mathcal{S}^* = \arg\max_{\mathcal{S} \subseteq \mathcal{F}, |\mathcal{S}| \le k} J(f_{\theta}(X_{\mathcal{S}}), Y)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. قاعده اضافه کردن حریصانه در روش انتخاب رو به جلو (Sequential Forward Selection):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$x^+ = \arg\max_{x \in \mathcal{F} \setminus \mathcal{S}_t} J(\mathcal{S}_t \cup \{x\}), \quad \mathcal{S}_{t+1} = \mathcal{S}_t \cup \{x^+\}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. قاعده حذف بازگشتی ویژگی‌ها (Recursive Feature Elimination - RFE) بر اساس ضریب اهمیت:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$x^- = \arg\min_{x_i \in \mathcal{S}_t} (w_i)^2, \quad \mathcal{S}_{t+1} = \mathcal{S}_t \setminus \{x^-\}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. ارزیابی عملکرد زیرمجموعه با اعتبارسنجی متقاطع چندلایه ($K$-Fold Cross Validation):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$J(\mathcal{S}) = \frac{1}{K} \sum_{k=1}^{K} \text{F1-Score}\left(Y_{\text{val}}^{(k)}, \hat{Y}^{(k)}(X_{\mathcal{S},\text{val}}^{(k)})\right)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: نتفلیکس با روش Wrapper Methods، ویژگی‌های گمراه‌کننده را از مدل حذف کرده و با کاهش ابعاد بردار تعاملات، بدون افت دقت پیش‌بینی، پیچیدگی محاسباتی سرویس را به شدت کاهش می‌دهد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: کشف بهترین تعاملات چندویژگی با استفاده از بازخورد مستقیم خطای مدل یادگیری
    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "بهینه‌سازی زیرمجموعه ویژگی‌ها در «نتفلیکس» (Wrapper Methods From Scratch) 🎬", 
    "main_title"
)

display_academic_text(
    "در سرویس نتفلیکس، پیش‌بینی ریزش (Churn) کاربران بسیار حیاتی است. ده‌ها متغیر رفتارِ کاربر را ثبت می‌کنند، اما "
    "اثر این ویژگی‌ها مستقل از یکدیگر نیست. مهندسان به جای روش‌های فیلتر، از <b>Wrapper Methods</b> استفاده می‌کنند. "
    "در این رویکرد، ما الگوریتم یادگیری ماشین را داخل یک «حلقه جستجو» قرار می‌دهیم و زیرمجموعه‌های مختلف را بر اساس "
    "«دقتِ واقعیِ پیش‌بینیِ مدل» می‌سنجیم تا طلایی‌ترین ترکیب متغیرها کشف شود.", 
    "paragraph"
)

display_academic_text("معادلات و مفاهیم ریاضی روش‌های لفاف‌پیچ (Wrapper):", "sub_title")

display_academic_text("۱. تابع هدف بهینه‌سازی برای یافتن زیرمجموعه بهینه ویژگی‌ها ($\mathcal{S}^*$):", "paragraph")
display_academic_text(r"$$\mathcal{S}^* = \arg\max_{\mathcal{S} \subseteq \mathcal{F}, |\mathcal{S}| \le k} J(f_{\theta}(X_{\mathcal{S}}), Y)$$", "paragraph")

display_academic_text("۲. قاعده اضافه کردن حریصانه در روش انتخاب رو به جلو (Sequential Forward Selection - SFS):", "paragraph")
display_academic_text(r"$$x^+ = \arg\max_{x \in \mathcal{F} \setminus \mathcal{S}_t} J(\mathcal{S}_t \cup \{x\}), \quad \mathcal{S}_{t+1} = \mathcal{S}_t \cup \{x^+\}$$", "paragraph")

# ==============================================================================
# 2. NETFLIX USER ENGAGEMENT DATASET SIMULATION
# ==============================================================================
np.random.seed(42)
n_users = 600

# Informative behavioral features
completion_rate = np.random.uniform(0.1, 1.0, size=n_users)
rewatch_frequency = np.random.poisson(lam=2.5, size=n_users)
weekend_ratio = np.random.uniform(0.2, 0.9, size=n_users)
search_success_rate = np.random.uniform(0.3, 1.0, size=n_users)
active_devices = np.random.choice([1, 2, 3, 4], size=n_users, p=[0.4, 0.3, 0.2, 0.1])

# Non-informative noise features (We expect SFS to ignore these!)
random_noise_id = np.random.normal(0, 1, size=n_users)
ui_theme_code = np.random.choice([0, 1, 2], size=n_users)
temp_client_ping = np.random.uniform(20, 120, size=n_users)

# Ground truth probability of subscription renewal (Target = 1: Renew, 0: Churn)
# Complex multi-feature interaction logic:
log_odds = (
    3.5 * completion_rate +
    0.6 * rewatch_frequency +
    1.8 * search_success_rate +
    0.4 * active_devices -
    2.8
)
probabilities = 1.0 / (1.0 + np.exp(-log_odds))
target_renew = (np.random.rand(n_users) < probabilities).astype(int)

df_netflix = pd.DataFrame({
    'Completion_Rate': completion_rate,
    'Rewatch_Freq': rewatch_frequency,
    'Weekend_Ratio': weekend_ratio,
    'Search_Success': search_success_rate,
    'Active_Devices': active_devices,
    'Noise_Random_ID': random_noise_id,
    'Noise_UI_Theme': ui_theme_code,
    'Noise_Ping_Time': temp_client_ping
})

display_side_by_side_datasets([df_netflix.head(5)], ["دیتاست شبیه‌سازی‌شده رفتار کاربران نتفلیکس (حاوی ویژگی‌های کلیدی و نویزِ مزاحم)"])

# ==============================================================================
# 3. LOGISTIC CLASSIFIER & SFS WRAPPER FROM SCRATCH
# ==============================================================================
class LogisticRegressionScratch:
    """Logistic Regression implemented from scratch using Gradient Descent."""
    def __init__(self, lr=0.05, n_epochs=120):
        self.lr = lr
        self.n_epochs = n_epochs
        self.weights = None
        self.bias = 0.0

    def _sigmoid(self, z):
        return 1.0 / (1.0 + np.exp(-np.clip(z, -25, 25))) # Clipped for numerical stability

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0.0

        for _ in range(self.n_epochs):
            linear_model = np.dot(X, self.weights) + self.bias
            y_predicted = self._sigmoid(linear_model)

            # Gradient computation
            dw = (1.0 / n_samples) * np.dot(X.T, (y_predicted - y))
            db = (1.0 / n_samples) * np.sum(y_predicted - y)

            self.weights -= self.lr * dw
            self.bias -= self.lr * db

    def predict_proba(self, X):
        return self._sigmoid(np.dot(X, self.weights) + self.bias)

    def predict(self, X):
        return (self.predict_proba(X) >= 0.5).astype(int)

def cross_validate_score(X, y, feature_indices, k_folds=5):
    """Computes K-Fold Cross-Validation Accuracy to evaluate a specific feature subset."""
    if len(feature_indices) == 0:
        return 0.0
    
    X_sub = X[:, feature_indices]
    n_samples = len(y)
    fold_size = n_samples // k_folds
    indices = np.arange(n_samples)
    scores = []

    for k in range(k_folds):
        val_idx = indices[k * fold_size : (k + 1) * fold_size]
        train_idx = np.setdiff1d(indices, val_idx)

        X_tr, y_tr = X_sub[train_idx], y[train_idx]
        X_val, y_val = X_sub[val_idx], y[val_idx]

        # Standardize features (Mean=0, Std=1) to prevent Gradient Descent explosion
        mean = np.mean(X_tr, axis=0)
        std = np.std(X_tr, axis=0)
        std = np.where(std == 0, 1.0, std)

        X_tr_scaled = (X_tr - mean) / std
        X_val_scaled = (X_val - mean) / std

        model = LogisticRegressionScratch(lr=0.08, n_epochs=150)
        model.fit(X_tr_scaled, y_tr)
        preds = model.predict(X_val_scaled)
        
        acc = np.mean(preds == y_val)
        scores.append(acc)

    return np.mean(scores)

class SequentialForwardSelector:
    """Sequential Forward Selection (SFS) Wrapper built entirely from scratch."""
    def __init__(self, data_df, target_array):
        self.feature_names = list(data_df.columns)
        self.X = data_df.values
        self.y = target_array
        self.history = []

    def run(self):
        selected_idx = []
        remaining_idx = list(range(self.X.shape[1]))

        # We keep adding one feature at a time until all features are selected
        for step in range(len(self.feature_names)):
            best_score = -1.0
            best_candidate = None

            # Test every remaining feature by adding it to the currently selected subset
            for candidate in remaining_idx:
                test_set = selected_idx + [candidate]
                # Evaluate this subset using the ACTUAL Machine Learning Model via CV
                score = cross_validate_score(self.X, self.y, test_set)
                
                if score > best_score:
                    best_score = score
                    best_candidate = candidate

            # Permanently add the winner to the selected subset
            selected_idx.append(best_candidate)
            remaining_idx.remove(best_candidate)

            self.history.append({
                'Step': step + 1,
                'Num_Features': len(selected_idx),
                'Added_Feature': self.feature_names[best_candidate],
                'CV_Accuracy': best_score
            })

        return pd.DataFrame(self.history)

# Execute SFS Wrapper Search
wrapper_sfs = SequentialForwardSelector(df_netflix, target_renew)
df_sfs_results = wrapper_sfs.run()

# Display trace
display_side_by_side_datasets([df_sfs_results.head(8)], ["گزارش گام‌به‌گام الگوریتم SFS برای انتخاب بهترین ویژگی‌ها (حداکثر شدن دقت در گام ۴)"])

# ==============================================================================
# 4. VISUALIZATIONS AND PERFORMANCE TRACE CHARTS (ACADEMIC BLUE PALETTE)
# ==============================================================================
# Find the peak performance step
best_step = df_sfs_results.loc[df_sfs_results['CV_Accuracy'].idxmax()]
opt_k = best_step['Num_Features']
opt_acc = best_step['CV_Accuracy'] * 100

fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

#  CHART 1: SFS Accuracy Progression vs Number of Features 
axes[0].plot(
    df_sfs_results['Num_Features'], 
    df_sfs_results['CV_Accuracy'] * 100,
    marker='o', linewidth=3, color='#2980b9', markersize=9, label='5-Fold CV Accuracy'
)

# Highlight optimal feature subset point
axes[0].scatter(
    [opt_k], [opt_acc], 
    color='#e74c3c', s=250, zorder=5, edgecolors='black', marker='*', 
    label=f'Optimal Peak (k={opt_k})'
)
axes[0].axvline(opt_k, color='#c0392b', linestyle='--', linewidth=2)

axes[0].set_xlabel('Number of Selected Features', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Cross-Validation Accuracy (%)', fontsize=12, fontweight='bold')
axes[0].set_title('1. Wrapper SFS Search: Model Performance vs Feature Set Size\n(Curse of Dimensionality Demonstration)', fontsize=13, fontweight='bold', pad=12)
axes[0].set_xticks(df_sfs_results['Num_Features'])
axes[0].legend(loc='lower right', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
axes[0].grid(True, linestyle='--', alpha=0.6)

# Annotate which feature was added at each step
for _, row in df_sfs_results.iterrows():
    y_offset = -12 if row['Num_Features'] % 2 == 0 else 10
    axes[0].annotate(
        f"+{row['Added_Feature'][:12]}",
        xy=(row['Num_Features'], row['CV_Accuracy'] * 100),
        xytext=(0, y_offset), textcoords="offset points",
        ha='center', fontsize=9, fontweight='bold', color='#2c3e50', rotation=15
    )


#  CHART 2: Performance Comparison (Singles vs Full vs Optimal) 
full_set_score = df_sfs_results.iloc[-1]['CV_Accuracy'] * 100

# Calculate baseline performance of each individual feature alone
single_feature_scores = []
for i, name in enumerate(df_netflix.columns):
    sc = cross_validate_score(df_netflix.values, target_renew, [i]) * 100
    single_feature_scores.append({'Feature': name, 'Accuracy': sc})

df_singles = pd.DataFrame(single_feature_scores).sort_values(by='Accuracy', ascending=True)

comparison_names = list(df_singles['Feature']) + ['ALL Features Combined (Full Model)', f'SFS Optimal Subset ({opt_k} Features)']
comparison_values = list(df_singles['Accuracy']) + [full_set_score, opt_acc]

# Color coding: Singles = Gray/Navy, Full = Red, Optimal = Academic Blue
bar_colors = ['#bdc3c7'] * (len(df_singles) - 3) + ['#2c3e50'] * 3 + ['#e74c3c', '#2980b9']

bars = axes[1].barh(comparison_names, comparison_values, color=bar_colors, edgecolor='black', alpha=0.9, height=0.6)

axes[1].set_xlabel('5-Fold CV Accuracy (%)', fontsize=12, fontweight='bold')
axes[1].set_title('2. Comparison: Individual Features vs Full Set vs SFS Optimal Subset', fontsize=13, fontweight='bold', pad=12)
axes[1].set_xlim(50, 90)
axes[1].grid(axis='x', linestyle='--', alpha=0.7)

for bar in bars:
    w = bar.get_width()
    axes[1].annotate(
        f'{w:.1f}%', 
        xy=(w, bar.get_y() + bar.get_height()/2),
        xytext=(5, 0), textcoords="offset points", 
        ha='left', va='center', fontweight='bold', fontsize=10, color='#2c3e50'
    )

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (خطر داده‌های مازاد):</b> بررسی گراف ۱ و جدولِ گام‌به‌گام به ما درسی حیاتی در داده‌کاوی می‌دهد. "
    r"الگوریتم SFS ابتدا <code>Completion_Rate</code> را پیدا کرد، سپس متوجه شد که ترکیب آن با <code>Search_Success</code> "
    r"دقت را بالا می‌برد. این روند صعودی تا گام چهارم (ستاره قرمز) ادامه داشت و به اوج دقت رسید. اما به محض اینکه الگوریتم "
    r"مجبور شد نویزهایی مثل <code>Noise_Ping_Time</code> یا <code>UI_Theme</code> را به مدل اضافه کند، "
    r"مدل رگرسیون دچار <b>بیش‌برازش (Overfitting)</b> شد و دقت آن شروع به ریزش کرد! گراف ۲ تأیید می‌کند که «مدل کامل» (نوار قرمز) "
    r"دقت پایین‌تری نسبت به «زیرمجموعه طلایی» (نوار آبی) دارد. نتفلیکس با روش Wrapper، نه تنها سرعت پردازش سرورها را افزایش می‌دهد، "
    r"بلکه با حذف متغیرهای گیج‌کننده، توانایی کشف ریزش (Churn) کاربران را نیز به حداکثر می‌رساند."
)
display_academic_text(insight_conclusion, "insight")
```

<li>پیاده‌سازی کامل استراتژی انتخاب ویژگی لفاف‌پیچ رو به جلو (SFS) از صفر (From Scratch):
  <ul>
    <li><b>مدل رگرسیون لجستیک پایه‌ای:</b> در کلاس <code>LogisticRegressionScratch</code>، الگوریتم بهینه‌سازی گرادیان کاهشی به همراه تابع سیگموئید و گرادیان وزن‌ها و بایاس پیاده شد تا به عنوان موتور ارزیابی درون لفاف‌پیچ عمل کند.</li>
    <li><b>اعتبارسنجی متقاطع چندلایه:</b> در تابع <code>cross_validate_score</code>، تقسیم داده‌ها به ۵ بخش (5-Fold CV)، استانداردسازی مستقل در هر فولد و محاسبه میانگین دقت تست پیاده‌سازی گردید تا از نشت اطلاعات (Data Leakage) جلوگیری شود.</li>
    <li><b>موتور جستجوی حریصانه رو به جلو:</b> در کلاس <code>SequentialForwardSelector</code>، در هر مرحله تمام متغیرهای باقی‌مانده آزموده شده و بهترین متغیری که بیشترین جهش را در دقت اعتبارسنجی ایجاد کند، به زیرمجموعه افزوده شد.</li>

<li>تفسیر خروجی و نمودارهای تحلیلی:
  <ul>
