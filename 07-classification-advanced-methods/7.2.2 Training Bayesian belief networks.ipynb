<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: یادگیری پارامترهای جداول CPT و برازش توزیع احتمالات در «آمازون» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در زیرساخت ابری خدمات وب <b>آمازون</b>، پایش سلامت سرورها و پیش‌بینی قطعی سرویس‌ها (Service Outage) از طریق تحلیل لاگ‌های عملیاتی انجام می‌پذیرد. ساختار علّی وابستگی میان متغیرها (نظیر بار پردازنده، نشت حافظه، تاخیر دیسک و قطعی پایگاه داده) در قالب یک گراف مشخص است، اما مقادیر جداول احتمال شرطی (CPTs) باید به صورت مداوم از میلیون‌ها رکورد تله‌متری آموزش داده شوند. مهندسان داده آمازون با استفاده از <b>تخمین درست‌نمایی بیشینه (MLE)</b> و <b>تخمین بیزی با توزیع پیشین دیریکله (Dirichlet Smoothing)</b> پارامترهای شبکه را بدون افتادن در دام احتمالات صفر تخمین می‌زنند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی آموزش پارامترهای شبکه باور بیزی:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. تخمین درست‌نمایی بیشینه (MLE) برای پارامتر $\theta_{ijk} = P(X_i = k \mid \text{Parents}(X_i) = j)$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\hat{\theta}_{ijk}^{\text{MLE}} = \frac{N_{ijk}}{\sum_{k'=1}^{r_i} N_{ijk'}} = \frac{N_{ijk}}{N_{ij}}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. تخمین بیزی با هموارسازی لاپلاس / دیریکله برای جلوگیری از بیش‌برازش در نمونه‌های اندک:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\hat{\theta}_{ijk}^{\text{Bayes}} = \frac{N_{ijk} + \alpha_{ijk}}{N_{ij} + \sum_{k'=1}^{r_i} \alpha_{ijk'}} = \frac{N_{ijk} + \alpha}{N_{ij} + r_i \alpha}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. تابع لگاریتم درست‌نمایی (Log-Likelihood Score) کل مجموعه داده لاگ‌ها ($D$) بر روی گراف ($G$):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\log L(\Theta \mid D) = \sum_{i=1}^{n} \sum_{j=1}^{q_i} \sum_{k=1}^{r_i} N_{ijk} \log \hat{\theta}_{ijk}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. معیار ارزیابی فشردگی و اطلاعات بیزی (Bayesian Information Criterion - BIC):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{BIC}(\Theta \mid D) = \log L(\Theta \mid D) - \frac{d}{2} \log |D|, \quad d = \sum_{i=1}^n q_i(r_i - 1)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: آمازون با به‌روزرسانی برخط پارامترهای شبکه باور بیزی بر پایه شمارش‌های تجربی لاگ‌ها، سیستم‌های هشدار خودکار را با کمترین نرخ هشدار کاذب کالیبره می‌کند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: تخمین مقاوم و بدون سوگیری احتمالات شرطی از داده‌های تله‌متری با مهار احتمال صفر
    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "یادگیری پارامترهای شبکه‌ بیزی و برازش توزیع در «آمازون» (MLE & Dirichlet Smoothing) ☁️", 
    "main_title"
)

display_academic_text(
    "در دیتاسنترهای آمازون (AWS)، تشخیص زودهنگام قطعی سرویس‌ها با تحلیل لاگ‌ها انجام می‌شود. "
    "در این تمرین، دو روش تخمین بیشینه درست‌نمایی (MLE) و تخمین بیزی با هموارسازی دیریکله را از پایه برنامه‌نویسی می‌کنیم.", 
    "paragraph"
)
# ==============================================================================
# 2. AWS SERVER TELEMETRY LOGS SIMULATION (GROUND TRUTH GENERATION)
# ==============================================================================
np.random.seed(42)
n_records = 2000

p_true_cpu = 0.25
p_true_mem = 0.15
p_true_io = {0: 0.10, 1: 0.75}
p_true_outage = {
    (0, 0): 0.02,
    (0, 1): 0.45,
    (1, 0): 0.60,
    (1, 1): 0.95 
}

cpu = (np.random.rand(n_records) < p_true_cpu).astype(int)
mem = (np.random.rand(n_records) < p_true_mem).astype(int)
io_probs = np.array([p_true_io[c] for c in cpu])
io_delay = (np.random.rand(n_records) < io_probs).astype(int)
outage_probs = np.array([p_true_outage[(i, m)] for i, m in zip(io_delay, mem)])
outage = (np.random.rand(n_records) < outage_probs).astype(int)

df_aws_logs = pd.DataFrame({
    'High_CPU_Load': cpu,
    'Memory_Leak_Detected': mem,
    'Storage_IO_Delay': io_delay,
    'System_Outage': outage
})

display_side_by_side_datasets([df_aws_logs.head(5)], ["نمونه‌ای از لاگ‌های تله‌متری سرورهای AWS (شبیه‌سازی‌شده)"])

# ==============================================================================
# 3. BAYESIAN NETWORK PARAMETER LEARNING (MLE & DIRICHLET) FROM SCRATCH
# ==============================================================================
class BBNParameterLearner:
    def __init__(self, dag_structure):
        self.dag = dag_structure
        self.cpt_mle = {}
        self.cpt_bayes = {}

    def fit(self, data, alpha=1.0):
        for node, parents in self.dag.items():
            self.cpt_mle[node] = {}
            self.cpt_bayes[node] = {}

            if not parents:
                counts = data[node].value_counts().to_dict()
                total = len(data)
                n_states = 2 

                for state in [0, 1]:
                    cnt = counts.get(state, 0)
                    self.cpt_mle[node][state] = cnt / total if total > 0 else 0
                    self.cpt_bayes[node][state] = (cnt + alpha) / (total + n_states * alpha)
            else:
                n_parents = len(parents)
                
                if n_parents == 1:
                    all_combos = [(0,), (1,)]
                else:
                    all_combos = list(itertools.product([0, 1], repeat=n_parents))
                
                for combo in all_combos:
                    key = combo[0] if len(combo) == 1 else combo
                    
                    mask = np.ones(len(data), dtype=bool)
                    for p_idx, p_col in enumerate(parents):
                        val = combo[p_idx] if isinstance(combo, tuple) else combo
                        mask = mask & (data[p_col] == val)
                    
                    sub_data = data[mask]
                    n_parent_occurrences = len(sub_data)
                    n_states = 2
            
                    for state in [0, 1]:
                        cnt = np.sum(sub_data[node] == state) if n_parent_occurrences > 0 else 0
                        
                        if n_parent_occurrences > 0:
                            self.cpt_mle[node][(key, state)] = cnt / n_parent_occurrences
                        else:
                            self.cpt_mle[node][(key, state)] = 0.0 
            
                        self.cpt_bayes[node][(key, state)] = (
                            (cnt + alpha) /
                            (n_parent_occurrences + n_states * alpha)
                        )

    def compute_log_likelihood(self, data, method='bayes'):
        cpt_table = self.cpt_bayes if method == 'bayes' else self.cpt_mle
        total_log_l = 0.0

        for _, row in data.iterrows():
            for node, parents in self.dag.items():
                state = int(row[node])
                if not parents:
                    prob = cpt_table[node][state]
                else:
                    if len(parents) == 1:
                        key = int(row[parents[0]])
                    else:
                        key = tuple(int(row[p]) for p in parents)
                    
                    prob = cpt_table[node].get((key, state), 1e-12)
                
                total_log_l += np.log(max(prob, 1e-12))
                
        return total_log_l

aws_dag = {
    'High_CPU_Load': [],
    'Memory_Leak_Detected': [],
    'Storage_IO_Delay': ['High_CPU_Load'],
    'System_Outage': ['Storage_IO_Delay', 'Memory_Leak_Detected']
}

learner = BBNParameterLearner(aws_dag)
learner.fit(df_aws_logs, alpha=1.0)

# ==============================================================================
# 4. TRAINING SAMPLE SIZE CONVERGENCE SIMULATION
# ==============================================================================
sample_sizes = [30, 75, 150, 300, 600, 1200, 2000]
ll_mle = []
ll_bayes = []

val_df = df_aws_logs.iloc[1500:]

for size in sample_sizes:
    sub_df = df_aws_logs.iloc[:size]
    sub_learner = BBNParameterLearner(aws_dag)
    sub_learner.fit(sub_df, alpha=1.0)
    
    ll_mle.append(sub_learner.compute_log_likelihood(val_df, method='mle') / len(val_df))
    ll_bayes.append(sub_learner.compute_log_likelihood(val_df, method='bayes') / len(val_df))

# ==============================================================================
# 5. VISUALIZATIONS AND ESTIMATION CHARTS (ACADEMIC BLUE PALETTE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

#  CHART 1: Parameter Estimation Accuracy P(Outage = 1 | IO, Mem) 
parent_labels = ['Normal\n(IO=0, Mem=0)', 'Leak Only\n(IO=0, Mem=1)', 'Delay Only\n(IO=1, Mem=0)', 'Critical\n(IO=1, Mem=1)']
keys = [(0, 0), (0, 1), (1, 0), (1, 1)]

true_vals = [p_true_outage[k] for k in keys]
mle_vals = [learner.cpt_mle['System_Outage'][(k, 1)] for k in keys]
bayes_vals = [learner.cpt_bayes['System_Outage'][(k, 1)] for k in keys]

x = np.arange(len(parent_labels))
width = 0.26

r1 = axes[0].bar(x - width, true_vals, width, label='Ground Truth Base Probability', color='#2c3e50', edgecolor='black', alpha=0.95)
r2 = axes[0].bar(x, mle_vals, width, label='MLE Calculation (No Prior)', color='#e74c3c', edgecolor='black', alpha=0.85)
r3 = axes[0].bar(x + width, bayes_vals, width, label=r'Bayesian Dirichlet Smoothing ($\alpha=1$)', color='#2980b9', edgecolor='black', alpha=0.9)

axes[0].set_ylabel('Conditional Probability $P(Outage = 1 | Parents)$', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Parent States Configuration (IO Delay, Memory Leak)', fontsize=12, fontweight='bold')
axes[0].set_title('1. Bayesian Network CPT Learning Comparison\n(Ground Truth vs MLE vs Bayesian Estimates)', fontsize=13, fontweight='bold', pad=12)
axes[0].set_xticks(x)
axes[0].set_xticklabels(parent_labels, fontsize=11, fontweight='bold')
axes[0].set_ylim(0, 1.15)
axes[0].legend(loc='upper left', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
axes[0].grid(axis='y', linestyle='--', alpha=0.6)

for rect in r2:
    h = rect.get_height()
    axes[0].annotate(
        f'{h:.2f}', 
        xy=(rect.get_x() + rect.get_width()/2, h),
        xytext=(0, 4), textcoords="offset points", 
        ha='center', va='bottom', fontsize=9, fontweight='bold', color='#c0392b'
    )


#  CHART 2: Log-Likelihood Convergence vs Log Data Volume 
axes[1].plot(
    sample_sizes, ll_bayes, 
    marker='o', linewidth=3.5, color='#2980b9', markersize=9, 
    label='Bayesian Learner (Smooth, robust to zero counts)'
)
axes[1].plot(
    sample_sizes, ll_mle, 
    marker='s', linewidth=2.5, linestyle='--', color='#e74c3c', markersize=8, 
    label='MLE Learner (Fails on unseen validation events)'
)

axes[1].set_xlabel('Amount of Training Log Records Provided ($N$)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Validation Log-Likelihood Score (Higher is Better)', fontsize=12, fontweight='bold')
axes[1].set_title('2. Learning Curve & Generalization to Unseen Data\n(Notice MLE crash when data is scarce)', fontsize=13, fontweight='bold', pad=12)
axes[1].legend(loc='lower right', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=11)
axes[1].grid(True, linestyle='--', alpha=0.6)
axes[1].set_xscale('log')
axes[1].set_xticks(sample_sizes)
axes[1].set_xticklabels(sample_sizes, fontweight='bold')

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 6: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (برتری بیزی در داده‌های کم):</b> گراف ۲ قدرت مطلقِ آمار بیزی را در شرایط بحرانی نشان می‌دهد. "
    r"وقتی فقط ۳۰ تا ۷۵ لاگ آموزشی داشتیم (سمت چپ نمودار)، مدل <b>MLE</b> به شدت سقوط کرد (خط قرمز). چرا؟ "
    r"چون در این لاگ‌های محدود، اتفاق نیفتاده بود که «هم تاخیر IO داشته باشیم و هم نشت حافظه». MLE احتمالِ این "
    r"سناریو را <b>دقیقاً صفر</b> در نظر گرفت. وقتی در داده‌های Validation این اتفاقِ نادر رخ داد، لگاریتمِ صفر بی‌نهایت "
    r"منفی شد و سیستم هشدار MLE به طور کامل از کار افتاد! اما مدل <b>Bayesian</b> با تکنیک هموارسازی "
    r"دیریکله (اضافه کردن فرضیه $\alpha=1$)، هرگز به هیچ رخدادی احتمال صفر مطلق نداد. در نتیجه خط آبی از همان روزهای اولِ "
    r"کسب‌وکار مقاومت کرد و با دریافت داده‌های بیشتر، هر دو مدل به توافق رسیدند."
)
display_academic_text(insight_conclusion, "insight")
```

<li>پیاده‌سازی کامل موتور یادگیری پارامترهای شبکه باور بیزی از صفر (From Scratch):

    <li><b>هموارسازی بیزی دیریکله (Laplace Smoothing):</b> فرمول تخمین پارامتر با مقدار شبه‌شمارش $\alpha = 1.0$ پیاده شد تا در صورت عدم مشاهده یک ترکیب خاص از والدین، احتمال خروجی به صفر تبدیل نگردد و مدل از پدیده واگرایی لگاریتمی محافظت شود.</li>
    <li><b>محاسبه امتیاز لگاریتم درست‌نمایی (Log-Likelihood):</b> در متد <code>compute_log_likelihood</code>، میزان برازش پارامترهای آموزش‌دیده روی رکوردهای اعتبارسنجی ارزیابی گردید.</li>
  </ul>
</li>
<li>تفسیر خروجی و نمودارهای تحلیلی:
  <ul>
    <li><b>نمودار مقایسه‌ای چپ (CPT Learning Accuracy):</b> دقت فوق‌العاده تخمین پارامترهای CPT را برای گره تصمیم‌گیری قطعی سرور (Outage) نشان می‌دهد؛ مقادیر استخراج‌شده با هر دو روش MLE و بیزی بر مقادیر واقعی (Ground Truth) منطبق شده‌اند.</li>
    <li><b>نمودار یادگیری و همگرایی راست (Generalization Curve):</b> پایداری روش بیزی در حجم داده‌های بسیار کوچک (زیر ۱۰۰ نمونه) را اثبات می‌کند؛ در نمونه‌های کم، روش MLE به دلیل احتمال صفر دچار خطای تعمیم می‌شود در حالی که روش بیزی همگرایی نرم و پیوسته‌ای به سمت درست‌نمایی بیشینه به نمایش می‌گذارد.</li>
