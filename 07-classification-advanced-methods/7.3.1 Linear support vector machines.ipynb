<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: تفکیک بیشینه حاشیه و طبقه‌بندی خطی در «اوبر» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در سامانه مدیریت کیفیت و ایمنی ناوگان <b>اوبر</b>، ارزیابی بلادرنگ رفتار رانندگان بر اساس شاخص‌های تله‌متری (مانند انحراف معیار سرعت، شتاب‌های ناگهانی و ترمزهای شدید) برای تفکیک رانندگان ایمن از رانندگان پرخطر انجام می‌شود. چالش اصلی در این طبقه‌بندی، یافتن مرز تصمیمی است که علاوه بر جداسازی صحیح دو دسته، بیشترین فاصله اطمینان (Margin) را نسبت به داده‌های مرزی حفظ کند تا مدل در مواجهه با خطاهای جزئی GPS و نویز سنسورها دچار خطا نشود. مهندسان اوبر برای این منظور از <b>ماشین بردار پشتیبان خطی (Linear SVM)</b> با هدف بیشینه‌سازی حاشیه هندسی استفاده می‌کنند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی ماشین بردار پشتیبان خطی با حاشیه نرم (Soft Margin SVM):
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. فرم اولیه مسئله بهینه‌سازی هموار محدب (Primal Optimization Problem):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\min_{\mathbf{w}, b, \boldsymbol{\xi}} \frac{1}{2} \|\mathbf{w}\|_2^2 + C \sum_{i=1}^{N} \xi_i \quad \text{subject to} \quad y_i(\mathbf{w}^T \mathbf{x}_i + b) \ge 1 - \xi_i, \quad \xi_i \ge 0$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. بازنویسی تابع هزینه بر اساس تابع زیان لولا (Hinge Loss):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$J(\mathbf{w}, b) = \frac{1}{2} \|\mathbf{w}\|_2^2 + C \sum_{i=1}^{N} \max\left(0, 1 - y_i(\mathbf{w}^T \mathbf{x}_i + b)\right)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. پهنای حاشیه هندسی (Geometric Margin) میان دو ابرصفحه پشتیبان:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{Margin} = \frac{2}{\|\mathbf{w}\|_2}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. قاعده گرادیان فرعی (Subgradient) برای به‌روزرسانی وزن‌ها:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\nabla_{\mathbf{w}} J = \mathbf{w} - C \sum_{i: y_i(\mathbf{w}^T \mathbf{x}_i + b) < 1} y_i \mathbf{x}_i, \quad \nabla_b J = - C \sum_{i: y_i(\mathbf{w}^T \mathbf{x}_i + b) < 1} y_i$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: اوبر با متمرکز ساختن مرز تصمیم بر روی بحرانی‌ترین نمونه‌ها (بردارهای پشتیبان)، مدلی پایدار و با بیشترین قابلیت تعمیم به دست می‌آورد که در برابر نوسانات ترافیکی و خطاهای داده‌ای مقاوم است.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: جداسازی بهینه کلاس‌ها با ایجاد حداکثر حاشیه اطمینان و استخراج بردارهای پشتیبان
    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "تفکیک بیشینه حاشیه و طبقه‌بندی خطی در «اوبر» (Soft-Margin SVM From Scratch) 🚕", 
    "main_title"
)

display_academic_text(
    "در سامانه ایمنی ناوگان اوبر، ارزیابی رفتار رانندگان بر اساس شاخص‌های تله‌متری (مثل واریانس سرعت) برای "
    "تفکیک رانندگان ایمن از پرخطر انجام می‌شود. ماشین بردار پشتیبان (SVM) با ایجاد <b>بیشترین حاشیه (Max-Margin)</b> "
    "اطمینان حاصل می‌کند که خطاهای جزئی GPS باعث طبقه‌بندی اشتباه نشوند. در این تمرین، این الگوریتم قدرتمند "
    "را به همراه توابع Hinge Loss و نزول زیرگرادیان (Subgradient Descent) مستقیماً کدنویسی می‌کنیم.", 
    "paragraph"
)

display_academic_text("فرمول‌های پایه در یادگیری SVM:", "sub_title")

display_academic_text("۱. تابع هزینه (Primal Objective) بر اساس تابع زیان لولا (Hinge Loss) و جریمه L2:", "paragraph")
display_academic_text(r"$$J(\mathbf{w}, b) = \frac{1}{2} \|\mathbf{w}\|_2^2 + C \sum_{i=1}^{N} \max\left(0, 1 - y_i(\mathbf{w}^T \mathbf{x}_i + b)\right)$$", "paragraph")

display_academic_text("۲. محاسبه پهنای حاشیه هندسی (Geometric Margin) میان دو ابرصفحه پشتیبان:", "paragraph")
display_academic_text(r"$$\text{Margin} = \frac{2}{\|\mathbf{w}\|_2}$$", "paragraph")

# ==============================================================================
# 2. UBER DRIVER TELEMETRY DATASET GENERATION
# ==============================================================================
np.random.seed(42)
n_per_class = 60

# Class +1: Safe / Standard Drivers (Low speed variance, low harsh braking)
mean_safe = [2.2, 2.5]
cov_safe = [[0.4, 0.15], [0.15, 0.4]]
X_safe = np.random.multivariate_normal(mean_safe, cov_safe, n_per_class)
y_safe = np.ones(n_per_class)

# Class -1: High-Risk Drivers (High speed variance, frequent harsh braking)
mean_risky = [4.8, 5.0]
cov_risky = [[0.5, 0.1], [0.1, 0.5]]
X_risky = np.random.multivariate_normal(mean_risky, cov_risky, n_per_class)
y_risky = -np.ones(n_per_class)

# Combine into single dataset
X = np.vstack((X_safe, X_risky))
y = np.hstack((y_safe, y_risky))

# Shuffle dataset
shuffle_idx = np.random.permutation(len(y))
X = X[shuffle_idx]
y = y[shuffle_idx]
# ==============================================================================
# 3. LINEAR SUPPORT VECTOR MACHINE (SUBGRADIENT DESCENT) FROM SCRATCH
# ==============================================================================
class LinearSVMScratch:
    """Soft-Margin Linear Support Vector Machine built from scratch."""

    def __init__(self, C=1.0, learning_rate=0.001, n_epochs=1000):
        self.C = C
        self.lr = learning_rate
        self.n_epochs = n_epochs
        self.weights = None
        self.bias = 0.0
        self.loss_history = []
        self.support_vectors_idx = []

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0.0
        self.loss_history = []

        for epoch in range(self.n_epochs):
            # Compute geometric margins for all samples: y_i * (w^T x_i + b)
            linear_outputs = np.dot(X, self.weights) + self.bias
            margins = y * linear_outputs
            
            # Hinge loss: max(0, 1 - margin)
            hinge_losses = np.maximum(0, 1 - margins)
            
            # Total Primal Objective = Regularization + C * Loss
            total_loss = 0.5 * np.dot(self.weights, self.weights) + self.C * np.sum(hinge_losses)
            self.loss_history.append(total_loss)

            # Subgradient Descent initialization
            grad_w = self.weights.copy()
            grad_b = 0.0

            # Compute Subgradient ONLY for samples that violate the margin (margin < 1)
            for i in range(n_samples):
                if margins[i] < 1:
                    grad_w -= self.C * y[i] * X[i]
                    grad_b -= self.C * y[i]

            # Parameter updates
            self.weights -= self.lr * grad_w
            self.bias -= self.lr * grad_b

        # Identify Support Vectors: points lying exactly on or inside the margin
        final_margins = y * (np.dot(X, self.weights) + self.bias)
        self.support_vectors_idx = np.where(final_margins <= 1.01)[0]
        return self

    def decision_function(self, X):
        return np.dot(X, self.weights) + self.bias

    def predict(self, X):
        return np.sign(self.decision_function(X))


# Train Linear SVM from scratch
svm = LinearSVMScratch(C=1.5, learning_rate=0.0008, n_epochs=1200)
svm.fit(X, y)

# Calculate Geometric Margin Width (2 / ||w||)
w_norm = np.linalg.norm(svm.weights)
margin_width = 2.0 / w_norm

# ==============================================================================
# 4. VISUALIZATIONS AND DECISION BOUNDARY PLOTS (ACADEMIC BLUE PALETTE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

#  CHART 1: Optimal Hyperplane, Margins, and Support Vectors 
x_min, x_max = X[:, 0].min() - 0.8, X[:, 0].max() + 0.8
y_min, y_max = X[:, 1].min() - 0.8, X[:, 1].max() + 0.8
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 300), np.linspace(y_min, y_max, 300))
grid_points = np.c_[xx.ravel(), yy.ravel()]
Z = svm.decision_function(grid_points).reshape(xx.shape)

# Draw decision boundary and margin contours (Academic Teal and Navy)
axes[0].contour(
    xx, yy, Z, levels=[-1, 0, 1], linestyles=['--', '-', '--'],
    colors=['#2c3e50', '#2980b9', '#16a085'], linewidths=[2, 3, 2]
)

# Fill margin area (The "Street")
axes[0].contourf(xx, yy, Z, levels=[-1, 1], colors=['#ecf0f1'], alpha=0.5)

# Plot Data Points
safe_mask = (y == 1)
risky_mask = (y == -1)

axes[0].scatter(X[safe_mask, 0], X[safe_mask, 1], color='#16a085', s=60, edgecolors='black', label='Safe Drivers (+1)', zorder=3)
axes[0].scatter(X[risky_mask, 0], X[risky_mask, 1], color='#2c3e50', s=60, edgecolors='black', label='Risky Drivers (-1)', zorder=3)

# Highlight Support Vectors (Red Circles)
sv_idx = svm.support_vectors_idx
axes[0].scatter(
    X[sv_idx, 0], X[sv_idx, 1], s=180, facecolors='none', edgecolors='#e74c3c', 
    linewidths=2.0, label='Support Vectors', zorder=4
)

axes[0].set_title(f'1. Uber Linear SVM Decision Boundary & Margin (Width: {margin_width:.2f})', fontsize=13, fontweight='bold', pad=12)
axes[0].set_xlabel('Normalized Speed Variance Metric ($x_1$)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Harsh Braking Frequency Metric ($x_2$)', fontsize=12, fontweight='bold')
axes[0].legend(loc='lower right', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
axes[0].grid(True, linestyle='--', alpha=0.6)

# Annotate hyperplanes with 'r' strings to prevent LaTeX errors
axes[0].text(x_min + 0.2, y_max - 0.4, r'$\mathbf{w}^T \mathbf{x} + b = 0$ (Boundary)', fontsize=10, fontweight='bold', color='#2980b9')
axes[0].text(x_min + 0.2, y_max - 0.8, r'$\mathbf{w}^T \mathbf{x} + b = +1$ (Margin)', fontsize=10, fontweight='bold', color='#16a085')
axes[0].text(x_min + 0.2, y_max - 1.2, r'$\mathbf{w}^T \mathbf{x} + b = -1$ (Margin)', fontsize=10, fontweight='bold', color='#2c3e50')


#  CHART 2: Regularized Hinge Loss Convergence 
axes[1].plot(range(len(svm.loss_history)), svm.loss_history, color='#2980b9', linewidth=3, label='Primal Cost $J(\mathbf{w}, b)$')

axes[1].set_title('2. Linear SVM Subgradient Descent Loss Convergence Curve', fontsize=13, fontweight='bold', pad=12)
axes[1].set_xlabel('Training Iteration (Epochs)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Primal Objective Loss Value', fontsize=12, fontweight='bold')
axes[1].legend(loc='upper right', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=11)
axes[1].grid(True, linestyle='--', alpha=0.7)

# Highlight convergence annotation
final_loss = svm.loss_history[-1]
axes[1].annotate(
    f'Converged Loss: {final_loss:.2f}',
    xy=(len(svm.loss_history) - 1, final_loss),
    xytext=(-160, 40), textcoords="offset points",
    arrowprops=dict(arrowstyle="->", color='#e74c3c', lw=2),
    fontsize=11, fontweight='bold', color='#2c3e50'
)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (اهمیت بردارهای پشتیبان):</b> به گراف ۱ نگاه کنید؛ دایره‌های قرمزرنگ که دور برخی نقاط کشیده شده‌اند، "
    r"همان <b>بردارهای پشتیبان (Support Vectors)</b> هستند. این‌ها نقاطی هستند که روی خط‌چین حاشیه قرار دارند یا از آن عبور کرده‌اند "
    r"(Margin < 1). نکته جادویی ماشین‌های بردار پشتیبان این است که تمام رانندگانی که کاملاً ایمن (بالای خط‌چین سبز) یا "
    r"کاملاً خطرناک (پایین خط‌چین سرمه‌ای) هستند، از فرآیند محاسبه شیب (گرادیان) <b>حذف می‌شوند</b>! مدل SVM اوبر فقط و فقط "
    r"به داده‌های لبه مرز و گیج‌کننده توجه می‌کند تا مرز (خیابان خاکستری) را در ایمن‌ترین نقطه ممکن رسم کند. "
    r"این یعنی مدل ما در برابر هرگونه نویز اضافی در داده‌های واضح، کاملاً مقاوم (Robust) است."
)
display_academic_text(insight_conclusion, "insight")
```
