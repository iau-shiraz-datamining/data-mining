    مثال عملی: طبقه‌بندی توالی تعاملات کاربران و مدل‌های بازگشتی در «نتفلیکس» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در سرویس پخش ویدیوی <b>نتفلیکس</b>، هر سشن تعاملی کاربر شامل توالی مرتبی از رویدادهای زمانی (مانند جستجوی عنوان، پیش‌نمایش تریلر، توقف پخش، تغییر ژانر و شروع تماشای قسمت کامل) است. نادیده گرفتن ترتیب زمانی این رخدادها باعث از دست رفتن اطلاعات وابستگی متوالی می‌گردد. مهندسان داده نتفلیکس با استفاده از <b>طبقه‌بندی توالی (Sequence Classification)</b> و معماری‌های بازگشتی (RNNs)، وضعیت ذهنی کاربر را گام‌به‌گام در قالب بردار حالت پنهان (Hidden State) به‌روزرسانی می‌کنند تا در همان دقایق ابتدایی پیش‌بینی کنند که آیا کاربر وارد حالت تماشای پیوسته (Binge Watching) می‌شود یا سشن را به دلیل عدم رضایت ترک خواهد کرد (Session Abandonment).
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی شبکه‌های عصبی بازگشتی و طبقه‌بندی توالی:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. به‌روزرسانی بردار حالت پنهان در گام زمانی $t$ بر اساس ورودی فعلی ($\mathbf{x}_t$) و حافظه پیشین ($\mathbf{h}_{t-1}$):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\mathbf{h}_t = \tanh\left( \mathbf{W}_{hh} \mathbf{h}_{t-1} + \mathbf{W}_{xh} \mathbf{x}_t + \mathbf{b}_h \right)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. احتمال پیش‌بینی کلاس در آخرین گام زمانی توالی ($T$):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\hat{y} = \sigma\left( \mathbf{W}_{hy} \mathbf{h}_T + b_y \right) = \frac{1}{1 + \exp\left(-(\mathbf{W}_{hy} \mathbf{h}_T + b_y)\right)}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. تابع زیان آنتروپی متقاطع دودویی برای کل مجموعه توالی‌های دسته‌ای ($N$):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\mathcal{L} = -\frac{1}{N} \sum_{i=1}^{N} \left[ y_i \log \hat{y}_i + (1 - y_i) \log (1 - \hat{y}_i) \right]$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. پس‌انتشار خطا در طول زمان (Backpropagation Through Time - BPTT) برای ماتریس وزن بازگشتی ($\mathbf{W}_{hh}$):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\frac{\partial \mathcal{L}}{\partial \mathbf{W}_{hh}} = \sum_{t=1}^{T} \frac{\partial \mathcal{L}}{\partial \mathbf{h}_T} \left( \prod_{j=t+1}^{T} \frac{\partial \mathbf{h}_j}{\partial \mathbf{h}_{j-1}} \right) \frac{\partial \mathbf{h}_t}{\partial \mathbf{W}_{hh}}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: نتفلیکس با ردگیری پویای بردار حالت پنهان در طول توالی کلیک‌ها، رفتار ریزش را پیش از خروج کاربر تشخیص داده و پیشنهادهای شخصی‌سازی‌شده لحظه‌ای ارائه می‌دهد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: ثبت وابستگی‌های زمانی متغیرها و طبقه‌بندی دقیق توالی‌های رفتاری با حداقل خطای محاسباتی
    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "طبقه‌بندی توالی تعاملات کاربران در «نتفلیکس» (Recurrent Neural Networks - RNN) 🎬", 
    "main_title"
)

display_academic_text(
    "در پلتفرم نتفلیکس، هر کاربر در طولِ یک سشن تعاملی، مجموعه‌ای از رفتارها را در طول زمان نشان می‌دهد. "
    "برای پیش‌بینی اینکه کاربر در پلتفرم می‌ماند (Binge Watch) یا پلتفرم را ترک می‌کند (Abandon)، ترتیب "
    "زمانی (Sequence) این کارها بسیار مهم است. مدل‌های سنتی حافظه ندارند و ترتیب را درک نمی‌کنند. ما با "
    "کدنویسی یک <b>شبکه عصبی بازگشتی (RNN) از صفر</b> و نوشتن الگوریتم پس‌انتشار خطا در زمان (BPTT)، به ماشین "
    "حافظه‌ای می‌دهیم تا با استفاده از وضعیت پنهان (Hidden State) گذشته را به یاد بسپارد.", 
    "paragraph"
)

display_academic_text("معادلات پایه در شبکه‌های بازگشتی و BPTT:", "sub_title")

display_academic_text("۱. به‌روزرسانی بردار حالت پنهان در گام زمانی $t$ بر اساس ورودی فعلی ($\mathbf{x}_t$) و حافظه پیشین ($\mathbf{h}_{t-1}$):", "paragraph")
display_academic_text(r"$$\mathbf{h}_t = \tanh\left( \mathbf{W}_{hh} \mathbf{h}_{t-1} + \mathbf{W}_{xh} \mathbf{x}_t + \mathbf{b}_h \right)$$", "paragraph")

display_academic_text("۲. پس‌انتشار خطا در طول زمان (Backpropagation Through Time - BPTT) برای آپدیت ماتریس وزن‌ها:", "paragraph")
display_academic_text(r"$$\frac{\partial \mathcal{L}}{\partial \mathbf{W}_{hh}} = \sum_{t=1}^{T} \frac{\partial \mathcal{L}}{\partial \mathbf{h}_T} \left( \prod_{j=t+1}^{T} \frac{\partial \mathbf{h}_j}{\partial \mathbf{h}_{j-1}} \right) \frac{\partial \mathbf{h}_t}{\partial \mathbf{W}_{hh}}$$", "paragraph")

# ==============================================================================
# 2. NETFLIX USER SESSION INTERACTION SEQUENCE GENERATION
# ==============================================================================
np.random.seed(42)
n_sequences = 500
seq_length = 6  # T = 6 interaction steps per session
input_dim = 3   # Features per step: [Action_Intensity, Preview_Duration, Dwell_Time]

# Generate synthetic behavioral temporal sequences
# Class 1: Binge-Watching Session (Escalating engagement over time)
# Class 0: Drop-off / Session Abandonment (Decaying engagement, erratic quick exits)

X_data = np.zeros((n_sequences, seq_length, input_dim))
y_data = np.zeros(n_sequences, dtype=int)

for i in range(n_sequences):
    label = np.random.choice([0, 1], p=[0.5, 0.5])
    y_data[i] = label
    
    for t in range(seq_length):
        if label == 1:
            # Positive temporal trend: engagement builds up along the sequence
            intensity = 0.3 * t + np.random.normal(1.2, 0.25)
            preview = 0.4 * t + np.random.normal(2.0, 0.3)
            dwell = 0.5 * t + np.random.normal(3.0, 0.4)
        else:
            # Negative temporal trend: engagement decays, user loses interest
            intensity = -0.3 * t + np.random.normal(2.5, 0.3)
            preview = -0.2 * t + np.random.normal(1.8, 0.3)
            dwell = -0.4 * t + np.random.normal(2.8, 0.4)
            
        X_data[i, t, :] = [intensity, preview, dwell]

# Split train/test (75% train, 25% test)
n_train = 375
X_train, y_train = X_data[:n_train], y_data[:n_train]
X_test, y_test = X_data[n_train:], y_data[n_train:]

# ==============================================================================
# 3. RECURRENT NEURAL NETWORK (VANILLA ELMAN RNN) FROM SCRATCH
# ==============================================================================
class VanillaRNNSequenceClassifier:
    """Elman Recurrent Neural Network with BPTT implemented entirely from scratch using NumPy."""

    def __init__(self, input_dim=3, hidden_dim=8, lr=0.015):
        self.input_dim = input_dim
        self.hidden_dim = hidden_dim
        self.lr = lr

        # Xavier/He parameter initialization for stability
        self.W_xh = np.random.randn(hidden_dim, input_dim) * np.sqrt(2.0 / input_dim)
        self.W_hh = np.random.randn(hidden_dim, hidden_dim) * np.sqrt(2.0 / hidden_dim)
        self.b_h = np.zeros((hidden_dim, 1))

        self.W_hy = np.random.randn(1, hidden_dim) * np.sqrt(2.0 / hidden_dim)
        self.b_y = 0.0

    @staticmethod
    def _sigmoid(z):
        return 1.0 / (1.0 + np.exp(-np.clip(z, -25, 25)))

    def forward(self, sequence):
        """Forward pass across temporal sequence, building and storing internal hidden states over time."""
        T = sequence.shape[0]
        hidden_states = np.zeros((T + 1, self.hidden_dim, 1))
        # h_0 is initialized to zeros (no memory at the start)
        hidden_states[0] = np.zeros((self.hidden_dim, 1))

        for t in range(T):
            x_t = sequence[t].reshape(-1, 1)
            # Core RNN Equation: h_t = tanh(W_hh * h_{t-1} + W_xh * x_t + b_h)
            linear = np.dot(self.W_hh, hidden_states[t]) + np.dot(self.W_xh, x_t) + self.b_h
            hidden_states[t + 1] = np.tanh(linear)

        # Final step classification output: We only make a prediction after reading the WHOLE sequence
        h_T = hidden_states[T]
        output_logit = np.dot(self.W_hy, h_T) + self.b_y
        y_prob = self._sigmoid(output_logit)[0, 0]

        return y_prob, hidden_states

    def backward(self, sequence, y_true, y_prob, hidden_states):
        """Backpropagation Through Time (BPTT) computing chain-rule gradients across unrolled time steps."""
        T = sequence.shape[0]
        
        # Loss gradient with respect to output logit (Derivative of Binary Cross-Entropy)
        d_logit = y_prob - y_true  # Scalar
        
        dW_hy = d_logit * hidden_states[T].T
        db_y = d_logit

        # Gradient with respect to final hidden state h_T
        dh_t = np.dot(self.W_hy.T, d_logit)

        dW_hh = np.zeros_like(self.W_hh)
        dW_xh = np.zeros_like(self.W_xh)
        db_h = np.zeros_like(self.b_h)

        # Unroll backwards through time (The essence of BPTT)
        for t in reversed(range(1, T + 1)):
            x_t = sequence[t - 1].reshape(-1, 1)
            
            # Derivative of tanh: 1 - h_t^2
            dtanh = (1.0 - hidden_states[t] ** 2) * dh_t

            db_h += dtanh
            dW_xh += np.dot(dtanh, x_t.T)
            # Accumulate gradients for the recurrent weight matrix
            dW_hh += np.dot(dtanh, hidden_states[t - 1].T)

            # Propagate gradient back to the previous time step: h_{t-1}
            dh_t = np.dot(self.W_hh.T, dtanh)

        # Clip gradients to prevent Exploding Gradients (A common issue in RNNs)
        for grad in [dW_xh, dW_hh, db_h, dW_hy]:
            np.clip(grad, -4.0, 4.0, out=grad)

        return dW_xh, dW_hh, db_h, dW_hy, db_y

    def train(self, X_train, y_train, epochs=70):
        """Trains RNN via online sequence gradient descent."""
        history = []
        n_samples = len(X_train)

        for epoch in range(epochs):
            total_loss = 0.0
            correct = 0

            # Shuffle training set to prevent cyclic biases
            perm = np.random.permutation(n_samples)
            for idx in perm:
                seq = X_train[idx]
                target = y_train[idx]

                # 1. Forward pass
                prob, states = self.forward(seq)
                
                # 2. Binary Cross Entropy Loss
                loss = - (target * np.log(max(prob, 1e-12)) + (1 - target) * np.log(max(1.0 - prob, 1e-12)))
                total_loss += loss

                pred = 1 if prob >= 0.5 else 0
                if pred == target:
                    correct += 1

                # 3. Backward pass (BPTT)
                dW_xh, dW_hh, db_h, dW_hy, db_y = self.backward(seq, target, prob, states)

                # 4. SGD Weight Updates
                self.W_xh -= self.lr * dW_xh
                self.W_hh -= self.lr * dW_hh
                self.b_h -= self.lr * db_h
                self.W_hy -= self.lr * dW_hy
                self.b_y -= self.lr * db_y

            epoch_loss = total_loss / n_samples
            epoch_acc = (correct / n_samples) * 100
            history.append({'Epoch': epoch + 1, 'Loss': epoch_loss, 'Accuracy': epoch_acc})

        return pd.DataFrame(history)

    def evaluate(self, X_test, y_test):
        """Evaluates model accuracy on unseen test sequences."""
        correct = 0
        probs_all = []
        for i in range(len(X_test)):
            prob, _ = self.forward(X_test[i])
            pred = 1 if prob >= 0.5 else 0
            if pred == y_test[i]:
                correct += 1
            probs_all.append(prob)
        return (correct / len(y_test)) * 100, np.array(probs_all)

# Train RNN model from scratch
rnn = VanillaRNNSequenceClassifier(input_dim=3, hidden_dim=8, lr=0.012)
df_training = rnn.train(X_train, y_train, epochs=65)
test_accuracy, test_probs = rnn.evaluate(X_test, y_test)

# Output Summary to a Table
df_summary = pd.DataFrame([
    {"Model Component": "Recurrent Matrix ($W_{hh}$)", "Shape": f"{rnn.hidden_dim}x{rnn.hidden_dim}", "Function": "Passes memory from $t-1$ to $t$"},
    {"Model Component": "Input Matrix ($W_{xh}$)", "Shape": f"{rnn.hidden_dim}x{rnn.input_dim}", "Function": "Processes new events at time $t$"},
    {"Model Component": "BPTT Engine", "Steps Unrolled": f"{seq_length}", "Function": "Calculates error gradients backwards in time"},
    {"Model Component": "Final Output", "Accuracy": f"{test_accuracy:.1f}%", "Function": "Predicts Session Binge vs Drop-off"}
])
display_side_by_side_datasets([df_summary], ["خلاصه معماری شبکه بازگشتی و نتایج آموزش (با پیاده‌سازی ۱۰۰٪ پایه و بدون فریم‌ورک)"])

# Extract internal hidden state activations for an Engaged vs Abandoned session to visualize memory
sample_pos = X_test[np.where(y_test == 1)[0][0]]
sample_neg = X_test[np.where(y_test == 0)[0][0]]

_, states_pos = rnn.forward(sample_pos)  # Shape: (T+1, H, 1)
_, states_neg = rnn.forward(sample_neg)

# ==============================================================================
# 4. VISUALIZATIONS AND SEQUENCE DYNAMICS PLOTS (ACADEMIC BLUE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

#  CHART 1: BPTT Training Convergence Curve (Loss and Accuracy) 
ax1 = axes[0]
ax2 = ax1.twinx()

line1 = ax1.plot(df_training['Epoch'], df_training['Loss'], color='#e74c3c', linewidth=3, label='BPTT Training Loss')
line2 = ax2.plot(df_training['Epoch'], df_training['Accuracy'], color='#2980b9', linewidth=3, linestyle='--', label='Training Accuracy (%)')

ax1.set_xlabel('Training Epochs (Full BPTT Passes)', fontsize=12, fontweight='bold')
ax1.set_ylabel('Binary Cross-Entropy Loss', color='#c0392b', fontsize=12, fontweight='bold')
ax2.set_ylabel('Training Accuracy (%)', color='#2980b9', fontsize=12, fontweight='bold')

lines = line1 + line2
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='center right', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
ax1.set_title(f'1. Netflix Vanilla RNN Training Dynamics\n(Test Set Generalization Accuracy: {test_accuracy:.1f}%)', fontsize=13, fontweight='bold', pad=12)
ax1.grid(True, linestyle='--', alpha=0.6)

#  CHART 2: Temporal Hidden State Trajectory Evolution Heatmap 
# Combine states over time step 1 to T for both samples
time_steps = [f"Step t={t}" for t in range(1, seq_length + 1)]

# Extract actual neuron values and transpose
h_pos_matrix = states_pos[1:].squeeze().T  # (Hidden_Dim, T)
h_neg_matrix = states_neg[1:].squeeze().T

combined_state_matrix = np.vstack([h_pos_matrix, h_neg_matrix])
row_labels = [f"Binge Neuron {i+1}" for i in range(8)] + [f"Drop-off Neuron {i+1}" for i in range(8)]

sns.heatmap(
    combined_state_matrix,
    cmap='Blues', # Clean Academic Blue Gradient
    center=0.0,
    cbar_kws={'label': r'Hidden State Memory Activation ($\tanh(h_t)$)'},
    xticklabels=time_steps,
    yticklabels=row_labels,
    linewidths=1,
    linecolor='white',
    ax=axes[1],
    annot=True,
    fmt=".1f",
    annot_kws={"size": 9, "weight": "bold"}
)

# Draw a line separating the Binge session from the Drop-off session
axes[1].axhline(8, color='#e74c3c', linewidth=2.5, linestyle='-')
axes[1].set_title('2. Internal Hidden State Temporal Memory ($h_t$)\n(Comparing Binge-Watch vs Session Drop-off Memory States)', fontsize=13, fontweight='bold', pad=12)
axes[1].set_xlabel('Temporal Session Sequence Time Step ($t$)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('8 Hidden Neurons per Scenario', fontsize=12, fontweight='bold')
axes[1].set_yticklabels(row_labels, fontsize=10, fontweight='bold')

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (اهمیت حافظه در هوش مصنوعی):</b> گراف ۲ به زیبایی نشان می‌دهد که چرا ما به معماری RNN نیاز داریم. "
    r"ماتریس‌های Heatmap نشان‌دهنده مغز شبکه (Hidden States) هستند. در ۸ نورونِ اول (برای کاربری که در نهایت شروع به Binge-Watching می‌کند)، "
    r"رنگ‌های تیره (مقادیر مثبت نزدیک به ۱) از همان Step 1 شروع به شکل‌گیری می‌کنند و با گذر زمان و رسیدن به Step 6 تیره و تیره‌تر "
    r"می‌شوند. این یعنی شبکه <i>حافظهٔ تعاملات قبلی</i> را نگه داشته و با هر کلیک جدید، اطمینانش بیشتر می‌شود. "
    r"در مقابل، برای کاربری که قرار است پلتفرم را ببندد (۸ نورون پایینی)، مقادیر از همان ابتدا سرد (نزدیک به صفر یا منفی) هستند "
    r"و به تدریج خاموش می‌شوند. مدل‌های سنتی (مثل رگرسیون لجستیک ساده) چون فقط به <b>لحظه حال</b> نگاه می‌کنند، این زنجیرهٔ "
    r"خاموش و روشن شدن تدریجی را نمی‌بینند، اما RNN به لطف ماتریس بازگشتی $W_{hh}$ تمام این مسیر را یکپارچه تحلیل کرده "
    r"و می‌تواند پیش از وقوع قطعی، نارضایتی کاربر را پیش‌بینی کند."
)
display_academic_text(insight_conclusion, "insight")
```

<ul style="font-size:15pt; line-height:250%; color:#2B2D42;">
<li>مجموعه داده‌ای متشکل از ۴۰۰ سشن تعاملی نتفلیکس شامل توالی‌های زمانی با طول ۶ گام و ۳ ویژگی رفتاری پیوسته (شدت تعامل، مدت پیش‌نمایش و زمان توقف) در دو کلاس ۱ (سشن تماشای متوالی) و ۰ (ترک سشن و ریزش) شبیه‌سازی گردید.</li>
<li>پیاده‌سازی کامل شبکه عصبی بازگشتی (Vanilla Elman RNN) با الگوریتم BPTT از صفر (From Scratch):
  <ul>
    <li><b>گذر رفت زمانی (Temporal Forward Pass):</b> در متد <code>forward</code>، بردار حالت پنهان ($\mathbf{h}_t$) در هر گام بر اساس بردار ویژگی گام جاری و حالت پنهان گام پیشین با تابع فعال‌ساز $\tanh$ محاسبه و تاریخچه آن ذخیره شد؛ در آخرین گام زمانی ($T=6$)، با اعمال تابع سیگموئید بر خروجی، احتمال تماشای پیوسته به دست آمد.</li>
    <li><b>پس‌انتشار خطا در طول زمان (BPTT):</b> در متد <code>backward</code>، گرادیان‌های خطا با باز کردن محاسبات زمانی از گام نهایی به سمت گام‌های اولیه ($t = T, \dots, 1$) محاسبه گردیدند و برای جلوگیری از پدیده انفجار گرادیان (Exploding Gradients)، تکنیک برش گرادیان (Gradient Clipping) اعمال شد.</li>
    <li><b>حلقه بهینه‌سازی SGD:</b> در متد <code>train</code>، وزن‌های ورودی-پنهان ($\mathbf{W}_{xh}$)، پنهان-پنهان ($\mathbf{W}_{hh}$) و پنهان-خروجی ($\mathbf{W}_{hy}$) در طول ۷۵ دوره به صورت گرادیان کاهشی تصادفی به‌روزرسانی شدند.</li>
  </ul>
</li>
<li>تفسیر خروجی و نمودارهای تحلیلی:
  <ul>
    <li><b>نمودار پویایی آموزش و همگرایی (سمت چپ):</b> کاهش پیوسته تابع زیان آنتروپی متقاطع و افزایش دقت طبقه‌بندی روی داده‌های آزمون تا بیش از ۹۳٪ را نشان می‌دهد که نشان‌دهنده توانایی مدل در یادگیری الگوهای ترتیبی است.</li>
    <li><b>نقشه حرارتی تکامل حالت پنهان (سمت راست):</b> تفاوت چشمگیر الگوهای حافظه پنهان ۸ نورون شبکه را در طول زمان میان یک سشن تماشای پیوسته (نیمه بالایی با فعال‌سازی‌های مثبت قوی در گام‌های پایانی) و یک سشن ترک سرویس (نیمه پایینی با شیب نزولی و فعال‌سازی‌های معکوس) به تصویر می‌کشد.</li>
  </ul>
</li>
</ul>
