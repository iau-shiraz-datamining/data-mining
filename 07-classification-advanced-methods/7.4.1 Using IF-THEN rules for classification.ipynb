<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: طبقه‌بندی شفاف سفارش‌ها با قوانین IF-THEN در «اینستاکارت» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پلتفرم خرید آنلاین خواربار <b>اینستاکارت</b>، سامانه عملیاتی باید تصمیم بگیرد که آیا یک سفارش در معرض خطر تاخیر و نارضایتی مشتری قرار دارد یا خیر. از آنجا که تیم‌های پشتیبانی و مدیران فروشگاه‌ها باید بتوانند منطق تصمیم سیستم را به زبان ساده درک و بازبینی کنند، استفاده از مدل‌های جعبه‌سیاه امکان‌پذیر نیست. به همین دلیل از <b>طبقه‌بندی مبتنی بر قوانین (Rule-based Classification)</b> در قالب یک فهرست تصمیم (Decision List) از قوانین «اگر ... آنگاه ...» استفاده می‌شود که هر قانون بر اساس معیارهای آماری پوشش و دقت رتبه‌بندی می‌گردد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی ارزیابی کیفیت قوانین $R: \text{Antecedent} \rightarrow \text{Class}_k$:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. معیار پوشش (Coverage) - نسبت نمونه‌هایی که شرط قانون را ارضا می‌کنند:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{Coverage}(R) = \frac{n_{\text{covers}}}{|D|}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. معیار دقت (Accuracy) - نسبت پیش‌بینی صحیح در میان نمونه‌های پوشش‌داده‌شده:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{Accuracy}(R) = \frac{n_{\text{correct}}}{n_{\text{covers}}}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. تخمین لاپلاس (Laplace Estimate) برای پایدارسازی قوانین با پوشش اندک و $k$ کلاس:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{Laplace}(R) = \frac{n_{\text{correct}} + 1}{n_{\text{covers}} + k}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. بهره اطلاعاتی FOIL برای سنجش بهبود حاصل از افزودن یک شرط جدید به قانون:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{FOIL\_Gain} = p_1 \left( \log_2 \frac{p_1}{p_1 + n_1} - \log_2 \frac{p_0}{p_0 + n_0} \right)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۵. اندازه‌گیری کیفیت هرس قانون با معیار FOIL_Prune روی مجموعه اعتبارسنجی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{FOIL\_Prune}(R) = \frac{p - n}{p + n}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: اینستاکارت با مرتب‌سازی قوانین بر مبنای تخمین لاپلاس و اعمال ترتیبی آن‌ها، تصمیم‌های عملیاتی خود را کاملاً قابل توضیح و قابل ممیزی نگه می‌دارد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: طبقه‌بندی شفاف و قابل تفسیر با فهرست مرتب قوانین منطقی «اگر - آنگاه»
    </div>
  </p>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "طبقه‌بندی شفاف سفارش‌ها با قوانین IF-THEN در «اینستاکارت» (Rule-based Classification) 🛒", 
    "main_title"
)

display_academic_text(
    "در پلتفرم خواربار اینستاکارت، اگر سیستمی یک سفارش را پرریسک اعلام کند، مدیران و تیم پشتیبانی "
    "باید بتوانند به زبان ساده بپرسند: «چرا؟». در این شرایط، مدل‌های جعبه‌سیاه (مثل جنگل تصادفی یا شبکه‌های عصبی) "
    "کنار گذاشته می‌شوند و از <b>فهرست تصمیم (Decision List) مبتنی بر قانون</b> استفاده می‌گردد. در این تمرین، "
    "الگوریتمی را از پایه می‌نویسیم که قوانین منطقی (اگر...آنگاه...) را مستقیماً از داده‌ها استخراج و رتبه‌بندی می‌کند.", 
    "paragraph"
)

display_academic_text("معادلات پایه ارزیابی کیفیت قوانین ($R: \text{Antecedent} \to \text{Class}_k$):", "sub_title")

display_academic_text("۱. معیار پوشش (Coverage) و دقت خام (Accuracy) قانون:", "paragraph")
display_academic_text(r"$$\text{Coverage}(R) = \frac{n_{\text{covers}}}{|D|}, \quad \text{Accuracy}(R) = \frac{n_{\text{correct}}}{n_{\text{covers}}}$$", "paragraph")

display_academic_text("۲. تخمین لاپلاس (Laplace Estimate) برای پایدارسازی قوانین و جلوگیری از اطمینان کاذب به قوانین کم‌تکرار:", "paragraph")
display_academic_text(r"$$\text{Laplace}(R) = \frac{n_{\text{correct}} + 1}{n_{\text{covers}} + k}$$", "paragraph")

# ==============================================================================
# 2. INSTACART ORDER RISK DATASET (CATEGORICAL ATTRIBUTES)
# ==============================================================================
np.random.seed(42)
n_orders = 1000

# Categorical operational attributes of Instacart orders
basket_size = np.random.choice(['Small', 'Medium', 'Large'], n_orders, p=[0.4, 0.4, 0.2])
time_slot = np.random.choice(['Morning', 'Afternoon', 'PeakEvening'], n_orders, p=[0.35, 0.35, 0.3])
shopper_rating = np.random.choice(['Low', 'High'], n_orders, p=[0.3, 0.7])
store_traffic = np.random.choice(['Quiet', 'Busy'], n_orders, p=[0.6, 0.4])
distance_tier = np.random.choice(['Near', 'Far'], n_orders, p=[0.65, 0.35])

# Ground truth risk logic (with stochastic noise)
def generate_label(bs, ts, sr, st, dt):
    risk = 0.05
    if ts == 'PeakEvening': risk += 0.30
    if st == 'Busy':        risk += 0.25
    if bs == 'Large':       risk += 0.20
    if sr == 'Low':         risk += 0.18
    if dt == 'Far':         risk += 0.15
    return 1 if np.random.rand() < min(risk, 0.97) else 0

labels = [generate_label(b, t, s, st, d) 
          for b, t, s, st, d in zip(basket_size, time_slot, shopper_rating, store_traffic, distance_tier)]

df_orders = pd.DataFrame({
    'BasketSize': basket_size,
    'TimeSlot': time_slot,
    'ShopperRating': shopper_rating,
    'StoreTraffic': store_traffic,
    'DistanceTier': distance_tier,
    'DelayRisk': labels  # 1 = Delayed / At-Risk, 0 = On-Time
})

# Split into training and testing sets manually
split_point = int(0.75 * len(df_orders))
df_train = df_orders.iloc[:split_point].reset_index(drop=True)
df_test = df_orders.iloc[split_point:].reset_index(drop=True)

# ==============================================================================
# 3. IF-THEN RULE-BASED CLASSIFIER IMPLEMENTED FROM SCRATCH
# ==============================================================================
class IfThenRuleClassifier:
    """Rule-based Decision List classifier generating IF-THEN rules from scratch."""

    def __init__(self, max_conditions=2, min_coverage=15, n_classes=2):
        self.max_conditions = max_conditions
        self.min_coverage = min_coverage
        self.n_classes = n_classes
        self.rules = []
        self.default_class = 0

    def _matches(self, row, antecedent):
        """Checks whether a data row satisfies all conditions of a rule antecedent."""
        for attr, val in antecedent:
            if row[attr] != val:
                return False
        return True

    def _evaluate_rule(self, data, antecedent, target_class):
        """Computes coverage, accuracy and Laplace estimate for a candidate rule."""
        mask = np.ones(len(data), dtype=bool)
        for attr, val in antecedent:
            mask = mask & (data[attr].values == val)

        n_covers = int(np.sum(mask))
        if n_covers == 0:
            return None

        n_correct = int(np.sum(data['DelayRisk'].values[mask] == target_class))

        coverage = n_covers / len(data)
        accuracy = n_correct / n_covers
        
        # Laplace Smoothing: (correct + 1) / (total + k)
        laplace = (n_correct + 1) / (n_covers + self.n_classes)

        return {
            'Antecedent': antecedent,
            'Class': target_class,
            'N_Covers': n_covers,
            'N_Correct': n_correct,
            'Coverage': coverage,
            'Accuracy': accuracy,
            'Laplace': laplace
        }

    def generate_candidate_rules(self, data):
        """Enumerates all IF-THEN candidate rules up to max_conditions length."""
        attributes = [c for c in data.columns if c != 'DelayRisk']

        # Build attribute-value literal pool
        literals = []
        for attr in attributes:
            for val in sorted(data[attr].unique()):
                literals.append((attr, val))

        candidates = []
        for depth in range(1, self.max_conditions + 1):
            for combo in itertools.combinations(literals, depth):
                # Reject combinations using the same attribute twice (e.g. Traffic=Busy AND Traffic=Quiet)
                used_attrs = [c[0] for c in combo]
                if len(set(used_attrs)) != len(used_attrs):
                    continue

                for target_class in [0, 1]:
                    stats = self._evaluate_rule(data, list(combo), target_class)
                    # Filter by minimum support (coverage)
                    if stats and stats['N_Covers'] >= self.min_coverage:
                        candidates.append(stats)

        return candidates

    def fit(self, data):
        """Builds an ordered decision list using sequential covering (greedy approach)."""
        remaining = data.copy().reset_index(drop=True)
        self.rules = []

        while len(remaining) > self.min_coverage:
            candidates = self.generate_candidate_rules(remaining)
            if not candidates:
                break

            # Select best rule by Laplace estimate (tie-breaker: coverage)
            best_rule = max(candidates, key=lambda r: (r['Laplace'], r['N_Covers']))

            # Stop if the best rule is barely better than random guessing
            if best_rule['Accuracy'] < 0.60:
                break

            self.rules.append(best_rule)

            # Sequential Covering: Remove instances covered by this rule
            mask = np.ones(len(remaining), dtype=bool)
            for attr, val in best_rule['Antecedent']:
                mask = mask & (remaining[attr].values == val)
            remaining = remaining[~mask].reset_index(drop=True)

        # Default rule: majority class of whatever instances are left over
        if len(remaining) > 0:
            self.default_class = int(remaining['DelayRisk'].mode()[0])
        else:
            self.default_class = int(data['DelayRisk'].mode()[0])

        return self

    def predict_row(self, row):
        """Applies rules sequentially; first matching rule fires and stops the evaluation."""
        for rule in self.rules:
            if self._matches(row, rule['Antecedent']):
                return rule['Class'], rule
        return self.default_class, None

    def predict(self, data):
        preds = []
        for _, row in data.iterrows():
            cls, _ = self.predict_row(row)
            preds.append(cls)
        return np.array(preds)

    def rule_to_text(self, rule):
        """Converts a rule dictionary into a human-readable IF-THEN string."""
        conds = " AND ".join([f"{a}='{v}'" for a, v in rule['Antecedent']])
        outcome = "Delayed Risk" if rule['Class'] == 1 else "On-Time"
        return f"IF {conds} \u2794 THEN {outcome}"

# Train the rule-based classifier
rule_clf = IfThenRuleClassifier(max_conditions=2, min_coverage=20)
rule_clf.fit(df_train)

# Evaluate on the held-out test set
y_pred = rule_clf.predict(df_test)
y_true = df_test['DelayRisk'].values
test_accuracy = np.mean(y_pred == y_true) * 100

# ==============================================================================
# 4. RULE PRESENTATION AND METRICS
# ==============================================================================
# Build reporting dataframe for the learned rule set
rule_records = []
for i, r in enumerate(rule_clf.rules):
    rule_records.append({
        '#': f"R{i+1}",
        'Extracted Human-Readable Logic Rule': rule_clf.rule_to_text(r),
        'Data Cov': f"{r['N_Covers']} orders",
        'Accuracy': f"{r['Accuracy']*100:.1f}%",
        'Laplace Est': f"{r['Laplace']*100:.1f}%",
        '_raw_acc': r['Accuracy'],
        '_raw_lap': r['Laplace'],
        '_raw_cov': r['Coverage']
    })

# Add Default Rule
rule_records.append({
    '#': "DEFAULT",
    'Extracted Human-Readable Logic Rule': f"IF No Prior Rule Matched \u2794 THEN {'Delayed Risk' if rule_clf.default_class == 1 else 'On-Time'}",
    'Data Cov': "Remainder",
    'Accuracy': "-",
    'Laplace Est': "-",
    '_raw_acc': 0,
    '_raw_lap': 0,
    '_raw_cov': 0
})

df_rules = pd.DataFrame(rule_records)

# Display the rules as a clean, dashboard-like table
display_cols = ['#', 'Extracted Human-Readable Logic Rule', 'Data Cov', 'Accuracy', 'Laplace Est']
display_side_by_side_datasets(
    [df_rules[display_cols]], 
    ["فهرست تصمیم‌گیری عملیاتی اینستاکارت (کاملاً قابل تفسیر برای انسان)"]
)

# ==============================================================================
# 5. VISUALIZATIONS OF RULE QUALITY AND CLASSIFIER PERFORMANCE
# ==============================================================================
# Confusion matrix computed from scratch
conf_matrix = np.zeros((2, 2), dtype=int)
for t, p in zip(y_true, y_pred):
    conf_matrix[int(t), int(p)] += 1

fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

#  CHART 1: Rule Quality Metrics (Grouped Bar Chart) 
# Exclude the Default rule for metric plotting
plot_rules = df_rules[:-1]
y_idx = np.arange(len(plot_rules))
bar_h = 0.28

# Academic Blue Palette for metric bars
b1 = axes[0].barh(y_idx - bar_h, plot_rules['_raw_acc'] * 100, height=bar_h, label='Raw Accuracy (%)', color='#2980b9', edgecolor='black', alpha=0.9)
b2 = axes[0].barh(y_idx, plot_rules['_raw_lap'] * 100, height=bar_h, label='Laplace Estimate (%)', color='#2c3e50', edgecolor='black', alpha=0.9)
b3 = axes[0].barh(y_idx + bar_h, plot_rules['_raw_cov'] * 100, height=bar_h, label='Relative Coverage (%)', color='#bdc3c7', edgecolor='black', alpha=0.9)

axes[0].axvline(50, color='#e74c3c', linestyle='--', linewidth=2, label='Random Baseline (50%)')
axes[0].set_yticks(y_idx)
# Shorten rules for Y-axis display
short_rules = [r.replace(" \u2794 THEN Delayed Risk", "").replace(" \u2794 THEN On-Time", "") for r in plot_rules['Extracted Human-Readable Logic Rule']]
axes[0].set_yticklabels(short_rules, fontsize=9.5, fontweight='bold')
axes[0].set_xlabel('Metric Value (%)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Sequential Rules', fontsize=12, fontweight='bold')
axes[0].set_title('1. Rule Set Quality (Why Laplace > Raw Accuracy)', fontsize=13, fontweight='bold', pad=12)
axes[0].legend(loc='lower right', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
axes[0].grid(axis='x', linestyle='--', alpha=0.6)
axes[0].invert_yaxis() # Display R1 at the top

# Annotate Laplace vs Accuracy to show the penalization of small coverage
for bar_a, bar_l in zip(b1, b2):
    w_a, w_l = bar_a.get_width(), bar_l.get_width()
    axes[0].annotate(f'{w_a:.1f}', xy=(w_a, bar_a.get_y() + bar_a.get_height() / 2), xytext=(4, 0), textcoords='offset points', ha='left', va='center', fontsize=9, fontweight='bold', color='#2980b9')
    axes[0].annotate(f'{w_l:.1f}', xy=(w_l, bar_l.get_y() + bar_l.get_height() / 2), xytext=(4, 0), textcoords='offset points', ha='left', va='center', fontsize=9, fontweight='bold', color='#2c3e50')


#  CHART 2: Confusion Matrix Heatmap 
labels_axis = ['Actual: On-Time', 'Actual: Delayed']
cols_axis = ['Pred: On-Time', 'Pred: Delayed']

sns.heatmap(
    pd.DataFrame(conf_matrix, index=labels_axis, columns=cols_axis),
    annot=True, fmt='d', cmap='Blues', linewidths=1.5, linecolor='white',
    cbar_kws={'label': 'Number of Test Orders'}, annot_kws={'size': 14, 'weight': 'bold'},
    ax=axes[1]
)

# Derived quality metrics computed from scratch
tp, fp, fn, tn = conf_matrix[1, 1], conf_matrix[0, 1], conf_matrix[1, 0], conf_matrix[0, 0]
precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0.0

axes[1].set_title(
    f'2. Rule-Based Decision List Performance on Test Set\n'
    f'Accuracy: {test_accuracy:.1f}% | Precision: {precision*100:.1f}% | Recall: {recall*100:.1f}%',
    fontsize=13, fontweight='bold', pad=12
)
axes[1].set_xlabel('Predicted Class by First Firing Rule', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Ground Truth Order Outcome', fontsize=12, fontweight='bold')
axes[1].set_xticklabels(cols_axis, fontsize=11, fontweight='bold')
axes[1].set_yticklabels(labels_axis, fontsize=11, fontweight='bold', rotation=0)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 6: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (شفافیت و قانون لاپلاس):</b> در جدول استخراج شده، سیستم ما به صریح‌ترین شکل ممکن "
    r"به مدیران اینستاکارت می‌گوید: <i>اگر شیفت کاری عصر (PeakEvening) است و ترافیک فروشگاه بالاست، سفارش قطعا با تاخیر مواجه می‌شود.</i> "
    r"هیچ بلک‌باکسی در کار نیست! اما نکته کلیدی در گراف ۱ نهفته است: مقایسه نوار آبی روشن (دقت خام) با نوار سورمه‌ای (تخمین لاپلاس). "
    r"در قوانین انتهایی (مثل R4)، دقت خام ممکن است ۱۰۰٪ باشد، اما از آنجا که این قانون فقط تعداد کمی از داده‌ها را "
    r"پوشش داده است (نوار خاکستری کوچک)، <b>فرمول لاپلاس آن را جریمه می‌کند</b> و مقدار آن را به زیر ۸۵٪ می‌کشاند. "
    r"این مکانیسم محافظتی باعث می‌شود که سیستم، قوانینِ عمومی و قدرتمند را بر قوانینِ خاص اما شانسی، ارجحیت دهد."
)
display_academic_text(insight_conclusion, "insight")
```

<li>مجموعه داده‌ای شامل ۸۰۰ سفارش شبیه‌سازی‌شده اینستاکارت با ۵ ویژگی کاملاً گسسته (اندازه سبد، بازه زمانی، امتیاز خریدار، شلوغی فروشگاه و فاصله) و برچسب دودویی ریسک تاخیر تولید و به دو بخش آموزش و آزمون تقسیم شد.</li>
<li>پیاده‌سازی کامل طبقه‌بند مبتنی بر قوانین IF-THEN از صفر (From Scratch):
  <ul>
    <li><b>تولید فضای کاندیدای قوانین:</b> در متد <code>generate_candidate_rules</code>، تمام ترکیب‌های ممکن از گزاره‌های ویژگی-مقدار تا طول دو شرط با استفاده از <code>itertools.combinations</code> ساخته شد و ترکیب‌های تکراری روی یک ویژگی حذف گردیدند.</li>
    <li><b>محاسبه معیارهای کیفی قانون:</b> در متد <code>_evaluate_rule</code>، سه شاخص پوشش، دقت و تخمین لاپلاس مستقیماً از شمارش نمونه‌های پوشش‌داده‌شده و صحیح استخراج شد تا قوانین با پوشش کم به صورت مصنوعی دقت ۱۰۰ درصدی نگیرند.</li>
    <li><b>ساخت فهرست تصمیم ترتیبی:</b> در متد <code>fit</code>، در هر تکرار بهترین قانون بر مبنای تخمین لاپلاس انتخاب شده، به فهرست افزوده می‌شود و سپس نمونه‌های پوشش‌داده‌شده از داده حذف می‌گردند تا قانون بعدی روی باقیمانده آموزش ببیند.</li>
    <li><b>استنتاج ترتیبی و قانون پیش‌فرض:</b> در متد <code>predict_row</code>، قوانین به ترتیب اولویت بررسی می‌شوند و اولین قانون منطبق شلیک می‌کند؛ در صورت عدم تطابق هیچ قانونی، کلاس اکثریت باقیمانده به عنوان قانون پیش‌فرض اعمال می‌شود.</li>
  </ul>
</li>
<li>تفسیر خروجی و نمودارهای تحلیلی:
  <ul>
    <li><b>نمودار میله‌ای سمت چپ (Rule Set Quality):</b> سه معیار دقت، تخمین لاپلاس و پوشش را برای هر قانون یادگرفته‌شده کنار هم مقایسه می‌کند؛ مشاهده می‌شود قوانین ابتدایی فهرست پوشش گسترده‌تری دارند و قوانین پایانی تخصصی‌تر و با پوشش کمتر هستند و فاصله میان دقت و تخمین لاپلاس در قوانین کم‌پوشش بیشتر می‌شود.</li>
    <li><b>نقشه حرارتی سمت راست (Confusion Matrix):</b> عملکرد فهرست تصمیم روی داده‌های آزمون دیده‌نشده را همراه با شاخص‌های دقت، صحت، فراخوانی و امتیاز F1 نمایش می‌دهد و نشان می‌دهد مدل با وجود سادگی و شفافیت کامل، توان تفکیک قابل قبولی میان سفارش‌های به‌موقع و پرریسک دارد.</li>
    <li><b>خروجی متنی فهرست قوانین:</b> در انتها مجموعه قوانین به صورت جملات خوانا برای انسان چاپ می‌شود که همان مزیت اصلی طبقه‌بندی مبتنی بر قانون یعنی تفسیرپذیری و قابلیت ممیزی توسط تیم‌های عملیاتی است.</li>
  </ul>
</li>
</ul>
