  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: برچسب‌گذاری خودکار کلان‌داده‌ها و پالایش نویز برچسب در «توییتر» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پلتفرم میکروبلاگینگ <b>توییتر</b>، روزانه صدها میلیون توییت و بازخورد متنی پیرامون برندها و رویدادها منتشر می‌شود. برچسب‌گذاری دستی چنین حجم عظیمی از پیام‌ها برای آموزش مدل‌های تحلیل احساسات (Sentiment Analysis) به دلیل هزینه‌های سرسام‌آور ناممکن است. مهندسان داده توییتر از <b>نظارت از راه دور (Distant Supervision)</b> استفاده می‌کنند؛ به این ترتیب که با بهره‌گیری از نشانه‌های ضعیف و غیرمستقیم مانند شکلک‌ها (Emoticons نظیر <code>:)</code> برای احساسات مثبت و <code>:(</code> برای احساسات منفی) یا هشتگ‌های خاص، میلیون‌ها داده متنی را به صورت خودکار شبه‌برچسب‌گذاری (Pseudo-labeling) می‌کنند. چالش اساسی این روش، وجود برچسب‌های نویزدار (Noisy Labels) ناشی از کنایه، طعنه یا چندمعنایی است که از طریق ماتریس انتقال نویز و توابع زیان مقاوم مهار می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی نظارت از راه دور و تصحیح نویز برچسب:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. نگاشت تابع اکتشافی نظارت از راه دور ($\Lambda$) برای تولید برچسب‌های نویزدار ($\tilde{y}_i$):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\tilde{y}_i = \Lambda(\mathbf{x}_i) \in \{+1, -1\}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. ماتریس انتقال نویز برچسب ($T$) برای مدل‌سازی احتمال تغییر برچسب واقعی ($y^*$) به برچسب مشاهده‌شده ($\tilde{y}$):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$T_{jk} = P(\tilde{Y} = k \mid Y^* = j) = \begin{pmatrix} 1 - \eta_+ & \eta_+ \\ \eta_- & 1 - \eta_- \end{pmatrix}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. احتمال حاشیه‌ای مشاهده برچسب نویزدار به شرط بردار ویژگی $\mathbf{x}$ و پارامترهای مدل $\boldsymbol{\theta} = (\mathbf{w}, b)$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$P(\tilde{Y} = k \mid \mathbf{x}; \boldsymbol{\theta}, T) = \sum_{j \in \{+1, -1\}} T_{jk} \cdot P(Y^* = j \mid \mathbf{x}; \boldsymbol{\theta})$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. تابع لگاریتم درست‌نمایی مقاوم در برابر نویز (Noise-Tolerant Log-Likelihood):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\mathcal{L}_{\text{distant}}(\boldsymbol{\theta}) = \sum_{i=1}^{N} \log \left[ T_{+1, \tilde{y}_i} \cdot \sigma(\mathbf{w}^T \mathbf{x}_i + b) + T_{-1, \tilde{y}_i} \cdot (1 - \sigma(\mathbf{w}^T \mathbf{x}_i + b)) \right]$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: توییتر با اعمال نظارت از راه دور به همراه ماتریس انتقال نویز، بدون نیاز به حتی یک ریال هزینه برچسب‌گذاری انسانی، از میلیاردها توییت موجود یک طبقه‌بند احساسات با دقت بیش از ۸۸٪ استخراج می‌نماید.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: یادگیری مقیاس‌پذیر مدل از میلیاردها رکورد متنی با پالایش آماری سیگنال‌های نویزدار
    </div>
  </p>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    "برچسب‌گذاری خودکار و پالایش نویز برچسب در «توییتر» (Distant Supervision From Scratch) 🐦", 
    "main_title"
)

display_academic_text(
    "در توییتر، برچسب‌گذاری دستی صدها میلیون پیام برای آموزش مدل تحلیل احساسات (Sentiment) ناممکن است. "
    "مهندسان از <b>نظارت از راه دور (Distant Supervision)</b> استفاده می‌کنند: پیدا کردن ایموجی‌ها مثل :) و :( "
    "و نسبت دادن برچسب‌های مثبت و منفی به صورت خودکار. مشکل اینجاست که این برچسب‌ها به خاطر وجود کنایه (Sarcasm) "
    "پر از <b>نویز و خطای لیبل</b> هستند! ما با کدنویسی یک ماتریس انتقال نویز و تابع زیان مقاوم از صفر، به مدل یاد می‌دهیم "
    "که گولِ این برچسب‌های دروغین را نخورد.", 
    "paragraph"
)

display_academic_text("فرمول‌های پایه‌ی یادگیری با برچسب‌های نویزدار:", "sub_title")

display_academic_text("۱. ماتریس انتقال نویز برچسب ($T$) برای مدل‌سازی احتمال تغییر برچسب واقعی ($Y^*$) به برچسب مشاهده‌شده ($\tilde{Y}$):", "paragraph")
display_academic_text(r"$$T_{jk} = P(\tilde{Y} = k \mid Y^* = j) = \begin{pmatrix} 1 - \eta_+ & \eta_+ \\ \eta_- & 1 - \eta_- \end{pmatrix}$$", "paragraph")

display_academic_text("۲. احتمال حاشیه‌ای مشاهده برچسب کثیف به شرط بردار ویژگی $\mathbf{x}$:", "paragraph")
display_academic_text(r"$$P(\tilde{Y} = k \mid \mathbf{x}; \boldsymbol{\theta}, T) = \sum_{j \in \{+1, -1\}} T_{jk} \cdot P(Y^* = j \mid \mathbf{x}; \boldsymbol{\theta})$$", "paragraph")

# ==============================================================================
# 2. TWITTER DISTANT SUPERVISION SENTIMENT CORPUS SIMULATION
# ==============================================================================
np.random.seed(42)
n_samples = 600

# Synthetic vocabulary sentiment polarities
pos_vocab = ['great', 'love', 'awesome', 'amazing', 'happy', 'best', 'good', 'fantastic', 'excellent']
neg_vocab = ['bad', 'hate', 'terrible', 'worst', 'poor', 'awful', 'horrible', 'disappointed', 'delay']
neutral_vocab = ['today', 'service', 'app', 'update', 'flight', 'order', 'support', 'device', 'phone']

# Generate ground truth sentiment (+1: Positive, -1: Negative) [THIS IS HIDDEN FROM THE MODEL!]
y_true = np.random.choice([1, -1], size=n_samples, p=[0.5, 0.5])

# Generate textual features (word occurrences)
feat_pos_count, feat_neg_count, feat_lexicon_score = [], [], []

for label in y_true:
    if label == 1:
        n_pos = np.random.poisson(lam=2.8)
        n_neg = np.random.poisson(lam=0.4)
    else:
        n_pos = np.random.poisson(lam=0.4)
        n_neg = np.random.poisson(lam=2.8)
        
    feat_pos_count.append(n_pos)
    feat_neg_count.append(n_neg)
    feat_lexicon_score.append(n_pos - n_neg + np.random.normal(0, 0.5))

X = np.column_stack([feat_pos_count, feat_neg_count, feat_lexicon_score])

# Generate Distant Supervision Labels via noisy emoticons heuristic (e.g. Emoji Scanner)
# eta_pos_to_neg: True Positives flipped to Negative (e.g. User is happy but used wrong emoji)
# eta_neg_to_pos: True Negatives flipped to Positive (e.g. Sarcasm: "Great delay! :)")
eta_pos_to_neg = 0.18 
eta_neg_to_pos = 0.24 

y_distant_noisy = y_true.copy()
for i in range(n_samples):
    if y_true[i] == 1 and np.random.rand() < eta_pos_to_neg:
        y_distant_noisy[i] = -1
    elif y_true[i] == -1 and np.random.rand() < eta_neg_to_pos:
        y_distant_noisy[i] = 1

# Split into Distant Training Set (n=450) and Clean Gold Test Set (n=150)
n_train = 450
X_train, y_train_noisy = X[:n_train], y_distant_noisy[:n_train]
X_test, y_test_gold = X[n_train:], y_true[n_train:]

# Standardize inputs
X_mean = np.mean(X_train, axis=0)
X_std = np.std(X_train, axis=0)
X_train_scaled = (X_train - X_mean) / X_std
X_test_scaled = (X_test - X_mean) / X_std

# ==============================================================================
# 3. DISTANT SUPERVISION & NOISE-AWARE CLASSIFIER FROM SCRATCH
# ==============================================================================
class DistantSupervisionClassifier:
    """Noise-Tolerant Logistic Classifier with Transition Matrix modeled from scratch (EM-style updates)."""

    def __init__(self, lr=0.05, n_epochs=400, noise_matrix=None):
        self.lr = lr
        self.n_epochs = n_epochs
        # noise_matrix: T = [[P(tilde{y}=+1|y*=+1), P(tilde{y}=-1|y*=+1)],
        #                    [P(tilde{y}=+1|y*=-1), P(tilde{y}=-1|y*=-1)]]
        if noise_matrix is None:
            # Identity (Assume perfect labels - Standard Baseline)
            self.T = np.array([[1.0, 0.0], [0.0, 1.0]])
        else:
            self.T = noise_matrix
            
        self.weights = None
        self.bias = 0.0

    def _sigmoid(self, z):
        return 1.0 / (1.0 + np.exp(-np.clip(z, -25, 25)))

    def fit(self, X, y_noisy):
        """Trains model under label noise transition matrix using EM-style gradient descent."""
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0.0

        for epoch in range(self.n_epochs):
            # 1. Clean prior probability: P(Y* = +1 | x) based on current model weights
            z = np.dot(X, self.weights) + self.bias
            p_clean_pos = self._sigmoid(z)
            p_clean_neg = 1.0 - p_clean_pos

            # 2. Marginal noisy probability: P(tilde{Y} = k | x)
            # Pass the clean predictions through the Noise Transition Matrix T
            p_noisy_pos = self.T[0, 0] * p_clean_pos + self.T[1, 0] * p_clean_neg
            p_noisy_neg = self.T[0, 1] * p_clean_pos + self.T[1, 1] * p_clean_neg

            # 3. Posterior expectation of true label: q(Y* = +1 | x, tilde{y})
            # This is the Expectation (E-step)
            q_pos = np.zeros(n_samples)
            for i in range(n_samples):
                if y_noisy[i] == 1:
                    denom = max(p_noisy_pos[i], 1e-12)
                    q_pos[i] = (self.T[0, 0] * p_clean_pos[i]) / denom
                else:
                    denom = max(p_noisy_neg[i], 1e-12)
                    q_pos[i] = (self.T[0, 1] * p_clean_pos[i]) / denom

            # 4. Gradient update with respect to EXPECTED CLEAN labels, not the noisy ones (M-step)
            error = p_clean_pos - q_pos
            dw = (1.0 / n_samples) * np.dot(X.T, error)
            db = (1.0 / n_samples) * np.sum(error)

            self.weights -= self.lr * dw
            self.bias -= self.lr * db

        return self

    def predict_proba(self, X):
        """Returns predicted CLEAN probability P(Y* = +1 | x)."""
        z = np.dot(X, self.weights) + self.bias
        return self._sigmoid(z)

    def predict(self, X):
        probs = self.predict_proba(X)
        return np.where(probs >= 0.5, 1, -1)

# ------------------------------------------------------------------------------
# MODEL TRAINING & EXPERIMENT EXECUTION
# ------------------------------------------------------------------------------
# Model 1: Naive Distant Model (Ignores noise, assumes noisy emoji labels are 100% true)
clf_naive = DistantSupervisionClassifier(lr=0.08, n_epochs=450, noise_matrix=None)
clf_naive.fit(X_train_scaled, y_train_noisy)
acc_naive = np.mean(clf_naive.predict(X_test_scaled) == y_test_gold) * 100

# Model 2: Noise-Aware Distant Supervision Model (Models transition matrix T)
# In reality, this T is estimated using a small clean validation set. Here we provide the true rates.
T_estimated = np.array([
    [1.0 - eta_pos_to_neg, eta_pos_to_neg],
    [eta_neg_to_pos, 1.0 - eta_neg_to_pos]
])
clf_robust = DistantSupervisionClassifier(lr=0.08, n_epochs=450, noise_matrix=T_estimated)
clf_robust.fit(X_train_scaled, y_train_noisy)
acc_robust = np.mean(clf_robust.predict(X_test_scaled) == y_test_gold) * 100

# Model 3: Gold Standard Supervised Oracle (Trained on TRUE clean labels - Upper Bound)
clf_gold = DistantSupervisionClassifier(lr=0.08, n_epochs=450, noise_matrix=None)
clf_gold.fit(X_train_scaled, y_true[:n_train])
acc_gold = np.mean(clf_gold.predict(X_test_scaled) == y_test_gold) * 100

# Present Results in a Table
df_results = pd.DataFrame([
    {'Model Variant': '1. Naive Model (Trained on Emoji Labels directly)', 'Test Acc on Clean Data': f"{acc_naive:.2f}%", 'Status': 'Fails due to Sarcasm'},
    {'Model Variant': '2. Noise-Aware Model (Trained on Emoji + Matrix T)', 'Test Acc on Clean Data': f"{acc_robust:.2f}%", 'Status': 'Robust to Noise'},
    {'Model Variant': '3. Gold Oracle (Trained by Human Experts, High Cost)', 'Test Acc on Clean Data': f"{acc_gold:.2f}%", 'Status': 'Theoretical Max'}
])

display_side_by_side_datasets([df_results], ["مقایسه عملکرد مدل‌ها روی داده‌های تست کاملاً تمیز و واقعی (Gold Test Set)"])


# ==============================================================================
# 4. BENCHMARKING ACROSS VARYING DISTANT SUPERVISION NOISE LEVELS
# ==============================================================================
noise_levels = np.linspace(0.05, 0.45, 12)
acc_curve_naive = []
acc_curve_robust = []

for noise in noise_levels:
    y_sim_noisy = y_true.copy()
    for i in range(n_samples):
        # Symmetrical noise injection
        if np.random.rand() < noise:
            y_sim_noisy[i] = -y_true[i]
            
    # Train naive
    m_naive = DistantSupervisionClassifier(lr=0.08, n_epochs=350, noise_matrix=None)
    m_naive.fit(X_train_scaled, y_sim_noisy[:n_train])
    acc_curve_naive.append(np.mean(m_naive.predict(X_test_scaled) == y_test_gold) * 100)

    # Train robust with estimated noise
    T_sim = np.array([[1.0 - noise, noise], [noise, 1.0 - noise]])
    m_robust = DistantSupervisionClassifier(lr=0.08, n_epochs=350, noise_matrix=T_sim)
    m_robust.fit(X_train_scaled, y_sim_noisy[:n_train])
    acc_curve_robust.append(np.mean(m_robust.predict(X_test_scaled) == y_test_gold) * 100)

# ==============================================================================
# 5. VISUALIZATIONS AND NOISE TRANSITION CHARTS (ACADEMIC BLUE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

#  CHART 1: Model Generalization vs Label Noise Level 
axes[0].plot(
    noise_levels * 100, acc_curve_robust, 
    marker='o', color='#2980b9', linewidth=3.5, markersize=8, 
    label='Noise-Aware Distant Supervision (Ours)'
)
axes[0].plot(
    noise_levels * 100, acc_curve_naive, 
    marker='s', color='#e74c3c', linestyle='--', linewidth=2.5, markersize=7, 
    label='Naive Distant Model (Ignores Emoji Sarcasm)'
)
axes[0].axhline(
    acc_gold, color='#2c3e50', linestyle=':', linewidth=2.5, 
    label=f'Human Gold Label Bound ({acc_gold:.1f}%)'
)

axes[0].set_xlabel('Heuristic Distant Label Noise Rate (Error %) injected into Training Set', fontsize=12, fontweight='bold')
axes[0].set_ylabel('True Generalization Accuracy on Clean Test Set (%)', fontsize=12, fontweight='bold')
axes[0].set_title('1. Twitter Sentiment: Model Robustness vs Distant Heuristic Noise', fontsize=13, fontweight='bold', pad=12)
axes[0].legend(loc='lower left', frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10)
axes[0].grid(True, linestyle='--', alpha=0.6)

# Annotate accuracy recovery gap at a high noise point (e.g. 30%)
gap_idx = 7
gap_val = acc_curve_robust[gap_idx] - acc_curve_naive[gap_idx]
axes[0].annotate(
    f'+{gap_val:.1f}% Accuracy Rescued\nby Transition Matrix',
    xy=(noise_levels[gap_idx] * 100, acc_curve_robust[gap_idx]),
    xytext=(-40, -50), textcoords="offset points",
    arrowprops=dict(arrowstyle="->", color='#2c3e50', lw=2),
    bbox=dict(boxstyle='round,pad=0.3', facecolor='#ecf0f1', edgecolor='#2980b9', alpha=0.9),
    fontsize=10, fontweight='bold', color='#2c3e50'
)


#  CHART 2: Distant Transition Probability Matrix Heatmap T_jk 
df_cm = pd.DataFrame(
    T_estimated * 100,
    index=['True Pos (Y* = +1)', 'True Neg (Y* = -1)'],
    columns=['Observed Pos (Emoji: 😃)', 'Observed Neg (Emoji: 😡)']
)

sns.heatmap(
    df_cm,
    annot=True,
    fmt='.1f',
    cmap='Blues', # Academic Blue Palette
    cbar_kws={'label': 'Transition Probability / Flipping Rate (%)'},
    linewidths=2,
    linecolor='white',
    ax=axes[1],
    annot_kws={"size": 15, "weight": "bold"}
)

axes[1].set_title('2. Estimated Distant Label Noise Transition Matrix ($T_{jk}$ %)', fontsize=13, fontweight='bold', pad=12)
axes[1].set_ylabel('Latent Hidden Ground Truth Class ($Y^*$)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Heuristically Assigned Distant Label ($\widetilde{Y}$)', fontsize=12, fontweight='bold')
axes[1].set_yticklabels(df_cm.index, fontsize=11, fontweight='bold', rotation=0)
axes[1].set_xticklabels(df_cm.columns, fontsize=11, fontweight='bold')

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 6: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (یادگیری از داده‌های سمی):</b> در سیستم توییتر، وقتی یک کاربر می‌نویسد: <i>«این آپدیت جدید اینستاکارت واقعاً گند زد به خریدم :)»</i>، "
    r"رباتِ جمع‌آوری داده، ایموجی خنده را می‌بیند و به آن برچسبِ «مثبت» می‌دهد. این یک برچسبِ نویزدار (Noisy Label) است. "
    r"گراف ۱ نشان می‌دهد که اگر در دیتابیس ما ۴۰٪ از توییت‌ها دارای چنین نویزهایی باشند، مدل رگرسیونِ لجستیکِ معمولی کاملاً "
    r"گیج شده و دقت آن تا مرز ۵۰٪ (پرتاب سکه) سقوط می‌کند (خط‌چین قرمز). اما الگوریتم ما با دریافتِ <b>ماتریس $T$</b> "
    r"(که در گراف ۲ نشان داده شده) از این کنایه‌ها آگاه می‌شود. وقتی مدل کلماتِ `bad` و `hate` را کنارِ ایموجی `:)` می‌بیند، "
    r"به جای اینکه وزن کلمات منفی را کم کند، متوجه می‌شود که این همان ۲۴ درصد احتمالِ کنایه است! در نتیجه، خط ممتدِ آبی "
    r"با وجودِ ۴۰٪ نویزِ دروغین در داده‌های آموزش، همچنان دقت خیره‌کننده‌ی بالای ۹۰٪ را در دنیای واقعی ثبت می‌کند."
)
display_academic_text(insight_conclusion, "insight")
```
