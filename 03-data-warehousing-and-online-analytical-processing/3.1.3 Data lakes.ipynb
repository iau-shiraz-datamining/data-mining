    مثال عملی: ذخیره‌سازی داده‌های خام و ناهمگن در «نتفلیکس» (Netflix) 🌊
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    پلتفرم «نتفلیکس» (Netflix) روزانه چندین پتابایت داده از منابع کاملاً متفاوت تولید می‌کند:
  </p>

  <ul style="text-align: right; margin: 0 0 10px; padding-right: 22px;">
    <li><b>داده‌های ساخت‌یافته (Structured):</b> اطلاعات حساب و اشتراک کاربران در پایگاه داده SQL.</li>
    <li><b>داده‌های نیمه‌ساخت‌یافته (Semi-Structured):</b> لوگ‌های کلیک و مکث ویدیوها به فرمت JSON.</li>
    <li><b>داده‌های بدون ساختار (Unstructured):</b> نظرات متنی کاربران، فریم‌های پوستر فیلم‌ها و فایل‌های صوتی.</li>
  </ul>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>چالش انبار داده سنتی (Data Warehouse):</b><br>
    انبار داده نیازمند الگوی <b>Schema-on-Write</b> است؛ یعنی داده‌ها باید قبل از ذخیره‌سازی کاملاً پاک‌سازی، تبدیل و ساختاردهی شوند (فرآیند ETL). تبدیل فوری پتابایت‌ها داده نیمه‌ساخت‌یافته و متنی، فوق‌العاده گران و بسیار زمان‌بر است.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>راه‌حل — دریاچه داده (Data Lake) و الگوی Schema-on-Read:</b><br>
    دریاچه داده بر اساس الگوی <b>Schema-on-Read</b> عمل می‌کند؛ یعنی تمام داده‌های خام از هر منبعی بدون تغییر و با فرمت اصلی وارد مخزن متمرکز (مثل Amazon S3) می‌شوند. ساختار و طرح داده فقط زمانی تعریف می‌شود که یک تحلیل‌گر یا الگوریتم هوش مصنوعی بخواهد داده‌ها را بخواند و پردازش کند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>مقایسه کلیدی انبار داده و دریاچه داده:</b><br>
    &nbsp;&nbsp;&nbsp;• <b>انبار داده (Data Warehouse):</b> مناسب گزارش‌های مدیریتی و BI، سرعت کوئری بالا، هزینه ذخیره‌سازی بالا، ساختار صلب.<br>
    &nbsp;&nbsp;&nbsp;• <b>دریاچه داده (Data Lake):</b> مناسب یادگیری ماشین و پردازش Big Data، هزینه ذخیره‌سازی بسیار ارزان، انعطاف‌پذیری نامحدود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> نتفلیکس با دریاچه داده، داده‌های خام تمام کلیک‌ها و لوگ‌ها را با هزینه بسیار کم ذخیره کرده و بعداً برای آموزش مدل‌های هوش مصنوعی پیشنهاد فیلم از آن‌ها استفاده می‌کند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: ذخیره‌سازی ارزان داده‌های خام (JSON, CSV, Text) و اعمال ساختار در زمان خواندن (Schema-on-Read)
    </div>
  </p>

</div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json        
from io import StringIO 

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})

plt.rcParams['figure.dpi'] = 135

np.random.seed(42)
```

```python
display_academic_text(
    "معماری کلان‌داده: ذخیره‌سازی داده‌های خام و ناهمگن در «نتفلیکس» (Netflix Data Lake) 🌊", 
    "main_title"
)

display_academic_text(
    "پلتفرم «نتفلیکس» روزانه چندین پتابایت داده از منابع کاملاً متفاوت و ناهمگن تولید می‌کند:<br>"
    "<b>۱. ساخت‌یافته (Structured):</b> اطلاعات حساب کاربری و اشتراک‌ها در پایگاه‌های SQL.<br>"
    "<b>۲. نیمه‌ساخت‌یافته (Semi-Structured):</b> لاگ‌های کلیک، جستجو و مکثِ ویدیوها (فرمت JSON).<br>"
    "<b>۳. بدون ساختار (Unstructured):</b> نظرات متنی، پوسترها، فریم‌های ویدیو و فایل‌های صوتی.", 
    "paragraph"
)

display_academic_text(
    "چالش انبار داده سنتی (Data Warehouse) و راه‌حل دریاچه داده (Data Lake):", 
    "sub_title"
)

display_academic_text(
    "انبار داده نیازمند الگوی <b>Schema-on-Write</b> است؛ یعنی داده‌ها باید قبل از ذخیره‌سازی کاملاً پاک‌سازی و در جداول از پیش‌تعریف‌شده ساختاردهی شوند (ETL). "
    "این فرآیند برای پتابایت‌ها دادهٔ نیمه‌ساخت‌یافته به شدت گران، محدودکننده و زمان‌بر است.<br><br>"
    "در مقابل، <b>دریاچه داده (Data Lake)</b> بر اساس <b>Schema-on-Read</b> عمل می‌کند؛ تمام داده‌های خام بدون تغییر و با فرمت اصلی وارد مخزن (مثلاً Amazon S3) می‌شوند "
    "و ساختار (Schema) فقط و فقط زمانی روی آن‌ها اِعمال می‌شود که یک تحلیل‌گر یا الگوریتمِ هوش مصنوعی بخواهد آن‌ها را پردازش کند.", 
    "paragraph"
)

# ==========================================================
# SECTION 2: SIMULATING RAW HETEROGENEOUS DATA LAKE INGESTION
# ==========================================================
# Source A: Semi-structured JSON (User Clickstream Events)
json_clickstream_raw = (
    '[\n'
    '  {"user_id": 101, "event": "click_play", "movie": "Stranger Things", "timestamp": "2026-10-01 10:15:00"},\n'
    '  {"user_id": 102, "event": "pause", "movie": "The Crown", "timestamp": "2026-10-01 10:16:30"},\n'
    '  {"user_id": 101, "event": "search", "query": "sci-fi movies", "timestamp": "2026-10-01 10:20:00"}\n'
    ']'
)

# Source B: Structured CSV (Server Performance Metrics)
csv_server_logs_raw = (
    "server_id,cpu_utilization_pct,latency_ms,region\n"
    "srv-us-east-1,45.2,12,us-east\n"
    "srv-us-west-2,88.9,45,us-west\n"
    "srv-eu-central-1,62.1,22,eu-central\n"
)

# Source C: Unstructured Text (User Feedback Reviews)
text_reviews_raw = [
    "User 101: The 4K streaming quality is absolutely amazing, no lag at all!",
    "User 102: Buffering occurred frequently during peak hours. Needs fix."
]

display_academic_text(
    "<b>وضعیت ورودِ بی‌درنگِ داده‌ها به دریاچه داده (Data Ingestion Pipeline):</b><br>"
    "✅ داده‌های خام JSON (رویدادهای کلیک) مستقیماً در فضای Object Storage بارگذاری شدند.<br>"
    "✅ داده‌های خام CSV (لاگ سخت‌افزاری سرورها) در فضای Object Storage بارگذاری شدند.<br>"
    "✅ داده‌های متنی بدون ساختار (نظرات کاربران) با موفقیت به صورت فایل‌های خام (Raw) ذخیره گردیدند.", 
    "insight"
)

# ==========================================================
# SECTION 3: IMPLEMENTING SCHEMA-ON-READ ANALYTICS
# ==========================================================
display_academic_text(
    "اجرای الگوی Schema-on-Read (پردازش و ساختاردهی منحصراً در زمان خواندن)", 
    "sub_title"
)

# Querying Source A (Clickstream JSON parsed On-Demand from the Data Lake)
data_clickstream = json.loads(json_clickstream_raw)
df_clickstream = pd.DataFrame(data_clickstream).reset_index(drop=True)

# Querying Source B (Server Metrics CSV parsed On-Demand from the Data Lake)
df_server = pd.read_csv(StringIO(csv_server_logs_raw)).reset_index(drop=True)

# Querying Source C (Unstructured Reviews assigned a minimal wrapper schema upon read)
df_reviews = pd.DataFrame({'raw_text_review': text_reviews_raw}).reset_index(drop=True)

# Step 3: Display Structured Views logically projected from the Raw Data Lake
display_side_by_side_datasets(
    [
        df_clickstream[['user_id', 'event', 'timestamp']], 
        df_server[['server_id', 'cpu_utilization_pct', 'region']]
    ],
    [
        "۱. اعمال شماتیکِ درلحظه: رویدادهای کلیک (اِستخراج‌شده از JSON خام)", 
        "۲. اعمال شماتیکِ درلحظه: وضعیت سرورها (اِستخراج‌شده از CSV خام)"
    ]
)

display_side_by_side_datasets(
    [df_reviews],
    ["۳. استخراج متون خام جهت پردازش NLP: نظرات کاربران (Unstructured Text)"]
)

# ==========================================================
# SECTION 4: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
# To provide a complete analytical dashboard, we visualize the architecture's capabilities
fig, axes = plt.subplots(1, 2, figsize=(18, 6.5), dpi=135)

# ----------------------------------------------------------
# Subplot 1: Conceptual Storage Volume by Data Type
# ----------------------------------------------------------
data_types = ['Structured\n(SQL / Tables)', 'Semi-Structured\n(JSON / XML / Logs)', 'Unstructured\n(Video / Audio / Text)']
# Netflix Data Lake conceptually holds mostly Unstructured (Videos) and Semi-Structured (Logs)
storage_volume_pb = [15, 120, 850] 

bars = axes[0].bar(
    data_types, 
    storage_volume_pb, 
    color=['#2c3e50', '#2980b9', '#1abc9c'], # Navy, Classic Blue, Turquoise
    edgecolor='#2c3e50', 
    linewidth=2.0,
    width=0.55
)

axes[0].annotate(
    "The 80% Rule:\nOver 80% of enterprise data today is unstructured.\nA traditional Data Warehouse cannot store video\nfiles or raw text documents, but a Data Lake can.",
    xy=(2.0, 425), xytext=(0.5, 600),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[0].set_title('Figure 1: Data Lake Storage Distribution (Conceptual Volume)', fontsize=14, fontweight='bold', pad=15)
axes[0].set_ylabel('Storage Volume (Petabytes)', fontsize=13, fontweight='bold')
axes[0].set_xticklabels(data_types, fontweight='bold', fontsize=12)
axes[0].grid(axis='y', linestyle='--', alpha=0.5)

for bar in bars:
    yval = bar.get_height()
    axes[0].text(
        bar.get_x() + bar.get_width() / 2, 
        yval + 20, 
        f'{yval} PB', 
        ha='center', 
        fontweight='bold',
        fontsize=12,
        color='#2c3e50'
    )

# ----------------------------------------------------------
# Subplot 2: Capabilities Score (Data Lake vs Data Warehouse)
# ----------------------------------------------------------
categories_eval = ['Ingestion (Write) Speed\n[Accepting Raw Data]', 'Schema Flexibility\n[Agility for Data Scientists]', 'Query (Read) Speed\n[BI Dashboards]']

# Schema-on-Read (Lake): Fast write, highly flexible, but slower to query raw files directly
lake_scores = [5, 5, 2.5]  
# Schema-on-Write (Warehouse): Slow to write (due to heavy ETL), rigid, but extremely fast to query
warehouse_scores = [2, 2, 5]  

x = np.arange(len(categories_eval))
width = 0.35

axes[1].bar(
    x - width/2, 
    lake_scores, 
    width, 
    label='Data Lake (Schema-on-Read)', 
    color='#1abc9c', # Turquoise
    edgecolor='#2c3e50',
    linewidth=2.0
)

axes[1].bar(
    x + width/2, 
    warehouse_scores, 
    width, 
    label='Data Warehouse (Schema-on-Write)', 
    color='#2c3e50', # Navy
    edgecolor='#2c3e50',
    linewidth=2.0
)

axes[1].annotate(
    "Data Swamp Risk:\nBecause lakes store raw data, reading it\nrequires heavy computing power (Spark/Hadoop).\nDashboards are slower compared to Warehouses.",
    xy=(2.1, 2.5), xytext=(0.8, 3.8),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[1].set_title('Figure 2: Architecture Capabilities Comparison (Score 1-5)', fontsize=14, fontweight='bold', pad=15)
axes[1].set_xticks(x)
axes[1].set_xticklabels(categories_eval, fontweight='bold', fontsize=11)
axes[1].set_ylabel('Capability Score (Higher is Better)', fontsize=13, fontweight='bold')
axes[1].set_ylim(0, 6)
axes[1].legend(loc='upper right', frameon=True, fontsize=11, edgecolor='#bdc3c7')
axes[1].grid(axis='y', linestyle='--', alpha=0.5)

# Clear top and right spines for a clean academic aesthetic
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 5: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (Strategic Implication):</b><br>"
    r"الگوی <b>Schema-on-Read</b> در دریاچهٔ داده (Data Lake)، شرکت‌هایی نظیر نتفلیکس را از گلوگاه‌های سنگینِ پیش‌پردازش و ETL نجات می‌دهد. "
    r"با ذخیرهٔ <i>ارزان، خام و بدونِ قید و شرطِ</i> هر نوع داده (از جمله فیلم‌ها و لاگ‌های JSON)، نتفلیکس دیگر نیازی ندارد که پیشاپیش بداند قرار است چه سوالی از داده‌ها بپرسد!<br>"
    r"دانشمندان داده (Data Scientists) می‌توانند سال‌ها بعد، الگوریتم‌های جدیدِ یادگیری ماشین را روی تاریخچهٔ خام و دست‌نخوردهٔ نظرات و کلیک‌ها اجرا کنند؛ "
    r"قابلیتی که در <b>انبارهای دادهٔ سفت‌وسخت (Rigid Data Warehouses)</b> – که داده‌ها را در قالب‌هایِ جبری خلاصه و نابود می‌کنند – عملاً غیرممکن است. "
    r"با این حال، برای اتصال مستقیم داده‌ها به داشبوردهای مدیریتی سریع (BI)، همچنان به انبار داده (یا ساختارهای ترکیبی مانند <i>Lakehouse</i>) نیاز خواهیم داشت."
)
display_academic_text(insight_conclusion, "insight")
```
