  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: دسته‌بندی و محاسبات ریاضی معیارها در «وال‌مارت» (Walmart) 📊
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در مکعب‌های داده انبار داده «وال‌مارت» (Walmart)، تمامی اعداد جدول واقعیت (Fact Table) ماهیت یکسانی ندارند. برای جلوگیری از محاسبات اشتباه در گزارشات مدیریتی، تحلیل‌گران داده معیارهای عددی را در ۳ دسته اصلی طبقه‌بندی می‌کنند:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۱. معیارهای جمع‌پذیر (Additive Measures):</b><br>
    عددهایی هستند که می‌توان آن‌ها را روی تمام ابعاد (زمان، شهر، نوع محصول) با تابع $\text{SUM}()$ جمع بست بدون اینکه اشتباه ریاضی رخ دهد:<br>
    &nbsp;&nbsp;&nbsp;• <b>کل درآمد فروش (Total Revenue):</b> مجموع درآمد شعبه نیویورک + شیکاگو.<br>
    &nbsp;&nbsp;&nbsp;• <b>تعداد کالای فروخته‌شده (Total Quantity):</b> مجموع کالاهای فروخته‌شده در تمام روزها.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۲. معیارهای غیرجمع‌پذیر / نسبت‌ها (Non-Additive Measures):</b><br>
    عددهایی مانند «نسبت‌ها»، «درصدها» یا «میانگین حاشیه سود» را نمی‌توان با جمع زدن ساده تجمیع کرد! یک اشتباه رایج آماری این است که درصد حاشیه سود شعب را با هم جمع کنیم یا میانگین ساده بگیریم.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>اثبات ریاضی خطای جمع‌زدن نسبت‌ها:</b><br>
    اگر شعبه A با فروش ۱۰۰ دلار حاشیه سود ۱۰٪ داشته باشد ($10\$ \text{ Profit}$) و شعبه B با فروش ۱۰۰,۰۰۰ دلار حاشیه سود ۵۰٪ داشته باشد ($50,000\$ \text{ Profit}$)، میانگین ساده درصدها می‌شود ۳۰٪ ($\frac{10\% + 50\%}{2}$)؛ که کاملاً غلط است! زیرا حاشیه سود واقعی کل مجموعه برابر است با:
  </p>

  <div style="text-align: center; margin: 15px 0;">
    $$\text{Actual Profit Margin \%} = \frac{\sum \text{Total Profit}}{\sum \text{Total Revenue}} \times 100 = \frac{50,010}{100,100} \times 100 \approx 49.96\%$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۳. معیارهای محاسباتی (Computed Measures):</b><br>
    معیارهای جدیدی که بر اساس فرمول‌های ریاضی از معیارهای پایه مشتق می‌شوند (مانند $\text{Profit} = \text{Revenue} - \text{Cost}$).
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> تشخیص دقیق نوع معیار مانع از خطاهای محاسبه‌ای در سیستم‌های گزارش‌گیری OLAP و هوش تجاری می‌گردد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: جداسازی معیارهای جمع‌پذیر از نسبت‌های غیرجمع‌پذیر و اجرای تجمیع وزن‌دار درست
    </div>
  </p>

</div>

```python
display_academic_text(
    "دسته‌بندی و محاسبات ریاضیِ معیارها در انبار داده «وال‌مارت» (Walmart) 🧮", 
    "main_title"
)

display_academic_text(
    "در مکعب‌های دادهِ انبار داده (Data Warehouse)، تمامی اعداد در جدول واقعیت (Fact Table) ماهیتِ یکسانی ندارند. "
    "برای جلوگیری از محاسباتِ فاجعه‌بار در گزارشات مدیریتی، تحلیل‌گرانِ BI معیارهای عددی را در ۳ دستهٔ اصلی طبقه‌بندی می‌کنند:", 
    "paragraph"
)

display_academic_text("۱. معیارهای جمع‌پذیر (Additive Measures):", "sub_title")
display_academic_text(
    "اعداد مطلقی هستند که می‌توان آن‌ها را روی <i>تمامیِ ابعاد</i> (زمان، شهر، نوع محصول) با تابع <code>SUM()</code> جمع بست، بدون اینکه اشتباه ریاضی رخ دهد:<br>"
    "• <b>کل درآمد فروش (Total Revenue):</b> مجموع درآمد شعبه نیویورک + شیکاگو.<br>"
    "• <b>تعداد کالای فروخته‌شده (Quantity):</b> مجموع کالاهای فروخته‌شده امروز + دیروز.", 
    "paragraph"
)

display_academic_text("۲. معیارهای غیرجمع‌پذیر / نسبت‌ها (Non-Additive Measures):", "sub_title")
display_academic_text(
    "اعدادی مانند <b>«نسبت‌ها»، «درصدها» یا «میانگین حاشیه سود»</b> را نمی‌توان با جمع زدنِ ساده تجمیع کرد! "
    "یک اشتباهِ رایجِ آماری در داشبوردها این است که درصدِ حاشیه سودِ شعبِ مختلف را با هم جمع کنیم یا میانگین ساده (Simple Average) بگیریم.", 
    "paragraph"
)

display_academic_text("اثبات ریاضیِ خطای جمع‌زدن نسبت‌ها (Simpson's Paradox):", "sub_title")
display_academic_text(
    "اگر شعبه A با فروش ۱۰۰ دلار حاشیه سود ۱۰٪ داشته باشد (۱۰ دلار سود) و شعبه B با فروش ۱۰۰,۰۰۰ دلار حاشیه سود ۵۰٪ داشته باشد (۵۰,۰۰۰ دلار سود)، "
    "میانگینِ سادهٔ درصدها می‌شود ۳۰٪؛ که کاملاً <b>غلط</b> است! زیرا حاشیه سودِ <i>واقعیِ</i> کل مجموعه (میانگین وزن‌دار) برابر است با:", 
    "paragraph"
)

display_academic_text(
    r"$$\text{Simple Average (Wrong)} = \frac{10\% + 50\%}{2} = 30\%$$", 
    "paragraph"
)

display_academic_text(
    r"$$\text{Weighted Average (Correct)} = \frac{\sum \text{Total Profit}}{\sum \text{Total Revenue}} \times 100 = \frac{50,010}{100,100} \times 100 \approx 49.96\%$$", 
    "paragraph"
)

display_academic_text("۳. معیارهای محاسباتی در زمان اجرا (Computed Measures):", "sub_title")
display_academic_text(
    r"معیارهای جدیدی که بر اساس فرمول‌های ریاضی از معیارهای پایه <i>مشتق</i> می‌شوند (مانند $Profit = Revenue - Cost$).", 
    "paragraph"
)

# ==========================================================
# SECTION 2: DATA GENERATION AND MATHEMATICAL FORMULATIONS
# ==========================================================
display_academic_text(
    "مرحله نخست: تولید داده‌های فروش و تجمیع (Aggregation)", 
    "sub_title"
)

# Step 1: Create Walmart sales & cost dataset across Stores and Categories
data = {
    'Store': ['New York', 'New York', 'Chicago', 'Chicago', 'Los Angeles', 'Los Angeles'],
    'Category': ['Electronics', 'Apparel', 'Electronics', 'Apparel', 'Electronics', 'Apparel'],
    # Additive Measures
    'Revenue_USD': [150000, 20000, 80000, 15000, 220000, 30000],
    'Cost_USD': [100000, 15000, 60000, 12000, 130000, 18000],
    'Quantity_Sold': [300, 500, 160, 400, 450, 750],
}
df = pd.DataFrame(data)

# Step 2: Compute Base Computed Measure (Profit = Revenue - Cost)
df['Profit_USD'] = df['Revenue_USD'] - df['Cost_USD']

# Step 3: Compute Item-Level Non-Additive Measure (Profit Margin %)
df['Profit_Margin_Pct'] = (df['Profit_USD'] / df['Revenue_USD']) * 100

# Step 4: Correct vs Incorrect Aggregation by Store
# 4.1 Correct Additive Aggregation (SUM) for Revenue & Profit
store_agg = df.groupby('Store')[['Revenue_USD', 'Cost_USD', 'Profit_USD', 'Quantity_Sold']].sum()

# 4.2 WRONG Non-Additive Aggregation (Naive Mean of Percentages)
store_agg['Naive_Avg_Margin_Pct'] = df.groupby('Store')['Profit_Margin_Pct'].mean()

# 4.3 CORRECT Computed Non-Additive Aggregation (Weighted Margin %)
store_agg['Correct_Weighted_Margin_Pct'] = (store_agg['Profit_USD'] / store_agg['Revenue_USD']) * 100

# Prepare DataFrames for display to avoid hidden index columns
df_disp = df[['Store', 'Category', 'Revenue_USD', 'Profit_USD', 'Profit_Margin_Pct']].reset_index(drop=True)
store_agg_disp = store_agg[['Revenue_USD', 'Profit_USD', 'Naive_Avg_Margin_Pct', 'Correct_Weighted_Margin_Pct']].reset_index()

display_side_by_side_datasets(
    [df_disp],
    ["جدول ۱: داده‌های سطح کالا (Item-level) شامل معیارهای پایه و محاسباتی"]
)

display_side_by_side_datasets(
    [store_agg_disp],
    ["جدول ۲: مقایسه تجمیعِ درست (وزن‌دار) و نادرست (میانگین ساده) حاشیه سود به تفکیک شعبه"]
)

# ==========================================================
# SECTION 3: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی - توهمِ میانگینِ ساده (The Aggregation Fallacy)", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکیِ زیر نشان می‌دهد که چرا <b>متغیرهای نسبتی (مانند درصد)</b> به هیچ‌وجه نباید با <code>AVG()</code> در دیتابیس تجمیع شوند. "
    "در نمودار سمت راست، اگر میانگین ساده بگیریم، حاشیه سود در نیویورک ۲۹٪ گزارش می‌شود. اما این یک خطای فاحش است، زیرا فروش ۱۵۰ هزار دلاریِ الکترونیک "
    "(با حاشیه سود متفاوت) وزن بسیار بیشتری از فروش ۲۰ هزار دلاری پوشاک دارد. حاشیه سودِ واقعیِ شعبه (آبی فیروزه‌ای) ۳۲.۳۵٪ است!", 
    "paragraph"
)

# Visualization using Matplotlib
fig, axes = plt.subplots(1, 2, figsize=(16, 6.5), dpi=135)

x = np.arange(len(store_agg.index))
width = 0.35

# ----------------------------------------------------------
# Subplot 1: Additive Measures (Revenue and Profit Bar Chart)
# ----------------------------------------------------------
axes[0].bar(
    x - width / 2,
    store_agg['Revenue_USD'] / 1000,
    width,
    label='Total Revenue ($k)',
    color='#2980b9', # Classic Blue
    edgecolor='#2c3e50',
    linewidth=1.5
)
axes[0].bar(
    x + width / 2,
    store_agg['Profit_USD'] / 1000,
    width,
    label='Total Profit ($k)',
    color='#2c3e50', # Navy
    edgecolor='#2c3e50',
    linewidth=1.5
)

axes[0].set_title('Figure 1: Additive Measures (Safe SUM Aggregation)', fontsize=14, fontweight='bold', pad=15)
axes[0].set_xticks(x)
axes[0].set_xticklabels(store_agg.index, fontweight='bold', fontsize=11)
axes[0].set_ylabel('Amount in $1,000s USD', fontweight='bold', fontsize=12)
axes[0].legend(loc='upper left', frameon=True, fontsize=11)
axes[0].grid(axis='y', linestyle='--', alpha=0.5)

# ----------------------------------------------------------
# Subplot 2: Non-Additive Comparison (Naive Mean vs Correct Weighted)
# ----------------------------------------------------------
axes[1].bar(
    x - width / 2,
    store_agg['Naive_Avg_Margin_Pct'],
    width,
    label='Naive Simple Average (INCORRECT)',
    color='#e74c3c', # Red Alert
    edgecolor='#c0392b',
    linewidth=1.5,
)
axes[1].bar(
    x + width / 2,
    store_agg['Correct_Weighted_Margin_Pct'],
    width,
    label='Weighted Margin (CORRECT)',
    color='#1abc9c', # Turquoise (Safe)
    edgecolor='#16a085',
    linewidth=1.5,
)

axes[1].annotate(
    "The Fallacy of Averages:\nThe simple average (red) ignores that Electronics\ngenerated $150k while Apparel generated only $20k.\nThe true branch margin (green) is heavily pulled\ntowards the Electronics margin.",
    xy=(0.15, 32.35), xytext=(0.0, 45),
    fontsize=11, fontweight='bold', color='#2c3e50',
    arrowprops=dict(arrowstyle="->", color="#34495e", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#ecf0f1", ec="#bdc3c7", lw=1.5)
)

axes[1].set_title('Figure 2: Non-Additive Measures (Aggregation Fallacy)', fontsize=14, fontweight='bold', pad=15)
axes[1].set_xticks(x)
axes[1].set_xticklabels(store_agg.index, fontweight='bold', fontsize=11)
axes[1].set_ylabel('Profit Margin (%)', fontweight='bold', fontsize=12)
axes[1].set_ylim(0, 55)
axes[1].legend(loc='lower right', frameon=True, fontsize=11)
axes[1].grid(axis='y', linestyle='--', alpha=0.5)

# Clear top and right spines for a clean academic aesthetic
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک در هوش تجاری (Strategic Implication):</b><br>"
    r"تشخیص دقیق نوع معیار، مانع از خطاهای مخرب محاسبه‌ای در داشبوردهای مدیریتی (BI) می‌گردد. تجمیع معیارهای <b>غیرجمع‌پذیر (Non-Additive)</b> مانند درصدها "
    r"نباید هرگز با توابع <code>SUM()</code> یا <code>AVG()</code> در دیتابیس انجام شود.<br>"
    r"سیستم‌های پردازش تحلیلی آنلاین (OLAP) همیشه باید در سطح خُرد، اعداد مطلق و جمع‌پذیر (مانند <i>Profit</i> و <i>Revenue</i>) را با تابع <code>SUM()</code> جمع کنند، "
    r"و فقط در <b>گام نهایی (در سطح داشبورد و پس از تجمیع)</b>، فرمولِ محاسبهٔ نسبت (تقسیم سود کل بر درآمد کل) را اجرا نمایند تا میانگینِ وزن‌دارِ دقیق به دست آید."
)
display_academic_text(insight_conclusion, "insight")
```
