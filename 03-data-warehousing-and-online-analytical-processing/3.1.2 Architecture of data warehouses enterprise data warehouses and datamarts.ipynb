  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: طراحی لایه‌های معماری انبار داده و بازارهای داده در شرکت «اوبر» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    شرکت هوشمند حمل‌ونقل «اوبر» (Uber) روزانه میلیون‌ها سفر را در سراسر جهان پردازش می‌کند. برای سازمان‌دهی این داده‌های عظیم و پاسخ به نیازهای تحلیلی واحدهای مختلف، سیستم انبار داده اوبر در چهار لایه اصلی معماری می‌شود:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    چهار لایه اصلی معماری انبار داده:<br>
    ۱. لایه منابع داده (Data Sources): جمع‌آوری داده‌ها از اپلیکیشن مسافران، سیستم رانندگان، تبلیغات و پشتیبانی.<br>
    ۲. لایه ETL یا ELT: استخراج، پاک‌سازی، یکپارچه‌سازی و تبدیل فرمت داده‌های خام.<br>
    ۳. لایه ذخیره‌سازی مرکزی (Enterprise Data Warehouse - EDW): مخزن متمرکز داده‌های هماهنگ‌شده سازمان در قالب شمای ستاره‌ای یا برف‌دانه.<br>
    ۴. لایه ارائه و بازارهای داده (Data Marts): تحویل زیرمجموعه‌های تخصصی داده به بخش‌های مختلف (مثل بازار داده مالی برای تیم حسابداری و بازار داده بازاریابی برای تیم تبلیغات).
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    رویکردهای اصلی در طراحی معماری انبار داده:<br>
    رویکرد اینمون (Bill Inmon - بالا به پایین): ابتدا یک انبار داده مرکزی متمرکز و جامع (EDW) به‌صورت کاملاً نرمال‌شده ساخته می‌شود و سپس بازارهای داده مورد نیاز هر واحد از آن استخراج می‌گردند. این روش یکپارچگی فوق‌العاده‌ای دارد اما ساخت اولیه آن زمان‌بر است.<br>
    رویکرد کیمبال (Ralph Kimball - پایین به بالا): ابتدا بازارهای داده مجزا (Data Marts) بر اساس مدل‌سازی چندبعدی ساخته می‌شوند و سپس با ابعاد هم‌شکل (Conformed Dimensions) به یکدیگر وصل شده و انبار داده را شکل می‌دهند. این روش بسیار سریع‌تر به بهره‌برداری می‌رسد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    معماری‌های مدرن ابری و دریاچه‌انبار (Lakehouse):<br>
    امروزه پلتفرم‌های ابری مدرن مانند Snowflake و BigQuery، پردازش محاسباتی را از ذخیره‌سازی جدا کرده‌اند. ترکیب انبار داده و دریاچه داده در قالب دریاچه‌انبار (Lakehouse) امکان تحلیل هم‌زمان داده‌های ساخت‌یافته و ناهمگن را با هزینه بسیار کم فراهم می‌سازد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: تفکیک انبار داده اصلی به بازارهای داده تخصصی، سرعت کوئری‌های تحلیلی هر واحد سازمان را به‌شدت افزایش می‌دهد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: جداسازی لایه‌ای داده‌ها و ایجاد بازارهای داده اختصاصی برای افزایش سرعت گزارش‌گیری تحلیلی
    </div>
  </p>

</div>

```python
display_academic_text(
    "طراحی لایه‌های معماری انبار داده و بازارهای داده (Data Marts) در شرکت «اوبر» 🚖", 
    "main_title"
)

display_academic_text(
    "شرکت هوشمند حمل‌ونقل «اوبر» (Uber) روزانه میلیون‌ها سفر را در سراسر جهان پردازش می‌کند. برای سازمان‌دهی این داده‌های عظیم "
    "و پاسخ سریع به نیازهای تحلیلی واحدهای مختلف، سیستم انبار داده اوبر در چهار لایهٔ اصلی معماری می‌شود.", 
    "paragraph"
)

display_academic_text("چهار لایه اصلی معماری انبار داده:", "sub_title")
display_academic_text(
    "۱. <b>لایه منابع داده (Data Sources):</b> جمع‌آوری داده‌ها از اپلیکیشن مسافران، سیستم رانندگان، تبلیغات و پشتیبانی.<br>"
    "۲. <b>لایه ETL یا ELT:</b> استخراج، پاک‌سازی، یکپارچه‌سازی و تبدیل فرمت داده‌های خام.<br>"
    "۳. <b>لایه ذخیره‌سازی مرکزی (EDW):</b> مخزن متمرکز داده‌های هماهنگ‌شده سازمان در قالب شمای ستاره‌ای (Star Schema).<br>"
    "۴. <b>لایه ارائه و بازارهای داده (Data Marts):</b> تحویل زیرمجموعه‌های تخصصی و سبک‌شدهٔ داده به بخش‌های مختلف (مثل تیم مالی یا بازاریابی).", 
    "paragraph"
)

display_academic_text("رویکردهای اصلی معماری:", "sub_title")
display_academic_text(
    "<b>رویکرد اینمون (Inmon - بالا به پایین):</b> ابتدا انبار دادهٔ مرکزی (EDW) ساخته می‌شود و سپس بازارهای داده (Marts) از آن استخراج می‌گردند. یکپارچگیِ بالا اما پیاده‌سازیِ زمان‌بر دارد.<br>"
    "<b>رویکرد کیمبال (Kimball - پایین به بالا):</b> ابتدا بازارهای دادهٔ مجزا ساخته شده و سپس به هم متصل می‌شوند تا انبار داده را شکل دهند.", 
    "paragraph"
)

# ==========================================================
# SECTION 2: SIMULATING CENTRAL EDW & EXTRACTING DATA MARTS
# ==========================================================
display_academic_text(
    "مرحله نخست: پیاده‌سازی رویکرد اینمون (استخراج Data Mart از EDW)", 
    "sub_title"
)

display_academic_text(
    "در این بخش، ابتدا یک انبار دادهٔ مرکزی (EDW) حاوی صد هزار رکورد با <b>تمامی ستون‌ها</b> (راننده، مالی، پروموشن و...) ساخته می‌شود. "
    "سپس برای جلوگیری از خواندنِ ستون‌های اضافی از روی هارد دیسک، یک <b>بازار داده مالی (Finance Data Mart)</b> که فقط حاوی ستون‌های مرتبط با پول است، استخراج می‌گردد.", 
    "paragraph"
)

# Step 1: Simulate Central Enterprise Data Warehouse (EDW)
n_trips = 100000

# SQLite In-memory Database for EDW and Data Marts
conn_edw = sqlite3.connect(':memory:')
conn_mart = sqlite3.connect(':memory:') # Dedicated connection for the Mart

# EDW Central Fact Table - Contains all enterprise metrics
df_edw = pd.DataFrame({
    'Trip_ID': range(100001, 100001 + n_trips),
    'Rider_ID': np.random.randint(1, 5000, size=n_trips),
    'Driver_ID': np.random.randint(101, 1000, size=n_trips),
    'City': np.random.choice(['New York', 'London', 'Tokyo', 'Paris'], size=n_trips),
    'Fare_USD': np.random.uniform(10.0, 150.0, size=n_trips),
    'Promo_Code_Used': np.random.choice(['SUMMER20', 'WELCOME', 'NONE'], size=n_trips),
    'Driver_Rating': np.random.uniform(3.5, 5.0, size=n_trips),
    'Trip_Distance_KM': np.random.uniform(2.0, 45.0, size=n_trips),
    'Tx_Date': np.random.choice(['2023-01-15', '2023-06-20', '2023-11-05'], size=n_trips)
})

df_edw.to_sql('Fact_Enterprise_Trips', conn_edw, index=False, if_exists='replace')

# Step 2: Build Specialized Data Marts from Central EDW (Inmon Approach)
# Data Mart 1: Finance Data Mart (Only extracts financial columns)
query_finance_mart = "SELECT Trip_ID, City, Fare_USD, Tx_Date FROM Fact_Enterprise_Trips"
df_mart_finance = pd.read_sql_query(query_finance_mart, conn_edw)

# Store the Data Mart in its own dedicated database to simulate isolated architecture
df_mart_finance.to_sql('Mart_Finance', conn_mart, index=False, if_exists='replace')

# Render sample rows to show the structural difference
display_side_by_side_datasets(
    [df_edw.head(4).reset_index(drop=True), df_mart_finance.head(4).reset_index(drop=True)],
    [
        f"انبار داده مرکزی (EDW) - مخزن سنگین با ۹ ستون", 
        f"بازار داده مالی (Finance Mart) - مخزن سبک با ۴ ستون"
    ]
)

# ==========================================================
# SECTION 3: BENCHMARK ANALYTICAL QUERY EXECUTION SPEED
# ==========================================================
display_academic_text(
    "مرحله دوم: بنچمارک سرعت (چرا Data Martها ضروری هستند؟)", 
    "sub_title"
)

display_academic_text(
    "اکنون تیم مالی می‌خواهد مجموع درآمد به تفکیک هر شهر را محاسبه کند. این کوئری را یک بار روی EDWِ سنگین، "
    "و یک بار روی Data Martِ سبک اجرا می‌کنیم تا تأثیر I/Oِ دیسک را بر سرعت تحلیل مشاهده نماییم.", 
    "paragraph"
)

# Query: Calculate Total Revenue by City
n_benchmark_runs = 50

# Benchmark 1: Query execution on full heavy EDW Table
query_edw = "SELECT City, SUM(Fare_USD) as Total_Revenue FROM Fact_Enterprise_Trips GROUP BY City"
start_time = time.time()
for _ in range(n_benchmark_runs):
    pd.read_sql_query(query_edw, conn_edw)
time_edw = (time.time() - start_time) * 1000

# Benchmark 2: Query execution on slim Finance Data Mart
query_mart = "SELECT City, SUM(Fare_USD) as Total_Revenue FROM Mart_Finance GROUP BY City"
start_time = time.time()
for _ in range(n_benchmark_runs):
    pd.read_sql_query(query_mart, conn_mart)
time_mart = (time.time() - start_time) * 1000

speedup = time_edw / time_mart

display_academic_text(
    f"<b>نتایج بنچمارک برای {n_benchmark_runs} کوئری تحلیلی:</b><br>"
    f"⏱️ زمان جستجو در انبار داده مرکزی (EDW سنگین): <b>{time_edw:.1f} میلی‌ثانیه</b><br>"
    f"⏱️ زمان جستجو در بازار داده مالی (Marts سبک): <b>{time_mart:.1f} میلی‌ثانیه</b><br>"
    f"🚀 <b>بهبود عملکرد:</b> <b>{speedup:.1f} برابر سریع‌تر</b> با استفاده از بازار داده (کاهش بار I/O).", 
    "insight"
)

# ==========================================================
# SECTION 4: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 6.5), dpi=135)

# --- SUBPLOT 1: Query Speed Comparison (EDW vs Data Mart) ---
architectures = ['Central EDW\n(9 Columns on Disk)', 'Finance Data Mart\n(4 Columns on Disk)']
times = [time_edw, time_mart]

bars1 = axes[0].bar(architectures, times, color=['#e74c3c', '#1abc9c'], edgecolor='#2c3e50', linewidth=2.0, width=0.5)

axes[0].annotate(
    "I/O Bottleneck:\nEven if we only select 2 columns in our SQL query,\nthe database engine still has to read all 9 heavy columns\nfrom the EDW hard disk, slowing down the system.",
    xy=(0.0, time_edw/2), xytext=(0.6, time_edw/1.5),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title(f'Figure 1: Analytical Query Execution Time ({n_benchmark_runs} Runs)', fontsize=14, fontweight='bold', pad=15)
axes[0].set_ylabel('Execution Time (Milliseconds) - Lower is Better', fontsize=13, fontweight='bold')
axes[0].grid(axis='y', linestyle='--', alpha=0.5)
axes[0].set_xticklabels(architectures, fontweight='bold', fontsize=11)

for bar in bars1:
    yval = bar.get_height()
    axes[0].text(bar.get_x() + bar.get_width() / 2, yval + (time_edw * 0.02), f'{yval:.1f} ms', ha='center', fontweight='bold', fontsize=13, color='#2c3e50')


# --- SUBPLOT 2: Data Mart Output (Revenue by City) ---
df_city_revenue = pd.read_sql_query(query_mart, conn_mart).sort_values(by='Total_Revenue', ascending=False)

bars2 = axes[1].bar(df_city_revenue['City'], df_city_revenue['Total_Revenue'] / 1000, color='#2980b9', edgecolor='#2c3e50', linewidth=2.0, width=0.6)

axes[1].set_title('Figure 2: Finance Data Mart Report (Total Revenue by City)', fontsize=14, fontweight='bold', pad=15)
axes[1].set_xlabel('Global Operating City', fontsize=13, fontweight='bold')
axes[1].set_ylabel('Total Revenue Generated ($k USD)', fontsize=13, fontweight='bold')
axes[1].grid(axis='y', linestyle='--', alpha=0.5)
axes[1].set_xticklabels(df_city_revenue['City'], fontweight='bold', fontsize=11)
axes[1].set_ylim(0, max(df_city_revenue['Total_Revenue'] / 1000) * 1.2) # Headroom for labels

for i, val in enumerate(df_city_revenue['Total_Revenue'] / 1000):
    axes[1].text(i, val + (max(df_city_revenue['Total_Revenue'] / 1000) * 0.03), f'${val:.1f}k', ha='center', fontweight='bold', fontsize=12, color='#2c3e50')

# Clear top and right spines for a clean academic aesthetic
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# Close database connections safely
conn_edw.close()
conn_mart.close()

# ==========================================================
# SECTION 5: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک معماری داده (Strategic Implication):</b><br>"
    r"تفکیک انبار داده اصلی به <b>بازارهای داده تخصصی (Data Marts)</b>، سرعت کوئری‌های تحلیلی هر واحد از سازمان را به‌شدت افزایش می‌دهد. "
    r"دلیل این امر، پدیدهٔ <b>I/O Bottleneck</b> در دیتابیس‌های رابطه‌ای (Row-oriented) است؛ موتور پایگاه داده برای خواندن ۲ ستون، مجبور است تمام ۹ ستون را از روی هارد دیسک اسکن کند (Figure 1).<br>"
    r"با ساخت یک Data Mart سبک (که در آن ستون‌های سنگین متنی مثل پروموشن‌ها و ریتینگ‌ها حذف شده‌اند)، معماریِ سیستم موفق شد "
    r"بار خواندن/نوشتن دیسک (I/O) را کاهش داده و زیرساختی مقیاس‌پذیر، پرسرعت و امن برای پردازش داده‌های کلان در شرکت‌هایی نظیر اوبر فراهم آورد."
)
display_academic_text(insight_conclusion, "insight")
```
