<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: پردازش هوشمند پرس‌وجوهای تحلیلی در «استارباکس» با استفاده از جوهره‌ها 📊
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    کافه‌های زنجیره‌ای «استارباکس» (Starbucks) روزانه میلیون‌ها رسید خرید ثبت می‌کنند. وقتی مدیران سیستم بخواهند گزارش‌های تحلیلی مختلفی تهیه کنند، موتور OLAP به جای اسکن مجدد تمام رسیدهای خام، پرس‌وجو را به مناسب‌ترین و کوچک‌ترین <b>جوهره (Cuboid)</b> از پیش محاسبه‌شده هدایت می‌کند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    شبکه جوهره‌ها (Lattice of Cuboids) در مکعب داده استارباکس:<br>
    فرض کنید داده‌ها دارای ۳ بعد شهر (Location)، محصول (Product) و زمان (Year) باشند. در این صورت شبکه جوهره‌ها شامل سطوح زیر است:
  </p>

  <ul style="text-align: right; margin: 0 0 10px; padding-right: 22px;">
    <li>جوهره پایه (۳ بعدی): ترکیب کامل (Location, Product, Year) شامل جزئی‌ترین داده‌ها.</li>
    <li>جوهره‌های ۲ بعدی: ترکیب‌های (Location, Product) یا (Location, Year) یا (Product, Year).</li>
    <li>جوهره‌های ۱ بعدی: تجمیع‌های تک بعدی روی (Location) یا (Product) یا (Year).</li>
    <li>جوهره رأس (۰ بعدی): تجمیع کل فروش تمام شعب در تمام زمان‌ها (Apex Cuboid).</li>
  </ul>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مکانیسم پردازش هوشمند پرس‌وجو (Query Matching):<br>
    اگر مدیری میزان فروش محصول لاتین در شهر لندن را بخواهد (بدون توجه به سال)، موتور OLAP هوشمندانه جوهره دو بعدی (Location, Product) را انتخاب می‌کند. این جوهره از قبل تجمیع شده و بسیار کوچک‌تر از جدول خام یا جوهره پایه سه بعدی است؛ در نتیجه کوئری با حداقل اسکن حافظه پاسخ داده می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: انتخاب متناظرترین جوهره پیش‌محاسبه‌شده برای هر پرس‌وجو، بار محاسباتی سرورها را به حداقل می‌رساند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: پاسخ‌گویی به پرس‌وجوهای OLAP با هدایت مستقیم به کوچک‌ترین جوهره مرتبط در شبکه مکعب‌واره‌ها
    </div>
  </p>

</div>

```python
display_academic_text(
    "پردازش هوشمند پرس‌وجوهای تحلیلی در «استارباکس» با استفاده از شبکه جوهره‌ها (Lattice of Cuboids) 📊", 
    "main_title"
)

display_academic_text(
    "کافه‌های زنجیره‌ای «استارباکس» (Starbucks) روزانه میلیون‌ها رسید خرید ثبت می‌کنند. "
    "وقتی مدیران بخواهند گزارش‌های تحلیلی مختلفی تهیه کنند، موتور هوشمند OLAP به جای اسکن مجدد تمام رسیدهای خام در دیتابیس، "
    "<b>پرس‌وجو (Query) را به مناسب‌ترین، مرتبط‌ترین و کوچک‌ترین جوهرهٔ (Cuboid) از پیش محاسبه‌شده هدایت (Route) می‌کند</b>.", 
    "paragraph"
)

display_academic_text("شبکه جوهره‌ها در مکعب دادهٔ استارباکس (دارای ۳ بُعد: مکان، محصول، زمان):", "sub_title")

display_academic_text(
    "• <b>جوهره پایه (۳ بُعدی - Base Cuboid):</b> ترکیب کامل ابعاد شامل جزئی‌ترین داده‌ها.<br>"
    "• <b>جوهره‌های ۲ بُعدی (2D Cuboids):</b> ترکیب‌های دوگانه مانند (مکان، محصول) یا (مکان، سال). بُعد سوم در این مکعب‌ها فشرده شده است.<br>"
    "• <b>جوهره‌های ۱ بُعدی (1D Cuboids):</b> تجمیع‌های کلان روی یک بُعدِ تنها، مثلاً کُلِ فروش هر محصول.<br>"
    "• <b>جوهره رأس (۰ بُعدی - Apex):</b> تجمیع کُلِ فروش تمامِ شعب در تمام زمان‌ها و محصولات (تنها یک عدد مطلق).", 
    "paragraph"
)


# ==========================================================
# SECTION 2: DATA GENERATION AND CUBOIDS PRECOMPUTATION
# ==========================================================
display_academic_text("مرحله نخست: تولید تراکنش‌ها و متریالیزه کردن شبکه جوهره‌ها در رَم", "sub_title")

# Increase rows to 500,000 for realistic performance testing
n_receipts = 500000

cities = ['London', 'New York', 'Tokyo', 'Paris']
products = ['Latte', 'Espresso', 'Mocha', 'Cappuccino']
years = [2024, 2025, 2026]

# Step 1: Create Starbucks Raw Transaction Dataset
df_starbucks = pd.DataFrame({
    'Location': np.random.choice(cities, size=n_receipts),
    'Product': np.random.choice(products, size=n_receipts),
    'Year': np.random.choice(years, size=n_receipts),
    'Amount_USD': np.random.uniform(3.5, 8.5, size=n_receipts)
})

display_side_by_side_datasets(
    [df_starbucks.head(5).reset_index(drop=True)],
    [f"رسیدهای خام استارباکس (پایگاه داده اصلی حاوی {n_receipts:,} رکورد تراکنشی)"]
)

# Step 2: Build Precomputed Lattice of Cuboids (In-Memory Materialization)

# 3D Base Cuboid (Year is kept)
cuboid_3d = df_starbucks.groupby(['Location', 'Product', 'Year'])['Amount_USD'].sum()

# 2D Cuboids (Year is completely rolled-up / summed out)
cuboid_2d_loc_prod = df_starbucks.groupby(['Location', 'Product'])['Amount_USD'].sum()

# 1D Cuboids
cuboid_1d_product = df_starbucks.groupby('Product')['Amount_USD'].sum()

display_academic_text(
    f"<b>حجم جوهره‌های محاسبه‌شده در حافظه (Search Space Size):</b><br>"
    f"حجم جدول خام و رابطه‌ای (Raw Table): <b>{len(df_starbucks):,} ردیف</b><br>"
    f"حجم جوهره ۳ بُعدی (Base Cuboid): <b>{len(cuboid_3d)} سلول حافظه</b><br>"
    f"حجم جوهره ۲ بُعدی (مکان + محصول): <b>{len(cuboid_2d_loc_prod)} سلول حافظه</b><br>"
    f"حجم جوهره ۱ بُعدی (فقط محصول): <b>{len(cuboid_1d_product)} سلول حافظه</b>", 
    "insight"
)

# ==========================================================
# SECTION 3: SIMULATING QUERY PROCESSING & OPTIMAL ROUTING
# ==========================================================
display_academic_text("مرحله دوم: تست سرعت مسیریابی (Query Routing Benchmark)", "sub_title")

# Query: "Find total sales of Latte in London (ACROSS ALL YEARS)"
n_runs = 100

# ----------------------------------------------------------
# Method 1: Scan Raw Fact Table (O(N) full scan)
# ----------------------------------------------------------
start = time.perf_counter()
for _ in range(n_runs):
    res_raw = df_starbucks[
        (df_starbucks['Location'] == 'London') & 
        (df_starbucks['Product'] == 'Latte')
    ]['Amount_USD'].sum()
time_raw = (time.perf_counter() - start) * 1000

# ----------------------------------------------------------
# Method 2: Query 3D Base Cuboid (Sub-optimal routing)
# Since the query doesn't specify 'Year', the engine must extract the 3 years 
# and AGGREGATE them on the fly.
# ----------------------------------------------------------
start = time.perf_counter()
for _ in range(n_runs):
    res_3d = cuboid_3d.loc[('London', 'Latte')].sum()
time_3d = (time.perf_counter() - start) * 1000

# ----------------------------------------------------------
# Method 3: Query Optimal 2D Cuboid (Perfect O(1) matching)
# The 2D cuboid has already aggregated the 'Year' dimension in the background!
# ----------------------------------------------------------
start = time.perf_counter()
for _ in range(n_runs):
    res_2d = cuboid_2d_loc_prod.loc[('London', 'Latte')]
time_2d = (time.perf_counter() - start) * 1000

speedup = time_raw / time_2d

display_academic_text(
    f"<b>نتایج بنچمارک - پاسخ به کوئری (Total Latte in London) برای {n_runs} بار اجرا:</b><br>"
    f"🔴 زمان اسکنِ جدول خام (Raw Scan): <b>{time_raw:.2f} میلی‌ثانیه</b><br>"
    f"🔵 زمان پیمایش مکعب ۳ بُعدی (مسیریابی <i>غیربهینه</i> - نیاز به تجمیع درلحظه): <b>{time_3d:.2f} میلی‌ثانیه</b><br>"
    f"🟢 زمان دسترسی به مکعب ۲ بُعدی (مسیریابی <b>بهینه</b> - خواندن مستقیم): <b>{time_2d:.2f} میلی‌ثانیه</b><br>"
    f"🚀 <b>بهبود عملکرد سیستم:</b> مسیریابی هوشمند در شبکه لتیس، پاسخگویی را <b>{speedup:.1f} برابر سریع‌تر</b> کرد.", 
    "paragraph"
)

# ==========================================================
# SECTION 4: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 6.5), dpi=135)

structures = ['Raw Fact Table\n(Unindexed)', '3D Base Cuboid\n(Loc, Prod, Year)', 'Optimal 2D Cuboid\n(Loc, Prod)']

# ----------------------------------------------------------
# Subplot 1: Rows/Cells Scanned per Cuboid (Log Scale)
# ----------------------------------------------------------
rows_scanned = [len(df_starbucks), len(cuboid_3d), len(cuboid_2d_loc_prod)]

bars1 = axes[0].bar(
    structures, 
    rows_scanned, 
    color=['#e74c3c', '#2980b9', '#1abc9c'], # Red, Classic Blue, Turquoise
    edgecolor='#2c3e50', 
    linewidth=2.0,
    width=0.5
)

axes[0].annotate(
    "Search Space Reduction:\nBy routing the query to the pre-aggregated\n2D Cuboid, the system only had to search\nthrough 16 cells instead of 500,000 rows.",
    xy=(2.0, len(cuboid_2d_loc_prod)), xytext=(0.5, 1000),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[0].set_title('Figure 1: Search Space Size / Cells Scanned per Query', fontsize=14, fontweight='bold', pad=15)
axes[0].set_ylabel('Number of Scanned Records (Logarithmic Scale)', fontsize=13, fontweight='bold')
axes[0].set_yscale('log')
axes[0].grid(axis='y', linestyle='--', alpha=0.5)
axes[0].set_xticklabels(structures, fontweight='bold', fontsize=12)

for bar in bars1:
    yval = bar.get_height()
    axes[0].text(
        bar.get_x() + bar.get_width() / 2, 
        yval * 1.5, # Offset for log scale
        f'{yval:,} Cells', 
        ha='center', fontweight='bold', fontsize=13, color='#2c3e50'
    )

# ----------------------------------------------------------
# Subplot 2: Execution Time Comparison
# ----------------------------------------------------------
exec_times = [time_raw, time_3d, time_2d]

bars2 = axes[1].bar(
    structures, 
    exec_times, 
    color=['#e74c3c', '#2980b9', '#1abc9c'], 
    edgecolor='#2c3e50', 
    linewidth=2.0,
    width=0.5
)

axes[1].annotate(
    "Sub-Optimal Routing Cost:\nThe 3D Cuboid is fast, but because it still\ncontains the 'Year' dimension, the CPU had\nto aggregate 3 years on-the-fly. The 2D Cuboid\nhad already squashed the years, so it was instant.",
    xy=(1.0, time_3d), xytext=(0.1, time_raw * 0.6),
    fontsize=11, fontweight='bold', color='#2980b9',
    arrowprops=dict(arrowstyle="->", color="#2980b9", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#ecf0f1", ec="#bdc3c7", lw=1.5)
)

axes[1].set_title(f'Figure 2: Query Execution Time Comparison ({n_runs} Runs)', fontsize=14, fontweight='bold', pad=15)
axes[1].set_ylabel('Execution Time (Milliseconds)', fontsize=13, fontweight='bold')
axes[1].grid(axis='y', linestyle='--', alpha=0.5)
axes[1].set_xticklabels(structures, fontweight='bold', fontsize=12)

for bar in bars2:
    yval = bar.get_height()
    axes[1].text(
        bar.get_x() + bar.get_width() / 2, 
        yval + (max(exec_times) * 0.02), 
        f'{yval:.1f} ms', 
        ha='center', fontweight='bold', fontsize=13, color='#2c3e50'
    )

# Clear top and right spines for a clean academic aesthetic
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout(pad=2.0)
plt.show()

# ==========================================================
# SECTION 5: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک در معماری موتورهای تحلیلی (Strategic Implication):</b><br><br>"
    r"مکانیسم <b>مسیریابی هوشمند پرس‌وجو (Query Routing)</b> قلب تپندهٔ موتورهای تجاریِ OLAP (مانند SSAS یا ClickHouse) است.<br><br>"
    r"اگر یک مدیر، کُلِ فروش محصولِ لاته را در شهر لندن بخواهد (بدون اشاره به سال)، موتور OLAP از خواندنِ مکعب ۳ بُعدیِ حاوی سال امتناع کرده "
    r"و کوئری را مستقیماً به جوهرهٔ ۲ بُعدی (مکان، محصول) هدایت می‌کند. انتخابِ متناظرترین و فشرده‌ترین جوهرهٔ پیش‌محاسبه‌شده، بار محاسباتی (Aggregation) را در زمانِ اجرایِ داشبورد "
    r"کاملاً حذف کرده و زمان اسکن را از مرزِ $O(N)$ به یک جستجوی <b>حافظه‌ایِ بی‌درنگ با زمانِ $O(1)$</b> ارتقا می‌دهد."
)
display_academic_text(insight_conclusion, "insight")
```
