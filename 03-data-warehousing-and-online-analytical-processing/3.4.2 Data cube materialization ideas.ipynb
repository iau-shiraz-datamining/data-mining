  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: راهبردهای متریالیزه کردن مکعب داده در وال‌مارت برای افزایش سرعت کوئری 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در سیستم‌های انبار داده بزرگ مانند وال‌مارت (Walmart)، اگر یک مکعب داده دارای $N$ بعد مختلف باشد، تعداد کل ترکیب‌های تجمیعی ممکن (مکعب‌واره‌ها یا Cuboids) برابر با $2^N$ خواهد بود. برای مثال اگر ۱۰ بعد داشته باشیم، $2^{10} = 1024$ مکعب‌واره مختلف تولید می‌شود. محاسبه آنی این تجمیع‌ها در زمان اجرای کوئری بسیار کند است. برای حل این چالش از راهکارهای <b>متریالیزه کردن (Materialization)</b> یعنی پیش‌محاسبه و ذخیره‌سازی تجمیع‌ها استفاده می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. متریالیزه کامل (Full Materialization): تمام $2^N$ حالت تجمیع ممکن از قبل محاسبه و روی دیسک ذخیره می‌شوند. سرعت کوئری فوق‌العاده بالاست، اما فضای ذخیره‌سازی بسیار زیادی مصرف می‌کند (انفجار فضای ذخیره‌سازی).
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. متریالیزه جزئی (Partial Materialization): تنها مکعب‌واره‌هایی که بیشترین الگوی کوئری و تقاضا را دارند از قبل محاسبه می‌شوند. این روش تعادلی بهینه بین سرعت اجرای کوئری و حافظه مصرفی ایجاد می‌کند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. متریالیزه Top-N: تنها N تجمیع پرکاربرد و پرپرسش متریالیزه شده و بقیه تجمیع‌ها در صورت نیاز در زمان اجرا محاسبه می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. متریالیزه افزایشی (Incremental Materialization): هنگام ورود داده‌های جدید، به جای محاسبه مجدد کل مکعب داده، تنها تغییرات جدید به تجمیع‌های قبلی اضافه می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: انتخاب درست استراتژی متریالیزه کردن مانع از اتلاف حافظه دیسک شده و پاسخ به پرس‌وجوهای OLAP را آنی می‌سازد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: ایجاد موازنه هوشمند بین سرعت پاسخ‌گویی به کوئری و حافظه ذخیره‌سازی
    </div>
  </p>

</div>

```python
display_academic_text(
    "موازنه فضا در برابر زمان: راهبردهای متریالیزه کردن مکعب داده (Data Cube Materialization) 🧊", 
    "main_title"
)

display_academic_text(
    "در سیستم‌های انبار داده بزرگ مانند وال‌مارت (Walmart)، اگر یک مکعب داده دارای $N$ بُعد (Dimension) مختلف باشد، "
    "تعداد کُلِ ترکیب‌های تجمیعیِ ممکن (زیرمکعب‌ها یا Cuboids) به صورت <b>نمایی (Exponential)</b> رشد می‌کند.", 
    "paragraph"
)

display_academic_text(
    r"برای مثال اگر سیستم دارای ۱۰ بُعدِ تحلیلی باشد، $2^{10} = 1024$ مکعب‌واره مختلف تولید می‌شود. محاسبهٔ درلحظه (On-the-fly) "
    r"این تجمیع‌ها در زمانِ اجرای کوئریِ مدیران، بسیار کُند و فلج‌کننده است. برای حل این چالش از راهکارهای <b>متریالیزه کردن (Materialization)</b> "
    r"یعنی <i>پیش‌محاسبه (Pre-computing) و ذخیره‌سازیِ فیزیکیِ تجمیع‌ها روی هارد دیسک</i> استفاده می‌شود.", 
    "paragraph"
)

display_academic_text("استراتژی‌های متریالیزه کردن (Materialization Trade-offs):", "sub_title")

display_academic_text(
    "۱. <b>متریالیزه کردن کامل (Full Materialization):</b> تمام $2^N$ حالت پیش‌محاسبه می‌شود (سرعت بی‌نظیر است، اما منجر به انفجارِ حجم دیسک می‌شود).<br>"
    "۲. <b>متریالیزه کردن جزئی (Partial Materialization):</b> فقط مکعب‌واره‌هایی که <i>بیشترین تقاضا</i> را دارند پیش‌محاسبه می‌شوند (تعادلِ طلایی).<br>"
    "۳. <b>بدون متریالیزه (No Materialization):</b> هیچ تجمیعی ذخیره نمی‌شود و هربار روی جداولِ خام (Base Tables) محاسبه صورت می‌گیرد (کندترین حالت).", 
    "paragraph"
)

# ==========================================================
# SECTION 2: GENERATING BASE SALES DATA AND CUBOIDS
# ==========================================================
display_academic_text("مرحله نخست: تولید جداول پایه و استراتژی‌های پیش‌محاسبه (Caching)", "sub_title")

# Step 1: Generate Walmart Base Sales Dataset (500,000 Rows for noticeable benchmarking)
# 3 Dimensions: Store, Category, Year -> Total 2^3 = 8 Cuboids
n_rows = 500000

stores = [f'Store_{i}' for i in range(1, 11)]
categories = [f'Category_{i}' for i in range(1, 21)]
years = [2024, 2025, 2026]

df_base = pd.DataFrame({
    'Store': np.random.choice(stores, size=n_rows),
    'Category': np.random.choice(categories, size=n_rows),
    'Year': np.random.choice(years, size=n_rows),
    'Sales_USD': np.random.uniform(20.0, 500.0, size=n_rows)
})

display_side_by_side_datasets(
    [df_base.head(5).reset_index(drop=True)],
    [f"جدول پایه تراکنش‌های فروش (Base Fact Table - دارای {n_rows:,} رکورد خام)"]
)

# Step 2: Implement 3 Materialization Strategies

# Strategy 1: No Materialization (On-the-Fly Compute)
# Stores 0 precomputed tables in memory/disk

# Strategy 2: Partial Materialization
# Precompute ONLY the top 2 most popular business queries: Store-Year & Category-Year
cache_partial = {
    'Store_Year': df_base.groupby(['Store', 'Year'])['Sales_USD'].sum().reset_index(),
    'Category_Year': df_base.groupby(['Category', 'Year'])['Sales_USD'].sum().reset_index()
}

# Strategy 3: Full Materialization (Precompute ALL 2^3 = 8 possible cuboid combinations)
cache_full = {
    'Base_Store_Cat_Year': df_base.groupby(['Store', 'Category', 'Year'])['Sales_USD'].sum().reset_index(),
    'Store_Cat': df_base.groupby(['Store', 'Category'])['Sales_USD'].sum().reset_index(),
    'Store_Year': df_base.groupby(['Store', 'Year'])['Sales_USD'].sum().reset_index(),
    'Cat_Year': df_base.groupby(['Category', 'Year'])['Sales_USD'].sum().reset_index(),
    'Store': df_base.groupby('Store')['Sales_USD'].sum().reset_index(),
    'Category': df_base.groupby('Category')['Sales_USD'].sum().reset_index(),
    'Year': df_base.groupby('Year')['Sales_USD'].sum().reset_index(),
    'All_Total': df_base['Sales_USD'].sum() # Apex cuboid
}

# ==========================================================
# SECTION 3: BENCHMARK QUERY PERFORMANCE
# ==========================================================
# Query execution simulator: 50 Random Requests for "Sales by Store & Year"
n_queries = 50

# Benchmark 1: No Materialization Time (Must scan 500k rows & aggregate 50 times)
start = time.time()
for _ in range(n_queries):
    res_1 = df_base.groupby(['Store', 'Year'])['Sales_USD'].sum()
time_no_mat = (time.time() - start) * 1000

# Benchmark 2: Partial Materialization Time (Retrieve instantly from small, targeted cache)
start = time.time()
for _ in range(n_queries):
    res_2 = cache_partial['Store_Year']
time_partial_mat = (time.time() - start) * 1000

# Benchmark 3: Full Materialization Time (Retrieve instantly from heavy, complete cache)
start = time.time()
for _ in range(n_queries):
    res_3 = cache_full['Store_Year']
time_full_mat = (time.time() - start) * 1000

display_academic_text(
    f"<b>نتایج بنچمارک سرعت برای اجرای پشتِ سرِ هَمِ {n_queries} کوئری:</b><br>"
    f"🔴 بدون پیش‌محاسبه (پردازشِ زنده روی جدول پایه): <b>{time_no_mat:.1f} میلی‌ثانیه</b><br>"
    f"🟢 متریالیزه جزئی (Partial - خواندن از کَشِ سبک): <b>{time_partial_mat:.1f} میلی‌ثانیه</b><br>"
    f"🟢 متریالیزه کامل (Full - خواندن از کَشِ سنگین): <b>{time_full_mat:.1f} میلی‌ثانیه</b>", 
    "insight"
)

# ==========================================================
# SECTION 4: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
# Step 4: Storage Overhead Metrics
cuboids_stored = [0, len(cache_partial), len(cache_full)]
execution_times = [time_no_mat, time_partial_mat, time_full_mat]
strategies = ['No Materialization\n(On-the-Fly)', 'Partial Materialization\n(The Golden Balance)', 'Full Materialization\n(Precomputed All)']

# Step 5: Visualization using Matplotlib
fig, axes = plt.subplots(1, 2, figsize=(16, 6.5), dpi=135)

# ----------------------------------------------------------
# Subplot 1: Query Execution Time Comparison
# ----------------------------------------------------------
bars1 = axes[0].bar(
    strategies, 
    execution_times, 
    color=['#e74c3c', '#2980b9', '#1abc9c'], # Red, Classic Blue, Turquoise
    edgecolor='#2c3e50', 
    linewidth=2.0,
    width=0.5
)

axes[0].annotate(
    "CPU Bottleneck:\nRe-calculating the sums over 500,000 rows\nevery single time a dashboard refreshes\nis extremely slow and inefficient.",
    xy=(0.0, time_no_mat/2), xytext=(0.4, time_no_mat*0.7),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title(f'Figure 1: Total Execution Time for {n_queries} Queries (Lower = Better)', fontsize=14, fontweight='bold', pad=15)
axes[0].set_ylabel('Query Execution Time (Milliseconds)', fontsize=13, fontweight='bold')
axes[0].grid(axis='y', linestyle='--', alpha=0.5)
axes[0].set_xticklabels(strategies, fontweight='bold', fontsize=11)

for bar in bars1:
    yval = bar.get_height()
    axes[0].text(
        bar.get_x() + bar.get_width() / 2, 
        yval + (max(execution_times) * 0.02), 
        f'{yval:.1f} ms', 
        ha='center', fontweight='bold', fontsize=13, color='#2c3e50'
    )

# ----------------------------------------------------------
# Subplot 2: Storage Overhead Comparison
# ----------------------------------------------------------
bars2 = axes[1].bar(
    strategies, 
    cuboids_stored, 
    color=['#e74c3c', '#2980b9', '#1abc9c'], 
    edgecolor='#2c3e50', 
    linewidth=2.0,
    width=0.5
)

axes[1].annotate(
    "Storage Explosion:\nFull Materialization gives fast speeds,\nbut storing all 8 possible combinations\nwastes massive amounts of Disk Space.",
    xy=(2.0, 6.5), xytext=(0.3, 7.0),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[1].set_title('Figure 2: Storage Space Overhead (Precomputed Cuboids)', fontsize=14, fontweight='bold', pad=15)
axes[1].set_ylabel('Number of Extra Tables Stored on Disk', fontsize=13, fontweight='bold')
axes[1].grid(axis='y', linestyle='--', alpha=0.5)
axes[1].set_ylim(0, 9)
axes[1].set_xticklabels(strategies, fontweight='bold', fontsize=11)

for bar in bars2:
    yval = bar.get_height()
    axes[1].text(
        bar.get_x() + bar.get_width() / 2, 
        yval + 0.2, 
        f'{yval} Extra Tables', 
        ha='center', fontweight='bold', fontsize=13, color='#2c3e50'
    )

# Clear top and right spines for a clean academic aesthetic
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 5: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (Strategic Implication):</b><br>"
    r"طراحی پایگاه داده همواره یک بازیِ <b>«موازنه فضا در برابر زمان» (Space-Time Trade-off)</b> است.<br>"
    r"متریالیزه کردن کامل (Full) اگرچه سرعت جستجو را آنی می‌کند، اما ذخیره‌سازیِ تمامِ $2^N$ حالت ممکن روی دیسک، در دنیای واقعی یک فاجعهٔ معماری محسوب می‌شود (انفجار حافظه). "
    r"در مقابل، <b>متریالیزه جزئی (Partial Materialization)</b> یک موازنهٔ بی‌نظیر ایجاد می‌کند؛ این روش با پیش‌محاسبه و ذخیرهٔ صرفاً چند کوئریِ پُرتقاضا (مثلاً گزارش فروش ماهانه که مدیران هر روز می‌بینند)، "
    r"هم مانع از اتلاف حافظه دیسک شده و هم سرعت پاسخگوییِ سیستم را به سطح ایده‌آلی ارتقا می‌دهد (Materialized Views در دیتابیس‌های مدرن)."
)
display_academic_text(insight_conclusion, "insight")
```
