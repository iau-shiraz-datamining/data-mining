<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: تحلیل فروش چندبعدی در «وال‌مارت» (Walmart) با مکعب داده 🧊
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    تیم تحلیل هوش تجاری فروشگاه زنجیره‌ای «وال‌مارت» (Walmart) نمی‌تواند فروش را فقط در یک جدول دو‌بعدی (سطر و ستون ساده) تحلیل کند. مدیران ارشد نیاز دارند میزان فروش را به‌طور هم‌زمان بر اساس <b>۳ بعد (Dimension) مختلف</b> تحلیل کنند:
  </p>

  <ul style="text-align: right; margin: 0 0 10px; padding-right: 22px;">
    <li><b>بعد اول (Product / محصول):</b> الکترونیک، پوشاک، لوازم خانگی.</li>
    <li><b>بعد دوم (Location / شهر):</b> نیویورک، شیکاگو، لس‌آنجلس.</li>
    <li><b>بعد سوم (Time / زمان):</b> سه ماهه اول (Q1)، سه ماهه دوم (Q2)، سه ماهه سوم (Q3).</li>
  </ul>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>مفهوم ریاضی مکعب داده (Data Cube):</b><br>
    مکعب داده این ۳ بعد را در یک فضای سه‌بعدی ترکیب می‌کند. هر خانه یا سلول ($Cell$) در این مکعب، تقاطع یک محصول مشخص، در یک شهر مشخص و در یک فصل مشخص است که مقدار معیار عددی (Measure) مانند «مبلغ کل فروش» را در خود ذخیره می‌کند:
  </p>

  <div style="text-align: center; margin: 15px 0;">
    $$\text{Cell}(d_1, d_2, d_3) = \text{Measure (Total Sales)}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>مثال عددی خانه مکعب:</b><br>
    $$\text{Cell}(\text{'Electronics'}, \text{'New York'}, \text{'Q1'}) = 150,000 \text{ USD}$$
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> مدل مکعب داده به مدیران اجازه می‌دهد تا با یک نگاه، میزان فروش در هر محل، هر زمان و هر گروه کالایی را استخراج کرده و پایه و اساس تحلیل‌های OLAP را شکل دهند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: سازمان‌دهی داده‌های تجاری در ساختار چندبعدی برای تحلیل‌های سریع و جامع
    </div>
  </p>

```python
display_academic_text(
    "تحلیل فروش چندبُعدی در «وال‌مارت» (Walmart) با مکعب داده (OLAP Data Cube) 🧊", 
    "main_title"
)

display_academic_text(
    "تیم تحلیل هوش تجاری فروشگاه زنجیره‌ای «وال‌مارت» نمی‌تواند فروش را فقط در یک جدول دو‌بُعدیِ سنتی (سطر و ستون سادهٔ رابطه‌ای) تحلیل کند. "
    "مدیران ارشد نیاز دارند میزان فروش را به‌طور هم‌زمان بر اساس <b>۳ بُعد (Dimension)</b> مختلف تحلیل و برش (Slice) زنند:<br>"
    "<b>بُعد اول (Product):</b> الکترونیک، پوشاک، لوازم خانگی.<br>"
    "<b>بُعد دوم (Location):</b> نیویورک، شیکاگو، لس‌آنجلس.<br>"
    "<b>بُعد سوم (Time):</b> سه ماهه اول (Q1)، سه ماهه دوم (Q2)، سه ماهه سوم (Q3).", 
    "paragraph"
)

display_academic_text(
    "مفهوم ریاضی مکعب داده (Data Cube):", 
    "sub_title"
)

display_academic_text(
    "مکعب داده، این ۳ بُعد را در یک فضای سه‌بُعدی ترکیب می‌کند. هر خانه یا سلول (Cell) در این مکعب، دقیقاً <b>تقاطع یک محصول، یک شهر و یک فصل مشخص</b> است "
    "که مقدارِ معیارِ عددی (Measure) مانند «مبلغ کل فروش» را در خود ذخیره می‌کند.", 
    "paragraph"
)

display_academic_text(
    r"$$Cell(d_1, d_2, d_3) = \text{Measure (Total Sales)}$$", 
    "paragraph"
)

display_academic_text(
    "مثال عددیِ یک خانه از مکعب:", 
    "sub_title"
)

display_academic_text(
    r"$$Cell(\text{'Electronics'}, \text{'New York'}, \text{'Q1'}) = 150,000 \text{ USD}$$", 
    "paragraph"
)


# ==========================================================
# SECTION 2: DATA CUBE GENERATION AND OLAP PIVOT TRANSFORMATION
# ==========================================================
display_academic_text(
    "مرحله نخست: ساخت مکعب داده و اعمال عملیات بُرش (Slice & Dice)", 
    "sub_title"
)

# Step 1: Create synthetic Walmart 3D Data Cube dataset
# Dimensions: Product (3), City (3), Quarter (3) -> Total 27 Cells
products = ['Electronics', 'Apparel', 'Home']
cities = ['New York', 'Chicago', 'Los Angeles']
quarters = ['Q1', 'Q2', 'Q3']

# Generate combinations
cube_records = []

for p_idx, prod in enumerate(products):
    for c_idx, city in enumerate(cities):
        for q_idx, qtr in enumerate(quarters):
            
            # Sales measure in kUSD (random value)
            sales = np.random.randint(50, 300)
            
            cube_records.append({
                'Product': prod,
                'City': city,
                'Quarter': qtr,
                'Prod_Idx': p_idx,
                'City_Idx': c_idx,
                'Qtr_Idx': q_idx,
                'Sales_kUSD': sales,
            })

df_cube = pd.DataFrame(cube_records)

# Step 2: Create Pivot Table (2D Representation of a Cube Slicing Operation)
pivot_view = df_cube.pivot_table(
    index='Product', 
    columns=['City', 'Quarter'], 
    values='Sales_kUSD'
).fillna(0)

df_cube_disp = df_cube[['Product', 'City', 'Quarter', 'Sales_kUSD']].head(9).reset_index(drop=True)

# Render the initial Flat records vs the 2D Pivot slice
display_side_by_side_datasets(
    [df_cube_disp, pivot_view],
    [
        "جدول ۱: رکوردِ سلول‌های مکعب (نمای خطی و رابطه‌ای)", 
        "جدول ۲: نمای دوبُعدی از بُرشِ مکعب (OLAP Pivot/Slice Operation)"
    ]
)

# ==========================================================
# SECTION 3: 3D VISUALIZATION (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی - تجسم فضای سه‌بُعدی و پراکندگیِ سلول‌ها", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکی زیر ماهیت واقعی یک مکعب داده (Data Cube) را ترسیم می‌کند. <b>محورهای ۳گانه</b> نشان‌دهندهٔ ابعاد (شهر، محصول، فصل) بوده "
    "و <b>اندازه و تیرگیِ رنگِ هر حباب</b>، مقدارِ متریک فروش (Measure) در آن تقاطعِ خاص را نمایش می‌دهد. این معماری سه‌بُعدی، "
    "بستر اصلی تحلیل‌های پیچیده و داشبوردهای تعاملیِ مدیران سازمان است.", 
    "paragraph"
)

# Step 3: Visualization using Matplotlib 3D Scatter Plot
fig = plt.figure(figsize=(13, 9), dpi=135)
ax = fig.add_subplot(111, projection='3d')

# Scatter points in 3D Space
# X: Product, Y: City, Z: Quarter, Color & Size: Sales Measure
scatter = ax.scatter(
    df_cube['Prod_Idx'],
    df_cube['City_Idx'],
    df_cube['Qtr_Idx'],
    c=df_cube['Sales_kUSD'],
    s=df_cube['Sales_kUSD'] * 3.5,  # Bubble size significantly amplified for visual clarity
    cmap='Blues',                   # Academic Blue Palette for uniformity
    alpha=0.9,
    edgecolors='#2c3e50',           # Navy border for sharp contrast
    linewidth=1.0
)

# Set axis tick labels correctly matching categorical indices
ax.set_xticks(range(len(products)))
ax.set_xticklabels(products, fontweight='bold', fontsize=11)

ax.set_yticks(range(len(cities)))
ax.set_yticklabels(cities, fontweight='bold', fontsize=11)

ax.set_zticks(range(len(quarters)))
ax.set_zticklabels(quarters, fontweight='bold', fontsize=11)

# Formatting Axis labels and title
ax.set_xlabel('Dimension 1: Product Category', labelpad=15, fontweight='bold', fontsize=12)
ax.set_ylabel('Dimension 2: Geographic City', labelpad=15, fontweight='bold', fontsize=12)
ax.set_zlabel('Dimension 3: Time (Quarter)', labelpad=15, fontweight='bold', fontsize=12)

ax.set_title(
    'Figure 1: Walmart 3D Data Cube Architecture\n(Bubble Size & Darkness = Total Sales Measure in kUSD)',
    fontsize=14,
    fontweight='bold',
    pad=20
)

# Add colorbar to interpret the measure mapping
cbar = fig.colorbar(scatter, ax=ax, shrink=0.5, pad=0.1)
cbar.set_label('Sales Measure Magnitude ($k USD)', fontweight='bold', fontsize=11)

# Adjust viewing angle for best 3D perspective
ax.view_init(elev=20, azim=45)

# Adjust plot boundaries cleanly
fig.subplots_adjust(left=0.0, right=1.0, bottom=0.0, top=1.0)

plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک سیستم‌های چندبُعدی (Strategic Implication):</b><br>"
    r"مدل <b>مکعب داده (Data Cube)</b> به مدیران وال‌مارت اجازه می‌دهد تا محدودیت‌های صفحات گستردهٔ دوبُعدی (همانند Excel) را درهم‌شکسته و با یک نگاه، میزان فروش را "
    r"در تقاطع‌های ۳بُعدیِ جغرافیایی، زمانی و کالایی ارزیابی کنند (گراف سه‌بُعدی بالا).<br>"
    r"اگر یک مدیر بخواهد <i>فقط فروش وسایل الکترونیک را در فصل اول (Q1)</i> بررسی کند، به سادگی یک لایه از این مکعب را برش می‌زند (عملیات <b>Slice</b>). "
    r"اگر بخواهد <i>فروش لوازم خانگی و پوشاک را در شیکاگو و نیویورک</i> تحلیل کند، یک مکعب کوچک‌تر استخراج می‌کند (عملیات <b>Dice</b>). "
    r"این معماری داده، هستهٔ مرکزی <b>پردازش تحلیلی آنلاین (OLAP)</b> بوده و سرعت گزارش‌گیری در داشبوردهای BI را به شکل چشمگیری افزایش می‌دهد."
)
display_academic_text(insight_conclusion, "insight")
```
