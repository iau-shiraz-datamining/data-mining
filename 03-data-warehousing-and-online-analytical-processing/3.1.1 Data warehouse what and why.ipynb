<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: چرا سیستم‌های فروش روزانه پاسخ‌گوی تحلیل نیستند و به انبار داده نیاز داریم؟ 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    فرض کنید مدیر یک فروشگاه زنجیره‌ای بزرگ هستید. سیستم عملیاتی ثبت فروش (OLTP)، هر ثانیه یک تراکنش ثبت می‌کند؛ مثلاً «خرید یک عدد عطر در ساعت ۱۰:۳۰». این سیستم برای ثبت سریع خرید مشتریان عالی است، اما اگر از آن بپرسید «در سه سال گذشته، کدام گروه سنی بیشترین خرید را از برند مشخصی داشته‌اند؟»، سیستم به‌شدت کند شده و حتی فرآیند ثبت فروش در صندوق‌ها متوقف می‌شود؛ زیرا سیستم‌های عملیاتی برای کوئری‌های سنگین و تحلیل‌های تاریخی طراحی نشده‌اند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مشکل دوم: پراکندگی و جزیره‌ای بودن داده‌ها (Data Silos)<br>
    حتی اگر سیستم فروش سریع باشد، اطلاعات کامل ندارد؛ زیرا گروه سنی مشتری در سیستم بازاریابی ذخیره شده، اطلاعات برند در سیستم انبارداری قرار دارد و اطلاعات پرداخت در سیستم مالی است. داده‌ها به‌صورت پراکنده و ناهمگن در سیستم‌های مجزا قفل شده‌اند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    راهکار انبار داده (Data Warehouse):<br>
    انبار داده یک مخزن متمرکز و جداگانه است که داده‌ها را از تمام سیستم‌های عملیاتی جمع‌آوری، یکدست و هماهنگ می‌کند. ویژگی‌های کلیدی انبار داده عبارت‌اند از:
  </p>

  <ul style="text-align: right; margin: 0 0 10px; padding-right: 22px;">
    <li>موضوع‌مدار (Subject-Oriented): داده‌ها دور موضوعات اصلی تجارت مانند مشتری و محصول سازمان‌دهی می‌شوند.</li>
    <li>یکپارچه (Integrated): تمام تعارضات فرمت داده‌ها از منبع‌های مختلف یکسان‌سازی می‌شوند.</li>
    <li>متغیر با زمان (Time-Variant): تاریخچه کامل داده‌ها برای چندین سال صیانت و نگهداری می‌شود.</li>
    <li>غیرقابل تغییر (Non-Volatile): داده‌ها هرگز ویرایش یا حذف نمی‌شوند، بلکه فقط ثبت‌های جدید به آن‌ها اضافه می‌گردد.</li>
  </ul>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه کاربردی: یک فروشگاه زنجیره‌ای با بررسی فروش چند سال گذشته در انبار داده، الگوی فصلی خرید را شناسایی می‌کند و پیش از شروع فصل، موجودی مناطق پرمصرف را افزایش می‌دهد تا از فروش ازدست‌رفته جلوگیری کند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: جداسازی پردازش تحلیلی از تراکنشی و ایجاد دیدگاه یکپارچه و تاریخی از کل سازمان
    </div>
  </p>

</div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display, HTML

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})

plt.rcParams['figure.dpi'] = 120

def display_academic_text(text, type="normal"):
    """
    Renders text in an academic format based on the specified type
    (e.g., title, paragraph, insight).
    Uses strict typography and styling to avoid informal elements.
    """
    styles = {
        "main_title": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 24px; "
            "font-weight: bold; text-align: center; border-bottom: 2px solid #34495e; "
            "padding-bottom: 15px; margin: 30px 0;"
        ),
        "sub_title": (
            "font-family: Tahoma, sans-serif; color: #34495e; font-size: 19px; "
            "font-weight: bold; border-right: 5px solid #2980b9; padding-right: 15px; "
            "margin: 40px 0px 20px 0px; background-color: #ecf0f1; padding: 12px; "
            "border-radius: 4px;"
        ),
        "paragraph": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 16px; "
            "margin-bottom: 20px; line-height: 1.8; text-align: justify;"
        ),
        "insight": (
            "font-family: Tahoma, sans-serif; color: #16a085; font-size: 16px; "
            "font-weight: bold; background-color: #e8f8f5; padding: 20px; "
            "border-left: 5px solid #1abc9c; margin: 30px 0; line-height: 1.8; "
            "text-align: justify;"
        )
    }
    display(HTML(f"<div dir='rtl' style='text-align: right; {styles[type]}'>{text}</div>"))

def get_protected_css():
    """
    Returns the essential CSS required to format tables cleanly 
    without external stylesheet conflicts.
    """
    return """
    <style>
    .academic-table-wrapper { 
        direction: ltr !important; 
        overflow-x: auto !important; 
        width: 100% !important; 
        background-color: #fff; 
        border: 1px solid #bdc3c7; 
        border-radius: 4px; 
        box-shadow: 0 3px 6px rgba(0,0,0,0.08); 
    }
    table.strict-tbl { 
        border-collapse: collapse !important; 
        width: 100% !important; 
        margin: 0 !important; 
        font-family: inherit !important;
        font-size: 14px !important; 
        table-layout: auto !important; 
        border: none !important; 
    }
    table.strict-tbl thead th { 
        background-color: #2c3e50 !important; 
        color: #ffffff !important; 
        padding: 12px 18px !important; 
        text-align: center !important; 
        border: 1px solid #34495e !important; 
        white-space: nowrap !important; 
        font-weight: bold !important; 
    }
    table.strict-tbl tbody td { 
        padding: 10px 18px !important; 
        text-align: center !important; 
        border: 1px solid #bdc3c7 !important; 
        white-space: nowrap !important; 
        color: #2c3e50 !important; 
    }
    table.strict-tbl tbody tr:nth-child(even) td { 
        background-color: #f8f9fa !important; 
    }
    table.strict-tbl tbody tr:hover td { 
        background-color: #e8ecef !important; 
    }
    </style>
    """

def display_side_by_side_datasets(dfs, titles, *args):
    """
    Renders one or multiple pandas DataFrames.
    
    - If only one dataframe: displays it full-width.
    - If multiple dataframes: displays them side-by-side.
    
    Parameters:
    -----------
    dfs : list or DataFrame
        List of dataframes, or a single dataframe
    titles : list or str
        List of titles, or a single title string
    *args : (optional) for backward compatibility with old four-argument calls
    """
    if not isinstance(dfs, list) and not isinstance(titles, list):
        df1, title1, df2, title2 = dfs, titles, args[0], args[1]
        dfs = [df1, df2]
        titles = [title1, title2]
    else:
        if not isinstance(dfs, list) or not isinstance(titles, list):
            raise ValueError("دو آرگومان اول باید لیست باشند (یا چهار آرگومان جداگانه).")
        if len(dfs) != len(titles):
            raise ValueError("تعداد دیتافریم‌ها و عنوان‌ها باید برابر باشد.")
        if len(dfs) == 0:
            raise ValueError("حداقل یک دیتافریم برای نمایش نیاز است.")
    
    if len(dfs) == 1:
        df = dfs[0]
        title = titles[0]
        table_html = df.to_html(index=False, classes=['strict-tbl'], border=0)
        html = f"""
        <div style="width: 100%; margin: 20px 0 35px 0;">
            <div dir="rtl" style="background-color: #2980b9; color: white; padding: 12px; 
                                 text-align: center; font-family: Tahoma, sans-serif; 
                                 font-weight: bold; font-size: 15px; border-radius: 4px 4px 0 0;">
                {title}
            </div>
            <div class="academic-table-wrapper" style="margin-top: 0; border-top: none;">
                {table_html}
            </div>
        </div>
        """
        display(HTML(get_protected_css() + html))
        return
    
    num_cols = len(dfs)
    col_html_list = []
    for i, (df, title) in enumerate(zip(dfs, titles)):
        table_html = df.to_html(index=False, classes=['strict-tbl'], border=0)
        bg_color = "#2980b9" if i % 2 == 0 else "#27ae60"
        col_html = f"""
        <div class="flex-column" style="flex: 1 1 0; min-width: 0; display: flex; flex-direction: column;">
            <div style="background-color: {bg_color}; color: white; padding: 12px; 
                        text-align: center; font-family: Tahoma, sans-serif; 
                        font-weight: bold; font-size: 15px; border-radius: 4px 4px 0 0;">
                {title}
            </div>
            <div class="academic-table-wrapper" style="margin-top: 0; border-top: none;">
                {table_html}
            </div>
        </div>
        """
        col_html_list.append(col_html)
    
    columns_joined = "".join(col_html_list)
    html = f"""
    <div style="display: flex; flex-direction: row; justify-content: space-between; 
                gap: 20px; width: 100%; max-width: 100%; margin: 20px 0 35px 0;
                box-sizing: border-box;">
        {columns_joined}
    </div>
    """
    display(HTML(get_protected_css() + html))


display(HTML("""
<style>
@import url('https://cdn.jsdelivr.net/gh/rastikerdar/vazirmatn@v33.003/Vazirmatn-font-face.css');

* {
    font-family: 'Vazirmatn', 'Vazir', Tahoma, sans-serif !important;
}
div, p, h1, h2, h3, h4, td, th, span, li, .strict-tbl, .academic-table-wrapper {
    font-family: 'Vazirmatn', 'Vazir', Tahoma, sans-serif !important;
}
</style>
"""))



np.random.seed(42)
```

```python
display_academic_text(
    "چرا سیستم‌های فروشِ روزانه پاسخ‌گوی تحلیل نیستند و به انبار داده (Data Warehouse) نیاز داریم؟ 🗄️", 
    "main_title"
)

display_academic_text(
    "فرض کنید مدیر یک فروشگاه زنجیره‌ای بزرگ هستید. سیستم عملیاتی ثبت فروش <b>(OLTP: Online Transaction Processing)</b>، هر ثانیه صدها تراکنش ثبت می‌کند؛ "
    "مثلاً <i>«خرید یک عدد عطر در ساعت ۱۰:۳۰»</i>. این سیستم برای <b>ثبت سریعِ خرید مشتریان</b> عالی است، اما اگر از آن بپرسید "
    "«در سه سال گذشته، کدام گروه سنی بیشترین خرید را از یک برند مشخص داشته‌اند؟»، سیستم به‌شدت کُند شده و حتی فرآیند ثبت فروش در صندوق‌ها متوقف می‌شود! "
    "زیرا سیستم‌های عملیاتی برای کوئری‌های سنگین (JOIN های پیچیده) و تحلیل‌های تاریخی طراحی نشده‌اند.", 
    "paragraph"
)

display_academic_text("مشکل دوم: پراکندگی و جزیره‌ای بودن داده‌ها (Data Silos)", "sub_title")
display_academic_text(
    "حتی اگر سیستم فروش سریع باشد، اطلاعاتِ کاملی ندارد؛ زیرا گروه سنی مشتری در <i>سیستم بازاریابی</i> ذخیره شده، اطلاعات برند در <i>سیستم انبارداری</i> قرار دارد "
    "و اطلاعات پرداخت در <i>سیستم مالی</i> است. داده‌ها به‌صورت پراکنده و ناهمگن در سیستم‌های مجزا قفل شده‌اند.", 
    "paragraph"
)

display_academic_text("راهکار قطعی: انبار داده (Data Warehouse / OLAP)", "sub_title")
display_academic_text(
    "انبار داده یک مخزن <b>متمرکز و جداگانه</b> است که داده‌ها را از تمام سیستم‌های عملیاتی استخراج، یکدست و هماهنگ می‌کند (فرآیند ETL). ویژگی‌های کلیدی انبار داده عبار‌ت‌اند از:<br>"
    "۱. <b>موضوع‌مدار (Subject-Oriented):</b> سازمان‌دهی بر اساس موجودیت‌های اصلی (مثل فروش، مشتری).<br>"
    "۲. <b>یکپارچه (Integrated):</b> رفع تعارضات و تجمیع داده‌ها از منابع گوناگون در یک جدول مسطح (Denormalized).<br>"
    "۳. <b>متغیر با زمان (Time-Variant):</b> نگهداری تاریخچهٔ کامل داده‌ها برای چندین سال.<br>"
    "۴. <b>غیرقابل تغییر (Non-Volatile):</b> عدم ویرایش؛ فقط رکوردهای جدید اضافه می‌شوند (سرعت بی‌نظیر در خواندن).", 
    "paragraph"
)

# ==========================================================
# SECTION 2: SIMULATING OLTP SILOS & ETL PROCESS
# ==========================================================
# Step 1: Create Simulated Operational Data Silos (OLTP Normalized Tables)
n_transactions = 50000

# SQLite in-memory databases for OLTP and DW
conn_oltp = sqlite3.connect(':memory:')
conn_dw = sqlite3.connect(':memory:')

# Generate OLTP Normalized Tables (Separated Data Silos)
df_customers_oltp = pd.DataFrame({
    'Customer_ID': range(1, 5001),
    'Age_Group': np.random.choice(['18-25', '26-35', '36-50', '50+'], size=5000),
})

df_products_oltp = pd.DataFrame({
    'Product_ID': range(101, 601),
    'Brand': np.random.choice(['Brand_A', 'Brand_B', 'Brand_C', 'Brand_D'], size=500),
})

df_sales_oltp = pd.DataFrame({
    'Transaction_ID': range(1, n_transactions + 1),
    'Customer_ID': np.random.randint(1, 5001, size=n_transactions),
    'Product_ID': np.random.randint(101, 601, size=n_transactions),
    'Amount_USD': np.random.uniform(10.0, 300.0, size=n_transactions),
    'Tx_Date': np.random.choice(['2024-05-10', '2025-08-15', '2026-11-20'], size=n_transactions),
})

# Save to OLTP Database (Simulating isolated silos)
df_customers_oltp.to_sql('Customers_Marketing', conn_oltp, index=False, if_exists='replace')
df_products_oltp.to_sql('Products_Inventory', conn_oltp, index=False, if_exists='replace')
df_sales_oltp.to_sql('Sales_Operations', conn_oltp, index=False, if_exists='replace')

# Step 2: Build Integrated Data Warehouse Table (Denormalized DW Fact Table)
# ETL Process: Extract from 3 OLTP silos, Transform (JOIN), and Load into 1 Unified Table
query_etl = """
    SELECT 
        s.Transaction_ID, c.Age_Group, p.Brand, s.Amount_USD, s.Tx_Date 
    FROM Sales_Operations s 
    JOIN Customers_Marketing c ON s.Customer_ID = c.Customer_ID 
    JOIN Products_Inventory p ON s.Product_ID = p.Product_ID
"""

df_dw_integrated = pd.read_sql_query(query_etl, conn_oltp)
df_dw_integrated.to_sql('Fact_Sales_DW', conn_dw, index=False, if_exists='replace')

# Render a preview of the scattered OLTP data vs Integrated DW
display_side_by_side_datasets(
    [df_sales_oltp.head(5).reset_index(drop=True), df_dw_integrated.head(5).reset_index(drop=True)],
    [
        "پایگاه عملیاتی (OLTP): پراکنده - فقط شامل کُدها است", 
        "انبار داده (DW): یکپارچه - جدول مسطح (Denormalized) با اطلاعات کامل"
    ]
)

# ==========================================================
# SECTION 3: BENCHMARK ANALYTICAL QUERY PERFORMANCE
# ==========================================================
display_academic_text(
    "مرحله دوم: بنچمارک و مقایسهٔ سرعت کوئری‌های تحلیلی", 
    "sub_title"
)

display_academic_text(
    "اکنون سؤال مدیر فروشگاه را از هر دو سیستم می‌پرسیم: <i>«درآمد کل برای Brand_A به تفکیک گروه سنی چقدر است؟»</i><br>"
    "سیستم OLTP مجبور است ۳ جدولِ مجزا را هم‌زمان با اجرای کوئری <b>JOIN</b> کند (فاجعهٔ عملکردی برای سرور لایو). اما انبار داده مستقیماً و بدون JOIN از جدول مسطحِ خود، "
    "داده‌ها را با سرعت نور تجمیع می‌کند (Aggregations).", 
    "paragraph"
)

n_benchmark_runs = 50

# Benchmark 1: Querying OLTP (Requires 3-table JOIN across silos at runtime)
query_oltp = """
    SELECT c.Age_Group, SUM(s.Amount_USD) as Total_Revenue 
    FROM Sales_Operations s 
    JOIN Customers_Marketing c ON s.Customer_ID = c.Customer_ID 
    JOIN Products_Inventory p ON s.Product_ID = p.Product_ID 
    WHERE p.Brand = 'Brand_A' 
    GROUP BY c.Age_Group
"""

start_time = time.time()
for _ in range(n_benchmark_runs):
    pd.read_sql_query(query_oltp, conn_oltp)
time_oltp_query = (time.time() - start_time) * 1000

# Benchmark 2: Querying Integrated DW Fact Table (Direct Aggregation - NO JOINs needed)
query_dw = """
    SELECT Age_Group, SUM(Amount_USD) as Total_Revenue 
    FROM Fact_Sales_DW 
    WHERE Brand = 'Brand_A' 
    GROUP BY Age_Group
"""

start_time = time.time()
for _ in range(n_benchmark_runs):
    pd.read_sql_query(query_dw, conn_dw)
time_dw_query = (time.time() - start_time) * 1000

speedup = time_oltp_query / time_dw_query

# Display benchmark results dynamically
display_academic_text(
    f"<b>نتایج بنچمارک برای اجرای پشت‌سرهمِ {n_benchmark_runs} کوئری تحلیلی:</b><br>"
    f"⏱️ زمان پاسخ‌گویی OLTP (جداول پراکنده): <b>{time_oltp_query:.1f} میلی‌ثانیه</b><br>"
    f"⏱️ زمان پاسخ‌گویی Data Warehouse (جدول متمرکز): <b>{time_dw_query:.1f} میلی‌ثانیه</b><br>"
    f"🚀 <b>بهبود عملکرد (Speedup):</b> انبار داده <b>{speedup:.1f} برابر سریع‌تر</b> عمل کرد!", 
    "insight"
)

# ==========================================================
# SECTION 4: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 6.5), dpi=135)

# --- SUBPLOT 1: Query Time Comparison ---
systems = ['Live Operational Server\n(OLTP 3-Table JOIN)', 'Dedicated Data Warehouse\n(Unified Fact Table)']
times = [time_oltp_query, time_dw_query]

# Palette: Red for slow/dangerous OLTP, Turquoise for fast DW
bars1 = axes[0].bar(systems, times, color=['#e74c3c', '#1abc9c'], edgecolor='#2c3e50', linewidth=2.0, width=0.5)

axes[0].annotate(
    "Server Overload:\nRunning analytical joins on the Live OLTP\nserver consumes massive CPU/RAM and can\ncause checkout counters to crash!",
    xy=(0.0, time_oltp_query/2), xytext=(0.6, time_oltp_query/1.5),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title(f'Figure 1: Analytical Query Execution Time ({n_benchmark_runs} Runs)', fontsize=14, fontweight='bold', pad=15)
axes[0].set_ylabel('Execution Time (Milliseconds) - Lower is Better', fontsize=12, fontweight='bold')
axes[0].grid(axis='y', linestyle='--', alpha=0.5)
axes[0].set_xticklabels(systems, fontweight='bold', fontsize=11)

for bar in bars1:
    yval = bar.get_height()
    axes[0].text(bar.get_x() + bar.get_width() / 2, yval + (time_oltp_query * 0.02), f'{yval:.1f} ms', ha='center', fontweight='bold', fontsize=13, color='#2c3e50')

# --- SUBPLOT 2: Analytical Results (Revenue by Age Group for Brand A) ---
df_result = pd.read_sql_query(query_dw, conn_dw).sort_values(by='Age_Group')

# Academic Blue Palette for the business chart
bars2 = axes[1].bar(df_result['Age_Group'], df_result['Total_Revenue'] / 1000, color='#2980b9', edgecolor='#2c3e50', linewidth=2.0, width=0.6)

axes[1].annotate(
    "Business Intelligence:\nThe DW safely calculated that the 26-35\nage group generates the most revenue\nfor Brand_A across all historical records.",
    xy=(1.0, max(df_result['Total_Revenue'] / 1000)/2), xytext=(2.0, max(df_result['Total_Revenue'] / 1000)/1.5),
    fontsize=11, fontweight='bold', color='#2980b9',
    arrowprops=dict(arrowstyle="->", color="#2980b9", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#2980b9", lw=1.5)
)

axes[1].set_title('Figure 2: DW Historical Revenue Analysis for Brand_A', fontsize=14, fontweight='bold', pad=15)
axes[1].set_xlabel('Customer Demographic (Age Group)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Total Revenue Generated ($k USD)', fontsize=12, fontweight='bold')
axes[1].grid(axis='y', linestyle='--', alpha=0.5)
axes[1].set_xticklabels(df_result['Age_Group'], fontweight='bold', fontsize=11)
axes[1].set_ylim(0, max(df_result['Total_Revenue'] / 1000) * 1.2) # Give headroom for labels

for i, val in enumerate(df_result['Total_Revenue'] / 1000):
    axes[1].text(i, val + (max(df_result['Total_Revenue'] / 1000) * 0.03), f'${val:.1f}k', ha='center', fontweight='bold', fontsize=12, color='#2c3e50')

# Clear top and right spines for a clean academic aesthetic
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# Close database connections safely
conn_oltp.close()
conn_dw.close()

# ==========================================================
# SECTION 5: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه کاربردی مهندسی (Strategic Implication):</b><br>"
    r"خروجی این معماری، <b>جداسازیِ مطلقِ پردازش تحلیلی (OLAP) از پردازش تراکنشی (OLTP)</b> و ایجاد یک دیدگاه یکپارچه (Single Source of Truth) از کل سازمان است.<br>"
    r"اجرای کدهای تحلیلیِ پایتون یا SQL روی دیتابیسِ لایو فروشگاه، باعث کُند شدن سرور و از کار افتادن صندوق‌های پرداخت می‌گردد (Figure 1). "
    r"با استخراج داده‌ها (ETL) و انتقال آن‌ها به یک <b>انبار داده (Data Warehouse)</b>، تحلیلگران داده می‌توانند میلیون‌ها رکورد تاریخی را با سرعت بالا کاوش کنند "
    r"(Figure 2)، الگوهای فصلی خرید را شناسایی نموده و بدون هیچ خطری برای سیستم‌های فروش روزانه، تصمیمات استراتژیک (Business Intelligence) اتخاذ نمایند."
)
display_academic_text(insight_conclusion, "insight")
```
