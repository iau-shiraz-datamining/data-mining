<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: محاسبه کامل مکعب داده با جمع‌بندی آرایه چندطرفه در آمازون 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پردازش تحلیلی آنی (MOLAP)، داده‌ها به جای جداول رابطه‌ای، در قالب آرایه‌های چندبعدی فشرده به نام تنسور (Tensor) نگهداری می‌شوند. سیستم تحلیلی آمازون برای محاسبه کامل تمام مکعب‌واره‌های موجود در شبکه لتیس (Lattice of Cuboids)، از روش جمع‌بندی آرایه چندطرفه (Multiway Array Aggregation) استفاده می‌کند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم کلیدی این الگوریتم:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. نمایش داده به صورت آرایه یا تنسور (Tensor Representation):<br>
    داده‌های سه بعدی به صورت یک آرایه سه‌بعدی متراکم در حافظه نگهداری می‌شوند که سه بعد آن به ترتیب نماینده زمان (سال)، منطقه جغرافیایی و دسته‌بندی کالا است. هر خانه آرایه اندیس مشخصی دارد و مستقیم در حافظه آدرس‌دهی می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. تجمیع مستقیم در امتداد محورها (Axis Aggregation):<br>
    به جای اجرای دستورات سنگین SQL و فیلتر کردن جداول، الگوریتم تنها با جمع زدن مقادیر آرایه در امتداد یک یا چند محور (Axis)، تمام ۲ به توان N مکعب‌واره فرعی را به صورت هم‌زمان محاسبه می‌کند:
  </p>

  <ul style="text-align: right; margin: 0 0 10px; padding-right: 22px;">
    <li><b>مکعب‌واره پایه (۳ بعدی):</b> داده‌های کامل بدون تجمیع روی محورها.</li>
    <li><b>مکعب‌واره‌های ۲ بعدی:</b> تجمیع در امتداد ۱ محور (مثلاً جمع زدن روی محور سال برای رسیدن به جدول منطقه در دسته‌بندی).</li>
    <li><b>مکعب‌واره‌های ۱ بعدی:</b> تجمیع در امتداد ۲ محور (مثلاً جمع زدن روی سال و منطقه برای رسیدن به فروش کل هر دسته‌بندی).</li>
    <li><b>مکعب‌واره رأس (۰ بعدی):</b> تجمیع در امتداد تمام محورها برای رسیدن به عدد کل فروش آمازون.</li>
  </ul>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. محاسبه کارآمد و هم‌زمان (Single-Pass Computation):<br>
    آرایه‌ها بر اساس ترتیب بهینه‌ای در حافظه پیمایش می‌شوند تا تجمیع‌های فرعی با حداقل پیمایش حافظه و بدون ایجاد جداول واسط فورا محاسبه شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: این روش نیاز به جداول رابطه‌ای موقت را از بین برده و تمام زیرمکعب‌ها را با سرعت بالایی در حافظه محاسبه می‌کند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: محاسبه کامل شبکه مکعب‌واره‌ها با عملیات تجمیع محورهای آرایه چندبعدی
    </div>
  </p>

</div>

```python
display_academic_text(
    "محاسبه کامل مکعب داده در MOLAP با جمع‌بندی آرایه چندطرفه (Multiway Array Aggregation) 🧊", 
    "main_title"
)

display_academic_text(
    "در سیستم‌های پردازش تحلیلیِ آنی (MOLAP)، داده‌ها به جای جداول رابطه‌ای (SQL)، در قالبِ <b>آرایه‌های چندبُعدیِ فشرده به نام تنسور (Tensor)</b> در حافظهٔ رَم (RAM) نگهداری می‌شوند. "
    "سیستمِ تحلیلی برای محاسبهٔ کامل تمامِ تجمیع‌های ممکن در شبکه لتیس (Lattice of Cuboids)، از روشِ بسیار پُرسرعتِ <b>جمع‌بندی آرایه چندطرفه (Multiway Array Aggregation)</b> استفاده می‌کند.", 
    "paragraph"
)

display_academic_text("مفاهیم کلیدی این الگوریتم سخت‌افزاری:", "sub_title")

display_academic_text(
    "۱. <b>نمایش داده به صورت تنسور (Tensor Representation):</b> داده‌های ۳ بُعدی به صورت یک آرایه متراکم (Dense Array) ذخیره می‌شوند.<br>"
    "۲. <b>تجمیعِ مستقیم در امتداد محورها (Axis Aggregation):</b> به جای اجرای دستوراتِ سنگین <code>GROUP BY</code>، مقادیرِ آرایه صرفاً در امتداد یک یا چند محور (Axis) جمع زده می‌شوند:<br>"
    "&nbsp;&nbsp;&nbsp;• <b>پایه (۳ بُعدی):</b> داده‌های کامل بدون تجمیع (Base Cuboid).<br>"
    "&nbsp;&nbsp;&nbsp;• <b>۲ بُعدی (2D Cuboids):</b> تجمیع در امتداد ۱ محور (مثلاً جمع زدن روی محور سال).<br>"
    "&nbsp;&nbsp;&nbsp;• <b>۱ بُعدی (1D Cuboids):</b> تجمیع در امتداد ۲ محور.<br>"
    "&nbsp;&nbsp;&nbsp;• <b>مکعب‌واره رأس (Apex / 0D):</b> تجمیع در امتدادِ <i>تمامِ محورها</i> (یک عددِ واحد مثل کُلِ فروش سیستم).<br>"
    "۳. <b>محاسبه هم‌زمان (Single-Pass Computation):</b> آرایه‌ها با ترتیبی بهینه پردازش می‌شوند تا کمترین نیاز به جابجایی در حافظه وجود داشته باشد.", 
    "paragraph"
)


# ==========================================================
# SECTION 2: SIMULATING DATA AS A MULTI-DIMENSIONAL TENSOR
# ==========================================================
display_academic_text("گام اول: ایجاد تنسور پایه (Base Cuboid در حافظه)", "sub_title")

# Step 1: Define Tensor Dimensions for Sales
# Axis 0: Year (3), Axis 1: Region (4), Axis 2: Category (3)
years = ['2024', '2025', '2026']
regions = ['North America', 'Europe', 'Asia Pacific', 'Latin America']
categories = ['Electronics', 'Books', 'Fashion']

shape = (len(years), len(regions), len(categories))  # Shape: (3, 4, 3)

# Generate 3D Sales Tensor (Base Cuboid: 3D NumPy Array)
sales_tensor = np.random.randint(50, 250, size=shape)

display_academic_text(
    f"<b>وضعیت حافظه (MOLAP Engine):</b><br>"
    f"ابعاد تنسور (<code>Year $\\times$ Region $\\times$ Category</code>): <b>{sales_tensor.shape}</b><br>"
    f"تعداد کل سلول‌های پایهٔ رزروشده در حافظه RAM: <b>{sales_tensor.size} واحد</b>", 
    "insight"
)

# ==========================================================
# SECTION 3: MULTIWAY ARRAY AGGREGATION (LATTICE GENERATION)
# ==========================================================
display_academic_text("گام دوم: نتایج تجمیع چندمحوری در شبکه لتیس (Lattice Generation)", "sub_title")

# Step 2: Aggregation across Axes using C-level Vectorized NumPy operations

# 2.1 Compute 2D Cuboids (Compressing 1 Axis)
cuboid_region_category = np.sum(sales_tensor, axis=0)  # Sum over Year (Axis 0) -> New Shape: (4, 3)
cuboid_year_category = np.sum(sales_tensor, axis=1)    # Sum over Region (Axis 1) -> New Shape: (3, 3)
cuboid_year_region = np.sum(sales_tensor, axis=2)      # Sum over Category (Axis 2) -> New Shape: (3, 4)

# 2.2 Compute 1D Cuboids (Compressing 2 Axes)
cuboid_category = np.sum(sales_tensor, axis=(0, 1))    # Sum over Year & Region -> New Shape: (3,)
cuboid_region = np.sum(sales_tensor, axis=(0, 2))      # Sum over Year & Category -> New Shape: (4,)
cuboid_year = np.sum(sales_tensor, axis=(1, 2))        # Sum over Region & Category -> New Shape: (3,)

# 2.3 Compute 0D Apex Cuboid (Compressing All 3 Axes)
apex_total_sales = np.sum(sales_tensor, axis=(0, 1, 2))  # Total Grand Sum

display_academic_text(
    f"فروش کل سیستم (Apex Cuboid - تجمیعِ همزمانِ ۳ محور به کُلِ شبکه): <b>${apex_total_sales:,}k USD</b>", 
    "paragraph"
)

# Step 3: Convert the 2D and 1D numpy arrays back to Pandas DataFrames purely for UI Display
df_region_cat = pd.DataFrame(
    cuboid_region_category, 
    index=regions, 
    columns=categories
).reset_index().rename(columns={'index': 'Region'})

df_1d_category = pd.DataFrame({
    'Category': categories, 
    'Sales_Aggregated_kUSD': cuboid_category
})

# Display 2D and 1D Cuboids
display_side_by_side_datasets(
    [df_region_cat, df_1d_category],
    [
        "مکعب‌واره ۲ بُعدی (Region vs Category) - بُعد Year فشرده شده", 
        "مکعب‌واره ۱ بُعدی (فروش کل هر دسته) - بُعدهای Year و Region فشرده شده"
    ]
)

# ==========================================================
# SECTION 4: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
display_academic_text("مصورسازی تحلیلی: خروجیِ عملیات تجمیع آرایه‌ای", "sub_title")

# Step 4: Visualization using Matplotlib
fig, axes = plt.subplots(1, 2, figsize=(16, 6.5), dpi=135)

# ----------------------------------------------------------
# Subplot 1: Heatmap of 2D Cuboid (Region x Category)
# ----------------------------------------------------------
im = axes[0].imshow(
    cuboid_region_category, 
    cmap='Blues', # Academic Blue palette
    aspect='auto'
)

axes[0].annotate(
    "Tensor Compression:\nThe CPU simply summed the matrices of\n2024, 2025, and 2026 exactly on top of\neach other to generate this 2D view.",
    xy=(0.5, 0.5), xytext=(1.0, 1.5),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title('Figure 1: 2D Cuboid (Region vs Category)', fontsize=14, fontweight='bold', pad=15)
axes[0].set_xlabel('Dimension: Category', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Dimension: Region', fontsize=13, fontweight='bold')

axes[0].set_xticks(range(len(categories)))
axes[0].set_xticklabels(categories, fontweight='bold', fontsize=11)
axes[0].set_yticks(range(len(regions)))
axes[0].set_yticklabels(regions, fontweight='bold', fontsize=11)

# Annotate heatmap values with intelligent text color contrast
for i in range(len(regions)):
    for j in range(len(categories)):
        val = cuboid_region_category[i, j]
        # Calculate dynamic threshold based on array max to decide text color (white on dark blue)
        threshold = np.max(cuboid_region_category) * 0.6
        text_color = 'white' if val > threshold else '#2c3e50'
        axes[0].text(
            j, i, 
            f'${val}k',
            ha='center', va='center',
            color=text_color,
            fontweight='bold',
            fontsize=12
        )

# ----------------------------------------------------------
# Subplot 2: 1D Cuboids Aggregated Sales (Category)
# ----------------------------------------------------------
x_cat = np.arange(len(categories))
width = 0.5

axes[1].bar(
    x_cat,
    cuboid_category,
    color='#1abc9c', # Turquoise
    edgecolor='#2c3e50',
    width=width,
    linewidth=2.0
)

axes[1].annotate(
    "1D Flattening:\nWe further compressed the 2D array by\nsumming it across the Region axis,\nleaving only 1 dimension: Category.",
    xy=(2.0, 1500), xytext=(0.5, 2000),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[1].set_title('Figure 2: 1D Cuboid (Total Sales per Category)', fontsize=14, fontweight='bold', pad=15)
axes[1].set_xlabel('Dimension: Category', fontsize=13, fontweight='bold')
axes[1].set_ylabel('Total Aggregated Sales (kUSD)', fontsize=13, fontweight='bold')
axes[1].set_xticks(x_cat)
axes[1].set_xticklabels(categories, fontweight='bold', fontsize=12)
axes[1].grid(axis='y', linestyle='--', alpha=0.5)

# Annotate Bars
for i, val in enumerate(cuboid_category):
    axes[1].text(
        i, val + (max(cuboid_category) * 0.02), 
        f'${val:,}k', 
        ha='center', fontweight='bold', fontsize=13, color='#2c3e50'
    )

# Clear top and right spines for academic aesthetic
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout(pad=2.0)
plt.show()

# ==========================================================
# SECTION 5: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک در طراحی سیستم‌های OLAP (Strategic Implication):</b><br>"
    r"الگوریتم تجمیعِ آرایه‌های چندطرفه (Multiway Aggregation)، نیاز به ایجادِ جداول موقت و عملیات پُرهزینهٔ <code>JOIN</code> در پایگاه‌های داده رابطه‌ای (SQL) را کاملاً از بین می‌برد.<br>"
    r"با بهره‌گیری از ریاضیات جبر خطی (تنسورها) و نگاشت‌های محوری، تمامِ شبکهٔ لتیس (Lattice) صرفاً از طریق جمع‌زدنِ لایه‌های آرایه روی یکدیگر در فضای RAM پردازش می‌شود. "
    r"این مکانیسم باعث می‌شود زمان پاسخ‌گویی به کوئری‌های چندبُعدی در نرم‌افزارهای BI، از مقیاس دقیقه به <b>کسری از ثانیه</b> کاهش یابد."
)
display_academic_text(insight_conclusion, "insight")
```
