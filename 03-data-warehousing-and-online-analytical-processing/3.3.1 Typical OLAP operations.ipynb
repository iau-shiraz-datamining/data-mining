<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: تحلیل تعاملی فروش آمازون با عملیات پنج‌گانه OLAP 📊
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    تیم تحلیل هوش تجاری آمازون برای بررسی فروش جهانی خود، داده‌ها را در یک مکعب داده چندبعدی شامل ابعاد زمان، منطقه جغرافیایی و دسته‌بندی کالا ذخیره می‌کند. مدیران برای پاسخ به سوالات مختلف، ۵ عملیات اصلی OLAP را روی این مکعب اجرا می‌کنند:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. عملیات فشردن (Roll-up): تجمیع داده‌ها و حرکت به سمت سطوح بالاتر انتزاع. مثلاً تبدیل داده‌های فروش ماهانه به فروش سه ماهه (فصلی) یا سالانه.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. عملیات جزئی‌سازی (Drill-down): باز کردن داده‌های خلاصه‌شده و حرکت به سمت جزئیات دقیق‌تر. مثلاً شکستن فروش سالانه به فروش ماهانه برای پیدا کردن علت افت فروش در یک ماه خاص.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. عملیات برش (Slice): برش دادن مکعب روی یک بعد خاص برای بررسی یک لایه دو بعدی. مثلاً فیلتر کردن کل مکعب فقط برای سال ۲۰۲۳.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. عملیات تاس‌زدن (Dice): انتخاب یک زیرمکعب کوچک‌تر با مشخص کردن فیلتر روی چند بعد مختلف. مثلاً انتخاب داده‌های فروش سال ۲۰۲۳، منطقه آمریکا و دسته کالای الکترونیک.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۵. عملیات چرخش (Pivot): چرخاندن محورهای مکعب داده برای تغییر زاویه دید و نمایش داده‌ها از یک دیدگاه جدید (مثلاً جا‌به‌جا کردن جای سطرها و ستون‌ها در جدول گزارش).
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: این عملیات پنج‌گانه به مدیران اجازه می‌دهد در چند ثانیه از کلی‌ترین گزارش‌های مدیریتی به جزیی‌ترین داده‌های عملیاتی دسترسی پیدا کنند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: تحلیل تعاملی و چندبعدی داده‌ها با ترکیب ۵ عملیات پایه OLAP
    </div>
  </p>

</div>

```python
display_academic_text(
    "تحلیل تعاملی فروشِ آمازون با عملیات پنج‌گانه OLAP 🎲", 
    "main_title"
)

display_academic_text(
    "تیم تحلیل هوش تجاری (BI) آمازون برای بررسی فروش جهانی خود، داده‌ها را در یک <b>مکعب دادهٔ چندبُعدی (Data Cube)</b> شامل ابعادِ زمان، "
    "منطقه جغرافیایی و دسته‌بندیِ کالا ذخیره می‌کند. مدیران برای پاسخ به سوالات مختلف و چرخش در این فضا، <b>۵ عملیاتِ اصلی OLAP</b> را روی این مکعب اجرا می‌کنند:", 
    "paragraph"
)

display_academic_text(
    "۱. <b>فشردن (Roll-up):</b> تجمیع داده‌ها و حرکت به سطوح بالاتر انتزاع. مثلاً تبدیل فروش ماهانه به سالانه.<br>"
    "۲. <b>جزئی‌سازی (Drill-down):</b> باز کردن داده‌های خلاصه‌شده و حرکت به سمت جزئیات. مثلاً شکستن فروش سالانه به فصلی برای کشف علت افت فروش.<br>"
    "۳. <b>بُرش (Slice):</b> برش دادنِ مکعب روی <i>یک بُعدِ خاص</i> برای بررسی لایهٔ دو بُعدی. مثلاً فیلتر کردن کل مکعب <i>فقط</i> برای سال ۲۰۲۶.<br>"
    "۴. <b>تاس‌زدن (Dice):</b> انتخاب یک زیرمکعب کوچک‌تر با مشخص کردن فیلتر روی <i>چندین بُعد هم‌زمان</i>. مثلاً سال ۲۰۲۶، <b>و</b> منطقه آمریکا، <b>و</b> دسته الکترونیک.<br>"
    "۵. <b>چرخش (Pivot):</b> چرخاندن محورهای مکعب برای تغییر زاویه دید تحلیلی (مثلاً جا‌به‌جا کردن جای سطرها و ستون‌ها در گزارش).", 
    "paragraph"
)

# ==========================================================
# SECTION 2: DATA GENERATION AND OLAP OPERATIONS PIPELINE
# ==========================================================
# Step 1: Create Amazon 3D Sales Dataset
# Dimensions: Year (2025, 2026), Region (US, Europe), Category (Electronics, Apparel)

records = []
years = [2025, 2026]
quarters = ['Q1', 'Q2', 'Q3', 'Q4']
regions = ['US', 'Europe']
categories = ['Electronics', 'Apparel']

# Generating the Multidimensional Cube Data
for y in years:
    for q in quarters:
        for r in regions:
            for c in categories:
                sales = np.random.randint(100, 500)
                records.append({
                    'Year': y,
                    'Quarter': q,
                    'Region': r,
                    'Category': c,
                    'Sales_kUSD': sales,
                })

df = pd.DataFrame(records)

# Step 2: Implement 5 Typical OLAP Operations in Pandas

# 1. Roll-up: Aggregating from Quarter to Year level (Discarding Quarter info)
rollup_df = df.groupby(['Year', 'Region'])['Sales_kUSD'].sum().reset_index()

# 2. Drill-down: Expanding Year 2026 into Quarter details
drilldown_df = df[df['Year'] == 2026].groupby(['Quarter', 'Category'])['Sales_kUSD'].sum().reset_index()

# 3. Slice: Filtering on a single dimension (e.g., Year = 2026) -> Makes a 3D cube into a 2D slice
slice_df = df[df['Year'] == 2026]

# 4. Dice: Filtering on multiple dimensions (Year = 2026 AND Region = 'US' AND Category = 'Electronics') -> Makes a sub-cube
dice_df = df[
    (df['Year'] == 2026) & 
    (df['Region'] == 'US') & 
    (df['Category'] == 'Electronics')
]

# 5. Pivot: Rotating axes (Category vs Region for the 2026 Slice)
pivot_df = pd.pivot_table(
    slice_df,
    values='Sales_kUSD',
    index='Category',
    columns='Region',
    aggfunc='sum',
).reset_index() # Flatten the MultiIndex for clean display

# Display preparation to prevent empty index columns in HTML
df_disp = df.head(6).reset_index(drop=True)
rollup_disp = rollup_df.head(4).reset_index(drop=True)

# Display sample results of the operations
display_side_by_side_datasets(
    [df_disp, rollup_disp, pivot_df],
    [
        "جدول ۱: پایگاه داده اصلی (شامل جزئیات فصل)", 
        "جدول ۲: عملیات Roll-up (تجمیع فصل‌ها به کُل سال)",
        "جدول ۳: عملیات Pivot (چرخش دسته‌بندی و منطقه)"
    ]
)

# ==========================================================
# SECTION 3: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
fig, axes = plt.subplots(2, 2, figsize=(19, 12), dpi=135)

# ----------------------------------------------------------
# Plot 1: Roll-up View (Sales by Year and Region)
# ----------------------------------------------------------
pivot_rollup = rollup_df.pivot(index='Year', columns='Region', values='Sales_kUSD')
pivot_rollup.plot(
    kind='bar', 
    ax=axes[0, 0], 
    color=['#2980b9', '#1abc9c'], # Classic Blue, Turquoise
    edgecolor='#2c3e50',
    linewidth=2.0
)

axes[0, 0].annotate(
    "Roll-Up (Aggregation):\nQuarters are completely hidden.\nWe aggregated up to the Year level\nto see the macro-trend of regions.",
    xy=(0.0, 1500), xytext=(-0.4, 2100),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[0, 0].set_title('Figure 1: Roll-Up Operation (Yearly Sales by Region)', fontsize=14, fontweight='bold', pad=15)
axes[0, 0].set_ylabel('Total Sales (kUSD)', fontsize=13, fontweight='bold')
axes[0, 0].set_xlabel('Macro Dimension: Year', fontsize=13, fontweight='bold')
axes[0, 0].grid(axis='y', linestyle='--', alpha=0.5)
axes[0, 0].tick_params(axis='x', rotation=0, labelsize=12)
axes[0, 0].set_xticklabels(axes[0, 0].get_xticklabels(), fontweight='bold')
axes[0, 0].legend(title='Region', frameon=True, fontsize=11)

# ----------------------------------------------------------
# Plot 2: Drill-down View (Quarterly details for 2026 by Category)
# ----------------------------------------------------------
pivot_drill = drilldown_df.pivot(index='Quarter', columns='Category', values='Sales_kUSD')
pivot_drill.plot(
    kind='line', 
    marker='o', 
    ax=axes[0, 1], 
    color=['#2c3e50', '#5DADE2'], # Navy, Light Blue
    linewidth=3.0,
    markersize=10
)

axes[0, 1].annotate(
    "Drill-Down (Detailing):\nWe expanded 2026 back into Quarters\nto find out exactly WHEN 'Apparel'\nsales spiked.",
    xy=(1.0, 520), xytext=(1.5, 450),
    fontsize=11, fontweight='bold', color='#2c3e50',
    arrowprops=dict(arrowstyle="->", color="#34495e", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#ecf0f1", ec="#bdc3c7", lw=1.5)
)

axes[0, 1].set_title('Figure 2: Drill-Down Operation (2026 Quarterly Detail)', fontsize=14, fontweight='bold', pad=15)
axes[0, 1].set_ylabel('Quarterly Sales (kUSD)', fontsize=13, fontweight='bold')
axes[0, 1].set_xlabel('Micro Dimension: Quarter', fontsize=13, fontweight='bold')
axes[0, 1].grid(True, linestyle='--', alpha=0.5)
axes[0, 1].tick_params(axis='x', labelsize=12)
axes[0, 1].set_xticklabels(axes[0, 1].get_xticklabels(), fontweight='bold')
axes[0, 1].legend(title='Product Category', frameon=True, fontsize=11)

# ----------------------------------------------------------
# Plot 3: Slice & Dice View (Dice: US Electronics in 2026)
# ----------------------------------------------------------
axes[1, 0].bar(
    dice_df['Quarter'],
    dice_df['Sales_kUSD'],
    color='#1abc9c', # Turquoise
    edgecolor='#2c3e50',
    linewidth=2.0,
    width=0.5,
)

axes[1, 0].annotate(
    "Slice & Dice (Multi-Filtering):\nWe sliced Year=2026. Then we diced\nRegion=US AND Category=Electronics.\nWe are looking at a very tiny sub-cube.",
    xy=(2.0, 200), xytext=(0.5, 300),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[1, 0].set_title('Figure 3: Slice & Dice (2026 + US + Electronics)', fontsize=14, fontweight='bold', pad=15)
axes[1, 0].set_ylabel('Isolated Sales (kUSD)', fontsize=13, fontweight='bold')
axes[1, 0].set_xlabel('Time Dimension', fontsize=13, fontweight='bold')
axes[1, 0].grid(axis='y', linestyle='--', alpha=0.5)
axes[1, 0].tick_params(axis='x', labelsize=12)
axes[1, 0].set_xticklabels(axes[1, 0].get_xticklabels(), fontweight='bold')

for p in axes[1, 0].patches:
    axes[1, 0].annotate(
        f"${int(p.get_height())}k", 
        (p.get_x() + p.get_width() / 2., p.get_height()), 
        ha='center', va='center', xytext=(0, -12), textcoords='offset points', 
        fontweight='bold', fontsize=12, color='white'
    )

# ----------------------------------------------------------
# Plot 4: Pivot View (Rotated Category vs Region)
# ----------------------------------------------------------
# Exclude the 'Category' column from plotting, use it as the index
pivot_df_plot = pivot_df.set_index('Category')

pivot_df_plot.plot(
    kind='barh', 
    ax=axes[1, 1], 
    color=['#2980b9', '#2c3e50'], # Classic Blue, Navy
    edgecolor='white',
    linewidth=1.5
)

axes[1, 1].annotate(
    "Pivot (Rotation):\nWe rotated the visual perspective.\nInstead of putting Regions on the axis,\nwe made Category the Y-axis.",
    xy=(1000, 1.0), xytext=(1200, 0.5),
    fontsize=11, fontweight='bold', color='#2c3e50',
    arrowprops=dict(arrowstyle="->", color="#34495e", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#ecf0f1", ec="#bdc3c7", lw=1.5)
)

axes[1, 1].set_title('Figure 4: Pivot Operation (Rotated Perspective)', fontsize=14, fontweight='bold', pad=15)
axes[1, 1].set_xlabel('Total Aggregated Sales (kUSD)', fontsize=13, fontweight='bold')
axes[1, 1].set_ylabel('Dimension: Category', fontsize=13, fontweight='bold')
axes[1, 1].grid(axis='x', linestyle='--', alpha=0.5)
axes[1, 1].tick_params(axis='y', labelsize=12)
axes[1, 1].set_yticklabels(axes[1, 1].get_yticklabels(), fontweight='bold')
axes[1, 1].legend(title='Dimension: Region', loc='upper right', frameon=True, fontsize=11)

# Clear top and right spines for a clean academic aesthetic
for ax in axes.flat:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.suptitle('OLAP Cubes: Visual Proof of the 5 Core Analytical Operations', fontsize=17, fontweight='bold', y=1.02)

plt.tight_layout(pad=3.0)
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه کاربردی (Strategic Implication):</b><br>"
    r"این <b>عملیاتِ ۵گانه</b> دقیقاً همان مکانیسمی است که در پس‌زمینهٔ داشبوردهای هوش تجاری (نظیر Tableau یا PowerBI) اجرا می‌شود. "
    r"مدیران می‌توانند در چند ثانیه از کُلی‌ترین گزارش‌های استراتژیک (<b>Roll-up</b> - گراف ۱) به جزیی‌ترین داده‌های عملیاتی و فصلی (<b>Drill-down</b> - گراف ۲) دست پیدا کنند.<br>"
    r"موتورهای پردازشیِ OLAP با پیش‌محاسبهٔ این ابعاد در ساختارهایی به نام <i>مکعب داده (Data Cubes)</i>، نیاز به کوئری‌های کُند، پیچیده و پُر از JOIN در SQL را از بین برده "
    r"و تجربهٔ <b>تحلیلِ تعاملی (Interactive Analytics)</b> را خلق می‌نمایند. اگر مدیری فقط بخواهد یک گوشهٔ خاص از بازار را ببیند، کافیست روی آن <b>Slice & Dice</b> کند (گراف ۳) و در صورت نیاز، زاویهٔ دید گزارش را <b>Pivot</b> نماید (گراف ۴)."
)
display_academic_text(insight_conclusion, "insight")
```
