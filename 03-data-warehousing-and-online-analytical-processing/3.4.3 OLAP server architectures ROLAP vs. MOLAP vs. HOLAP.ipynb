<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: انتخاب معماری سرور OLAP در آمازون (مقایسه ROLAP، MOLAP و HOLAP) 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مهندسان انبار داده شرکت آمازون برای پاسخ‌گویی به پرس‌وجوهای چندبعدی مدیران، باید معماری مناسبی برای سرور OLAP انتخاب کنند. بسته به حجم داده‌ها و نیاز به سرعت، ۳ معماری اصلی وجود دارد:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. معماری ROLAP (Relational OLAP) — مبتنی بر پایگاه داده رابطه‌ای:<br>
    در این معماری، داده‌های واقعیت و ابعاد در همان پایگاه‌های داده رابطه‌ای سنتی (مثل PostgreSQL یا SQLite) ذخیره می‌شوند. کوئری‌های چندبعدی در زمان اجرا به دستورات SQL تبدیل شده و با عملیات JOIN و GROUP BY محاسبه می‌شوند.<br>
    مزیت: قابلیت مقیاس‌پذیری تا حجم‌های بسیار بزرگ (پتابایت) بدون محدودیت حافظه.<br>
    عیب: سرعت پایین‌تر کوئری‌ها به دلیل محاسبات سنگین در زمان اجرا.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. معماری MOLAP (Multidimensional OLAP) — مبتنی بر آرایه‌های چندبعدی:<br>
    داده‌ها از قبل تجمیع شده و در ساختارهای آرایه‌ای اختصاصی چندبعدی (In-Memory Arrays) ذخیره می‌شوند. دستیابی به داده‌ها مستقیم و با آدرس‌دهی اندیس آرایه انجام می‌شود.<br>
    مزیت: سرعت فوق‌العاده بالا و پاسخگویی آنی به کوئری‌ها در چند میلی‌ثانیه.<br>
    عیب: نیاز به حافظه زیاد و محدودیت در پذیریش حجم‌های بسیار عظیم داده‌های جزئی.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. معماری HOLAP (Hybrid OLAP) — معماری ترکیبی بهینه:<br>
    این روش مزایای ROLAP و MOLAP را با هم ترکیب می‌کند. خلاصه‌ها و تجمیع‌های سطح بالا (مثل فروش کل سالانه یا فصلی) در مکعب‌های چندبعدی سریع (MOLAP) ذخیره می‌شوند، اما داده‌های جزئی و ریزتراکنش‌ها در پایگاه داده رابطه‌ای (ROLAP) نگهداری می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: معماران داده با به‌کارگیری HOLAP، تعادلی هوشمندانه بین سرعت پاسخ‌دهی به داشبوردهای مدیریتی و حفظ حجم زیاد داده‌های جزئی ایجاد می‌کنند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: شبیه‌سازی و مقایسه عملکرد ROLAP، MOLAP و HOLAP در اجرای کوئری‌های تحلیلی
    </div>
  </p>

```python
display_academic_text(
    "انتخاب معماری سرور OLAP در آمازون (مقایسه ROLAP، MOLAP و HOLAP) 🏛️", 
    "main_title"
)

display_academic_text(
    "مهندسان انبار داده شرکت آمازون برای پاسخ‌گویی به پرس‌وجوهای چندبُعدی مدیران، باید معماری مناسبی برای <b>سرور تحلیلی (OLAP Server)</b> انتخاب کنند. "
    "بسته به حجم داده‌ها و نیاز به سرعت، ۳ معماریِ اصلی در صنعت وجود دارد:", 
    "paragraph"
)

display_academic_text("۱. معماری ROLAP (Relational OLAP):", "sub_title")
display_academic_text(
    "کوئری‌های چندبُعدی در زمانِ اجرا (Run-time) به دستورات SQL تبدیل شده و با عملیات سنگین JOIN و GROUP BY روی دیتابیسِ رابطه‌ای سنتی محاسبه می‌شوند.<br>"
    "<b>مزیت:</b> مقیاس‌پذیری پتابایتی بدون محدودیت حافظه (مناسب برای جزئیاتِ خُرد).<br>"
    "<b>عیب:</b> سرعتِ پایین‌تر به دلیل تفسیر SQL و اسکن دیسک در زمان اجرا.", 
    "paragraph"
)

display_academic_text("۲. معماری MOLAP (Multidimensional OLAP):", "sub_title")
display_academic_text(
    "داده‌ها از قبل تجمیع شده و در ساختارهای آرایه‌ای اختصاصی درون حافظهٔ رَم (In-Memory Arrays / Cubes) ذخیره می‌شوند.<br>"
    "<b>مزیت:</b> پاسخگوییِ آنی در چند میلی‌ثانیه با آدرس‌دهی مستقیمِ اندیس ($O(1)$).<br>"
    "<b>عیب:</b> نیاز به حافظهٔ RAM زیاد و ناتوانی در ذخیرهٔ حجم عظیم داده‌های جزئی (Sparsity Problem).", 
    "paragraph"
)

display_academic_text("۳. معماری HOLAP (Hybrid OLAP):", "sub_title")
display_academic_text(
    "این روشِ ترکیبی (Hybrid)، مزایای دو مدل قبل را در هم می‌آمیزد. خلاصه‌ها و تجمیع‌های سطح بالا در مکعب‌های سریع MOLAP ذخیره شده، "
    "و داده‌های ریزتراکنش‌ها برای زمانی که مدیر نیاز به Drill-Down دارد، در دیتابیسِ ارزان‌قیمتِ ROLAP نگهداری می‌شوند.", 
    "paragraph"
)

# ==========================================================
# SECTION 2: DATA GENERATION AND ARCHITECTURE SIMULATIONS
# ==========================================================
display_academic_text("مرحله نخست: شبیه‌سازی معماری‌های سرور روی داده‌های تراکنشی", "sub_title")

# Step 1: Prepare Synthetic Sales Dataset (Increased to 100,000 for better benchmark scale)
n_records = 100000

years = [2024, 2025, 2026]
regions = ['North', 'South', 'East', 'West']
categories = ['Electronics', 'Apparel', 'Home']

df_sales = pd.DataFrame({
    'Year': np.random.choice(years, size=n_records),
    'Region': np.random.choice(regions, size=n_records),
    'Category': np.random.choice(categories, size=n_records),
    'Sales_USD': np.random.uniform(10.0, 500.0, size=n_records)
})

display_side_by_side_datasets(
    [df_sales.head(5).reset_index(drop=True)],
    [f"نمونه داده‌های خام تراکنش‌های فروش (پایگاه داده اصلی شامل {n_records:,} رکورد)"]
)

# ------------------------------------------------------------------------------
# Architecture 1: ROLAP Simulation (Relational Database via SQLite on Disk/Memory)
# ------------------------------------------------------------------------------
conn = sqlite3.connect(':memory:')
df_sales.to_sql('fact_sales', conn, index=False, if_exists='replace')

def query_rolap(year, region):
    """ROLAP Query: Dynamic SQL interpretation, filtering, and Group By over relational table."""
    sql = f"SELECT Year, Region, SUM(Sales_USD) as Total_Sales FROM fact_sales WHERE Year = {year} AND Region = '{region}' GROUP BY Year, Region"
    return pd.read_sql_query(sql, conn)

# ------------------------------------------------------------------------------
# Architecture 2: MOLAP Simulation (Pre-computed 3D NumPy Array in RAM)
# ------------------------------------------------------------------------------
# Dimensions mapping for O(1) array access
year_map = {y: i for i, y in enumerate(years)}
region_map = {r: i for i, r in enumerate(regions)}
cat_map = {c: i for i, c in enumerate(categories)}

# Precompute full 3D array in memory (MOLAP Cube)
molap_cube = np.zeros((len(years), len(regions), len(categories)))

for _, row in df_sales.iterrows():
    y_idx = year_map[row['Year']]
    r_idx = region_map[row['Region']]
    c_idx = cat_map[row['Category']]
    molap_cube[y_idx, r_idx, c_idx] += row['Sales_USD']

def query_molap(year, region):
    """MOLAP Query: Direct O(1) array slicing in memory. No SQL interpretation."""
    y_idx = year_map[year]
    r_idx = region_map[region]
    total_sales = np.sum(molap_cube[y_idx, r_idx, :])
    return total_sales

# ------------------------------------------------------------------------------
# Architecture 3: HOLAP Simulation (Hybrid Model Strategy)
# ------------------------------------------------------------------------------
# Pre-computed MOLAP cache for High-level Aggregates (Year x Region)
holap_summary_cache = df_sales.groupby(['Year', 'Region'])['Sales_USD'].sum().to_dict()

def query_holap(year, region, detailed_category=None):
    """HOLAP Query: Routes high-level to MOLAP cache, routes detailed drill-down to ROLAP."""
    if detailed_category is None:
        # High-level summary -> Instant return from MOLAP cache
        return holap_summary_cache.get((year, region), 0.0)
    else:
        # Detailed drill-down -> Processed by ROLAP SQL engine
        sql = f"SELECT Year, Region, Category, SUM(Sales_USD) as Total_Sales FROM fact_sales WHERE Year = {year} AND Region = '{region}' AND Category = '{detailed_category}' GROUP BY Year, Region, Category"
        return pd.read_sql_query(sql, conn)

# ==========================================================
# SECTION 3: EXECUTION BENCHMARKING
# ==========================================================
# Step 5: Benchmark Execution Times over 100 Summary Queries
n_queries = 100

# Benchmark ROLAP
start = time.time()
for _ in range(n_queries):
    query_rolap(2026, 'North')
time_rolap = (time.time() - start) * 1000

# Benchmark MOLAP
start = time.time()
for _ in range(n_queries):
    query_molap(2026, 'North')
time_molap = (time.time() - start) * 1000

# Benchmark HOLAP (Summary Query Mode)
start = time.time()
for _ in range(n_queries):
    query_holap(2026, 'North')
time_holap = (time.time() - start) * 1000

display_academic_text(
    f"<b>نتایج بنچمارک برای اجرایِ تکراریِ {n_queries} کوئری خلاصه‌سازی (Summary):</b><br>"
    f"🔴 زمان اجرای ROLAP (تفسیر کُدهای SQL و اسکن جدول): <b>{time_rolap:.1f} میلی‌ثانیه</b><br>"
    f"🟢 زمان اجرای MOLAP (آدرس‌دهیِ مستقیمِ آرایه در رَم): <b>{time_molap:.1f} میلی‌ثانیه</b><br>"
    f"🟢 زمان اجرای HOLAP (مسیریابی هوشمند به کَشِ MOLAP): <b>{time_holap:.1f} میلی‌ثانیه</b>", 
    "insight"
)

# ==========================================================
# SECTION 4: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
# Step 6: Visualization using Matplotlib
fig, axes = plt.subplots(1, 2, figsize=(16, 6.5), dpi=135)

# ----------------------------------------------------------
# Subplot 1: Query Execution Time Comparison
# ----------------------------------------------------------
architectures = ['ROLAP\n(Relational SQL)', 'MOLAP\n(In-Memory Array)', 'HOLAP\n(Hybrid Router)']
times = [time_rolap, time_molap, time_holap]

bars1 = axes[0].bar(
    architectures, 
    times, 
    color=['#e74c3c', '#1abc9c', '#2980b9'], # Red, Turquoise, Classic Blue
    edgecolor='#2c3e50', 
    linewidth=2.0,
    width=0.5
)

axes[0].annotate(
    "Query Translation Overhead:\nROLAP is slow because every query requires\nthe database engine to parse SQL, build an\nexecution plan, and scan the raw table rows.",
    xy=(0.0, time_rolap/2), xytext=(0.4, time_rolap*0.7),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title(f'Figure 1: High-Level Summary Query Time ({n_queries} Runs)', fontsize=14, fontweight='bold', pad=15)
axes[0].set_ylabel('Execution Time (Milliseconds) - Lower is Better', fontsize=13, fontweight='bold')
axes[0].grid(axis='y', linestyle='--', alpha=0.5)
axes[0].set_xticklabels(architectures, fontweight='bold', fontsize=12)

for bar in bars1:
    yval = bar.get_height()
    axes[0].text(
        bar.get_x() + bar.get_width() / 2, 
        yval + (max(times) * 0.02), 
        f'{yval:.1f} ms', 
        ha='center', fontweight='bold', fontsize=13, color='#2c3e50'
    )

# ----------------------------------------------------------
# Subplot 2: Architecture Characteristics (Qualitative Ratings)
# ----------------------------------------------------------
categories_eval = ['Query Speed\n[Fast is 5]', 'Storage Scalability\n[Petabytes is 5]', 'Drill-Down Detail\n[Item-level is 5]']
# Score from 1 to 5 (5 is Best)
rolap_scores = [2, 5, 5]
molap_scores = [5, 2, 1]
holap_scores = [4.5, 4.5, 5] # HOLAP brings balance

x = np.arange(len(categories_eval))
width = 0.25

axes[1].bar(
    x - width, 
    rolap_scores, 
    width, 
    label='ROLAP (Relational DB)', 
    color='#e74c3c', # Red
    edgecolor='#2c3e50',
    linewidth=1.5
)
axes[1].bar(
    x, 
    molap_scores, 
    width, 
    label='MOLAP (In-Memory Cube)', 
    color='#1abc9c', # Turquoise
    edgecolor='#2c3e50',
    linewidth=1.5
)
axes[1].bar(
    x + width, 
    holap_scores, 
    width, 
    label='HOLAP (Hybrid Strategy)', 
    color='#2980b9', # Classic Blue
    edgecolor='#2c3e50',
    linewidth=1.5
)

axes[1].annotate(
    "The HOLAP Sweet Spot:\nHOLAP acts as a smart router. It provides the\nblazing speed of MOLAP for dashboard summaries,\nwhile preserving ROLAP's ability to drill-down\ninto infinite petabytes of detail.",
    xy=(2.2, 5.0), xytext=(0.6, 5.5),
    fontsize=11, fontweight='bold', color='#2980b9',
    arrowprops=dict(arrowstyle="->", color="#2980b9", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#ecf0f1", ec="#bdc3c7", lw=1.5)
)

axes[1].set_title('Figure 2: Architectural Trade-off Evaluation (Score 1 to 5)', fontsize=14, fontweight='bold', pad=15)
axes[1].set_xticks(x)
axes[1].set_xticklabels(categories_eval, fontweight='bold', fontsize=11)
axes[1].set_ylabel('Performance / Capability Score', fontsize=13, fontweight='bold')
axes[1].set_ylim(0, 6.5) # Extra headroom for legend and annotations
axes[1].legend(loc='lower left', frameon=True, fontsize=11, edgecolor='#bdc3c7')
axes[1].grid(axis='y', linestyle='--', alpha=0.5)

# Clear top and right spines for a clean academic aesthetic
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# Clean up database connection
conn.close()

# ==========================================================
# SECTION 5: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (Strategic Implication):</b><br>"
    r"معماران داده در سیستم‌های کلان‌داده همواره با یک موازنه (Trade-off) میان «سرعت» و «حجم ذخیره‌سازی» روبه‌رو هستند. "
    r"ذخیره کردن پتابایت‌ها داده با تمام جزئیاتِ خُرد (Item-level) در رم (MOLAP) عملاً غیرممکن و به‌شدت گران‌قیمت است. "
    r"از طرفی، کوئری زدنِ مستقیم روی دیتابیس‌های رابطه‌ای (ROLAP) داشبوردها را کُند می‌کند.<br>"
    r"معماری <b>HOLAP</b> راه‌حلی هوشمندانه و استراتژیک برای این چالش است؛ سیستمی که به عنوان یک <b>مسیریابِ هوشمند (Smart Router)</b> عمل می‌کند: "
    r"این معماری با استفاده از کَشِ MOLAP داشبوردهایِ کلان مدیریتی را در کسری از ثانیه بارگذاری کرده و هم‌زمان با اتکا به ROLAP، محدودیت‌های ذخیره‌سازیِ جزئیاتِ دقیق را دور می‌زند."
)
display_academic_text(insight_conclusion, "insight")
```
