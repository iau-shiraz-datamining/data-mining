<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: تحلیل فروش در سطوح مختلف زمانی در «وال‌مارت» (Walmart) 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مدیران «وال‌مارت» (Walmart) نیاز دارند گزارش‌های فروش را در سطوح مختلف جزئیات بررسی کنند. مثلاً مدیرعامل می‌خواهد فروش کل سالانه را ببیند، اما مدیر بازاریابی نیاز دارد فروش ماه به ماه را تحلیل کند. این سطوح تودرتوی جزئیات را <b>سلسله مراتب مفاهیم (Concept Hierarchy)</b> می‌نامند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>مثال عملی: سلسله مراتب بعد زمان (Time Dimension Hierarchy):</b><br>
    داده‌های فروش روزانه در پایین‌ترین سطح جزئیات ثبت شده‌اند، اما می‌توان آن‌ها را در سطوح انتزاعی‌تری خلاصه کرد:
  </p>

  <div style="text-align: center; margin: 15px 0; font-weight: bold; font-size: 14pt;">
    Day → Month → Quarter → Year
  </div>

  <ul style="text-align: right; margin: 0 0 10px; padding-right: 22px;">
    <li><b>سطح ۱ (پایین‌ترین / Day):</b> فروش هر روز مشخص (مثلاً ۲۰۲۳-۰۱-۰۵).</li>
    <li><b>سطح ۲ (Month):</b> جمع فروش یک ماه (مثلاً January 2023).</li>
    <li><b>سطح ۳ (Quarter):</b> جمع فروش یک فصل (مثلاً Q1-2023).</li>
    <li><b>سطح ۴ (بالاترین / Year):</b> جمع فروش کل سال (مثلاً 2023).</li>
  </ul>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>ارتباط سلسله مراتب با عملیات OLAP:</b><br>
    &nbsp;&nbsp;&nbsp;• <b>Roll-Up (جمع‌بندی):</b> حرکت از سطح جزئی‌تر به سطح خلاصه‌تر (مثلاً از ماه به فصل).<br>
    &nbsp;&nbsp;&nbsp;• <b>Drill-Down (ریزنگری):</b> حرکت معکوس از سطح خلاصه به سطح جزئی‌تر (مثلاً از سال به ماه).
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> سلسله مراتب مفاهیم امکان ناوبری چندسطحی در داده‌ها را فراهم می‌سازد تا هر مدیر در سطح مورد نیاز خود، گزارش‌ دقیق و معناداری دریافت کند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: تجمیع تدریجی داده‌ها از سطح روزانه تا سالانه بر اساس سلسله مراتب بعد زمان
    </div>
  </p>

</div>

```python
display_academic_text(
    "تحلیل چندسطحی فروش در «وال‌مارت» با استفاده از سلسله‌مراتب مفاهیم (Concept Hierarchy) 📊", 
    "main_title"
)

display_academic_text(
    "مدیران «وال‌مارت» (Walmart) نیاز دارند گزارش‌های فروش را در سطوح مختلفِ جزئیات بررسی کنند. مثلاً مدیرعامل می‌خواهد فروش کل سالانه را ببیند، "
    "اما مدیر بازاریابی نیاز دارد فروش ماه به ماه را تحلیل کند. این سطوحِ تودرتوی جزئیات را در انبارهای داده، <b>سلسله‌مراتب مفاهیم (Concept Hierarchy)</b> می‌نامند.", 
    "paragraph"
)

display_academic_text("سلسله‌مراتب بُعد زمان (Time Dimension Formal Hierarchy):", "sub_title")

display_academic_text(
    r"$$\text{Hierarchy}_{Time} = \{ \text{Day} \xrightarrow{\text{Roll-Up}} \text{Month} \xrightarrow{\text{Roll-Up}} \text{Quarter} \xrightarrow{\text{Roll-Up}} \text{Year} \}$$", 
    "paragraph"
)

display_academic_text(
    "ارتباط سلسله‌مراتب با عملیات OLAP:", 
    "sub_title"
)

display_academic_text(
    "<b>• جمع‌بندی (Roll-Up):</b> حرکت از سطح جزئی‌تر به سطح خلاصه‌تر (مثلاً از ماه به فصل). باعث کاهش ابعاد و نویز می‌شود.<br>"
    "<b>• ریزنگری (Drill-Down):</b> حرکت معکوس از سطح خلاصه به سطح جزئی‌تر (مثلاً از سال به ماه). باعث کشف الگوهای پنهان در داده‌های خُرد می‌شود.", 
    "paragraph"
)

# ==========================================================
# SECTION 2: DATA GENERATION & HIERARCHICAL AGGREGATION
# ==========================================================
display_academic_text(
    "مرحله نخست: تولید داده‌های روزانه و اجرای عملیات Roll-Up", 
    "sub_title"
)

# Step 1: Create daily-level sales data (Lowest granularity)
dates = pd.date_range(start='2026-01-01', end='2026-12-31', freq='D')
daily_sales = np.random.uniform(50, 300, size=len(dates))

df = pd.DataFrame({'Date': dates, 'Daily_Sales_kUSD': daily_sales})

# Step 2: Build Concept Hierarchy Levels from the Time Dimension

# Level 2: Month (Roll-up from Day)
df['Month'] = df['Date'].dt.to_period('M')
monthly_sales = df.groupby('Month')['Daily_Sales_kUSD'].sum().reset_index()
monthly_sales.columns = ['Month', 'Monthly_Sales_kUSD']

df_day_disp = df[['Date', 'Daily_Sales_kUSD']].head(6).reset_index(drop=True)
df_month_disp = monthly_sales.head(6).reset_index(drop=True)

# Display Level 1 and 2
display_side_by_side_datasets(
    [df_day_disp, df_month_disp],
    [
        "سطح ۱: فروش روزانه (پایین‌ترین سطح جزئیات)", 
        "سطح ۲: فروش ماهانه (Roll-Up از روزانه)"
    ]
)

# Level 3: Quarter (Roll-up from Month/Day)
df['Quarter'] = df['Date'].dt.to_period('Q')
quarterly_sales = df.groupby('Quarter')['Daily_Sales_kUSD'].sum().reset_index()
quarterly_sales.columns = ['Quarter', 'Quarterly_Sales_kUSD']

# Level 4: Year (Highest Granularity)
df['Year'] = df['Date'].dt.to_period('Y')
yearly_sales = df.groupby('Year')['Daily_Sales_kUSD'].sum().reset_index()
yearly_sales.columns = ['Year', 'Yearly_Sales_kUSD']

df_qtr_disp = quarterly_sales.reset_index(drop=True)
df_yr_disp = yearly_sales.reset_index(drop=True)

# Display Level 3 and 4
display_side_by_side_datasets(
    [df_qtr_disp, df_yr_disp],
    [
        "سطح ۳: فروش فصلی (Roll-Up از ماهانه)", 
        "سطح ۴: فروش سالانه (بالاترین سطح تجمیع)"
    ]
)

# ==========================================================
# ==========================================================
# SECTION 3: VISUAL ANALYTICS (ACADEMIC BLUE) - CORRECTED
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی - تکامل هندسی با حرکت در سلسله‌مراتب", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکیِ زیر تأثیر عملیات <b>Roll-Up</b> را به صورت بصری اثبات می‌کند. با حرکت از سمت چپ به راست و از بالا به پایین، "
    "نوساناتِ شدید و نویزهای روزانه (نمودار اول) در یکدیگر ادغام شده و الگوهایِ کلان (نمودارهای فصل و سال) پدیدار می‌گردند. "
    "<i>رنگ‌بندی نمودارها با افزایش سطحِ تجمیع، از فیروزه‌ای به سرمه‌ایِ تیره تغییر می‌کند تا سنگینیِ داده‌های تجمیع‌شده را تداعی نماید.</i>", 
    "paragraph"
)

# Step 3: Visualization of Concept Hierarchy Levels
fig, axes = plt.subplots(2, 2, figsize=(19, 11), dpi=135)

# ----------------------------------------------------------
# Subplot 1: Daily Sales (Level 1 - Most Detailed)
# ----------------------------------------------------------
axes[0, 0].plot(
    df['Date'], 
    df['Daily_Sales_kUSD'], 
    color='#1abc9c', # Turquoise
    alpha=0.9, 
    linewidth=1.5
)

axes[0, 0].annotate(
    "High Variance (Noise):\nAt the daily level, sales fluctuate wildly.\nIt's hard to see the big picture.",
    xy=(df['Date'].iloc[60], 280), xytext=(df['Date'].iloc[100], 310),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[0, 0].set_title('Figure 1A: Level 1 - Daily Sales (Finest Granularity)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[0, 0].set_ylabel('Sales Volume (kUSD)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0, 0].grid(True, linestyle='--', alpha=0.5)
axes[0, 0].tick_params(axis='x', rotation=15, labelsize=11)
axes[0, 0].set_xticklabels(axes[0, 0].get_xticklabels(), fontweight='bold') # FIX: Bold formatting

# ----------------------------------------------------------
# Subplot 2: Monthly Sales (Level 2 - Roll-Up)
# ----------------------------------------------------------
axes[0, 1].bar(
    monthly_sales['Month'].astype(str),
    monthly_sales['Monthly_Sales_kUSD'],
    color='#5DADE2', # Light/Medium Blue
    edgecolor='#2c3e50',
    linewidth=1.5
)

axes[0, 1].set_title('Figure 1B: Level 2 - Monthly Sales (Roll-Up from Day)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[0, 1].set_ylabel('Aggregated Sales (kUSD)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[0, 1].tick_params(axis='x', rotation=45, labelsize=11)
axes[0, 1].set_xticklabels(axes[0, 1].get_xticklabels(), fontweight='bold') # FIX: Bold formatting
axes[0, 1].grid(axis='y', linestyle='--', alpha=0.5)

# ----------------------------------------------------------
# Subplot 3: Quarterly Sales (Level 3 - Roll-Up)
# ----------------------------------------------------------
axes[1, 0].bar(
    quarterly_sales['Quarter'].astype(str),
    quarterly_sales['Quarterly_Sales_kUSD'],
    color='#2980b9', # Classic Blue
    edgecolor='#2c3e50',
    width=0.5,
    linewidth=1.5
)

axes[1, 0].annotate(
    "Pattern Emergence:\nBy rolling up to Quarters, we can easily\ncompare macro-level performance across seasons.",
    xy=(2.0, 15500), xytext=(0.5, 17000),
    fontsize=11, fontweight='bold', color='#2980b9',
    arrowprops=dict(arrowstyle="->", color="#2980b9", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#ecf0f1", ec="#bdc3c7", lw=1.5)
)

axes[1, 0].set_title('Figure 1C: Level 3 - Quarterly Sales (Roll-Up from Month)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[1, 0].set_ylabel('Aggregated Sales (kUSD)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1, 0].tick_params(axis='x', labelsize=12) # FIX: Removed labelweight
axes[1, 0].set_xticklabels(axes[1, 0].get_xticklabels(), fontweight='bold') # FIX: Applied fontweight properly
axes[1, 0].grid(axis='y', linestyle='--', alpha=0.5)

for p in axes[1, 0].patches:
    axes[1, 0].annotate(
        f"{int(p.get_height())}k", 
        (p.get_x() + p.get_width() / 2., p.get_height()), 
        ha='center', va='center', xytext=(0, 9), textcoords='offset points', 
        fontweight='bold', fontsize=12, color='#2c3e50'
    )


# ----------------------------------------------------------
# Subplot 4: Yearly Sales (Level 4 - Highest Summary)
# ----------------------------------------------------------
axes[1, 1].bar(
    yearly_sales['Year'].astype(str),
    yearly_sales['Yearly_Sales_kUSD'],
    color='#2c3e50', # Navy Blue
    edgecolor='black',
    width=0.25,
    linewidth=2.0
)

axes[1, 1].set_title('Figure 1D: Level 4 - Yearly Sales (Highest Roll-Up)', fontdict={'fontsize': 14, 'fontweight': 'bold'}, pad=15)
axes[1, 1].set_ylabel('Total Annual Sales (kUSD)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
axes[1, 1].tick_params(axis='x', labelsize=13) # FIX: Removed labelweight
axes[1, 1].set_xticklabels(axes[1, 1].get_xticklabels(), fontweight='bold') # FIX: Applied fontweight properly
axes[1, 1].grid(axis='y', linestyle='--', alpha=0.5)

for p in axes[1, 1].patches:
    axes[1, 1].annotate(
        f"${int(p.get_height()):,}k", 
        (p.get_x() + p.get_width() / 2., p.get_height()), 
        ha='center', va='center', xytext=(0, 12), textcoords='offset points', 
        fontweight='bold', fontsize=14, color='#2c3e50'
    )


# Clear top and right spines for a clean academic aesthetic
for ax in axes.flat:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.suptitle('OLAP Concept Hierarchy: Time Dimension Data Aggregation', fontsize=17, fontweight='bold', y=1.02)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه کاربردی (Strategic Implication):</b><br>"
    r"سلسله‌مراتب مفاهیم (Concept Hierarchy) امکان <b>ناوبریِ چندسطحی</b> در داده‌های کلان را فراهم می‌سازد. همان‌طور که نمودارها از بالا-چپ به پایین-راست نشان می‌دهند، "
    r"با اجرای عملیات <b>Roll-Up</b>، نویزهای پرنوسانِ روزانه فیلتر شده و الگوهایِ کلان (نظیر عملکرد یکپارچه فصلی) نمایان می‌شوند.<br>"
    r"این معماری در سیستم‌های هوش تجاری (BI) تضمین می‌کند که هر کاربر سازمانی، گزارش‌ها را دقیقاً در <i>سطحِ جزئیاتِ (Granularity)</i> مورد نیاز خود دریافت نماید: "
    r"مدیرعامل به گرافِ سالانه (سرمه‌ای رنگ) نگاه می‌کند، در حالی که مدیر عملیاتی روی گرافِ روزانه (فیروزه‌ای رنگ) <b>Drill-Down</b> می‌کند تا علت اُفت فروش در یک روزِ خاص را بررسی کند."
)
display_academic_text(insight_conclusion, "insight")
```

    <li><b>نمودار پایین-چپ (Bar Chart بنفش):</b> Roll-Up به ۴ فصل؛ سطح خلاصه‌تر برای تحلیل فصلی.</li>
    <li><b>نمودار پایین-راست (Bar Chart قرمز):</b> Roll-Up به ۱ عدد سالانه؛ بالاترین سطح خلاصه‌سازی.</li>
  </ul>
