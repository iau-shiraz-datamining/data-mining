  <span style="font-size: 20pt;"></span> 3.4.1 Terminology of data cube computation

<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: واژه‌شناسی و اصطلاحات کلیدی مکعب داده در «آمازون» 📚
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    تیم‌های هوش تجاری و تحلیل داده در آمازون برای بررسی دقیق عملکرد فروش، اصطلاحات استاندارد مکعب داده را به کار می‌برند. این مفاهیم نحوه تغییر گرانولاریته و زاویه دید روی داده‌های چندبعدی را توصیف می‌کنند:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. برش (Slice): ثابت نگه داشتن یک بعد خاص برای رسیدن به یک تصویر دوبعدی از مکعب.
    مثال: فیلتر کردن کل مکعب فروش فقط برای سال ۲۰۲۳.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. تاس‌زدن (Dice): اعمال فیلتر روی دو یا چند بعد به صورت هم‌زمان برای برش یک زیرمکعب کوچک‌تر.
    مثال: فیلتر کردن فروش سال ۲۰۲۳ در منطقه آمریکا و فقط برای گروه کالای کالای دیجیتال.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. فشردن (Roll-up): صعود در سلسله‌مراتب ابعاد و تجمیع داده‌ها برای کاهش جزئیات و رسیدن به خلاصه مدیوم.
    مثال: جمع زدن داده‌های ماهانه برای رسیدن به فروش سالانه.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. جزئی‌سازی (Drill-down): نزول در سلسله‌مراتب ابعاد و شکستن داده‌های خلاصه‌شده به اجزای دقیق‌تر.
    مثال: شکستن فروش یک سال به ۴ فصل یا ۱۲ ماه.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۵. چرخش (Pivot): تغییر چیدمان و جای‌گذاری سطرها و ستون‌های جدول برای مشاهده داده‌ها از یک دیدگاه جدید.
    مثال: جابه‌جا کردن بعد منطقه از سطر به ستون جهت مقایسه آسان‌تر.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: تسلط بر این اصطلاحات پایه، تحلیل داده‌های چندبعدی را در سیستم‌های OLAP دقیق و استاندارد می‌سازد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: واژه‌شناسی کاربردی عملیات تعاملی روی ابعاد مکعب داده
    </div>
  </p>

</div>

```python
display_academic_text(
    "واژه‌شناسی و اصطلاحات کلیدی مکعب داده (Data Cube) در «آمازون» 📚", 
    "main_title"
)

display_academic_text(
    "تیم‌های هوش تجاری و تحلیل داده در آمازون برای بررسی دقیق عملکرد فروش، اصطلاحات استانداردِ مکعب داده را به کار می‌برند. "
    "این مفاهیم، نحوهٔ تغییر گرانولاریته (سطح جزئیات) و زاویهٔ دید روی داده‌های چندبُعدی را توصیف می‌کنند:", 
    "paragraph"
)

display_academic_text(
    "<b>۱. بُرش (Slice):</b> ثابت نگه داشتن <i>یک بُعد خاص</i> برای رسیدن به تصویر دوبُعدی (مثال: فیلتر کردن کُل فروش فقط برای سال ۲۰۲۳).<br>"
    "<b>۲. تاس‌زدن (Dice):</b> اعمال فیلتر روی <i>دو یا چند بُعد به‌صورت هم‌زمان</i> برای ساخت یک زیرمکعب (مثال: فروش سال ۲۰۲۳، منطقه آمریکا، گروه Tech).<br>"
    "<b>۳. فشردن (Roll-up):</b> تجمیع داده‌ها برای کاهش جزئیات و رسیدن به خلاصهٔ مدیریتی (مثال: جمع زدن داده‌های فصلی به سالانه).<br>"
    "<b>۴. جزئی‌سازی (Drill-down):</b> شکستن داده‌های خلاصه‌شده به اجزای دقیق‌تر (مثال: شکستن فروش یک سال به ۴ فصل).<br>"
    "<b>۵. چرخش (Pivot):</b> تغییر چیدمان و جای‌گذاری سطرها و ستون‌های جدول برای مشاهده داده‌ها از زاویه‌ای جدید.", 
    "paragraph"
)

# ==========================================================
# SECTION 2: DATA GENERATION AND DIMENSIONAL OPERATIONS
# ==========================================================
display_academic_text("مرحله نخست: تولید مکعب داده و شبیه‌سازی اصطلاحات پایه‌ای", "sub_title")

# Step 1: Generate Synthetic Amazon 3D Cube Dataset
years = [2025, 2026]
quarters = ['Q1', 'Q2', 'Q3', 'Q4']
regions = ['US', 'EU']
categories = ['Tech', 'Fashion']

records = []
for y in years:
    for q in quarters:
        for r in regions:
            for c in categories:
                sales = np.random.randint(150, 600)
                records.append({
                    'Year': y,
                    'Quarter': q,
                    'Region': r,
                    'Category': c,
                    'Sales_kUSD': sales,
                })

df_cube = pd.DataFrame(records)

# Step 2: Implement Terminology Operations

# 1. Slice: Fix exactly 1 dimension (Year = 2026) -> Flattens the cube
slice_data = df_cube[df_cube['Year'] == 2026]

# 2. Dice: Fix multiple dimensions (Year=2026, Region='US', Category='Tech') -> Extracts a sub-cube
dice_data = df_cube[
    (df_cube['Year'] == 2026) & 
    (df_cube['Region'] == 'US') & 
    (df_cube['Category'] == 'Tech')
]

# 3. Roll-up: Aggregate from Quarter to Year (Reduce dimensions)
rollup_data = df_cube.groupby(['Year', 'Category'])['Sales_kUSD'].sum().reset_index()

# 4. Drill-down: Show Quarterly breakdown for 2026 (Increase dimensions)
drilldown_data = df_cube[df_cube['Year'] == 2026].groupby(['Quarter', 'Category'])['Sales_kUSD'].sum().reset_index()

# 5. Pivot: Rotate layout (Category vs Region for 2026)
pivot_data = pd.pivot_table(
    slice_data,
    values='Sales_kUSD',
    index='Category',
    columns='Region',
    aggfunc='sum',
).reset_index() # Reset index for clean HTML rendering

# Render Data Samples securely
display_side_by_side_datasets(
    [df_cube.head(5).reset_index(drop=True), slice_data.head(5).reset_index(drop=True), pivot_data],
    [
        "پایگاه داده اصلی (Original 3D Cube)", 
        "عملیات Slice (ثابت کردن بُعد زمان = ۲۰۲۶)",
        "عملیات Pivot (چرخش محورهای Region و Category)"
    ]
)

# ==========================================================
# SECTION 3: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
display_academic_text("مرحله دوم: مصورسازی تحلیلی مفاهیم OLAP", "sub_title")

# Step 3: Visualization using Matplotlib
fig, axes = plt.subplots(2, 2, figsize=(18, 12), dpi=135)

# ----------------------------------------------------------
# Plot 1: Slice (2026 Sales across Categories)
# ----------------------------------------------------------
slice_summary = slice_data.groupby('Category')['Sales_kUSD'].sum().reset_index()

axes[0, 0].bar(
    slice_summary['Category'],
    slice_summary['Sales_kUSD'],
    color='#1abc9c', # Turquoise
    edgecolor='#2c3e50',
    linewidth=2.0,
    width=0.4,
)

axes[0, 0].annotate(
    "Slice Concept:\nBy fixing Year=2026, we sliced a 2D flat\nplane out of the 3D data cube.",
    xy=(1.0, 2500), xytext=(0.5, 3000),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[0, 0].set_title('Figure 1: Slice Operation (Fixed Dimension: Year = 2026)', fontsize=14, fontweight='bold', pad=15)
axes[0, 0].set_ylabel('Total Sales (kUSD)', fontsize=13, fontweight='bold')
axes[0, 0].grid(axis='y', linestyle='--', alpha=0.5)
axes[0, 0].tick_params(axis='x', labelsize=12)
axes[0, 0].set_xticklabels(slice_summary['Category'], fontweight='bold')

for p in axes[0, 0].patches:
    axes[0, 0].annotate(f"${int(p.get_height()):,}k", (p.get_x() + p.get_width() / 2., p.get_height()), ha='center', va='center', xytext=(0, -15), textcoords='offset points', fontweight='bold', fontsize=12, color='white')

# ----------------------------------------------------------
# Plot 2: Dice (Fixed Year = 2026, Region = US, Category = Tech)
# ----------------------------------------------------------
axes[0, 1].bar(
    dice_data['Quarter'].astype(str),
    dice_data['Sales_kUSD'],
    color='#2980b9', # Classic Blue
    edgecolor='#2c3e50',
    linewidth=2.0,
    width=0.4,
)

axes[0, 1].annotate(
    "Dice Concept:\nBy filtering Year, Region AND Category,\nwe isolated a tiny sub-cube for precise analysis.",
    xy=(2.0, 300), xytext=(0.5, 450),
    fontsize=11, fontweight='bold', color='#2980b9',
    arrowprops=dict(arrowstyle="->", color="#2980b9", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#ecf0f1", ec="#bdc3c7", lw=1.5)
)

axes[0, 1].set_title('Figure 2: Dice Operation (US Tech in 2026 by Quarter)', fontsize=14, fontweight='bold', pad=15)
axes[0, 1].set_ylabel('Sales (kUSD)', fontsize=13, fontweight='bold')
axes[0, 1].grid(axis='y', linestyle='--', alpha=0.5)
axes[0, 1].tick_params(axis='x', labelsize=12)
axes[0, 1].set_xticklabels(dice_data['Quarter'], fontweight='bold')

for p in axes[0, 1].patches:
    axes[0, 1].annotate(f"${int(p.get_height())}k", (p.get_x() + p.get_width() / 2., p.get_height()), ha='center', va='center', xytext=(0, 10), textcoords='offset points', fontweight='bold', fontsize=12, color='#2c3e50')

# ----------------------------------------------------------
# Plot 3: Roll-up (Yearly Aggregated Sales)
# ----------------------------------------------------------
pivot_rollup = rollup_data.pivot(index='Year', columns='Category', values='Sales_kUSD')
pivot_rollup.plot(
    kind='bar', 
    ax=axes[1, 0], 
    color=['#2c3e50', '#1abc9c'], # Navy, Turquoise
    edgecolor='#2c3e50',
    linewidth=2.0
)

axes[1, 0].annotate(
    "Roll-Up Concept:\nWe climbed up the Concept Hierarchy,\nhiding the Quarter noise to see the yearly trend.",
    xy=(0.0, 2000), xytext=(-0.35, 3500),
    fontsize=11, fontweight='bold', color='#2c3e50',
    arrowprops=dict(arrowstyle="->", color="#34495e", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#bdc3c7", lw=1.5)
)

axes[1, 0].set_title('Figure 3: Roll-Up Operation (Yearly Aggregated Macro-Trend)', fontsize=14, fontweight='bold', pad=15)
axes[1, 0].set_ylabel('Total Sales (kUSD)', fontsize=13, fontweight='bold')
axes[1, 0].set_xlabel('Macro Dimension: Year', fontsize=13, fontweight='bold')
axes[1, 0].tick_params(axis='x', rotation=0, labelsize=12)
axes[1, 0].set_xticklabels(axes[1, 0].get_xticklabels(), fontweight='bold')
axes[1, 0].grid(axis='y', linestyle='--', alpha=0.5)
axes[1, 0].legend(title='Category', frameon=True, fontsize=11)

# ----------------------------------------------------------
# Plot 4: Pivot (Rotated View of Category vs Region)
# ----------------------------------------------------------
pivot_data_plot = pivot_data.set_index('Category')
pivot_data_plot.plot(
    kind='barh', 
    ax=axes[1, 1], 
    color=['#5DADE2', '#2980b9'], # Light Blue, Classic Blue
    edgecolor='#2c3e50',
    linewidth=1.5
)

axes[1, 1].annotate(
    "Pivot Concept:\nSimply rotating the presentation axes.\nThe data doesn't change, only the viewing angle.",
    xy=(1500, 1.0), xytext=(1600, 0.4),
    fontsize=11, fontweight='bold', color='#2980b9',
    arrowprops=dict(arrowstyle="->", color="#2980b9", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#bdc3c7", lw=1.5)
)

axes[1, 1].set_title('Figure 4: Pivot Operation (Rotated Dimensional Layout)', fontsize=14, fontweight='bold', pad=15)
axes[1, 1].set_xlabel('Sales (kUSD)', fontsize=13, fontweight='bold')
axes[1, 1].set_ylabel('Dimension: Category', fontsize=13, fontweight='bold')
axes[1, 1].tick_params(axis='y', labelsize=12)
axes[1, 1].set_yticklabels(axes[1, 1].get_yticklabels(), fontweight='bold')
axes[1, 1].grid(axis='x', linestyle='--', alpha=0.5)
axes[1, 1].legend(title='Dimension: Region', loc='lower right', frameon=True, fontsize=11)

# Clear top and right spines for a clean academic aesthetic
for ax in axes.flat:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.suptitle('OLAP Dimensional Terminology & Operations Visualized', fontsize=17, fontweight='bold', y=1.02)

plt.tight_layout(pad=3.0)
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه کاربردی (Strategic Implication):</b><br>"
    r"تسلط بر این اصطلاحاتِ پایه، زیربنای مهندسیِ داده‌های چندبُعدی را شکل می‌دهد. استفاده درست از عملیات <b>Slice</b> و <b>Dice</b> "
    r"بارِ پردازشیِ کوئری‌ها را با حذفِ نویزهای ناخواسته و ایزوله‌سازیِ قطعاتِ مرتبط به حداقل می‌رساند.<br>"
    r"در عین حال، ترکیب هنرمندانهٔ <b>Roll-up</b> و <b>Drill-down</b> به تحلیل‌گران این قدرت را می‌بخشد که داشبوردهای تعاملی خلق کنند؛ "
    r"داشبوردهایی که کاربر در آن‌ها می‌تواند با عملیاتِ Roll-up به <i>عمقِ استراتژیک (Strategic Depth)</i> دست یابد، و با Drill-down به <i>جزئیاتِ عملیاتی (Operational Granularity)</i> نفوذ کند."
)
display_academic_text(insight_conclusion, "insight")
```
