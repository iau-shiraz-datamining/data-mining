    مثال عملی: راهبردهای سه‌گانه محاسبه و مدیریت مکعب داده در آمازون 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در سیستم‌های انبار داده بزرگ مانند آمازون (Amazon)، روزانه مگابایت‌ها و گیکابایت‌ها داده خام تراکنشی وارد سیستم می‌شود. محاسبه مکعب‌های داده چندبعدی روی چنین حجم عظیمی نیازمند راهبردهای بهینه‌سازی محاسباتی است. مهندسان داده برای افزایش سرعت محاسبه و پاسخ‌گویی به کوئری‌ها از ۳ راهبرد اصلی استفاده می‌کنند:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. تجمیع و پیش‌محاسبه (Aggregation and Precomputation):<br>
    به جای محاسبه آنی در زمان اجرای کوئری، حالت‌های تجمیعی پرکاربرد از قبل در فواصل زمانی خلوتی سیستم محاسبه و در قالب مکعب‌واره‌های خلاصه‌شده ذخیره می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. ایندکس‌گذاری و پارتیشن‌بندی (Indexing and Partitioning):<br>
    داده‌های عظیم بر اساس یک بعد اصلی (مثلاً سال یا منطقه) به بخش‌های کوچک‌تر به نام پارتیشن تقسیم می‌شوند و روی کلیدها ایندکس چندسطحی ساخته می‌شود. هنگام کوئری‌گیری، سیستم فقط پارتیشن مرتبط را می‌خواند و از پیمایش کل انبار داده جلوگیری می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. پردازش موازی (Parallel Processing):<br>
    وظیفه سنگین محاسبه مکعب داده به چند تکه کوچک‌تر (Chunks) تقسیم شده و به صورت هم‌زمان روی چند هسته پردازنده (CPU Cores) یا گره‌های محاسباتی اجرا می‌شود تا زمان محاسبه کل مکعب به شدت کاهش یابد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: ترکیب این ۳ راهبرد باعث می‌شود محاسبه مکعب داده‌های پتابایتی امکان‌پذیر شده و زمان پاسخ به داشبوردهای مدیریتی آنی گردد.<br><br>

      خروجی: بهینه‌سازی کامل محاسبه مکعب داده با ترکیب پیش‌محاسبه، پارتیشن‌بندی و پردازش موازی
    </div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import time 
from concurrent.futures import ThreadPoolExecutor 

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11
})

plt.rcParams['figure.dpi'] = 135

np.random.seed(42)
```

```python
display_academic_text(
    "راهبردهای سه‌گانه محاسبه و مدیریت مکعب داده (Data Cube) در آمازون 📦", 
    "main_title"
)

display_academic_text(
    "در سیستم‌های انبار داده بزرگ مانند آمازون، روزانه ترابایت‌ها دادهٔ تراکنشی وارد سیستم می‌شود. "
    "برای پاسخ‌گوییِ سریع به داشبوردهای مدیریتی روی این حجم عظیم از داده، موتورهای OLAP نمی‌توانند صرفاً جداول خام را اسکن کنند. "
    "مهندسان داده از <b>۳ راهبرد سخت‌افزاری و نرم‌افزاریِ</b> اصلی استفاده می‌کنند:", 
    "paragraph"
)

display_academic_text("۱. تجمیع و پیش‌محاسبه (Precomputation / Materialization):", "sub_title")
display_academic_text(
    "محاسبهٔ حالت‌های پُرکاربردِ گزارش‌گیری (مثل فروش سالانه هر منطقه) از قبل، و ذخیره‌سازی آن‌ها در <i>مکعب‌واره‌های خلاصه‌شده (Cuboids)</i> در حافظهٔ رَم.", 
    "paragraph"
)

display_academic_text("۲. پارتیشن‌بندی و ایندکس‌گذاری (Partitioning):", "sub_title")
display_academic_text(
    "تقسیم فیزیکیِ داده‌های عظیم روی هارد دیسک بر اساس یک بُعد اصلی (مثلاً <code>Region</code>). در این حالت، اگر مدیر، فروشِ منطقهٔ «آسیا» را بخواهد، "
    "سیستم اصلاً فایل‌هایِ مربوط به «اروپا» و «آمریکا» را نمی‌خواند (کاهش شدید I/O).", 
    "paragraph"
)

display_academic_text("۳. پردازش موازی (Parallel / Distributed Processing):", "sub_title")
display_academic_text(
    "وظیفهٔ سنگین محاسبهٔ یک مکعب داده به چند تکهٔ کوچک‌تر (Chunks) تقسیم شده و به‌طور هم‌زمان روی چند هستهٔ پردازنده (Multi-Core) یا چند سرورِ مجزا اجرا می‌شود.", 
    "paragraph"
)

# ==========================================================
# SECTION 2: CREATE SYNTHETIC AMAZON TRANSACTION DATASET
# ==========================================================
display_academic_text("مرحله نخست: تولید داده‌های حجیم خام", "sub_title")

# Increased transaction volume for realistic benchmarking
n_transactions = 500_000

regions = ['North America', 'Europe', 'Asia Pacific', 'Latin America']
categories = ['Electronics', 'Books', 'Fashion', 'Home']
years = [2024, 2025, 2026]

df_raw = pd.DataFrame({
    'Region': np.random.choice(regions, size=n_transactions),
    'Category': np.random.choice(categories, size=n_transactions),
    'Year': np.random.choice(years, size=n_transactions),
    'Sales_USD': np.random.uniform(10.0, 500.0, size=n_transactions)
})

display_side_by_side_datasets(
    [df_raw.head(5).reset_index(drop=True)],
    [f"مجموعه داده خام تراکنش‌های آمازون (تعداد: {n_transactions:,} رکورد پخش‌شده در حافظه)"]
)

# ==========================================================
# SECTION 3: IMPLEMENTING OPTIMIZATION STRATEGIES
# ==========================================================
display_academic_text("مرحله دوم: بنچمارک استراتژی‌های سه‌گانه و مقایسه با اسکن سنتی", "sub_title")

# ----------------------------------------------------------
# Baseline: Raw Table Scan (The naive approach)
# ----------------------------------------------------------
start_time = time.perf_counter()

raw_result = df_raw[
    (df_raw['Region'] == 'North America') &
    (df_raw['Category'] == 'Electronics') &
    (df_raw['Year'] == 2026)
]['Sales_USD'].sum()

time_raw_scan = (time.perf_counter() - start_time) * 1000

# ----------------------------------------------------------
# Strategy 1: Aggregation & Precomputation (Materialized Cube)
# ----------------------------------------------------------
# Building the cube happens in the background (ETL time)
precomputed_cube = df_raw.groupby(['Region', 'Category', 'Year'])['Sales_USD'].sum()

# Query Simulation on Precomputed Cuboid (User time)
start_time = time.perf_counter()

precomputed_result = precomputed_cube.loc[('North America', 'Electronics', 2026)]

time_precomputed_lookup = (time.perf_counter() - start_time) * 1000

# ----------------------------------------------------------
# Strategy 2: Partitioning (Pruning irrelevant data)
# ----------------------------------------------------------
# Dictionary as a simulation of physical partition folders on disk
df_partitioned = {region: group for region, group in df_raw.groupby('Region')}

# Query ONLY the relevant Partition
start_time = time.perf_counter()

north_america_partition = df_partitioned['North America'] # Pruning step
partition_result = north_america_partition[
    (north_america_partition['Category'] == 'Electronics') & 
    (north_america_partition['Year'] == 2026)
]['Sales_USD'].sum()

time_partition_query = (time.perf_counter() - start_time) * 1000

# ----------------------------------------------------------
# Strategy 3: Parallel Processing (Divide and Conquer)
# ----------------------------------------------------------
num_chunks = 4
chunk_size = len(df_raw) // num_chunks
chunks = [df_raw.iloc[i * chunk_size : (i + 1) * chunk_size].copy() for i in range(num_chunks)]

# Append remainder to the last chunk
if len(df_raw) > chunk_size * num_chunks:
    chunks[-1] = pd.concat([chunks[-1], df_raw.iloc[chunk_size * num_chunks:]])

def process_chunk(chunk_df):
    """Aggregate a single chunk."""
    return chunk_df.groupby(['Region', 'Category', 'Year'])['Sales_USD'].sum()

start_time = time.perf_counter()

with ThreadPoolExecutor(max_workers=num_chunks) as executor:
    results = list(executor.map(process_chunk, chunks))

# Combine results from all threads (Reduce step in MapReduce)
parallel_cube = pd.concat(results).groupby(level=['Region', 'Category', 'Year']).sum()
parallel_result = parallel_cube.loc[('North America', 'Electronics', 2026)]

time_parallel = (time.perf_counter() - start_time) * 1000

# Verify logic completeness (Sanity check)
is_valid = np.isclose(precomputed_result, raw_result) and np.isclose(partition_result, raw_result)

# Display Summarized Metrics
display_academic_text(
    f"<b>گزارش عملکرد سیستم (تأیید یکسان بودن خروجی محاسبات: {is_valid}):</b><br>"
    f"🔴 زمان اسکنِ کُل جدول خام (Raw Table Scan): <b>{time_raw_scan:.2f} میلی‌ثانیه</b><br>"
    f"🔵 زمان اجرای پردازش موازی (Parallel Compute): <b>{time_parallel:.2f} میلی‌ثانیه</b><br>"
    f"🟢 زمان استخراج از پارتیشن ایزوله (Partition Pruning): <b>{time_partition_query:.2f} میلی‌ثانیه</b><br>"
    f"🚀 زمان دسترسی به پیش‌محاسبه در رَم (O(1) Lookup): <b>{time_precomputed_lookup:.4f} میلی‌ثانیه</b>", 
    "insight"
)

# ==========================================================
# SECTION 4: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
strategies = ['Raw Scan\n(Inefficient)', 'Parallel Compute\n(MapReduce)', 'Partition Query\n(Pruning)', 'Precomputed\nLookup (O(1))']
execution_times = [time_raw_scan, time_parallel, time_partition_query, time_precomputed_lookup]

fig, axes = plt.subplots(1, 2, figsize=(17, 6.5), dpi=135)

# ----------------------------------------------------------
# Plot 1: Execution Time Comparison
# ----------------------------------------------------------
bars1 = axes[0].bar(
    strategies, 
    execution_times, 
    color=['#e74c3c', '#2c3e50', '#2980b9', '#1abc9c'], # Red, Navy, Classic Blue, Turquoise
    edgecolor='#2c3e50', 
    linewidth=2.0,
    width=0.55
)

axes[0].annotate(
    "Algorithmic Complexity:\nThe Raw Scan has $O(N)$ complexity, iterating\nover half a million rows. The Precomputed\nLookup uses an Index Dictionary to jump\nstraight to the answer in $O(1)$ time!",
    xy=(3.0, time_precomputed_lookup), xytext=(1.2, time_raw_scan * 0.7),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[0].set_title(f'Figure 1: Query Execution Time Comparison ({n_transactions:,} Rows)', fontsize=14, fontweight='bold', pad=15)
axes[0].set_ylabel('Execution Time (Milliseconds) - Lower is Better', fontsize=13, fontweight='bold')
axes[0].grid(axis='y', linestyle='--', alpha=0.5)
axes[0].set_xticklabels(strategies, fontweight='bold', fontsize=11)

for bar, value in zip(bars1, execution_times):
    axes[0].text(
        bar.get_x() + bar.get_width() / 2, 
        bar.get_height() + (max(execution_times) * 0.02), 
        f'{value:.2f} ms', 
        ha='center', va='bottom', fontweight='bold', fontsize=12, color='#2c3e50'
    )

# ----------------------------------------------------------
# Plot 2: Parallel Processing Workload Distribution
# ----------------------------------------------------------
chunk_sizes = [len(chunk) for chunk in chunks]
chunk_names = [f'CPU Core {i}' for i in range(1, num_chunks + 1)]

bars2 = axes[1].bar(
    chunk_names, 
    chunk_sizes, 
    color='#5DADE2', # Light Blue
    edgecolor='#2c3e50', 
    linewidth=2.0,
    width=0.5
)

axes[1].annotate(
    "MapReduce Concept:\nInstead of 1 CPU core analyzing 500,000 rows,\nwe mapped the data across 4 independent cores,\neach processing ~125,000 rows simultaneously.",
    xy=(1.5, chunk_sizes[0]), xytext=(0.1, chunk_sizes[0] * 1.15),
    fontsize=11, fontweight='bold', color='#2c3e50',
    arrowprops=dict(arrowstyle="->", color="#34495e", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#bdc3c7", lw=1.5)
)

axes[1].set_title('Figure 2: Parallel Processing Workload Distribution', fontsize=14, fontweight='bold', pad=15)
axes[1].set_ylabel('Number of Transactions Processed per Thread', fontsize=13, fontweight='bold')
axes[1].grid(axis='y', linestyle='--', alpha=0.5)
axes[1].set_ylim(0, max(chunk_sizes) * 1.35) # Headroom for annotations
axes[1].set_xticklabels(chunk_names, fontweight='bold', fontsize=12)

for bar, value in zip(bars2, chunk_sizes):
    axes[1].text(
        bar.get_x() + bar.get_width() / 2, 
        bar.get_height() + (max(chunk_sizes) * 0.02), 
        f'{value:,}', 
        ha='center', va='bottom', fontweight='bold', fontsize=12, color='#2c3e50'
    )

# Clear top and right spines for academic aesthetic
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 5: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک در مهندسی کلان‌داده (Strategic Implication):</b><br>"
    r"با ترکیبِ هوشمندانهٔ این ۳ راهبرد، محاسبهٔ مکعب‌هایِ داده‌هایِ پتابایتی امکان‌پذیر شده و زمانِ پاسخ به داشبوردهایِ مدیریتی کاملاً <b>آنی (Real-Time)</b> می‌گردد:<br>"
    r"۱. <b>پارتیشن‌بندی (Partitioning):</b> بارِ خواندنِ دیسک (I/O) را با فیلتر کردن و نادیده گرفتن بلوک‌های نامربوط کاهش داد.<br>"
    r"۲. <b>پردازش موازی (Parallel Processing):</b> زمان تجمیع و محاسبات سنگین را بین هسته‌های مختلف پخش کرد (قلبِ موتورهای Spark و Hadoop).<br>"
    r"۳. <b>پیش‌محاسبه (Materialization / Precomputation):</b> سرعت زمانِ پاسخ به کاربر نهایی را از مرتبهٔ $O(N)$ (اسکن کُل جدول) به یک <i>جستجویِ مستقیمِ حافظه‌ای (Lookup) از مرتبهٔ $O(1)$</i> ارتقا داد!"
)
display_academic_text(insight_conclusion, "insight")
```
