  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: مقایسه کارایی پایگاه داده ستونی ClickHouse و BigQuery در تحلیل‌های OLAP 📊
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در سیستم‌های پردازش آنلاین تراکنش (OLTP) مانند دیتابیس ثبت سفارشات، داده‌ها به‌صورت سطری (Row-oriented) ذخیره می‌شوند؛ یعنی تمام مشخصات یک کاربر (شناسه، نام، آدرس، ایمیل و مبلغ خرید) پشت سر هم روی دیسک قرار می‌گیرند. این ساختار برای ثبت یک سفارش جدید فوق‌العاده است. اما در تحلیل‌های بزرگ انبار داده (OLAP)، ذخیره‌سازی سطری با یک چالش بزرگ مواجه می‌شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    چالش تحلیل در دیتابیس سطری:<br>
    فرض کنید می‌خواهیم مجموع مبلغ کل خرید ۱۰۰ میلیون کاربر را محاسبه کنیم. در دیتابیس سطری، سیستم مجبور است تمام اطلاعات غیرضروری مثل نام، آدرس و ایمیل را هم از روی دیسک بخواند تا فقط به ستون مبلغ برسد! این یعنی خواندن ۹۰٪ داده‌های اضافی و کندی شدید سیستم.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    راه‌حل — پایگاه داده مبتنی بر ستون (Columnar Database):<br>
    در پایگاه‌های داده تحلیلی مانند ClickHouse، Google BigQuery یا Amazon Redshift، تمام مقادیر یک ستون کنار هم روی دیسک ذخیره می‌شوند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مزایای کلیدی معماری ستونی:<br>
    ۱. خواندن هوشمند دیسک (Selective Read): برای محاسبه مجموع خریه‌ها، سیستم فقط فایل مربوط به ستون مبلغ خرید را می‌خواند و بقیه ستون‌ها اصلاً بارگذاری نمی‌شوند.<br>
    ۲. نرخ فشرده‌سازی فوق‌العاده بالا: چون تمام داده‌های یک ستون از یک جنس هستند (مثلاً همه عدد یا همه تاریخ)، الگوریتم‌های فشرده‌سازی حجم ذخیره‌سازی را تا ۱۰ برابر کاهش می‌دهند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: معماری ستونی سرعت اجرای کوئری‌های تحلیلی و تجمیعی روی حجم‌های پتابایتی داده را تا ۱۰۰ برابر افزایش می‌دهد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: کاهش حجم داده‌های خوانده‌شده از دیسک و افزایش چشمگیر سرعت محاسبات OLAP
    </div>
  </p>

</div>

```python
display_academic_text(
    "تفاوت معماری: چرا دیتابیس‌های ستونی (مثل ClickHouse) در تحلیل‌های OLAP پادشاهی می‌کنند؟ 🚀", 
    "main_title"
)

display_academic_text(
    "در سیستم‌های پردازش آنلاین تراکنش (OLTP) مانند دیتابیسِ ثبت سفارشات یک فروشگاه، داده‌ها به‌صورت <b>سطری (Row-oriented)</b> روی هارد دیسک ذخیره می‌شوند؛ "
    "یعنی تمامِ مشخصاتِ یک کاربر (آیدی، سن، ایمیل، خرید) پشت سر هم قرار می‌گیرند. این برای خواندن و نوشتنِ سریعِ «یک تک‌مشتری» عالی است، "
    "اما در تحلیل‌های کُلانِ انبار داده (OLAP)، ذخیره‌سازی سطری با چالشی بزرگ مواجه می‌شود.", 
    "paragraph"
)

display_academic_text("چالش مهلک در دیتابیس سطری (I/O Bottleneck):", "sub_title")
display_academic_text(
    "فرض کنید می‌خواهیم مجموع مبلغ کُل خریدِ ۱۰۰ میلیون کاربر را محاسبه کنیم. در دیتابیس سطری، سیستم مجبور است تمام اطلاعات غیرضروری مثل "
    "<i>نام، آدرس و ایمیل</i> را هم از روی دیسک به حافظهٔ RAM بیاورد تا فقط به ستون «مبلغ» برسد! این یعنی خواندن ۸۰٪ داده‌های اضافی و کُندیِ شدید.", 
    "paragraph"
)

display_academic_text("راه‌حل جادویی: پایگاه داده مبتنی بر ستون (Column-oriented):", "sub_title")
display_academic_text(
    "در پایگاه‌های داده تحلیلی مانند ClickHouse ،Google BigQuery یا Amazon Redshift، تمام مقادیرِ یک ستون در فایل‌های مجزا و کنار هم روی دیسک ذخیره می‌شوند.<br>"
    "۱. <b>خواندن هوشمند دیسک (Selective Read):</b> سیستم فقط و فقط فایلِ مربوط به ستونِ «مبلغ خرید» را می‌خواند (I/O به شدت کاهش می‌یابد).<br>"
    "۲. <b>فشرده‌سازی و برداری‌سازی (Vectorization):</b> چون تمام داده‌های یک ستون از یک جنس (مثلاً عدد) هستند، CPU می‌تواند از دستوراتِ برداری (SIMD) استفاده کرده و داده‌ها را تا ۱۰ برابر بیشتر فشرده کند.", 
    "paragraph"
)

# ==========================================================
# SECTION 2: DATA GENERATION AND ARCHITECTURAL MODELING
# ==========================================================
# Increase size to 1,000,000 to make the benchmark more realistic
n_rows = 1000000

# Generating Data Columns
ids = np.arange(n_rows)
ages = np.random.randint(18, 70, size=n_rows)
cities = np.array([f'City_{i}' for i in np.random.randint(1, 20, size=n_rows)])
emails = np.array([f'user_{i}@domain.com' for i in range(n_rows)])
purchases = np.random.uniform(10.0, 500.0, size=n_rows)

# Step 2: Represent Data in Row-Oriented Layout (List of Tuples / Rows)
# Memory structure: [(id1, age1, city1, email1, purchase1), (id2, age2...), ...]
row_oriented_db = list(zip(ids, ages, cities, emails, purchases))

# Step 3: Represent Data in Column-Oriented Layout (Dict of Arrays / Columns)
# Memory structure: {'ID': [id1, id2...], 'Purchase': [purchase1, purchase2...], ...}
column_oriented_db = {
    'ID': ids,
    'Age': ages,
    'City': cities,
    'Email': emails,
    'Purchase_Amount': purchases,
}

# Display a preview of the dataset to the user securely
df_preview = pd.DataFrame(column_oriented_db).head(5).reset_index(drop=True)
display_side_by_side_datasets(
    [df_preview],
    [f"نمونه پایگاه داده شبیه‌سازی‌شده (شامل {n_rows:,} رکورد و ۵ ستون)"]
)

# ==========================================================
# SECTION 3: BENCHMARKING QUERY PERFORMANCE
# ==========================================================
display_academic_text("اجرای کوئری تحلیلی: محاسبه کُل درآمد <code>SUM(Purchase_Amount)</code>", "sub_title")

# ----------------------------------------------------------
# Approach A: Row-Oriented Engine
# Must iterate through entire row tuples and unpack everything just to reach the 5th element
# ----------------------------------------------------------
start_time = time.time()
row_sum = 0.0
elements_scanned_row = 0

# Simulating CPU loading the full row into cache
for row in row_oriented_db:
    row_sum += row[4]            # Access 5th element of every row
    elements_scanned_row += 5    # The CPU had to read all 5 elements from memory

row_time = (time.time() - start_time) * 1000  # Convert to ms

# ----------------------------------------------------------
# Approach B: Column-Oriented Engine
# Direct vector access to Purchase_Amount array only (SIMD Processing)
# ----------------------------------------------------------
start_time = time.time()
col_sum = np.sum(column_oriented_db['Purchase_Amount']) # NumPy uses C-level vectorization
elements_scanned_col = len(column_oriented_db['Purchase_Amount'])

col_time = (time.time() - start_time) * 1000  # Convert to ms

speedup_factor = row_time / col_time

display_academic_text(
    f"<b>نتایج بنچمارک برای {n_rows:,} رکورد:</b><br>"
    f"🔴 موتور سطری (سنتی): <b>{row_time:.2f} میلی‌ثانیه</b> | تعداد عناصر خوانده‌شده: {elements_scanned_row:,} واحد<br>"
    f"🟢 موتور ستونی (تحلیلی): <b>{col_time:.2f} میلی‌ثانیه</b> | تعداد عناصر خوانده‌شده: {elements_scanned_col:,} واحد<br>"
    f"🚀 <b>بهبود عملکرد:</b> موتور ستونی <b>{speedup_factor:.1f} برابر سریع‌تر</b> عمل کرد!", 
    "insight"
)

# ==========================================================
# SECTION 4: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 6.5), dpi=135)

engines = ['Row-Oriented DB\n(e.g. MySQL/PostgreSQL)', 'Column-Oriented DB\n(e.g. ClickHouse/BigQuery)']

# ----------------------------------------------------------
# Subplot 1: Memory Elements Scanned Comparison
# ----------------------------------------------------------
scanned_counts = [elements_scanned_row, elements_scanned_col]

bars1 = axes[0].bar(
    engines, 
    scanned_counts, 
    color=['#e74c3c', '#1abc9c'], # Red for inefficient, Turquoise for efficient
    edgecolor='#2c3e50', 
    linewidth=2.0,
    width=0.5
)

axes[0].annotate(
    "I/O Waste:\nThe Row engine had to scan 5 million elements\n(loading Names, Cities, etc.) just to sum 1 column.\nThe Column engine strictly scanned 1 million.",
    xy=(0.0, elements_scanned_row/2), xytext=(0.5, elements_scanned_row*0.7),
    fontsize=11, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title(f'Figure 1: Disk/Memory Elements Scanned ({n_rows:,} Rows)', fontsize=14, fontweight='bold', pad=15)
axes[0].set_ylabel('Number of Memory Elements Read (Lower is Better)', fontsize=13, fontweight='bold')
axes[0].grid(axis='y', linestyle='--', alpha=0.5)
axes[0].set_xticklabels(engines, fontweight='bold', fontsize=12)

# Format y-axis to show millions cleanly
def millions_formatter(x, pos):
    return f'{x / 1e6:.1f}M'
axes[0].yaxis.set_major_formatter(plt.FuncFormatter(millions_formatter))

for bar in bars1:
    yval = bar.get_height()
    axes[0].text(
        bar.get_x() + bar.get_width() / 2, 
        yval + (max(scanned_counts) * 0.02), 
        f'{yval/1e6:.1f}M Units', 
        ha='center', fontweight='bold', fontsize=13, color='#2c3e50'
    )

# ----------------------------------------------------------
# Subplot 2: Query Execution Time Comparison
# ----------------------------------------------------------
exec_times = [row_time, col_time]

bars2 = axes[1].bar(
    engines, 
    exec_times, 
    color=['#e74c3c', '#1abc9c'], 
    edgecolor='#2c3e50', 
    linewidth=2.0,
    width=0.5
)

axes[1].annotate(
    "Vectorization (SIMD):\nBecause data in a column is contiguous in RAM,\nthe CPU can process chunks of numbers simultaneously,\nresulting in a massive speedup.",
    xy=(1.0, col_time*2), xytext=(1.0, row_time*0.5),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

axes[1].set_title('Figure 2: Query Execution Time Comparison', fontsize=14, fontweight='bold', pad=15)
axes[1].set_ylabel('Execution Time (Milliseconds) - Lower is Better', fontsize=13, fontweight='bold')
axes[1].grid(axis='y', linestyle='--', alpha=0.5)
axes[1].set_xticklabels(engines, fontweight='bold', fontsize=12)

for bar in bars2:
    yval = bar.get_height()
    axes[1].text(
        bar.get_x() + bar.get_width() / 2, 
        yval + (max(exec_times) * 0.02), 
        f'{yval:.1f} ms', 
        ha='center', fontweight='bold', fontsize=13, color='#2c3e50'
    )

# Clear top and right spines for a clean academic aesthetic
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 5: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (Strategic Implication):</b><br>"
    r"معماری ستونی (Columnar Architecture)، با حذفِ کاملِ عملیات خواندن داده‌های غیرضروری از روی دیسک (کاهش چشمگیر I/O bottleneck) و بهره‌گیری از "
    r"<b>پردازش برداری (Vectorized Execution / SIMD)</b> در CPU، سرعت اجرای کوئری‌های تحلیلی و تجمیعی (نظیر SUM و AVG) روی حجم‌های پتابایتی داده را به‌طرز دیوانه‌واری افزایش می‌دهد.<br>"
    r"این معماری، قلبِ تپندهٔ پلتفرم‌های هوش تجاریِ ابریِ مدرن نظیر <b>Google BigQuery</b>، <b>Amazon Redshift</b> و <b>Snowflake</b> محسوب می‌گردد و دلیلِ اصلیِ "
    r"سرعتِ بالای داشبوردها در مواجهه با میلیاردها رکورد تراکنش است."
)
display_academic_text(insight_conclusion, "insight")
```
