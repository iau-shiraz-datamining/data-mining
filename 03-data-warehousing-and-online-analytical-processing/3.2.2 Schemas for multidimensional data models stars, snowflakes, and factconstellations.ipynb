  <span style="font-size: 20pt;"></span> 3.2.2 Schemas for multidimensional data models: stars, snowflakes, and fact constellations
</div>

<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: طراحی شماهای انبار داده در «آمازون» (Amazon Data Warehouse) 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    تیم مهندسی داده «آمازون» (Amazon) برای ذخیره‌سازی داده‌های تحلیل هوش تجاری (BI)، جداول را به دو دسته اصلی تقسیم می‌کند: جداول واقعیت (Fact Tables) که حاوی اعداد و معیارهای قابل جمع بستن (مثل مبلغ فروش و هزینه پستی) هستند، و جداول ابعاد (Dimension Tables) که متن‌ها و ویژگی‌های توصیفی (مثل نام مشتری و دسته‌بندی کالا) را نگه می‌دارند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۱. شمای ستاره‌ای (Star Schema) ✨ — ساده‌ترین و سریع‌ترین ساختار:</b><br>
    یک جدول واقعیت مرکزی (<code>Fact_Sales</code>) مستقیماً به جداول ابعاد غیرنرمال‌شده (<code>Dim_Customer</code>, <code>Dim_Product</code>, <code>Dim_Time</code>) وصل است. این ساختار شبیه ستاره است، سرعت کوئری‌گیری بالایی دارد و نیاز به JOINهای پیچیده ندارد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۲. شمای برف‌دانه (Snowflake Schema) ❄️ — ساختار نرمال‌شده بدون همبستگی تکراری:</b><br>
    در این شما، جداول ابعاد نرمال‌سازی می‌شوند. برای مثال جدول <code>Dim_Product</code> به زیر‌ابعاد <code>Dim_Subcategory</code> و <code>Dim_Category</code> شکسته می‌شود. فضای ذخیره‌سازی کمتری مصرف می‌شود اما کوئری‌ها نیازمند JOINهای متوالی هستند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    <b>۳. شمای صورت فلکی (Fact Constellation Schema) 🌠 — چند جدول واقعیت با ابعاد مشترک:</b><br>
    در سازمان‌های بزرگ مثل آمازون، چندین جدول واقعیت وجود دارد (مثلاً <code>Fact_Sales</code> برای فروش و <code>Fact_Shipping</code> برای هزینه‌های ارسال). این جداول واقعیت از ابعاد هم‌شکل و مشترک (Conformed Dimensions) مثل <code>Dim_Time</code> و <code>Dim_Customer</code> استفاده می‌کنند تا امکان تحلیل‌های متقاطع فراهم شود.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    <b>نتیجه:</b> انتخاب شما بستگی به اولویت سیستم دارد؛ شمای ستاره‌ای برای سرعت کوئری، برف‌دانه برای بهینه‌سازی حافظه، و صورت فلکی برای پوشش کل فرآیندهای تجاری سازمان به‌کار می‌رود.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: سازمان‌دهی منطقی جداول واقعیت و ابعاد برای گزارش‌گیری‌های پیچیده OLAP
    </div>
  </p>

</div>

```python
display_academic_text("۱. پیاده‌سازی شمای ستاره‌ای (Star Schema)", "sub_title")

# Dimension Tables (Flat & Denormalized)
dim_customer = pd.DataFrame({
    'Customer_Key': [1, 2],
    'Customer_Name': ['Alice', 'Bob'],
    'City': ['Seattle', 'New York'],
})

dim_product_star = pd.DataFrame({
    'Product_Key': [101, 102],
    'Product_Name': ['Echo Dot', 'Kindle Paperwhite'],
    'Category': ['Electronics', 'Electronics'],  # Denormalized (Repeated strings)
})

# Central Fact Table
fact_sales = pd.DataFrame({
    'Sales_ID': [10001, 10002],
    'Customer_Key': [1, 2],
    'Product_Key': [101, 102],
    'Time_Key': [20261001, 20261002],
    'Quantity_Sold': [2, 1],
    'Revenue_USD': [99.98, 139.99],
})

# Perform Single-level Star Schema Joins
star_query = fact_sales.merge(dim_customer, on='Customer_Key').merge(dim_product_star, on='Product_Key')

display_side_by_side_datasets(
    [star_query[['Sales_ID', 'Customer_Name', 'Product_Name', 'Category', 'Revenue_USD']].reset_index(drop=True)],
    ["خروجی شمای ستاره‌ای (Star): یک Fact متصل به ابعاد مسطح (سریع‌ترین نوع جستجو)"]
)

# ------------------------------------------------------------------------------
# 2. SIMULATING SNOWFLAKE SCHEMA ❄️
# ------------------------------------------------------------------------------
display_academic_text("۲. پیاده‌سازی شمای برف‌دانه (Snowflake Schema)", "sub_title")

# Normalized Dimensions (Sub-dimensions hierarchy to save storage)
dim_category_snow = pd.DataFrame({
    'Category_Key': [50], 
    'Category_Name': ['Electronics']
})

dim_product_snow = pd.DataFrame({
    'Product_Key': [101, 102],
    'Product_Name': ['Echo Dot', 'Kindle Paperwhite'],
    'Category_Key': [50, 50],  # Foreign Key pointing to dim_category_snow instead of repeating strings
})

# Perform Multi-level Snowflake Schema Joins
snowflake_query = (
    fact_sales
    .merge(dim_product_snow, on='Product_Key')
    .merge(dim_category_snow, on='Category_Key')  # Additional Hop (Extra JOIN)
    .merge(dim_customer, on='Customer_Key')
)

display_side_by_side_datasets(
    [snowflake_query[['Sales_ID', 'Customer_Name', 'Product_Name', 'Category_Name', 'Revenue_USD']].reset_index(drop=True)],
    ["خروجی شمای برف‌دانه (Snowflake): نیازمند JOINهای متوالی به‌خاطر نرمال‌سازی ابعاد"]
)

# ------------------------------------------------------------------------------
# 3. SIMULATING FACT CONSTELLATION SCHEMA 🌠
# ------------------------------------------------------------------------------
display_academic_text("۳. پیاده‌سازی شمای صورت فلکی (Fact Constellation Schema)", "sub_title")

# Second Fact Table (Sharing Conformed Dimensions: Customer and Time)
fact_shipping = pd.DataFrame({
    'Shipping_ID': [9001, 9002],
    'Customer_Key': [1, 2],
    'Time_Key': [20261001, 20261002],
    'Shipping_Cost_USD': [5.50, 12.00],
    'Delivery_Days': [2, 1],
})

# Cross-Fact Analysis (Comparing Sales Revenue vs Shipping Costs per Customer)
sales_summary = fact_sales.groupby('Customer_Key')['Revenue_USD'].sum()
shipping_summary = fact_shipping.groupby('Customer_Key')['Shipping_Cost_USD'].sum()

fact_constellation_summary = pd.DataFrame({
    'Total_Revenue_USD': sales_summary,
    'Total_Shipping_Cost_USD': shipping_summary,
}).reset_index()

# Merge summaries with Conformed Dimension
constellation_query = fact_constellation_summary.merge(dim_customer, on='Customer_Key')

display_side_by_side_datasets(
    [constellation_query[['Customer_Name', 'City', 'Total_Revenue_USD', 'Total_Shipping_Cost_USD']].reset_index(drop=True)],
    ["خروجی صورت فلکی (Constellation): تحلیل متقاطع دو جدول Fact (فروش و ارسال) بر روی بُعد مشترکِ مشتری"]
)

# ==========================================================
# SECTION 3: VISUAL ANALYTICS (ACADEMIC BLUE)
# ==========================================================
display_academic_text(
    "مرحله دوم: مصورسازی تحلیلی - موازنه (Trade-offs) در معماری‌های مختلف", 
    "sub_title"
)

display_academic_text(
    "بوم گرافیکیِ زیر نشان می‌دهد که در طراحی پایگاه داده، هیچ «بهترینِ مطلقی» وجود ندارد. <b>شمای ستاره‌ای</b> بالاترین سرعت کوئری (به دلیل JOINهای کمتر) را ارائه می‌دهد اما فضای بیشتری اشغال می‌کند. "
    "<b>شمای برف‌دانه</b> با نرمال‌سازیِ دقیق، در مصرف هارد دیسک صرفه‌جویی می‌کند اما سرعت استخراج را قربانی می‌کند. <b>صورت فلکی</b> نیز پادشاهِ تحلیل‌هایِ پیچیده و فراسازمانی (Cross-Process) است.", 
    "paragraph"
)

# Data for visualization (Conceptual scoring out of 10)
schemas = ['Star Schema\n(Denormalized)', 'Snowflake Schema\n(Normalized)', 'Fact Constellation\n(Multi-Fact)']
metrics = {
    'Query Execution Speed (Fewer JOINs)': [9.5, 4.0, 7.0],
    'Storage Efficiency (No Redundancy)': [4.0, 9.5, 6.0],
    'Analytical Depth (Cross-Process)': [4.0, 4.0, 9.5]
}
df_metrics = pd.DataFrame(metrics, index=schemas).reset_index().rename(columns={'index': 'Architecture'})

# Melt for seaborn grouped bar chart
df_melted = df_metrics.melt(id_vars='Architecture', var_name='Metric', value_name='Score')

# Initialize canvas
plt.figure(figsize=(15, 6.5), dpi=135)

# Grouped bar chart with Academic Blue palette
sns.barplot(
    data=df_melted, 
    x='Architecture', 
    y='Score', 
    hue='Metric',
    palette=['#1abc9c', '#2980b9', '#2c3e50'], # Turquoise, Classic Blue, Navy
    edgecolor='#2c3e50',
    linewidth=1.5
)

# Intelligent Annotations
plt.annotate(
    "Speed vs Storage:\nStar Schema is extremely fast\nbecause it avoids cascading JOINs,\nbut it wastes storage space.",
    xy=(0.0, 9.5), xytext=(-0.4, 6.5),
    fontsize=11, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#e8f8f5", ec="#1abc9c", lw=1.5)
)

plt.annotate(
    "Enterprise Analytics:\nConstellation is the only schema that\nallows comparing multiple business\nprocesses (e.g., Sales vs Shipping).",
    xy=(2.2, 9.5), xytext=(1.8, 6.5),
    fontsize=11, fontweight='bold', color='#2c3e50',
    arrowprops=dict(arrowstyle="->", color="#34495e", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.4", fc="#ecf0f1", ec="#bdc3c7", lw=1.5)
)

# Formatting
plt.title('Figure 1: Architectural Trade-offs in Data Warehouse Modeling', fontdict={'fontsize': 15, 'fontweight': 'bold'}, pad=15)
plt.ylabel('Capability Score (0 to 10)', fontdict={'fontsize': 13, 'fontweight': 'bold'})
plt.xlabel('Data Warehouse Schema Topology', fontdict={'fontsize': 13, 'fontweight': 'bold'})
plt.xticks(fontweight='bold', fontsize=12)
plt.ylim(0, 11)
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Adjust Legend
plt.legend(title='Evaluation Metric', bbox_to_anchor=(1.01, 1), loc='upper left', frameon=True, fontsize=11, title_fontproperties={'weight':'bold'})

# Clear top and right spines
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# SECTION 4: STRATEGIC CONCLUSION
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه کاربردی (Strategic Implication):</b><br>"
    r"انتخاب معماری پایگاه داده، یک موازنه (Trade-off) استراتژیک است.<br>"
    r"<b>شمای ستاره‌ای (Star)</b> به دلیل سادگی مدل و کاهش بار پردازشی JOINها، سریع‌ترین گزینه برای تغذیهٔ داشبوردهای مدیریتی (BI) است. "
    r"درحالی‌که <b>شمای برف‌دانه (Snowflake)</b> با ایزوله‌سازی جداول، فضای ذخیره‌سازی را بهینه‌تر مصرف می‌کند و از خطای افزونگی داده جلوگیری می‌نماید. "
    r"در نهایت، <b>شمای صورت فلکی (Fact Constellation)</b> برای پوشش ۳۶۰ درجه کلِ فرآیندهای تجاری سازمان (مانند بررسی سودآوری با کسر هزینه‌های پستی از مبلغ فروش) "
    r"در پلتفرم‌های کلانی نظیر آمازون به‌کار می‌رود."
)
display_academic_text(insight_conclusion, "insight")
```

در این مثال:
<ul>
<li>ما با استفاده از دیتافریم‌های کتابخانه <b>pandas</b>، جداول واقعیت (Fact) و ابعاد (Dimension) انبار داده آمازون را در ۳ شما شبیه‌سازی کردیم:</li>
<li><b>۱. شمای ستاره‌ای (Star Schema):</b> جدول مرکزی <b>fact_sales</b> شامل کلیدهای خارجی و معیارهای عددی (فروش و تعداد) است که با یک مرحله دستور <b>merge()</b> مستقیماً به جداول غیرنرمال‌شده <b>dim_customer</b> و <b>dim_product_star</b> متصل شد.</li>
<li><b>۲. شمای برف‌دانه (Snowflake Schema):</b> بعد محصول نرمال‌سازی شد؛ جدول <b>dim_product_snow</b> با کلید خارجی <code>Category_Key</code> به جدول سطح بالاتر <b>dim_category_snow</b> لینک گردید و کوئری نهایی با دو سطح ادغام (JOIN) اجرا شد.</li>

<li><b>نتیجه خروجی:</b> کد نشان داد که چطور در شمای صورت فلکی می‌توان تحلیل‌های متقاطع انجام داد و کل درآمد فروش هر مشتری را در کنار هزینه‌های ارسال او ارزیابی کرد.</li>
