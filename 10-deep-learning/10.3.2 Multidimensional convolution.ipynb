  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: استخراج الگوهای فضازمانی در پردازش ابر نقاط حجمی ۳ بعدی در «وایمو» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در ناوگان خودروهای خودران <b>وایمو</b>، حسگرهای لیدار (LiDAR) و دوربین‌های پیرامونی در هر ثانیه میلیون‌ها نقطه سه‌بعدی از محیط اطراف را ثبت می‌کنند. داده‌های خام حسگرها به شبکه‌های حجمی سه‌بعدی شامل وکسل‌ها (Voxels) و فریم‌های متوالی زمانی تبدیل می‌شوند. اعمال کانولوشن دوبعدی سنتی بر روی تک‌تک فریم‌ها، پیوستگی زمانی و همبستگی ساختاری در بعد عمق را نادیده می‌گیرد و پیش‌بینی مسیر حرکت عابران پیاده و موانع پویا را با خطا روبه‌رو می‌کند. مهندسان داده وایمو با به‌کارگیری <b>کانولوشن چندبعدی (3D Convolution)</b> و لغزش هسته‌های مکعبی فیلتر در طول سه بعد فضایی و زمانی ($D \times H \times W$)، بردار سرعت و ساختار حجمی موانع را به صورت یکپارچه استخراج می‌کنند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    معادلات و فرمول‌های ریاضی عملگر کانولوشن گسسته سه‌بعدی و انتشار گرادیان:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. فرمول همبستگی متقاطع سه‌بعدی برای تانسور ورودی حجمی $V$ و هسته فیلتر مکعبی $K$ با ابعاد $(2P+1) \times (2M+1) \times (2N+1)$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$S(d, i, j) = (V * K)(d, i, j) = \sum_{p=-P}^{P} \sum_{m=-M}^{M} \sum_{n=-N}^{N} V(d+p, i+m, j+n) \cdot K(p, m, n)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. محاسبه ابعاد تانسور خروجی نقشه ویژگی حجمی ($D_{\text{out}} \times H_{\text{out}} \times W_{\text{out}}$) با ابعاد هسته $F$، گام $S$ و پدینگ $P$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$D_{\text{out}} = \left\lfloor \frac{D_{\text{in}} - F_d + 2P_d}{S_d} \right\rfloor + 1, \quad H_{\text{out}} = \left\lfloor \frac{H_{\text{in}} - F_h + 2P_h}{S_h} \right\rfloor + 1, \quad W_{\text{out}} = \left\lfloor \frac{W_{\text{in}} - F_w + 2P_w}{S_w} \right\rfloor + 1$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. نگاشت نقشه ویژگی سه‌بعدی در حضور چند کانال ورودی $C_{\text{in}}$، بایاس $b_k$ و تابع فعال‌سازی غیرخطی $\sigma$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$A_{d, i, j, k} = \sigma\left( \sum_{c=1}^{C_{\text{in}}} \sum_{p} \sum_{m} \sum_{n} V_{d+p, i+m, j+n, c} \cdot W_{p, m, n, c, k} + b_k \right)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. محاسبه گرادیان پس‌انتشار خطا نسبت به وزن‌های مکعبی هسته فیلتر:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\frac{\partial \mathcal{L}}{\partial K_{p, m, n}} = \sum_{d} \sum_{i} \sum_{j} \frac{\partial \mathcal{L}}{\partial S(d, i, j)} \cdot V(d+p, i+m, j+n)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: وایمو با استفاده از کانولوشن سه‌بعدی بر روی توالی وکسل‌ها، نرخ خطای تشخیص اشیاء متحرک در شرایط نوری پیچیده و مه‌گرفتگی را تا ۳۸٪ کاهش داده و ایمنی هدایت خودکار خودرو را تثبیت می‌نماید.<br><br>

    </div>
  </p>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    r"استخراج الگوهای فضازمانی در پردازش ابر نقاط حجمی با کانولوشن ۳‌بعدی در «وایمو» 🚗", 
    "main_title"
)

display_academic_text(
    r"داده‌های لیدار (LiDAR) در خودروهای خودران مجموعه‌ای از ابرنقاط (Point Clouds) هستند که به صورت بلوک‌های سه‌بعدیِ فیزیکی (وکسل/Voxel) "
    r"رندر می‌شوند. اگر ما این حجم سه‌بعدی را فِلت (Flatten) کنیم، ماشین درکی از «عمق» و «سرعت» نخواهد داشت! "
    r"راه‌حل، استفاده از <b>کانولوشن ۳‌بعدی (3D Convolution)</b> است. در این روش، یک هسته فیلترِ مکعبی ($3 \times 3 \times 3$) در فضایِ ۳بُعدی می‌لغزد "
    r"و به جای یافتن خطوط دوبعدی، <i>تغییرات مکعب‌ها در طول زمان</i> (وکتور سرعت) را استخراج می‌کند.", 
    "paragraph"
)

display_academic_text(r"معادلات پایه در عملگر کانولوشن گسسته سه‌بعدی (Spatiotemporal Filtering):", "sub_title")

display_academic_text(r"۱. فرمول همبستگی متقاطع سه‌بعدی (لغزش هسته مکعبی $K$ در طول تانسور حجمی $V$):", "paragraph")
# MathJax Safe Formatting
display_academic_text(r"$$ S(d, i, j) = (V * K)(d, i, j) = \sum_{p=-P}^{P} \sum_{m=-M}^{M} \sum_{n=-N}^{N} V(d+p, i+m, j+n) K(p, m, n) $$", "paragraph")

# ==============================================================================
# 2. WAYMO SPATIOTEMPORAL 3D VOXEL DATASET GENERATION
# ==============================================================================
seed_val = 42
np.random.seed(seed_val)

# Simulating 600 volumetric samples: Shape = (N, Depth/Time, Height, Width)
n_samples = 600
depth_dim = 6   # 6 temporal frames / depth slices
height_dim = 16 # 16 spatial rows
width_dim = 16  # 16 spatial columns

X_volumetric = np.zeros((n_samples, depth_dim, height_dim, width_dim), dtype=np.float32)
y_labels = np.zeros(n_samples, dtype=np.int32)

for i in range(n_samples):
    # Add baseline sensory noise (Random light Voxels simulating fog/dust)
    X_volumetric[i] = np.random.normal(loc=0.02, scale=0.05, size=(depth_dim, height_dim, width_dim))
    
    if i % 2 == 0:
        # Class 0: Static Object (e.g., Traffic Light) at fixed coordinates across ALL time slices
        center_r, center_c = np.random.randint(6, 10, size=2)
        for d in range(depth_dim):
            X_volumetric[i, d, center_r-1:center_r+2, center_c-1:center_c+2] += np.random.uniform(0.7, 0.9)
        y_labels[i] = 0
    else:
        # Class 1: Dynamic Object (Pedestrian/Car) moving diagonally across time slices
        start_r = np.random.randint(2, 5)
        start_c = np.random.randint(2, 5)
        for d in range(depth_dim):
            cur_r = int(start_r + d * 1.5)
            cur_c = int(start_c + d * 1.5)
            X_volumetric[i, d, cur_r-1:cur_r+2, cur_c-1:cur_c+2] += np.random.uniform(0.7, 0.9)
        y_labels[i] = 1

X_volumetric = np.clip(X_volumetric, 0.0, 1.0)

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X_volumetric, y_labels, test_size=0.25, random_state=seed_val, stratify=y_labels
)

# ==============================================================================
# 3. 3D MULTIDIMENSIONAL CONVOLUTION FORWARD PASS FROM SCRATCH
# ==============================================================================
def conv3d_from_scratch(volume_3d, kernel_3d, stride=(1, 1, 1), padding=1):
    """Executes a 3D volumetric convolution operation across depth(time), height, and width."""
    d_in, h_in, w_in = volume_3d.shape
    k_d, k_h, k_w = kernel_3d.shape
    s_d, s_h, s_w = stride

    # Apply zero-padding across spatial-temporal dimensions
    if padding > 0:
        padded_vol = np.pad(
            volume_3d,
            ((padding, padding), (padding, padding), (padding, padding)),
            mode='constant',
            constant_values=0.0
        )
    else:
        padded_vol = volume_3d

    # Calculate 3D output dimensions
    d_out = int(np.floor((padded_vol.shape[0] - k_d) / s_d) + 1)
    h_out = int(np.floor((padded_vol.shape[1] - k_h) / s_h) + 1)
    w_out = int(np.floor((padded_vol.shape[2] - k_w) / s_w) + 1)

    output_3d_map = np.zeros((d_out, h_out, w_out), dtype=np.float32)

    # 3D sliding volume dot product (The core computational heavy lifter)
    for z in range(d_out):
        for y in range(h_out):
            for x in range(w_out):
                z_start = z * s_d
                z_end = z_start + k_d
                y_start = y * s_h
                y_end = y_start + k_h
                x_start = x * s_w
                x_end = x_start + k_w

                receptive_subvolume = padded_vol[z_start:z_end, y_start:y_end, x_start:x_end]
                output_3d_map[z, y, x] = np.sum(receptive_subvolume * kernel_3d)

    return output_3d_map

# Define a 3x3x3 Spatiotemporal Velocity Kernel (Derivative across Time)
# This kernel subtracts the previous frame from the next frame, highlighting only MOVING voxels!
motion_detector_kernel_3d = np.zeros((3, 3, 3), dtype=np.float32)
motion_detector_kernel_3d[0] = -1.0 * np.array([[0, 1, 0], [1, 2, 1], [0, 1, 0]]) # Frame t-1 (Subtract)
motion_detector_kernel_3d[1] =  0.0 * np.ones((3, 3))                             # Frame t (Ignore)
motion_detector_kernel_3d[2] =  1.0 * np.array([[0, 1, 0], [1, 2, 1], [0, 1, 0]]) # Frame t+1 (Add)

# Apply from-scratch 3D convolution on a DYNAMIC sample (Class 1 - Pedestrian)
sample_dynamic_volume = X_train[1] # A moving object
# We process it through our custom 3D Kernel
fmap_3d_dynamic = conv3d_from_scratch(sample_dynamic_volume, motion_detector_kernel_3d, padding=1)

# ==============================================================================
# 4. CUSTOM "3D CNN" TRAINING ENGINE (NO TENSORFLOW)
# ==============================================================================
def extract_3d_features_batch(volume_batch):
    """Applies the 3D Motion kernel to the entire batch and flattens the result."""
    N = volume_batch.shape[0]
    # Output of conv3d is (6, 16, 16). Flattened = 1536 features
    features = np.zeros((N, 6 * 16 * 16))
    
    for idx in range(N):
        fmap = conv3d_from_scratch(volume_batch[idx], motion_detector_kernel_3d, padding=1)
        features[idx] = fmap.flatten()
        
    return features

# Pre-extract features for train and test sets
X_train_3d = extract_3d_features_batch(X_train)
X_test_3d = extract_3d_features_batch(X_test)

# Flatten raw volumes for the baseline "Blind" dense network
X_train_flat = X_train.reshape(X_train.shape[0], -1)
X_test_flat = X_test.reshape(X_test.shape[0], -1)

# Training Loop using Scikit-Learn
epochs_count = 25
classes_unique = np.unique(y_labels)

# Baseline Model: Trained on Raw Voxels (No 3D Spatiotemporal awareness)
mlp_baseline = MLPClassifier(
    hidden_layer_sizes=(64,), activation='relu', solver='adam',
    learning_rate_init=0.01, max_iter=1, warm_start=True, random_state=seed_val
)

# 3D Convolutional Model: Trained on Extracted 3D Spatiotemporal Features
mlp_3d_conv = MLPClassifier(
    hidden_layer_sizes=(32,), activation='relu', solver='adam',
    learning_rate_init=0.01, max_iter=1, warm_start=True, random_state=seed_val
)

acc_baseline_history = []
acc_3d_conv_history = []

for epoch in range(epochs_count):
    mlp_baseline.partial_fit(X_train_flat, y_train, classes=classes_unique)
    mlp_3d_conv.partial_fit(X_train_3d, y_train, classes=classes_unique)
    
    acc_baseline_history.append(accuracy_score(y_test, mlp_baseline.predict(X_test_flat)))
    acc_3d_conv_history.append(accuracy_score(y_test, mlp_3d_conv.predict(X_test_3d)))

final_baseline_acc = acc_baseline_history[-1] * 100
final_3d_conv_acc = acc_3d_conv_history[-1] * 100

# Benchmark Output
df_benchmark = pd.DataFrame([
    {"Architecture": "Dense Flattened Network", "Spatiotemporal Awareness": "Blind (Treats time as static pixels)", "Validation Accuracy": f"{final_baseline_acc:.2f}%"},
    {"Architecture": "Custom 3D-CNN (Convolution Filters)", "Spatiotemporal Awareness": "High (Extracts Motion Velocity)", "Validation Accuracy": f"{final_3d_conv_acc:.2f}%"}
])
display_side_by_side_datasets([df_benchmark], ["بنچمارک دقت: مقایسه شبکه کور با شبکه مجهز به کانولوشن سه‌بعدی (3D Spatiotemporal)"])

# ==============================================================================
# 5. TRUE 3D VISUALIZATIONS AND ANALYTICAL CHARTS (ACADEMIC BLUE)
# ==============================================================================
# Setup a figure with 3D projection for true Voxel rendering
fig = plt.figure(figsize=(18, 7), dpi=120)

# We use gridspec to manage the layout: Left (3D Raw), Center (3D Convolved), Right (2D Line Plot)
gs = fig.add_gridspec(1, 3, width_ratios=[1, 1, 1.5], wspace=0.2)

ax_raw_3d = fig.add_subplot(gs[0, 0], projection='3d')
ax_fmap_3d = fig.add_subplot(gs[0, 1], projection='3d')
ax_curve = fig.add_subplot(gs[0, 2])

def plot_voxels(ax, volumetric_data, title, color, alpha=0.6, threshold=0.4):
    """Renders a 3D NumPy array as actual physical voxels in a 3D grid."""
    # Create coordinates for the voxel grid
    z, y, x = np.indices((volumetric_data.shape[0]+1, volumetric_data.shape[1]+1, volumetric_data.shape[2]+1))
    
    # Define which voxels are "solid" based on an intensity threshold
    # (Removes noise and shows only the actual objects/edges)
    filled_voxels = np.abs(volumetric_data) > threshold
    
    # Render voxels
    ax.voxels(x, y, z, filled_voxels, facecolors=color, edgecolor='k', linewidth=0.3, alpha=alpha)
    
    # Formatting
    ax.set_title(title, fontsize=11, fontweight="bold", color="#2c3e50")
    ax.set_xlabel("X (Width)", fontsize=9)
    ax.set_ylabel("Y (Height)", fontsize=9)
    ax.set_zlabel("Z (Depth/Time Frame)", fontsize=9)
    # Set a good viewing angle
    ax.view_init(elev=20, azim=45)

#  CHART 1: Render Raw 3D Volumetric Input 
plot_voxels(
    ax_raw_3d, 
    sample_dynamic_volume, 
    "Raw Input Volume\n(Moving Object + Noise)", 
    color="#2980b9", 
    threshold=0.5 # Higher threshold to cut through baseline Gaussian noise
)

#  CHART 2: Render 3D Convolved Feature Map 
plot_voxels(
    ax_fmap_3d, 
    fmap_3d_dynamic, 
    "3D Convolution Output\n(Motion Signature Extracted!)", 
    color="#e74c3c", 
    threshold=1.5 # Only show high magnitude changes (Velocity)
)

#  CHART 3: 3D CNN Training and Validation Convergence 
epochs_axis = np.arange(1, epochs_count + 1)

def smooth(points, factor=0.5):
    smoothed = []
    for p in points:
        if smoothed:
            smoothed.append(smoothed[-1] * factor + p * (1 - factor))
        else:
            smoothed.append(p)
    return smoothed

ax_curve.plot(
    epochs_axis, smooth(acc_3d_conv_history),
    "-", color="#2c3e50", lw=3.5, label=f"3D Convolutional Network Acc: {final_3d_conv_acc:.1f}%"
)
ax_curve.plot(
    epochs_axis, smooth(acc_baseline_history),
    ":", color="#bdc3c7", lw=3.0, label=f"Baseline (Flattened) Acc: {final_baseline_acc:.1f}%"
)

ax_curve.annotate(
    f"Spatiotemporal 3D filters instantly\nisolate moving obstacles from noise,\nreaching {final_3d_conv_acc:.1f}% accuracy effortlessly.",
    xy=(epochs_count, smooth(acc_3d_conv_history)[-1]),
    xytext=(epochs_count - 12, smooth(acc_3d_conv_history)[-1] - 0.20),
    fontsize=10, fontweight="bold", color="#2c3e50",
    arrowprops=dict(arrowstyle="->", color="#34495e", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.3", fc="#ebf5fb", ec="#2980b9", lw=1.5)
)

ax_curve.set_title("Waymo Classifier: Convergence Metrics", fontsize=13, fontweight="bold", pad=12)
ax_curve.set_xlabel("Training Epochs (Forward + Backward Passes)", fontsize=12, fontweight="bold")
ax_curve.set_ylabel("Validation Categorical Accuracy", fontsize=12, fontweight="bold")
ax_curve.set_ylim(0.4, 1.05)
ax_curve.legend(loc="lower right", frameon=True, facecolor="#f8f9fa", edgecolor="#bdc3c7", fontsize=10)
ax_curve.grid(True, linestyle="--", alpha=0.6)

ax_curve.spines['top'].set_visible(False)
ax_curve.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 6: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (درک عمق و زمان با رندرهای ۳‌بعدی):</b> در گراف‌های سه‌بعدیِ بالا (سمت چپ و وسط)، شما دقیقاً ساختارِ هندسیِ "
    r"داده‌های لیدار (LiDAR) را مشاهده می‌کنید. "
    r"محور Z نشان‌دهندهٔ <i>عمق یا زمان (فریم‌های متوالی)</i> است. در گراف سمت چپ (وکسل‌های آبی)، می‌بینیم که یک شیء (مثلاً عابر پیاده) "
    r"در طول زمان در فضای سه‌بعدی حرکت کرده و یک مسیر مورب ایجاد کرده است. مقداری نویز هم در فضا پراکنده است. "
    r"شبکه‌های عصبی سنتی این مکعبِ زیبا را لِه می‌کنند (Flatten) و در نتیجه ماشین درکی از «حرکت» نخواهد داشت. "
    r"اما وقتی ما <b>فیلتر کانولوشن ۳‌بعدی (3D Convolution)</b> را به شکل یک مکعب ($3 \times 3 \times 3$) در این فضای آبی لغزاندیم، "
    r"الگوریتم به طور خودکار تمام نویزهایِ ثابت را پاک کرد و دقیقاً محل‌هایی که شیء تغییرِ سرعت یا مکان داده بود را شکار کرد "
    r"(وکسل‌های قرمز رنگ در گراف میانی). ماشین اکنون به جای دیدنِ پیکسلِ خام، <i>ردپای حرکتی (Motion Signature)</i> را به صورت فضایی می‌بیند. "
    r"در گراف سمت راست (منحنی سرمه‌ای ضخیم) می‌بینیم که تزریقِ این ویژگی‌های نابِ ۳‌بعدی به شبکه، باعث می‌شود ماشین با سرعتی خارق‌العاده همگرا شود و مانعِ متحرک را شناسایی کند."
)
display_academic_text(insight_conclusion, "insight")
```
