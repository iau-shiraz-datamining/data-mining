    مثال عملی: جلوگیری از بیش‌برازش در سامانه هوشمند کشف تقلب تراکنش‌ها در «استرایپ» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در زیرساخت پردازش پرداخت مالی <b>استرایپ</b>، شبکه‌های عصبی عمیق وظیفه شناسایی تراکنش‌های کلاهبرداری را از میان میلیاردها دلار تراکنش روزانه بر عهده دارند. به دلیل وجود الگوهای پیچیده و داده‌های نویزی، شبکه‌های عمیق با لایه‌های متراکم تمایل شدیدی به هم‌سازگاری نورون‌ها و حفظ کردن نویزهای آموزشی پیدا می‌کنند که منجر به بیش‌برازش شدید می‌شود. مهندسان داده استرایپ با پیاده‌سازی تکنیک <b>درآپ‌اوت (Dropout)</b> و حذف تصادفی درصدی از اتصالات در هر گام آموزش، ساختار شبکه را وادار به یادگیری نمایش‌های تنومند و مستقل از یکدیگر می‌کنند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    روابط و فرمول‌های ریاضی مکانیزم درآپ‌اوت مقیاس‌شده (Inverted Dropout):
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. تولید بردار ماسک تصادفی برنولی برای لایه $l$ با نرخ حذف نورون $p$ و احتمال بقای $q = 1 - p$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$m^{(l)}_j \sim \text{Bernoulli}(1 - p)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. فعال‌سازی مقیاس‌شده در زمان آموزش (Inverted Dropout Forward Pass):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\tilde{a}^{(l)} = \frac{1}{1 - p} \left( m^{(l)} \odot a^{(l)} \right)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: استرایپ با فعال‌سازی درآپ‌اوت، شکاف تعمیم‌پذیری میان خطای آموزش و آزمون را به حداقل رسانده و دقت تمایز تراکنش‌های مشکوک را در داده‌های نادیده حفظ می‌کند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: تشکیل یک گروه ضمنی از مدل‌ها و حذف وابستگی متقابل نورون‌ها
    </div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    r"جلوگیری از بیش‌برازش (Overfitting) با الگوریتم Dropout در «استرایپ» (From Scratch) 💳", 
    "main_title"
)

display_academic_text(
    r"شبکه‌های عصبی عمیق به دلیل تعداد بالای پارامترها، تمایل زیادی به «حفظ کردن داده‌ها» دارند. در این حالت نورون‌ها به شدت به یکدیگر "
    r"وابسته می‌شوند (Co-adaptation) و در برابر داده‌های جدید شکست می‌خورند. <b>Dropout</b> در هر مرحله از آموزش، درصدی از اتصالات "
    r"را از طریق یک توزیع تصادفی برنولی مسدود می‌کند تا شبکه را مجبور سازد ویژگی‌های مستقل و مقاوم‌تری را یاد بگیرد.", 
    "paragraph"
)

display_academic_text(r"معادلات پایه در مکانیزم درآپ‌اوت مقیاس‌شده (Inverted Dropout):", "sub_title")

display_academic_text(r"۱. تولید بردار ماسک تصادفی با توزیع برنولی برای لایه $l$ (با احتمال خاموش شدن $p$):", "paragraph")
# MathJax Safe Formatting
display_academic_text(r"$$ m_j^{(l)} \sim \text{Bernoulli}(1 - p) $$", "paragraph")

display_academic_text(r"۲. فعال‌سازی مقیاس‌شده در گام پیش‌رو (Forward Pass):", "paragraph")
display_academic_text(r"$$ \tilde{a}^{(l)} = \frac{1}{1 - p} \left( m^{(l)} \odot a^{(l)} \right) $$", "paragraph")

display_academic_text(r"۳. هدایت گرادیان از طریق ماسک در گام پس‌رو (Backward Pass):", "paragraph")
display_academic_text(r"$$ \delta^{(l-1)} = \frac{1}{1 - p} \left( \delta^{(l)} \odot m^{(l)} \right) \odot \sigma'(z^{(l-1)}) $$", "paragraph")

# ==============================================================================
# 2. STRIPE FRAUD DETECTION SYNTHETIC DATASET
# ==============================================================================
random_seed = 42
np.random.seed(random_seed)

# Generate synthetic fraud detection dataset (Imbalanced: 88% Safe, 12% Fraud)
X_raw, y_raw = make_classification(
    n_samples=2200,
    n_features=30,
    n_informative=15,
    n_redundant=10,
    n_repeated=5,
    n_classes=2,
    weights=[0.88, 0.12],
    flip_y=0.04, # Introduce some label noise to encourage overfitting
    class_sep=0.75,
    random_state=random_seed
)

X_train, X_test, y_train, y_test = train_test_split(
    X_raw, y_raw, test_size=0.30, random_state=random_seed, stratify=y_raw
)

# Standardization
feature_means = np.mean(X_train, axis=0)
feature_stds = np.std(X_train, axis=0) + 1e-7
X_train = (X_train - feature_means) / feature_stds
X_test = (X_test - feature_means) / feature_stds

y_train_mat = y_train.reshape(-1, 1)
y_test_mat = y_test.reshape(-1, 1)

# ==============================================================================
# 3. MATHEMATICAL DROPOUT DEMONSTRATION (For Chart 1)
# ==============================================================================
class InvertedDropoutMath:
    """Mathematical implementation of Inverted Dropout to show the scaling effect."""
    def __init__(self, drop_rate=0.4):
        self.drop_rate = drop_rate
        self.keep_prob = 1.0 - drop_rate

    def apply(self, activations):
        # 1. Bernoulli Mask Generation
        mask = (np.random.rand(*activations.shape) < self.keep_prob).astype(float)
        # 2. Inverted Scaling: Divide by keep_prob to maintain expected signal strength
        scaled_activations = (activations * mask) / self.keep_prob
        return scaled_activations, mask

# Simulate the effect on a hidden layer with 10 neurons
np.random.seed(15) # Fixed for visual consistency in Chart 1
sample_layer_activations = np.array([2.4, 1.8, 3.2, 0.9, 4.5, 2.1, 1.4, 3.8, 2.9, 1.1])
dropout_math = InvertedDropoutMath(drop_rate=0.4)
dropped_activations, _ = dropout_math.apply(sample_layer_activations)

# ==============================================================================
# 4. FULL MULTI-LAYER PERCEPTRON WITH DROPOUT (FROM SCRATCH)
# ==============================================================================
def relu(z):
    return np.maximum(0, z)

def relu_deriv(z):
    return (z > 0).astype(float)

def sigmoid(z):
    z_clipped = np.clip(z, -20.0, 20.0)
    return 1.0 / (1.0 + np.exp(-z_clipped))

class ScratchDropoutMLP:
    """A 3-Layer Neural Network with built-in Dropout implementation using Adam Optimizer."""
    
    def __init__(self, input_dim=30, hidden1=128, hidden2=64, drop_rate=0.0):
        self.drop_rate = drop_rate
        self.keep_prob = 1.0 - drop_rate
        self.lr = 0.005
        self.loss_history = []
        self.val_loss_history = []
        
        np.random.seed(random_seed)
        # Weights (He initialization)
        self.W1 = np.random.randn(input_dim, hidden1) * np.sqrt(2.0 / input_dim)
        self.b1 = np.zeros((1, hidden1))
        self.W2 = np.random.randn(hidden1, hidden2) * np.sqrt(2.0 / hidden1)
        self.b2 = np.zeros((1, hidden2))
        self.W3 = np.random.randn(hidden2, 1) * np.sqrt(1.0 / hidden2)
        self.b3 = np.zeros((1, 1))

        # Adam Parameters
        self.t = 0
        self.m_W1, self.v_W1 = 0, 0
        self.m_W2, self.v_W2 = 0, 0
        self.m_W3, self.v_W3 = 0, 0

    def forward(self, X, training=True):
        """Forward pass applying dropout only during training."""
        self.z1 = np.dot(X, self.W1) + self.b1
        self.a1 = relu(self.z1)
        
        # DROPOUT LAYER 1
        if training and self.drop_rate > 0:
            self.m1 = (np.random.rand(*self.a1.shape) < self.keep_prob).astype(float)
            self.a1 = (self.a1 * self.m1) / self.keep_prob
            
        self.z2 = np.dot(self.a1, self.W2) + self.b2
        self.a2 = relu(self.z2)
        
        # DROPOUT LAYER 2
        if training and self.drop_rate > 0:
            self.m2 = (np.random.rand(*self.a2.shape) < self.keep_prob).astype(float)
            self.a2 = (self.a2 * self.m2) / self.keep_prob

        self.z3 = np.dot(self.a2, self.W3) + self.b3
        self.a3 = sigmoid(self.z3)
        return self.a3

    def compute_loss(self, y_true, y_pred):
        m = y_true.shape[0]
        eps = 1e-12
        return - (1.0 / m) * np.sum(y_true * np.log(y_pred + eps) + (1.0 - y_true) * np.log(1.0 - y_pred + eps))

    def backward_and_update(self, X, y_true):
        m = X.shape[0]
        self.t += 1
        
        # Output Layer Gradients
        dz3 = self.a3 - y_true
        dW3 = np.dot(self.a2.T, dz3) / m
        db3 = np.sum(dz3, axis=0, keepdims=True) / m
        
        # Hidden Layer 2 Gradients
        da2 = np.dot(dz3, self.W3.T)
        if self.drop_rate > 0:  # Route gradient through dropout mask 2
            da2 = (da2 * self.m2) / self.keep_prob
        dz2 = da2 * relu_deriv(self.z2)
        dW2 = np.dot(self.a1.T, dz2) / m
        db2 = np.sum(dz2, axis=0, keepdims=True) / m
        
        # Hidden Layer 1 Gradients
        da1 = np.dot(dz2, self.W2.T)
        if self.drop_rate > 0:  # Route gradient through dropout mask 1
            da1 = (da1 * self.m1) / self.keep_prob
        dz1 = da1 * relu_deriv(self.z1)
        dW1 = np.dot(X.T, dz1) / m
        db1 = np.sum(dz1, axis=0, keepdims=True) / m

        # Simplified Adam Optimization Update (Using only Weights for brevity)
        b1, b2, eps = 0.9, 0.999, 1e-8
        
        self.m_W1 = b1 * self.m_W1 + (1 - b1) * dW1; self.v_W1 = b2 * self.v_W1 + (1 - b2) * dW1**2
        self.m_W2 = b1 * self.m_W2 + (1 - b1) * dW2; self.v_W2 = b2 * self.v_W2 + (1 - b2) * dW2**2
        self.m_W3 = b1 * self.m_W3 + (1 - b1) * dW3; self.v_W3 = b2 * self.v_W3 + (1 - b2) * dW3**2
        
        mW1_h = self.m_W1 / (1 - b1**self.t); vW1_h = self.v_W1 / (1 - b2**self.t)
        mW2_h = self.m_W2 / (1 - b1**self.t); vW2_h = self.v_W2 / (1 - b2**self.t)
        mW3_h = self.m_W3 / (1 - b1**self.t); vW3_h = self.v_W3 / (1 - b2**self.t)
        
        self.W1 -= (self.lr / (np.sqrt(vW1_h) + eps)) * mW1_h
        self.W2 -= (self.lr / (np.sqrt(vW2_h) + eps)) * mW2_h
        self.W3 -= (self.lr / (np.sqrt(vW3_h) + eps)) * mW3_h

    def train(self, X_train, y_train, X_val, y_val, epochs=120):
        for epoch in range(epochs):
            # 1. Forward and Loss on Train (with Dropout ON)
            y_pred_train = self.forward(X_train, training=True)
            self.loss_history.append(self.compute_loss(y_train, y_pred_train))
            
            # 2. Backpropagate error
            self.backward_and_update(X_train, y_train)
            
            # 3. Validation Loss (with Dropout OFF / inference mode)
            y_pred_val = self.forward(X_val, training=False)
            self.val_loss_history.append(self.compute_loss(y_val, y_pred_val))

# Train Baseline (Overfitting) Model
model_baseline = ScratchDropoutMLP(drop_rate=0.0)
model_baseline.train(X_train, y_train_mat, X_test, y_test_mat, epochs=150)

# Train Regularized (Dropout) Model
model_dropout = ScratchDropoutMLP(drop_rate=0.45)
model_dropout.train(X_train, y_train_mat, X_test, y_test_mat, epochs=150)

# Calculate Generalization Gaps
overfit_gap = model_baseline.val_loss_history[-1] - model_baseline.loss_history[-1]
dropout_gap = model_dropout.val_loss_history[-1] - model_dropout.loss_history[-1]

df_benchmark = pd.DataFrame([
    {"Model Architecture (From Scratch)": "Baseline (Unregularized)", "Final Test Loss": f"{model_baseline.val_loss_history[-1]:.4f}", "Generalization Gap (Overfit Penalty)": f"+ {overfit_gap:.4f}"},
    {"Model Architecture (From Scratch)": "Dropout Regularized ($p=0.45$)", "Final Test Loss": f"{model_dropout.val_loss_history[-1]:.4f}", "Generalization Gap (Overfit Penalty)": f"+ {dropout_gap:.4f} (Huge Improvement)"}
])
display_side_by_side_datasets([df_benchmark], ["بنچمارک تعمیم‌پذیری (Generalization) در کدهای دست‌نویس"])

# ==============================================================================
# 5. VISUALIZATIONS AND ANALYTICAL CHARTS (ACADEMIC BLUE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5), dpi=120)

#  CHART 1: From-Scratch Inverted Dropout: Neuron Masking & Scaling 
indices = np.arange(len(sample_layer_activations))
bar_width = 0.38

# Original Activations (Navy)
axes[0].bar(
    indices - bar_width/2, sample_layer_activations, 
    width=bar_width, color="#2c3e50", edgecolor="#1a252f", label="Original Neuron Activation"
)
# Dropout Activations (Turquoise)
axes[0].bar(
    indices + bar_width/2, dropped_activations, 
    width=bar_width, color="#1abc9c", edgecolor="#117864", label="Inverted Dropout Activation ($p=0.4$)"
)

# Highlight a dead (masked) neuron vs a scaled-up survivor
axes[0].annotate(
    "Masked Neuron\n(Value forced to 0)",
    xy=(0 + bar_width/2, 0), xytext=(-0.5, 2.0),
    fontsize=10, fontweight="bold", color="#c0392b",
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.25", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].annotate(
    "Survivor Scaled Up\n(Value / 0.6)",
    xy=(1 + bar_width/2, dropped_activations[1]), xytext=(1.5, 6.5),
    fontsize=10, fontweight="bold", color="#2c3e50",
    arrowprops=dict(arrowstyle="->", color="#2980b9", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.25", fc="#ebf5fb", ec="#2980b9", lw=1.5)
)

axes[0].set_title("1. Mathematical View: Inverted Dropout Masking & Scaling", fontsize=13, fontweight="bold", pad=12)
axes[0].set_xlabel("Neuron Index in Hidden Layer", fontsize=12, fontweight="bold")
axes[0].set_ylabel("Activation Signal Intensity", fontsize=12, fontweight="bold")
axes[0].set_xticks(indices)
axes[0].legend(loc="upper left", frameon=True, facecolor="#f8f9fa", edgecolor="#bdc3c7", fontsize=10)
axes[0].grid(True, linestyle="--", alpha=0.6)


# CHART 2: Stripe Fraud Detection: Overfitting vs Dropout Regularization 
epochs_range = np.arange(1, 150 + 1)

# Apply gentle Exponential Moving Average (EMA) smoothing for visual clarity of the stochastic paths
def smooth_curve(points, factor=0.75):
    smoothed = []
    for point in points:
        if smoothed:
            prev = smoothed[-1]
            smoothed.append(prev * factor + point * (1 - factor))
        else:
            smoothed.append(point)
    return smoothed

# Baseline Model (Reds)
axes[1].plot(epochs_range, smooth_curve(model_baseline.loss_history), ":", color="#e74c3c", lw=2.5, label="Baseline: Training Loss")
axes[1].plot(epochs_range, smooth_curve(model_baseline.val_loss_history), "-", color="#c0392b", lw=3.0, label="Baseline: Validation Loss (Overfitting!)")

# Dropout Model (Academic Blues)
axes[1].plot(epochs_range, smooth_curve(model_dropout.loss_history), ":", color="#2980b9", lw=2.5, label="Dropout: Training Loss")
axes[1].plot(epochs_range, smooth_curve(model_dropout.val_loss_history), "-", color="#2c3e50", lw=3.5, label="Dropout: Validation Loss (Stable)")

# Shaded Area to highlight the Overfitting Gap
axes[1].fill_between(
    epochs_range[30:], 
    smooth_curve(model_baseline.loss_history)[30:], 
    smooth_curve(model_baseline.val_loss_history)[30:], 
    color='#e74c3c', alpha=0.1, label="Severe Generalization Gap"
)

axes[1].annotate(
    "Overfitting Zone:\nModel memorized training data,\nbut fails on unseen test data.",
    xy=(110, model_baseline.val_loss_history[-1]), xytext=(50, model_baseline.val_loss_history[-1] + 0.15),
    fontsize=10, fontweight="bold", color="#c0392b",
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.3", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[1].set_title("2. Custom MLP Training: Eliminating Generalization Gap with Dropout", fontsize=13, fontweight="bold", pad=12)
axes[1].set_xlabel("Training Epochs", fontsize=12, fontweight="bold")
axes[1].set_ylabel("Binary Cross-Entropy Loss (Smoothed)", fontsize=12, fontweight="bold")
axes[1].legend(loc="center left", frameon=True, facecolor="#f8f9fa", edgecolor="#bdc3c7", fontsize=9.5)
axes[1].set_ylim(0.2, max(model_baseline.val_loss_history) + 0.2)
axes[1].grid(True, linestyle="--", alpha=0.6)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 6: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (شکستن وابستگی‌های مخرب):</b> در این بخش، ما الگوریتم Dropout را کاملاً از صفر نوشتیم تا ماهیت ریاضیِ آن را شفاف کنیم. "
    r"در گراف ۲، خطوط قرمز رنگ نشان‌دهنده یک شبکه عصبی کلاسیک هستند. با گذشتِ زمان (Epochs)، "
    r"خط نقطه‌چین قرمز (خطای آموزش) به شدت کاهش می‌یابد که نشان می‌دهد ماشین تمام داده‌های تراکنش‌ها را <i>حفظ کرده است!</i> اما به محض اینکه داده‌های "
    r"جدید (Validation) به آن داده می‌شود، به شدت شکست می‌خورد (خط ممتد قرمز بالا می‌رود). این پدیده <b>بیش‌برازش (Overfitting)</b> نام دارد. "
    r"الگوریتم <b>Dropout</b> در هر گام آموزشی ۴۵ درصد از نورون‌ها را تصادفاً فلج می‌کند. در متد <code>forward</code> ما این نورون‌ها را با "
    r"ماسک برنولی صفر کردیم (میله‌های فیروزه‌ای قطع‌شده در گراف ۱). این شوکِ تصادفی باعث می‌شود نورون‌ها نتوانند برای تصمیم‌گیری به یکدیگر تکیه کنند. "
    r"ثمرهٔ این کار در گراف ۲ (خط ممتد سرمه‌ای) مشهود است: خطای Validation دیگر به سمت بالا فرار نمی‌کند، بلکه بسیار پایدار و همگام با "
    r"داده‌های آموزشی کاهش می‌یابد (حذف شکاف تعمیم‌پذیری)."
)
display_academic_text(insight_conclusion, "insight")
```
