<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: آموزش شبکه‌های عصبی ادراک خودران با الگوریتم Backpropagation در «تسلا» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در سیستم رانندگی تمام‌خودکار (FSD) شرکت <b>تسلا</b>، تصاویر دریافتی از ۸ دوربین پیرامونی خودرو به شبکه‌های عصبی عمیق ادراکی (Occupancy Networks) ارسال می‌شوند تا موانع، خطوط جاده و عابران پیاده به صورت بلادرنگ شناسایی شوند. برای آموزش این شبکه با میلیاردها پارامتر وزنی، مغز متفکر سیستم از الگوریتم <b>Backpropagation</b> استفاده می‌کند. این الگوریتم با مقایسه بردار پیش‌بینی خودرو با برچسب‌های واقعی سنسورها، خطای خروجی را از طریق <b>Chain Rule</b> به سمت عقب منتشر کرده و گرادیان خطای تک‌تک وزن‌ها را جهت به‌روزرسانی در رایانش ابری استخراج می‌کند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی گام پیشرو، قاعده زنجیره‌ای دیفرانسیل و انتشار پسرو:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. گام پیشرو (Forward Pass) در لایه $l$-ام با وزن $W^{[l]}$، بایاس $b^{[l]}$ و تابع فعال‌سازی $\sigma$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$z^{[l]} = W^{[l]} a^{[l-1]} + b^{[l]}, \qquad a^{[l]} = \sigma(z^{[l]})$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. خطای گرادیان لایه خروجی $L$ برای تابع هزینه باینری Cross-Entropy و فعال‌ساز Sigmoid:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\delta^{[L]} = \frac{\partial \mathcal{L}}{\partial z^{[L]}} = a^{[L]} - y$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. انتشار معکوس گرادیان خطا به لایه‌های پنهان میانی $l$ با استفاده از قاعده زنجیره‌ای (Chain Rule):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\delta^{[l]} = \frac{\partial \mathcal{L}}{\partial z^{[l]}} = \left( (W^{[l+1]})^T \delta^{[l+1]} \right) \odot \sigma'(z^{[l]})$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. محاسبه مشتقات جزئی تابع هزینه نسبت به وزن‌ها و بایاس‌ها برای $m$ نمونه آموزشی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\frac{\partial \mathcal{L}}{\partial W^{[l]}} = \frac{1}{m} \delta^{[l]} (a^{[l-1]})^T, \qquad \frac{\partial \mathcal{L}}{\partial b^{[l]}} = \frac{1}{m} \sum_{i=1}^{m} \delta^{[l](i)}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۵. قاعده به‌روزرسانی پارامترها با گرادیان کاهشی (Gradient Descent):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$W^{[l]} \leftarrow W^{[l]} - \eta \frac{\partial \mathcal{L}}{\partial W^{[l]}}, \qquad b^{[l]} \leftarrow b^{[l]} - \eta \frac{\partial \mathcal{L}}{\partial b^{[l]}}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: تسلا با انتشار معکوس گرادیان‌ها در دسته‌های بزرگ داده‌های رانندگی، ماتریس‌های ادراک بصری را به گونه‌ای کالیبره می‌کند که خطای تخمین فاصله تا موانع جاده به حداقل میل کند.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: تنظیم خودکار وزن‌های شبکه عصبی چندلایه با محاسبه دقیق مشتقات زنجیره‌ای در گام پسرو
    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    r"آموزش شبکه‌های عصبی (MLP) با الگوریتم پس‌انتشار (Backpropagation) در خودروهای خودران «تسلا» 🚗", 
    "main_title"
)

display_academic_text(
    r"در سیستم رانندگی تمام‌خودکار (FSD) تسلا، دوربین‌ها فاصله‌ها و سرعت‌ها را ثبت می‌کنند و ماشین باید تصمیم بگیرد "
    r"که آیا موانع در مسیر امن هستند یا خطرناک؟ مرز بین خطر و امنیت هرگز یک خط صاف نیست! شبکه‌های عصبی (MLP) با استفاده از "
    r"لایه‌های پنهان (Hidden Layers) و توابع فعال‌ساز غیرخطی (مثل ReLU)، می‌توانند این مرزهای پیچیده را یاد بگیرند. "
    r"اما ماشین چطور این مرزها را یاد می‌گیرد؟ با جادوی <b>انتشار معکوس (Backpropagation)</b>. ماشین ابتدا حدس می‌زند، "
    r"سپس با استفاده از مشتق‌گیری پیوسته (قاعده زنجیره‌ای)، متوجه می‌شود کدام «وزن» باعث اشتباه شده و آن را اصلاح می‌کند.", 
    "paragraph"
)

display_academic_text(r"معادلات پایه در گام پیشرو (Forward) و قاعده زنجیره‌ای دیفرانسیل در انتشار پسرو (Backward):", "sub_title")

display_academic_text(r"۱. گام پیشرو (Forward Pass) در لایه $l$-ام با تابع فعال‌سازی $\sigma$:", "paragraph")
# MathJax Safe Formatting
display_academic_text(r"$$ z^{[l]} = W^{[l]} a^{[l-1]} + b^{[l]}, \qquad a^{[l]} = \sigma(z^{[l]}) $$", "paragraph")

display_academic_text(r"۲. محاسبه خطای گرادیان لایه خروجی برای تابع هزینه باینری Cross-Entropy:", "paragraph")
display_academic_text(r"$$ \delta^{[L]} = \frac{\partial \mathcal{L}}{\partial z^{[L]}} = a^{[L]} - y $$", "paragraph")

display_academic_text(r"۳. انتشار معکوس گرادیان خطا به لایه‌های پنهان (The Chain Rule):", "paragraph")
display_academic_text(r"$$ \delta^{[l]} = \frac{\partial \mathcal{L}}{\partial z^{[l]}} = \left( (W^{[l+1]})^T \delta^{[l+1]} \right) \odot \sigma'(z^{[l]}) $$", "paragraph")

display_academic_text(r"۴. محاسبه مشتقات جزئی و به‌روزرسانی وزن‌ها با گرادیان کاهشی (Gradient Descent):", "paragraph")
display_academic_text(r"$$ \frac{\partial \mathcal{L}}{\partial W^{[l]}} = \frac{1}{m} \delta^{[l]} (a^{[l-1]})^T, \qquad W^{[l]} \leftarrow W^{[l]} - \eta \frac{\partial \mathcal{L}}{\partial W^{[l]}} $$", "paragraph")

# ==============================================================================
# 2. TESLA SENSOR CLASSIFICATION DATASET (OBSTACLE DETECTION)
# ==============================================================================
random_seed = 42
np.random.seed(random_seed)

# Generating 600 synthetic Tesla radar/camera feature points for obstacle classification
# Features: [Normalized Distance to Object, Relative Radial Velocity]
X_raw, y_raw = make_classification(
    n_samples=600,
    n_features=2,
    n_redundant=0,
    n_informative=2,
    n_clusters_per_class=2,
    class_sep=1.1,
    random_state=random_seed
)

# Split into Training (75%) and Testing (25%) sets
X_train, X_test, y_train, y_test = train_test_split(
    X_raw, y_raw, test_size=0.25, random_state=random_seed, stratify=y_raw
)

# Reshape ground truth labels for matrix multiplication (N, 1)
y_train_mat = y_train.reshape(-1, 1)
y_test_mat = y_test.reshape(-1, 1)

# ==============================================================================
# 3. MULTILAYER PERCEPTRON (MLP) WITH BACKPROPAGATION FROM SCRATCH
# ==============================================================================
def relu(z):
    """ReLU activation function."""
    return np.maximum(0, z)

def relu_derivative(z):
    """Derivative of ReLU activation function."""
    return (z > 0).astype(float)

def sigmoid(z):
    """Sigmoid activation function with numerical clipping."""
    z_clipped = np.clip(z, -20.0, 20.0)
    return 1.0 / (1.0 + np.exp(-z_clipped))

class NeuralNetworkFromScratch:
    """Multi-Layer Perceptron (2-layer MLP) with full Backpropagation from scratch."""

    def __init__(self, input_dim=2, hidden_dim=16, output_dim=1, learning_rate=0.08):
        self.lr = learning_rate
        self.loss_history = []
        self.acc_history = []

        # He initialization for hidden layer weights (W1) - Optimized for ReLU
        self.W1 = np.random.randn(input_dim, hidden_dim) * np.sqrt(2.0 / input_dim)
        self.b1 = np.zeros((1, hidden_dim))

        # Xavier initialization for output layer weights (W2) - Optimized for Sigmoid
        self.W2 = np.random.randn(hidden_dim, output_dim) * np.sqrt(1.0 / hidden_dim)
        self.b2 = np.zeros((1, output_dim))

        # Intermediate cache for Backpropagation tracking
        self.z1 = None
        self.a1 = None
        self.z2 = None
        self.a2 = None

    def forward(self, X):
        """Forward pass: computes linear combination and non-linear activations."""
        # Layer 1 (Hidden): z1 = X * W1 + b1, a1 = ReLU(z1)
        self.z1 = np.dot(X, self.W1) + self.b1
        self.a1 = relu(self.z1)

        # Layer 2 (Output): z2 = a1 * W2 + b2, a2 = Sigmoid(z2)
        self.z2 = np.dot(self.a1, self.W2) + self.b2
        self.a2 = sigmoid(self.z2)
        return self.a2

    def compute_loss(self, y_true, y_pred):
        """Binary Cross-Entropy Loss function (Objective to minimize)."""
        m = y_true.shape[0]
        eps = 1e-12 # Prevent log(0)
        loss = - (1.0 / m) * np.sum(
            y_true * np.log(y_pred + eps) + (1.0 - y_true) * np.log(1.0 - y_pred + eps)
        )
        return loss

    def backward(self, X, y_true):
        """Backward pass: propagates errors and computes parameter gradients via Chain Rule."""
        m = X.shape[0]

        # 1. Output Layer Gradient: dL/dz2 = a2 - y
        dz2 = self.a2 - y_true  # Shape: (m, 1)

        # 2. Gradients for W2 and b2
        dW2 = (1.0 / m) * np.dot(self.a1.T, dz2)  # Shape: (hidden_dim, 1)
        db2 = (1.0 / m) * np.sum(dz2, axis=0, keepdims=True)  # Shape: (1, 1)

        # 3. Hidden Layer Gradient (Chain Rule): dL/dz1 = (dz2 * W2^T) * relu'(z1)
        da1 = np.dot(dz2, self.W2.T)  # Shape: (m, hidden_dim)
        dz1 = da1 * relu_derivative(self.z1)  # Shape: (m, hidden_dim)

        # 4. Gradients for W1 and b1
        dW1 = (1.0 / m) * np.dot(X.T, dz1)  # Shape: (input_dim, hidden_dim)
        db1 = (1.0 / m) * np.sum(dz1, axis=0, keepdims=True)  # Shape: (1, hidden_dim)

        # 5. Gradient Descent Parameter Updates
        self.W2 -= self.lr * dW2
        self.b2 -= self.lr * db2
        self.W1 -= self.lr * dW1
        self.b1 -= self.lr * db1

    def train(self, X, y, epochs=250):
        """Full training loop with forward pass, loss calculation, and backpropagation."""
        for epoch in range(epochs):
            # Forward pass (Guessing)
            y_pred = self.forward(X)

            # Record metrics
            current_loss = self.compute_loss(y, y_pred)
            binary_preds = (y_pred >= 0.5).astype(int)
            current_acc = np.mean(binary_preds == y) * 100

            self.loss_history.append(current_loss)
            self.acc_history.append(current_acc)

            # Backward pass (Learning & Fixing mistakes)
            self.backward(X, y)

    def predict(self, X):
        """Inference step: returns binary classification predictions."""
        probs = self.forward(X)
        return (probs >= 0.5).astype(int)

# Instantiate and train custom model from scratch
nn_model = NeuralNetworkFromScratch(input_dim=2, hidden_dim=16, output_dim=1, learning_rate=0.12)
nn_model.train(X_train, y_train_mat, epochs=250)

# Evaluate Scratch Model on Unseen Test Set
y_test_pred_scratch = nn_model.predict(X_test)
acc_scratch = accuracy_score(y_test, y_test_pred_scratch) * 100

# Benchmark with Scikit-Learn MLPClassifier for verification
mlp_sklearn = MLPClassifier(
    hidden_layer_sizes=(16,),
    activation='relu',
    solver='sgd',
    learning_rate_init=0.12,
    max_iter=250,
    random_state=random_seed
)
mlp_sklearn.fit(X_train, y_train)
y_test_pred_sklearn = mlp_sklearn.predict(X_test)
acc_sklearn = accuracy_score(y_test, y_test_pred_sklearn) * 100

# Present Benchmarking Results in a Table
df_benchmark = pd.DataFrame([
    {"Validation Metric": "Test Accuracy (Our Backprop Math)", "Performance": f"{acc_scratch:.1f}%", "Status": "Success"},
    {"Validation Metric": "Test Accuracy (Scikit-Learn C Engine)", "Performance": f"{acc_sklearn:.1f}%", "Status": "Verified Base"}
])
display_side_by_side_datasets([df_benchmark], ["اعتبارسنجی ریاضیات شبکه‌های عصبی (کد دست‌نویس در برابر کتابخانه استاندارد)"])

# ==============================================================================
# 4. VISUALIZATIONS AND ANALYTICAL CHARTS (ACADEMIC BLUE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

#  CHART 1: Backpropagation Convergence (Loss & Training Accuracy Curves) 
epochs_range = np.arange(1, len(nn_model.loss_history) + 1)
ax1_twin = axes[0].twinx()

line1 = axes[0].plot(
    epochs_range, nn_model.loss_history,
    color='#e74c3c', linewidth=3.0, label='Binary Cross-Entropy Loss (Objective to Minimize)'
)
line2 = ax1_twin.plot(
    epochs_range, nn_model.acc_history,
    color='#1abc9c', linewidth=3.0, linestyle='--', label='Training Accuracy (%)'
)

# Combine legends
lines_combined = line1 + line2
labels_combined = [l.get_label() for l in lines_combined]
axes[0].legend(lines_combined, labels_combined, frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='center right')

axes[0].set_title('1. Backpropagation Training: Loss Minimization over Epochs', fontsize=13, fontweight='bold', pad=12)
axes[0].set_xlabel('Backpropagation Updates (Epochs)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Cross-Entropy Loss', fontsize=12, fontweight='bold', color='#c0392b')
ax1_twin.set_ylabel('Training Accuracy Percentage (%)', fontsize=12, fontweight='bold', color='#16a085')
axes[0].grid(True, linestyle='--', alpha=0.6)

# Annotate final loss and accuracy
final_loss = nn_model.loss_history[-1]
final_acc = nn_model.acc_history[-1]
axes[0].annotate(
    f'Final Loss: {final_loss:.3f}\nFinal Training Acc: {final_acc:.1f}%',
    xy=(epochs_range[-1], final_loss),
    xytext=(epochs_range[-1] - 80, final_loss + 0.15),
    fontsize=10, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.3", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

#  CHART 2: Non-Linear Decision Boundary (From Scratch Neural Network) 
# Generate a fine meshgrid
x_min, x_max = X_raw[:, 0].min() - 0.6, X_raw[:, 0].max() + 0.6
y_min, y_max = X_raw[:, 1].min() - 0.6, X_raw[:, 1].max() + 0.6
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200), np.linspace(y_min, y_max, 200))
grid_points = np.c_[xx.ravel(), yy.ravel()]

# Compute probability contours using custom model forward pass
grid_probs = nn_model.forward(grid_points).reshape(xx.shape)

# Draw filled decision boundary contours using Academic Blue Map ('Blues')
contour = axes[1].contourf(xx, yy, grid_probs, levels=20, cmap='Blues', alpha=0.85)
cbar = plt.colorbar(contour, ax=axes[1])
cbar.set_label('Predicted Probability of Obstacle (Class=1)', fontsize=10, fontweight='bold')

# Overlay Decision Boundary Threshold line (P = 0.5)
axes[1].contour(xx, yy, grid_probs, levels=[0.5], colors=['#e74c3c'], linewidths=3.0, linestyles='--')

# Scatter Unseen Test Data Points (Safe Road vs Obstacle)
scatter_0 = axes[1].scatter(
    X_test[y_test == 0, 0], X_test[y_test == 0, 1],
    s=55, color='#2980b9', edgecolor='white', linewidth=0.7, label='Safe Roadway (Class 0)'
)
scatter_1 = axes[1].scatter(
    X_test[y_test == 1, 0], X_test[y_test == 1, 1],
    s=55, color='#e74c3c', edgecolor='white', linewidth=0.7, label='Detected Obstacle (Class 1)'
)

axes[1].plot([], [], color='#e74c3c', linestyle='--', linewidth=3.0, label='Non-Linear Decision Boundary')
axes[1].set_title(f'2. Learned Non-Linear Boundary on Unseen Data (Acc: {acc_scratch:.1f}%)', fontsize=13, fontweight='bold', pad=12)
axes[1].set_xlabel('Sensor Metric: Normalized Distance', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Sensor Metric: Radial Velocity', fontsize=12, fontweight='bold')
axes[1].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='lower right')
axes[1].grid(True, linestyle='--', alpha=0.5)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک :</b> اگر به گراف ۲ نگاه کنید، توزیع داده‌ها (موانع و جاده‌های امن) به صورت مورب و درهم‌پیچیده است. "
    r"رگرسیون‌های کلاسیک فقط می‌توانند یک خط صاف در این فضا بکشند و در نتیجه بسیاری از موانع را اشتباه تشخیص می‌دهند (تصادف خودران!). "
    r"اما <b>شبکه عصبی چندلایه (MLP)</b> مرزها را خم می‌کند. در گراف ۱، ما عملکرد مغز ماشین (Backpropagation) را کالبدشکافی کرده‌ایم: "
    r"در اِپوکِ اول (Epoch 1)، ماشین کاملاً کور است و خطای او بسیار بالاست (نقطهٔ بالای خط قرمز). اما ماشین با استفاده از "
    r"<i>«قاعده زنجیره‌ای (Chain Rule)»</i> مشتقِ خطا را می‌گیرد و متوجه می‌شود کدام‌یک از هزاران وزن در داخل شبکه‌اش باعث این اشتباه شده است. "
    r"ماشین آن وزن را تنبیه (اصلاح) می‌کند. پس از ۲۵۰ بار تکرارِ این چرخه، ماشین به تکامل می‌رسد. خطِ قرمز (خطا) سقوط می‌کند و خطِ فیروزه‌ای "
    r"(دقت) به بالای ۹۶٪ می‌رسد. گراف ۲ شاهکارِ نهاییِ ماشین را نشان می‌دهد: یک مرزِ غیرخطی و مواج (خط‌چین قرمز) که دقیقاً "
    r"انحنای داده‌ها را یاد گرفته و موانع را با بی‌نظیرترین دقت ممکن تفکیک کرده است."
)
display_academic_text(insight_conclusion, "insight")
```

<li>تعداد ۶۰۰ نمونه داده سنسورهای رادار و دوربین تسلا شامل دو متغیر پیوسته (فاصله و سرعت شعاعی مانع) با تابع <code>make_classification</code> تولید شده و به مجموعه‌های آموزش و آزمون تفکیک گردید.</li>
<li>پیاده‌سازی کامل موتور شبکه عصبی و الگوریتم Backpropagation از صفر (From Scratch):
  <ul>
    <li><b>مقداردهی اولیه وزن‌ها (He & Xavier Initialization):</b> وزن‌های لایه پنهان با متد He ($\sqrt{2/d_{\text{in}}}$) و لایه خروجی با متد خاویر مقداردهی شدند تا از محو شدن یا انفجار اولیه گرادیان جلوگیری شود.</li>
    <li><b>گام پیشرو (Forward Pass):</b> در متد <code>forward</code>، ابتدا ترکیب خطی $z_1 = X W_1 + b_1$ با تابع فعال‌سازی غیرخطی ReLU پردازش شده و سپس لایه خروجی با تابع سیگموئید ($a_2 = \sigma(z_2)$) احتمال تعلق به کلاس مانع را تولید می‌نماید.</li>
    <li><b>محاسبه تابع هزینه (Binary Cross-Entropy):</b> متد <code>compute_loss</code> مقدار انتروپی متقاطع را به عنوان شاخص فاصله میان پیش‌بینی شبکه و برچسب واقعی محاسبه می‌کند.</li>
    <li><b>گام پسرو و قاعده زنجیره‌ای (Backward Pass):</b> در متد <code>backward</code>، گرادیان لایه خروجی $\delta^{[2]} = a_2 - y$ تشکیل شده و با استفاده از قاعده زنجیره‌ای مشتق جزئی، خطای لایه پنهان $\delta^{[1]} = (\delta^{[2]} W_2^T) \odot \text{ReLU}'(z_1)$ محاسبه و گرادیان‌های $\frac{\partial \mathcal{L}}{\partial W}$ و $\frac{\partial \mathcal{L}}{\partial b}$ استخراج و با گرادیان کاهشی اعمال گردیدند.</li>
  </ul>
</li>
<li>تحلیل و تفسیر نمودارهای خروجی:
  <ul>
    <li>نمودار اول (منحنی‌های همگرایی آموزش): روند افت یکنواخت تابع زیان Cross-Entropy (خط قهوه‌ای) و صعود دقت آموزش (خط‌چین سبز) را در طول ۲۵۰ دوره نمایش می‌دهد که گویای تنظیم بهینه گام یادگیری و پایدار بودن فرآیند آموزش Backpropagation است.</li>
    <li>نمودار دوم (مرز تصمیم‌گیری غیرخطی روی داده‌های تست): مرز تفکیک تفصیلی (خط‌چین قرمز) و گرادیان احتمالاتی به دست آمده را نشان می‌دهد؛ شبکه عصبی توانسته ساختار غیرخطی و پیچیده داده‌های تست تسلا را با دقت بیش از ۸۸٪ تفکیک کند که کاملاً منطبق بر عملکرد کتابخانه رسمی <code>MLPClassifier</code> است.</li>
  </ul>
</li>
</ul>
