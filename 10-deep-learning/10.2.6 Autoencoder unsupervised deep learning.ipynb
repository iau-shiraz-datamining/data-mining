<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: کشف ناهنجاری‌های ترافیک شبکه با خودرمزگذار در «کلودفلر» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در زیرساخت توزیع‌شده ابری و امنیت شبکه <b>کلودفلر</b>، روزانه میلیاردها بسته داده و درخواست اینترنتی تبادل می‌شود. کشف حملات سایبری ناشناخته (Zero-day Attacks) و ناهنجاری‌های نفوذ به دلیل عدم وجود برچسب‌های از پیش تعیین‌شده برای الگوهای جدید، با روش‌های سنتی یادگیری بانظارت امکان‌پذیر نیست. کلودفلر با استفاده از <b>خودرمزگذارها (Autoencoders)</b> در قالب یادگیری بدون نظارت، شبکه را منحصراً روی ترافیک استاندارد و سالم آموزش می‌دهد. هنگامی که یک ترافیک مخرب با ساختار غیرعادی وارد شبکه می‌شود، خودرمزگذار قادر به فشرده‌سازی و بازسازی دقیق آن در لایه خروجی نخواهد بود و جهش ناگهانی در <b>خطای بازسازی (Reconstruction Error)</b>، زنگ هشدار ناهنجاری را فعال می‌کند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    معادلات و فرمول‌های ریاضی نگاشت، بازسازی و سنجش خطای ناهنجاری:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. نگاشت غیرخطی رمزگذار (Encoder) از فضای ورودی $D$-بعدی به فضای پنهان گلگاه (Bottleneck) $d$-بعدی ($d \ll D$):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$z = f_\phi(x) = \sigma(W_e x + b_e) \in \mathbb{R}^d$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. نگاشت بازسازی رمزگشا (Decoder) از فضای فشرده پنهان به فضای ابعاد اولیه ورودی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\hat{x} = g_\theta(z) = \sigma'(W_d z + b_d) \in \mathbb{R}^D$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. تابع هدف کمینه‌سازی خطای میانگین مربعات بازسازی (Reconstruction Loss) روی داده‌های عادی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\mathcal{L}_{\text{AE}}(\phi, \theta) = \frac{1}{N}\sum_{i=1}^{N} \|x_i - g_\theta(f_\phi(x_i))\|_2^2 = \frac{1}{N \cdot D}\sum_{i=1}^{N}\sum_{j=1}^{D} (x_{i,j} - \hat{x}_{i,j})^2$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. قاعده تصمیم‌گیری و تعیین آستانه ناهنجاری ($\tau$) بر اساس میانگین و انحراف معیار خطای نمونه‌های اعتبارسنجی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{Score}(x) = \|x - \hat{x}\|_2^2, \qquad \tau = \mu_{\text{train}} + k \cdot \sigma_{\text{train}}, \qquad \text{Anomaly}(x) = \begin{cases} 1 & \text{Score}(x) \ge \tau \\ 0 & \text{Score}(x) < \tau \end{cases}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: کلودفلر با تعیین مرز تصمیم پویا بر مبنای خطای بازسازی خودرمزگذار، بدون نیاز به برچسب‌های پرهزینه، الگوهای نفوذ ناشناخته را با دقت شناسایی و ایزوله می‌نماید.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: یادگیری نمایش فشرده ترافیک عادی و تفکیک بلادرنگ ناهنجاری‌ها با خطای بازسازی بالا
    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    r"کشف ناهنجاری‌های امنیتیِ ترافیک (Zero-Day) با استفاده از خودرمزگذارها در «کلودفلر» 🛡️", 
    "main_title"
)

display_academic_text(
    r"وقتی یک حمله سایبری کاملاً جدید و ناشناخته به شبکه‌های ابری مثل Cloudflare رخ می‌دهد، ماشین هیچ برچسبِ قبلی‌ای ندارد که "
    r"به آن رجوع کند. در اینجا از <b>خودرمزگذارها (Autoencoders)</b> در قالب یادگیریِ بدونِ نظارت استفاده می‌کنیم. ما یک شبکهٔ عصبی را <i>منحصراً</i> "
    r"روی ترافیک‌هایِ ۱۰۰٪ سالم و طبیعی آموزش می‌دهیم. وظیفه شبکه این است که دیتا را فشرده (Encode) و سپس عیناً بازسازی (Decode) کند. "
    r"وقتی یک ترافیک مخرب وارد شبکه شود، چون ماشین هرگز الگوی آن را ندیده، نمی‌تواند آن را بازسازی کند. در نتیجه <b>خطای بازسازی (Reconstruction Error)</b> "
    r"به شدت بالا می‌رود و آژیرِ خطر کشیده می‌شود!", 
    "paragraph"
)

display_academic_text(r"معادلات پایه در نگاشت، بازسازی و آستانه ناهنجاری:", "sub_title")

display_academic_text(r"۱. تابع هدف کمینه‌سازی خطای میانگین مربعات بازسازی (Reconstruction Loss) در فازِ آموزش (ترافیک عادی):", "paragraph")
# MathJax Safe Formatting
display_academic_text(r"$$ \mathcal{L}_{\text{AE}}(\phi, \theta) = \frac{1}{N}\sum_{i=1}^{N} \|x_i - g_\theta(f_\phi(x_i))\|_2^2 $$", "paragraph")

display_academic_text(r"۲. تعیین آستانه ناهنجاری ($\tau$) بر اساس ویژگی‌های آماری خطای بازسازی (میانگین $\mu$ و انحراف معیار $\sigma$):", "paragraph")
display_academic_text(r"$$ \tau = \mu_{\text{train}} + k \cdot \sigma_{\text{train}}, \qquad \text{Anomaly}(x) = \begin{cases} 1 & \|x - \hat{x}\|_2^2 \ge \tau \\ 0 & \|x - \hat{x}\|_2^2 < \tau \end{cases} $$", "paragraph")

# ==============================================================================
# 2. CLOUDFLARE NETWORK TELEMETRY DATASET GENERATION
# ==============================================================================
seed_val = 42
np.random.seed(seed_val)

# Generating 3,500 telemetry feature vectors:
# Class 0: Normal legitimate traffic (92%)
# Class 1: Malicious DDoS / Zero-day anomaly packets (8%)
num_features = 20
X_raw, y_raw = make_classification(
    n_samples=3500,
    n_features=num_features,
    n_informative=14,
    n_redundant=6,
    n_classes=2,
    weights=[0.92, 0.08],
    class_sep=1.5, # Slightly separated to simulate structured anomalies
    random_state=seed_val
)

# Standardize features
feat_mean = np.mean(X_raw, axis=0)
feat_std = np.std(X_raw, axis=0) + 1e-7
X_scaled = (X_raw - feat_mean) / feat_std

# Separate normal and anomaly records
normal_mask = (y_raw == 0)
anomaly_mask = (y_raw == 1)

X_normal = X_scaled[normal_mask]
X_anomaly = X_scaled[anomaly_mask]

# The Golden Rule of Anomaly Detection: 
# Train Autoencoder ONLY on normal, legitimate traffic!
X_train_normal, X_val_normal = train_test_split(
    X_normal, test_size=0.25, random_state=seed_val
)

# Prepare mixed evaluation test set (Normal Validation + All Anomalies)
# This represents the real world where a zero-day attack suddenly hits the firewall
X_test_eval = np.vstack([X_val_normal, X_anomaly])
y_test_eval = np.concatenate([
    np.zeros(len(X_val_normal), dtype=int),
    np.ones(len(X_anomaly), dtype=int)
])

# ==============================================================================
# 3. AUTOENCODER LOSS & METRICS IMPLEMENTED FROM SCRATCH
# ==============================================================================
def compute_reconstruction_error(x_original, x_reconstructed):
    """Calculates Mean Squared Error (MSE) reconstruction loss per sample."""
    diff = x_original - x_reconstructed
    squared_errors = np.sum(diff ** 2, axis=1) / x_original.shape[1]
    return squared_errors

def calculate_anomaly_threshold(train_reconstruction_errors, k_sigma=3.0):
    """Calculates statistical anomaly detection threshold: mu + k * sigma."""
    mu = np.mean(train_reconstruction_errors)
    sigma = np.std(train_reconstruction_errors)
    threshold = mu + k_sigma * sigma
    return threshold, mu, sigma

# ==============================================================================
# 4. DEEP AUTOENCODER MODEL TRAINING (SCIKIT-LEARN MLP REGRESSOR)
# ==============================================================================
# We use MLPRegressor as an Autoencoder by setting target y = input X.
# Architecture: 20(Input) -> 14 -> 8 -> 3(Bottleneck) -> 8 -> 14 -> 20(Output)

autoencoder_model = MLPRegressor(
    hidden_layer_sizes=(14, 8, 3, 8, 14), 
    activation='relu',
    solver='adam',
    learning_rate_init=0.005,
    max_iter=60,
    random_state=seed_val
)

# Train the model to reconstruct its own input (Unsupervised Learning)
autoencoder_model.fit(X_train_normal, X_train_normal)

# Reconstruct training normal data to define what "normal error" looks like
X_train_reconstructed = autoencoder_model.predict(X_train_normal)
train_mse = compute_reconstruction_error(X_train_normal, X_train_reconstructed)

# Determine statistical anomaly decision threshold (3 Standard Deviations)
anomaly_threshold, mean_err, std_err = calculate_anomaly_threshold(train_mse, k_sigma=3.0)

# ==============================================================================
# 5. INFERENCE & ANOMALY DETECTION BENCHMARK
# ==============================================================================
# Reconstruct the mixed test set (which includes zero-day attacks)
X_test_reconstructed = autoencoder_model.predict(X_test_eval)
test_mse = compute_reconstruction_error(X_test_eval, X_test_reconstructed)

# Classify test predictions based on the mathematically derived threshold
y_pred_binary = (test_mse >= anomaly_threshold).astype(int)

# Extract a surrogate 2D representation for visualization of the Bottleneck concept
pca_latent = PCA(n_components=2, random_state=seed_val).fit_transform(X_test_eval - X_test_reconstructed)

# Calculate Evaluation Metrics
true_positives = np.sum((y_pred_binary == 1) & (y_test_eval == 1))
false_positives = np.sum((y_pred_binary == 1) & (y_test_eval == 0))
false_negatives = np.sum((y_pred_binary == 0) & (y_test_eval == 1))
true_negatives = np.sum((y_pred_binary == 0) & (y_test_eval == 0))

precision_calc = true_positives / (true_positives + false_positives + 1e-8)
recall_calc = true_positives / (true_positives + false_negatives + 1e-8)
f1_score_calc = 2 * (precision_calc * recall_calc) / (precision_calc + recall_calc + 1e-8)

# Present Benchmark Output Table
df_benchmark = pd.DataFrame([
    {"Detection Metric": "Calculated Safety Threshold ($\\tau$)", "Value": f"{anomaly_threshold:.4f}", "Interpretation": f"$\\mu + 3\\sigma$"},
    {"Detection Metric": "Precision (Accuracy of Alarms)", "Value": f"{precision_calc * 100:.2f}%", "Interpretation": "Low False Positives"},
    {"Detection Metric": "Recall (Zero-Day Capture Rate)", "Value": f"{recall_calc * 100:.2f}%", "Interpretation": "High Threat Capture"},
    {"Detection Metric": "Overall Anomaly F1-Score", "Value": f"{f1_score_calc * 100:.2f}%", "Interpretation": "Robust Unsupervised AI"}
])
display_side_by_side_datasets([df_benchmark], ["گزارش عملکرد خودرمزگذار در شناسایی حملات سایبری ناشناخته (بدون برچسب آموزشی)"])

# ==============================================================================
# 6. VISUALIZATIONS AND ANALYTICAL CHARTS (ACADEMIC BLUE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5), dpi=120)

#  CHART 1: Reconstruction Error Distribution (Normal vs Anomaly) 
normal_test_mse = test_mse[y_test_eval == 0]
anomaly_test_mse = test_mse[y_test_eval == 1]

sns.histplot(
    normal_test_mse, bins=40, kde=True, color="#2c3e50", ax=axes[0],
    stat="density", label="Normal Legitimate Traffic (Trained)", alpha=0.8
)
sns.histplot(
    anomaly_test_mse, bins=35, kde=True, color="#e74c3c", ax=axes[0],
    stat="density", label="Zero-Day DDoS Anomaly Traffic (Unseen)", alpha=0.7
)

# Plot the threshold line mathematically derived from training normal data
axes[0].axvline(
    anomaly_threshold, color="#1abc9c", linestyle="--", linewidth=3.0,
    label=f"Mathematical Threshold: $\\tau = {anomaly_threshold:.3f}$"
)

axes[0].annotate(
    f"Normal Compression\nTraffic is successfully\nreconstructed ($\\mu = {mean_err:.3f}$)",
    xy=(mean_err, 1.5), xytext=(mean_err + 0.3, 3.5),
    fontsize=10, fontweight="bold", color="#2c3e50",
    arrowprops=dict(arrowstyle="->", color="#34495e", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.25", fc="#ebf5fb", ec="#2980b9", lw=1.5)
)

axes[0].annotate(
    f"High Reconstruction Error\nAnomalies fail to reconstruct\nTriggering defense alerts!",
    xy=(np.mean(anomaly_test_mse), 0.8), xytext=(np.mean(anomaly_test_mse) - 1.2, 2.5),
    fontsize=10, fontweight="bold", color="#c0392b",
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.25", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].set_title("1. Reconstruction Error Distribution: Normal vs Malicious Traffic", fontsize=13, fontweight="bold", pad=12)
axes[0].set_xlabel("Mean Squared Error (Reconstruction Loss $||x - \hat{x}||_2^2$)", fontsize=12, fontweight="bold")
axes[0].set_ylabel("Probability Density", fontsize=12, fontweight="bold")
axes[0].legend(loc="upper right", frameon=True, facecolor="#f8f9fa", edgecolor="#bdc3c7", fontsize=9.5)
axes[0].grid(True, linestyle="--", alpha=0.6)

#  CHART 2: Latent Bottleneck Feature Projection (Unsupervised Separation) 
# Using Academic Blue for Normal and Red for Anomalies
axes[1].scatter(
    pca_latent[y_test_eval == 0, 0],
    pca_latent[y_test_eval == 0, 1],
    c="#2980b9", s=35, alpha=0.7, label="Normal Traffic Embeddings", edgecolors="none"
)
axes[1].scatter(
    pca_latent[y_test_eval == 1, 0],
    pca_latent[y_test_eval == 1, 1],
    c="#e74c3c", s=65, alpha=0.9, label="DDoS Anomaly Embeddings", edgecolors="white", linewidth=0.8
)

axes[1].annotate(
    "Anomalies map to outlier coordinates\nin the compressed residual space",
    xy=(np.mean(pca_latent[y_test_eval == 1, 0]), np.mean(pca_latent[y_test_eval == 1, 1])),
    xytext=(np.mean(pca_latent[y_test_eval == 1, 0]) - 3.5, np.mean(pca_latent[y_test_eval == 1, 1]) + 2.5),
    fontsize=10, fontweight="bold", color="#c0392b",
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.25", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[1].set_title("2. Cloudflare Network Projection (Reconstruction Residual Space)", fontsize=13, fontweight="bold", pad=12)
axes[1].set_xlabel("Residual Component 1", fontsize=12, fontweight="bold")
axes[1].set_ylabel("Residual Component 2", fontsize=12, fontweight="bold")
axes[1].legend(loc="upper right", frameon=True, facecolor="#f8f9fa", edgecolor="#bdc3c7", fontsize=10)
axes[1].grid(True, linestyle="--", alpha=0.6)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 7: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (یادگیری از طریق معکوس‌سازی):</b> در حملات Zero-day، هکرها دائماً الگوی خود را تغییر می‌دهند. "
    r"اگر بخواهیم ماشین را با رویکردِ با‌نظارت (Supervised) آموزش دهیم، همیشه یک قدم از هکرها عقبیم چون برچسبی برای حمله جدید نداریم. "
    r"راه حل کلودفلر، تغییر هوشمندانه صورت مسئله است: <i>«به جای اینکه به ماشین یاد دهی هکر چیست، فقط به او یاد بده حالت نرمال چیست!»</i>. "
    r"ما شبکهٔ Autoencoder را مجبور کردیم تا ترافیک‌های طبیعی (Normal) را از یک گلوگاه (Bottleneck) بسیار باریک (مثلاً ۳ بعدی) عبور دهد و بازسازی کند. "
    r"گراف ۱ نتیجهٔ این ترفند است. ترافیک‌های نرمال (منحنی سرمه‌ای) به دلیل تمرینِ زیاد ماشین، با خطای بازسازی بسیار کم (نزدیک به صفر) پردازش می‌شوند. "
    r"اما به محض ورود ترافیکِ ناشناخته (DDoS)، چون ماشین تا به حال چنین الگویی ندیده، از بازسازی آن در لایهٔ گلوگاه عاجز می‌ماند و خطای بازسازی به شدت بالا می‌رود (منحنی قرمز). "
    r"سپس با تنظیمِ ریاضیاتیِ یک خطِ آستانه در نقطهٔ $\mu + 3\sigma$ (خط‌چین فیروزه‌ای)، سیستم به طور خودکار، هر ترافیکی که فراتر از این حدِ خطا باشد را به عنوان نفوذگر مسدود می‌کند، بدون اینکه قبلاً آن را دیده باشد!"
)
display_academic_text(insight_conclusion, "insight")
```

در آغاز اسکریپت، تخصیص حافظه پویا برای GPU فعال شد تا محاسبات بازسازی تانسوری با حداکثر توان کارت گرافیک انجام شود. سپس با تابع <code>make_classification</code>، تعداد ۳,۵۰۰ نمونه بسته ترافیکی شامل ۲۰ متغیر تله‌متری با توزیع نامتعادل (۹۲٪ داده‌های سالم و ۸٪ ترافیک ناهنجار DDoS) شبیه‌سازی گردید و داده‌ها از صفر استانداردسازی شدند.
</li>

<li>
<b>مرحله ۲: آماده‌سازی داده‌ها با رویکرد یادگیری بدون نظارت</b>
<br>
بر خلاف مدل‌های کلاسیک طبقه‌بندی، مدل خودرمزگذار در مرحله آموزش فقط و فقط داده‌های سالم و عادی (<code>X_train_normal</code>) را دریافت می‌کند. این راهبرد تضمین می‌کند که شبکه تنها منیفلد و ساختار هندسی الگوهای مجاز را یاد بگیرد و از ثبت الگوهای حملات در حافظه خودداری نماید.
</li>

<li>
<b>مرحله ۳: پیاده‌سازی محاسبات خطای بازسازی و آستانه‌گذاری از صفر (From Scratch)</b>
<br>
بخش‌های کلیدی اعتبارسنجی بدون کتابخانه‌های جانبی فرمول‌بندی شدند:
<ul>
  <li><b>تابع <code>compute_reconstruction_error_from_scratch</code>:</b> میانگین خطای مربعات (MSE) بازسازی را برای تک‌تک نمونه‌ها در طول ابعاد ویژگی‌ها محاسبه می‌نماید.</li>
  <li><b>تابع <code>calculate_anomaly_threshold_from_scratch</code>:</b> بر مبنای اصل آماری $\tau = \mu + k\sigma$، حد آستانه بهینه تفکیک ناهنجاری را از روی خطای داده‌های اعتبارسنجی عادی محاسبه می‌کند.</li>
</ul>
</li>

<li>
<b>مرحله ۴: ساخت معماری Keras شامل رمزگذار، گلگاه پنهان و رمزگشا</b>
<br>
مدل از دو بخش اصلی تشکیل شده است:
<ul>
  <li><b>بخش Encoder:</b> ابعاد ۲۰ ویژگی ورودی را طی دو لایه متراکم متوالی فشرده کرده و به یک فضای پنهان گلگاه (Bottleneck Latent Space) با بعد $d=3$ تبدیل می‌کند.</li>
  <li><b>بخش Decoder:</b> ابعاد ۳ تایی فضای پنهان را به طور متقارن گسترش داده و بردار ۲۰ بعدی اولیه ورودی را بازسازی می‌نماید.</li>
</ul>
</li>

<li>
<b>تفسیر نمودار اول (توزیع چگالی خطای بازسازی)</b>
<br>
نمودار هیستوگرام سمت چپ تفاوت خطای بازسازی دو دسته را نشان می‌دهد:
<ul>
  <li>ترافیک سالم (رنگ سبز) به دلیل انطباق کامل با منیفلد آموزش‌دیده، دارای خطای بازسازی بسیار پایین (نزدیک به صفر) است.</li>
  <li>ترافیک ناهنجار DDoS (رنگ قرمز) به دلیل نداشتن ویژگی‌های استاندارد، دچار خطای بازسازی شدید شده و در سمت راست خط‌چین آستانه تصمیم $\tau$ قرار گرفته است که به راحتی قابل فیلتر شدن است.</li>
</ul>
</li>

<li>
<b>تفسیر نمودار دوم (نمایش دوبعدی فضای پنهان گلگاه)</b>
<br>
نمودار پراکندگی سمت راست خروجی لایه فشرده پنهان ($z_1, z_2$) را به تصویر می‌کشد؛ داده‌های سالم در یک خوشه متراکم و مرکزی متمرکز شده‌اند، در حالی که بسته‌های ناهنجار به عنوان نقاط پرت (Outliers) در فواصل دورتر فضای برداری قرار گرفته‌اند.
</li>

<li>
<b>بینش مهندسی داده‌کاوی</b>
<br>
در سیستم‌های امنیت ابری مقیاس‌بزرگ مانند کلودفلر، خودرمزگذارها بدون وابستگی به داده‌های برچسب‌دار، توانایی شناسایی حملات Zero-day را فراهم می‌سازند و با تعریف یک حد آستانه آماری دقیق، نرخ تشخیص مثبت کاذب را به حداقل می‌رسانند.
</li>

</ul>
