  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: استخراج ویژگی‌های بصری و ساخت نقشه‌های ویژگی در «اسنپ‌چت» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در سامانه پردازش بلادرنگ تصویر و فیلترهای واقعیت افزوده <b>اسنپ‌چت</b>، میلیون‌ها فریم ویدیویی در هر ثانیه برای شناسایی چهره، برآورد موقعیت چشم‌ها و لب‌ها و ردیابی حرکات اجزای صورت تحلیل می‌شوند. شبکه‌های عصبی متراکم (Fully Connected) به دلیل حساسیت شدید به تغییر زاویه یا جابه‌جایی پیکسل‌ها و تعداد بسیار بالای پارامترها، کارایی لازم را روی دستگاه‌های تلفن همراه ندارند. مهندسان داده اسنپ‌چت با پیاده‌سازی <b>لایه‌های کانولوشنال (Convolutional Layers)</b> متشکل از بانک فیلترهای چندکاناله، میدان‌های گیرنده محلی (Local Receptive Fields) و اشتراک وزن‌ها، نقشه‌های ویژگی چندبعدی (Feature Maps) تولید می‌کنند که ویژگی‌های اساسی تصویر را مستقل از موقعیت مکانی استخراج می‌نمایند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    معادلات و فرمول‌های ریاضی انتقال پیش‌رو، نگاشت ابعاد و محاسبات پارامترها در لایه کانولوشنال:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. تبدیل تانسوری چندکاناله ورودی $X \in \mathbb{R}^{H_{\text{in}} \times W_{\text{in}} \times C_{\text{in}}}$ به نقشه ویژگی خروجی $Y \in \mathbb{R}^{H_{\text{out}} \times W_{\text{out}} \times C_{\text{out}}}$ با هسته‌های فیلتر $W \in \mathbb{R}^{K_h \times K_w \times C_{\text{in}} \times C_{\text{out}}}$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$Y_{i, j, k} = \sigma\left( \sum_{c=1}^{C_{\text{in}}} \sum_{m=-M}^{M} \sum_{n=-N}^{N} X_{i\cdot S + m, j\cdot S + n, c} \cdot W_{m, n, c, k} + b_k \right)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. محاسبه ابعاد فضایی نقشه ویژگی با در نظر گرفتن اندازه گام لغزش $S$، پدینگ $P$ و اندازه هسته فیلتر $K$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$H_{\text{out}} = \left\lfloor \frac{H_{\text{in}} - K_h + 2P_h}{S_h} \right\rfloor + 1, \qquad W_{\text{out}} = \left\lfloor \frac{W_{\text{in}} - K_w + 2P_w}{S_w} \right\rfloor + 1$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. فرمول کل پارامترهای قابل آموزش لایه کانولوشنال (وزن‌ها و بایاس‌ها):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\text{Total Parameters} = \left( K_h \times K_w \times C_{\text{in}} + 1 \right) \times C_{\text{out}}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. گسترش شعاع میدان گیرنده مؤثر ($RF_l$) در لایه‌های پیاپی کانولوشن:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$RF_l = RF_{l-1} + (K_l - 1) \cdot \prod_{i=1}^{l-1} S_i$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: اسنپ‌چت با تنظیم بهینه عمق فیلترها و گام‌های لایه کانولوشن، حجم محاسبات استخراج لندمارک‌های صورت را به میزان چشمگیری کاهش داده و امکان اعمال افکت‌های بلادرنگ با ۶۰ فریم بر ثانیه را فراهم می‌سازد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: استخراج هم‌زمان نقشه‌های ویژگی چندگانه و کاهش پارامترها با اشتراک‌گذاری وزن‌ها
    </div>
  </p>

</div>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    r"استخراج ویژگی‌های بصری و ساخت نقشه‌های ویژگی (Feature Maps) با کانولوشن چندکاناله در «اسنپ‌چت» 👻", 
    "main_title"
)

display_academic_text(
    r"در اسنپ‌چت، برای ردیابی چهره و اعمال فیلترها، تصویر به صورت بلادرنگ پردازش می‌شود. ما نمی‌توانیم فقط به دنبال یک نوع ویژگی بگردیم. "
    r"ماشین باید <i>همزمان</i> لبه‌های چشم، انحنای لب و بافت پوست را پیدا کند. یک لایه <b>کانولوشن چندکاناله (Multi-Channel Convolution)</b> "
    r"به جای یک فیلتر، از ده‌ها فیلتر مختلف (Bank of Filters) استفاده می‌کند. هر فیلتر وظیفه دارد یک ویژگی خاص را شکار کند و یک خروجی مستقل بسازد. "
    r"بنابراین یک عکسِ ورودی، تبدیل به یک مکعب از چندین <b>نقشهٔ ویژگی (Feature Map)</b> می‌شود.", 
    "paragraph"
)

display_academic_text(r"معادلات پایه در کانولوشن چندکاناله و نگاشت ابعاد فضایی:", "sub_title")

display_academic_text(r"۱. تبدیل تانسور ورودی به نقشه ویژگی خروجی با استفاده از $k$-امین فیلتر (کرنل):", "paragraph")
# MathJax Safe Formatting
display_academic_text(r"$$ Y_{i, j, k} = \sigma\left( \sum_{c=1}^{C_{\text{in}}} \sum_{m=-M}^{M} \sum_{n=-N}^{N} X_{i\cdot S + m, j\cdot S + n, c} \cdot W_{m, n, c, k} + b_k \right) $$", "paragraph")

display_academic_text(r"۲. محاسبه ابعاد فضایی نقشه ویژگی با در نظر گرفتن پدینگ ($P$) و استراید ($S$):", "paragraph")
display_academic_text(r"$$ H_{\text{out}} = \left\lfloor \frac{H_{\text{in}} - K_h + 2P_h}{S_h} \right\rfloor + 1, \qquad W_{\text{out}} = \left\lfloor \frac{W_{\text{in}} - K_w + 2P_w}{S_w} \right\rfloor + 1 $$", "paragraph")

# ==============================================================================
# 2. SNAPCHAT OPTICAL DIGIT DATASET PREPARATION
# ==============================================================================
seed_val = 42
np.random.seed(seed_val)

# Load 8x8 pixel grayscale digits dataset
digits = load_digits()
X_raw_images = digits.images.astype("float32") / 16.0  # Normalize to [0, 1]
y_labels = digits.target

# Reshape to standard 4D tensor: (Batch, Height, Width, Channels)
X_tensors = np.expand_dims(X_raw_images, axis=-1)

X_train, X_test, y_train, y_test = train_test_split(
    X_tensors, y_labels, test_size=0.25, random_state=seed_val, stratify=y_labels
)

# ==============================================================================
# 3. MULTI-CHANNEL CONVOLUTIONAL LAYER IMPLEMENTED FROM SCRATCH
# ==============================================================================
class MultiChannelConv2DFromScratch:
    """Implements a full 2D Convolutional Layer with MULTIPLE filters (Feature Maps)."""
    def __init__(self, num_filters=4, kernel_size=(3, 3), stride=1, padding='same'):
        self.num_filters = num_filters
        self.k_h, self.k_w = kernel_size
        self.stride = stride
        self.padding = padding
        
        # Initialize deterministic custom filter kernels (Edges, Corners, Diagonal)
        # Shape: (Height, Width, Input_Channels, Output_Channels)
        self.weights = np.zeros((self.k_h, self.k_w, 1, num_filters), dtype=np.float32)
        
        # Filter 0: Horizontal Edge Detector
        self.weights[:, :, 0, 0] = np.array([[-1, -2, -1], [0, 0, 0], [1, 2, 1]])
        # Filter 1: Vertical Edge Detector
        self.weights[:, :, 0, 1] = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]])
        # Filter 2: Diagonal Texture Detector
        self.weights[:, :, 0, 2] = np.array([[2, -1, -1], [-1, 2, -1], [-1, -1, 2]])
        # Filter 3: Corner / High-Frequency Detector
        self.weights[:, :, 0, 3] = np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]])

        self.biases = np.zeros(num_filters, dtype=np.float32)

    def forward(self, input_batch):
        """Forward pass executing multi-channel convolution and ReLU activation."""
        batch_size, in_h, in_w, in_c = input_batch.shape
        
        # Calculate padding dimensions (Zero Padding)
        if self.padding == 'same':
            pad_h = (self.k_h - 1) // 2
            pad_w = (self.k_w - 1) // 2
            padded_input = np.pad(
                input_batch,
                ((0, 0), (pad_h, pad_h), (pad_w, pad_w), (0, 0)),
                mode='constant',
                constant_values=0.0
            )
        else:
            pad_h, pad_w = 0, 0
            padded_input = input_batch

        out_h = int(np.floor((padded_input.shape[1] - self.k_h) / self.stride) + 1)
        out_w = int(np.floor((padded_input.shape[2] - self.k_w) / self.stride) + 1)
        
        # Output Tensor: (Batch, Height, Width, Num_Filters)
        feature_maps = np.zeros((batch_size, out_h, out_w, self.num_filters), dtype=np.float32)

        # Slide filters across local receptive fields (Mathematically pure)
        for b in range(batch_size):
            for f in range(self.num_filters):
                for i in range(out_h):
                    for j in range(out_w):
                        r_start = i * self.stride
                        r_end = r_start + self.k_h
                        c_start = j * self.stride
                        c_end = c_start + self.k_w

                        receptive_field = padded_input[b, r_start:r_end, c_start:c_end, :]
                        linear_response = np.sum(receptive_field * self.weights[:, :, :, f]) + self.biases[f]
                        # Apply Non-Linearity (ReLU)
                        feature_maps[b, i, j, f] = max(0.0, linear_response)

        return feature_maps

# Initialize our custom Convolution Layer
conv_layer = MultiChannelConv2DFromScratch(num_filters=4, kernel_size=(3, 3), stride=1, padding='same')

# Execute from-scratch Conv2D layer on a sample image (Digit '5')
sample_idx = 5
sample_tensor = X_train[sample_idx:sample_idx+1]
scratch_feature_maps = conv_layer.forward(sample_tensor)

# ==============================================================================
# 4. CUSTOM TRAINING ENGINE (NO TENSORFLOW)
# ==============================================================================
def extract_all_feature_maps(image_tensor_batch):
    """Processes the entire dataset through the Conv Layer and flattens for the Classifier."""
    fmaps = conv_layer.forward(image_tensor_batch)
    # Flatten the (Batch, 8, 8, 4) tensor into (Batch, 256)
    return fmaps.reshape(fmaps.shape[0], -1)

# Pre-extract spatial feature maps for train and test sets
X_train_fmaps = extract_all_feature_maps(X_train)
X_test_fmaps = extract_all_feature_maps(X_test)

# Flatten raw images for the baseline network (Blind to 2D structure)
X_train_flat = X_train.reshape(X_train.shape[0], -1)
X_test_flat = X_test.reshape(X_test.shape[0], -1)

# Scikit-Learn Classifiers
epochs_count = 35
classes_unique = np.unique(y_labels)

# Baseline Model: Trained on Raw Pixels
mlp_baseline = MLPClassifier(
    hidden_layer_sizes=(64,), activation='relu', solver='adam',
    learning_rate_init=0.005, max_iter=1, warm_start=True, random_state=seed_val
)

# Feature Map Model: Trained on 4 extracted Multi-Channel Features
mlp_fmaps = MLPClassifier(
    hidden_layer_sizes=(64,), activation='relu', solver='adam',
    learning_rate_init=0.005, max_iter=1, warm_start=True, random_state=seed_val
)

acc_baseline_history = []
acc_fmaps_history = []

# Manual epoch loop to track validation accuracy
for epoch in range(epochs_count):
    mlp_baseline.partial_fit(X_train_flat, y_train, classes=classes_unique)
    mlp_fmaps.partial_fit(X_train_fmaps, y_train, classes=classes_unique)
    
    acc_baseline_history.append(accuracy_score(y_test, mlp_baseline.predict(X_test_flat)))
    acc_fmaps_history.append(accuracy_score(y_test, mlp_fmaps.predict(X_test_fmaps)))

final_baseline_acc = acc_baseline_history[-1] * 100
final_fmaps_acc = acc_fmaps_history[-1] * 100

# Benchmark Output Table
df_benchmark = pd.DataFrame([
    {"Architecture": "Dense Network (Raw Pixels)", "Visual Features": "None (Blind to edges/shapes)", "Validation Accuracy": f"{final_baseline_acc:.2f}%"},
    {"Architecture": "ConvNet (Feature Maps)", "Visual Features": "4 Independent Spatial Maps (Edges, Corners)", "Validation Accuracy": f"{final_fmaps_acc:.2f}%"}
])
display_side_by_side_datasets([df_benchmark], ["بنچمارک دقت: تأثیر استخراج ویژگی‌های چندکاناله (Multi-Channel) بر عملکرد مدل"])

# ==============================================================================
# 5. VISUALIZATIONS AND ANALYTICAL CHARTS (ACADEMIC BLUE)
# ==============================================================================
# Using GridSpec to beautifully separate the feature maps from the line plot
fig = plt.figure(figsize=(18, 7), dpi=120)
gs = fig.add_gridspec(2, 4, width_ratios=[1, 1, 1, 3], wspace=0.25, hspace=0.4)

#  CHART 1: Multi-Channel Feature Maps Decomposition 
# Input Image
ax_in = fig.add_subplot(gs[0, 0])
ax_in.imshow(sample_tensor[0, :, :, 0], cmap="Blues", interpolation="nearest")
ax_in.set_title("Input Raw Image (Digit 5)", fontsize=10, fontweight="bold", color="#2c3e50")
ax_in.axis("off")

# Layer Metadata Box
ax_info = fig.add_subplot(gs[1, 0])
ax_info.axis("off")
ax_info.text(
    0.5, 0.5,
    f"Layer Architecture:\n• Input: (8x8x1)\n• Output: (8x8x4)\n• Stride/Padding: 1/Same\n• Weights: 3x3x1x4\n• Total Params: 40",
    fontsize=9.5, fontweight="bold", color="#2c3e50", va="center", ha="center",
    bbox=dict(boxstyle="round,pad=0.6", fc="#ebf5fb", ec="#2980b9", lw=1.5)
)

# Feature Maps Output
ax_fm1 = fig.add_subplot(gs[0, 1])
ax_fm1.imshow(scratch_feature_maps[0, :, :, 0], cmap="mako", interpolation="nearest")
ax_fm1.set_title("Map 1: Horizontal Edges", fontsize=9, fontweight="bold", color="#2c3e50")
ax_fm1.axis("off")

ax_fm2 = fig.add_subplot(gs[0, 2])
ax_fm2.imshow(scratch_feature_maps[0, :, :, 1], cmap="mako", interpolation="nearest")
ax_fm2.set_title("Map 2: Vertical Edges", fontsize=9, fontweight="bold", color="#2c3e50")
ax_fm2.axis("off")

ax_fm3 = fig.add_subplot(gs[1, 1])
ax_fm3.imshow(scratch_feature_maps[0, :, :, 2], cmap="vlag", interpolation="nearest")
ax_fm3.set_title("Map 3: Diagonal Textures", fontsize=9, fontweight="bold", color="#2c3e50")
ax_fm3.axis("off")

ax_fm4 = fig.add_subplot(gs[1, 2])
ax_fm4.imshow(scratch_feature_maps[0, :, :, 3], cmap="vlag", interpolation="nearest")
ax_fm4.set_title("Map 4: Corners & Peaks", fontsize=9, fontweight="bold", color="#2c3e50")
ax_fm4.axis("off")

fig.text(0.25, 0.94, "1. From-Scratch Conv2D: Multi-Channel Feature Maps Generation", fontsize=13, fontweight="bold", ha="center")


#  CHART 2: Training & Validation Accuracy Convergence 
ax_curve = fig.add_subplot(gs[:, 3])
epochs_axis = np.arange(1, epochs_count + 1)

def smooth(points, factor=0.6):
    smoothed = []
    for p in points:
        if smoothed:
            smoothed.append(smoothed[-1] * factor + p * (1 - factor))
        else:
            smoothed.append(p)
    return smoothed

ax_curve.plot(
    epochs_axis, smooth(acc_fmaps_history),
    "-", color="#1abc9c", lw=3.5, label=f"ConvNet (Feature Maps) Acc: {final_fmaps_acc:.1f}%"
)
ax_curve.plot(
    epochs_axis, smooth(acc_baseline_history),
    ":", color="#e74c3c", lw=3.0, label=f"Baseline (Raw Pixels) Acc: {final_baseline_acc:.1f}%"
)

ax_curve.annotate(
    f"Multi-Channel Feature Extraction\ncreates a richer representation of the image,\nleading to +{final_fmaps_acc - final_baseline_acc:.1f}% accuracy gain.",
    xy=(epochs_count, smooth(acc_fmaps_history)[-1]),
    xytext=(epochs_count - 18, smooth(acc_fmaps_history)[-1] - 0.15),
    fontsize=10, fontweight="bold", color="#2c3e50",
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.3", fc="#ebf5fb", ec="#2980b9", lw=1.5)
)

ax_curve.set_title("2. Snapchat Visual Classifier: Validation Convergence", fontsize=13, fontweight="bold", pad=12)
ax_curve.set_xlabel("Training Epochs", fontsize=12, fontweight="bold")
ax_curve.set_ylabel("Validation Classification Accuracy", fontsize=12, fontweight="bold")
ax_curve.set_ylim(0.2, 1.05)
ax_curve.legend(loc="lower right", frameon=True, facecolor="#f8f9fa", edgecolor="#bdc3c7", fontsize=10)
ax_curve.grid(True, linestyle="--", alpha=0.6)

ax_curve.spines['top'].set_visible(False)
ax_curve.spines['right'].set_visible(False)

plt.show()

# ==========================================================
# Section 6: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک (بینایی ماشینی چندبُعدی):</b> در گراف ۱، ما یک تصویر خام دو‌بعدی (یک کاناله) را به ماشین دادیم. "
    r"لایه‌های متراکم قدیمی با این تصویر مثل یک رشتهٔ طولانی از اعداد بی‌معنی برخورد می‌کنند. اما ما از یک لایهٔ <b>کانولوشن چندکاناله (Multi-Channel Convolution)</b> "
    r"استفاده کردیم که ۴ فیلتر (کرنل) مختلف داشت. ماشین این فیلترها را روی تصویر لغزاند و تصویر را به ۴ <b>نقشه ویژگی (Feature Map)</b> مختلف تجزیه کرد: "
    r"یک نقشه فقط لبه‌های افقی را می‌بیند، یک نقشه خطوط مورب را شکار کرده و نقشه دیگر گوشه‌ها را برجسته کرده است! "
    r"این دقیقاً روشی است که فیلترهای اسنپ‌چت کار می‌کنند؛ در یک کانال چشم‌ها را پیدا می‌کنند و در کانالی دیگر بافتِ پوست را شناسایی می‌کنند. "
    r"در گراف ۲ (منحنی فیروزه‌ای)، اثباتِ قدرتِ این معماری را می‌بینیم: شبکه‌ای که با این ۴ نقشهٔ ویژگیِ ثروتمند (Rich Representation) تغذیه شده است، "
    r"با دقتی بسیار بالاتر و خطای بسیار کمتر نسبت به رقیبِ کلاسیکِ خود همگرا می‌شود و نیازی به میلیاردها پارامتر برای یادگیری ندارد!"
)
display_academic_text(insight_conclusion, "insight")
```
