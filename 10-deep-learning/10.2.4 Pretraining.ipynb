<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: بازشناسی و رده‌بندی هوشمند تصاویر با یادگیری انتقالی در «پین‌ترست» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در پلتفرم کشف بصری <b>پین‌ترست</b>، روزانه صدها میلیون تصویر و پین جدید بارگذاری می‌شود. چالش اساسی، دسته‌بندی و برچسب‌گذاری خودکار این تصاویر در شرایطی است که برای بسیاری از دسته‌بندی‌های نوظهور، داده‌های دارای برچسب بسیار اندک و کمیاب هستند. آموزش یک شبکه کانولوشنی عمیق از ابتدا (From Scratch) روی داده‌های محدود منجر به بیش‌برازش شدید و عدم یادگیری الگوهای بصری می‌شود. مهندسان داده پین‌ترست با استفاده از رویکرد <b>پیش‌آموزش (Pretraining)</b> بر روی مجموعه داده عظیم ImageNet و سپس <b>تنظیم دقیق (Fine-Tuning)</b> لایه‌های انتهایی بر روی تصاویر هدف، دانش استخراج ویژگی‌های عمومی (مانند لبه‌ها، بافت‌ها و اشکال هندسی) را منتقل می‌کنند.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مبانی و فرمول‌های ریاضی فرآیند پیش‌آموزش و انتقال دانش به دامنه هدف:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. کمینه‌سازی خطای تجربی در فاز پیش‌آموزش روی دامنه منبع $D_{\text{src}}$ برای یادگیری پارامترهای پایه $\theta_{\text{base}}$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\min_{\theta_{\text{base}}, \phi_{\text{src}}} \frac{1}{|D_{\text{src}}|} \sum_{(x_i, y_i) \in D_{\text{src}}} \mathcal{L}_{\text{CE}}\left(g_{\phi_{\text{src}}}(f_{\theta_{\text{base}}}(x_i)), y_i\right)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. استخراج نگاشت ویژگی‌های متراکم (Latent Embeddings) برای نمونه‌های دامنه هدف $D_{\text{tgt}}$:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$z_j = f_{\theta_{\text{base}}^*}(x_j^{\text{tgt}}) \in \mathbb{R}^d$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. بهینه‌سازی سربرگ طبقه‌بندی هدف $\phi_{\text{tgt}}$ با انجماد لایه‌های پایه (Frozen Feature Extractor):
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\min_{\phi_{\text{tgt}}} \frac{1}{|D_{\text{tgt}}|} \sum_{(x_j, y_j) \in D_{\text{tgt}}} \mathcal{L}_{\text{CE}}\left(h_{\phi_{\text{tgt}}}(f_{\theta_{\text{base}}^*}(x_j^{\text{tgt}})), y_j\right) + \lambda \|\phi_{\text{tgt}}\|_2^2$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. قاعده به‌روزرسانی با نرخ یادگیری تفکیکی لایه‌ای (Discriminative Layer-wise Learning Rate) در فاز Fine-tuning:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$\theta_{l, t+1} = \theta_{l, t} - \eta_l \nabla_{\theta_l} \mathcal{L}(\theta), \qquad \eta_l = \eta_{\text{base}} \cdot \gamma^{L - l}, \quad \gamma \in (0, 1]$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: پین‌ترست با بهره‌گیری از وزن‌های پیش‌آموزش‌دیده، حتی با کمتر از ۱۰۰ تصویر نمونه برای هر دسته، به دقت بالای ۸۸٪ دست یافته و زمان همگرایی آموزش را بیش از ۷۰٪ کاهش می‌دهد.<br><br>

    </div>
  </p>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.neural_network import MLPClassifier
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from IPython.display import display, HTML
import warnings

warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", rc={"axes.titlesize": 14, "axes.labelsize": 12, "xtick.labelsize": 10, "ytick.labelsize": 10})
plt.rcParams['figure.dpi'] = 120

def display_academic_text(text, type="normal"):
    styles = {
        "main_title": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 24px; "
            "font-weight: bold; text-align: center; border-bottom: 2px solid #34495e; "
            "padding-bottom: 15px; margin: 30px 0;"
        ),
        "sub_title": (
            "font-family: Tahoma, sans-serif; color: #34495e; font-size: 19px; "
            "font-weight: bold; border-right: 5px solid #2980b9; padding-right: 15px; "
            "margin: 40px 0px 20px 0px; background-color: #ecf0f1; padding: 12px; "
            "border-radius: 4px;"
        ),
        "paragraph": (
            "font-family: Tahoma, sans-serif; color: #2c3e50; font-size: 16px; "
            "margin-bottom: 20px; line-height: 1.8; text-align: justify;"
        ),
        "insight": (
            "font-family: Tahoma, sans-serif; color: #16a085; font-size: 16px; "
            "font-weight: bold; background-color: #e8f8f5; padding: 20px; "
            "border-left: 5px solid #1abc9c; margin: 30px 0; line-height: 1.8; "
            "text-align: justify;"
        )
    }
    display(HTML(f"<div dir='rtl' style='text-align: right; {styles[type]}'>{text}</div>"))

def get_protected_css():
    return """
    <style>
    .academic-table-wrapper { 
        direction: ltr !important; 
        overflow-x: auto !important; 
        width: 100% !important; 
        background-color: #fff; 
        border: 1px solid #1abc9c; /* Bright turquoise border */
        border-radius: 4px; 
        box-shadow: 0 3px 6px rgba(0,0,0,0.08); 
    }
    table.strict-tbl { 
        border-collapse: collapse !important; 
        width: 100% !important; 
        margin: 0 !important; 
        font-family: inherit !important;
        font-size: 14px !important; 
        table-layout: auto !important; 
        border: none !important; 
    }
    table.strict-tbl thead th { 
        background-color: #2c3e50 !important; /* Deep navy blue header */
        color: #ffffff !important; 
        padding: 12px 18px !important; 
        text-align: center !important; 
        border: 1px solid #34495e !important; 
        white-space: nowrap !important; 
        font-weight: bold !important; 
    }
    table.strict-tbl tbody td { 
        padding: 10px 18px !important; 
        text-align: center !important; 
        border: 1px solid #bdc3c7 !important; 
        white-space: nowrap !important; 
        color: #2c3e50 !important; 
    }
    table.strict-tbl tbody tr:nth-child(even) td { 
        background-color: #ebf5fb !important; /* Light blue rows */
    }
    table.strict-tbl tbody tr:hover td { 
        background-color: #d6eaf8 !important; 
    }
    </style>
    """

def display_side_by_side_datasets(dfs, titles, *args):
    if not isinstance(dfs, list) and not isinstance(titles, list):
        dfs = [dfs]
        titles = [titles]
        
    col_html_list = []
    for i, (df, title) in enumerate(zip(dfs, titles)):
        table_html = df.to_html(index=False, classes=['strict-tbl'], border=0)
        bg_color = "#2c3e50" 
        col_html_list.append(f"""
        <div class="flex-column" style="flex: 1 1 0; min-width: 0; display: flex; flex-direction: column;">
            <div style="background-color: {bg_color}; color: white; padding: 12px; 
                        text-align: center; font-family: Tahoma, sans-serif; 
                        font-weight: bold; font-size: 15px; border-radius: 4px 4px 0 0;">
                {title}
            </div>
            <div class="academic-table-wrapper" style="margin-top: 0; border-top: none;">{table_html}</div>
        </div>
        """)
    
    display(HTML(get_protected_css() + f'<div style="display: flex; gap: 20px; width: 100%; margin: 20px 0 35px 0;">{"".join(col_html_list)}</div>'))

display(HTML("""
<style>
@import url('https://cdn.jsdelivr.net/gh/rastikerdar/vazirmatn@v33.003/Vazirmatn-font-face.css');
* { font-family: 'Vazirmatn', 'Vazir', Tahoma, sans-serif !important; }
</style>
"""))
```

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    r"رده‌بندی تصاویر با داده‌های اندک (Few-Shot) با استفاده از یادگیری انتقالی در «پین‌ترست» 📌", 
    "main_title"
)

display_academic_text(
    r"در پین‌ترست ده‌ها دسته‌بندی نوظهور وجود دارد که فقط تعداد کمی عکس برچسب‌دار برای آن‌ها در دسترس است. "
    r"اگر یک شبکهٔ عصبیِ خام را با این تعداد کم آموزش دهیم، شبکه سریعاً دچار بیش‌برازش (Overfit) می‌شود. "
    r"اما ما از <b>یادگیری انتقالی (Transfer Learning)</b> استفاده می‌کنیم. ما مدلی را فرض می‌کنیم که قبلاً روی میلیون‌ها عکس آموزش دیده و "
    r"ویژگی‌های بنیادین را یاد گرفته است. در این کد، ما تفاوت «یادگیری از روی فضای پیکسل‌های خام» در برابر «یادگیری از روی فضای جاسازی‌شده (Latent Space)» را مدل‌سازی و مقایسه می‌کنیم.", 
    "paragraph"
)

display_academic_text(r"معادلات پایه در فاز پیش‌آموزش و انتقال دانش:", "sub_title")

display_academic_text(r"۱. استخراج نگاشت ویژگی‌های متراکم (Latent Embeddings) توسط مدل از پیش‌آموزش‌دیده ($f_{\theta_{\text{base}}^*}$):", "paragraph")
# MathJax Safe Formatting
display_academic_text(r"$$ z_j = f_{\theta_{\text{base}}^*}(x_j^{\text{tgt}}) \in \mathbb{R}^d $$", "paragraph")

display_academic_text(r"۲. بهینه‌سازی سربرگ طبقه‌بندیِ هدف ($\phi_{\text{tgt}}$) روی ویژگی‌های استخراج‌شده:", "paragraph")
display_academic_text(r"$$ \min_{\phi_{\text{tgt}}} \frac{1}{|D_{\text{tgt}}|} \sum_{(x_j, y_j) \in D_{\text{tgt}}} \mathcal{L}_{\text{CE}}\left(h_{\phi_{\text{tgt}}}(z_j), y_j\right) $$", "paragraph")

# ==============================================================================
# 2. SYNTHETIC TRANSFER LEARNING DATASET PREPARATION (TF-INDEPENDENT)
# ==============================================================================
random_seed = 42
np.random.seed(random_seed)

# We mathematically simulate the difference between "Raw Pixels" and "Pretrained Embeddings"
n_samples = 900
n_classes = 3
class_names = ["Minimalist Decor", "Vintage Fashion", "Tech Gadgets"]

# 1. Base Latent Space (Simulating the clean, separated features a Pretrained model extracts)
# e.g., representing shapes, edges, and textures extracted by a frozen VGG16
X_latent, y_target = make_blobs(
    n_samples=n_samples, 
    n_features=25, 
    centers=n_classes, 
    cluster_std=2.2, 
    random_state=random_seed
)
# Add some ambient noise to the latent space
X_pretrained = X_latent + np.random.normal(0, 0.5, X_latent.shape)

# 2. Raw Pixel Space (Simulating raw images before processing)
# We project the latent space into a highly non-linear, high-dimensional space (e.g., 400 pixels) 
# and add massive noise so the classes are heavily entangled.
random_projection_matrix = np.random.randn(25, 400)
X_raw = np.sin(np.dot(X_latent, random_projection_matrix)) + np.random.normal(0, 2.5, (n_samples, 400))

# Simulate "Few-Shot Learning": We only have 60 labeled images to train on (20 per class)!
X_raw_train, X_raw_test, X_pre_train, X_pre_test, y_train, y_test = train_test_split(
    X_raw, X_pretrained, y_target, train_size=60, random_state=random_seed, stratify=y_target
)

# Standardize both spaces
from sklearn.preprocessing import StandardScaler
scaler_raw = StandardScaler().fit(X_raw_train)
X_raw_train, X_raw_test = scaler_raw.transform(X_raw_train), scaler_raw.transform(X_raw_test)

scaler_pre = StandardScaler().fit(X_pre_train)
X_pre_train, X_pre_test = scaler_pre.transform(X_pre_train), scaler_pre.transform(X_pre_test)

# ==============================================================================
# 3. CUSTOM TRAINING LOOP FOR VALIDATION CURVES (FROM SCRATCH LOGIC)
# ==============================================================================
# Model 1: Training From Scratch (on Raw Pixels)
mlp_scratch = MLPClassifier(
    hidden_layer_sizes=(64,),
    activation='relu',
    solver='adam',
    learning_rate_init=0.005,
    max_iter=1,  # We will manually iterate to track test accuracy!
    warm_start=True,
    random_state=random_seed
)

# Model 2: Transfer Learning (on Pretrained Latent Features)
mlp_transfer = MLPClassifier(
    hidden_layer_sizes=(64,),
    activation='relu',
    solver='adam',
    learning_rate_init=0.005,
    max_iter=1,
    warm_start=True,
    random_state=random_seed
)

epochs_count = 45
acc_scratch_history = []
acc_transfer_history = []
classes_unique = np.unique(y_target)

# Manual Training Loop to extract exact Validation Accuracy per epoch
for epoch in range(epochs_count):
    # Train for one epoch
    mlp_scratch.partial_fit(X_raw_train, y_train, classes=classes_unique)
    mlp_transfer.partial_fit(X_pre_train, y_train, classes=classes_unique)
    
    # Evaluate strictly on unseen Test Data
    pred_scratch = mlp_scratch.predict(X_raw_test)
    pred_transfer = mlp_transfer.predict(X_pre_test)
    
    acc_scratch_history.append(accuracy_score(y_test, pred_scratch))
    acc_transfer_history.append(accuracy_score(y_test, pred_transfer))

# Present Benchmark Accuracy Table
final_acc_scratch = acc_scratch_history[-1] * 100
final_acc_transfer = acc_transfer_history[-1] * 100

df_benchmark = pd.DataFrame([
    {"Model Approach": "Trained From Scratch (Raw Pixels)", "Visual Knowledge Base": "Random Init (Blind)", "Test Accuracy (60 Samples)": f"{final_acc_scratch:.2f}%"},
    {"Model Approach": "Transfer Learning (Latent Space)", "Visual Knowledge Base": "Pretrained Feature Extractor", "Test Accuracy (60 Samples)": f"{final_acc_transfer:.2f}%"}
])
display_side_by_side_datasets([df_benchmark], ["بنچمارک دقت مدل‌ها در طبقه‌بندی تصاویر (یادگیری Few-Shot با تنها ۶۰ نمونه)"])

# ==============================================================================
# 4. VISUALIZATIONS AND ANALYTICAL CHARTS (ACADEMIC BLUE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5), dpi=120)

# Academic Blue Palette for Classes
palette_colors = ["#2c3e50", "#1abc9c", "#2980b9"]

#  CHART 1: PCA Projections of Feature Spaces (Raw vs Pretrained) 
# We use PCA to compress both spaces to 2D to show how "entangled" the raw data is
pca_raw = PCA(n_components=2).fit_transform(X_raw)
pca_pre = PCA(n_components=2).fit_transform(X_pretrained)

# Plot Raw Pixels (Background/Faded to show entanglement)
axes[0].scatter(
    pca_raw[:, 0], pca_raw[:, 1],
    color='#bdc3c7', s=25, alpha=0.6, edgecolor='none', label="Raw Pixels (Highly Entangled)"
)

# Plot Pretrained Features
for cls_idx, cls_label in enumerate(class_names):
    idx_mask = (y_target == cls_idx)
    axes[0].scatter(
        pca_pre[idx_mask, 0],
        pca_pre[idx_mask, 1],
        color=palette_colors[cls_idx],
        label=f"Pretrained Space: {cls_label}",
        s=45, alpha=0.9, edgecolor="#ffffff", linewidth=0.5
    )

axes[0].annotate(
    "Raw Pixels (Gray) are completely mixed.\nPretrained features (Colors) naturally\ncluster shapes before ANY training!",
    xy=(float(np.mean(pca_pre[y_target == 1, 0])), float(np.mean(pca_pre[y_target == 1, 1]))),
    xytext=(0.0, 10.0),
    fontsize=10, fontweight='bold', color='#2c3e50',
    arrowprops=dict(arrowstyle="->", color="#34495e", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.25", fc="#ebf5fb", ec="#2980b9", lw=1.5)
)

axes[0].set_title("1. Latent Space Comparison: Raw Pixels vs Pretrained Embeddings", fontsize=13, fontweight='bold', pad=12)
axes[0].set_xlabel("Principal Component 1", fontsize=12, fontweight='bold')
axes[0].set_ylabel("Principal Component 2", fontsize=12, fontweight='bold')
axes[0].legend(loc="lower right", frameon=True, facecolor="#f8f9fa", edgecolor="#bdc3c7", fontsize=9.5)
axes[0].grid(True, linestyle="--", alpha=0.6)


#  CHART 2: Validation Accuracy Trajectory (Transfer vs Scratch) 
epoch_axis = np.arange(1, epochs_count + 1)

# Apply gentle exponential smoothing for better visual flow
def smooth_curve(points, factor=0.65):
    smoothed = []
    for p in points:
        if smoothed:
            smoothed.append(smoothed[-1] * factor + p * (1 - factor))
        else:
            smoothed.append(p)
    return smoothed

axes[1].plot(
    epoch_axis,
    np.array(smooth_curve(acc_transfer_history)) * 100,
    "-", color="#1abc9c", linewidth=3.5,
    label=f"Transfer Learning (Final: {final_acc_transfer:.1f}%)"
)
axes[1].plot(
    epoch_axis,
    np.array(smooth_curve(acc_scratch_history)) * 100,
    ":", color="#e74c3c", linewidth=3.0,
    label=f"Trained From Scratch (Final: {final_acc_scratch:.1f}%)"
)

axes[1].annotate(
    f"Transfer Learning Gain: +{final_acc_transfer - final_acc_scratch:.1f}%\nMassive gap achieved due to pretrained\nfeature extraction on scarce data.",
    xy=(epochs_count, smooth_curve(acc_transfer_history)[-1] * 100),
    xytext=(epochs_count - 25, smooth_curve(acc_transfer_history)[-1] * 100 - 18.0),
    fontsize=10, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.25", fc="#ebf5fb", ec="#1abc9c", lw=1.5)
)

axes[1].set_title("2. Pinterest Few-Shot Classification: Validation Accuracy", fontsize=13, fontweight='bold', pad=12)
axes[1].set_xlabel("Training Epochs (Fine-Tuning the Classification Head)", fontsize=12, fontweight='bold')
axes[1].set_ylabel("Categorical Validation Accuracy (%)", fontsize=12, fontweight='bold')
axes[1].set_ylim(20, 105)
axes[1].legend(loc="lower right", frameon=True, facecolor="#f8f9fa", edgecolor="#bdc3c7", fontsize=10)
axes[1].grid(True, linestyle="--", alpha=0.6)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک :</b> در پین‌ترست، وقتی دستهٔ جدیدی ایجاد می‌شود، تصاویر زیادی برای آموزش ماشین نداریم. "
    r"در این شبیه‌سازی ما <b>تنها ۶۰ عکس</b> برای آموزش شبکه فراهم کردیم (محیط Few-Shot). "
    r"اگر یک شبکه عصبی را به صورت خام و از صفر روی فضای پیکسل‌ها آموزش دهیم (خط‌چین قرمز در گراف ۲)، شبکه در انبوهی از متغیرهای بی‌معنی و نویز غرق می‌شود. "
    r"به نقاط طوسی در گراف ۱ نگاه کنید؛ داده‌های خام کاملاً در هم گره خورده‌اند! در نتیجه ماشین هیچ الگوی معناداری پیدا نمی‌کند و دقت آن در حدس زدن باقی می‌ماند. "
    r"اما معجزهٔ <b>یادگیری انتقالی (Transfer Learning)</b> این است که ما فضای مسئله را تغییر می‌دهیم. یک مدلِ از پیش‌آموزش‌دیده (Pretrained)، عکس‌ها را به یک "
    r"فضای جدید (نقاط رنگی در گراف ۱) می‌برد که در آن، ویژگی‌ها از قبل دسته‌بندی و شفاف شده‌اند. "
    r"گراف ۲ نشان می‌دهد که شبکه ما وقتی روی این فضایِ از قبل آماده‌شده (خط فیروزه‌ای ضخیم) آموزش می‌بیند، با سرعتی باورنکردنی و تنها با ۶۰ عکس، "
    r"به دقتی بالای ۸۰ درصد پرواز می‌کند!"
)
display_academic_text(insight_conclusion, "insight")
```
