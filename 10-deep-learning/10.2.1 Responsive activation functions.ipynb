<div style="font-family: 'Vazirmatn', Tahoma, sans-serif; direction: rtl; border-radius: 0 10px 10px 0; padding: 16px 20px; margin: 14px 0; border-right: 5px solid #B45309; font-size: 15pt; line-height: 2.2; background: #FFFBEB; color: #78350F; box-sizing: border-box;">

  <div style="font-size: 16pt; text-align: center; font-weight: 700; margin-bottom: 8px; color: #B45309;">
    مثال عملی: تسریع یادگیری عمیق در تشخیص حملات سایبری با توابع فعال‌سازی واکنشی در «کلودفلر» 
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    در زیرساخت ابری و امنیت شبکه <b>کلودفلر</b>، روزانه ده‌ها میلیارد درخواست اینترنتی به منظور شناسایی حملات منع سرویس توزیع‌شده (DDoS) و ترافیک مخرب ربات‌ها توسط شبکه‌های عصبی عمیق تحلیل می‌شوند. در شبکه‌های چندلایه بسیار عمیق، استفاده از تابع فعال‌سازی سنتی <b>ReLU</b> به دلیل تولید شیب صفر برای ورودی‌های منفی منجر به پدیده <b>Dying ReLU</b> می‌شود (وضعیتی که در آن نورون‌ها برای همیشه غیرفعال شده و یادگیری متوقف می‌شود). کلودفلر با جایگزینی توابع فعال‌سازی واکنشی نظیر <b>Leaky ReLU</b>، <b>ELU</b> و <b>PReLU</b>، جریان پیوسته گرادیان را در لایه‌های زیرین برقرار نگه می‌دارد.
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    مفاهیم و فرمول‌های ریاضی توابع فعال‌سازی واکنشی و مشتقات جهت‌دار آن‌ها:
  </p>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۱. تابع فعال‌سازی پایه ReLU و گرادیان ناپدیدشونده در ناحیه منفی:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$f_{\text{ReLU}}(z) = \max(0, z), \qquad f'_{\text{ReLU}}(z) = \begin{cases} 1 & z > 0 \\ 0 & z < 0 \end{cases}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۲. تابع فعال‌سازی Leaky ReLU با شیب نشتی ثابت $\alpha \approx 0.01$ برای جلوگیری از مرگ نورون‌ها:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$f_{\text{LeakyReLU}}(z) = \max(\alpha z, z) = \begin{cases} z & z \ge 0 \\ \alpha z & z < 0 \end{cases}, \qquad f'_{\text{LeakyReLU}}(z) = \begin{cases} 1 & z \ge 0 \\ \alpha & z < 0 \end{cases}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۳. تابع فعال‌سازی نمایی هموار (Exponential Linear Unit - ELU) با میانگین فعال‌سازی نزدیک به صفر:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$f_{\text{ELU}}(z) = \begin{cases} z & z \ge 0 \\ \alpha (e^z - 1) & z < 0 \end{cases}, \qquad f'_{\text{ELU}}(z) = \begin{cases} 1 & z \ge 0 \\ f_{\text{ELU}}(z) + \alpha & z < 0 \end{cases}$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0 0 10px;">
    ۴. تابع فعال‌سازی پارامتریک PReLU با شیب یادگیرنده $a$ در فرآیند Backpropagation:
  </p>

  <div style="text-align: center; margin: 15px 0; direction: ltr;">
    $$f_{\text{PReLU}}(z) = \max(0, z) + a \min(0, z), \qquad \frac{\partial \mathcal{L}}{\partial a} = \sum_{i} \delta_i \cdot \min(0, z_i)$$
  </div>

  <p style="font-size: 15pt; text-align: right; margin: 0;">
    نتیجه: کلودفلر با استفاده از توابع واکنشی، مرگ بیش از ۳۰٪ از نورون‌های شبکه ادراک ترافیک را رفع کرده و نرخ همگرایی و دقت تمایز ترافیک قانونی از حملات سایبری را به طور چشمگیری بهبود می‌بخشد.<br><br>
    <div style="text-align:center; font-weight:bold;">
      خروجی: برقراری پیوسته جریان گرادیان در لایه‌های عمیق و حذف پدیده نورون‌های مرده
    </div>
  </p>

```python
# 1. CONCEPTUAL FRAMEWORK & MATHEMATICAL DEFINITIONS
display_academic_text(
    r"جلوگیری از مرگ نورون‌ها با توابع فعال‌سازی واکنشی در «کلودفلر» (Custom Neural Network) 🛡️", 
    "main_title"
)

display_academic_text(
    r"در شبکه‌های عمیق امنیتی کلودفلر، اگر خروجی یک نورونِ ReLU منفی شود، مشتق آن صفر شده و سیگنال خطای Backpropagation "
    r"متوقف می‌شود (نورون می‌میرد). ما در این مثال یک <b>شبکه عصبی از صفر (From Scratch)</b> نوشته‌ایم که مشتقات توابعِ پیشرفته‌ای "
    r"مانند <b>Leaky ReLU</b> و <b>ELU</b> را در خود دارد. این توابع با حفظِ یک شیبِ ملایم در ناحیه منفی، جریان خونِ شبکه (گرادیان) را "
    r"تضمین می‌کنند.", 
    "paragraph"
)

display_academic_text(r"معادلات پایه در توابع فعال‌سازی واکنشی و مشتقات جهت‌دار آن‌ها:", "sub_title")

display_academic_text(r"۱. تابع فعال‌سازی پایه ReLU (خطر گرادیان صفر در ناحیه منفی):", "paragraph")
# MathJax Safe Formatting
display_academic_text(r"$$ f_{\text{ReLU}}(z) = \max(0, z), \qquad f'_{\text{ReLU}}(z) = \begin{cases} 1 & z > 0 \\ 0 & z < 0 \end{cases} $$", "paragraph")

display_academic_text(r"۲. تابع فعال‌سازی Leaky ReLU با شیب نشتی ثابت $\alpha \approx 0.1$ برای حفظ جریان گرادیان:", "paragraph")
display_academic_text(r"$$ f_{\text{LeakyReLU}}(z) = \max(\alpha z, z), \qquad f'_{\text{LeakyReLU}}(z) = \begin{cases} 1 & z \ge 0 \\ \alpha & z < 0 \end{cases} $$", "paragraph")

display_academic_text(r"۳. تابع نمایی هموار (ELU) با مشتق‌پذیریِ پیوسته در مبدأ:", "paragraph")
display_academic_text(r"$$ f_{\text{ELU}}(z) = \begin{cases} z & z \ge 0 \\ \alpha (e^z - 1) & z < 0 \end{cases}, \qquad f'_{\text{ELU}}(z) = \begin{cases} 1 & z \ge 0 \\ f_{\text{ELU}}(z) + \alpha & z < 0 \end{cases} $$", "paragraph")

# ==============================================================================
# 2. CLOUDFLARE CYBERSECURITY TRAFFIC DATASET GENERATION
# ==============================================================================
random_seed = 42
np.random.seed(random_seed)

# Simulating 1,200 network traffic packets (Class 0: Legitimate, Class 1: DDoS)
X_traffic, y_traffic = make_classification(
    n_samples=1200, n_features=12, n_informative=8, n_redundant=4,
    n_classes=2, class_sep=0.9, random_state=random_seed
)

X_train, X_test, y_train, y_test = train_test_split(
    X_traffic, y_traffic, test_size=0.25, random_state=random_seed, stratify=y_traffic
)
y_train_mat = y_train.reshape(-1, 1)

# ==============================================================================
# 3. RESPONSIVE ACTIVATION FUNCTIONS (Mathematical Profiles for Plotting)
# ==============================================================================
z_domain = np.linspace(-4.0, 4.0, 500)

# Function 1: Standard ReLU
relu_vals = np.maximum(0, z_domain)

# Function 2: Leaky ReLU (alpha = 0.1)
alpha_leaky = 0.1
leaky_relu_vals = np.where(z_domain >= 0, z_domain, alpha_leaky * z_domain)

# Function 3: Exponential Linear Unit (ELU, alpha = 1.0)
alpha_elu = 1.0
elu_vals = np.where(z_domain >= 0, z_domain, alpha_elu * (np.exp(z_domain) - 1.0))

# Function 4: Parametric ReLU (PReLU, a = 0.25)
a_param = 0.25
prelu_vals = np.where(z_domain >= 0, z_domain, a_param * z_domain)

# ==============================================================================
# 4. DEEP NEURAL NETWORK ENGINE (IMPLEMENTED FROM SCRATCH)
# ==============================================================================
class CustomDeepNetwork:
    """A 2-Layer Neural Network implemented from scratch with interchangeable activation functions."""
    
    def __init__(self, input_dim=12, hidden_dim=64, activation='relu', alpha=0.1, lr=0.015):
        self.act = activation
        self.alpha = alpha  # Used for Leaky ReLU and ELU
        self.lr = lr
        self.loss_history = []
        
        # He Initialization
        self.W1 = np.random.randn(input_dim, hidden_dim) * np.sqrt(2.0 / input_dim)
        self.b1 = np.zeros((1, hidden_dim))
        self.W2 = np.random.randn(hidden_dim, 1) * np.sqrt(2.0 / hidden_dim)
        self.b2 = np.zeros((1, 1))

    def forward(self, X):
        self.z1 = np.dot(X, self.W1) + self.b1
        
        # Forward Activations
        if self.act == 'relu':
            self.a1 = np.maximum(0, self.z1)
        elif self.act == 'leaky_relu':
            self.a1 = np.where(self.z1 > 0, self.z1, self.alpha * self.z1)
        elif self.act == 'elu':
            self.a1 = np.where(self.z1 > 0, self.z1, self.alpha * (np.exp(self.z1) - 1.0))
            
        self.z2 = np.dot(self.a1, self.W2) + self.b2
        # Sigmoid for binary classification
        self.a2 = 1.0 / (1.0 + np.exp(-np.clip(self.z2, -20, 20)))
        return self.a2

    def backward(self, X, y):
        m = X.shape[0]
        # Cross-Entropy + Sigmoid derivative simplifies to:
        dz2 = self.a2 - y
        
        dW2 = np.dot(self.a1.T, dz2) / m
        db2 = np.sum(dz2, axis=0, keepdims=True) / m
        
        da1 = np.dot(dz2, self.W2.T)
        
        # Backward Activations (Derivatives / Chain Rule)
        if self.act == 'relu':
            dz1 = da1 * (self.z1 > 0).astype(float)
        elif self.act == 'leaky_relu':
            dz1 = da1 * np.where(self.z1 > 0, 1.0, self.alpha)
        elif self.act == 'elu':
            # derivative of ELU for z < 0 is alpha * e^z = f(z) + alpha = a1 + alpha
            dz1 = da1 * np.where(self.z1 > 0, 1.0, self.a1 + self.alpha)
            
        dW1 = np.dot(X.T, dz1) / m
        db1 = np.sum(dz1, axis=0, keepdims=True) / m
        
        # Gradient Descent Updates
        self.W1 -= self.lr * dW1
        self.b1 -= self.lr * db1
        self.W2 -= self.lr * dW2
        self.b2 -= self.lr * db2

    def train(self, X, y, epochs=150):
        for _ in range(epochs):
            y_pred = self.forward(X)
            # Binary Cross Entropy Loss
            loss = -np.mean(y * np.log(y_pred + 1e-12) + (1 - y) * np.log(1 - y_pred + 1e-12))
            self.loss_history.append(loss)
            self.backward(X, y)

    def predict(self, X):
        return (self.forward(X) >= 0.5).astype(int)

# Train the three network variants from scratch!
epochs_count = 150

# 1. Standard ReLU Network
net_relu = CustomDeepNetwork(activation='relu', lr=0.05)
net_relu.train(X_train, y_train_mat, epochs=epochs_count)

# 2. Leaky ReLU Network
net_leaky = CustomDeepNetwork(activation='leaky_relu', alpha=0.1, lr=0.05)
net_leaky.train(X_train, y_train_mat, epochs=epochs_count)

# 3. ELU Network
net_elu = CustomDeepNetwork(activation='elu', alpha=1.0, lr=0.05)
net_elu.train(X_train, y_train_mat, epochs=epochs_count)

# Calculate Accuracy on Unseen Test Data
acc_relu = accuracy_score(y_test, net_relu.predict(X_test)) * 100
acc_leaky = accuracy_score(y_test, net_leaky.predict(X_test)) * 100
acc_elu = accuracy_score(y_test, net_elu.predict(X_test)) * 100

# Present Benchmark Accuracy Table
df_benchmark = pd.DataFrame([
    {"Activation Function": "Standard ReLU", "Mathematical Derivative ($z < 0$)": "Zero (Dead Neurons)", "Test Accuracy": f"{acc_relu:.2f}%"},
    {"Activation Function": "Leaky ReLU", "Mathematical Derivative ($z < 0$)": "Constant $\\alpha = 0.1$", "Test Accuracy": f"{acc_leaky:.2f}%"},
    {"Activation Function": "Exponential Linear Unit (ELU)", "Mathematical Derivative ($z < 0$)": "Smooth $f(z) + \\alpha$", "Test Accuracy": f"{acc_elu:.2f}%"}
])
display_side_by_side_datasets([df_benchmark], ["بنچمارک دقت توابع فعال‌سازی در یادگیری ساختار ترافیک (DDoS)"])

# ==============================================================================
# 5. VISUALIZATIONS AND ANALYTICAL CHARTS (ACADEMIC BLUE)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))

#  CHART 1: Mathematical Profile of Responsive Activations 
axes[0].plot(z_domain, relu_vals, color='#7f8c8d', linestyle=':', linewidth=2.5, label='Standard ReLU $f(z)$')
axes[0].plot(z_domain, leaky_relu_vals, color='#1abc9c', linewidth=3.0, label=f'Leaky ReLU ($\\alpha={alpha_leaky}$)')
axes[0].plot(z_domain, elu_vals, color='#2c3e50', linewidth=3.0, label=f'ELU ($\\alpha={alpha_elu}$)')
axes[0].plot(z_domain, prelu_vals, color='#2980b9', linestyle='--', linewidth=2.5, label=f'PReLU ($a={a_param}$)')

# Annotate Active Gradient in Negative Subspace
axes[0].axvspan(-4.0, 0.0, color='#bdc3c7', alpha=0.3, label='Negative Subspace (Dying Zone for ReLU)')
axes[0].axhline(0, color='black', linewidth=1.0)
axes[0].axvline(0, color='black', linewidth=1.0)

axes[0].annotate(
    'Dying ReLU Problem\nZero Gradient for $z < 0$',
    xy=(-2.5, 0.0), xytext=(-3.8, 1.8),
    fontsize=10, fontweight='bold', color='#c0392b',
    arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.25", fc="#fdf2e9", ec="#e74c3c", lw=1.5)
)

axes[0].annotate(
    'Non-Zero Responsive Slope\nPrevents Dead Neurons',
    xy=(-2.5, -0.25), xytext=(-3.8, -1.8),
    fontsize=10, fontweight='bold', color='#16a085',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.25", fc="#ebf5fb", ec="#1abc9c", lw=1.5)
)

axes[0].set_title('1. Mathematical Behavior: ReLU vs Responsive Activations', fontsize=13, fontweight='bold', pad=12)
axes[0].set_xlabel('Pre-Activation Input Net Value ($z$)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Activation Output Value $f(z)$', fontsize=12, fontweight='bold')
axes[0].set_ylim(-2.2, 4.2)
axes[0].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=9.5, loc='upper left')
axes[0].grid(True, linestyle='--', alpha=0.6)


#  CHART 2: Loss Convergence Curve (Custom Neural Network) 
epochs_axis = np.arange(1, epochs_count + 1)

axes[1].plot(epochs_axis, net_relu.loss_history, color='#7f8c8d', linestyle=':', linewidth=2.5, label='Standard ReLU Loss (Stalling/Dead Neurons)')
axes[1].plot(epochs_axis, net_leaky.loss_history, color='#1abc9c', linewidth=3.5, label='Leaky ReLU Loss (Fastest Descent)')
axes[1].plot(epochs_axis, net_elu.loss_history, color='#2c3e50', linewidth=3.5, label='ELU Loss (Smooth Convergence)')

axes[1].annotate(
    f'Leaky ReLU Test Acc: {acc_leaky:.1f}%\nStandard ReLU Acc: {acc_relu:.1f}%',
    xy=(epochs_count, net_leaky.loss_history[-1]),
    xytext=(epochs_count - 50, net_leaky.loss_history[-1] + 0.15),
    fontsize=10, fontweight='bold', color='#2c3e50',
    arrowprops=dict(arrowstyle="->", color="#1abc9c", lw=2.0),
    bbox=dict(boxstyle="round,pad=0.25", fc="#ebf5fb", ec="#1abc9c", lw=1.5)
)

axes[1].set_title('2. Loss Curve Convergence across Epochs (DDoS Detection)', fontsize=13, fontweight='bold', pad=12)
axes[1].set_xlabel('Training Epochs (Backpropagation Updates)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Binary Cross-Entropy Loss (Objective)', fontsize=12, fontweight='bold')
axes[1].legend(frameon=True, facecolor='#f8f9fa', edgecolor='#bdc3c7', fontsize=10, loc='upper right')
axes[1].grid(True, linestyle='--', alpha=0.6)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

# ==========================================================
# Section 5: Strategic Conclusion
# ==========================================================
insight_conclusion = (
    r"<b>نتیجه استراتژیک :</b> الگوریتم Backpropagation با ضرب کردنِ مشتقات کار می‌کند. در گراف ۱، "
    r"می‌بینیم که خطِ طوسی‌رنگ (تابع استاندارد ReLU) در سمت چپ محور Y کاملاً افقی است (مشتق = صفر). در این کد ما شبکه عصبی را <b>از صفر</b> نوشتیم "
    r"تا دقیقاً ببینیم چه اتفاقی می‌افتد: در متد <code>backward</code>، اگر خروجی نورون منفی شود، خطای مشتق‌گیری صفر می‌شود و دیگر هیچ سیگنالِ "
    r"خطایی به لایه‌های پایین‌تر منتقل نمی‌گردد (فلج شدنِ بخشی از مغز ماشین). "
    r"اما توابعِ واکنشی مثل <b>Leaky ReLU</b> (خط فیروزه‌ای) و <b>ELU</b> (خط سرمه‌ای) یک شیبِ بسیار کوچک در ناحیه منفی تعبیه کرده‌اند. "
    r"این نشتیِ کوچک تضمین می‌کند که جریان گرادیان هرگز متوقف نشود. ثمرهٔ این معماریِ دست‌نویس در گراف ۲ کاملاً مشهود است: خط فیروزه‌ای "
    r"و سرمه‌ای با سرعت و پایداریِ بسیار بیشتری نسبت به خطِ نقطه‌چینِ طوسی افت می‌کنند و به دقت بالاتری می‌رسند، زیرا تمام نورون‌های شبکه "
    r"تا آخرین اپوکِ آموزشی بیدار و در حال یادگیری باقی مانده‌اند."
)
display_academic_text(insight_conclusion, "insight")
```

در این پیاده‌سازی کاربردی و مقایسه‌ای از توابع فعال‌سازی واکنشی (Responsive Activation Functions):
</p>
<ul style="font-size:15pt; line-height:250%; color:#2B2D42;">
<li>تعداد ۱,۲۰۰ بسته ترافیک شبکه کلودفلر شامل ۱۲ ویژگی با تابع <code>make_classification</code> برای تفکیک حملات سایبری DDoS از ترافیک سالم کاربران شبیه‌سازی شد.</li>
<li>پیاده‌سازی فرمول‌های ریاضی و مشتقات از صفر (From Scratch):
  <ul>
    <li><b>مدل‌سازی تابع Leaky ReLU:</b> با در نظر گرفتن ضریب نشتی $\alpha = 0.1$، شیب مثبت در ناحیه منفی ورودی ($z < 0$) حفظ شد تا نورون‌ها در صورت دریافت مقادیر منفی وارد حالت کما یا مرگ دائم نشوند.</li>
    <li><b>مدل‌سازی تابع نمایی ELU:</b> با استفاده از فرمول $\alpha (e^z - 1)$، انتقال ملایم و پیوسته در نقطه صفر مدل‌سازی گردید که میانگین فعال‌سازی‌ها را به صفر نزدیک‌تر کرده و اثر نوسانات گرادیان را کاهش می‌دهد.</li>
    <li><b>مدل‌سازی تابع PReLU:</b> با معرفی ضریب شیب متغیر $a=0.25$، انعطاف‌پذیری تطبیقی شبکه در یادگیری شیب بهینه نمایش داده شد.</li>
  </ul>
</li>
<li>ساخت مدل‌های یادگیری عمیق Keras و تحلیل نمودارها:
  <ul>
    <li>نمودار اول (رفتار ریاضی توابع در ناحیه ورودی): ناحیه قرمزرنگ ورودی‌های منفی را به تصویر می‌کشد؛ خط‌چین خاکستری (ReLU استاندارد) شیب کاملاً صفر تولید می‌کند، در حالی که خطوط سبز (Leaky ReLU) و قهوه‌ای (ELU) گرادیان‌های غیرصفر و پیوسته برقرار کرده و جریان یادگیری پس‌انتشار را تضمین می‌نمایند.</li>
    <li>نمودار دوم (منحنی همگرایی زیان اعتبارسنجی): روند افت خطای دسته‌بندی ترافیک کلودفلر را مقایسه می‌کند؛ مدل‌های مجهز به <code>LeakyReLU</code> و <code>ELU</code> با جلوگیری از پدیده Dying ReLU، سرعت همگرایی را بیش از ۳۰٪ افزایش داده و به دقت نهایی بالاتری در تفکیک حملات سایبری دست یافته‌اند.</li>
  </ul>
</li>
</ul>
